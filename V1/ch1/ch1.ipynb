<p style="font-size:18pt; text-align:center;">
<span id="section10" style="font-size:24pt; font-weight:bold; color:#FF6700;">۱. مقدمه‌ای بر بازیابی اطلاعات: سفر به قلب وب امروزی</span> 🧠
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 تصور کنید یک مغز دیجیتال فوق‌العاده</p>
<p style="font-size:16pt; text-align:right;">
شما هر روز چند بار از خود می‌پرسید: «اون ایمیل مهم کجا رفت؟»، «مقاله‌ای که دیروز خوندم رو کجا ذخیره کردم؟» یا «چطور می‌تونم یه فیلم خوب با بازیگرای مورد علاقه‌م پیدا کنم؟». در پشت هر یک از این سوالات ساده، یک غول نامرئی به نام <strong>بازیابی اطلاعات</strong> (Information Retrieval) کار می‌کند. این علم، هنر یافتن سوزنِ اطلاعات در انبار کاه دیجیتال ماست.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
دهه‌ها پیش، این یک مهارت تخصصی برای کتابداران و وکلای پژوهشگر بود. امروز، اما صدها میلیون نفر هر روز از آن استفاده می‌کنند—هر بار که در گوگل عبارتی را تایپ می‌کنند یا در صندوق ورودی ایمیل خود کلمه‌ای را جستجو می‌نمایند. بازیابی اطلاعات، دیگر یک فعالیت حاشیه‌ای نیست؛ بلکه به dominant form of information access تبدیل شده است.
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 چالش اصلی: داده‌های بی‌نظم و نامنظم</p>
<p style="font-size:16pt; text-align:right;">
قلب این علم، کار با داده‌های <strong>ناساختاریافته</strong> (Unstructured Data) است. داده‌های ساختاریافته، مانند یک جدول اکسل منظم یا یک پایگاه داده فروشگاه، ستون‌ها و ردیف‌های مشخصی دارند. اما بیشتر اطلاعات دنیای ما—مانند ایمیل‌ها، مقالات، اخبار و صفحات وب—مثل یک اتاق پر از وسایل هستند که هرچیز جایی ریخته و هیچ برچسبی ندارد. علم بازیابی اطلاعات به کامپیوتر یاد می‌دهد که چطور در این اتاق شلوغ، مورد مورد نظر ما را پیدا کند.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
البته، تقریباً هیچ داده‌ای کاملاً بی‌ساختار نیست. متون هم دارای ساختارهای پنهان زبانی هستند. سیستم‌های بازیابی مدرن از این ساختارها مانند عناوین، پاراگراف‌ها و حتی لینک‌ها (hyperlinks) بهره می‌برند تا جستجو را <strong>نیمه‌ساختاریافته</strong> (Semi-structured) و هوشمندانه‌تر کنند.
</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🗂️ فراتر از جستجو: سازماندهی و طبقه‌بندی</p>
<p style="font-size:16pt; text-align:right;">
بازیابی اطلاعات فقط پیدا کردن چند سند مرتبط نیست. این شامل وظایف هوشمندانه‌تری است:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>خوشه‌بندی (Clustering):</strong> تصور کنید صدها سند علمی در اختیار دارید. سیستم می‌تواند آن‌ها را به صورت خودکار به گروه‌های موضوعی تقسیم کند، درست مانند زمانی که کتاب‌های یک قفسه را بر اساس موضوع مرتب می‌کنید.</li>
<li><strong>طبقه‌بندی (Classification):</strong> سیستم ایمیل شما هر روز این کار را برای شما انجام می‌دهد. به صورت خودکار تشخیص می‌دهد که کدام ایمیل اسپم است و کدامیک مهم و آن را در پوشه مناسب قرار می‌دهد.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
با توجه به مقیاس عملکرد، سیستم‌های بازیابی اطلاعات به سه دسته اصلی تقسیم می‌شوند:</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>🌐 جستجوی وب (Web Scale):</strong> بزرگترین و پیچیده‌ترین حوزه. این سیستم‌ها باید بر روی میلیاردها سند در میلیون‌ها کامپیوتر جستجو کنند. چالش‌ها عظیم هستند: از جمع‌آوری اسناد با خزنده‌های وب (Crawlers) گرفته تا مقابله با ترفندهای سایتی برای دستکاری رتبه‌ها.</li>
<li><strong>💻 جستجوی شخصی (Personal Search):</strong> همان جستجویی که در کامپیوتر خود انجام می‌دهید (مانند Spotlight در مک یا Instant Search در ویندوز). اینجا تمرکز بر سرعت، سبکی و کارایی با منابع محدود است.</li>
<li><strong>🏢 جستجوی سازمانی/حوزه‌محور (Enterprise/Domain-Specific):</strong> جستجو در مجموعه‌های متمرکز مانند اسناد داخلی یک شرکت، پایگاه‌های ثبت اختراع یا مقالات تخصصی شیمی. این سیستم‌ها معمولاً روی چند سرور قدرتمند اجرا می‌شوند.</li>
</ol>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🖼️ یک تصویر به ارزش هزاران کلمه</p>
<p style="font-size:16pt; text-align:right;">
برای درک عمیق‌تر این مفهوم، تصور کنید کاربری که در اقیانوس بی‌پایان داده‌ها (سمت راست) یک نیاز اطلاعاتی دارد (سمت چپ). یک سیستم بازیابی اطلاعات هوشمند (مغز دیجیتال) باید بتواند از میان این حجم عظیم داده‌های نامنظم، دقیقاً همان اسناد مرتبط را استخراج و به کاربر ارائه دهد. این تصویر، قلب تلاش ما در این فصل است: چطور این مغز دیجیتال را بسازیم.
</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/1.png" alt="ocean of information and our role as a homo sapian" style="width:600px; border:1px solid #ccc; border-radius:12px;"
</div>
</div>

<p style="font-size:16pt; text-align:right;">
در این کتاب، ما به بررسی ایده‌ها و تکنیک‌هایی می‌پردازیم که این مغز دیجیتال را می‌سازند؛ از ساختارهای داده‌ای اولیه گرفته تا الگوریتم‌های پیچیده‌ای که دنیای وب را می‌چرخانند. 
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11" style="font-size:24pt; font-weight:bold; color:#FF6700;">۱.۱ یک مسئله نمونه از بازیابی اطلاعات: جستجو در اقیانوس دیجیتال</span> 🎭
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 تصور کنید مدیر یک سرویس استریم فیلمنامه</p>
<p style="font-size:16pt; text-align:right;">
شما مدیر آرشیو دیجیتال غول‌پیکری هستید که شامل تمام فیلمنامه‌های نوشته‌شده در تاریخ است. یک روز، یک کاربر پژوهشگر با یک درخواست بسیار خاص به شما مراجعه می‌کند: «می‌خواهم تمام فیلمنامه‌هایی را پیدا کنم که در آن‌ها <strong>بروتوس</strong> و <strong>سزار</strong> هر دو حضور دارند، اما <strong>کالپورنیا</strong> در هیچ‌کدام حضور ندارد.» چطور این جستجو را به صورت کارآمد انجام می‌دهید؟</p>
</div>

<p style="font-size:16pt; text-align:right;">
یک راه اولیه و شاید بلافاصله به ذهن می‌رسد، استخدام یک کتابدار بسیار پرتلاش و خواندن تک‌تک تمام فیلمنامه‌ها! این روش در دنیای کامپیوتر به <strong>پیمایش خطی</strong> یا <strong>گرِپ کردن (grepping)</strong> معروف است. برای مجموعه‌ای به کوچکی مانند آثار کامل شکسپیر، این روش حتی با سرعت کامپیوترهای امروزی هم قابل قبول است. اما برای میلیاردها سند یا برای پرسمان‌های پیچیده‌تر، این روش مانند تلاش برای پیدا کردن سوزن در انبار کاه است!
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧮 ایده اول: ماتریس وقوع بولی (یک برگه بازرسی غول‌پیکر)</p>
<p style="font-size:16pt; text-align:right;">
اولین راهکار هوشمندانه، ساخت یک <strong>ماتریس وقوع بولی</strong> است. تصور کنید یک برگه غول‌پیکر که در آن هر ردیف به یک واژه (مثل «بروتوس») و هر ستون به یک سند (مثل «هملت») اختصاص دارد. در خانه‌ای که واژه در سند حضور دارد، عدد ۱ و در غیر این صورت ۰ می‌نویسیم.
</p>
<p style="font-size:16pt; text-align:right;">
برای پاسخ به پرسمان <span style="font-family:monospace;">Brutus AND Caesar AND NOT Calpurnia</span>، کافی است بردارهای بولی مربوط به این سه واژه را برداشته و یک عملیات بیتی روی آن‌ها انجام دهیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
    
 $$ [1, 1, 0, 1, 0, 0] \land [1, 1, 0, 1, 1, 1] \land [1, 0, 1, 1, 1, 1] = [1, 0, 0, 1, 0, 0]
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
نتیجه این محاسبه به ما می‌گوید که سندهای شماره ۱ (هملت) و ۳ (مکبث) پاسخ‌های درست هستند. این روش، در تئوری بی‌نقص است، اما در عمل با یک مشکل اساسی روبرو است: <strong>اندازه!</strong>
</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚫 فاجعه مقیاس: ماتریس بیش از حد بزرگ!</p>
<p style="font-size:16pt; text-align:right;">
فرض کنید مجموعه ما فقط یک میلیون سند با ۵۰۰ هزار واژهٔ منحصربه‌فرد دارد. ماتریس ما ابعاد $500{,}000 \times 1{,}000{,}000$ خواهد داشت! این ماتریس حدود <strong>۵۰۰ میلیارد</strong> درایه دارد. حتی اگر هر درایه فقط یک بایت باشد، به بیش از ۴۶۰ گیگابایت فضا نیاز دارد. مشکل بزرگ‌تر این است که این ماتریس فوق‌العاده <strong>تنک (sparse)</strong> است. چون هر سند فقط حدود ۱۰۰۰ واژه دارد، از ۵۰۰ میلیارد درایه، کمتر از ۱ میلیارد آن‌ها ۱ هستند و بقیه (یعنی ۹۹.۸٪) همه صفر هستند! ما در حال ذخیره یک اقیانوس از صفرها هستیم!
</p>
</div>

<div style="background:#d4edda; border-right:4px solid #155724; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#155724;">💡 راه حل نهایی: ایندکس معکوس (نگاهی از زاویه‌ای دیگر)</p>
<p style="font-size:16pt; text-align:right;">
به‌جای پرسیدن «کدام اسناد شامل واژه X هستند؟»، مسئله را برعکس می‌کنیم و می‌پرسیم: «واژه X در کدام اسناد ظاهر شده است؟» این تغییر دیدگاه ساده، به <strong>ایندکس معکوس</strong> (Inverted Index) منجر می‌شود.
</p>
<p style="font-size:16pt; text-align:right;">
این ساختار هوشمندانه از دو بخش اصلی تشکیل شده است:
</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>📖 دیکشنری (Dictionary):</strong> یک فهرست مرتب‌شده از تمام واژه‌هایی که در مجموعه وجود دارند. این مانند فهرست مطالب کتاب است.</li>
<li><strong>📋 لیست ارسال (Postings List):</strong> برای هر واژه در دیکشنری، لیستی از تمام اسنادی که آن واژه در آن‌ها یافت شده است. این لیست‌ها معمولاً بر اساس شماره سند (docID) مرتب هستند.</li>
</ol>

<p style="font-size:16pt; text-align:right;">

<p style="font-size:16pt; text-align:right;">
تصویر زیر (شکل ۱.۲ در کتاب) دقیقاً همان نتیجه‌ای است که از روش ایندکس معکوس به دست می‌آید. این تصویر، لحظه‌ای است که تلاش فکری ما به نتیجه ملموس تبدیل می‌شود. این تصویر نشان می‌دهد که اسناد ۱ و ۳ (هملت و مکبث) پاسخ صحیح پرسمان ما هستند، دقیقاً همان چیزی که با محاسبات بیتی به دست آوردیم. این تصویر، پلی است بین تئوری ماتریس و کاربرد عملی ایندکس معکوس.

<p style="font-size:16pt; text-align:right;">
ایدهٔ ایندکس معکوس، سنگ بنای بسیاری از سیستم‌های بازیابی اطلاعات مدرن، از گوگل گرفته تا جستجوی فایل‌های محلی شما، است. شما در حال یادگیری یکی از بنیادی‌ترین و قدرتمندترین مفاهیم در دنیای دیجیتال هستید. 🚀

```python
######## اگر یکسری از جاهای کد را متوجه نمیشید، نگران نباشید چون در ادامه‌ی جزوه در مورد تمام کارایی که اینجا داریم میکنیم بیشتر صحبت میکنیم. ########



# --- تعریف ساختار داده‌ها ---

# نمونه‌ای از مجموعه اسناد (در اینجا: نمایشنامه‌های شکسپیر)
documents = {
    "Antony": "Agrippa [Aside to Domitius Enobarbus]: Why, Enobarbus, When Antony found Julius Caesar dead, He cried almost to roaring; and he wept When at Philippi he found Brutus slain.",
    "Hamlet": "Lord Polonius: I did enact Julius Caesar: I was killed i' the Capitol; Brutus killed me.",
    "Othello": "Had I but one hour tonight-- Othello, I am not well.",
    "Macbeth": "If it were done when 'twere done, then 'twere well it were done quickly."
}
print(documents)
print()

# --- پیاده‌سازی الگوریتم‌ها ---

def build_term_document_matrix(docs):
    """
    این تابع ماتریس وقوع بولی (term-document incidence matrix) را می‌سازد.
    خروجی: یک دیکشنری که کلیدها واژه‌ها و مقادیر دیکشنری‌های دیگر هستند
    """
    print("مرحله ۱: ساخت ماتریس وقوع بولی...")
    all_terms = set()
    # استخراج تمام واژه‌های یکتا از همه اسناد
    for doc_name, content in docs.items():
        all_terms.update(content.lower().split())

    # ساخت ماتریس با مقادیر اولیه صفر
    term_doc_matrix = {term: {} for term in all_terms}

    # پر کردن ماتریس بر اساس وجود واژه در هر سند
    for term in all_terms:
        for doc_name, content in docs.items():
            term_doc_matrix[term][doc_name] = 1 if term in content.lower() else 0

    print("ماتریس وقوع بولی ساخته شد.")
    return term_doc_matrix

def boolean_query(matrix, query):
    """
    این تابع یک پرسمان بولی را بر اساس ماتریس وقوع اجرا می‌کند.
    ورودی: ماتریس وقوع و پرسمان (مثال: ['Brutus', 'AND', 'Caesar', 'AND', 'NOT', 'Calpurnia'])
    """
    print(f"\nمرحله ۲: اجرای پرسمان بولی: {' '.join(query)}")
    
    # نرمال‌سازی پرسمان: تبدیل واژه‌ها به حروف کوچک (اما نه عملگرها)
    normalized_query = []
    for token in query:
        if token in ['AND', 'OR', 'NOT']:
            normalized_query.append(token)
        else:
            normalized_query.append(token.lower())  # واژه‌ها را کوچک می‌کنیم
    
    # جداسازی پرسمان به بخش‌های مثبت و منفی
    positive_terms = []
    negative_terms = []
    i = 0
    while i < len(normalized_query):
        token = normalized_query[i]
        if token == 'NOT':
            if i + 1 < len(normalized_query) and normalized_query[i + 1] not in ['AND', 'OR', 'NOT']:
                negative_terms.append(normalized_query[i + 1])
                i += 2
                continue
        elif token not in ['AND', 'OR']:
            positive_terms.append(token)
        i += 1

    # اگر واژه‌ای وجود نداشت، هیچ سندی برگردانده نشود
    if not positive_terms:
        print("نتیجه پرسمان: []")
        return []

    # شروع با اسناد حاوی اولین واژه مثبت
    first_term = positive_terms[0]
    if first_term not in matrix:
        result_docs = set()
    else:
        result_docs = set(doc for doc, exists in matrix[first_term].items() if exists)

    # اعمال AND برای سایر واژه‌های مثبت
    for term in positive_terms[1:]:
        if term in matrix:
            term_docs = set(doc for doc, exists in matrix[term].items() if exists)
            result_docs = result_docs.intersection(term_docs)
        else:
            result_docs = set()  # واژه‌ای نیست → نتیجه خالی
            break

    # اعمال NOT
    for term in negative_terms:
        if term in matrix:
            term_docs = set(doc for doc, exists in matrix[term].items() if exists)
            result_docs = result_docs.difference(term_docs)

    print(f"نتیجه پرسمان: {sorted(result_docs)}")
    return sorted(result_docs)

def build_inverted_index(matrix):
    """
    این تابع یک ایندکس معکوس از روی ماتریس وقوع می‌سازد.
    این تابع، پل ارتباطی بین ماتریس و ایندکس معکوس را نشان می‌دهد.
    """
    print("\nمرحله ۳: ساخت ایندکس معکوس از ماتریس...")
    inverted_index = {}

    for term, doc_dict in matrix.items():
        # استخراج اسنادی که واژه در آن‌ها وجود دارد
        docs_with_term = [doc for doc, exists in doc_dict.items() if exists]
        if docs_with_term:
            inverted_index[term] = sorted(docs_with_term)

    print("ایندکس معکوس ساخته شد.")
    # print(inverted_index)
    return inverted_index

# --- اجرای مثال ---

if __name__ == "__main__":
    # مرحله ۱: ساخت ماتریس وقوع بولی
    term_doc_matrix = build_term_document_matrix(documents)

    # نمایش ماتریس برای چند واژه نمونه
    sample_terms = ['Brutus', 'Caesar', 'Calpurnia']
    print(f"\nنمونه‌ای از ماتریس وقوع برای واژه‌های {sample_terms}:")
    for term in sample_terms:
        if term in term_doc_matrix:
            print(f"{term}: {term_doc_matrix[term]}")

    # مرحله ۲: اجرای پرسمان بولی
    query = ['Brutus', 'AND', 'Caesar', 'AND', 'NOT', 'Calpurnia']
    result = boolean_query(term_doc_matrix, query)

    # مرحله ۳: ساخت ایندکس معکوس
    inverted_index = build_inverted_index(term_doc_matrix)

    # نمایش ساختار ایندکس معکوس
    sample_terms = ['Brutus', 'Caesar', 'Calpurnia']
    print(f"\nنمونه‌ای از ماتریس وقوع برای واژه‌های {sample_terms}:")
    for term in sample_terms:
        lower_term = term.lower()
        if lower_term in term_doc_matrix:
            print(f"{term}: {term_doc_matrix[lower_term]}")
        else:
            print(f"{term}: (وجود ندارد در ماتریس)")

    print("\nاین فرآیند، چگونگی تبدیل یک مسئله جستجوی خطی به یک راه‌حل بهینه (ایندکس معکوس) را نشان می‌دهد.")
```

<p style="font-size:16pt; text-align:right;">
برای بهره‌گیری از سرعت فوق‌العاده در زمان جستجو، باید این فهرست را <strong>از قبل</strong> بسازیم. این فرآیند مانند تهیه یک دستور آشپزی دقیق است و چهار مرحله اصلی دارد:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>۱. جمع‌آوری مواد اولیه (اسناد)</strong><br>
همه پست‌های IG را جمع می‌کنیم. این «مجموعه اسناد» ماست.</li>
<li><strong>۲. خرد کردن (توکن‌سازی)</strong><br>
هر پست را به واحدهای کوچک‌تری به نام «توکن» (Token) می شکنیم. در دنیای ما، این یعنی جدا کردن تمام کلمات و هشتگ‌ها. 🧩</li>
<li><strong>۳. تمیزکاری و آماده‌سازی (پیش‌پردازش زبانی)</strong><br>
توکن‌ها را نرمال‌سازی می‌کنیم: همه را به حروف کوچک تبدیل می‌کنیم، علائم نگارشی مانند «!» و «#» را حذف می‌کنیم و ریشه کلمات را پیدا می‌کنیم. 🧠</li>
<li><strong>۴. ساخت دفترچه تلفن هوشمند (ایندکس‌سازی)</strong><br>
حالا مهم‌ترین بخش کار: ساختن خود فهرست معکوس. 📚</li>
</ol>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎩 جادوی مرتب‌سازی: از آشفتگی تا نظم</p>
<p style="font-size:16pt; text-align:right;">
چگونه این دفترچه تلفن را می‌سازیم؟ روش اولیه و هوشمندانه، <strong>ایندکس‌سازی مبتنی بر مرتب‌سازی</strong> است:
</p>

<p style="font-size:16pt; text-align:right;">
<strong>مرحله الف: آشفت‌گاه اولیه</strong><br>
ابتدا یک لیست بزرگ از تمام جفت‌های (توکن، شناسه_پست) برای کل مجموعه اسناد می‌سازیم. مثلاً:

<p style="font-size:16pt; text-align:right;">
<strong>مرحله ب: نظم جادویی</strong><br>
حالا این لیست شلوغ را بر اساس توکن‌ها (الفبا) مرتب می‌کنیم. سپس موارد تکراری را گروه‌بندی می‌کنیم. نتیجه چیزی شبیه به این می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">

این ساختار مرتب، قلب تپنده فهرست معکوس ماست.

<p style="font-size:16pt; text-align:right;">
در نهایت، این دفترچه تلفن هوشمند به دو بخش اصلی تقسیم می‌شود:
</p>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">📚 دو بخش اصلی فهرست معکوس</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>دیکشنری (Dictionary):</strong> این بخش مانند فهرست نام‌ها در دفترچه تلفن است. شامل تمام توکن‌های منحصربه‌فرد (مانند <code>سفر</code>، <code>غروب_آفتاب</code>) است. برای هر توکن، یک آمار کلیدی هم ذخیره می‌شود: <strong>فراوانی سند</strong> (Document Frequency یا $df_t$)؛ یعنی تعداد کل پست‌هایی که آن توکن در آن‌ها وجود دارد. این دیکشنری آنقدر کوچک است که می‌توانیم آن را کاملاً در حافظه اصلی (RAM) نگه داریم تا دسترسی به آن فوری باشد.</li>
<li><strong>لیست ارسال (Postings List):</strong> این بخش مانند شماره تلفن‌های مربوط به هر نام در دفترچه تلفن است. برای هر توکن در دیکشنری، لیستی از شناسه‌های پست‌هایی (docID) که آن توکن در آن‌ها وجود دارد، ذخیره می‌شود. این لیست‌ها بسیار حجیم هستند و روی دیسک ذخیره می‌شوند، اما <strong>مرتب‌شده بر اساس docID</strong> هستند. این مرتب‌سازی، کلید سرعت پردازش پرسمان‌های بولی است که در بخش قبل دیدیم.</li>
</ol>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🤓 عمیق‌تر: بهترین ساختار برای لیست ارسال چیست؟</p>
<p style="font-size:16pt; text-align:right;">
برای ذخیره لیست ارسال در حافظه، دو گزینه اصلی وجود دارد و انتخاب بین آن‌ها یک معامله بین انعطاف‌پذیری، سرعت و فضا است:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>لیست پیوندی تک‌طرفه (Singly Linked List):</strong> مثل قطاری از واگن‌ها که به هم متصل‌اند. <strong>مزیت:</strong> اضافه کردن یک پست جدید به لیست بسیار سریع و ارزان است. <strong>عیب:</strong> به حافظه پراکنده نیاز دارد و پیمایش آن ممکن است کندتر باشد. این ساختار، پایه‌ای برای تکنیک‌های پیشرفته‌تر مانند <em>Skip Lists</em> است.</li>
<li><strong>آرایه با طول متغیر (Variable Length Array):</strong> مثل یک قطار یکپارچه با تعداد واگن متغیر. <strong>مزیت:</strong> بسیار فشرده‌تر است و چون در حافظه پشت سر هم قرار گرفته، پیمایش آن در پردازنده‌های مدرن بسیار سریع‌تر است. <strong>عیب:</strong> اضافه کردن عضو جدید ممکن است نیازمند ساختن کل آرایه جدیدی باشد.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در دنیای واقعی، اگر به‌روزرسانی زیاد باشد، از لیست پیوندی استفاده می‌کنیم. اما اگر اسناد کمتر تغییر کنند، آرایه‌ها به دلیل فشردگی و سرعت بالا، برنده هستند! 💾
</p>

این ساختار به ظاهر ساده، <strong>بی‌رقیبترین و کارآمدترین روش</strong> برای پشتیبانی از جستجوی متنی موردی (ad-hoc) است. هر بار که در گوگل عبارتی را جستجو می‌کنید یا در یک پایگاه داده حقوقی به دنبال یک سند خاص می‌گردید، در حال استفاده از نسخه‌ای بسیار پیچیده‌تر از همین ایده ساده و زیبا هستید.

اگر به این موضوع علاقه‌مند شدی، پیشنهاد می‌کنم مقاله "Beyond Boolean: The Evolution of Information Retrieval Models" را مطالعه کنی. این مقاله نشان می‌دهد که چگونه مدل‌های بولی به مدل‌های پیچیده‌تر مانند مدل فضای برداری و مدل‌های احتمالی تبدیل شده‌اند. 📖

```python
# --- تعریف ساختار داده‌ها ---

# نمونه‌ای از یک ایندکس معکوس (ساده‌شده)
# کلیدها واژه‌ها و مقادیر، لیست‌های ارسال (postings lists) مرتب‌شده هستند.
# این ساختار، قلب یک موتور جستجوی بولی است.
inverted_index = {
    "Brutus": [1, 2, 4, 11, 31, 45, 173, 174],
    "Calpurnia": [2, 31, 54, 101],
    "Caesar": [1, 2, 4, 5, 6, 16, 57, 132],
    "Antony": [5, 10, 31],
    "Cleopatra": [2, 3, 31, 64]
}

# فرض می‌کنیم مجموعه اسناد ما شامل شناسه‌های ۱ تا ۲۰۰ است.
# این لیست برای عملگر NOT ضروری است تا بتوانیم مکمل یک مجموعه را محاسبه کنیم.
ALL_DOCUMENTS = list(range(1, 201))


# --- پیاده‌سازی الگوریتم‌های اصلی ---

def intersect(p1, p2):
    """
    الگوریتم ادغام دوطرفه برای یافتن اشتراک دو لیست مرتب‌شده.
    این پیاده‌سازی دقیقاً مطابق با شبه‌کد کتاب و توضیحات مارک‌دان سل است.
    این الگوریتم به صورت خطی و با حداکثر کارایی O(len(p1) + len(p2)) عمل می‌کند.
    """
    print(f"در حال محاسبه اشتراک لیست‌ها: {p1} و {p2}")
    i, j = 0, 0  # دو نشانگر (pointer) برای حرکت در لیست‌ها
    result = []
    # تا زمانی که هر دو نشانگر به انتها نرسیده‌اند، به مقایسه ادامه می‌دهیم
    while i < len(p1) and j < len(p2):
        # اگر آیتم‌ها برابر بودند، در نتیجه ذکر شده و هر دو نشانگر جلو می‌روند
        if p1[i] == p2[j]:
            result.append(p1[i])
            i += 1
            j += 1
        # اگر آیتم لیست اول کوچکتر بود، فقط نشانگر اول را جلو می‌بریم
        elif p1[i] < p2[j]:
            i += 1
        # در غیر این صورت، نشانگر دوم را جلو می‌بریم
        else:
            j += 1
    print(f"نتیجه اشتراک: {result}")
    return result

def union(p1, p2):
    """
    الگوریتم ادغام برای یافتن اجتماع دو لیست مرتب‌شده (بدون تکرار).
    این تابع برای پردازش عملگر OR استفاده می‌شود.
    """
    print(f"در حال محاسبه اجتماع لیست‌ها: {p1} و {p2}")
    i, j = 0, 0
    result = []
    # تا زمانی که هر دو لیست آیتم دارند
    while i < len(p1) and j < len(p2):
        if p1[i] == p2[j]:
            result.append(p1[i])
            i += 1
            j += 1
        elif p1[i] < p2[j]:
            result.append(p1[i])
            i += 1
        else:
            result.append(p2[j])
            j += 1
    # اضافه کردن آیتم‌های باقی‌مانده در هر دو لیست
    result.extend(p1[i:])
    result.extend(p2[j:])
    print(f"نتیجه اجتماع: {result}")
    return result

def complement(postings_list):
    """
    تابع مکمل (NOT) برای پیاده‌سازی عملگر نفی.
    این عملیات هزینه‌بر است چون به کل مجموعه اسناد نیاز دارد.
    """
    print(f"در حال محاسبه مکمل لیست: {postings_list}")
    p1_iter = iter(postings_list)
    current_doc_to_exclude = next(p1_iter, None)
    result = []
    for doc_id in ALL_DOCUMENTS:
        if current_doc_to_exclude is not None and doc_id == current_doc_to_exclude:
            current_doc_to_exclude = next(p1_iter, None)
        else:
            result.append(doc_id)
    # برای نمایش بهتر، فقط ۱۰ سند اول نتیجه را چاپ می‌کنیم
    print(f"نتیجه مکمل (۱۰ سند اول): {result[:10]}...")
    return result

def get_postings(term):
    """تابع کمکی برای گرفتن لیست ارسال یک واژه."""
    return inverted_index.get(term, [])

def process_conjunctive_query(terms):
    """
    پردازش یک پرسمان ترکیبی (conjunctive query) با بهینه‌سازی.
    واژه‌ها بر اساس فراوانی (طول لیست ارسال) به ترتیب صعودی پردازش می‌شوند.
    این مهم‌ترین تکنیک بهینه‌سازی برای پرسمان‌های بولی است.
    """
    print(f"\n--- پردازش پرسمان AND: {' AND '.join(terms)} ---")
    
    # مرحله ۱: مرتب‌سازی واژه‌ها بر اساس طول لیست ارسال (بهینه‌سازی کلیدی)
    terms_with_freq = [(term, len(get_postings(term))) for term in terms]
    terms_with_freq.sort(key=lambda x: x[1])
    sorted_terms = [term for term, freq in terms_with_freq]
    print(f"ترتیب بهینه پردازش واژه‌ها (بر اساس فراوانی صعودی): {sorted_terms}")

    # مرحله ۲: شروع با کوچکترین لیست و اشتراک گرفتن تدریجی
    result = get_postings(sorted_terms[0])
    print(f"شروع با لیست کوچکترین واژه '{sorted_terms[0]}': {result}")
    
    for term in sorted_terms[1:]:
        if not result:  # اگر نتیجه میانی خالی شد، دیگر ادامه نده
            print("نتیجه میانی خالی شد، پردازش متوقف شد.")
            break
        result = intersect(result, get_postings(term))
        
    print(f"نتیجه نهایی پرسمان: {result}")
    return result

# --- اجرای مثال‌ها ---

if __name__ == "__main__":
    
    # مثال ۱: پرسمان ساده (Brutus AND Calpurnia)
    print("="*50)
    print("مثال ۱: پرسمان ساده")
    brutus_postings = get_postings("Brutus")
    calpurnia_postings = get_postings("Calpurnia")
    intersect(brutus_postings, calpurnia_postings)

    # مثال ۲: پرسمان ترکیبی با بهینه‌سازی (Brutus AND Caesar AND Calpurnia)
    print("\n" + "="*50)
    print("مثال ۲: پرسمان ترکیبی با بهینه‌سازی")
    process_conjunctive_query(["Brutus", "Caesar", "Calpurnia"])

    # مثال ۳: پرسمان پیچیده با OR و NOT
    print("\n" + "="*50)
    print("مثال ۳: پرسمان پیچیده با OR و NOT")
    # (Brutus OR Caesar) AND NOT (Antony OR Cleopatra)
    
    # مرحله ۱: محاسبه (Brutus OR Caesar)
    brutus_caesar_union = union(get_postings("Brutus"), get_postings("Caesar"))
    
    # مرحله ۲: محاسبه (Antony OR Cleopatra)
    antony_cleopatra_union = union(get_postings("Antony"), get_postings("Cleopatra"))
    
    # مرحله ۳: محاسبه NOT (Antony OR Cleopatra)
    # هشدار: این عملیات بسیار سنگین است!
    # antony_cleopatra_complement = complement(antony_cleopatra_union)
    
    # مرحله ۴: محاسبه نهایی (اشتراک)
    # نتیجه نهایی = intersect(brutus_caesar_union, antony_cleopatra_complement)
    # برای جلوگیری از اجرای عملیات سنگین NOT، آن را کامنت می‌کنیم.
    print("برای جلوگیری از عملیات سنگین NOT، محاسبه نهایی انجام نمی‌شود.")
    print("در عمل، موتورهای جستجو از تکنیک‌های هوشمندانه‌تری برای پردازش NOT استفاده می‌کنند.")
```

<li>استفاده زیاد از <code>OR</code> → نتایج بسیار <strong>زیاد</strong> ولی بسیاری از آن‌ها <strong>غیرمربوط</strong>. (دقت پایین، بازیابی بالا)</li>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>🔍 جستجوی بخشنده (Tolerant Search):</strong> سیستمی که اشتباهات املایی شما را می‌بخشد و می‌فهمد که «ماشین» و «دستگاه» می‌توانند مترادف باشند.</li>
<li><strong>🧩 جستجوی بافتار (Proximity & Phrase Search):</strong> سیستمی که تفاوت بین «سگ گاز می‌گیرد» و «انسان به سگ گاز می‌گیرد» را می‌فهمد. این یعنی درک نزدیکی و ترتیب کلمات.</li>
<li><strong>📊 اهمیت فراوانی (Term Frequency):</strong> سیستمی که می‌داند اگر کلمه «هوش مصنوعی» در یک سند ۱۰ بار تکرار شده، آن سند احتمالاً از سندی که فقط یک بار به آن اشاره کرده، مرتبط‌تر است.</li>
<li><strong>🥇 هنر رتبه‌بندی (Ranking):</strong> سیستمی که به جای گفتن «بله/خیر»، به هر سند یک امتیاز از ۱۰۰ می‌دهد و بهترین‌ها را در بالای لیست قرار می‌دهد.</li>
</ol>

این چهار ستون، فناوری است که امروزه دنیا را تسخیر کرده است. بار بعدی که در گوگل عبارتی را جستجو می‌کنید، به یاد داشته باشید که در پشت این کادر ساده، این ایده‌ها با قدرت در حال کار هستند. گوگل ممکن است ظاهراً یک جستجوی متنی آزاد ارائه دهد، اما در پشت صحنه از جستجوی عبارت (دستور دوم) و نسخه‌های ساده‌شده‌ای از منطق بولی استفاده می‌کند.

```python
# --- تعریف ساختار داده‌ها ---

# کلاس ساده برای نمایش یک سند با محتوا و پاراگراف‌های آن
class Document:
    def __init__(self, doc_id, content):
        self.doc_id = doc_id
        # تقسیم محتوا به پاراگراف‌ها بر اساس خطوط خالی
        self.paragraphs = [p.strip() for p in content.split('\n\n') if p.strip()]
        # ایجاد لیستی از کلمات در هر پاراگراف برای جستجوی سریع‌تر
        self.paragraph_words = [p.lower().split() for p in self.paragraphs]
    
    def __str__(self):
        return f"سند {self.doc_id}: {self.paragraphs[0][:50]}..."

# نمونه‌ای از مجموعه اسناد کوچک
documents = [
    Document(1, "مهمانان در مهمانی مست بودند. میزبان مسئولیت مواد الکلی را بر عهده داشت. مهمانان پس از ترک مهمانی تصادف کردند."),
    Document(2, "میزبان مهمانی دیگری برگزار کرد. در این مهمانی، مهمانان الکل مصرف نکردند. مسئولیت میزبان در این مورد کمتر بود."),
    Document(3, "قوانین مربوط به مسئولیت میزبان در بسیاری از کشورها وجود دارد. اگر مهمان مست باشد و حادثه‌ای رخ دهد، میزبان مسئول است."),
    Document(4, "مهمان و میزبان توافق‌نامه‌ای قبل از مهمانی امضا کردند. این توافق‌نامه مسئولیت‌ها را مشخص می‌کرد."),
    Document(5, "در این مهمانی، میزبان به مهمانان الکل نداد. با این حال، یکی از مهمانان مشروب خود را آورده بود و مست شد.")
]

# --- پیاده‌سازی الگوریتم‌های جستجو ---

def boolean_search(documents, query_terms):
    """
    جستجوی بولی ساده: اسنادی را برمی‌گرداند که تمام واژه‌های پرسمان را دارند.
    """
    results = []
    for doc in documents:
        # تبدیل کل سند به حروف کوچک برای جستجوی حساس‌به‌حروف بزرگ/کوچک نباشیم
        doc_text = ' '.join(doc.paragraphs).lower()
        if all(term.lower() in doc_text for term in query_terms):
            results.append(doc)
    return results

def proximity_search_paragraph(doc, term1, term2):
    """
    بررسی می‌کند که آیا دو واژه در یک پاراگراف از سند وجود دارند یا نه.
    این پیاده‌سازی برای عملگر /p (همان پاراگراف) است.
    """
    for words in doc.paragraph_words:
        has_term1 = any(term1.lower() in word for word in words)
        has_term2 = any(term2.lower() in word for word in words)
        if has_term1 and has_term2:
            return True
    return False

def extended_boolean_search(documents, term1, term2, operator):
    """
    جستجوی بولی گسترش‌یافته با عملگرهای نزدیکی.
    """
    results = []
    for doc in documents:
        if operator == "AND":
            # جستجوی بولی ساده
            doc_text = ' '.join(doc.paragraphs).lower()
            if term1.lower() in doc_text and term2.lower() in doc_text:
                results.append(doc)
        elif operator == "/p":
            # جستجوی نزدیکی در همان پاراگراف
            if proximity_search_paragraph(doc, term1, term2):
                results.append(doc)
    return results

def term_frequency(doc, term):
    """
    تعداد تکرار یک واژه در یک سند را می‌شمارد.
    """
    count = 0
    for paragraph in doc.paragraphs:
        words = paragraph.lower().split()
        count += sum(1 for word in words if term.lower() in word)
    return count

def ranked_retrieval(documents, query_terms):
    """
    بازیابی رتبه‌بندی‌شده: اسناد را بر اساس مجموع فراوانی واژه‌های پرسمان رتبه‌بندی می‌کند.
    """
    doc_scores = []
    for doc in documents:
        score = sum(term_frequency(doc, term) for term in query_terms)
        if score > 0:  # فقط اسنادی که حداقل یکی از واژه‌ها را دارند
            doc_scores.append((doc, score))
    
    # مرتب‌سازی اسناد بر اساس امتیاز (از بیشترین به کمترین)
    doc_scores.sort(key=lambda x: x[1], reverse=True)
    return doc_scores

# --- اجرای مثال‌ها ---

if __name__ == "__main__":
    
    # مثال ۱: مقایسه جستجوی بولی ساده و جستجوی نزدیکی
    print("="*60)
    print("مثال ۱: مقایسه جستجوی بولی ساده و جستجوی نزدیکی")
    term1 = "میزبان"
    term2 = "مهمان"
    
    # جستجوی بولی ساده (AND)
    boolean_results = boolean_search(documents, [term1, term2])
    print(f"\nنتایج جستجوی بولی ساده ({term1} AND {term2}):")
    for doc in boolean_results:
        print(f"  - {doc}")
    
    # جستجوی نزدیکی (/p)
    proximity_results = extended_boolean_search(documents, term1, term2, "/p")
    print(f"\nنتایج جستجوی نزدیکی ({term1} /p {term2}):")
    for doc in proximity_results:
        print(f"  - {doc}")
    
    # مثال ۲: مقایسه مدل بولی و مدل رتبه‌بندی‌شده
    print("\n" + "="*60)
    print("مثال ۲: مقایسه مدل بولی و مدل رتبه‌بندی‌شده")
    query_terms = ["میزبان", "مسئولیت"]
    
    # نتایج مدل بولی
    boolean_results = boolean_search(documents, query_terms)
    print(f"\nنتایج مدل بولی (بدون رتبه‌بندی):")
    for doc in boolean_results:
        print(f"  - {doc}")
    
    # نتایج مدل رتبه‌بندی‌شده
    ranked_results = ranked_retrieval(documents, query_terms)
    print(f"\nنتایج مدل رتبه‌بندی‌شده (با امتیاز):")
    for doc, score in ranked_results:
        print(f"  - امتیاز {score}: {doc}")
    
    # مثال ۳: نمایش فراوانی واژه در اسناد مختلف
    print("\n" + "="*60)
    print("مثال ۳: فراوانی واژه در اسناد مختلف")
    term = "میزبان"
    print(f"\nفراوانی واژه '{term}' در اسناد:")
    for doc in documents:
        freq = term_frequency(doc, term)
        print(f"  - {doc}: {freq} بار")
```

<li>مدل بولی را به‌عنوان سنگ بنای سیستم‌های جستجو معرفی کردیم. این مدل همان منطقی است که در کتابخانه‌ها با جستجوی «و» و «یا» استفاده می‌کنیم - مثلاً جستجوی «هوش مصنوعی و یادگیری ماشین» در پایگاه داده مقالات علمی.</li>
<li>ماتریس وقوع بولی و محدودیت‌های آن را بررسی کردیم. تصور کنید یک کتابخانه با هزاران کتاب دارید و می‌خواهید بدانید کدام کتاب‌ها کلمات کلیدی خاصی دارند - ماتریس وقوع این کار را انجام می‌دهد، اما برای کتابخانه‌های بزرگ بسیار ناکارآمد است!</li>
<li>فهرست معکوس را به‌عنوان راه‌حلی کارآمد برای ذخیره‌سازی و جست‌وجو معرفی کردیم. این همان فهرستی است که در انتهای کتاب‌ها می‌بینیم، اما برای جستجوی کامپیوتری بهینه شده است. گوگل در نسخه‌های اولیه خود از این روش استفاده می‌کرد.</li>
<li>الگوریتم‌های اشتراک (AND)، ادغام (OR) و بهینه‌سازی ترتیب پردازش را پیاده‌سازی کردیم. وقتی در اسنپ فود «پیتزا و نوشابه» را جستجو می‌کنید، در واقع از الگوریتم AND استفاده می‌کنید - فقط نتایجی نمایش داده می‌شوند که هر دو کلمه را داشته باشند.</li>
<li>تفاوت مدل بولی ساده، بولی گسترش‌یافته، و بازیابی رتبه‌بندی‌شده را بررسی کردیم. این همان تفاوتی است که بین جستجوی ساده در فایل‌های ویندوز و جستجوی پیشرفته در گوگل وجود دارد.</li>

<p style="font-size:16pt; text-align:right;">
<b>بیایید این مفاهیم را با چند مثال واقعی مقایسه کنیم:</b>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازیابی بولی ساده:</b> تصور کنید در دیجی‌کالا به دنبال «گوشی سامسونگ با حافظه ۲۵۶ گیگابایت و قیمت زیر ۱۰ میلیون تومان» می‌گردید. این جستجو دقیقاً از منطق بولی استفاده می‌کند: (گوشی AND سامسونگ) AND (حافظه ۲۵۶) AND (قیمت زیر ۱۰ میلیون). سیستم ابتدا تمام گوشی‌های سامسونگ را پیدا می‌کند، سپس از میان آن‌ها موارد با حافظه ۲۵۶ را جدا کرده و در نهایت آن‌هایی را نمایش می‌دهد که قیمت مناسب داشته باشند. نتیجه یا هست یا نیست؛ هیچ رتبه‌بندی وجود ندارد.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازیابی بولی گسترش‌یافته:</b> یک وکیل در پایگاه داده قوانین و آراء دادگاه به دنبال یک مورد خاص می‌گردد. او جستجوی زیر را انجام می‌دهد: <br>
«<code>مسئولیت قراردادی NEAR/10 خسارت</code>» <br>
در اینجا، `NEAR/10` یک عملگر گسترش‌یافته است که به سیستم می‌گوید فقط سندهایی را پیدا کن که عبارت «مسئولیت قراردادی» و «خسارت» در فاصله کمتر از ۱۰ کلمه از هم قرار دارند. این کار از پیدا کردن سندهایی که این دو کلمه را به صورت پراکنده در آن آورده‌اند، بسیار دقیق‌تر است.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازیابی رتبه‌بندی‌شده:</b> شما در گوگل عبارت «بهترین کافه‌های قصرالدشت» را جستجو می‌کنید. گوگل هزاران صفحه را پیدا می‌کند که این کلمات را دارند. اما به جای نمایش همه آن‌ها، آن‌ها را <b>رتبه‌بندی</b> می‌کند. نتیجه اول احتمالاً کافه‌ای است که امتیاز بالایی دارد، نظرات مثبت زیادی گرفته، به موقعیت مکانی شما نزدیک است و در وب‌سایت‌های معتبر به آن اشاره شده است. نتیجه پنجم ممکن است کافه خوبی باشد اما امتیاز کمتری دارد. این مدل، نتایج را بر اساس «مرتبطی» به شما نمایش می‌دهد، نه فقط بر اساس تطابق کلمات.
</p>

<li>خوشه‌بندی و طبقه‌بندی متن - روشی که گوگل اخبار را به دسته‌های ورزشی، سیاسی و اقتصادی تقسیم می‌کند</li>
</ul>
