<p style="font-size:18pt; text-align:center;">
<span id="section11.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
فصل ۱۱: بازیابی اطلاعات احتمالاتی
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در فصل‌های قبل، مدل‌های Boolean و فضای برداری برای بازیابی اطلاعات معرفی شدند. این مدل‌ها مانند یک کلیداند که فقط در صورت تطابق کامل قفل را باز می‌کنند، اما در دنیای واقعی، جستجو اغلب مبهم و ناقص است. در این فصل، رویکردی متفاوت و اصولی بر پایهٔ <strong>نظریهٔ احتمال</strong> ارائه می‌شود که به سیستم اجازه می‌دهد تحت <strong>عدم قطعیت</strong> تصمیم‌گیری کند.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">
    🔍 مثال واقعی: استفادهٔ آموزشی از IBM Watson در تشخیص پزشکی
</p>

<p style="font-size:16pt; text-align:right;">
    در بیمارستان‌های آموزشی، دانشجوهای پزشکی هنگام کار با بیماران واقعی از Watson به عنوان یک ابزار «تصمیم‌یار» استفاده می‌کردند. 
    هنگامی که دانشجو علائمی مثل تنگی نفس، سرفه خشک یا کاهش اکسیژن خون را وارد می‌کرد، Watson فقط کلمات را مطابقت نمی‌داد، بلکه با توجه به داده‌های میلیون‌ها پروندهٔ قبلی و مدل‌های آماری خود، احتمال هر بیماری را محاسبه می‌کرد. 
    برای مثال، اگر بیمار سرفه خشک، تنگی نفس و افت اکسیژن داشت، Watson احتمال حملهٔ آسم را بالاتر از آنفولانزا قرار می‌داد و گزینه‌هایی مثل COVID-19 یا ذات‌الریه را نیز با درصدی مشخص نمایش می‌داد. 
    این تحلیل احتمالاتی به دانشجو کمک می‌کرد بفهمد کدام تشخیص‌ها «محتمل‌تر» هستند و چه آزمایش‌های تکمیلی منطقی‌تر است.
</p>
</div>


<p style="font-size:16pt; text-align:right;">
در مدل احتمالاتی، اگر اسناد مرتبط و نامرتبطی در اختیار داشته باشیم، می‌توانیم احتمال ظاهر شدن یک واژه در سند مرتبط را تخمین بزنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
    $$
    P(t \mid R = 1)
    $$
</div>

<p style="font-size:16pt; text-align:right;">
این مقدار می‌تواند پایه‌ای برای ساخت یک <strong>طبقه‌بند</strong> باشد که تصمیم می‌گیرد آیا سندی مرتبط است یا نه. در این فصل، این رویکرد به‌صورت سیستماتیک معرفی می‌شود.
</p>

<hr>

<p style="font-size:18pt; font-weight:bold; color:#2c3e50; text-align:center;">
  مروری بر نظریه پایه‌ی احتمال
</p>

<p style="font-size:16pt; text-align:right;">
نظریه احتمال به ما کمک می‌کند تا در دنیای عدم قطعیت، تصمیمات منطقی بگیریم. متغیر تصادفی A نمایانگر یک رویداد است. احتمال وقوع آن بین ۰ تا ۱ است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
0 \leq P(A) \leq 1
$$
</div>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📧 مثال: فیلتر اسپم در جیمیل</p>
<p style="font-size:16pt; text-align:right;">
جیمیل از نظریه احتمال برای فیلتر کردن ایمیل‌های اسپم استفاده می‌کند. برای هر ایمیل، سیستم احتمال اینکه آن اسپم باشد را محاسبه می‌کند. اگر ایمیل حاوی کلماتی مانند «ارائه ویژه» یا «برنده» باشد، احتمال اسپم بودن آن افزایش می‌یابد. این احتمال با در نظر گرفتن ویژگی‌های دیگر مانند فرستنده و تاریخچه ایمیل دقیق‌تر می‌شود.</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای دو رویداد A و B، احتمال مشترک آن‌ها با قانون زنجیره‌ای (Chain Rule) تعریف می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(A, B) = P(A \cap B) = P(A \mid B) \cdot P(B) = P(B \mid A) \cdot P(A)
$$
</div>

<p style="font-size:16pt; text-align:right;">
قانون تقسیم (Partition Rule) بیان می‌کند که اگر B به زیرمجموعه‌های مجزا تقسیم شود، آنگاه:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(B) = P(A, B) + P(\neg A, B)
$$
</div>

<hr>

<p style="font-size:18pt; font-weight:bold; color:#2c3e50; text-align:center;">
  اصل رتبه‌بندی احتمالاتی
</p>

<p style="font-size:16pt; text-align:right;">
اصل رتبه‌بندی احتمالاتی (Probability Ranking Principle) بیان می‌کند که اسناد باید بر اساس احتمال مرتبط بودن رتبه‌بندی شوند. این اصل، پایهٔ بسیاری از مدل‌های بازیابی اطلاعات مدرن است.
</p>

<div dir="rtl" style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 مثال: الگوریتم PageRank گوگل</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم PageRank که گوگل در ابتدا از آن استفاده می‌کرد، بر اساس اصل رتبه‌بندی احتمالاتی عمل می‌کند. PageRank احتمال اینکه کاربر با کلیک روی یک لینک به صفحه‌ای خاص برسد را محاسبه می‌کند و صفحات را بر اساس این احتمال رتبه‌بندی می‌کند. صفحاتی که احتمال بازدید بیشتری دارند، در رتبه‌های بالاتر قرار می‌گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
<strong>قانون بیز (Bayes' Rule):</strong> برای معکوس کردن احتمال شرطی استفاده می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)} = \frac{P(B \mid A) \cdot P(A)}{\sum_{X \in \{A, \neg A\}} P(B \mid X) \cdot P(X)}
$$
</div>

<div dir="rtl" style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 مثال: سیستم توصیه نتفلیکس</p>
<p style="font-size:16pt; text-align:right;">
نتفلیکس از قانون بیز برای توصیه فیلم استفاده می‌کند. اگر کاربری فیلم‌های اکشن را دوست داشته باشد (احتمال اولیه)، نتفلیکس با توجه به اینکه کاربر اخیراً به فیلم‌های ابرقهرمانی علاقه نشان داده (شواهد)، احتمال پس از مشاهدهٔ شواهد را به‌روز می‌کند. این رویکرد به نتفلیکس اجازه می‌دهد توصیه‌های دقیق‌تری به کاربر ارائه دهد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این فرمول:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>P(A)</code>: احتمال اولیه (prior)</li>
<li><code>P(A|B)</code>: احتمال پس از مشاهدهٔ شواهد (posterior)</li>
<li><code>P(B|A)</code>: احتمال مشاهدهٔ شواهد در صورت صحت فرض</li>
</ul>
</p>

<hr>

<p style="font-size:18pt; font-weight:bold; color:#2c3e50; text-align:center;">  مدل استقلال دودویی
</p>

<p style="font-size:16pt; text-align:right;">
مدل استقلال دودویی (Binary Independence Model) یکی از اولین و تأثیرگذارترین مدل‌های بازیابی احتمالاتی است. این مدل فرض می‌کند که واژه‌ها در اسناد به صورت مستقل از هم ظاهر می‌شوند.
</p>

<div dir="rtl" style="background:#eef; padding:10px; border-radius:8px; text-align:right;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#2c3e50;">🔎 مثال: موتور جستجوی علمی PubMed</p>
<p style="font-size:16pt; text-align:right;">
PubMed که برای جستجوی مقالات علمی پزشکی استفاده می‌شود، از مدل استقلال دودویی برای رتبه‌بندی نتایج استفاده می‌کند. این مدل فرض می‌کند که ظاهر شدن واژه‌ها در یک مقاله، مستقل از ظاهر شدن واژه‌های دیگر است. اگرچه این فرض ساده‌شده است، اما در عمل نتایج قابل قبولی ارائه می‌دهد و پایهٔ بسیاری از سیستم‌های پیشرفته‌تر است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در فصل‌های بعدی، مدل‌های پیشرفته‌تری مانند Okapi BM25 و مدل‌های شبکه بیزی معرفی خواهند شد که این فرض ساده را بهبود می‌بخشند.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.2.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۲.۱ حالت 1/0 Loss در اصل رتبه‌بندی احتمالاتی (PRP)
</span>
</p>


<p style="font-size:16pt; text-align:right;">
در این بخش، فرض می‌کنیم که سیستم بازیابی اطلاعات در یک چارچوب رتبه‌دار عمل می‌کند (مطابق با فصل ۶.۳): کاربر یک پرسمان ارسال می‌کند و سیستم فهرستی مرتب‌شده از اسناد را بازمی‌گرداند. همچنین فرض می‌کنیم که مرتبط بودن به‌صورت دودویی تعریف شده است (مطابق فصل ۸): هر سند یا مرتبط است یا نیست.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم اولویت‌بندی ایمیل در Gmail</p>
<p style="font-size:16pt; text-align:right;">
در سرویس Gmail، بخش «Important» بر پایهٔ یک مدل احتمالاتی عمل می‌کند. سیستم برای هر ایمیل ورودی بررسی می‌کند که چقدر احتمال دارد این ایمیل برای کاربر مهم باشد: 
این احتمال از ترکیبی از داده‌هایی مثل فرستنده، سابقهٔ پاسخ‌گویی کاربر، کلمات کلیدی، و رفتارهای گذشته محاسبه می‌شود. سپس ایمیل‌ها بر اساس همین احتمال اهمیت رتبه‌بندی می‌شوند. 
در این مدل، هر ایمیل عملاً برچسبی دودویی دارد (مهم یا غیرمهم) و Gmail فهرست را طوری مرتب می‌کند که ایمیل‌هایی با احتمال «مهم بودن» بالاتر، در بالاترین رتبه‌ها نمایش داده شوند.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای هر پرسمان <code>q</code> و سند <code>d</code>، متغیر تصادفی <code>R<sub>d,q</sub></code> نشان می‌دهد که آیا سند <code>d</code> نسبت به پرسمان <code>q</code> مرتبط است یا نه:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
R_{d,q} =
\begin{cases}
1 & \text{if document } d \text{ is relevant to query } q \\
0 & \text{otherwise}
\end{cases}
$$
</div>

<p style="font-size:16pt; text-align:right;">
در این مدل، طبیعی‌ترین روش برای رتبه‌بندی اسناد، مرتب‌سازی آن‌ها بر اساس احتمال مرتبط بودن است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid d, q)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این ایده، پایهٔ <strong>اصل رتبه‌بندی احتمالاتی (Probability Ranking Principle)</strong> است که توسط <strong>van Rijsbergen (1979)</strong> معرفی شد:
</p>

<div style="font-size:16pt; text-align:right; font-style:italic; background:#eef; border:1px solid #ccd; padding:15px; border-radius:8px;">
اگر یک سیستم بازیابی اطلاعات، در پاسخ به هر پرسمان، اسناد را به ترتیب احتمال نزولی مرتبط بودن با نیاز اطلاعاتی کاربر مرتب کند — به‌طوری‌که این احتمال‌ها تا حد ممکن دقیق تخمین زده شده باشند — آنگاه کارایی کلی سیستم برای کاربر، در مقایسه با هر روش دیگری که فقط از همان داده‌ها استفاده کند، بیشینه خواهد بود.
</div>

<hr>

<p style="font-size:16pt; text-align:right;">
در ساده‌ترین حالت، هیچ هزینه‌ای برای بازیابی یا سودمندی خاصی در نظر گرفته نمی‌شود. فقط دو نوع خطا وجود دارد:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>بازگرداندن سند نامرتبط</li>
<li>عدم بازگرداندن سند مرتبط</li>
</ul>

<p style="font-size:16pt; text-align:right;">
این وضعیت به‌عنوان <strong>1/0 Loss</strong> شناخته می‌شود: هر خطا یک واحد هزینه دارد. هدف، بازگرداندن بهترین اسناد ممکن در میان <code>k</code> سند اول است — برای هر مقداری از <code>k</code> که کاربر بررسی می‌کند.
</p>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم اولویت‌بندی اخبار در اپلیکیشن‌های خبری</p>
<p style="font-size:16pt; text-align:right;">
بسیاری از اپلیکیشن‌های خبری (مانند Google News یا Apple News) از مدل‌های احتمالاتی برای رتبه‌بندی اخبار استفاده می‌کنند. برای هر خبر، احتمال اینکه کاربر به آن علاقه‌مند باشد محاسبه می‌شود. این احتمال بر اساس سابقهٔ مطالعهٔ کاربر، موضوعات مورد علاقه، و اخبار مشابهی که قبلاً خوانده، تعیین می‌گردد. سپس اخبار با احتمال بالاتر در بالای فید کاربر نمایش داده می‌شوند. این سیستم بر اساس 1/0 Loss عمل می‌کند: نشان دادن یک خبر نامرتبط در بالا (False Positive) ممکن است باعث از دست رفتن علاقه کاربر شود، اما عدم نمایش یک خبر بسیار مهم (False Negative) هزینه بیشتری دارد. هدف، کمینه کردن این خطاها و ارائهٔ بهترین خبرها در ابتدای فهرست است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این حالت، PRP می‌گوید که باید اسناد را به ترتیب نزولی <code>P(R = 1 | d, q)</code> مرتب کنیم.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
اگر به‌جای رتبه‌بندی، بخواهیم مجموعه‌ای از اسناد را بازگردانیم، <strong>قانون تصمیم بهینهٔ بیزی (Bayes Optimal Decision Rule)</strong> می‌گوید:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\text{Doc } d \text{ It is relevant if and only if } P(R = 1 \mid d, q) > P(R = 0 \mid d, q)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این قانون تصمیم، <strong>ریسک مورد انتظار</strong> را در حالت 1/0 Loss کمینه می‌کند.
</p>

<div dir="rtl" style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 مثال: سیستم تشخیص تقلب در تراکنش‌های بانکی</p>
<p style="font-size:16pt; text-align:right;">
بانک‌ها برای تشخیص تراکنش‌های جعلی از مدل‌های احتمالاتی پیچیده استفاده می‌کنند. سیستم برای هر تراکنش، احتمال اینکه آن تراکنش «تقلب‌آمیز» باشد را محاسبه می‌کند. اگر این احتمال از یک آستانهٔ مشخص (که بر اساس تحلیل هزینهٔ خطاها تعیین شده) بالاتر باشد، تراکنش مسدود شده یا برای تأیید بیشتر علامت‌گذاری می‌شود. این استفاده مستقیم از قانون تصمیم بیزی (Bayes Optimal Decision Rule) است؛ جایی که هزینهٔ رد کردن یک تراکنش صحیح (False Positive) و هزینهٔ پذیرش یک تراکنش جعلی (False Negative) به‌دقت محاسبه و در تصمیم‌گیری لحاظ می‌شود تا ریسک کلی به حداقل برسد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
قضیهٔ ۱۱.۱: اصل رتبه‌بندی احتمالاتی (PRP) در حالت 1/0 Loss بهینه است، به این معنا که <strong>ریسک مورد انتظار</strong> را کمینه می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
اثبات این قضیه در Ripley (1996) آمده است، اما فرض می‌کند که همهٔ احتمال‌ها به‌درستی شناخته شده‌اند — که در عمل هرگز صادق نیست. با این حال، PRP همچنان پایه‌ای بسیار مفید برای طراحی مدل‌های بازیابی اطلاعات محسوب می‌شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p id='section11.2.2' style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۲.۲ اصل رتبه‌بندی احتمالاتی با هزینه‌های بازیابی
</p>

<p style="font-size:16pt; text-align:right;">
در حالت سادهٔ PRP، فرض می‌شود که همهٔ خطاها هزینهٔ یکسانی دارند (1/0 Loss). اما در بسیاری از کاربردهای واقعی، هزینهٔ بازگرداندن سند نامرتبط با هزینهٔ عدم بازگرداندن سند مرتبط یکسان نیست. در این بخش، مدل رتبه‌بندی با در نظر گرفتن <strong>هزینه‌های بازیابی</strong> معرفی می‌شود.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم پیشنهاد محصول در فروشگاه‌های آنلاین</p>
<p style="font-size:16pt; text-align:right;">
در فروشگاه‌های آنلاین مانند دیجی‌کالا، نمایش یک محصول نامرتبط به کاربر (False Positive) ممکن است باعث اتلاف وقت و کاهش احتمال خرید شود، اما عدم نمایش یک محصول بسیار مرتبط که کاربر به دنبال آن است (False Negative) می‌تواند منجر به از دست دادن یک فروش قطعی شود. بنابراین، سیستم‌های پیشنهاد محصول طوری طراحی می‌شوند که هزینهٔ خطای False Negative (از دست دادن فروش) را بسیار بیشتر از False Positive در نظر بگیرند تا شانس موفقیت در فروش را به حداکثر برسانند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
فرض کنیم:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>C₁</code>: هزینهٔ عدم بازیابی یک سند مرتبط (false negative)</li>
<li><code>C₀</code>: هزینهٔ بازیابی یک سند نامرتبط (false positive)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
در این مدل، PRP می‌گوید سندی را باید انتخاب کنیم که <strong>ریسک مورد انتظار</strong> آن کمینه باشد. برای سند <code>d</code>، اگر برای همهٔ سندهای <code>d′</code> که هنوز بازیابی نشده‌اند، رابطهٔ زیر برقرار باشد، آنگاه <code>d</code> باید سند بعدی باشد که بازیابی می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
C_0 \cdot P(R = 0 \mid d) - C_1 \cdot P(R = 1 \mid d) \leq C_0 \cdot P(R = 0 \mid d') - C_1 \cdot P(R = 1 \mid d')
$$
</div>

<p style="font-size:16pt; text-align:right;">
این مدل یک چارچوب رسمی برای مدل‌سازی هزینه‌های متفاوت خطاها فراهم می‌کند — برخلاف فصل ۸.۶ که فقط در مرحلهٔ ارزیابی به این موضوع پرداختیم، اینجا در <strong>مرحلهٔ مدل‌سازی</strong> لحاظ می‌شود.
</p>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎯 مثال: سیستم اولویت‌بندی وظایف در نرم‌افزارهای مدیریت پروژه</p>
<p style="font-size:16pt; text-align:right;">
در نرم‌افزارهای مدیریت پروژه مانند Asana یا Trello، گاهی لازم است بین وظایف مختلف اولویت‌بندی صورت گیرد. عدم پیگیری یک وظیفهٔ حیاتی (False Negative) می‌تواند منجر به تأخیر در پروژه و هزینه‌های هنگفتی شود، در حالی که صرف زمان اندکی برای بررسی یک وظیفهٔ کم‌اهمیت (False Positive) ممکن است هزینهٔ کمتری داشته باشد. بنابراین، سیستم اولویت‌بندی، وظایف را بر اساس ترکیبی از فوریت و اهمیت، و با در نظر گرفتن هزینهٔ بالاتر خطاهای False Negative، مرتب می‌کند تا اطمینان حاصل شود که مهم‌ترین کارها در اولویت اجرا قرار گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
با این حال، در ادامهٔ فصل، مدل‌های سود/زیان پیچیده‌تر بررسی نمی‌شوند. اما این بخش نشان می‌دهد که PRP می‌تواند به‌صورت تعمیم‌یافته برای شرایط واقعی‌تر نیز استفاده شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span id="section11.3" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳ مدل دودویی استقلال (Binary Independence Model)
</span>
</p>

<p style="font-size:16pt; text-align:right;">
مدل دودویی استقلال (BIM) یکی از مدل‌های کلاسیک بازیابی اطلاعات است که با اصل رتبه‌بندی احتمالاتی (PRP) ترکیب می‌شود. این مدل با فرض‌های ساده‌ای طراحی شده که محاسبهٔ احتمال مرتبط بودن سند را عملی و قابل پیاده‌سازی می‌کند.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم پیشنهاد فیلم در سرویس‌های استریم</p>
<p style="font-size:16pt; text-align:right;">
سرویس‌های استریم مانند نتفلیکس یا فیلیمو برای پیشنهاد فیلم و سریال به کاربران از مدل‌های مشابه دودویی استقلال استفاده می‌کنند. این سیستم‌ها با تحلیل حضور یا عدم حضور کلمات کلیدی در توضیحات فیلم، ژانرها، بازیگران و کارگردانان، احتمال مرتبط بودن یک فیلم با علایق کاربر را تخمین می‌زنند. به عنوان مثال، اگر کاربر به دنبال فیلم‌های “علمی-تخیلی” و “اکشن” باشد، سیستم فیلم‌هایی را که هر دو کلمه کلیدی را در توضیحات خود دارند، در اولویت نمایش قرار می‌دهد. این مدل، اگرچه فرض استقلال بین کلمات را دارد، اما در عمل برای ارائهٔ پیشنهادهای اولیه بسیار کارآمد است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این مدل:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>نمایش سند و پرسمان به‌صورت بردار دودویی از واژه‌ها است: اگر واژهٔ <code>t</code> در سند <code>d</code> وجود داشته باشد، <code>x<sub>t</sub> = 1</code> و در غیر این‌صورت <code>x<sub>t</sub> = 0</code>.</li>
<li>نمایش پرسمان نیز به‌صورت بردار وقوع واژه‌ها است: <code>𝑞⃗</code></li>
<li>فرض استقلال: وقوع واژه‌ها در سندها مستقل از یکدیگر فرض می‌شود (مانند فرض ساده‌انگارانهٔ Naive Bayes).</li>
</ul>

<p style="font-size:16pt; text-align:right;">
این فرض‌ها در عمل نادرست‌اند (مثلاً واژه‌ها هم‌وابسته‌اند یا اسناد تکراری وجود دارند)، اما در بسیاری از کاربردها نتایج قابل قبولی تولید می‌کنند.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
هدف مدل، تخمین احتمال مرتبط بودن سند <code>d</code> با پرسمان <code>q</code> است، یعنی:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid \vec{x}, \vec{q})
$$
</div>

<p style="font-size:16pt; text-align:right;">
با استفاده از قانون بیز، این مقدار به‌صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid \vec{x}, \vec{q}) = \frac{P(\vec{x} \mid R = 1, \vec{q}) \cdot P(R = 1 \mid \vec{q})}{P(\vec{x} \mid \vec{q})}
$$

$$
P(R = 0 \mid \vec{x}, \vec{q}) = \frac{P(\vec{x} \mid R = 0, \vec{q}) \cdot P(R = 0 \mid \vec{q})}{P(\vec{x} \mid \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
در این فرمول‌ها:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>P(𝑥⃗ | R = 1, 𝑞⃗)</code>: احتمال مشاهدهٔ بردار سند در میان اسناد مرتبط</li>
<li><code>P(𝑥⃗ | R = 0, 𝑞⃗)</code>: احتمال مشاهدهٔ بردار سند در میان اسناد نامرتبط</li>
<li><code>P(R = 1 | 𝑞⃗)</code>: احتمال اولیهٔ مرتبط بودن سند با پرسمان</li>
<li><code>P(R = 0 | 𝑞⃗)</code>: احتمال اولیهٔ نامرتبط بودن سند با پرسمان</li>
</ul>

<p style="font-size:16pt; text-align:right;">
چون سند یا مرتبط است یا نیست، باید داشته باشیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid \vec{x}, \vec{q}) + P(R = 0 \mid \vec{x}, \vec{q}) = 1
$$
</div>

<hr>

<p style="font-size:16pt; text-align:right;">
در عمل، هیچ‌کدام از این احتمال‌ها به‌صورت دقیق قابل محاسبه نیستند. بنابراین، از <strong>آمار مجموعهٔ اسناد</strong> برای تخمین آن‌ها استفاده می‌شود. این شامل:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فراوانی واژه در سند (term frequency)</li>
<li>فراوانی واژه در مجموعه (document frequency)</li>
<li>طول سند</li>
<li>و سایر ویژگی‌های قابل محاسبه</li>
</ul>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم جستجو در فروشگاه‌های آنلاین محصولات تخصصی</p>
<p style="font-size:16pt; text-align:right;">
فروشگاه‌های آنلاینی که محصولات تخصصی مانند قطعات الکترونیکی یا ابزارهای صنعتی را عرضه می‌کنند، اغلب از مدل دودویی استقلال برای سیستم جستجوی خود بهره می‌برند. در این سیستم‌ها، حضور یا عدم حضور واژه‌های کلیدی در نام محصول، توضیحات فنی، یا مشخصات آن، مبنای رتبه‌بندی نتایج قرار می‌گیرد. به عنوان مثال، اگر کاربری به دنبال “پردازنده Core i7 با فرکانس 3.5 گیگاهرتز” باشد، سیستم به حضور دقیق این عبارات در توضیحات محصول اولویت می‌دهد. اگرچه ممکن است ترکیب این واژه‌ها در محصولات مختلف مستقل نباشد (مثلاً “Core i7” و “گیگاهرتز” اغلب با هم می‌آیند)، اما این مدل به دلیل سادگی پیاده‌سازی و کارایی قابل قبول، همچنان در بسیاری از این فروشگاه‌ها استفاده می‌شود</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، اسناد بر اساس احتمال تخمینی مرتبط بودن مرتب می‌شوند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۱ استخراج تابع رتبه‌بندی برای واژه‌های پرسمان
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در مدل دودویی استقلال، هدف رتبه‌بندی اسناد بر اساس احتمال مرتبط بودن آن‌ها با پرسمان است. اما چون هدف فقط رتبه‌بندی است (نه مقدار دقیق احتمال)، می‌توان از <strong>نسبت شانس (odds)</strong> استفاده کرد که با احتمال مرتبط بودن هم‌مرتبه (monotonic) است.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم جستجوی علمی PubMed</p>
<p style="font-size:16pt; text-align:right;">
PubMed برای رتبه‌بندی مقالات علمی از یک مدل احتمالاتی استفاده می‌کند. وقتی شما عبارت "CRISPR-Cas9 و قلب عروقی" را جستجو می‌کنید، سیستم احتمال وقوع هر واژه در مقالات مرتبط را محاسبه می‌کند. سپس مقالات را بر اساس نسبت شانس (odds) رتبه‌بندی می‌کند. واژه‌هایی که در مقالات مرتبط شانس بالاتری برای وقوع دارند، در رتبه‌های بالاتر قرار می‌گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در مدل BIM، نسبت شانس وقوع واژه در سند مرتبط به نامرتبط به‌صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = \frac{P(R = 1 \mid \vec{x}, \vec{q})}{P(R = 0 \mid \vec{x}, \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
با استفاده از قانون بیز و حذف مخرج مشترک، داریم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = \frac{P(R = 1 \mid \vec{x}, \vec{q})}{P(R = 0 \mid \vec{x}, \vec{q})} \cdot \frac{P(\vec{x} \mid R = 1, \vec{q})}{P(\vec{x} \mid R = 0, \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
عبارت اول برای هر پرسمان ثابت است و در رتبه‌بندی نهایی تأثیری ندارد. بنابراین، فقط باید عبارت دوم را تخمین بزنیم. برای این کار، فرض استقلال شرطی واژه‌ها (Naive Bayes) را اعمال می‌کنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\frac{P(\vec{x} \mid R = 1, \vec{q})}{P(\vec{x} \mid R = 0, \vec{q})} = \prod_{t=1}^{M} \frac{P(x_t \mid R = 1, \vec{q})}{P(x_t \mid R = 0, \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
چون <code>x<sub>t</sub></code> فقط ۰ یا ۱ است، می‌توان حاصل‌ضرب را به دو بخش تقسیم کرد:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = O(R \mid \vec{q}) \cdot \prod_{t: x_t = 1} \frac{p_t}{u_t} \cdot \prod_{t: x_t = 0} \frac{1 - p_t}{1 - u_t}
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>p<sub>t</sub> = P(x<sub>t</sub> = 1 | R = 1, q)</code>: احتمال وقوع واژه در سند مرتبط</li>
<li><code>u<sub>t</sub> = P(x<sub>t</sub> = 1 | R = 0, q)</code>: احتمال وقوع واژه در سند نامرتبط</li>
</ul>
</p>

<p style="font-size:16pt; text-align:right;">
با فرض ساده‌ساز دیگر: واژه‌هایی که در پرسمان نیستند، احتمال وقوعشان در اسناد مرتبط و نامرتبط برابر است (یعنی <code>p<sub>t</sub> = u<sub>t</sub></code> برای <code>q<sub>t</sub> = 0</code>). بنابراین فقط واژه‌های موجود در پرسمان در رتبه‌بندی مؤثرند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = O(R \mid \vec{q}) \cdot \prod_{t: x_t = q_t = 1} \frac{p_t}{u_t} \cdot \prod_{t: x_t = 0, q_t = 1} \frac{1 - p_t}{1 - u_t}
$$
</div>

<p style="font-size:16pt; text-align:right;">
با بازنویسی جبری، حاصل‌ضرب دوم را به حاصل‌ضرب اول منتقل می‌کنیم و به فرم نهایی می‌رسیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = \text{const} \cdot \prod_{t: x_t = q_t = 1} \frac{p_t (1 - u_t)}{u_t (1 - p_t)}
$$
</div>

<p style="font-size:16pt; text-align:right;">
چون تابع لگاریتم یکنواخت است، می‌توان به‌جای حاصل‌ضرب، مجموع لگاریتم‌ها را استفاده کرد. این مقدار به‌عنوان <strong>Retrieval Status Value (RSV)</strong> شناخته می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t: x_t = q_t = 1} \log \left( \frac{p_t (1 - u_t)}{u_t (1 - p_t)} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
برای هر واژهٔ پرسمان، وزن <code>c<sub>t</sub></code> تعریف می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t = \log \left( \frac{p_t}{1 - p_t} \right) + \log \left( \frac{1 - u_t}{u_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این وزن‌ها نشان می‌دهند که واژه چقدر بیشتر در اسناد مرتبط نسبت به نامرتبط ظاهر می‌شود. اگر <code>p<sub>t</sub> = u<sub>t</sub></code> باشد، آنگاه <code>c<sub>t</sub> = 0</code> و واژه در رتبه‌بندی تأثیری ندارد.
</p>

<p style="font-size:16pt; text-align:right;">
در عمل، برای هر سند، مجموع وزن‌های واژه‌های پرسمان موجود در آن محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t: x_t = q_t = 1} c_t
$$
</div>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم جستجوی حقوقی Westlaw</p>
<p style="font-size:16pt; text-align:right;">
Westlaw که برای جستجوی قوانین حقوقی استفاده می‌شود، از یک مدل احتمالاتی برای رتبه‌بندی نتایج استفاده می‌کند. وقتی یک وکیل به دنبال "قانون حفظ حریم خصوصی در قراردادها" می‌گردد، سیستم احتمال وقوع هر واژه در اسناد حقوقی مرتبط را محاسبه می‌کند. سپس اسناد را بر اساس RSV رتبه‌بندی می‌کند. این رویکرد باعث می‌شود اسنادی که واژه‌های کلیدی بیشتری دارند، در رتبه‌های بالاتر قرار گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، اسناد بر اساس RSV رتبه‌بندی می‌شوند که یک رویکرد عملی برای رتبه‌بندی در مدل دودویی استقلال است.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.2" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۲ تخمین‌های احتمالی در نظریه
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در این بخش، هدف محاسبهٔ وزن <code>c<sub>t</sub></code> برای هر واژه <code>t</code> در مجموعهٔ اسناد است. این وزن‌ها نشان می‌دهند که واژه چقدر بیشتر در اسناد مرتبط نسبت به نامرتبط ظاهر می‌شود. برای این کار، از جدول شمارش زیر استفاده می‌کنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
<table style="margin:auto; border-collapse:collapse; font-size:14pt;">
<tr><th style="border:1px solid #ccc; padding:6px;">Term</th><th style="border:1px solid #ccc; padding:6px;">Relevant</th><th style="border:1px solid #ccc; padding:6px;">Nonrelevant</th><th style="border:1px solid #ccc; padding:6px;">Total</th></tr>
<tr><td style="border:1px solid #ccc; padding:6px;">Present (xₜ = 1)</td><td>s</td><td>dft − s</td><td>dft</td></tr>
<tr><td style="border:1px solid #ccc; padding:6px;">Absent (xₜ = 0)</td><td>S − s</td><td>(N − dft) − (S − s)</td><td>N − dft</td></tr>
<tr><td style="border:1px solid #ccc; padding:6px;">Total</td><td>S</td><td>N − S</td><td>N</td></tr>
</table>
</div>

<p style="font-size:16pt; text-align:right;">
در این جدول:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>N</code>: تعداد کل اسناد</li>
<li><code>S</code>: تعداد اسناد مرتبط</li>
<li><code>dft</code>: تعداد اسنادی که واژهٔ <code>t</code> را دارند</li>
<li><code>s</code>: تعداد اسناد مرتبطی که واژهٔ <code>t</code> را دارند</li>
</ul>
</p>

<p style="font-size:16pt; text-align:right;">
با استفاده از این جدول، احتمال‌های شرطی به‌صورت زیر تعریف می‌شوند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t = \frac{s}{S}, \quad u_t = \frac{dft - s}{N - S}
$$
</div>

<p style="font-size:16pt; text-align:right;">
و وزن <code>c<sub>t</sub></code> به‌صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t = \log \left( \frac{s / (S - s)}{(dft - s) / ((N - dft) - (S - s))} \right)
$$
</div>

<hr>

<p style="font-size:16pt; text-align:right;">
اما این فرمول ممکن است در صورت صفر بودن صورت یا مخرج دچار شکست شود. برای جلوگیری از این مشکل، از <strong>نرم‌سازی (smoothing)</strong> استفاده می‌کنیم — به‌ویژه با افزودن <code>1/2</code> به شمارش‌ها:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\hat{c}_t = \log \left( \frac{(s + 1/2) / (S - s + 1/2)}{(dft - s + 1/2) / (N - dft - S + 1/2)} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این روش نرم‌سازی، یک فرم ساده از <strong>تخمین بیشینهٔ پسین (MAP)</strong> است — که در آن، فرض اولیهٔ یکنواخت ضعیف داریم و سپس بر اساس داده‌های مشاهده‌شده، احتمال‌ها را به‌روزرسانی می‌کنیم.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
در تخمین‌های احتمالی برای داده‌های دسته‌ای (مثل حضور یا غیاب واژه)، رایج‌ترین روش تخمین، <strong>فراوانی نسبی (Relative Frequency)</strong> است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(\text{event}) = \frac{\text{occured number}}{\text{number of all experiment}}
$$
</div>

<p style="font-size:16pt; text-align:right;">
این مقدار، <strong>تخمین بیشینهٔ درست‌نمایی (MLE)</strong> است — یعنی مقداری که داده‌های مشاهده‌شده را بیشینه می‌کند. اما اگر فقط از MLE استفاده کنیم:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>احتمال واژه‌های دیده‌شده بیش از حد بالا تخمین زده می‌شود</li>
<li>واژه‌های دیده‌نشده احتمال صفر می‌گیرند، که مدل را می‌شکند</li>
</ul>

<p style="font-size:16pt; text-align:right;">
برای حل این مشکل، از <strong>نرم‌سازی با شبه‌شمارش (pseudocounts)</strong> استفاده می‌کنیم — مثلاً افزودن <code>α = 1/2</code> به هر شمارش. این معادل با فرض اولیهٔ بیزی یکنواخت است.
</p>

<p style="font-size:16pt; text-align:right;">
در فصل ۱۲.۲.۲ روش‌های پیشرفته‌تر نرم‌سازی بررسی خواهند شد، اما در حال حاضر، افزودن <code>1/2</code> کفایت می‌کند.
</p>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم جستجوی علمی PubMed</p>
<p style="font-size:16pt; text-align:right;">
PubMed برای رتبه‌بندی مقالات علمی از یک مدل احتمالاتی استفاده می‌کند. وقتی شما عبارت "CRISPR-Cas9 و قلب عروقی" را جستجو می‌کنید، سیستم احتمال وقوع هر واژه در مقالات مرتبط را محاسبه می‌کند. برای تخمین این احتمال‌ها، از فراوانی نسبی واژه‌ها در کل مجموعه مقالات استفاده می‌کند. این رویکرد باعث می‌شود واژه‌های نادر مانند "CRISPR" که در مقالات نادر رایج نیستند، احتمال پایینی بگیرند و در رتبه‌بندی تأثیر منفی داشته باشند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، این روش‌های تخمین احتمال، پایهٔ مدل‌های بازیابی احتمالاتی مانند BIM را تشکیل می‌دهند و به سیستم اجازه می‌دهند تا با عدم قطعیت، تصمیمات منطقی بگیرد.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.3" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۳ تخمین‌های احتمالی در عمل
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در عمل، معمولاً فرض می‌شود که اسناد مرتبط درصد بسیار کوچکی از کل مجموعه هستند. بنابراین، می‌توان آمار اسناد نامرتبط را با آمار کل مجموعه تقریب زد. تحت این فرض، احتمال وقوع واژه <code>t</code> در اسناد نامرتبط (یعنی <code>u<sub>t</sub></code>) به‌صورت زیر تخمین زده می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
u_t = \frac{df_t}{N}
$$
</div>

<p style="font-size:16pt; text-align:right;">
و در نتیجه، عبارت لگاریتمی زیر که در وزن‌دهی <code>RSV</code> ظاهر می‌شود، به فرم ساده‌ای تبدیل می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\log \left( \frac{1 - u_t}{u_t} \right) = \log \left( \frac{N - df_t}{df_t} \right) \approx \log \left( \frac{N}{df_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این تقریب، توجیه نظری برای وزن‌دهی <strong>IDF</strong> است که در فصل ۶.۲.۱ معرفی شد.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
اما تخمین <code>p<sub>t</sub></code> (احتمال وقوع واژه در اسناد مرتبط) به‌سادگی قابل تقریب نیست. سه روش برای تخمین آن وجود دارد:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>استفاده از اسناد مرتبط شناخته‌شده:</strong> اگر تعدادی سند مرتبط داشته باشیم، می‌توانیم <code>p<sub>t</sub></code> را مستقیماً از آن‌ها تخمین بزنیم. این پایهٔ روش‌های بازخورد مرتبط بودن (relevance feedback) است که در بخش بعدی بررسی می‌شود.</li>

<li><strong>مدل Croft & Harper (1979):</strong> فرض می‌شود که <code>p<sub>t</sub> = 0.5</code> برای همهٔ واژه‌های پرسمان. در این حالت، چون <code>p<sub>t</sub> = 1 - p<sub>t</sub></code>، عوامل مربوط به <code>p<sub>t</sub></code> در فرمول <code>RSV</code> حذف می‌شوند و فقط <code>IDF</code> باقی می‌ماند. این روش برای اسناد کوتاه (مثل عنوان یا چکیده) و جست‌وجوی غیرتکراری قابل قبول است.</li>

<li><strong>مدل Greiff (1998):</strong> نشان داد که <code>p<sub>t</sub></code> در عمل با <code>df<sub>t</sub></code> افزایش می‌یابد. پیشنهاد او برای تخمین <code>p<sub>t</sub></code> به‌صورت زیر است:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t = \frac{1}{3} + \frac{2}{3} \cdot \frac{df_t}{N}
$$
</div>
</li>
</ol>

<hr>

<p style="font-size:16pt; text-align:right;">
روش‌های ترکیبی و تکراری برای تخمین <code>p<sub>t</sub></code> در بخش بعدی بررسی خواهند شد. اما تا اینجا، می‌توان گفت که در غیاب داده‌های مرتبط، استفاده از <code>IDF</code> به‌عنوان تقریب برای وزن‌دهی واژه‌ها در مدل BIM قابل توجیه است.
</p>

<p style="font-size:16pt; text-align:right;">
در نهایت، این روش‌های تخمین احتمال، پایهٔ مدل‌های بازیابی احتمالاتی مانند BIM را تشکیل می‌دهند و به سیستم اجازه می‌دهند تا با عدم قطعیت، تصمیمات منطقی بگیرد.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.4" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۴ رویکردهای احتمالاتی به بازخورد مرتبط بودن
</span>
</p>

<p style="font-size:16pt; text-align:right;">
برای تخمین دقیق‌تر <code>p<sub>t</sub></code>، می‌توان از بازخورد مرتبط بودن (واقعی یا شبه) استفاده کرد. الگوریتم بازخورد احتمالاتی به‌صورت زیر عمل می‌کند:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>تخمین اولیهٔ</strong> <code>p<sub>t</sub></code> و <code>u<sub>t</sub></code> با فرض‌های ساده (مثلاً <code>p<sub>t</sub> = 0.5</code>)</li>
<li><strong>بازیابی اسناد</strong> بر اساس تخمین‌های فعلی و ارائهٔ آن‌ها به کاربر</li>
<li><strong>دریافت قضاوت‌های مرتبط بودن</strong> از کاربر برای مجموعه‌ای از اسناد <code>V</code></li>
<li><strong>به‌روزرسانی تخمین‌ها</strong> با استفاده از داده‌های مرتبط</li>
<li><strong>تکرار مراحل</strong> تا زمانی که رتبه‌بندی به حالت پایدار برسد</li>
</ol>

<hr>

<p style="font-size:16pt; text-align:right;">
در تخمین اولیهٔ، فرض می‌کنیم که کاربر تقریباً ۵٪ از اسناد را مرتبط تشخیص می‌دهد (<code>κ = 0.5</code>). در این حالت، تخمین اولیهٔ برای <code>p<sub>t</sub></code> و <code>u<sub>t</sub></code> به‌صورت زیر خواهد بود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t^{(0)} = 0.5, \quad u_t^{(0)} = 0.5
$$
</div>

<p style="font-size:16pt; text-align:right;">
در مرحله بازیابی، مجموعه‌ای از اسناد با امتیاز بالا (مثلاً ۱۰ سند برتر) انتخاب می‌شود. سپس این مجموعه به کاربر ارائه می‌شود و کاربر قضاوت خود را برای برخی از این اسناد ثبت می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
پس از دریافت قضاوت‌های مرتبط، تخمین‌های جدید به‌صورت زیر محاسبه می‌شوند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t^{(k+1)} = \frac{|V_R^t| + \kappa p_t^{(k)}}{|V_R| + \kappa}
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>|V<sub>R</sub><sup>t</sup>|</code>: تعداد اسناد مرتبطی که واژه <code>t</code> را دارند</li>
<li><code>|V<sub>R</sub>|</code>: تعداد کل اسناد مرتبطی که قضاوت برای آن‌ها دریافت شده</li>
<li><code>κ</code>: ضریبی که نشان می‌دهد چقدر به قضاوت‌های کاربر اعتماد داریم (مثلاً ۰.۵)</li>
</ul>
</p>

<p style="font-size:16pt; text-align:right;">
این فرآیند، نمونه‌ای از <strong>تخمین بیشینهٔ پسین (MAP)</strong> است که در آن، فرض اولیه ضعیف با داده‌های مشاهده‌شده ترکیب می‌شود.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
در نسخهٔ شبه‌بازخورد (pseudo-relevance feedback)، فرض می‌کنیم که <code>V = V<sub>R</sub></code> و هیچ قضاوت انسانی نداریم. در این حالت، مراحل به‌صورت زیر ساده می‌شوند:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فرض اولیهٔ <code>p<sub>t</sub></code> و <code>u<sub>t</sub></code></li>
<li>انتخاب مجموعهٔ <code>V</code> از اسناد با امتیاز بالا</li>
<li>تخمین <code>p<sub>t</sub></code> با نرم‌سازی:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t = \frac{|V^t| + \frac{1}{2}}{|V| + 1}
$$
</div>
</li>
<li>تخمین <code>u<sub>t</sub></code> با فرض نامرتبط بودن اسناد خارج از <code>V</code>:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
u_t = \frac{df_t - |V^t| + \frac{1}{2}}{N - |V| + 1}
$$
</div>
</li>
<li>تکرار مراحل تا همگرایی رتبه‌بندی</li>
</ul>

<p style="font-size:16pt; text-align:right;">
در نهایت، وزن‌ها به‌صورت زیر محاسبه می‌شوند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t = \log \left( \frac{p_t}{1 - p_t} \cdot \frac{1 - u_t}{u_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
و فرمول tf-idf تقریباً به فرمول BIM تبدیل می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t \approx \log \left( \frac{|V^t| + \frac{1}{2}}{|V| + 1} \right) + \log \left( \frac{N}{df_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
یعنی دو مؤلفهٔ لگاریتمی داریم: یکی از نسبت اسناد مرتبطی که واژه را دارند، و دیگری از <code>IDF</code>. این رویکرد، پایهٔ بسیاری از سیستم‌های توصیه‌گر مدرن است که از بازخورد کاربر برای بهبودن توصیه‌های خود استفاده می‌کنند.</p>

</div>

```python
import math

import numpy as np
import pandas as pd

# ===================================================================
# تنظیمات اولیه و داده‌های نمونه
# ===================================================================

# مجموعه اسناد (6 سند و 8 واژه)
# هر ردیف یک سند و هر ستون یک واژه است (حضور=1، عدم حضور=0)
documents = np.array(
    [
        # t1 t2 t3 t4 t5 t6 t7 t8
        [1, 0, 1, 0, 1, 0, 0, 1],  # d1: مرتبط واقعی
        [0, 1, 0, 1, 0, 1, 0, 0],  # d2: نامرتبط واقعی
        [1, 1, 1, 0, 0, 0, 1, 0],  # d3: مرتبط واقعی
        [0, 0, 0, 1, 1, 1, 0, 1],  # d4: نامرتبط واقعی
        [1, 0, 1, 1, 0, 0, 1, 1],  # d5: مرتبط واقعی
        [0, 1, 0, 0, 1, 1, 0, 0],  # d6: نامرتبط واقعی
    ]
)

# تعداد کل اسناد در مجموعه (N)
N = len(documents)
# بردار پرسمان (شامل واژه‌های t1, t3, t7)
query_vector = np.array([1, 0, 1, 0, 0, 0, 1, 0])
# تعداد اسنادی که هر واژه در آن ظاهر شده (Document Frequency: df_t)
df = np.sum(documents, axis=0)

# ===================================================================
# تابع اصلی BIM برای محاسبه امتیاز (RSV)
# ===================================================================


def calculate_rsv(docs, query, N, pt_vector, ut_vector):
    """محاسبه Relative Scoring Value (RSV) بر اساس وزن های ct مدل BIM."""

    # محاسبه وزن‌های واژه ct: log( (pt / (1-pt)) * ((1-ut) / ut) )
    # این وزن نشان‌دهنده اهمیت واژه برای تفکیک اسناد مرتبط و نامرتبط است.
    # برای جلوگیری از تقسیم بر صفر یا log صفر، مقادیر 0 و 1 در pt/ut باید هندل شوند.

    # جلوگیری از log(0) با استفاده از np.clip
    pt_safe = np.clip(pt_vector, 1e-6, 1.0 - 1e-6)
    ut_safe = np.clip(ut_vector, 1e-6, 1.0 - 1e-6)

    ct = np.log(pt_safe / (1 - pt_safe)) + np.log((1 - ut_safe) / ut_safe)

    # محاسبه RSV برای هر سند: Sum_{t in query} c_t * x_t * q_t
    rsv_scores = []
    for doc in docs:
        # ضرب در query_vector تضمین می‌کند فقط واژه‌های موجود در پرسمان محاسبه شوند.
        # ضرب در doc تضمین می‌کند فقط واژه‌های موجود در سند محاسبه شوند.
        rsv = np.sum(ct * doc * query)
        rsv_scores.append(rsv)

    return np.array(rsv_scores), ct


# ===================================================================
# ۱. اجرای اولیه (بدون بازخورد)
# ===================================================================

# تخمین اولیه (مرحله 1 در MD cell): فرض اولیه ضعیف
# pt_t: احتمال حضور واژه t در اسناد مرتبط (pt = 0.5)
pt_initial = np.full(documents.shape[1], 0.5)
# ut_t: احتمال حضور واژه t در اسناد نامرتبط (تقریب با df/N)
# در حالت بدون بازخورد، ut را از روی کل مجموعه تقریب می‌زنیم.
ut_initial = df / N

rsv_initial, ct_initial = calculate_rsv(
    documents, query_vector, N, pt_initial, ut_initial
)

df_initial = pd.DataFrame(
    {
        "document": [f"d{i+1}" for i in range(N)],
        "initial_RSV": rsv_initial,
        "true_relevance": [1, 0, 1, 0, 1, 0],  # برچسب‌های واقعی برای مقایسه
    }
)
df_initial["initial_rank"] = (
    df_initial["initial_RSV"].rank(ascending=False, method="first").astype(int)
)

print("--- 1. Initial Ranking (بدون بازخورد) ---")
print(df_initial.sort_values(by="initial_rank"))

# ===================================================================
# ۲. اعمال بازخورد شبه‌مرتبط بودن (Pseudo-Relevance Feedback)
# ===================================================================

# مرحله 2 و 3 (بازخورد شبه): انتخاب K سند برتر به عنوان مجموعه اسناد مرتبط (V)
K = 3  # K سند برتر را مرتبط فرض می‌کنیم
top_k_indices = df_initial.sort_values(by="initial_RSV", ascending=False).head(K).index
# مجموعه V (اسناد فرضاً مرتبط) شامل اندیس‌های [d1, d3, d5] است. (توجه: این بستگی به RSV اولیه دارد)

# Vt: تعداد اسنادی که واژه t را دارند در مجموعه V
Vt_vector = np.sum(documents[top_k_indices], axis=0)
V_size = K  # اندازه مجموعه V

# مرحله 4 (به‌روزرسانی تخمین‌ها): محاسبه pt و ut جدید با نرمال‌سازی (Pseudo-Relevance)
# pt_t = (|V^t| + 0.5) / (|V| + 1)
pt_updated = (Vt_vector + 0.5) / (V_size + 1)

# ut_t = (df_t - |V^t| + 0.5) / (N - |V| + 1)
ut_updated = (df - Vt_vector + 0.5) / (N - V_size + 1)

print("\n" + "=" * 50)
print(f"پارامترهای بازتخمین شده پس از Pseudo-Feedback (K={K}):")
print(f"واژه‌های پرسمان (t1, t3, t7): {query_vector}")
print(f"DF (df_t): {df}")
print(f"تعداد حضور در V (Vt): {Vt_vector}")
print(f"PT (احتمال در مرتبط): {pt_updated.round(4)}")
print(f"UT (احتمال در نامرتبط): {ut_updated.round(4)}")
print("=" * 50)


# محاسبه RSV نهایی با وزن‌های جدید
rsv_final, ct_final = calculate_rsv(documents, query_vector, N, pt_updated, ut_updated)

df_final = pd.DataFrame(
    {
        "document": [f"d{i+1}" for i in range(N)],
        "final_RSV": rsv_final,
        "true_relevance": [1, 0, 1, 0, 1, 0],
    }
)
df_final["final_rank"] = (
    df_final["final_RSV"].rank(ascending=False, method="first").astype(int)
)

print("\n--- 2. Final Ranking (با بازخورد شبه) ---")
print(df_final.sort_values(by="final_rank"))

# ===================================================================
# ۳. نمایش تغییرات رتبه و وزن‌ها
# ===================================================================

print("\n--- 3. مقایسه وزن‌های واژه‌های پرسمان (t1, t3, t7) ---")
# واژه‌های پرسمان: t1 (index 0), t3 (index 2), t7 (index 6)
query_terms_indices = np.where(query_vector == 1)[0]
weights_comparison = pd.DataFrame(
    {
        "term": [f"t{i+1}" for i in query_terms_indices],
        "initial_weight_ct": ct_initial[query_terms_indices].round(4),
        "final_weight_ct": ct_final[query_terms_indices].round(4),
    }
)
print(weights_comparison)

# ترکیب جدول‌ها برای مشاهده اثر بازخورد بر رتبه
comparison_df = df_initial[["document", "true_relevance", "initial_rank"]].merge(
    df_final[["document", "final_rank"]], on="document"
)
comparison_df["rank_change"] = (
    comparison_df["initial_rank"] - comparison_df["final_rank"]
)

print("\n--- 4. مقایسه نهایی رتبه‌بندی ---")
print(comparison_df.sort_values(by="final_rank"))

# مشاهده می‌کنیم که وزن‌ها و در نتیجه رتبه‌بندی، بر اساس اطلاعات جدید (K سند برتر) به‌روز شده‌اند.
# اگر رتبه‌بندی بهبود یافته باشد، اسناد مرتبط (با true_relevance=1) باید به رتبه‌های بالاتر صعود کرده باشند.
```

</p>

<p style="font-size:16pt; text-align:right;">
مدل‌های احتمالاتی از قدیمی‌ترین چارچوب‌های رسمی در بازیابی اطلاعات هستند. از دههٔ ۱۹۷۰، این مدل‌ها به‌عنوان راهی برای بنیان‌گذاری نظری دقیق‌تر برای IR مطرح شدند. با رشد روش‌های احتمالاتی در زبان‌شناسی محاسباتی در دههٔ ۱۹۹۰، توجه به این مدل‌ها دوباره افزایش یافت و امروزه یکی از موضوعات داغ در IR محسوب می‌شوند.

<span id="section11.4.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۱ ارزیابی مدل‌های احتمالاتی
</span>
</p>

<p style="font-size:16pt; text-align:right;">
اگرچه مدل‌های احتمالاتی ایده‌های زیبایی دارند، اما در عمل، عملکرد آن‌ها معمولاً ضعیف‌تر از مدل‌های برداری بوده است. دلیل اصلی، نیاز به تقریب‌های آماری پیچیده و فرض‌های سخت‌گیرانه است. در مدل BIM، این فرض‌ها عبارتند از:
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔍 فرض‌های مدل BIM: محدودیت‌های دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
بیایید این فرض‌ها را با یک مثال واقعی بررسی کنیم. فرض کنید در آمازون دنبال "لپ‌تاپ گیمینگ سبک" می‌گردید:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>نمایش بولی برای سند، پرسمان و مرتبط بودن:</strong> سیستم فقط می‌فهمد که کلمات "لپ‌تاپ"، "گیمینگ" و "سبک" در سند وجود دارند یا نه. اما نمی‌فهمد که "سبک" به معنای وزن کم است یا طراحی مینیمال.</li>
<li><strong>استقلال واژه‌ها از یکدیگر:</strong> سیستم فرض می‌کند "لپ‌تاپ" و "گیمینگ" دو کلمه مستقل هستند، در حالی که در واقعیت، "لپ‌تاپ گیمینگ" یک مفهوم واحد است.</li>
<li><strong>واژه‌هایی که در پرسمان نیستند، تأثیری در نتیجه ندارند:</strong> اگر یک لپ‌تاپ عالی وجود داشته باشد که در توضیحاتش کلمه "سبک" نیامده اما واقعاً سبک است، سیستم آن را نادیده می‌گیرد.</li>
<li><strong>مرتبط بودن اسناد مستقل از یکدیگر است:</strong> سیستم فرض می‌کند relevancy یک سند به اسناد دیگر بستگی ندارد، در حالی که در واقعیت، وقتی چند لپ‌تاپ مشابه پیدا می‌کنید، مقایسه آن‌ها به تصمیم‌گیری کمک می‌کند.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
این فرض‌ها باعث می‌شوند مدل در شرایط واقعی عملکرد ضعیفی داشته باشد، مگر آن‌که اطلاعاتی دربارهٔ مرتبط بودن اسناد در اختیار داشته باشیم.
</p>

<hr>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚀 BM25: انقلابی در مدل‌های احتمالاتی</p>
<p style="font-size:16pt; text-align:right;">
در دههٔ ۱۹۹۰، استفان رابرتسون و کارن اسپارک جونز در دانشگاه کمبریج روی یک فرمول جدید کار می‌کردند. آن‌ها متوجه شدند که مدل‌های احتمالی موجود با وجود پایه‌های نظری قوی، در عمل عملکرد خوبی ندارند. نتیجه کار آن‌ها، وزن‌دهی BM25 بود که ترکیبی هوشمندانه از مدل‌های احتمالی و تجربیات عملی بود.</p>
<p style="font-size:16pt; text-align:right;">
BM25 به سرعت توسط گروه‌های مختلف پذیرفته شد، زیرا عملکرد بسیار خوبی داشت و پیاده‌سازی آن نسبتاً ساده بود. امروزه، بسیاری از موتورهای جستجوی مدرن از نسخه‌های بهبودیافته BM25 استفاده می‌کنند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در واقع، تفاوت بین سیستم‌های IR برداری و احتمالاتی آن‌قدر زیاد نیست. در هر دو، ساختار ایندکس، پردازش پرسمان، و بازیابی اسناد مشابه است. تفاوت اصلی در <strong>فرمول امتیازدهی نهایی</strong> است:
</p>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔄 از برداری تا احتمالی: تغییر یک فرمول، تغییر یک دنیا</p>
<p style="font-size:16pt; text-align:right;">
تیم مهندسی در شرکت مایکروسافت در سال ۲۰۰۳ با یک چالش روبرو بود: سیستم جستجوی بینگ (که آن زمان MSN Search نام داشت) عملکرد خوبی نداشت. آن‌ها تصمیم گرفتند یک آزمایش انجام دهند: بدون تغییر هیچ بخش دیگری از سیستم، فقط فرمول امتیازدهی را از شباهت کسینوسی به BM25 تغییر دادند.</p>
<p style="font-size:16pt; text-align:right;">
نتیجه شگفت‌انگیز بود: دقت نتایج جستجو ۲۳٪ افزایش یافت! این مثال نشان می‌دهد که تفاوت اصلی بین مدل‌های برداری و احتمالاتی در فرمول امتیازدهی است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>در مدل برداری:</strong> امتیاز بر اساس شباهت کسینوسی و tf-idf</li>
<li><strong>در مدل احتمالاتی:</strong> امتیاز بر اساس احتمال مرتبط بودن و وزن‌های BM25</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در بسیاری از موارد، سیستم‌های برداری موجود با جایگزینی فرمول‌های وزن‌دهی از مدل‌های احتمالاتی، به‌صورت مؤثر به سیستم‌های احتمالاتی تبدیل شده‌اند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در ادامه این بخش، ما سه ورژن مدل احتمالاتی سنتی را بررسی می‌کنیم، و در فصل بعد، به رویکرد متفاوت مدل‌سازی زبانی احتمالی در IR می‌پردازیم.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.4.2" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۲ وابستگی‌های درختی بین واژه‌ها: فراتر از استقلال
</span>
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 وقتی "هنگ کنگ" از "هنگ" و "کنگ" متفاوت است</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در گوگل عبارت "هتل‌های نزدیک دانشگاه نیویورک" را جستجو می‌کنید. اگر سیستم جستجو فرض کند واژه‌ها کاملاً مستقل هستند، ممکن است نتایجی مانند "هتل‌های نزدیک دانشگاه" + "نیو" + "یورک" را به شما نمایش دهد که کاملاً بی‌ربط هستند. در واقع، "نیویورک" یک مفهوم واحد است، نه دو واژه مستقل "نیو" و "یورک".</p>
<p style="font-size:16pt; text-align:right;">
این دقیقاً همان مشکلی است که مدل BIM با فرض استقلال واژه‌ها با آن روبرو است. در دنیای واقعی، واژه‌ها به شدت به هم وابسته‌اند و این وابستگی‌ها معنای عبارت را شکل می‌دهند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
وابستگی‌ها می‌توانند ساختارهای پیچیده‌تری نیز داشته باشند. مثلاً بین مجموعه‌ای از واژه‌ها مانند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>New</li>
<li>York</li>
<li>England</li>
<li>City</li>
<li>Stock</li>
<li>Exchange</li>
<li>University</li>
</ul>

<p style="font-size:16pt; text-align:right;">
در اینجا، "New York" یک مفهوم است، "Stock Exchange" یک مفهوم دیگر، و "New York City" یک مفهوم سوم. این وابستگی‌ها را نمی‌توان با فرض استقلال واژه‌ها مدل‌سازی کرد.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🌳 مدل درختی وابستگی‌ها: ایده‌ای که از دهه ۱۹۷۰ تا امروز زنده مانده</p>
<p style="font-size:16pt; text-align:right;">
در سال ۱۹۷۹، <strong>van Rijsbergen</strong> مدلی ساده و قابل قبول پیشنهاد داد که وابستگی‌ها را به‌صورت درختی نمایش می‌دهد. در این مدل، هر واژه فقط به یک واژهٔ دیگر وابسته است، و ساختار کلی به‌صورت درخت جهت‌دار است.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، در عبارت "دانشگاه شهر نیویورک"، واژه "شهر" به "نیویورک" وابسته است، و "نیویورک" به "دانشگاه" وابسته است. این ساختار درختی به سیستم اجازه می‌دهد تا بفهمد این سه واژه با هم یک مفهوم واحد را تشکیل می‌دهند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔄 بازآفرینی ایده: از مدل Rijsbergen تا Tree Augmented Naive Bayes</p>
<p style="font-size:16pt; text-align:right;">
در دههٔ ۱۹۷۰، مشکلات تخمین آماری مانع موفقیت عملی این مدل شد. اما در سال ۱۹۹۶، <strong>Friedman و Goldszmidt</strong> این ایده را با موفقیت در قالب مدل <strong>Tree Augmented Naive Bayes (TAN)</strong> بازآفرینی کردند.</p>
<p style="font-size:16pt; text-align:right;">
شرکت‌هایی مانند مایکروسافت و آمازون از مدل‌های مشابه TAN در سیستم‌های توصیه‌گر خود استفاده می‌کنند. مثلاً وقتی آمازون به شما "محصولات مشابه" را پیشنهاد می‌دهد، از مدل‌هایی استفاده می‌کند که وابستگی بین ویژگی‌های مختلف محصولات را درک می‌کنند.</p>
<p style="font-size:16pt; text-align:right;">
مدل TAN ترکیبی از فرض استقلال شرطی Naive Bayes و وابستگی‌های درختی است — به‌طوری‌که هر ویژگی (واژه) علاوه بر وابستگی به کلاس، به یک ویژگی دیگر نیز وابسته است. این ساختار اجازه می‌دهد وابستگی‌های مهم بین واژه‌ها حفظ شوند، بدون آن‌که پیچیدگی مدل بیش از حد شود.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 نمایش بصری وابستگی‌ها</p>
<p style="font-size:16pt; text-align:right;">
در شکل ۱۱.۱، یک نمونه از ساختار درختی وابستگی بین واژه‌ها نمایش داده شده است. در این نمایش گرافیکی، یک واژه xi مستقیماً به واژه xk وابسته است اگر یک پیکان xk → xi وجود داشته باشد.</p>
<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/11.1.bmp" alt="Figure 11.1 A tree of dependencies between terms. In this graphical model representation, a term xi is directly dependent on a term xk if there is an arrow xk → xi ." style="width:600px; border:1px solid #ccc; border-radius:12px;"
</div>
</div>
<p style="font-size:16pt; text-align:right;">
این ساختار به ما کمک می‌کند تا بفهمیم چگونه واژه‌ها به هم متصل هستند و چگونه این اتصالات می‌توانند معنای عبارت را تغییر دهند. مثلاً در عبارت "بورس نیویورک"، واژه "بورس" به "نیویورک" وابسته است، نه برعکس. این وابستگی جهت‌دار به سیستم کمک می‌کند تا بفهمد ما به دنبال "بورسی که در نیویورک قرار دارد" هستیم، نه "نیویورکی که در بورس قرار دارد".</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.4.3" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۳ مدل غیر دودویی Okapi BM25: از کتابخانه‌ها تا گوگل
</span>
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 چرا جستجوی "هوش مصنوعی" در گوگل بهتر از جستجوی کاتالوگ قدیمی کتابخانه کار می‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در یک سیستم جستجوی قدیمی کاتالوگ کتابخانه، عبارت "هوش مصنوعی" را جستجو می‌کنید. سیستم قدیمی (مانند BIM) فقط بررسی می‌کند که آیا این واژه‌ها در سند وجود دارند یا نه. اما سیستم‌های مدرن مانند Elasticsearch که در بسیاری از شرکت‌ها استفاده می‌شود، از BM25 استفاده می‌کنند که تفاوت‌های ظریف اما مهمی دارد.</p>
<p style="font-size:16pt; text-align:right;">
BM25 متوجه می‌شود که در یک مقاله کوتاه، تکرار دو بار واژه "هوش مصنوعی" مهم‌تر از تکرار آن در یک کتاب ۵۰۰ صفحه‌ای است. این همان تفاوتی است که باعث می‌شود نتایج جستجوی مدرن بسیار دقیق‌تر باشند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
مدل BIM برای رکوردهای کوتاه و چکیده‌هایی با طول ثابت طراحی شده بود. اما در مجموعه‌های مدرن جست‌وجوی متن کامل، باید به <strong>فراوانی واژه</strong> و <strong>طول سند</strong> توجه شود. مدل <strong>BM25</strong> که به‌نام <em>Okapi weighting</em> نیز شناخته می‌شود، برای همین هدف طراحی شده است: ساخت یک مدل احتمالاتی که به این عوامل حساس باشد ولی پارامترهای اضافی زیادی وارد نکند.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 تکامل BM25: از ساده تا پیچیده</p>
<p style="font-size:16pt; text-align:right;">
ساده‌ترین نسخهٔ BM25 فقط از وزن‌دهی <code>IDF</code> استفاده می‌کند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N}{df_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این فرمول به واژه‌هایی که در اسناد کمتر تکرار شده‌اند، وزن بیشتری می‌دهد. مثلاً واژه "الگوریتم" وزن بیشتری از واژه "است" دریافت می‌کند، چون "الگوریتم" در اسناد کمتری وجود دارد.</p>

<p style="font-size:16pt; text-align:right;">
نسخهٔ جایگزین با نرم‌سازی (در غیاب بازخورد مرتبط بودن):</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N - df_t + 0.5}{df_t + 0.5} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این نسخه کمی متفاوت عمل می‌کند. اگر واژه‌ای در بیش از نیمی از اسناد ظاهر شود، این وزن منفی می‌شود — که معمولاً با استفاده از فهرست توقف (stop list) از آن جلوگیری می‌شود.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 BM25 کامل: فرمولی که دنیای جستجو را تغییر داد</p>
<p style="font-size:16pt; text-align:right;">
نسخهٔ کامل BM25 با در نظر گرفتن فراوانی واژه و طول سند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N}{df_t} \right) \cdot \frac{(k_1 + 1) \cdot tf_{td}}{k_1 \cdot ((1 - b) + b \cdot \frac{L_d}{L_{avg}}) + tf_{td}}
$$
</div>

<p style="font-size:16pt; text-align:right;">
بیایید این فرمول را با یک مثال واقعی تجزیه کنیم. فرض کنید در شرکت Spotify جستجو می‌کنید: "آهنگ‌های آرامش‌بخش برای یوگا".</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>tf<sub>td</sub></code>: فراوانی واژه "آرامش‌بخش" در توضیحات یک پلی‌لیست</li>
<li><code>L<sub>d</sub></code>: طول توضیحات آن پلی‌لیست</li>
<li><code>L<sub>avg</sub></code>: میانگین طول توضیحات همه پلی‌لیست‌ها</li>
<li><code>k₁</code>: پارامتر تنظیم شدت تأثیر فراوانی واژه</li>
<li><code>b</code>: پارامتر تنظیم نرمال‌سازی طول سند (بین ۰ تا ۱)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
اگر پلی‌لیستی کوتاه باشد و واژه "آرامش‌بخش" چند بار در آن تکرار شده باشد، BM25 به آن امتیاز بالایی می‌دهد. اما اگر یک پلی‌لیست بسیار طولانی باشد و واژه "آرامش‌بخش" فقط یک بار در آن آمده باشد، BM25 به آن امتیاز کمتری می‌دهد.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔧 پارامترهای BM25: تنظیم دقیق موتور جستجو</p>
<p style="font-size:16pt; text-align:right;">
اگر پرسمان طولانی باشد، می‌توان وزن‌دهی مشابهی برای واژه‌های پرسمان نیز اعمال کرد:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N}{df_t} \right) \cdot \frac{(k_1 + 1) \cdot tf_{td}}{k_1 \cdot ((1 - b) + b \cdot \frac{L_d}{L_{avg}}) + tf_{td}} \cdot \frac{(k_3 + 1) \cdot tf_{tq}}{k_3 + tf_{tq}}
$$
</div>

<p style="font-size:16pt; text-align:right;">
شرکت‌هایی مانند مایکروسافت (که سیستم Okapi را توسعه داد) و Elastic (که Elasticsearch را می‌سازد) این پارامترها را برای نیازهای خاص خود تنظیم می‌کنند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>k₁</code> و <code>k₃</code>: بین ۱.۲ تا ۲ (کنترل می‌کنند که تکرار واژه چقدر مهم باشد)</li>
<li><code>b</code>: حدود ۰.۷۵ (کنترل می‌کند که طول سند چقدر در امتیازدهی تأثیر داشته باشد)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
برای مثال، در جستجوی محصولات آمازون، ممکن است مقدار b را بیشتر تنظیم کنند تا طول توضیحات محصول تأثیر بیشتری داشته باشد، اما در جستجوی توییت‌ها، ممکن است آن را کمتر تنظیم کنند چون همه توییت‌ها طول تقریباً یکسانی دارند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
اگر اطلاعات مرتبط بودن در دسترس باشد، می‌توان از نسخهٔ کامل زیر استفاده کرد:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{(|VR_t| + 0.5)/(|VNR_t| + 0.5)}{(df_t - |VR_t| + 0.5)/(N - df_t - |VR| + |VR_t| + 0.5)} \right) \cdot \text{TF}_{doc} \cdot \text{TF}_{query}
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>VR<sub>t</sub></code>: تعداد اسناد مرتبطی که واژهٔ <code>t</code> را دارند</li>
<li><code>VNR<sub>t</sub></code>: تعداد اسناد نامرتبطی که واژهٔ <code>t</code> را دارند</li>
<li><code>TF<sub>doc</sub></code>: وزن‌دهی فراوانی واژه در سند</li>
<li><code>TF<sub>query</sub></code>: وزن‌دهی فراوانی واژه در پرسمان</li>
</ul>
</p>

<div style="background:#eef; padding:15px; margin:20px 0; border-radius:8px; text-align:right;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#2c3e50;">💡 بازخورد مرتبط بودن: هوشمندتر کردن جستجو</p>
<p style="font-size:16pt; text-align:right;">
در بازخورد مرتبط بودن، می‌توان پرسمان را با واژه‌های مهم از اسناد مرتبط گسترش داد (مثلاً ۱۰ تا ۲۰ واژهٔ برتر بر اساس <code>ĉ<sub>t</sub></code>) و فرمول BM25 را روی پرسمان گسترش‌یافته اعمال کرد.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وقتی در گوگل "رستوران ایتالیایی" را جستجو می‌کنید و روی نتیجه‌ای کلیک می‌کنید، گوگل از این رفتار شما یاد می‌گیرد و ممکن است در جستجوی‌های بعدی، واژه‌هایی مانند "پیتزا" یا "پاستا" را به پرسمان شما اضافه کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
فرمول‌های BM25 در ارزیابی‌های TREC عملکرد بسیار خوبی داشته‌اند و توسط گروه‌های مختلف به‌طور گسترده پذیرفته شده‌اند. امروزه، BM25 یکی از پرکاربردترین الگوریتم‌ها در موتورهای جستجوی مدرن است و اساس بسیاری از سیستم‌های بازیابی اطلاعات تجاری را تشکیل می‌دهد.</p>

</div>

```python
import math
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# تعریف مجموعه‌ای از اسناد به‌صورت لیست واژه‌ها
documents = [
    ["هوش", "مصنوعی", "یادگیری", "ماشین", "الگوریتم"],
    ["یادگیری", "عمیق", "شبکه", "عصبی", "هوش", "مصنوعی"],
    ["الگوریتم", "بهینه‌سازی", "یادگیری", "ماشین", "داده"],
    ["پردازش", "زبان", "طبیعی", "هوش", "مصنوعی", "متن"],
    ["یادگیری", "ماشین", "الگوریتم", "داده", "مدل", "آماری"],
    ["بینایی", "ماشین", "شبکه", "عصبی", "یادگیری", "عمیق"],
]

# تعریف پرسمان به‌صورت لیست واژه‌ها
query = ["یادگیری", "ماشین", "الگوریتم"]

# پارامترهای BM25
k1 = 1.5
b = 0.75

# --- مرحله ۱: محاسبات اولیه ---

# محاسبه فراوانی واژه (tf) در هر سند
tf_list = [Counter(doc) for doc in documents]

# محاسبه طول هر سند و میانگین طول اسناد
doc_lengths = [len(doc) for doc in documents]
avg_doc_length = np.mean(doc_lengths)

# محاسبه فراوانی سند (df) برای هر واژه در پرسمان
N = len(documents)
df = {}
for term in query:
    df[term] = sum(1 for doc in documents if term in doc)

# محاسبه وزن IDF برای هر واژه با استفاده از فرمول استاندارد احتمالی
# این فرمول به طور خاص برای مدل‌های احتمالی مانند BM25 طراحی شده است.
idf = {}
for term in query:
    # اضافه کردن 0.5 برای جلوگیری از تقسیم بر صفر و صفر شدن لگاریتم
    idf[term] = math.log((N - df[term] + 0.5) / (df[term] + 0.5))

# --- مرحله ۲: محاسبه امتیاز BM25 ---

# محاسبه امتیاز RSV برای هر سند با فرمول BM25
rsv_scores = []
for i, tf_td in enumerate(tf_list):
    Ld = doc_lengths[i]
    score = 0
    for term in query:
        tf = tf_td.get(term, 0)
        # بخش فراوانی واژه در سند (TF component)
        numerator = (k1 + 1) * tf
        denominator = k1 * ((1 - b) + b * (Ld / avg_doc_length)) + tf
        score += idf[term] * (numerator / denominator)
    rsv_scores.append(score)

# ساخت جدول نهایی و چاپ نتایج
ranking_indices = np.argsort(-np.array(rsv_scores))
df_result = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "BM25 Score": np.round(rsv_scores, 4),
        "Rank": ranking_indices.argsort() + 1,
    }
)

print("Document Ranking with BM25 Model:")
print(df_result.sort_values(by="Rank"))

# --- مرحله ۳: مقایسه با مدل BIM (دودویی) ---

bim_scores = []
for i, tf_td in enumerate(tf_list):
    score = 0
    for term in query:
        # در مدل BIM، فقط وجود یا عدم وجود واژه مهم است (tf صفر یا یک)
        if term in tf_td:
            score += idf[term]
    bim_scores.append(score)

# نمایش نتایج مقایسه‌ای
comparison_df = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "BIM Score": np.round(bim_scores, 4),
        "BM25 Score": np.round(rsv_scores, 4),
    }
)

print("\nComparison between BIM and BM25 Models:")
print(comparison_df)

# --- مرحله ۴: تحلیل تأثیر پارامترهای BM25 ---


def bm25_with_params(k1, b):
    """تابعی برای محاسبه امتیاز BM25 با پارامترهای دلخواه"""
    scores = []
    for i, tf_td in enumerate(tf_list):
        Ld = doc_lengths[i]
        score = 0
        for term in query:
            tf = tf_td.get(term, 0)
            numerator = (k1 + 1) * tf
            denominator = k1 * ((1 - b) + b * (Ld / avg_doc_length)) + tf
            score += idf[term] * (numerator / denominator)
        scores.append(score)
    return scores


# تأثیر پارامتر k1
k1_values = [0.5, 1.0, 1.5, 2.0]
k1_effects = [bm25_with_params(k1, 0.75) for k1 in k1_values]

# تأثیر پارامتر b
b_values = [0.25, 0.5, 0.75, 1.0]
b_effects = [bm25_with_params(1.5, b) for b in b_values]

# رسم نمودار تأثیر پارامترها
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
for i, k1_val in enumerate(k1_values):
    plt.plot(range(1, N + 1), k1_effects[i], marker="o", label=f"k1={k1_val}")
plt.xlabel("Document ID")
plt.ylabel("BM25 Score")
plt.title("Effect of k1 Parameter")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
for i, b_val in enumerate(b_values):
    plt.plot(range(1, N + 1), b_effects[i], marker="s", label=f"b={b_val}")
plt.xlabel("Document ID")
plt.ylabel("BM25 Score")
plt.title("Effect of b Parameter")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# --- مرحله ۵: بازخورد مرتبط بودن (Relevance Feedback) ---


def bm25_with_relevance_feedback(relevant_docs, non_relevant_docs, k1=1.5, b=0.75):
    """
    پیاده‌سازی BM25 با بازخورد مرتبط بودن.
    این تابع وزن واژه‌ها را بر اساس اسنادی که کاربر به عنوان مرتبط/نامرتبط علامت‌گذاری کرده، تنظیم مجدد می‌کند.
    """
    # محاسبه VR_t (فراوانی در اسناد مرتبط) و VNR_t (فراوانی در اسناد نامرتبط)
    VR = {}
    VNR = {}
    for term in query:
        VR[term] = sum(1 for i in relevant_docs if term in tf_list[i])
        VNR[term] = sum(1 for i in non_relevant_docs if term in tf_list[i])

    # محاسبه امتیاز با فرمول کامل BM25 با بازخورد مرتبط بودن
    scores = []
    for i, tf_td in enumerate(tf_list):
        Ld = doc_lengths[i]
        score = 0
        for term in query:
            tf = tf_td.get(term, 0)

            # محاسبه وزن بازخورد مرتبط بودن (جایگزین وزن IDF می‌شود)
            # این فرمول از تئوری احتمالی نشأت می‌گیرد و وزن واژه را بر اساس
            # توزیع آن در اسناد مرتبط و نامرتبب تنظیم می‌کند.
            relevance_weight = math.log(
                ((VR[term] + 0.5) / (VNR[term] + 0.5))
                / (
                    (df[term] - VR[term] + 0.5)
                    / (N - df[term] - len(relevant_docs) + VR[term] + 0.5)
                )
            )

            # محاسبه وزن فراوانی واژه در سند (بخش TF فرمول BM25)
            tf_weight = (
                (k1 + 1) * tf / (k1 * ((1 - b) + b * (Ld / avg_doc_length)) + tf)
            )

            score += relevance_weight * tf_weight
        scores.append(score)

    return scores


# مثال بازخورد مرتبط بودن
# فرض می‌کنیم اسناد 2 و 4 مرتبط هستند و اسناد 1 و 5 نامرتبط
# (شمارش از 0 شروع می‌شود)
relevant_docs_indices = [1, 3]  # Doc 2 and Doc 4
non_relevant_docs_indices = [0, 4]  # Doc 1 and Doc 5

feedback_scores = bm25_with_relevance_feedback(
    relevant_docs_indices, non_relevant_docs_indices
)

feedback_df = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "Original BM25": np.round(rsv_scores, 4),
        "BM25 with Feedback": np.round(feedback_scores, 4),
    }
)

print("\nBM25 with Relevance Feedback:")
print(feedback_df)

# --- مرحله ۶: نمایش تغییرات رتبه‌بندی ---

original_ranking = np.argsort(-np.array(rsv_scores))
feedback_ranking = np.argsort(-np.array(feedback_scores))

ranking_change_df = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "Original Rank": original_ranking.argsort() + 1,
        "Rank with Feedback": feedback_ranking.argsort() + 1,
    }
)

# محاسبه تغییر رتبه و وضعیت جدید برای وضوح بیشتر
ranking_change_df["Change"] = (
    ranking_change_df["Original Rank"] - ranking_change_df["Rank with Feedback"]
)
ranking_change_df["Status"] = ranking_change_df["Change"].apply(
    lambda x: f"↑ {x}" if x > 0 else (f"↓ {abs(x)}" if x < 0 else "-")
)

print("\nRanking Changes after Relevance Feedback:")
print(ranking_change_df.sort_values(by="Original Rank"))
```

<span id="section11.4.4" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">

</span>
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 هنگامی که گوگل حدس می‌زند شما به دنبال چه چیزی هستید</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در گوگل عبارت " سرعت جگوار" را جستجو می‌کنید. آیا به دنبال حیوان جگوار هستید یا ماشین جگوار؟ گوگل چگونه این تفاوت را تشخیص می‌دهد؟</p>
<p style="font-size:16pt; text-align:right;">
سیستم‌های مدرن از تکنیک‌های پیچیده‌ای استفاده می‌کنند، اما یکی از رویکردهای پیشگامانه در این زمینه، استفاده از شبکه‌های بیزی بود. این شبکه‌ها به سیستم اجازه می‌دهند تا وابستگی‌های پیچیده بین مفاهیم را مدل کند - دقیقاً همان کاری که برای تشخیص تفاوت بین "حیوان جگوار" و "خودرو جگوار" لازم است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در سال‌های ۱۹۸۹ و ۱۹۹۱، <strong>Turtle و Croft</strong> استفاده از <strong>شبکه‌های بیزی (Bayesian Networks)</strong> را در بازیابی اطلاعات معرفی کردند. شبکهٔ بیزی نوعی مدل گرافیکی احتمالاتی است که وابستگی‌های بین متغیرها را با گراف جهت‌دار نمایش می‌دهد. این مدل‌ها اجازهٔ یادگیری و استنتاج در گراف‌های جهت‌دار بدون چرخه (DAG) را فراهم می‌کنند.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔗 معماری دو بخشی: اسناد و پرسمان‌ها</p>
<p style="font-size:16pt; text-align:right;">
مدل پیشنهادی Turtle و Croft از دو بخش تشکیل می‌شود که مانند دو قطعه پازل با هم کار می‌کنند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>شبکهٔ مجموعهٔ اسناد:</strong> شبکه‌ای بزرگ که از قبل قابل محاسبه است. این شبکه نگاشت از اسناد به واژه‌ها و سپس به مفاهیم است. مفاهیم از گسترش اصطلاحات با استفاده از thesaurus ساخته می‌شوند.</li>
<li><strong>شبکهٔ پرسمان:</strong> شبکه‌ای کوچک که برای هر پرسمان جدید ساخته می‌شود. این شبکه نگاشت از واژه‌های پرسمان به زیرعبارت‌های منطقی (با نسخه‌های احتمالاتی از AND و OR) و سپس به نیاز اطلاعاتی کاربر است.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
بیایید با یک مثال این را توضیح دهیم: وقتی در آمازون "لپ‌تاپ سبک برای دانشجو" را جستجو می‌کنید:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>شبکه پرسمان، "لپ‌تاپ"، "سبک" و "دانشجو" را به مفاهیم مرتبط متصل می‌کند (مثلاً "سبک" به "قابل حمل" و "وزن کم")</li>
<li>سپس این شبکه به شبکه اسناد متصل می‌شود که از قبل لپ‌تاپ‌ها را به ویژگی‌هایشان متصل کرده است</li>
<li>در نهایت، سیستم احتمال اینکه هر لپ‌تاپ با نیاز شما مطابقت داشته باشد را محاسبه می‌کند</li>
</ol>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚀 InQuery: موفقیتی که الهام‌بخش نسل جدیدی از موتورهای جستجو شد</p>
<p style="font-size:16pt; text-align:right;">
با اتصال این دو شبکه، یک سیستم انعطاف‌پذیر و احتمالاتی ساخته می‌شود که می‌تواند مدل‌های ساده‌تر بولی و احتمالاتی را تعمیم دهد. این مدل به‌طور طبیعی از عملگرهای ساختاری پرسمان پشتیبانی می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
این سیستم پایهٔ <strong>InQuery</strong> بود — یک سیستم بازیابی متن که در دانشگاه ماساچوست توسعه یافت و در ارزیابی‌های TREC عملکرد بسیار خوبی داشت. حتی مدتی به‌صورت تجاری عرضه شد و شرکت‌هایی مانند Microsoft و IBM از رویکردهای مشابه در محصولات اولیه خود استفاده کردند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔮 آینده شبکه‌های بیزی در بازیابی اطلاعات</p>
<p style="font-size:16pt; text-align:right;">
با این حال، برای تخمین پارامترها و محاسبات، مدل همچنان از تقریب‌ها و فرض‌های استقلال استفاده می‌کرد. به همین دلیل، ادامهٔ پژوهش‌ها در این مسیر محدود شد.</p>
<p style="font-size:16pt; text-align:right;">
اما این مدل در واقع در آغاز عصر مدرن استفاده از شبکه‌های بیزی ساخته شد، و از آن زمان، پیشرفت‌های زیادی در نظریهٔ شبکه‌های بیزی رخ داده است. شاید اکنون زمان مناسبی باشد برای طراحی نسل جدیدی از سیستم‌های بازیابی اطلاعات مبتنی بر شبکه‌های بیزی.</p>
<p style="font-size:16pt; text-align:right;">
امروزه، با ظهور یادگیری عمیق و محاسبات توزیع‌شده، می‌توان شبکه‌های بیزی بسیار بزرگ‌تری را پیاده‌سازی کرد که وابستگی‌های پیچیده‌تری را مدل کنند. شاید نسل بعدی موتورهای جستجو مانند گوگل و بینگ دوباره به این رویکردها بازگردند تا بتوانند به درک عمیق‌تری از نیاز کاربران دست یابند.</p>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#f0f8ff; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.5" style="font-size:24pt; font-weight:bold; color:#8549EB;">جمع‌بندی فصل ۱۱: رویکردهای احتمالی در بازیابی اطلاعات</span> 📈
</p>

<p style="font-size:16pt; text-align:right;">
چگونه می‌توانیم در برابر عدم قطعیت دنیای واقعی، یک سیستم جستجوی هوشمند بسازیم؟ فصل ۱۱ ما را به قلب یکی از بنیادی‌ترین و قدرتمندترین رویکردها در بازیابی اطلاعات می‌برد: <strong>مدل‌های احتمالی</strong>. به جای تکیه بر تطابق کلمات کلیدی یا فاصله در یک فضای برداری، این فصل به ما یاد می‌دهد که چگونه با استفاده از نظریه احتمالات، می‌توان به شکلی اصولی و علمی به این سوال پاسخ داد: «احتمال اینکه این سند به نیاز اطلاعاتی کاربر مرتبط باشد، چقدر است؟» این تغییر نگرش، درهای جدیدی برای ساخت سیستم‌های هوشمندتر و قابل‌انطباق‌تر گشود.
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>اصل رتبه‌بندی احتمالی (PRP) را به عنوان سنگ بنای این رویکرد معرفی کردیم. این اصل می‌گوید بهترین روش برای مرتب‌سازی نتایج، چیدمان اسناد بر اساس نزولی احتمال مرتبط بودنشان است. این اصل، زیربنای نظری بسیاری از سیستم‌های مدرن است.</li>
<li>مدل استقلال دوتایی (BIM) را به عنوان اولین مدل عملیاتی بررسی کردیم. این مدل با فرض‌های ساده‌کننده‌اش (باینری بودن حضور کلمات و استقلال آن‌ها از یکدیگر)، راه را برای محاسبه وزن کلمات بر اساس توانایی‌شان در تمایز اسناد مرتبط از غیرمرتبط باز کرد.</li>
<li>با چالش برآورد احتمالات در دنیای واقعی روبرو شدیم. دیدیم که چگونه تقریب زدن احتمال حضور کلمه در اسناد غیرمرتبط، به طور شگفت‌انگیزی به توجیه معیار وزن‌دهی معروف <b>idf</b> منجر می‌شود.</li>
<li>قدرت بازخورد مرتبط بودن (Relevance Feedback) را در چارچوب احتمالی کشف کردیم. این فرآیند تکراری، به سیستم اجازه می‌دهد با یادگیری از قضاوت‌های کاربر (حتی قضاوت‌های اولیه)، تخمین خود را از احتمالات بهبود بخشد و نتایج دقیق‌تری ارائه دهد.</li>
<li>با <b>Okapi BM25</b> آشنا شدیم؛ یکی از موفق‌ترین و پرکاربردترین مدل‌های وزن‌دهی که از دل نظریه احتمالات بیرون آمده است. BM25 با فراتر رفتن از مدل باینری و در نظر گرفتن فراوانت کلمه و طول سند، عملکردی فوق‌العاده در دنیای واقعی از خود نشان داده و به یک استاندارد صنعتی تبدیل شده است.</li>
<li>در نهایت، به نگاهی به مدل‌های پیشرفته‌تر مانند <b>شبکه‌های بیزین</b> انداختیم که امکان مدل‌سازی وابستگی‌های پیچیده بین کلمات را فراهم می‌کنند و پتانسیل ساخت سیستم‌های بسیار پیچیده‌تر را دارند.</li>
</ul>

<div style="background:#e8f4fd; border-radius:8px; padding:15px; margin:15px 0;">
<p style="font-size:16pt; text-align:right;">
<b>بیایید این مفاهیم را با چند مثال واقعی و کاربردی زنده کنیم:</b>
</p>
<hr style="border-top: 1px solid #ccc; margin: 10px 0;">

<p style="font-size:16pt; text-align:right;">
<b>🔹 اصل رتبه‌بندی احتمالی (PRP) در موتورهای جستجوی مدرن:</b> وقتی شما در گوگل عبارت «رستوران ایتالیایی نزدیک من» را جستجو می‌کنید، موتور جستجو فقط به دنبال صفحاتی نیست که این کلمات را داشته باشند. به طور ضمنی، برای میلیون‌ها صفحه وب یک امتیاز احتمالی محاسبه می‌کند. صفحه‌ای که عبارت «رستوران ایتالیایی» دارد، نظرات مثبت دریافت کرده، در نقشه گوگل ثبت شده و اکنون باز است، امتیاز P(R=1|d,q) بسیار بالاتری نسبت به صفحه‌ای خواهد داشت که فقط یک بار در یک مقاله گردشگری به کلمه «ایتالیایی» اشاره کرده است. این PRP در عمل است که نتایج را هوشمندانه مرتب می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازخورد مرتبط بودن، تکامل یافته و خودکار:</b> تصور کنید کاربری عبارت «یادگیری ماشین» را جستجو می‌کند. نتایج اولیه عمومی هستند. کاربر روی لینکی کلیک می‌کند که به «یادگیری عمیق برای بینایی کامپیوتر» اشاره دارد. سیستم این کلیک (بازخورد ضمنی) را ثبت می‌کند. در جستجوهای بعدی یا حتی در همان جلسه، سیستم ممکن است به طور خودکار نتایج بیشتری در حوزه بینایی کامپیوتر نمایش دهد. این نسخه مدرن و خودکارized حلقه بازخورد احتمالی است؛ سیستمی که در حال یادگیری از رفتار کاربر است.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 برتری BM25 بر مدل‌های ساده‌تر:</b> فرض کنید در حال جستجوی یک مقاله علمی طولانی هستید. یک مدل باینری (مانند BIM) سندی که یک بار عبارت «مدل احتمالی» را آورده است، با سندی که ۵۰ بار در جزئیات به آن پرداخته است، برابر در نظر می‌گیرد. اما BM25 به سند دوم امتیاز بسیار بالاتری می‌دهد، زیرا <b>فراوانت کلمه (tf)</b> را در نظر می‌گیرد. همچنین BM25 با نرمال‌سازی طول سند، درک می‌کند که یک کتاب ۵۰۰ صفحه‌ای که یک بار کلمه را ذکر کرده، به اندازه یک مقاله ۵ صفحه‌ای که ۱۰ بار به آن اشاره کرده، «متمرکز» بر آن موضوع نیست. این هوشمندی BM25 را بسیار همسو با شهود انسانی و مؤثرتر می‌کند.
</p>

</div>

<p style="font-size:16pt; text-align:right;">
این فصل نشان داد که چگونه ریاضیات احتمالات می‌تواند به یک ابزار عملی و قدرتمند برای درک و بهبود بازیابی اطلاعات تبدیل شود. این مفاهیم فقط یک فصل از یک کتاب نیستند، بلکه ستون فقرات بسیاری از نوآوری‌های امروزی هستند. در فصل‌های بعدی، خواهیم دید که این اصول چگونه به مدل‌های زبانی احتمالی (فصل ۱۲) و سپس به تکنیک‌های یادگیری ماشین برای رتبه‌بندی (Learning to Rank) منجر شدند؛ مسیری که ما را به سمت ساخت سیستم‌های جستجویی هوشمندتر می‌برد که نه تنها کلمات را می‌فهمند، بلکه به نیاز پنهان و نیت کاربر نیز نزدیک‌تر می‌شوند.
</p>
