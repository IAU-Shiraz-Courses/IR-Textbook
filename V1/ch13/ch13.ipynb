<p style="font-size:18pt; text-align:center;">
<span id="section13.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">13 طبقه‌بندی متن و بیز ساده: از جستجوی گوگل تا فیلتر اسپم</span> 📍
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 از جستجوی یک‌باره تا پیگیری هوشمند</p>
<p style="font-size:16pt; text-align:right;">
تاکنون، این جزوه عمدتاً فرآیند بازیابی ادهاک (ad hoc retrieval) را بررسی کرده است - جایی که کاربران نیازهای اطلاعاتی موقتی دارند که با ارسال پرسمان به موتور جستجو تلاش می‌کنند به آن‌ها پاسخ دهند. اما بسیاری از کاربران نیازهای اطلاعاتی مداوم دارند.</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید بخواهید تحولات در حوزه تراشه‌های کامپیوتری چند هسته‌ای را دنبال کنید. یک راه این است که هر روز صبح پرسمان multicore AND computer AND chip را در شاخص مقالات خبری جدید جستجو کنید. اما این کار بسیار زمان‌بر است! گوگل نیوز چگونه این کار را برای شما خودکار می‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
در این فصل و دو فصل بعدی، به این سؤال می‌پردازیم: چگونه می‌توان این کار تکراری را خودکارسازی کرد؟ بسیاری از سیستم‌ها از <strong>پرسمان‌های ثابت</strong> (standing queries) پشتیبانی می‌کنند - پرسمان‌هایی که به صورت دوره‌ای روی مجموعه‌ای اجرا می‌شوند که اسناد جدید به تدریج به آن اضافه می‌شوند.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📧 چرا ایمیل‌های اسپم به پوشه اسپم می‌روند؟</p>
<p style="font-size:16pt; text-align:right;">
یک چالش مهم با پرسمان‌های ثابت این است که اگر پرسمان شما فقط multicore AND computer AND chip باشد، بسیاری از مقالات مرتبط را که از اصطلاحات دیگری مانند multicore processors استفاده می‌کنند، از دست خواهید داد. برای دستیابی به بازخوانی (recall) خوب، پرسمان‌های ثابت باید در طول زمان اصلاح شوند و می‌توانند به تدریج بسیار پیچیده شوند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
(multicore OR multi-core) AND (chip OR processor OR microprocessor)
</div>

<p style="font-size:16pt; text-align:right;">
برای درک گستردگی این مسئله، اکنون مفهوم کلی <strong>طبقه‌بندی</strong> (classification) را معرفی می‌کنیم. با توجه به مجموعه‌ای از کلاس‌ها، ما به دنبال تعیین این هستیم که یک شیء معین به کدام کلاس(ها) تعلق دارد. در مثال بالا، پرسمان ثابت برای تقسیم مقالات خبری جدید به دو کلاس عمل می‌کند: اسناد مربوط به تراشه‌های کامپیوتری چند هسته‌ای و اسناد غیرمرتبط با تراشه‌های کامپیوتری چند هسته‌ای. ما به این <strong>طبقه‌بندی دو کلاسی</strong> می‌گوییم. طبقه‌بندی با استفاده از پرسمان‌های ثابت همچنین <strong>مسیریابی</strong> (routing) یا <strong>فیلتر کردن</strong> (filtering) نامیده می‌شود.</p>
<p style="font-size:16pt; text-align:right;">
یک کلاس لزوماً نباید به اندازه پرسمان ثابت multicore computer chips محدود باشد. اغلب، یک کلاس یک حوزه موضوعی عمومی‌تر مانند چین یا قهوه است. چنین کلاس‌های عمومی‌تری معمولاً <strong>موضوعات</strong> (topics) نامیده می‌شوند و وظیفه طبقه‌بندی متن سپس <strong>طبقه‌بندی متن</strong> (text classification)، <strong>دسته‌بندی متن</strong> (text categorization)، <strong>طبقه‌بندی موضوع</strong> (topic classification) یا <strong>شناسایی موضوع</strong> (topic spotting) نامیده می‌شود.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🌐 کاربردهای واقعی طبقه‌بندی متن در دنیای دیجیتال</p>
<p style="font-size:16pt; text-align:right;">
مفهوم طبقه‌بندی بسیار کلی است و کاربردهای زیادی در داخل و خارج از بازیابی اطلاعات (IR) دارد. ما در اینجا بر مثال‌هایی از بازیابی اطلاعات تمرکز می‌کنیم:</p>

<p style="font-size:16pt; text-align:right;">
این فهرست اهمیت کلی طبقه‌بندی در IR را نشان می‌دهد. اکثر سیستم‌های بازیابی امروز حاوی چندین جزء هستند که به شکل طبقه‌بندی استفاده می‌کنند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🤖 سه رویکرد برای طبقه‌بندی متن</p>
<p style="font-size:16pt; text-align:right;">
برای پیاده‌سازی طبقه‌بندی متن، سه رویکرد اصلی وجود دارد:</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>طبقه‌بندی دستی</strong>: کتاب‌ها در یک کتابخانه توسط کتابدار به دسته‌بندی‌های کتابخانه کنگره اختصاص داده می‌شوند. این روش برای مقیاس‌بندی گران است.</li>
<li><strong>قوانین دستی</strong>: استفاده از پرسمان‌های ثابت که می‌توان به عنوان <strong>قوانین</strong> در نظر گرفت - که معمولاً به صورت دستی نوشته می‌شوند. مانند مثال ما<span dir='ltr'> (multicore OR multi-core) AND (chip OR processor OR microprocessor)</span>، قوانین گاهی معادل عبارات بولی هستند. یک قانون ترکیبی خاصی از کلمات کلیدی را که نشان‌دهنده یک کلاس است، ثبت می‌کند. قوانین کدگذاری شده دستی خواص مقیاس‌بندی خوبی دارند، اما ایجاد و نگهداری آن‌ها در طول زمان کار فشرده‌ای است.</li>
<li><strong>یادگیری ماشین</strong>: در این رویکرد، مجموعه‌ای از قوانین یا به طور کلی، معیار تصمیم‌گیری طبقه‌بند متن، به طور خودکار از داده‌های آموزشی یاد گرفته می‌شود. این رویکرد همچنین <strong>طبقه‌بندی آماری متن</strong> نامیده می‌شود اگر روش یادگیری آماری باشد.</li>
</ol>

<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی آماری متن، ما به تعدادی اسناد مثال خوب (یا اسناد آموزشی) برای هر کلاس نیاز داریم. نیاز به طبقه‌بندی دستی حذف نمی‌شود زیرا اسناد آموزشی از شخصی که آن‌ها را برچسب‌گذاری کرده است، می‌آیند - جایی که برچسب‌گذاری به فرآیند حاشیه‌نویسی هر سند با کلاس آن اشاره دارد. اما برچسب‌گذاری به طور قابل استدلالی وظیفه آسان‌تری از نوشتن قوانین است. تقریباً هر کسی می‌تواند به یک سند نگاه کند و تصمیم بگیرد که آیا به چین مرتبط است یا نه.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این فصل، ما با یک مقدمه کلی برای مسئله طبقه‌بندی متن آغاز می‌کنیم؛ سپس بیز ساده (Naive Bayes)، یک روش طبقه‌بندی به خصوص ساده و مؤثر را پوشش می‌دهیم. تمام الگوریتم‌های طبقه‌بندی که ما مطالعه می‌کنیم، اسناد را در فضاهای با ابعاد بالا نمایش می‌دهند. برای بهبود کارایی این الگوریتم‌ها، معمولاً مطلوب است که ابعاد این فضاها را کاهش دهیم؛ به این منظور، تکنیکی به نام <strong>انتخاب ویژگی</strong> (feature selection) معمولاً در طبقه‌بندی متن اعمال می‌شود.</p>

<p style="font-size:16pt; text-align:right;">
در فصل‌های بعدی، به دو خانواده دیگر از روش‌های طبقه‌ بندی، طبقه ‌بندهای فضای برداری و ماشین‌های بردار پشتیبان نگاه خواهیم کرد.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section13.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.1 مسئله طبقه‌ بندی متن: از ایمیل‌های اسپم تا دسته ‌بندی اخبار</span> 📚
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📮 چگونه جیمیل تشخیص می‌دهد ایمیل اسپم است؟</p>
<p style="font-size:16pt; text-align:right;">
در طبقه بندی متن، ما یک توصیف d ∈ X از یک سند داریم، که در آن X <strong>فضای سند</strong> (document space) است؛ و مجموعه‌ای ثابت از کلاس‌ها C = {c1, c2, ..., cJ}. کلاس‌ها همچنین به عنوان <strong>دسته‌ها</strong> (categories) یا <strong>برچسب‌ها</strong> (labels) شناخته می‌شوند.</p>
<p style="font-size:16pt; text-align:right;">
به طور معمول، فضای سند X نوعی فضای با ابعاد بالا است، و کلاس‌ها توسط انسان برای نیازهای یک کاربرد تعریف می‌شوند. برای مثال، در سیستم جیمل، هر ایمیل یک سند است و دو کلاس اصلی وجود دارد: "اسپم" و "غیراسپم". جیمل باید بتواند هر ایمیل جدید را به یکی از این دو کلاس اختصاص دهد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
ما یک <strong>مجموعه آموزشی</strong> (training set) D از اسناد برچسب‌دار ⟨d, c⟩ داریم که در آن ⟨d, c⟩ ∈ X × C. برای مثال:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
⟨Beijing joins the World Trade Organization, China⟩
</div>

<p style="font-size:16pt; text-align:right;">
برای سند تک جمله‌ای "Beijing joins the World Trade Organization" و کلاس (یا برچسب) China.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧠 یادگیری نظارت‌ شده: چگونه نتفلیکس سلیقه شما را یاد می‌گیرد؟</p>
<p style="font-size:16pt; text-align:right;">
با استفاده از یک <strong>روش یادگیری</strong> (learning method) یا <strong>الگوریتم یادگیری</strong> (learning algorithm)، ما سپس می‌خواهیم یک <strong>طبقه‌ بند</strong> (classifier) یا <strong>تابع طبقه‌ بندی</strong> (classification function) γ یاد بگیریم که اسناد را به کلاس‌ها نگاشت می‌کند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
γ : X → C
</div>

<p style="font-size:16pt; text-align:right;">
این نوع یادگیری <strong>یادگیری نظارت‌ شده</strong> (supervised learning) نامیده می‌شود زیرا یک ناظر (انسانی که کلاس‌ها را تعریف می‌کند و اسناد آموزشی را برچسب‌گذاری می‌کند) به عنوان یک معلم فرآیند یادگیری را هدایت می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
نتفلیکس از همین روش استفاده می‌کند: وقتی شما به فیلمی امتیاز می‌دهید، در واقع در حال برچسب‌گذاری داده‌های آموزشی هستید. الگوریتم نتفلیکس (روش یادگیری Γ) از این داده‌ها برای یادگیری تابع طبقه‌ بندی (γ) استفاده می‌کند تا بتواند پیش‌بینی کند کدام فیلم‌های دیگر را دوست خواهید داشت.</p>
<p style="font-size:16pt; text-align:right;">
ما روش یادگیری نظارت‌شده را با Γ نشان می‌دهیم و می‌نویسیم Γ(D) = γ. روش یادگیری Γ مجموعه آموزشی D را به عنوان ورودی می‌گیرد و تابع طبقه‌ بندی یادگرفته شده γ را برمی‌گرداند.</p>
</div>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig13.1.bmp" alt="Classes, training set, and test set in text classification." style="width:900px; border:1px solid #ccc; border-radius:12px;"
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📰 طبقه‌ بندی اخبار: از آموزش تا آزمون</p>
<p style="font-size:16pt; text-align:right;">
شکل 13.1 یک مثال از طبقه‌ بندی متن از مجموعه Reuters-RCV1 را نشان می‌دهد. شش کلاس (UK, China, ..., sports) وجود دارد، هر کدام با سه سند آموزشی. ما چند کلمه یادآوری برای محتوای هر سند نشان می‌دهیم. مجموعه آموزشی برخی مثال‌های معمولی برای هر کلاس فراهم می‌کند، به طوری که ما می‌توانیم تابع طبقه‌ بندی γ را یاد بگیریم.</p>
<p style="font-size:16pt; text-align:right;">
پس از یادگیری γ، ما می‌توانیم آن را به <strong>مجموعه آزمون</strong> (test set) (یا داده‌های آزمون) اعمال کنیم، برای مثال، سند جدید "first private Chinese airline" که کلاس آن ناشناخته است.</p>
<p style="font-size:16pt; text-align:right;">
در شکل 13.1، تابع طبقه ‌بندی سند جدید را به کلاس γ(d) = China اختصاص می‌دهد، که تخصیص صحیح است. این دقیقاً همان کاری است که خبرگزاری‌ها مانند گوگل نیوز با مقالات جدید انجام می‌دهند: آن‌ها با استفاده از مدل‌های آموزش‌دیده، مقالات جدید را به دسته‌بندی‌های مناسب اختصاص می‌دهند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🏷️ سلسله مراتب کلاس‌ها: از یک کلاس تا چندین کلاس</p>
<p style="font-size:16pt; text-align:right;">
کلاس‌ها در طبقه‌بندی متن اغلب دارای ساختار جالبی مانند سلسله مراتب در شکل 13.1 هستند. دو نمونه از هر یک از دسته‌بندی‌های منطقه، صنعت و حوزه موضوعی وجود دارد. یک سلسله مراتب می‌تواند یک کمک مهم در حل مسئله طبقه‌بندی باشد.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، در وب‌سایت آمازون، یک محصول می‌تواند هم در دسته "الکترونیک" و هم در دسته "لوازم خانگی" قرار گیرد. این سلسله مراتب به کاربران کمک می‌کند تا محصولات را بهتر پیدا کنند.</p>
<p style="font-size:16pt; text-align:right;">
تعریف (13.1) مقرر می‌دارد که یک سند عضو دقیقاً یک کلاس است. این مناسب‌ترین مدل برای سلسله مراتب در شکل 13.1 نیست. برای مثال، سندی در مورد المپیک 2008 باید عضو دو کلاس باشد: کلاس China و کلاس sports. این نوع مسئله طبقه ‌بندی به عنوان یک مسئله <strong>any-of</strong> نامیده می‌شود و ما به آن در بخش 14.5 بازمی‌گردیم. در حال حاضر، ما فقط مسائل <strong>one-of</strong> را در نظر می‌گیریم که در آن یک سند عضو دقیقاً یک کلاس است.</p>
<p style="font-size:16pt; text-align:right;">
یوتیوب از این مدل استفاده می‌کند: یک ویدیو می‌تواند هم به دسته "موسیقی" و هم به دسته "آموزشی" تعلق داشته باشد. این یک مسئله any-of است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
هدف ما در طبقه‌بندی متن، دقت بالا در داده‌های آزمون یا داده‌های جدید است - برای مثال، مقالات خبری که فردا صبح با آن‌ها مواجه خواهیم شد. دستیابی به دقت بالا در مجموعه آموزشی آسان است (مثلاً، ما می‌توانیم به سادگی برچسب‌ها را حفظ کنیم). اما دقت بالا در مجموعه آموزشی به طور کلی به این معنی نیست که طبقه‌بند در داده‌های جدید در یک کاربرد به خوبی کار خواهد کرد.</p>
<p style="font-size:16pt; text-align:right;">
وقتی ما از مجموعه آموزشی برای یادگیری یک طبقه‌بند برای داده‌های آزمون استفاده می‌کنیم، ما این فرض را می‌کنیم که داده‌های آموزشی و داده‌های آزمون شبیه هستند یا از همان توزیع هستند. این همان چیزی است که در دنیای واقعی اتفاق می‌افتد: الگوریتم‌های طبقه‌بندی فرض می‌کنند ایمیل‌های فردا شبیه به ایمیل‌های امروز خواهند بود.</p>

</div>



<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section13.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.2 طبقه ‌بندی متن با بیز ساده: چگونه جیمل ایمیل‌های اسپم را تشخیص می‌دهد؟</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 ایده اصلی: احتمالات در برابر کلمات</p>
<p style="font-size:16pt; text-align:right;">
مدل <strong>چندجمله‌ای بیز ساده</strong> (multinomial Naive Bayes) یک روش یادگیری احتمالی است که احتمال اینکه یک سند d در کلاس c باشد را محاسبه می‌کند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
P(c|d) ∝ P(c) ∏<sub>1≤k≤n<sub>d</sub></sub> P(t<sub>k</sub>|c)
</div>

<p style="font-size:16pt; text-align:right;">
بیایید این را با یک مثال واقعی بررسی کنیم: وقتی جیمل ایمیل شما را بررسی می‌کند، در واقع در حال محاسبه احتمال اینکه ایمیل "اسپم" است یا "غیراسپم" است. P(c) احتمال اولیه اینکه یک ایمیل به طور کلی اسپم باشد (مثلاً 30٪ ایمیل‌ها اسپم هستند). P(t<sub>k</sub>|c) احتمال اینکه کلمه‌ای مثل "ارزان" یا "کاهش وزن" در ایمیل‌های اسپم ظاهر شود.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🏆 بهترین کلاس: بیشترین احتمال پسین</p>
<p style="font-size:16pt; text-align:right;">
بهترین کلاس در طبقه ‌بندی NB، کلاس <strong>بیشترین احتمال پسین</strong> (maximum a posteriori یا MAP) است:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
c<sub>map</sub> = arg max<sub>c∈C</sub> P̂(c|d) = arg max<sub>c∈C</sub> P̂(c) ∏<sub>1≤k≤n<sub>d</sub></sub> P̂(t<sub>k</sub>|c)
</div>

<p style="font-size:16pt; text-align:right;">
در عمل، برای جلوگیری از سرریز اعشار، محاسبه با افزودن لگاریتم احتمالات انجام می‌شود:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
c<sub>map</sub> = arg max<sub>c∈C</sub> [log P̂(c) + ∑<sub>1≤k≤n<sub>d</sub></sub> log P̂(t<sub>k</sub>|c)]
</div>

<p style="font-size:16pt; text-align:right;">
هر پارامتر شرطی log P̂(t<sub>k</sub>|c) یک وزن است که نشان می‌دهد t<sub>k</sub> چقدر یک شاخص خوب برای c است. در سیستم فیلتر اسپم جیمل، کلماتی مانند "ارزان"، "کاهش وزن" و "کلیک کنید" وزن بالایی برای کلاس اسپم دارند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 تخمین پارامترها: از داده‌های واقعی به احتمالات</p>
<p style="font-size:16pt; text-align:right;">
پارامترها با استفاده از <strong>تخمین بیشترین درست نمایی</strong> (MLE) تخمین زده می‌شوند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
P̂(c) = N<sub>c</sub>/N<br>
P̂(t|c) = T<sub>ct</sub>/∑<sub>t′∈V</sub> T<sub>ct′</sub>
</div>

<p style="font-size:16pt; text-align:right;">
که در آن N<sub>c</sub> تعداد اسناد در کلاس c و T<sub>ct</sub> تعداد وقوع‌های t در اسناد آموزشی از کلاس c است.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚡ مشکل صفر و راه‌حل: هموارسازی یک افزودن</p>
<p style="font-size:16pt; text-align:right;">
مشکل با تخمین MLE این است که برای ترکیب واژه-کلاس که در داده‌های آموزشی رخ نداده است، صفر است. برای مثال، اگر کلمه "ویژه" فقط در ایمیل‌های غیراسپم ظاهر شده باشد، مدل MLE احتمال وقوع آن در ایمیل‌های اسپم را صفر در نظر می‌گیرد.</p>
<p style="font-size:16pt; text-align:right;">
برای حذف صفرها، از <strong>هموارسازی یک افزودن</strong> (add-one smoothing) استفاده می‌کنیم:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f0f0f0; padding:10px; border-radius:5px;">
P̂(t|c) = (T<sub>ct</sub> + 1)/(∑<sub>t′∈V</sub>(T<sub>ct′</sub>) + 1)
</div>

<p style="font-size:16pt; text-align:right;">
این تکنیک ساده اما مؤثر به جیمل اجازه می‌دهد حتی با کلماتی که قبلاً در ایمیل‌های اسپم دیده نشده، تصمیم‌گیری منطقی انجام دهد.</p>
</div>

<div style="background:#eef; padding:15px; margin:20px 0; border-radius:8px; text-align:right;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#2c3e50;">🚀 الگوریتم کامل: از آموزش تا طبقه‌بندی</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم کامل برای آموزش و اعمال طبقه‌بند NB:</p>

</div>

<p style="font-size:16pt; text-align:right;">
✎ مثال واقعی: برای طبقه‌بندی یک ایمیل با کلمات "چینی، توکیو، WTO"، سیستم بیز ساده ابتدا احتمال اولیه هر کلاس (اسپم در مقابل غیراسپم) را محاسبه کرده، سپس با توجه به فراوانی این کلمات در هر کلاس، احتمال نهایی را تخمین می‌زند.</p>

NB پیچیدگی زمانی بهینه دارد که یکی از دلایل محبوبیت آن به عنوان روش طبقه‌بندی متن است. همین الگوریتم ساده اما قدرتمند به جیمل اجازه می‌دهد میلیاردها ایمیل را در لحظه‌هایی با دقت بالای 99٪ طبقه‌بندی کند.</p>

</div>

```python
import math
from collections import defaultdict, Counter

class MultinomialNB:
    """
    پیاده‌سازی طبقه‌بند Naive Bayes چندجمله‌ای (Multinomial Naive Bayes)
    """
    def __init__(self):
        self.classes = []
        self.vocab = set()
        self.prior = {}
        self.cond_prob = defaultdict(dict)

    def train(self, documents, labels):
        """
        آموزش طبقه‌بند با استفاده از اسناد و برچسب‌های مربوطه
        documents: لیستی از اسناد، هر سند یک لیست از کلمات است
        labels: لیستی از برچسب‌های مربوط به هر سند
        """
        # استخراج کلاس‌ها و واژگان
        self.classes = list(set(labels))
        for doc in documents:
            self.vocab.update(doc)
        V = len(self.vocab)  # اندازه واژگان

        # محاسبه احتمال‌های اولیه (P̂(c))
        N = len(documents)
        for c in self.classes:
            N_c = sum(1 for label in labels if label == c)
            self.prior[c] = N_c / N

        # محاسبه احتمالات شرطی با هموارسازی یک افزودنی (Laplace Smoothing)
        # فرمول: P̂(t|c) = (T_ct + 1) / (sum(T_ct') + V)
        for c in self.classes:
            # الحاق تمام اسناد از کلاس c
            class_docs = [doc for doc, label in zip(documents, labels) if label == c]
            
            # استفاده از Counter برای شمارش کارآمدتر توکن‌ها در کلاس c
            text_c_counter = Counter(word for doc in class_docs for word in doc)
            
            # شمارش کل توکن‌ها در اسناد کلاس c (مجموع T_ct')
            total_tokens_in_c = sum(text_c_counter.values())
            
            # محاسبه احتمال شرطی برای هر واژه در واژگان
            for t in self.vocab:
                T_ct = text_c_counter.get(t, 0)  # تعداد تکرار واژه t در کلاس c
                # استفاده از هموارسازی یک افزودنی
                self.cond_prob[t][c] = (T_ct + 1) / (total_tokens_in_c + V)

    def predict(self, document):
        """
        پیش‌بینی کلاس یک سند جدید
        document: سند جدید، یک لیست از کلمات
        """
        scores = {}

        # محاسبه امتیاز (log-probability) هر کلاس
        for c in self.classes:
            # شروع با لگاریتم احتمال اولیه
            score = math.log(self.prior[c])
            
            # افزودن لگاریتم احتمالات شرطی برای هر واژه در سند
            for word in document:
                if word in self.vocab: # فقط واژه‌هایی که در واژگان آموزشی هستند را در نظر می‌گیریم
                    score += math.log(self.cond_prob[word][c])
            scores[c] = score

        # برای اهداف آموزشی، امتیازات نهایی را چاپ می‌کنیم
        print("\nامتیازات نهایی (log-probabilities):")
        for c, score in scores.items():
            print(f"  Score({c}): {score:.4f}")
            
        # بازگرداندن کلاس با بیشترین امتیاز
        return max(scores.items(), key=lambda x: x[1])[0]

# مثال از جدول 13.1 کتاب (Introduction to Information Retrieval)
# داده‌های آموزشی
train_docs = [
    ["Chinese", "Beijing", "Chinese"],
    ["Chinese", "Chinese", "Shanghai"],
    ["Chinese", "Macao"],
    ["Tokyo", "Japan", "Chinese"]
]
train_labels = ["China", "China", "China", "not China"]

# داده‌های آزمون
test_doc = ["Chinese", "Chinese", "Chinese", "Tokyo", "Japan"]

# ایجاد و آموزش طبقه‌بند
nb = MultinomialNB()
nb.train(train_docs, train_labels)

# پیش‌بینی کلاس سند آزمون
predicted_class = nb.predict(test_doc)
print(f"\nسند آزمون به کلاس '{predicted_class}' اختصاص داده شد.")

# چاپ احتمالات شرطی محاسبه شده برای واژه‌های کلیدی
print("\nاحتمالات شرطی محاسبه شده (با هموارسازی):")
for word in ["Chinese", "Tokyo", "Japan"]:
    print(f"P({word}|China) = {nb.cond_prob[word]['China']:.4f}")
    print(f"P({word}|not China) = {nb.cond_prob[word]['not China']:.4f}")
```

<span id="section13.3" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.3 مدل برنولی: برای طبقه‌بندی ایمیل‌های کوتاه کدام بهتر است ؟</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📧 دو رویکرد متفاوت برای طبقه‌بندی</p>
<p style="font-size:16pt; text-align:right;">
تا کنون دو مدل بیز ساده را بررسی کردیم: مدل <strong>چندجمله‌ای</strong> که تعداد وقوع کلمات را در نظر می‌گیرد و مدل <strong>برنولی</strong> که فقط وجود یا عدم وجود کلمات را بررسی می‌کند. بیایید این تفاوت را با یک مثال واقعی بررسی کنیم:</p>

<p style="font-size:16pt; text-align:right;">
فرض کنید در حال توسعه یک سیستم فیلتر اسپم برای پیام‌رسان‌های کوتاه هستید. در این سیستم، مهم است که بدانیم آیا پیام حاوی کلمات کلیدی مانند "کاهش وزن" یا "رایگان" است یا خیر. در این حالت، مدل برنولی انتخاب بهتری است، زیرا:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>در پیام‌های کوتاه، هر کلمه معمولاً فقط یک بار ظاهر می‌شود</li>
<li>وجود یا عدم وجود کلمه، اطلاعات بیشتری از تعداد تکرار آن ارائه می‌دهد</li>
<li>حجم محاسبات کمتری نیاز دارد (فقط نیاز به بررسی وجود کلمات)</li>
</ul>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔄 تفاوت در نحوه محاسبه احتمالات</p>
<p style="font-size:16pt; text-align:right;">
تفاوت اصلی بین دو مدل در نحوه محاسبه احتمالات شرطی P̂(t|c) است:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مدل چندجمله‌ای:</strong> P̂(t|c) به عنوان کسری از توکن‌ها یا موقعیت‌ها در اسناد کلاس c تخمین زده می‌شود.</li>
<li><strong>مدل برنولی:</strong> P̂(t|c) به عنوان کسری از اسناد کلاس c که شامل واژه t هستند تخمین زده می‌شود.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
این تفاوت به نتایج متفاوتی منجر می‌شود. برای مثال، اگر کلمه "چین" در یک سند ۵ بار تکرار شده باشد، مدل چندجمله‌ای این را یک شاخص بسیار قوی برای کلاس "چین" در نظر می‌گیرد، در حالی که مدل برنولی فقط وجود آن را بررسی می‌کند.</p>
</div>

نکتهای مهم در الگوریتم برنولی:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>در خط ۸، از هموارسازی یک افزودن با B=2 استفاده شده (دو حالت: وجود یا عدم وجود واژه)</li>
<li>در خط ۷، احتمال عدم وقوع واژه نیز در محاسبه نهایی لحاظ می‌شود (خط ۵-۷)</li>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🎯 مثال عملی: طبقه‌بندی پیام‌های کوتاه</p>
<p style="font-size:16pt; text-align:right;">
بیایید مثال 13.2 را با یک سناریوی واقعی بررسی کنیم. فرض کنید در حال توسعه یک سیستم طبقه‌بندی برای پیام‌های کوتاه هستید و باید تصمیم بگیرید که آیا پیام‌های دریافتی از طرف "بانک" هستند یا "تبلیغات" است.</p>

<p style="font-size:16pt; text-align:right;">
در این حالت، مدل برنولی انتخاب بهتری است زیرا:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>پیام‌های کوتاه معمولاً شامل کلمات کلیدی مشخصی هستند (مانند "مانده حساب"، "تراکنش")</li>
<li>وجود یا عدم وجود این کلمات، اطلاعات بیشتری از تعداد تکرار آن‌ها ارائه می‌دهد</li>
<li>برای طبقه‌بندی سریع، مدل برنولی نیاز به محاسبات کمتری دارد</li>
</ol>

<p style="font-size:16pt; text-align:right;">
با اعمال مدل برنولی بر داده‌های آموزشی و پیش‌بینی پیام جدید "کاهش وزن تراکنش"، سیستم به درستی آن را به کلاس "بانک" اختصاص می‌دهد، زیرا کلمات کلیدی بانک وجود دارند، در حالی که کلمات تبلیغاتی وجود ندارند.</p>
</div>

</div>

```python
import math
from collections import defaultdict

class BernoulliNB:
    """
    پیاده‌سازی طبقه‌بند برنولی بیز (Bernoulli Naive Bayes)
    """
    def __init__(self):
        self.classes = []
        self.vocab = set()
        self.prior = {}
        self.cond_prob = defaultdict(dict)

    def train(self, documents, labels):
        """
        آموزش طبقه‌بند برنولی با استفاده از اسناد و برچسب‌های مربوطه
        documents: لیستی از اسناد، هر سند یک لیست از کلمات است
        labels: لیستی از برچسب‌های مربوط به هر سند
        """
        # استخراج کلاس‌ها و واژگان
        self.classes = list(set(labels))
        for doc in documents:
            self.vocab.update(doc)
        V = len(self.vocab)

        # محاسبه احتمال‌های اولیه (P̂(c))
        N = len(documents)
        for c in self.classes:
            N_c = sum(1 for label in labels if label == c)
            self.prior[c] = N_c / N

        # محاسبه احتمالات شرطی با هموارسازی یک افزودنی (P̂(t|c))
        # فرمول: P̂(t|c) = (N_ct + 1) / (N_c + 2)
        for c in self.classes:
            N_c = sum(1 for label in labels if label == c)
            for t in self.vocab:
                # شمارش اسنادی که واژه t در آن کلاس c وجود دارد (N_ct)
                docs_with_term = sum(1 for doc, label in zip(documents, labels)
                                     if label == c and t in doc)
                # استفاده از هموارسازی یک افزودن با B=2 (دو حالت: وجود یا عدم وجود)
                self.cond_prob[t][c] = (docs_with_term + 1) / (N_c + 2)

    def predict(self, document):
        """
        پیش‌بینی کلاس یک سند جدید با مدل برنولی
        document: سند جدید، یک لیست از کلمات
        """
        scores = {}
        doc_set = set(document) # برای بررسی سریع‌تر وجود واژه

        # محاسبه امتیاز هر کلاس
        for c in self.classes:
            # شروع با لگاریتم احتمال اولیه
            score = math.log(self.prior[c])

            # برای تمام واژگان، احتمال وجود یا عدم وجود را در نظر می‌گیریم
            for t in self.vocab:
                if t in doc_set:
                    # اگر واژه در سند وجود دارد، از P(t|c) استفاده کن
                    score += math.log(self.cond_prob[t][c])
                else:
                    # اگر واژه در سند وجود ندارد، از 1 - P(t|c) استفاده کن
                    score += math.log(1 - self.cond_prob[t][c])

            
            scores[c] = score
            
        return scores


try:
    # پیام برای اطمینان از شروع اجرا
    print("شروع اجرای برنامه...")

    # مثال: طبقه‌بندی پیام‌های کوتاه به "بانکی" یا "تبلیغات"
    train_docs = [
        ["موجودی", "حساب", "شما", "مثبت", "است"],
        ["کاهش", "وزن", "تراکنش", "یافت", "شد"],
        ["جشن", "تخفیف", "فروش", "ویژه", "است"],
        ["برنده", "خوش", "شانس", "میلیون", "تومان"]
    ]
    train_labels = ["بانک", "بانک", "تبلیغات", "تبلیغات"]
    test_doc = ["کاهش", "وزن", "تراکنش", "شما"]

    # ایجاد و آموزش طبقه‌بند
    nb = BernoulliNB()
    nb.train(train_docs, train_labels)
    print("آموزش مدل با موفقیت انجام شد.")

    # پیش‌بینی کلاس سند آزمون
    scores = nb.predict(test_doc)
    predicted_class = max(scores.items(), key=lambda x: x[1])[0]
    
    # نمایش نتایج به صورت مستقیم و واضح
    print("\n--- نتایج ---")
    print("امتیازات نهایی (log-probabilities):")
    for c, score in scores.items():
        print(f"  Score({c}): {score:.4f}")
        
    print(f"\nپیام به کلاس '{predicted_class}' اختصاص داده شد.")

    # نمایش احتمالات شرطی برای کلمات کلیدی
    print("\nاحتمالات شرطی محاسبه شده (با هموارسازی):")
    for word in ["کاهش", "وزن", "تراکنش", "جشن", "برنده"]:
        print(f"P({word}|بانک) = {nb.cond_prob[word]['بانک']:.4f}")
        print(f"P({word}|تبلیغات) = {nb.cond_prob[word]['تبلیغات']:.4f}")
        
    print("\nاجرای برنامه به پایان رسید.")

except Exception as e:
    print(f"\nیک خطا رخ داد: {e}")
```

<p style="font-size:18pt; text-align:center;">
<span id="section13.4" style="font-size:24pt; font-weight:bold; color:#1B263B; text-align:center;">
13.4 ویژگی‌های بیز ساده: چرا روشی "ساده‌انگارانه" آنقدر قدرتمند است؟
</span>
</p>

<p style="font-size:16pt; text-align:right;">
بیز ساده بر اساس قانون بیز عمل می‌کند: سندی به کلاسی اختصاص می‌یابد که بیشترین احتمال پسین را داشته باشد:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
c<sub>map</sub> = arg max<sub>c∈C</sub> P(c|d) = arg max<sub>c∈C</sub> P(d|c)P(c)
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
در اینجا P(c|d) احتمال اینکه سند d به کلاس c تعلق دارد را نشان می‌دهد. با استفاده از قانون بیز، این احتمال به P(d|c)P(c) تبدیل می‌شود که مخرج آن (P(d)) برای همه کلاس‌ها یکسان است و می‌توان آن را نادیده گرفت.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🎯 دو رویکرد متفاوت برای مدل‌سازی اسناد</p>
<p style="font-size:16pt; text-align:right;">
بیز ساده دو مدل اصلی دارد که نحوه تولید سند را به شکلی متفاوت مدل می‌کنند:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
چندجمله‌ای: P(d|c) = P(h<sub>t<sub>1</sub>, ..., t<sub>k</sub>, ..., t<sub>n<sub>d</sub></sub>|c)<br>
برنولی: P(d|c) = P(h<sub>e<sub>1</sub>, ...,e<sub>i</sub>, ...,e<sub>M</sub></sub>|c)
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
در مدل چندجمله‌ای، ما به دنبال دنباله واژه‌ها در سند هستیم، در حالی که در مدل برنولی، فقط به وجود یا عدم وجود هر واژه توجه می‌کنیم.</p>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> برای طبقه‌بندی نظرات مشتریان آمازون، مدل چندجمله‌ای به تعداد تکرار واژه "عالی" در یک نظر توجه می‌کند، در حالی که مدل برنولی فقط بررسی می‌کند که آیا واژه "عالی" در نظر وجود دارد یا خیر.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🧩 فرض استقلال شرطی: ساده‌سازی هوشمندانه</p>
<p style="font-size:16pt; text-align:right;">
برای کاهش تعداد پارامترها، بیز ساده فرض می‌کند که ویژگی‌ها با توجه به کلاس از یکدیگر مستقل هستند:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
چندجمله‌ای: P(d|c) = ∏<sub>1≤k≤n<sub>d</sub></sub> P(X<sub>k</sub> = t<sub>k</sub>|c)<br>
برنولی: P(d|c) = ∏<sub>1≤i≤M</sub> P(U<sub>i</sub> = e<sub>i</sub>|c)
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
این فرض به ما اجازه می‌دهد تا به جای تخمین 2<sup>M</sup>|C| پارامتر، تنها Θ(M|C|) پارامتر را تخمین بزنیم، که در عمل بسیار کارآمدتر است.</p>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در طبقه‌بندی اخبار رویترز، با وجود اینکه واژه‌های "نیویورک" و "بورس" اغلب با هم می‌آیند، بیز ساده فرض می‌کند این دو واژه با توجه به کلاس "اقتصاد" از یکدیگر مستقل هستند. با وجود این فرض نادرست، الگوریتم همچنان به خوبی عمل می‌کند!</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📍 استقلال موقعیتی: نادیده گرفتن ترتیب واژه‌ها</p>
<p style="font-size:16pt; text-align:right;">
در مدل چندجمله‌ای، فرض دیگری به نام استقلال موقعیتی نیز وجود دارد: احتمال وقوع یک واژه در سند به موقعیت آن بستگی ندارد. این فرض معادل مدل bag of words است که ترتیب واژه‌ها را نادیده می‌گیرد.</p>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> برای سیستم طبقه‌بندی اسناد داخلی مایکروسافت، عبارت "مدیریت پروژه" و "پروژه مدیریت" به یک شکل در نظر گرفته می‌شوند، زیرا ترتیب واژه‌ها در مدل بیز ساده چندجمله‌ای نادیده گرفته می‌شود.</p>
</div>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🤔 چرا بیز ساده با وجود فرضیات اشتباه خوب کار می‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
بیز ساده به این نام خوانده می‌شود زیرا فرضیات استقلال آن برای مدل زبان طبیعی بسیار ساده‌انگارانه است. با این حال، تخمین‌های احتمالی بیز ساده کیفیت پایینی دارند، اما تصمیمات طبقه‌بندی آن به طرز شگفت‌آوری خوب هستند.</p>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان شرکت فیسبوک مارکت، یک نظر حاوی واژه‌های "عالی"، "خوب" و "پیشنهاد می‌کنم" با احتمال واقعی 0.6 به کلاس "مثبت" تعلق دارد. بیز ساده این احتمال را به اشتباه 0.99 تخمین می‌زند، اما همچنان نظر را به درستی به کلاس "مثبت" اختصاص می‌دهد.</p>

<p style="font-size:16pt; text-align:right;">
این پدیده به این دلیل رخ می‌دهد که در طبقه‌بندی بیز ساده، کلاس برنده معمولاً احتمال بسیار بزرگتری نسبت به کلاس‌های دیگر دارد، حتی اگر این احتمال به اشتباه تخمین زده شده باشد. تصمیم طبقه‌بندی بر اساس کلاسی است که بیشترین امتیاز را کسب می‌کند، نه بر اساس دقت تخمین احتمال.</p>

<div style="background:#d4edda; border-right:4px solid #155724; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#155724;">💡 مزایای بیز ساده در دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
بیز ساده دارای مزایای زیادی است که آن را به یک ابزار قدرتمند در صنعت تبدیل کرده است:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>در شرایطی که بسیاری از ویژگی‌های مهم مشترکاً به تصمیم طبقه‌بندی کمک می‌کنند، عالی عمل می‌کند. <b>مثال:</b> در سیستم طبقه‌بندی اسناد حقوقی شرکت اپل، واژه‌های متعددی مانند "قرارداد"، "حقوقی"، "توافق" و "قانون" با هم به تصمیم‌گیری کمک می‌کنند.</li>
<li>تا حدی در برابر ویژگی‌های نویز و تغییر مفهوم (concept drift) مقاوم است. <b>مثال:</b> سیستم فیلتر اسپم یاهو با گذشت زمان و تغییر تاکتیک‌های اسپمرها، همچنان به خوبی عمل می‌کند.</li>
<li>کارایی بالایی دارد: آموزش و طبقه‌بندی با یک عبور از داده‌ها قابل انجام است. <b>مثال:</b> گوگل می‌تواند میلیاردها وب‌صفحه را با استفاده از بیز ساده در زمان کوتاهی طبقه‌بندی کند.</li>
<li>اغلب به عنوان یک خط مبنا در تحقیقات طبقه‌بندی متن استفاده می‌شود. <b>مثال:</b> محققان آمازون قبل از آزمایش مدل‌های پیچیده، ابتدا عملکرد بیز ساده را به عنوان خط مبنا ارزیابی می‌کنند.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
در حالی که بیز ساده ممکن است دقیق‌ترین روش برای طبقه‌بندی متن نباشد، اما ترکیبی از کارایی و دقت خوب آن را به یک گزینه قوی برای بسیاری از کاربردهای تجاری تبدیل کرده است، از فیلتر کردن اسپم گرفته تا طبقه‌بندی اخبار و تحلیل احساسات مشتریان.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section13.4.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.4.1 یک نوع دیگر از مدل چندجمله‌ای</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 گوگل و دسته‌بندی ایمیل‌ها: چرا فراوانی کلمات مهم است؟</p>
<p style="font-size:16pt; text-align:right;">
وقتی گوگل ایمیل شما را دسته‌بندی می‌کند، به فراوانی کلمات در متن توجه می‌کند، نه فقط به وجود یا عدم وجود آن‌ها. ایمیل‌های اسپم معمولاً کلماتی مانند "ارزان"، "تخفیف"، "فوری" را چندین بار تکرار می‌کنند. این دقیقاً همان چیزی است که مدل چندجمله‌ای به ما می‌دهد - توانایی در نظر گرفتن تعداد تکرار هر کلمه.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این نوع از مدل چندجمله‌ای، هر سند d به صورت یک بردار M بعدی از فراوانی‌ها h<sub>tf<sub>t<sub>1</sub>,d</sub>, ..., tf<sub>t<sub>M</sub>,d</sub></sub> نمایش داده می‌شود که در آن tf<sub>t<sub>i</sub>,d</sub> تعداد دفعات تکرار واژه t<sub>i</sub> در سند d است.</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
P(d|c) = P(h<sub>tf<sub>t<sub>1</sub>,d</sub>, ..., tf<sub>t<sub>M</sub>,d</sub></sub>|c) ∝ ∏<sub>1≤i≤M</sub> P(X = t<sub>i</sub>|c)<sup>tf<sub>t<sub>i</sub>,d</sub></sup>
  </div>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🎯 تفاوت کلیدی: توان فراوانی</p>
<p style="font-size:16pt; text-align:right;">
نکته جالب در این فرمول، توان <sup>tf<sub>t<sub>i</sub>,d</sub></sup> است که به احتمال هر کلمه داده می‌شود. این یعنی هرچه یک کلمه بیشتر در سند تکرار شود، تأثیر بیشتری در احتمال نهایی خواهد داشت.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم دسته‌بندی نظرات مشتریان آمازون، اگر در یک نظر کلمه "عالی" ۵ بار تکرار شود و کلمه "بد" فقط یک بار، مدل چندجمله‌ای این تفاوت را در نظر می‌گیرد و احتمال مثبت بودن نظر را افزایش می‌دهد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
توجه کنید که در این فرمول، عامل چندجمله‌ای حذف شده است. این کار محاسبات را ساده‌تر می‌کند بدون اینکه تأثیر قابل توجهی بر دقت طبقه‌بندی داشته باشد.</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔄 معادل با مدل دنباله</p>
<p style="font-size:16pt; text-align:right;">
این مدل با مدل دنباله در معادله (13.2) معادل است، زیرا:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>برای واژه‌هایی که در سند وجود ندارند (tf<sub>t<sub>i</sub>,d</sub> = 0)، P(X = t<sub>i</sub>|c)<sup>0</sup> = 1 است و تأثیری در محاسبات ندارد.</li>
<li>برای واژه‌هایی که tf<sub>t<sub>i</sub>,d</sub> ≥ ۱ بار تکرار می‌شوند، این واژه در هر دو مدل به تعداد tf<sub>t<sub>i</sub>,d</sub> بار در محاسبات شرکت می‌کند.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
این رویکرد به ویژه در کاربردهایی که تعداد تکرار کلمات اهمیت دارد، مانند تحلیل احساسات در نظرات مشتریان یا تشخیص اسپم، بسیار مفید است.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section13.5" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.5 انتخاب ویژگی: چگونه کلمات مهم را از کلمات بی‌اهمیت تشخیص دهیم؟</span> 🔍
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📧 فیلتر اسپم گوگل و انتخاب ویژگی</p>
<p style="font-size:16pt; text-align:right;">
وقتی گوگل ایمیل شما را فیلتر می‌کند، از هزاران کلمه به عنوان ویژگی استفاده می‌کند. اما آیا همه این کلمات equally مهم هستند؟ کلماتی مانند "ارزان"، "تخفیف" و "برنده" در ایمیل‌های اسپم اهمیت بیشتری دارند. انتخاب ویژگی به گوگل کمک می‌کند تا روی این کلمات مهم تمرکز کند و نویز را نادیده بگیرد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
<strong>انتخاب ویژگی</strong> فرآیند انتخاب زیرمجموعه‌ای از واژه‌های موجود در مجموعه آموزشی و استفاده فقط از این زیرمجموعه به عنوان ویژگی در طبقه‌بندی متن است. این فرآیند دو هدف اصلی دارد:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>افزایش کارایی</strong> با کاهش اندازه واژگان مؤثر، آموزش و اعمال طبقه‌بند را کارآمدتر می‌کند. این موضوع برای طبقه‌بندهایی که برخلاف NB، آموزش آن‌ها پرهزینه است، اهمیت ویژه‌ای دارد.</li>
<li><strong>افزایش دقت طبقه‌بندی</strong> با حذف <strong>ویژگی‌های نویز</strong>. ویژگی نویز ویژگی است که وقتی به نمایش سند اضافه می‌شود، خطای طبقه‌بندی روی داده‌های جدید را افزایش می‌دهد.</li>
</ol>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">⚠️ خطر بیش‌برازش: چگونه یک کلمه نادر می‌تواند کل مدل را خراب کند؟</p>
<p style="font-size:16pt; text-align:right;">
یک واژه نادر مانند "arachnocentric" ممکن است هیچ اطلاعاتی درباره یک کلاس (مثلاً China) نداشته باشد، اما تمام نمونه‌های "arachnocentric" به صورت اتفاقی در اسناد China در مجموعه آموزشی ما رخ داده باشند. در این حالت، روش یادگیری ممکن است طبقه‌بندی تولید کند که اسناد آزمون حاوی "arachnocentric" را به اشتباه به کلاس China اختصاص می‌دهد. این تعمیم نادرست از یک ویژگی تصادفی مجموعه آموزشی <strong>بیش‌برازش</strong> (overfitting) نامیده می‌شود.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان آمازون، اگر واژه "خوشمزه" به اشتباه در چند نظر منفی درباره یک محصول ظاهر شود، ممکن است مدل یاد بگیرد که این واژه نشان‌دهنده نظر منفی است، در حالی که واقعیت چیز دیگری است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
الگوریتم انتخاب ویژگی (شکل 13.6) برای انتخاب k ویژگی برتر طراحی شده است:
</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig13.6.bmp" alt="" style="width:800px; border:1px solid #ccc; border-radius:12px;"
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 چگونه یک طبقه‌بند ضعیف‌تر می‌تواند بهتر عمل کند؟</p>
<p style="font-size:16pt; text-align:right;">
می‌توان انتخاب ویژگی را به عنوان روشی برای جایگزینی یک طبقه‌بند پیچیده (با استفاده از همه ویژگی‌ها) با یک طبقه‌بند ساده‌تر (با استفاده از زیرمجموعه‌ای از ویژگی‌ها) در نظر گرفت. اگرچه در نگاه اول ممکن است غیرمنطقی به نظر برسد که یک طبقه‌بند به ظاهر ضعیف‌تر در طبقه‌بندی آماری متن مطلوب است، اما در بحث معاوضه بایاس-واریانس (bias-variance tradeoff) خواهیم دید که مدل‌های ضعیف‌تر اغلب زمانی که داده‌های آموزشی محدودی در دسترس هستند، ترجیح داده می‌شوند.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم تشخیص اخبار جعلی فیسبوک، استفاده از تمام کلمات ممکن است منجر به بیش‌برازش شود، زیرا برخی کلمات به صورت اتفاقی در اخبار جعلی ظاهر می‌شوند. با انتخاب ویژگی‌های کلیدی مانند "شایعه"، "تأیید نشده" و "منبع نامعتبر"، مدل می‌تواند عملکرد بهتری داشته باشد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
الگوریتم اساسی انتخاب ویژگی در شکل 13.6 نشان داده شده است. برای یک کلاس مشخص c، یک معیار کاربردی A(t, c) برای هر واژه واژگان محاسبه کرده و k واژه با بالاترین مقادیر A(t, c) را انتخاب می‌کنیم. سایر واژه‌ها نادیده گرفته شده و در طبقه‌بندی استفاده نمی‌شوند.
</p>

<p style="font-size:16pt; text-align:right;">
در این بخش سه معیار کاربردی مختلف معرفی می‌شود:
</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
اطلاع متقابل: A(t, c) = I(U<sub>t</sub>; C<sub>c</sub>)<br>
آزمون χ²: A(t, c) = X²(t, c)<br>
فراوانی: A(t, c) = N(t, c)
  </div>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔍 مدل برنولی و حساسیت به نویز</p>
<p style="font-size:16pt; text-align:right;">
از دو مدل NB، مدل برنولی ویژه نسبت به ویژگی‌های نویز حساس است. یک طبقه‌بند برنولی NB به شکلی از انتخاب ویژگی نیاز دارد وگرنه دقت آن پایین خواهد بود.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم تحلیل احساسات توییتر، مدل برنولی اگر از انتخاب ویژگی استفاده نکند، ممکن است به کلمات نادر و بی‌اهمیت که به صورت اتفاقی در توییت‌های مثبت یا منفی ظاهر شده‌اند، بیش از حد حساس شود و دقت طبقه‌بندی کاهش یابد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
این بخش عمدتاً به انتخاب ویژگی برای وظایف طبقه‌بندی دو کلاسی مانند China در مقابل not-China می‌پردازد. بخش 13.5.5 به طور خلاصه به بهینه‌سازی برای سیستم‌های با بیش از دو کلاس می‌پردازد.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section13.5.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.5.1 اطلاعات متقابل: چگونه کلمات کلیدی را برای طبقه‌ بندی پیدا کنیم؟</span> 🔍
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📱 توئیتر و تشخیص موضوعات: چگونه کلمات مهم را از کلمات بی‌اهمیت تشخیص دهیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در توئیتر کار می‌کنید و باید سیستم طراحی کنید که توئیت‌ها را به موضوعات مختلفی مانند "ورزش"، "سیاست" یا "فناوری" دسته‌بندی کند. با میلیون‌ها توئیت در روز، چگونه می‌توانید کلماتی را پیدا کنید که بیشترین اطلاعات را درباره موضوع یک توئیت دارند؟</p>
<p style="font-size:16pt; text-align:right;">
این دقیقاً همان مشکلی است که اطلاعات متقابل (Mutual Information) حل می‌کند. این معیار به ما می‌گوید چقدر وجود یا عدم وجود یک کلمه در توئیت، به ما در تشخیص موضوع آن توئیت کمک می‌کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
یک روش متداول برای انتخاب ویژگی، محاسبه <strong>اطلاعات متقابل (Mutual Information - MI)</strong> برای واژه t و کلاس c است. MI میزان اطلاعاتی را اندازه‌گیری می‌کند که وجود/عدم وجود یک واژه به تصمیم طبقه‌بندی صحیح برای c کمک می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
فرمول MI به صورت زیر است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

 $$ \begin{aligned}
&I(U; C) = \sum_{e_t \in \{1,0\}} \sum_{e_c \in \{1,0\}} P(U = e_t, C = e_c) \log_2 \frac{P(U = e_t, C = e_c)}{P(U = e_t)P(C = e_c)}
\end{aligned}
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
که در آن U متغیر تصادفی است که مقادیر e<sub>t</sub> = 1 (سند شامل واژه t است) و e<sub>t</sub> = 0 (سند شامل واژه t نیست) می‌گیرد، و C متغیر تصادفی است که مقادیر e<sub>c</sub> = 1 (سند در کلاس c است) و e<sub>c</sub> = 0 (سند در کلاس c نیست) می‌گیرد.
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔢 از تئوری تا عمل: چگونه MI را محاسبه کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
در دنیای واقعی، ما از تخمین‌های بیشترین درست نمایی برای احتمالات استفاده می‌کنیم. این کار فرمول بالا را به معادله زیر تبدیل می‌کند که برای محاسبه عملی مناسب‌تر است:</p>
</div>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

 $$ \begin{aligned}
&I(U; C) = \frac{N_{11}}{N} \log_2 \frac{N \cdot N_{11}}{N_{1.} \cdot N_{.1}} \\
&+ \frac{N_{01}}{N} \log_2 \frac{N \cdot N_{01}}{N_{0.} \cdot N_{.1}} \\
&+ \frac{N_{10}}{N} \log_2 \frac{N \cdot N_{10}}{N_{1.} \cdot N_{.0}} \\
&+ \frac{N_{00}}{N} \log_2 \frac{N \cdot N_{00}}{N_{0.} \cdot N_{.0}}
\end{aligned}
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
که در آن N<sub>ij</sub> تعداد اسنادی است که مقادیر e<sub>t</sub> و e<sub>c</sub> را دارند که با دو زیرنشان مشخص شده‌اند. برای مثال، N<sub>10</sub> تعداد اسنادی است که شامل واژه t هستند (e<sub>t</sub> = 1) اما در کلاس c نیستند (e<sub>c</sub> = 0).
</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 مثال واقعی: طبقه‌بندی اخبار کشاورزی</p>
<p style="font-size:16pt; text-align:right;">
✎ مثال 13.3: برای کلاس poultry (طیور) و واژه export (صادرات) در مجموعه اخبار Reuters-RCV1، تعداد اسناد با چهار ترکیب ممکن از مقادیر شاخص به صورت زیر است:</p>
</div>

  <tr>
    <th></th>
    <th>e<sub>c</sub> = poultry = 1</th>
    <th>e<sub>c</sub> = poultry = 0</th>
  </tr>

    <td>e<sub>t</sub> = export = 1</td>
    <td>N<sub>11</sub> = 49</td>

  </tr>

  </tr>

<p style="font-size:16pt; text-align:right;">
با جایگذاری این مقادیر در معادله، به دست می‌آوریم: I(U; C) ≈ 0.0001105
</p>

<p style="font-size:16pt; text-align:right;">
این عدد کوچک نشان می‌دهد که واژه "export" به تنهایی اطلاعات کمی درباره کلاس "طیور" دارد. این منطقی است، زیرا "export" در بسیاری از زمینه‌ها استفاده می‌شود، نه فقط در اخبار مربوط به طیور.
</p>

<p style="font-size:16pt; text-align:right;">
برای انتخاب k واژه t<sub>1</sub>, ..., t<sub>k</sub> برای یک کلاس مشخص، از الگوریتم انتخاب ویژگی در شکل 13.6 استفاده می‌کنیم: معیار کاربردی را به صورت A(t, c) = I(U<sub>t</sub>, C<sub>c</sub>) محاسبه کرده و k واژه با بالاترین مقادیر را انتخاب می‌کنیم.
</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🌍 کلماتی که جهان را توصیف می‌کنند: MI در عمل</p>

<p style="font-size:16pt; text-align:right;">
اطلاعات متقابل میزان اطلاعاتی را اندازه‌گیری می‌کند که یک واژه درباره یک کلاس دارد. اگر توزیع یک واژه در کلاس مشابه کل مجموعه باشد، I(U; C) = 0 است. MI به حداکثر مقدار خود می‌رسد اگر واژه یک شاخص کامل برای عضویت در کلاس باشد.</p>
<p style="font-size:16pt; text-align:right;">
شگفت‌انگیز است که استفاده از زیرمجموعه‌ای کوچک از ویژگی‌ها می‌تواند دقت طبقه‌بندی را بهبود بخشد! شکل 13.8 تأثیر اندازه مجموعه ویژگی بر دقت را برای مدل‌های چندجمله‌ای و برنولی نشان می‌دهد. انتخاب ویژگی با استفاده از MI، F1 را برای مدل چندجمله‌ای حدود 0.1 و برای مدل برنولی بیش از 0.2 افزایش می‌دهد.</p>

<p style="font-size:16pt; text-align:right;">
برای مدل برنولی، F1 در زودهنگام، با ده ویژگی منتخب، به اوج می‌رسد. این نشان می‌دهد که برای مدل برنولی، استفاده از تعداد کمی از ویژگی‌های بسیار قوی، بهتر از استفاده از تعداد زیادی ویژگی با قدرت متغیر است.</p>
<p style="font-size:16pt; text-align:right;">
<b>کاربرد واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان آمازون، استفاده از تنها 20 کلمه با بالاترین امتیاز MI برای هر دسته (مثلاً "عالی"، "خوب"، "بد"، "ضعیف" برای نظرات مثبت و منفی) می‌تواند دقت طبقه‌بندی را به اندازه استفاده از تمام کلمات در نظر بهبود بخشد، در حالی که سرعت پردازش را به شدت افزایش می‌دهد.</p>

```python
# پیاده‌سازی انتخاب ویژگی با استفاده از اطلاعات متقابل (Mutual Information)

import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import f1_score

def calculate_mutual_information(documents, labels, term, target_class):
    """
    محاسبه اطلاعات متقابل بین یک واژه و یک کلاس
    
    این تابع فرمول MI را پیاده‌سازی می‌کند که در MD cell توضیح داده شد.
    MI میزان اطلاعاتی را اندازه‌گیری می‌کند که وجود یا عدم وجود یک واژه به تصمیم طبقه‌بندی صحیح کمک می‌کند.
    
    پارامترها:
    documents: لیستی از اسناد، هر سند یک لیست از کلمات است
    labels: لیستی از برچسب‌های مربوط به هر سند
    term: واژه مورد نظر
    target_class: کلاس مورد نظر
    """
    N = len(documents)  # تعداد کل اسناد
    
    # شمارش مقادیر N11, N10, N01, N00
    N11 = N10 = N01 = N00 = 0
    
    for doc, label in zip(documents, labels):
        term_in_doc = 1 if term in doc else 0
        is_target_class = 1 if label == target_class else 0
        
        if term_in_doc == 1 and is_target_class == 1:
            N11 += 1
        elif term_in_doc == 1 and is_target_class == 0:
            N10 += 1
        elif term_in_doc == 0 and is_target_class == 1:
            N01 += 1
        else:  # term_in_doc == 0 and is_target_class == 0
            N00 += 1
    
    # محاسبه مقادیر جانبی
    N1_ = N10 + N11  # تعداد اسناد شامل واژه
    N0_ = N00 + N01  # تعداد اسناد بدون واژه
    N_1 = N01 + N11  # تعداد اسناد در کلاس هدف
    N_0 = N00 + N10  # تعداد اسناد در کلاس غیرهدف
    
    # محاسبه اطلاعات متقابل با استفاده از فرمول
    mi = 0.0
    if N11 > 0:
        mi += (N11 / N) * np.log2((N * N11) / (N1_ * N_1))
    if N01 > 0:
        mi += (N01 / N) * np.log2((N * N01) / (N0_ * N_1))
    if N10 > 0:
        mi += (N10 / N) * np.log2((N * N10) / (N1_ * N_0))
    if N00 > 0:
        mi += (N00 / N) * np.log2((N * N00) / (N0_ * N_0))
    
    return mi

def select_features_mi(documents, labels, k, target_class):
    """
    انتخاب k ویژگی برتر با استفاده از اطلاعات متقابل
    
    این تابع MI را برای همه واژه‌ها محاسبه کرده و k واژه با بالاترین امتیاز را انتخاب می‌کند.
    این همان الگوریتمی است که در شکل 13.6 کتاب توضیح داده شد.
    """
    # استخراج واژگان
    vocab = set()
    for doc in documents:
        vocab.update(doc)
    
    # محاسبه اطلاعات متقابل برای هر واژه
    term_scores = {}
    for term in vocab:
        mi = calculate_mutual_information(documents, labels, term, target_class)
        term_scores[term] = mi
    
    # انتخاب k واژه با بالاترین امتیاز
    selected_features = sorted(term_scores.items(), key=lambda x: x[1], reverse=True)[:k]
    return [term for term, score in selected_features], [score for term, score in selected_features]

def create_document_representation(documents, selected_features):
    """
    ایجاد نمایش سند بر اساس ویژگی‌های منتخب
    برای مدل چندجمله‌ای: تعداد تکرار هر ویژگی
    برای مدل برنولی: وجود یا عدم وجود هر ویژگی
    """
    # نمایش برای مدل چندجمله‌ای
    multinomial_rep = []
    for doc in documents:
        doc_rep = [doc.count(term) for term in selected_features]
        multinomial_rep.append(doc_rep)
    
    # نمایش برای مدل برنولی
    bernoulli_rep = []
    for doc in documents:
        doc_rep = [1 if term in doc else 0 for term in selected_features]
        bernoulli_rep.append(doc_rep)
    
    return np.array(multinomial_rep), np.array(bernoulli_rep)

def evaluate_feature_selection(documents, labels, target_class, feature_counts):
    """
    ارزیابی تأثیر انتخاب ویژگی بر دقت طبقه‌بندی
    
    این تابع دقت مدل‌های چندجمله‌ای و برنولی را برای تعداد مختلفی از ویژگی‌ها محاسبه می‌کند.
    نتایج نشان می‌دهد چگونه انتخاب ویژگی می‌تواند دقت را بهبود بخشد، همانطور که در شکل 13.8 کتاب نشان داده شد.
    """
    # تقسیم داده‌ها به آموزش و آزمون
    X_train_docs, X_test_docs, y_train, y_test = train_test_split(
        documents, labels, test_size=0.3, random_state=42
    )
    
    # تبدیل برچسب‌ها به باینری (کلاس هدف در مقابل بقیه)
    y_train_binary = np.array([1 if label == target_class else 0 for label in y_train])
    y_test_binary = np.array([1 if label == target_class else 0 for label in y_test])
    
    # انتخاب ویژگی‌ها بر اساس داده‌های آموزشی
    vocab = set()
    for doc in X_train_docs:
        vocab.update(doc)
    
    # محاسبه امتیاز MI برای همه واژه‌ها
    term_scores = {}
    for term in vocab:
        mi = calculate_mutual_information(X_train_docs, y_train, term, target_class)
        term_scores[term] = mi
    
    # مرتب‌سازی واژه‌ها بر اساس امتیاز MI
    sorted_terms = sorted(term_scores.items(), key=lambda x: x[1], reverse=True)
    all_terms = [term for term, score in sorted_terms]
    
    # ارزیابی برای تعداد مختلفی از ویژگی‌ها
    multinomial_scores = []
    bernoulli_scores = []
    
    for k in feature_counts:
        # انتخاب k ویژگی برتر
        selected_features = all_terms[:k]
        
        # ایجاد نمایش سند برای مدل‌ها
        X_train_multi, X_train_bern = create_document_representation(X_train_docs, selected_features)
        X_test_multi, X_test_bern = create_document_representation(X_test_docs, selected_features)
        
        # آموزش و ارزیابی مدل چندجمله‌ای
        clf_multi = MultinomialNB()
        clf_multi.fit(X_train_multi, y_train_binary)
        y_pred_multi = clf_multi.predict(X_test_multi)
        f1_multi = f1_score(y_test_binary, y_pred_multi)
        multinomial_scores.append(f1_multi)
        
        # آموزش و ارزیابی مدل برنولی
        clf_bern = BernoulliNB()
        clf_bern.fit(X_train_bern, y_train_binary)
        y_pred_bern = clf_bern.predict(X_test_bern)
        f1_bern = f1_score(y_test_binary, y_pred_bern)
        bernoulli_scores.append(f1_bern)
    
    return multinomial_scores, bernoulli_scores

# ایجاد مجموعه داده نمونه (شبیه به مثال کتاب)
np.random.seed(42)

# کلاس‌ها
classes = ['poultry', 'coffee', 'sports', 'china', 'uk', 'elections']

# واژگان مرتبط با هر کلاس
class_vocab = {
    'poultry': ['poultry', 'chicken', 'meat', 'avian', 'bird', 'farm', 'egg', 'broiler', 'inspection', 'disease'],
    'coffee': ['coffee', 'bean', 'roast', 'brew', 'colombia', 'brazil', 'export', 'crop', 'harvest', 'arabica'],
    'sports': ['soccer', 'game', 'match', 'team', 'player', 'score', 'win', 'league', 'cup', 'championship'],
    'china': ['china', 'chinese', 'beijing', 'yuan', 'shanghai', 'hong', 'kong', 'xinhua', 'province', 'taiwan'],
    'uk': ['london', 'uk', 'british', 'britain', 'england', 'pound', 'pence', 'stg', 'plc', 'english'],
    'elections': ['election', 'vote', 'candidate', 'campaign', 'poll', 'party', 'democratic', 'republican', 'president', 'ballot']
}

# واژگان عمومی
general_vocab = ['report', 'said', 'year', 'today', 'new', 'government', 'market', 'company', 'people', 'world']

# ایجاد اسناد
documents = []
labels = []
num_docs_per_class = 100

for cls in classes:
    for i in range(num_docs_per_class):
        # هر سند شامل 5-15 واژه است
        doc_length = np.random.randint(5, 16)
        doc = []
        
        # اضافه کردن 2-4 واژه از کلاس
        num_class_terms = np.random.randint(2, 5)
        class_terms = np.random.choice(class_vocab[cls], num_class_terms, replace=False)
        doc.extend(class_terms)
        
        # اضافه کردن واژگان عمومی
        num_general_terms = doc_length - num_class_terms
        general_terms = np.random.choice(general_vocab, num_general_terms, replace=True)
        doc.extend(general_terms)
        
        # اضافه کردن برخی واژگان نویز (با احتمال کم)
        if np.random.random() < 0.1:
            noise_term = f"noise_term_{np.random.randint(100)}"
            doc.append(noise_term)
        
        documents.append(doc)
        labels.append(cls)

# محاسبه اطلاعات متقابل برای کلاس poultry و واژه export
mi_export = calculate_mutual_information(documents, labels, "export", "poultry")
print(f"اطلاعات متقابل برای واژه 'export' و کلاس 'poultry': {mi_export:.6f}")

# انتخاب 10 ویژگی برتر برای کلاس poultry
selected_features, scores = select_features_mi(documents, labels, 10, "poultry")
print("ویژگی‌های منتخب برای کلاس poultry:")
for feature, score in zip(selected_features, scores):
    print(f"{feature}: {score:.6f}")

# ارزیابی تأثیر انتخاب ویژگی بر دقت طبقه‌بندی
feature_counts = [1, 5, 10, 20, 50, 100]
multinomial_scores, bernoulli_scores = evaluate_feature_selection(documents, labels, "poultry", feature_counts)

# نمایش نتایج
plt.figure(figsize=(10, 6))
plt.plot(feature_counts, multinomial_scores, 'o-', label='Multinomial NB')
plt.plot(feature_counts, bernoulli_scores, 'x-', label='Bernoulli NB')
plt.xscale('log')
plt.xlabel('Number of Features')
plt.ylabel('F1 Score')
plt.title('Effect of Feature Selection on Classification Accuracy')
plt.legend()
plt.grid(True)
plt.show()

# نمایش نتایج عددی
print("\nنتایج ارزیابی:")
print("تعداد ویژگی\t\tF1 (چندجمله‌ای)\tF1 (برنولی)")
for i, k in enumerate(feature_counts):
    print(f"{k}\t\t\t{multinomial_scores[i]:.4f}\t\t{bernoulli_scores[i]:.4f}")
```

<p style="font-size:18pt; text-align:center;">
<span id="section13.5.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.5.2 آزمون χ²: چگونه کلمات مهم را از کلمات تصادفی تشخیص دهیم؟</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🛒 آمازون و تشخیص محصولات محبوب: آیا "خرید" واقعاً نشان‌دهنده محبوبیت است؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در آمازون کار می‌کنید و باید سیستم طراحی کنید که محصولات را به دسته‌بندی‌های مختلفی مانند "محبوب"، "متوسط" یا "نامحبوب" تقسیم کند. کلمه "خرید" در نظرات محصولات بسیار زیاد تکرار می‌شود، اما آیا این کلمه واقعاً به ما کمک می‌کند تا تشخیص دهیم یک محصول محبوب است یا خیر؟</p>
<p style="font-size:16pt; text-align:right;">
آزمون χ² به ما کمک می‌کند تا بفهمیم آیا یک کلمه واقعاً با یک کلاس مرتبط است یا فقط به صورت تصادفی در اسناد ظاهر شده است. این آزمون فراوانی مشاهده شده یک کلمه در یک کلاس را با فراوانی مورد انتظار تحت فرض استقلال مقایسه می‌کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
روش دیگر محبوب برای انتخاب ویژگی، استفاده از آزمون <strong>χ²</strong> است. در آمار، آزمون χ² برای بررسی استقلال دو رویداد به کار می‌رود. در انتخاب ویژگی، دو رویداد عبارتند از وقوع واژه و وقوع کلاس. ما واژه‌ها را بر اساس مقدار زیر رتبه‌بندی می‌کنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

 $$ \begin{aligned}
&\chi^2(D, t, c) = \sum_{e_t \in \{0,1\}} \sum_{e_c \in \{0,1\}} \frac{(N_{e_t e_c} - E_{e_t e_c})^2}{E_{e_t e_c}}
\end{aligned}
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
که در آن N فراوانی مشاهده شده و E فراوانی مورد انتظار است. برای مثال، E<sub>11</sub> فراوانی مورد انتظار وقوع همزمان واژه t و کلاس c است با فرض استقلال آن‌ها.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔢 از تئوری تا عمل: چگونه χ² را محاسبه کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
✎ مثال 13.4: بیایید به مثال اخبار کشاورزی برگردیم و χ² را برای کلاس poultry (طیور) و واژه export (صادرات) محاسبه کنیم. ابتدا E<sub>11</sub> را برای داده‌های مثال 13.3 محاسبه می‌کنیم:</p>
</div>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
E<sub>11</sub> = N × P(t) × P(c) = N × (49+141)/N × (49+27652)/N ≈ 6.6
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
با محاسبه سایر مقادیر E و جایگذاری در معادله، به دست می‌آوریم: χ²(D, t, c) ≈ 284</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 معنای آماری: چگونه χ² به ما کمک می‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
χ² معیاری از انحراف فراوانی‌های مورد انتظار (E) و مشاهده شده (N) از یکدیگر است. مقدار بالای χ² نشان می‌دهد که فرض استقلال، که به معنای شباهت فراوانی‌های مورد انتظار و مشاهده شده است، نادرست است.</p>
<p style="font-size:16pt; text-align:right;">
در مثال ما، χ² ≈ 284 > 10.83، بنابراین می‌توانیم فرض استقلال poultry و export را با احتمال خطای 0.001 رد کنیم. این به ما می‌گوید که واژه "export" و کلاس "poultry" به هم وابسته هستند و این وابستگی تصادفی نیست.</p>
<p style="font-size:16pt; text-align:right;">
<b>کاربرد واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان فیسبوک، کلمه "عالی" ممکن است در نظرات مثبت و منفی ظاهر شود. آزمون χ² به ما کمک می‌کند تا بفهمیم آیا این کلمه واقعاً با نظرات مثبت مرتبط است یا فقط به صورت تصادفی در هر دو نوع نظر ظاهر می‌شود.</p>
</div>

<p style="font-size:16pt; text-align:right;">
یک روش ساده‌تر برای محاسبه χ² به صورت زیر است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

 $$ \begin{aligned}
&\chi^2(D, t, c) = \frac{(N_{11} + N_{10} + N_{01} + N_{00}) \times (N_{11}N_{00} - N_{10}N_{01})^2}{(N_{11} + N_{01}) \times (N_{11} + N_{10}) \times (N_{10} + N_{00}) \times (N_{01} + N_{00})}
\end{aligned}
 $$ 
</div>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚠️ محدودیت‌های آماری: آیا χ² همیشه دقیق است؟</p>
<p style="font-size:16pt; text-align:right;">
از دیدگاه آماری، انتخاب ویژگی با χ² مشکل‌ساز است. برای یک آزمون با یک درجه آزادی، باید از اصلاح Yates استفاده شود که رسیدن به اهمیت آماری را دشوارتر می‌کند. همچنین، هرگاه یک آزمون آماری چندین بار استفاده شود، احتمال خطا افزایش می‌یابد.</p>
<p style="font-size:16pt; text-align:right;">
با این حال، در طبقه‌بندی متن، به ندرت اهمیت دارد که چند واژه اضافی به مجموعه ویژگی اضافه یا از آن حذف شوند. مهم‌تر، اهمیت نسبی ویژگی‌ها است. تا زمانی که انتخاب ویژگی با χ² فقط ویژگی‌ها را بر اساس کاربردی رتبه‌بندی کند و برای بیان بیانیه‌هایی درباره وابستگی آماری متغیرها استفاده نشود، نباید بیش از حد نگران عدم پیروی دقیق از نظریه آماری باشیم.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی اخبار گوگل، حتی اگر χ² برای برخی کلمات به دلیل محدودیت‌های آماری دقیق نباشد، سیستم همچنان می‌تواند با استفاده از رتبه‌بندی نسبی کلمات، اخبار را با دقت بالایی دسته‌بندی کند. در عمل، این کافی است که کلمات مرتبط‌تر بالاترین رتبه را داشته باشند، حتی اگر مقادیر دقیق χ² از نظر آماری کامل نباشند.</p>

```python
# پیاده‌سازی انتخاب ویژگی با استفاده از آزمون χ²

import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import f1_score

def calculate_chi2(documents, labels, term, target_class):
    """
    محاسبه مقدار χ² برای یک واژه و یک کلاس
    
    این تابع فرمول χ² را پیاده‌سازی می‌کند که در MD cell توضیح داده شد.
    χ² میزان انحراف فراوانی مشاهده شده از فراوانی مورد انتظار تحت فرض استقلال را اندازه‌گیری می‌کند.
    
    پارامترها:
    documents: لیستی از اسناد، هر سند یک لیست از کلمات است
    labels: لیستی از برچسب‌های مربوط به هر سند
    term: واژه مورد نظر
    target_class: کلاس مورد نظر
    """
    N = len(documents)  # تعداد کل اسناد
    
    # شمارش مقادیر N11, N10, N01, N00
    N11 = N10 = N01 = N00 = 0
    
    for doc, label in zip(documents, labels):
        term_in_doc = 1 if term in doc else 0
        is_target_class = 1 if label == target_class else 0
        
        if term_in_doc == 1 and is_target_class == 1:
            N11 += 1
        elif term_in_doc == 1 and is_target_class == 0:
            N10 += 1
        elif term_in_doc == 0 and is_target_class == 1:
            N01 += 1
        else:  # term_in_doc == 0 and is_target_class == 0
            N00 += 1
    
    # محاسبه χ² با استفاده از فرمول ساده‌شده (معادل فرمول اصلی)
    numerator = (N11 + N10 + N01 + N00) * ((N11 * N00) - (N10 * N01))**2
    denominator = (N11 + N01) * (N11 + N10) * (N10 + N00) * (N01 + N00)
    
    if denominator == 0:
        return 0
    
    chi2 = numerator / denominator
    return chi2

def calculate_mutual_information(documents, labels, term, target_class):
    """
    محاسبه اطلاعات متقابل بین یک واژه و یک کلاس برای مقایسه با χ²
    """
    N = len(documents)
    
    # شمارش مقادیر N11, N10, N01, N00
    N11 = N10 = N01 = N00 = 0
    
    for doc, label in zip(documents, labels):
        term_in_doc = 1 if term in doc else 0
        is_target_class = 1 if label == target_class else 0
        
        if term_in_doc == 1 and is_target_class == 1:
            N11 += 1
        elif term_in_doc == 1 and is_target_class == 0:
            N10 += 1
        elif term_in_doc == 0 and is_target_class == 1:
            N01 += 1
        else:  # term_in_doc == 0 and is_target_class == 0
            N00 += 1
    
    # محاسبه مقادیر جانبی
    N1_ = N10 + N11  # تعداد اسناد شامل واژه
    N0_ = N00 + N01  # تعداد اسناد بدون واژه
    N_1 = N01 + N11  # تعداد اسناد در کلاس هدف
    N_0 = N00 + N10  # تعداد اسناد در کلاس غیرهدف
    
    # محاسبه اطلاعات متقابل
    mi = 0.0
    if N11 > 0:
        mi += (N11 / N) * np.log2((N * N11) / (N1_ * N_1))
    if N01 > 0:
        mi += (N01 / N) * np.log2((N * N01) / (N0_ * N_1))
    if N10 > 0:
        mi += (N10 / N) * np.log2((N * N10) / (N1_ * N_0))
    if N00 > 0:
        mi += (N00 / N) * np.log2((N * N00) / (N0_ * N_0))
    
    return mi

def select_features_chi2(documents, labels, k, target_class):
    """
    انتخاب k ویژگی برتر با استفاده از آزمون χ²
    
    این تابع χ² را برای همه واژه‌ها محاسبه کرده و k واژه با بالاترین امتیاز را انتخاب می‌کند.
    """
    # استخراج واژگان
    vocab = set()
    for doc in documents:
        vocab.update(doc)
    
    # محاسبه χ² برای هر واژه
    term_scores = {}
    for term in vocab:
        chi2 = calculate_chi2(documents, labels, term, target_class)
        term_scores[term] = chi2
    
    # انتخاب k واژه با بالاترین امتیاز
    selected_features = sorted(term_scores.items(), key=lambda x: x[1], reverse=True)[:k]
    return [term for term, score in selected_features], [score for term, score in selected_features]

def select_features_mi(documents, labels, k, target_class):
    """
    انتخاب k ویژگی برتر با استفاده از اطلاعات متقابل برای مقایسه
    """
    # استخراج واژگان
    vocab = set()
    for doc in documents:
        vocab.update(doc)
    
    # محاسبه اطلاعات متقابل برای هر واژه
    term_scores = {}
    for term in vocab:
        mi = calculate_mutual_information(documents, labels, term, target_class)
        term_scores[term] = mi
    
    # انتخاب k واژه با بالاترین امتیاز
    selected_features = sorted(term_scores.items(), key=lambda x: x[1], reverse=True)[:k]
    return [term for term, score in selected_features], [score for term, score in selected_features]

def create_document_representation(documents, selected_features):
    """
    ایجاد نمایش سند بر اساس ویژگی‌های منتخب
    برای مدل چندجمله‌ای: تعداد تکرار هر ویژگی
    برای مدل برنولی: وجود یا عدم وجود هر ویژگی
    """
    # نمایش برای مدل چندجمله‌ای
    multinomial_rep = []
    for doc in documents:
        doc_rep = [doc.count(term) for term in selected_features]
        multinomial_rep.append(doc_rep)
    
    # نمایش برای مدل برنولی
    bernoulli_rep = []
    for doc in documents:
        doc_rep = [1 if term in doc else 0 for term in selected_features]
        bernoulli_rep.append(doc_rep)
    
    return np.array(multinomial_rep), np.array(bernoulli_rep)

def compare_feature_selection_methods(documents, labels, target_class, k=10):
    """
    مقایسه روش‌های انتخاب ویژگی χ² و MI
    
    این تابع نشان می‌دهد که چگونه دو روش مختلف می‌توانند ویژگی‌های متفاوتی را انتخاب کنند.
    """
    # انتخاب ویژگی با χ²
    chi2_features, chi2_scores = select_features_chi2(documents, labels, k, target_class)
    
    # انتخاب ویژگی با MI
    mi_features, mi_scores = select_features_mi(documents, labels, k, target_class)
    
    # نمایش نتایج
    print(f"Top {k} features selected by χ² for class '{target_class}':")
    for feature, score in zip(chi2_features, chi2_scores):
        print(f"{feature}: {score:.2f}")
    
    print(f"\nTop {k} features selected by MI for class '{target_class}':")
    for feature, score in zip(mi_features, mi_scores):
        print(f"{feature}: {score:.6f}")
    
    # مقایسه ویژگی‌های مشترک
    common_features = set(chi2_features) & set(mi_features)
    print(f"\nCommon features selected by both methods: {list(common_features)}")
    
    return chi2_features, mi_features

def evaluate_feature_selection_methods(documents, labels, target_class, feature_counts):
    """
    ارزیابی تأثیر روش‌های مختلف انتخاب ویژگی بر دقت طبقه‌بندی
    
    این تابع دقت مدل‌های چندجمله‌ای و برنولی را برای تعداد مختلفی از ویژگی‌ها محاسبه می‌کند.
    نتایج نشان می‌دهد چگونه روش‌های مختلف انتخاب ویژگی می‌توانند دقت را بهبود بخشند.
    """
    # تقسیم داده‌ها به آموزش و آزمون
    X_train_docs, X_test_docs, y_train, y_test = train_test_split(
        documents, labels, test_size=0.3, random_state=42
    )
    
    # تبدیل برچسب‌ها به باینری (کلاس هدف در مقابل بقیه)
    y_train_binary = np.array([1 if label == target_class else 0 for label in y_train])
    y_test_binary = np.array([1 if label == target_class else 0 for label in y_test])
    
    # انتخاب ویژگی‌ها بر اساس داده‌های آموزشی
    vocab = set()
    for doc in X_train_docs:
        vocab.update(doc)
    
    # محاسبه امتیاز χ² برای همه واژه‌ها
    chi2_scores = {}
    for term in vocab:
        chi2_scores[term] = calculate_chi2(X_train_docs, y_train, term, target_class)
    
    # محاسبه امتیاز MI برای همه واژه‌ها
    mi_scores = {}
    for term in vocab:
        mi_scores[term] = calculate_mutual_information(X_train_docs, y_train, term, target_class)
    
    # مرتب‌سازی واژه‌ها بر اساس امتیازها
    chi2_sorted_terms = sorted(chi2_scores.items(), key=lambda x: x[1], reverse=True)
    mi_sorted_terms = sorted(mi_scores.items(), key=lambda x: x[1], reverse=True)
    
    chi2_terms = [term for term, score in chi2_sorted_terms]
    mi_terms = [term for term, score in mi_sorted_terms]
    
    # ارزیابی برای تعداد مختلفی از ویژگی‌ها
    chi2_multinomial_scores = []
    chi2_bernoulli_scores = []
    mi_multinomial_scores = []
    mi_bernoulli_scores = []
    
    for k in feature_counts:
        # انتخاب k ویژگی برتر با χ²
        chi2_selected_features = chi2_terms[:k]
        
        # ایجاد نمایش سند برای مدل‌ها
        X_train_multi, X_train_bern = create_document_representation(X_train_docs, chi2_selected_features)
        X_test_multi, X_test_bern = create_document_representation(X_test_docs, chi2_selected_features)
        
        # آموزش و ارزیابی مدل چندجمله‌ای با χ²
        clf_multi = MultinomialNB()
        clf_multi.fit(X_train_multi, y_train_binary)
        y_pred_multi = clf_multi.predict(X_test_multi)
        f1_multi = f1_score(y_test_binary, y_pred_multi)
        chi2_multinomial_scores.append(f1_multi)
        
        # آموزش و ارزیابی مدل برنولی با χ²
        clf_bern = BernoulliNB()
        clf_bern.fit(X_train_bern, y_train_binary)
        y_pred_bern = clf_bern.predict(X_test_bern)
        f1_bern = f1_score(y_test_binary, y_pred_bern)
        chi2_bernoulli_scores.append(f1_bern)
        
        # انتخاب k ویژگی برتر با MI
        mi_selected_features = mi_terms[:k]
        
        # ایجاد نمایش سند برای مدل‌ها
        X_train_multi, X_train_bern = create_document_representation(X_train_docs, mi_selected_features)
        X_test_multi, X_test_bern = create_document_representation(X_test_docs, mi_selected_features)
        
        # آموزش و ارزیابی مدل چندجمله‌ای با MI
        clf_multi = MultinomialNB()
        clf_multi.fit(X_train_multi, y_train_binary)
        y_pred_multi = clf_multi.predict(X_test_multi)
        f1_multi = f1_score(y_test_binary, y_pred_multi)
        mi_multinomial_scores.append(f1_multi)
        
        # آموزش و ارزیابی مدل برنولی با MI
        clf_bern = BernoulliNB()
        clf_bern.fit(X_train_bern, y_train_binary)
        y_pred_bern = clf_bern.predict(X_test_bern)
        f1_bern = f1_score(y_test_binary, y_pred_bern)
        mi_bernoulli_scores.append(f1_bern)
    
    return (chi2_multinomial_scores, chi2_bernoulli_scores, 
            mi_multinomial_scores, mi_bernoulli_scores)

# ایجاد مجموعه داده نمونه (شبیه به مثال کتاب)
np.random.seed(42)

# کلاس‌ها
classes = ['poultry', 'coffee', 'sports', 'china', 'uk', 'elections']

# واژگان مرتبط با هر کلاس
class_vocab = {
    'poultry': ['poultry', 'chicken', 'meat', 'avian', 'bird', 'farm', 'egg', 'broiler', 'inspection', 'disease'],
    'coffee': ['coffee', 'bean', 'roast', 'brew', 'colombia', 'brazil', 'export', 'crop', 'harvest', 'arabica'],
    'sports': ['soccer', 'game', 'match', 'team', 'player', 'score', 'win', 'league', 'cup', 'championship'],
    'china': ['china', 'chinese', 'beijing', 'yuan', 'shanghai', 'hong', 'kong', 'xinhua', 'province', 'taiwan'],
    'uk': ['london', 'uk', 'british', 'britain', 'england', 'pound', 'pence', 'stg', 'plc', 'english'],
    'elections': ['election', 'vote', 'candidate', 'campaign', 'poll', 'party', 'democratic', 'republican', 'president', 'ballot']
}

# واژگان عمومی
general_vocab = ['report', 'said', 'year', 'today', 'new', 'government', 'market', 'company', 'people', 'world']

# ایجاد اسناد
documents = []
labels = []
num_docs_per_class = 100

for cls in classes:
    for i in range(num_docs_per_class):
        # هر سند شامل 5-15 واژه است
        doc_length = np.random.randint(5, 16)
        doc = []
        
        # اضافه کردن 2-4 واژه از کلاس
        num_class_terms = np.random.randint(2, 5)
        class_terms = np.random.choice(class_vocab[cls], num_class_terms, replace=False)
        doc.extend(class_terms)
        
        # اضافه کردن واژگان عمومی
        num_general_terms = doc_length - num_class_terms
        general_terms = np.random.choice(general_vocab, num_general_terms, replace=True)
        doc.extend(general_terms)
        
        # اضافه کردن برخی واژگان نویز (با احتمال کم)
        if np.random.random() < 0.1:
            noise_term = f"noise_term_{np.random.randint(100)}"
            doc.append(noise_term)
        
        documents.append(doc)
        labels.append(cls)

# مثال از کتاب: محاسبه χ² برای کلاس poultry و واژه export
chi2_export = calculate_chi2(documents, labels, "export", "poultry")
print(f"مقدار χ² برای واژه 'export' و کلاس 'poultry': {chi2_export:.2f}")

# مقایسه روش‌های انتخاب ویژگی χ² و MI
chi2_features, mi_features = compare_feature_selection_methods(documents, labels, "poultry", k=10)

# ارزیابی تأثیر روش‌های مختلف انتخاب ویژگی بر دقت طبقه‌بندی
feature_counts = [1, 5, 10, 20, 50, 100]
(chi2_multi_scores, chi2_bernoulli_scores, 
 mi_multi_scores, mi_bernoulli_scores) = evaluate_feature_selection_methods(
    documents, labels, "poultry", feature_counts)

# نمایش نتایج
plt.figure(figsize=(12, 8))
plt.plot(feature_counts, chi2_multi_scores, 'o-', label='Multinomial NB with χ²')
plt.plot(feature_counts, chi2_bernoulli_scores, 'x-', label='Bernoulli NB with χ²')
plt.plot(feature_counts, mi_multi_scores, 's-', label='Multinomial NB with MI')
plt.plot(feature_counts, mi_bernoulli_scores, 'd-', label='Bernoulli NB with MI')
plt.xscale('log')
plt.xlabel('Number of Features')
plt.ylabel('F1 Score')
plt.title('Comparison of Feature Selection Methods')
plt.legend()
plt.grid(True)
plt.ylim(-0.05, 1.05)
plt.show()

# نمایش نتایج عددی
print("\nنتایج ارزیابی:")
print("تعداد ویژگی\tF1 (چندجمله‌ای χ²)\tF1 (برنولی χ²)\tF1 (چندجمله‌ای MI)\tF1 (برنولی MI)")
for i, k in enumerate(feature_counts):
    print(f"{k}\t\t{chi2_multi_scores[i]:.4f}\t\t{chi2_bernoulli_scores[i]:.4f}\t\t{mi_multi_scores[i]:.4f}\t\t{mi_bernoulli_scores[i]:.4f}")
```

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎬 نتفلیکس و تحلیل نظرات کاربران: چگونه فیلم‌ها را دسته‌بندی کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم تحلیل داده‌های نتفلیکس کار می‌کنید و باید سیستم طراحی کنید که نظرات کاربران را به ژانرهای مختلف فیلم (اکشن، کمدی، درام، علمی-تخیلی) دسته‌بندی کند. یک رویکرد ساده این است که کلماتی را پیدا کنید که بیشترین تکرار را در نظرات هر ژانر دارند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، در نظرات فیلم‌های اکشن، کلماتی مانند "اکشن"، "هیجان‌انگیز"، "تعقیب" و "آتش" ممکن است بیشتر تکرار شوند. این روش ساده، انتخاب ویژگی مبتنی بر فراوانی نام دارد.</p>

<p style="font-size:16pt; text-align:right;">
روش سوم برای انتخاب ویژگی، <strong>انتخاب ویژگی مبتنی بر فراوانی</strong> است، یعنی انتخاب واژه‌هایی که در یک کلاس بیشترین فراوانی را دارند. این ساده‌ترین روش انتخاب ویژگی است، اما همیشه مؤثرترین نیست. فراوانی می‌تواند به دو صورت تعریف شود:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>فراوانی سندی</strong> (document frequency): تعداد اسناد در کلاس c که شامل واژه t هستند (مناسب برای مدل برنولی)</li>
<li><strong>فراوانی مجموعه</strong> (collection frequency): تعداد توکن‌های t که در اسناد کلاس c رخ می‌دهند (مناسب برای مدل چندجمله‌ای)</li>
</ul>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">⚠️ مزایا و معایب: چرا ساده‌ترین همیشه بهترین نیست؟</p>
<p style="font-size:16pt; text-align:right;">
انتخاب ویژگی مبتنی بر فراوانی ممکن است برخی واژه‌های پرتکرار را انتخاب کند که اطلاعات خاصی درباره کلاس ندارند، مانند روزهای هفته (Monday, Tuesday, ...) که در متن‌های خبری در همه کلاس‌ها پرتکرار هستند.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم تحلیل نظرات آمازون، کلماتی مانند "محصول"، "خرید" و "ارسال" در نظرات همه دسته‌ها (کتاب، الکترونیک، لباس) پرتکرار هستند، اما هیچ اطلاعات خاصی درباره دسته محصول به ما نمی‌دهند. این کلمات "نویز" می‌توانند دقت طبقه‌بندی را کاهش دهند.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 چه زمانی از این روش استفاده کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
وقتی هزاران ویژگی انتخاب می‌شوند، انتخاب ویژگی مبتنی بر فراوانی اغلب خوب عمل می‌کند. بنابراین، اگر دقت کمی زیربهینه قابل قبول باشد، انتخاب ویژگی مبتنی بر فراوانی می‌تواند جایگزین خوبی برای روش‌های پیچیده‌تر باشد.</p>
<p style="font-size:16pt; text-align:right;">
<b>کاربرد واقعی:</b> در سیستم تحلیل توییتر، وقتی میلیون‌ها توئیت باید به سرعت دسته‌بندی شوند، استفاده از روش مبتنی بر فراوانی می‌تواند یک راه حل سریع و کارآمد باشد، حتی اگر دقت کمی کمتر از روش‌های پیچیده‌تر باشد.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔍 مقایسه با روش‌های دیگر</p>
<p style="font-size:16pt; text-align:right;">
با این حال، شکل 13.8 موردی است که انتخاب ویژگی مبتنی بر فراوانی بسیار بدتر از MI و χ² عمل می‌کند و نباید استفاده شود. این نشان می‌دهد که سادگی همیشه بهترین راه حل نیست.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی مقالات علمی گوگل اسکالر، روش مبتنی بر فراوانی ممکن است کلمات عمومی مانند "تحقیق"، "نتایج" و "تحلیل" را انتخاب کند که در همه رشته‌ها پرتکرار هستند، در حالی که روش‌های MI و χ² می‌توانند کلمات تخصصی‌تری را شناسایی کنند که تمایز بهتری بین رشته‌ها ایجاد می‌کنند.</p>

```python
# پیاده‌سازی انتخاب ویژگی مبتنی بر فراوانی (Frequency-based Feature Selection)

import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import f1_score

def calculate_document_frequency(documents, labels, target_class):
    """
    محاسبه فراوانی سندی برای هر واژه در کلاس هدف
    
    فراوانی سندی: تعداد اسناد در کلاس c که شامل واژه t هستند
    این روش برای مدل برنولی مناسب است.
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    target_class: کلاسی که می‌خواهیم فراوانی واژه‌ها را برای آن محاسبه کنیم
    
    خروجی: دیکشنری که کلیدها واژه‌ها و مقادیر فراوانی سندی آن‌هاست
    """
    # فیلتر کردن اسناد مربوط به کلاس هدف
    target_docs = [doc for doc, label in zip(documents, labels) if label == target_class]
    
    # شمارش تعداد اسنادی که هر واژه در آن‌ها وجود دارد
    doc_freq = {}
    for doc in target_docs:
        unique_words = set(doc)  # هر واژه فقط یک بار در هر سند شمرش شود
        for word in unique_words:
            doc_freq[word] = doc_freq.get(word, 0) + 1
    
    return doc_freq

def calculate_collection_frequency(documents, labels, target_class):
    """
    محاسبه فراوانی مجموعه برای هر واژه در کلاس هدف
    
    فراوانی مجموعه: تعداد کل توکن‌های t که در اسناد کلاس c رخ می‌دهند
    این روش برای مدل چندجمله‌ای مناسب است.
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    target_class: کلاسی که می‌خواهیم فراوانی واژه‌ها را برای آن محاسبه کنیم
    
    خروجی: دیکشنری که کلیدها واژه‌ها و مقادیر فراوانی مجموعه آن‌هاست
    """
    # فیلتر کردن اسناد مربوط به کلاس هدف
    target_docs = [doc for doc, label in zip(documents, labels) if label == target_class]
    
    # شمارش تعداد کل تکرار هر واژه در تمام اسناد کلاس هدف
    all_words = [word for doc in target_docs for word in doc]
    coll_freq = Counter(all_words)
    
    return dict(coll_freq)

def select_features_by_frequency(documents, labels, k, target_class, frequency_type='document'):
    """
    انتخاب k ویژگی برتر بر اساس فراوانی
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    k: تعداد ویژگی‌هایی که می‌خواهیم انتخاب کنیم
    target_class: کلاس هدف برای انتخاب ویژگی
    frequency_type: نوع فراوانی ('document' برای فراوانی سندی، 'collection' برای فراوانی مجموعه)
    
    خروجی: لیستی از k واژه با بالاترین فراوانی
    """
    # محاسبه فراوانی بر اساس نوع درخواستی
    if frequency_type == 'document':
        term_frequencies = calculate_document_frequency(documents, labels, target_class)
    elif frequency_type == 'collection':
        term_frequencies = calculate_collection_frequency(documents, labels, target_class)
    else:
        raise ValueError("نوع فراوانی باید 'document' یا 'collection' باشد")
    
    # مرتب‌سازی واژه‌ها بر اساس فراوانی و انتخاب k واژه برتر
    sorted_terms = sorted(term_frequencies.items(), key=lambda x: x[1], reverse=True)
    selected_features = [term for term, freq in sorted_terms[:k]]
    
    return selected_features, [freq for term, freq in sorted_terms[:k]]

def compare_feature_selection_methods(documents, labels, target_class, k=10):
    """
    مقایسه روش‌های انتخاب ویژگی مبتنی بر فراوانی سندی و مجموعه
    
    این تابع نشان می‌دهد که چگونه دو نوع فراوانی می‌توانند ویژگی‌های متفاوتی را انتخاب کنند.
    """
    # انتخاب ویژگی با فراوانی سندی
    doc_features, doc_scores = select_features_by_frequency(documents, labels, k, target_class, 'document')
    
    # انتخاب ویژگی با فراوانی مجموعه
    coll_features, coll_scores = select_features_by_frequency(documents, labels, k, target_class, 'collection')
    
    # نمایش نتایج
    print(f"Top {k} features selected by document frequency for class '{target_class}':")
    for feature, score in zip(doc_features, doc_scores):
        print(f"{feature}: {score}")
    
    print(f"\nTop {k} features selected by collection frequency for class '{target_class}':")
    for feature, score in zip(coll_features, coll_scores):
        print(f"{feature}: {score}")
    
    # مقایسه ویژگی‌های مشترک
    common_features = set(doc_features) & set(coll_features)
    print(f"\nCommon features selected by both methods: {list(common_features)}")
    
    return doc_features, coll_features

def create_document_representation(documents, selected_features):
    """
    ایجاد نمایش سند بر اساس ویژگی‌های منتخب
    برای مدل چندجمله‌ای: تعداد تکرار هر ویژگی
    برای مدل برنولی: وجود یا عدم وجود هر ویژگی
    """
    # نمایش برای مدل چندجمله‌ای
    multinomial_rep = []
    for doc in documents:
        doc_rep = [doc.count(term) for term in selected_features]
        multinomial_rep.append(doc_rep)
    
    # نمایش برای مدل برنولی
    bernoulli_rep = []
    for doc in documents:
        doc_rep = [1 if term in doc else 0 for term in selected_features]
        bernoulli_rep.append(doc_rep)
    
    return np.array(multinomial_rep), np.array(bernoulli_rep)

def evaluate_frequency_feature_selection(documents, labels, target_class, feature_counts):
    """
    ارزیابی تأثیر انتخاب ویژگی مبتنی بر فراوانی بر دقت طبقه‌بندی
    
    این تابع دقت مدل‌های چندجمله‌ای و برنولی را برای تعداد مختلفی از ویژگی‌ها محاسبه می‌کند.
    نتایج نشان می‌دهد چگونه انتخاب ویژگی مبتنی بر فراوانی می‌تواند دقت را بهبود بخشد.
    """
    # تقسیم داده‌ها به آموزش و آزمون
    X_train_docs, X_test_docs, y_train, y_test = train_test_split(
        documents, labels, test_size=0.3, random_state=42
    )
    
    # تبدیل برچسب‌ها به باینری (کلاس هدف در مقابل بقیه)
    y_train_binary = np.array([1 if label == target_class else 0 for label in y_train])
    y_test_binary = np.array([1 if label == target_class else 0 for label in y_test])
    
    # استخراج واژگان از داده‌های آموزشی
    vocab = set()
    for doc in X_train_docs:
        vocab.update(doc)
    
    # محاسبه فراوانی سندی برای همه واژه‌ها
    doc_freq_scores = {}
    for term in vocab:
        doc_freq_scores[term] = calculate_document_frequency(X_train_docs, y_train, target_class).get(term, 0)
    
    # محاسبه فراوانی مجموعه برای همه واژه‌ها
    coll_freq_scores = {}
    for term in vocab:
        coll_freq_scores[term] = calculate_collection_frequency(X_train_docs, y_train, target_class).get(term, 0)
    
    # مرتب‌سازی واژه‌ها بر اساس فراوانی‌ها
    doc_sorted_terms = sorted(doc_freq_scores.items(), key=lambda x: x[1], reverse=True)
    coll_sorted_terms = sorted(coll_freq_scores.items(), key=lambda x: x[1], reverse=True)
    
    doc_terms = [term for term, score in doc_sorted_terms]
    coll_terms = [term for term, score in coll_sorted_terms]
    
    # ارزیابی برای تعداد مختلفی از ویژگی‌ها
    doc_multinomial_scores = []
    doc_bernoulli_scores = []
    coll_multinomial_scores = []
    coll_bernoulli_scores = []
    
    for k in feature_counts:
        # انتخاب k ویژگی برتر با فراوانی سندی
        doc_selected_features = doc_terms[:k]
        
        # ایجاد نمایش سند برای مدل‌ها
        X_train_multi, X_train_bern = create_document_representation(X_train_docs, doc_selected_features)
        X_test_multi, X_test_bern = create_document_representation(X_test_docs, doc_selected_features)
        
        # آموزش و ارزیابی مدل چندجمله‌ای با فراوانی سندی
        clf_multi = MultinomialNB()
        clf_multi.fit(X_train_multi, y_train_binary)
        y_pred_multi = clf_multi.predict(X_test_multi)
        f1_multi = f1_score(y_test_binary, y_pred_multi)
        doc_multinomial_scores.append(f1_multi)
        
        # آموزش و ارزیابی مدل برنولی با فراوانی سندی
        clf_bern = BernoulliNB()
        clf_bern.fit(X_train_bern, y_train_binary)
        y_pred_bern = clf_bern.predict(X_test_bern)
        f1_bern = f1_score(y_test_binary, y_pred_bern)
        doc_bernoulli_scores.append(f1_bern)
        
        # انتخاب k ویژگی برتر با فراوانی مجموعه
        coll_selected_features = coll_terms[:k]
        
        # ایجاد نمایش سند برای مدل‌ها
        X_train_multi, X_train_bern = create_document_representation(X_train_docs, coll_selected_features)
        X_test_multi, X_test_bern = create_document_representation(X_test_docs, coll_selected_features)
        
        # آموزش و ارزیابی مدل چندجمله‌ای با فراوانی مجموعه
        clf_multi = MultinomialNB()
        clf_multi.fit(X_train_multi, y_train_binary)
        y_pred_multi = clf_multi.predict(X_test_multi)
        f1_multi = f1_score(y_test_binary, y_pred_multi)
        coll_multinomial_scores.append(f1_multi)
        
        # آموزش و ارزیابی مدل برنولی با فراوانی مجموعه
        clf_bern = BernoulliNB()
        clf_bern.fit(X_train_bern, y_train_binary)
        y_pred_bern = clf_bern.predict(X_test_bern)
        f1_bern = f1_score(y_test_binary, y_pred_bern)
        coll_bernoulli_scores.append(f1_bern)
    
    return (doc_multinomial_scores, doc_bernoulli_scores, 
            coll_multinomial_scores, coll_bernoulli_scores)

# ایجاد مجموعه داده نمونه
np.random.seed(42)

# کلاس‌ها
classes = ['poultry', 'coffee', 'sports', 'china', 'uk', 'elections']

# واژگان مرتبط با هر کلاس
class_vocab = {
    'poultry': ['poultry', 'chicken', 'meat', 'avian', 'bird', 'farm', 'egg', 'broiler', 'inspection', 'disease'],
    'coffee': ['coffee', 'bean', 'roast', 'brew', 'colombia', 'brazil', 'export', 'crop', 'harvest', 'arabica'],
    'sports': ['soccer', 'game', 'match', 'team', 'player', 'score', 'win', 'league', 'cup', 'championship'],
    'china': ['china', 'chinese', 'beijing', 'yuan', 'shanghai', 'hong', 'kong', 'xinhua', 'province', 'taiwan'],
    'uk': ['london', 'uk', 'british', 'britain', 'england', 'pound', 'pence', 'stg', 'plc', 'english'],
    'elections': ['election', 'vote', 'candidate', 'campaign', 'poll', 'party', 'democratic', 'republican', 'president', 'ballot']
}

# واژگان عمومی (که در همه کلاس‌ها پرتکرار هستند)
general_vocab = ['report', 'said', 'year', 'today', 'new', 'government', 'market', 'company', 'people', 'world',
                'monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday', 'sunday']

# ایجاد اسناد
documents = []
labels = []
num_docs_per_class = 100

for cls in classes:
    for i in range(num_docs_per_class):
        # هر سند شامل 5-15 واژه است
        doc_length = np.random.randint(5, 16)
        doc = []
        
        # اضافه کردن 2-4 واژه از کلاس
        num_class_terms = np.random.randint(2, 5)
        class_terms = np.random.choice(class_vocab[cls], num_class_terms, replace=False)
        doc.extend(class_terms)
        
        # اضافه کردن واژگان عمومی
        num_general_terms = doc_length - num_class_terms
        general_terms = np.random.choice(general_vocab, num_general_terms, replace=True)
        doc.extend(general_terms)
        
        documents.append(doc)
        labels.append(cls)

# مقایسه روش‌های انتخاب ویژگی مبتنی بر فراوانی سندی و مجموعه
doc_features, coll_features = compare_feature_selection_methods(documents, labels, "poultry", k=10)

# ارزیابی تأثیر انتخاب ویژگی مبتنی بر فراوانی بر دقت طبقه‌بندی
feature_counts = [1, 5, 10, 20, 50, 100]
(doc_multi_scores, doc_bernoulli_scores, 
 coll_multi_scores, coll_bernoulli_scores) = evaluate_frequency_feature_selection(
    documents, labels, "poultry", feature_counts)

# نمایش نتایج
plt.figure(figsize=(12, 8))
plt.plot(feature_counts, doc_multi_scores, 'o-', label='Multinomial NB with Document Frequency')
plt.plot(feature_counts, doc_bernoulli_scores, 'x-', label='Bernoulli NB with Document Frequency')
plt.plot(feature_counts, coll_multi_scores, 's-', label='Multinomial NB with Collection Frequency')
plt.plot(feature_counts, coll_bernoulli_scores, 'd-', label='Bernoulli NB with Collection Frequency')
plt.xscale('log')
plt.xlabel('Number of Features')
plt.ylabel('F1 Score')
plt.title('Effect of Frequency-based Feature Selection on Classification Accuracy')
plt.legend()
plt.grid(True)
plt.show()

# نمایش نتایج عددی
print("\nنتایج ارزیابی:")
print("تعداد ویژگی\tF1 (چندجمله‌ای با فراوانی سندی)\tF1 (برنولی با فراوانی سندی)\tF1 (چندجمله‌ای با فراوانی مجموعه)\tF1 (برنولی با فراوانی مجموعه)")
for i, k in enumerate(feature_counts):
    print(f"{k}\t\t{doc_multi_scores[i]:.4f}\t\t\t{doc_bernoulli_scores[i]:.4f}\t\t\t{coll_multi_scores[i]:.4f}\t\t\t{coll_bernoulli_scores[i]:.4f}")

# نمایش ویژگی‌های منتخب با فراوانی سندی و مجموعه
print("\nویژگی‌های منتخب با فراوانی سندی برای کلاس poultry:")
for i, feature in enumerate(doc_features[:10]):
    print(f"{i+1}. {feature}")

print("\nویژگی‌های منتخب با فراوانی مجموعه برای کلاس poultry:")
for i, feature in enumerate(coll_features[:10]):
    print(f"{i+1}. {feature}")
```

در یک سیستم عملیاتی با تعداد زیادی طبقه‌بند، مطلوب است که یک مجموعه واحد از ویژگی‌ها انتخاب شود به جای مجموعه‌های متفاوت برای هر طبقه‌بند. این رویکرد به دلایل زیر مهم است:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>کاهش پیچیدگی محاسباتی و نیاز به حافظه</li>
<li>تسهیل نگهداری و به‌روزرسانی سیستم</li>
<li>امکان استفاده از نمایش سند مشترک برای چندین وظیفه</li>

<p style="font-size:16pt; text-align:right;">
دو رویکرد اصلی برای انتخاب ویژگی‌های مشترک برای چندین طبقه‌بند وجود دارد:</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 رویکرد اول: آماره χ² برای جدول n×2</p>
<p style="font-size:16pt; text-align:right;">
یک روش برای انجام این کار، محاسبه آماره χ² برای یک جدول n×2 است که در آن ستون‌ها وقوع و عدم وقوع یک واژه هستند و هر ردیف مربوط به یکی از کلاس‌ها است. سپس می‌توانیم k واژه با بالاترین آماره χ² را انتخاب کنیم.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی اسناد مایکروسافت، این رویکرد به تیم اجازه می‌دهد تا مجموعه‌ای از ویژگی‌ها را برای طبقه‌بندی انواع مختلف اسناد (Word، Excel، PowerPoint) انتخاب کند. این کار باعث می‌شود که نمایش سند مشترکی برای همه انواع اسناد وجود داشته باشد و جستجو و تحلیل آسان‌تر شود.</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔧 رویکرد دوم: ترکیب آماره‌های جداگانه</p>
<p style="font-size:16pt; text-align:right;">
روش متداول‌تر این است که آماره‌های انتخاب ویژگی ابتدا به صورت جداگانه برای هر کلاس در وظایف طبقه‌بندی دو کلاسی (c در مقابل c̄) محاسبه شده و سپس ترکیب شوند. دو روش ترکیب متداول وجود دارد:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>روش میانگین</strong>: یک شاخص واحد برای هر ویژگی محاسبه می‌شود (مثلاً با میانگین‌گیری از مقادیر A(t,c)) و k ویژگی با بالاترین شاخص‌ها انتخاب می‌شوند.</li>
<li><strong>روش ترکیبی</strong>: k/n ویژگی برتر برای هر یک از n طبقه‌بند انتخاب شده و سپس این n مجموعه در یک مجموعه ویژگی جهانی ترکیب می‌شوند.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان آمازون، تیم تحلیل داده‌ها ابتدا ویژگی‌های برتر را برای هر دسته (کتاب، الکترونیک، لباس) جداگانه محاسبه می‌کند. سپس با استفاده از روش ترکیبی، 20٪ از برترین ویژگی‌های هر دسته را انتخاب کرده و در یک مجموعه جهانی ترکیب می‌کند. این کار باعث می‌شود که سیستم بتواند به سرعت نظرات جدید را دسته‌بندی کند، حتی اگر برای هر دسته کمی دقت کمتری نسبت به استفاده از ویژگی‌های اختصاصی داشته باشد.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚖️ معامله بین دقت و کارایی</p>
<p style="font-size:16pt; text-align:right;">
دقت طبقه‌بندی اغلب کاهش می‌یابد هنگام انتخاب k ویژگی مشترک برای یک سیستم با n طبقه‌بند در مقابل n مجموعه متفاوت به اندازه k. اما حتی اگر دقت کاهش یابد، افزایش کارایی به دلیل نمایش سند مشترک ممکن است ارزش کاهش دقت را داشته باشد.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم فیلتر محتوای فیسبوک، که باید میلیاردها پست را در ثانیه طبقه‌بندی کند، استفاده از یک مجموعه ویژگی مشترک برای همه طبقه‌بندها (خشونت، هرزنامه، اخبار جعلی) ضروری است. حتی اگر این کار باعث کاهش 1-2 درصدی دقت شود، افزایش سرعت پردازش (تا 10 برابر) این کاهش دقت را جبران می‌کند و به فیسبوک اجازه می‌دهد تا محتوای مضر را به سرعت شناسایی و حذف کند.</p>

```python
# پیاده‌سازی انتخاب ویژگی برای چندین طبقه‌بند

import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
import math
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import f1_score

def calculate_chi2_for_multiple_classes(documents, labels, term):
    """
    محاسبه آماره χ² برای یک واژه در جدول n×2
    
    این تابع آماره χ² را برای یک واژه در جدول n×2 محاسبه می‌کند که در آن
    ستون‌ها وقوع و عدم وقوع واژه هستند و هر ردیف مربوط به یکی از کلاس‌هاست.
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    term: واژه مورد نظر
    
    خروجی: مقدار χ² برای واژه مورد نظر
    """
    N = len(documents)
    classes = list(set(labels))
    n_classes = len(classes)
    
    # ایجاد جدول n×2 برای واژه مورد نظر
    # هر ردیف: [تعداد اسناد در کلاس i که شامل واژه هستند, تعداد اسناد در کلاس i که شامل واژه نیستند]
    table = []
    
    for c in classes:
        N11 = N10 = N01 = N00 = 0
        
        for doc, label in zip(documents, labels):
            term_in_doc = 1 if term in doc else 0
            is_target_class = 1 if label == c else 0
            
            if term_in_doc == 1 and is_target_class == 1:
                N11 += 1
            elif term_in_doc == 1 and is_target_class == 0:
                N10 += 1
            elif term_in_doc == 0 and is_target_class == 1:
                N01 += 1
            else:  # term_in_doc == 0 and is_target_class == 0
                N00 += 1
        
        table.append([N11, N01])
    
    # محاسبه χ² برای جدول n×2
    chi2 = 0.0
    
    # محاسبه مجموع فراوانی‌های مشاهده شده در هر ستون
    col1_sum = sum(row[0] for row in table)  # مجموع وقوع واژه در همه کلاس‌ها
    col2_sum = sum(row[1] for row in table)  # مجموع عدم وقوع واژه در همه کلاس‌ها
    
    # محاسبه مجموع فراوانی‌های مشاهده شده در هر ردیف
    row_sums = [sum(row) for row in table]  # مجموع اسناد در هر کلاس
    
    # محاسبه فراوانی‌های مورد انتظار
    for i, row in enumerate(table):
        row_sum = row_sums[i]
        for j, cell in enumerate(row):
            col_sum = col1_sum if j == 0 else col2_sum
            expected = (row_sum * col_sum) / N
            
            if expected > 0:
                chi2 += ((cell - expected) ** 2) / expected
    
    return chi2

def calculate_feature_scores_per_class(documents, labels, score_func):
    """
    محاسبه امتیاز ویژگی‌ها برای هر کلاس به صورت جداگانه
    
    این تابع امتیاز ویژگی‌ها را برای هر کلاس به صورت جداگانه محاسبه می‌کند.
    این رویکرد برای روش ترکیبی مناسب است.
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    score_func: تابعی برای محاسبه امتیاز ویژگی (مانند اطلاعات متقابل یا χ²)
    
    خروجی: دیکشنری که کلیدها کلاس‌ها و مقادیر دیکشنری امتیاز ویژگی‌ها برای آن کلاس است
    """
    classes = list(set(labels))
    feature_scores = defaultdict(dict)
    
    # استخراج واژگان
    vocab = set()
    for doc in documents:
        vocab.update(doc)
    
    # محاسبه امتیاز برای هر واژه و هر کلاس
    for term in vocab:
        for c in classes:
            score = score_func(documents, labels, term, c)
            feature_scores[c][term] = score
    
    return feature_scores

def select_features_average_method(documents, labels, k, score_func):
    """
    انتخاب ویژگی با روش میانگین
    
    این روش یک شاخص واحد برای هر ویژگی محاسبه می‌کند (با میانگین‌گیری از مقادیر A(t,c))
    و k ویژگی با بالاترین شاخص‌ها را انتخاب می‌کند.
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    k: تعداد ویژگی‌هایی که می‌خواهیم انتخاب کنیم
    score_func: تابعی برای محاسبه امتیاز ویژگی (مانند اطلاعات متقابل یا χ²)
    
    خروجی: لیستی از k واژه با بالاترین امتیاز میانگین
    """
    # محاسبه امتیاز ویژگی‌ها برای هر کلاس
    feature_scores = calculate_feature_scores_per_class(documents, labels, score_func)
    classes = list(set(labels))
    
    # محاسبه میانگین امتیاز برای هر ویژگی
    avg_scores = {}
    for term in feature_scores[classes[0]].keys():  # همه واژه‌ها در همه کلاس‌ها وجود دارند
        avg_score = sum(feature_scores[c][term] for c in classes) / len(classes)
        avg_scores[term] = avg_score
    
    # انتخاب k واژه با بالاترین امتیاز میانگین
    selected_features = sorted(avg_scores.items(), key=lambda x: x[1], reverse=True)[:k]
    return [term for term, score in selected_features]

def select_features_union_method(documents, labels, k_per_class, score_func):
    """
    انتخاب ویژگی با روش ترکیبی
    
    این روش k/n ویژگی برتر برای هر یک از n طبقه‌بند انتخاب کرده
    و سپس این n مجموعه را در یک مجموعه ویژگی جهانی ترکیب می‌کند.
    
    پارامترها:
    documents: لیستی از اسناد (هر سند لیستی از کلمات است)
    labels: لیستی از برچسب‌های مربوط به اسناد
    k_per_class: تعداد ویژگی‌هایی که برای هر کلاس انتخاب می‌شوند
    score_func: تابعی برای محاسبه امتیاز ویژگی (مانند اطلاعات متقابل یا χ²)
    
    خروجی: لیستی از ویژگی‌های منتخب از ترکیب همه کلاس‌ها
    """
    # محاسبه امتیاز ویژگی‌ها برای هر کلاس
    feature_scores = calculate_feature_scores_per_class(documents, labels, score_func)
    classes = list(set(labels))
    
    # انتخاب k ویژگی برتر برای هر کلاس
    selected_per_class = {}
    for c in classes:
        class_scores = feature_scores[c]
        top_features = sorted(class_scores.items(), key=lambda x: x[1], reverse=True)[:k_per_class]
        selected_per_class[c] = [term for term, score in top_features]
    
    # ترکیب مجموعه‌های ویژگی
    combined_features = set()
    for features in selected_per_class.values():
        combined_features.update(features)
    
    return list(combined_features)

def mutual_information(documents, labels, term, target_class):
    """
    محاسبه اطلاعات متقابل بین یک واژه و یک کلاس
    """
    N = len(documents)
    
    # شمارش مقادیر N11, N10, N01, N00
    N11 = N10 = N01 = N00 = 0
    
    for doc, label in zip(documents, labels):
        term_in_doc = 1 if term in doc else 0
        is_target_class = 1 if label == target_class else 0
        
        if term_in_doc == 1 and is_target_class == 1:
            N11 += 1
        elif term_in_doc == 1 and is_target_class == 0:
            N10 += 1
        elif term_in_doc == 0 and is_target_class == 1:
            N01 += 1
        else:  # term_in_doc == 0 and is_target_class == 0
            N00 += 1
    
    # محاسبه مقادیر جانبی
    N1_ = N10 + N11  # تعداد اسناد شامل واژه
    N0_ = N00 + N01  # تعداد اسناد بدون واژه
    N_1 = N01 + N11  # تعداد اسناد در کلاس هدف
    N_0 = N00 + N10  # تعداد اسناد در کلاس غیرهدف
    
    # محاسبه اطلاعات متقابل
    mi = 0.0
    if N11 > 0:
        mi += (N11 / N) * math.log2((N * N11) / (N1_ * N_1))
    if N01 > 0:
        mi += (N01 / N) * math.log2((N * N01) / (N0_ * N_1))
    if N10 > 0:
        mi += (N10 / N) * math.log2((N * N10) / (N1_ * N_0))
    if N00 > 0:
        mi += (N00 / N) * math.log2((N * N00) / (N0_ * N_0))
    
    return mi

def chi2(documents, labels, term, target_class):
    """
    محاسبه آماره χ² برای یک واژه و یک کلاس
    """
    N = len(documents)
    
    # شمارش مقادیر N11, N10, N01, N00
    N11 = N10 = N01 = N00 = 0
    
    for doc, label in zip(documents, labels):
        term_in_doc = 1 if term in doc else 0
        is_target_class = 1 if label == target_class else 0
        
        if term_in_doc == 1 and is_target_class == 1:
            N11 += 1
        elif term_in_doc == 1 and is_target_class == 0:
            N10 += 1
        elif term_in_doc == 0 and is_target_class == 1:
            N01 += 1
        else:  # term_in_doc == 0 and is_target_class == 0
            N00 += 1
    
    # محاسبه χ² با استفاده از فرمول ساده‌شده
    numerator = (N11 + N10 + N01 + N00) * ((N11 * N00) - (N10 * N01))**2
    denominator = (N11 + N01) * (N11 + N10) * (N10 + N00) * (N01 + N00)
    
    if denominator == 0:
        return 0
    
    chi2_value = numerator / denominator
    return chi2_value

def evaluate_feature_selection_methods(documents, labels, k_values):
    """
    ارزیابی روش‌های مختلف انتخاب ویژگی برای چندین طبقه‌بند
    
    این تابع دقت طبقه‌بندی را برای روش‌های مختلف انتخاب ویژگی محاسبه می‌کند.
    نتایج نشان می‌دهد چگونه روش‌های مختلف می‌توانند دقت را تحت تأثیر قرار دهند.
    """
    # تقسیم داده‌ها به آموزش و آزمون
    X_train_docs, X_test_docs, y_train, y_test = train_test_split(
        documents, labels, test_size=0.3, random_state=42
    )
    
    # استخراج واژگان
    vocab = set()
    for doc in X_train_docs:
        vocab.update(doc)
    
    # ارزیابی برای تعداد مختلفی از ویژگی‌ها
    chi2_n2_scores = []
    mi_avg_scores = []
    mi_union_scores = []
    
    for k in k_values:
        # روش اول: χ² برای جدول n×2
        chi2_scores = {}
        for term in vocab:
            chi2_scores[term] = calculate_chi2_for_multiple_classes(X_train_docs, y_train, term)
        
        # انتخاب k ویژگی برتر
        selected_chi2 = sorted(chi2_scores.items(), key=lambda x: x[1], reverse=True)[:k]
        chi2_features = [term for term, score in selected_chi2]
        
        # ایجاد نمایش سند برای مدل چندجمله‌ای
        X_train = []
        X_test = []
        for doc in X_train_docs:
            doc_rep = [doc.count(term) for term in chi2_features]
            X_train.append(doc_rep)
        
        for doc in X_test_docs:
            doc_rep = [doc.count(term) for term in chi2_features]
            X_test.append(doc_rep)
        
        # آموزش و ارزیابی مدل
        clf = MultinomialNB()
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')
        chi2_n2_scores.append(f1)
        
        # روش دوم: میانگین اطلاعات متقابل
        selected_mi_avg = select_features_average_method(X_train_docs, y_train, k, mutual_information)
        
        # ایجاد نمایش سند برای مدل چندجمله‌ای
        X_train = []
        X_test = []
        for doc in X_train_docs:
            doc_rep = [doc.count(term) for term in selected_mi_avg]
            X_train.append(doc_rep)
        
        for doc in X_test_docs:
            doc_rep = [doc.count(term) for term in selected_mi_avg]
            X_test.append(doc_rep)
        
        # آموزش و ارزیابی مدل
        clf = MultinomialNB()
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')
        mi_avg_scores.append(f1)
        
        # روش سوم: ترکیب اطلاعات متقابل
        k_per_class = max(1, k // len(set(y_train)))  # تقسیم k بر تعداد کلاس‌ها
        selected_mi_union = select_features_union_method(X_train_docs, y_train, k_per_class, mutual_information)
        
        # ایجاد نمایش سند برای مدل چندجمله‌ای
        X_train = []
        X_test = []
        for doc in X_train_docs:
            doc_rep = [doc.count(term) for term in selected_mi_union]
            X_train.append(doc_rep)
        
        for doc in X_test_docs:
            doc_rep = [doc.count(term) for term in selected_mi_union]
            X_test.append(doc_rep)
        
        # آموزش و ارزیابی مدل
        clf = MultinomialNB()
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')
        mi_union_scores.append(f1)
    
    return chi2_n2_scores, mi_avg_scores, mi_union_scores

# ایجاد مجموعه داده نمونه
np.random.seed(42)

# کلاس‌ها
classes = ['poultry', 'coffee', 'sports', 'china', 'uk', 'elections']

# واژگان مرتبط با هر کلاس
class_vocab = {
    'poultry': ['poultry', 'chicken', 'meat', 'avian', 'bird', 'farm', 'egg', 'broiler', 'inspection', 'disease'],
    'coffee': ['coffee', 'bean', 'roast', 'brew', 'colombia', 'brazil', 'export', 'crop', 'harvest', 'arabica'],
    'sports': ['soccer', 'game', 'match', 'team', 'player', 'score', 'win', 'league', 'cup', 'championship'],
    'china': ['china', 'chinese', 'beijing', 'yuan', 'shanghai', 'hong', 'kong', 'xinhua', 'province', 'taiwan'],
    'uk': ['london', 'uk', 'british', 'britain', 'england', 'pound', 'pence', 'stg', 'plc', 'english'],
    'elections': ['election', 'vote', 'candidate', 'campaign', 'poll', 'party', 'democratic', 'republican', 'president', 'ballot']
}

# واژگان عمومی
general_vocab = ['report', 'said', 'year', 'today', 'new', 'government', 'market', 'company', 'people', 'world']

# ایجاد اسناد
documents = []
labels = []
num_docs_per_class = 100

for cls in classes:
    for i in range(num_docs_per_class):
        # هر سند شامل 5-15 واژه است
        doc_length = np.random.randint(5, 16)
        doc = []
        
        # اضافه کردن 2-4 واژه از کلاس
        num_class_terms = np.random.randint(2, 5)
        class_terms = np.random.choice(class_vocab[cls], num_class_terms, replace=False)
        doc.extend(class_terms)
        
        # اضافه کردن واژگان عمومی
        num_general_terms = doc_length - num_class_terms
        general_terms = np.random.choice(general_vocab, num_general_terms, replace=True)
        doc.extend(general_terms)
        
        documents.append(doc)
        labels.append(cls)

# مقایسه روش‌های انتخاب ویژگی برای چندین طبقه‌بند
print("مقایسه روش‌های انتخاب ویژگی برای چندین طبقه‌بند:")

# روش اول: χ² برای جدول n×2
chi2_scores = {}
for term in set(word for doc in documents for word in doc):
    chi2_scores[term] = calculate_chi2_for_multiple_classes(documents, labels, term)

# انتخاب 10 ویژگی برتر
selected_chi2 = sorted(chi2_scores.items(), key=lambda x: x[1], reverse=True)[:10]
chi2_features = [term for term, score in selected_chi2]

print("\n10 ویژگی برتر با استفاده از روش χ² برای جدول n×2:")
for i, (term, score) in enumerate(selected_chi2):
    print(f"{i+1}. {term}: {score:.4f}")

# روش دوم: میانگین اطلاعات متقابل
selected_mi_avg = select_features_average_method(documents, labels, 10, mutual_information)

print("\n10 ویژگی برتر با استفاده از روش میانگین اطلاعات متقابل:")
for i, term in enumerate(selected_mi_avg):
    print(f"{i+1}. {term}")

# روش سوم: ترکیب اطلاعات متقابل
k_per_class = 2  # 2 ویژگی برای هر کلاس
selected_mi_union = select_features_union_method(documents, labels, k_per_class, mutual_information)

print(f"\nویژگی‌های منتخب با استفاده از روش ترکیب اطلاعات متقابل (2 ویژگی برای هر کلاس):")
print(f"تعداد کل ویژگی‌ها: {len(selected_mi_union)}")
print(f"ویژگی‌ها: {selected_mi_union}")

# ارزیابی روش‌های مختلف انتخاب ویژگی
k_values = [10, 20, 50, 100]
chi2_n2_scores, mi_avg_scores, mi_union_scores = evaluate_feature_selection_methods(documents, labels, k_values)

# نمایش نتایج
plt.figure(figsize=(10, 6))
plt.plot(k_values, chi2_n2_scores, 'o-', label='χ² for n×2 table')
plt.plot(k_values, mi_avg_scores, 's-', label='Average MI')
plt.plot(k_values, mi_union_scores, 'd-', label='Union MI')
plt.xlabel('Number of Features')
plt.ylabel('Macro F1 Score')
plt.title('Comparison of Feature Selection Methods for Multiple Classifiers')
plt.legend()
plt.grid(True)
plt.show()

# نمایش نتایج عددی
print("\nنتایج ارزیابی:")
print("تعداد ویژگی\tχ² (جدول n×2)\tمیانگین MI\tترکیب MI")
for i, k in enumerate(k_values):
    print(f"{k}\t\t{chi2_n2_scores[i]:.4f}\t\t{mi_avg_scores[i]:.4f}\t\t{mi_union_scores[i]:.4f}")
```

<span id="section13.5.5" style="font-size:24pt; font-weight:bold; color:#1B263B;">13.5.5 مقایسه روش‌های انتخاب ویژگی: کدام روش برای کدام مشکل؟</span> ⚖️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 گوگل نیوز و چالش واژه‌های نادر: چگونه اخبار مهم را از اخبار عادی تشخیص دهیم؟</p>
<p style="font-size:16pt; text-align:right;">
تیم گوگل نیوز با چالشی جالب روبرو است: واژه‌ای مانند "پاندا" ممکن است فقط یک بار در اخبار مربوط به چین ظاهر شود، اما از نظر آماری بسیار معنادار باشد. آیا این واژه باید به عنوان یک ویژگی مهم برای شناسایی اخبار چین انتخاب شود؟</p>
<p style="font-size:16pt; text-align:right;">
این دقیقاً همان تفاوت بین اطلاعات متقابل (MI) و χ² است: χ² به دنبال واژه‌های معنادار از نظر آماری است، حتی اگر به ندرت ظاهر شوند، در حالی که MI به دنبال واژه‌هایی است که اطلاعات بیشتری را به همراه دارند، صرف نظر از تعداد تکرار آن‌ها.</p>
</div>

<p style="font-size:16pt; text-align:right;">
اطلاعات متقابل (MI) و χ² روش‌های نسبتاً متفاوتی برای انتخاب ویژگی هستند. تفاوت اصلی آن‌ها در نحوه برخورد با واژه‌های نادر است:</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 معناداری در مقابل اطلاعاتی بودن: تفاوت کلیدی χ² و MI</p>
<p style="font-size:16pt; text-align:right;">
استقلال واژه t و کلاس c گاهی می‌تواند با اطمینان بالایی رد شود حتی اگر t اطلاعات کمی درباره عضویت سند در c داشته باشد. این موضوع به ویژه برای واژه‌های نادر صادق است.</p>
<p style="font-size:16pt; text-align:right;">
اگر واژه‌ای یک بار در یک مجموعه بزرگ رخ دهد و آن یک وقوع در کلاس poultry باشد، این از نظر آماری معنادار است. اما یک وقوع واحد بر اساس تعریف نظریه اطلاعات، بسیار اطلاعاتی نیست. چون معیار χ² معناداری است، این روش واژه‌های نادر بیشتری را نسبت به اطلاعات متقابل انتخاب می‌کند (که اغلب شاخص‌های قابل اعتماد کمتری هستند).</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی مقالات علمی آرکایو (ArXiv)، واژه‌های بسیار تخصصی مانند "supersymmetry" ممکن است فقط در چند مقاله ظاهر شوند، اما از نظر χ² بسیار معنادار باشند. با این حال، این واژه‌ها ممکن است اطلاعات کمی برای طبقه‌بندی عمومی ارائه دهند، در حالی که MI به واژه‌های رایج‌تری مانند "quantum" یا "experiment" اهمیت بیشتری می‌دهد.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 عملکرد در عمل: آیا تفاوت‌ها واقعاً مهم هستند؟</p>
<p style="font-size:16pt; text-align:right;">
با وجود تفاوت‌ها، دقت طبقه‌بندی مجموعه‌های ویژگی منتخب با χ² و MI تفاوت سیستماتیک نشان نمی‌دهد. در اکثر مسائل طبقه‌بندی متن، چند شاخص قوی و بسیاری شاخص‌های ضعیف وجود دارد. تا زمانی که تمام شاخص‌های قوی و تعداد زیادی از شاخص‌های ضعیف انتخاب شوند، دقت انتظار می‌رود که خوب باشد. هر دو روش این کار را انجام می‌دهند.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان ایبی (eBay)، هر دو روش MI و χ² واژه‌هایی مانند "عالی"، "بد"، "ارسال سریع" و "بسته‌بندی" را به عنوان ویژگی‌های برتر شناسایی می‌کنند، حتی اگر اولویت‌بندی آن‌ها کمی متفاوت باشد. در نهایت، هر دو روش به دقت مشابهی در طبقه‌بندی نظرات مثبت و منفی منجر می‌شوند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📈 شکل 13.8: مقایسه عملکرد MI و χ²</p>
<p style="font-size:16pt; text-align:right;">
شکل 13.8 انتخاب ویژگی MI و χ² را برای مدل چندجمله‌ای مقایسه می‌کند. اوج کارایی تقریباً برای هر دو روش یکسان است. χ² این اوج را دیرتر، در 300 ویژگی، به دست می‌آورد، احتمالاً به این دلیل که ویژگی‌های نادر اما بسیار معناداری را که در ابتدا انتخاب می‌کند، تمام اسناد در کلاس را پوشش نمی‌دهند. با این حال، ویژگی‌های انتخابی بعدی (در محدوده 100-300) از کیفیت بهتری نسبت به ویژگی‌های انتخابی با MI برخوردارند.</p>
<p style="font-size:16pt; text-align:right;">
<b>تفسیر عملی:</b> در سیستم طبقه‌بندی اخبار توییتر، χ² ممکن است ابتدا واژه‌های نادر اما بسیار خاصی مانند "POTUS" (برای اخبار ریاست جمهوری) را انتخاب کند که فقط در چند توئیت ظاهر شده‌اند. این واژه‌ها به تنهایی پوشش کمی دارند، اما با افزایش تعداد ویژگی‌ها، χ² به واژه‌های رایج‌تر و کارآمدتر مانند "انتخابات" یا "رأی‌گیری" می‌رسد که پوشش بهتری ایجاد می‌کنند.</p>
</div>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🔄 محدودیت روش‌های حریصانه: مشکل ویژگی‌های زائد</p>
<p style="font-size:16pt; text-align:right;">
هر سه روش - MI، χ² و مبتنی بر فراوانی - روش‌های حریصانه هستند. ممکن است ویژگی‌هایی را انتخاب کنند که اطلاعات افزایشی نسبت به ویژگی‌های قبلی ندارند. در شکل 13.7، kog به عنوان هفتمین واژه انتخاب شده است با وجود اینکه با ویژگی قبلی hong همبستگی بالایی دارد و بنابراین زائد است.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی اخبار مربوط به هنگ کنگ، واژه‌های "hong" و "kong" تقریباً همیشه با هم ظاهر می‌شوند. اگر "hong" به عنوان یک ویژگی مهم انتخاب شده باشد، افزودن "kong" اطلاعات جدیدی به سیستم نمی‌دهد و فقط پیچیدگی را افزایش می‌دهد. این مشکل در شرکت‌هایی مانند بلومبرگ (Bloomberg) که باید اخبار مالی را به سرعت طبقه‌بندی کنند، اهمیت ویژه‌ای دارد.</p>
<p style="font-size:16pt; text-align:right;">
اگرچه چنین زائدی می‌تواند تأثیر منفی بر دقت داشته باشد، روش‌های غیرحریصانه (به بخش 13.7 مراجعه کنید) به دلیل هزینه محاسباتی به ندرست در طبقه‌بندی متن استفاده می‌شوند. در عمل، سادگی و سرعت روش‌های حریصانه اغلب بر مزایای بالقوه روش‌های پیچیده‌تر ارجحیت دارد.</p>

```python
# مقایسه روش‌های انتخاب ویژگی بر اساس تمرین 13.5

import math
import matplotlib.pyplot as plt
import numpy as np

def calculate_chi2_from_counts(N11, N10, N01, N00):
    """
    محاسبه امتیاز χ² بر اساس شمارش‌های Nij
    
    این تابع آماره χ² را بر اساس فرمول ساده‌شده محاسبه می‌کند.
    χ² معیاری از انحراف فراوانی مشاهده شده از فراوانی مورد انتظار تحت فرض استقلال است.
    
    پارامترها:
    N11: تعداد اسناد که شامل واژه و در کلاس هدف هستند
    N10: تعداد اسناد که شامل واژه اما در کلاس هدف نیستند
    N01: تعداد اسناد که در کلاس هدف هستند اما شامل واژه نیستند
    N00: تعداد اسناد که نه شامل واژه هستند و نه در کلاس هدف
    
    خروجی: مقدار χ²
    """
    N = N11 + N10 + N01 + N00
    numerator = N * ((N11 * N00) - (N10 * N01))**2
    denominator = (N11 + N01) * (N11 + N10) * (N10 + N00) * (N01 + N00)
    
    if denominator == 0:
        return 0  # اصلاح شده: return0 به return 0 تغییر یافت
    
    chi2_value = numerator / denominator
    return chi2_value

def calculate_mi_from_counts(N11, N10, N01, N00):
    """
    محاسبه امتیاز اطلاعات متقابل (MI) بر اساس شمارش‌های Nij
    
    این تابع اطلاعات متقابل را بر اساس تعریف نظریه اطلاعات محاسبه می‌کند.
    MI معیاری از اطلاعات مشترک بین دو متغیر تصادفی است.
    
    پارامترها:
    N11: تعداد اسناد که شامل واژه و در کلاس هدف هستند
    N10: تعداد اسناد که شامل واژه اما در کلاس هدف نیستند
    N01: تعداد اسناد که در کلاس هدف هستند اما شامل واژه نیستند
    N00: تعداد اسناد که نه شامل واژه هستند و نه در کلاس هدف
    
    خروجی: مقدار اطلاعات متقابل
    """
    N = N11 + N10 + N01 + N00
    
    # محاسبه مقادیر جانبی
    N1_ = N10 + N11  # تعداد اسناد شامل واژه
    N0_ = N00 + N01  # تعداد اسناد بدون واژه
    N_1 = N01 + N11  # تعداد اسناد در کلاس هدف
    N_0 = N00 + N10  # تعداد اسناد در کلاس غیرهدف
    
    # محاسبه اطلاعات متقابل
    mi = 0.0
    if N11 > 0:
        mi += (N11 / N) * math.log2((N * N11) / (N1_ * N_1))
    if N01 > 0:
        mi += (N01 / N) * math.log2((N * N01) / (N0_ * N_1))
    if N10 > 0:
        mi += (N10 / N) * math.log2((N * N10) / (N1_ * N_0))
    if N00 > 0:
        mi += (N00 / N) * math.log2((N * N00) / (N0_ * N_0))
    
    return mi

def calculate_document_frequency(N11, N10):
    """
    محاسبه فراوانی سندی (Document Frequency)
    
    این تابع تعداد اسنادی که واژه در آن‌ها وجود دارد را محاسبه می‌کند.
    این معیار برای مدل برنولی مناسب است.
    
    پارامترها:
    N11: تعداد اسناد که شامل واژه و در کلاس هدف هستند
    N10: تعداد اسناد که شامل واژه اما در کلاس هدف نیستند
    
    خروجی: فراوانی سندی واژه
    """
    return N11 + N10

def compare_feature_selection_methods(term_counts):
    """
    مقایسه روش‌های مختلف انتخاب ویژگی
    
    این تابع سه روش انتخاب ویژگی را مقایسه می‌کند:
    1. χ²
    2. اطلاعات متقابل (MI)
    3. فراوانی سندی
    
    پارامترها:
    term_counts: دیکشنری شامل شمارش‌های Nij برای هر واژه
    
    خروجی: دیکشنری شامل امتیاز هر واژه برای هر روش
    """
    results = {}
    
    for term, counts in term_counts.items():
        N11 = counts['N11']
        N10 = counts['N10']
        N01 = counts['N01']
        N00 = counts['N00']
        
        # محاسبه امتیازها برای هر روش
        chi2_score = calculate_chi2_from_counts(N11, N10, N01, N00)
        mi_score = calculate_mi_from_counts(N11, N10, N01, N00)
        doc_freq = calculate_document_frequency(N11, N10)
        
        results[term] = {
            'chi2': chi2_score,
            'mi': mi_score,
            'doc_freq': doc_freq,
            'N11': N11,
            'N10': N10,
            'N01': N01,
            'N00': N00
        }
    
    return results

def visualize_comparison(results):
    """
    مصورسازی مقایسه روش‌های انتخاب ویژگی
    
    این تابع نموداری برای مقایسه امتیاز واژه‌ها بر اساس روش‌های مختلف ایجاد می‌کند.
    """
    terms = list(results.keys())
    chi2_scores = [results[term]['chi2'] for term in terms]
    mi_scores = [results[term]['mi'] for term in terms]
    doc_freq_scores = [results[term]['doc_freq'] for term in terms]
    
    # ایجاد نمودار
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # نمودار مقایسه χ² و MI
    x = np.arange(len(terms))
    width = 0.35
    
    ax1.bar(x - width/2, chi2_scores, width, label='χ²')
    ax1.bar(x + width/2, mi_scores, width, label='Mutual Information')
    
    ax1.set_xlabel('Terms')
    ax1.set_ylabel('Score')
    ax1.set_title('Comparison of χ² and Mutual Information')
    ax1.set_xticks(x)
    ax1.set_xticklabels(terms)
    ax1.legend()
    
    # نمودار مقایسه فراوانی سندی
    ax2.bar(terms, doc_freq_scores, color='green')
    ax2.set_xlabel('Terms')
    ax2.set_ylabel('Document Frequency')
    ax2.set_title('Document Frequency Comparison')
    
    plt.tight_layout()
    plt.show()

# داده‌های تمرین 13.5 برای کلاس coffee
term_counts = {
    'brazil':    {'N00': 98012, 'N01': 102,  'N10': 1835, 'N11': 51},
    'council':   {'N00': 96322, 'N01': 133,  'N10': 3525, 'N11': 20},
    'producers': {'N00': 98524, 'N01': 119,  'N10': 1118, 'N11': 34},
    'roasted':   {'N00': 99824, 'N01': 143,  'N10': 23,   'N11': 10}
}

# محاسبه و مقایسه امتیازها
results = compare_feature_selection_methods(term_counts)

# نمایش نتایج در جدول
print("="*90)
print("مقایسه امتیاز ویژگی‌ها برای کلاس 'coffee' (تمرین 13.5)")
print("="*90)
print(f"{'واژه':<15} | {'فراوانی سندی':>15} | {'امتیاز χ²':>15} | {'امتیاز MI':>15}")
print("-"*90)

for term, scores in results.items():
    print(f"{term:<15} | {scores['doc_freq']:>15} | {scores['chi2']:>15.4f} | {scores['mi']:>15.6f}")

# انتخاب دو واژه برتر برای هر روش
print("\n" + "="*90)
print("دو واژه برتر انتخاب شده برای هر روش:")
print("="*90)

# (i) بر اساس χ²
sorted_chi2 = sorted(results.items(), key=lambda item: item[1]['chi2'], reverse=True)
selected_chi2 = [term for term, scores in sorted_chi2[:2]]
print(f"(i) χ²: {', '.join(selected_chi2)}")

# (ii) بر اساس اطلاعات متقابل
sorted_mi = sorted(results.items(), key=lambda item: item[1]['mi'], reverse=True)
selected_mi = [term for term, scores in sorted_mi[:2]]
print(f"(ii) اطلاعات متقابل: {', '.join(selected_mi)}")

# (iii) بر اساس فراوانی
sorted_freq = sorted(results.items(), key=lambda item: item[1]['doc_freq'], reverse=True)
selected_freq = [term for term, scores in sorted_freq[:2]]
print(f"(iii) فراوانی: {', '.join(selected_freq)}")

# مصورسازی نتایج
visualize_comparison(results)

# تحلیل نتایج
print("\n" + "="*90)
print("تحلیل نتایج:")
print("="*90)

# تحلیل چرا روش‌های مختلف واژه‌های متفاوتی را انتخاب می‌کنند
print("1. چرا χ² و MI واژه‌های متفاوتی را انتخاب می‌کنند؟")
print("   - χ² به واژه‌هایی که از نظر آماری معنادار هستند اهمیت می‌دهد، حتی اگر نادر باشند.")
print("   - MI به واژه‌هایی که اطلاعات بیشتری ارائه می‌دهند اهمیت می‌دهد، صرف نظر از معناداری آماری.")
print("   - برای مثال، 'brazil' امتیاز بالایی در χ² دارد چون همبستگی قوی با کلاس coffee دارد،")
print("     حتی اگر تعداد وقوع آن کم باشد.")
print("   - 'council' امتیاز بالایی در فراوانی سندی دارد چون در اسناد زیادی ظاهر می‌شود،")
print("     حتی اگر همبستگی آن با کلاس coffee ضعیف باشد.")

print("\n2. تفاوت‌های عملی در انتخاب ویژگی:")
print("   - χ² برای تشخیص واژه‌های نادر اما بسیار خاص مناسب است.")
print("   - MI برای تشخیص واژه‌هایی که اطلاعات عمومی و کاربردی ارائه می‌دهند مناسب است.")
print("   - فراوانی سندی ساده‌ترین روش است اما ممکن است واژه‌های عمومی و غیرخاص را انتخاب کند.")

print("\n3. کاربردهای عملی:")
print("   - در سیستم‌های فیلتر اسپم، χ² می‌تواند واژه‌های خاصی را شناسایی کند که در اسپم‌ها")
print("     به ندرت ظاهر می‌شوند اما بسیار مشخص‌کننده هستند.")
print("   - در سیستم‌های طبقه‌بندی اخبار، MI می‌تواند واژه‌های رایجی را انتخاب کند که در")
print("     انواع مختلف اخبار ظاهر می‌شوند و اطلاعات عمومی ارائه می‌دهند.")
print("   - در سیستم‌های طبقه‌بندی نظرات مشتریان، فراوانی سندی می‌تواند واژه‌هایی را")
print("     انتخاب کند که در نظرات زیاد تکرار می‌شوند، حتی اگر لزوماً مشخص‌کننده نباشند.")
```

<p style="font-size:16pt; text-align:right;">
تصور کنید در خبرگزاری رویترز کار می‌کنید و باید سیستمی طراحی کنید که هزاران خبر روزانه را به دسته‌های مختلفی (سیاست، ورزشی، اقتصادی، فرهنگی) اختصاص دهد. تیم رقیب شما در خبرگزاری گاردین از الگوریتم پیچیده‌تری استفاده می‌کند، اما شما باید سیستمی بسازید که هم سریع و هم قابل اعتماد باشد.</p>
<p style="font-size:16pt; text-align:right;">
این دقیقاً همان چالشی است که در ارزیابی طبقه‌بندی متن با آن روبرو هستیم: چگونه می‌توانیم بفهمیم که الگوریتم ما چقدر خوب کار می‌کند، به خصوص وقتی باید بین سرعت و دقت تعادل برقرار کنیم؟</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📏 چالش طبقه‌بندی چندکلاسی: یک در مقابل همه</p>
<p style="font-size:16pt; text-align:right;">
در ارزیابی طبقه‌بندی، ما به جای دقت (accuracy)، از معیارهای دقیق‌تری مانند <strong>دقت (precision)</strong>، <strong>بازیابی (recall)</strong> و <strong>F1</strong> استفاده می‌کنیم، به خصوص برای کلاس‌های کوچک که فراوانی کمی دارند.</p>
<p style="font-size:16pt; text-align:right;">
چرا دقت به تنهایی کافی نیست؟ تصور کنید سیستمی که 99% اسناد را به درستی طبقه‌بندی کند، اما این کار را با گفتن "هیچ سندی در کلاس X نیست" انجام دهد. اگر کلاس X فقط 1% از اسناد را تشکیل دهد، این سیستم 99% دقیق است اما کاملاً بی‌فایده است!</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم فیلتر محتوای یوتیوب، اگر سیستمی تمام محتوای مربوط به گروه‌های تروریستی را به درستی شناسایی کند اما در تشخیص محتوای مربوط به گروه‌های کوچک و جدید شکست بخورد، دقت بالای آن گمراه‌کننده است. در چنین مواردی، معیارهای F1 و بازیابی بهتر عملکرد واقعی سیستم را نشان می‌دهند.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚖️ میانگین‌گیری کلان در مقابل ریز: چگونه عملکرد کلی را گزارش کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
وقتی با چندین طبقه‌بند دو کلاسی کار می‌کنیم (مانند Reuters-21578 با 118 کلاس)، اغلب می‌خواهیم یک معیار تجمعی محاسبه کنیم که نتایج طبقه‌بندهای فردی را ترکیب کند. دو روش اصلی برای این کار وجود دارد:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>میانگین‌گیری کلان (Macroaveraging)</strong>: میانگین ساده روی کلاس‌ها که به هر کلاس وزن برابر می‌دهد.</li>
<li><strong>میانگین‌گیری ریز (Microaveraging)</strong>: تصمیمات طبقه‌بندی را در تمام کلاس‌ها ادغام کرده و سپس معیار را روی جدول ترکیبی محاسبه می‌کند.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
تفاوت بین این دو روش می‌تواند زیاد باشد. میانگین‌گیری کلان به هر کلاس وزن برابر می‌دهد، در حالی که میانگین‌گیری ریز به هر تصمیم طبقه‌بندی وزن برابر می‌دهد. از آنجایی که معیار F1 موارد منفی را نادیده می‌گیرد و عمدتاً توسط تعداد موارد مثبت تعیین می‌شود، کلاس‌های بزرگ در میانگین‌گیری ریز تسلط دارند.</p>

جدول 13.9 نتایج ارزیابی بیز ساده را در مقایسه با سایر طبقه‌بندها نشان می‌دهد. بیز ساده F1 میانگین‌گیری ریز 80% دارد که 9% کمتر از SVM (89%) است. با این حال، در کلاس‌های کوچک، عملکرد بیز ساده بسیار ضعیف‌تر است؛ F1 میانگین‌گیری کلان آن 13% کمتر از SVM است.</p>

<b>تفسیر عملی:</b> این نتایج نشان می‌دهند که اگرچه بیز ساده به اندازه SVM مؤثر نیست، اما سادگی و کارایی آن آن را به یک گزینه قوی برای مسائل طبقه‌بندی متن تبدیل کرده است. در دنیای واقعی، عواملی دیگری مانند ماهیت داده‌ها، تغییر مفهوم در طول زمان، و خطا در داده‌ها می‌توانند بر عملکرد طبقه‌بندها تأثیر گذارند.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان آمازون، بیز ساده ممکن است عملکرد خوبی در دسته‌بندی نظرات مثبت و منفی داشته باشد، اما در تشخیص زیردسته‌های خاص (مانند "کیفیت بسته‌بندی" در مقابل "کیفیت محصول") ضعیف عمل کند. در چنین مواردی، یک سیستم ترکیبی که از بیز ساده برای طبقه‌بندی کلی و از یک طبقه‌بند تخصصی برای زیردسته‌های خاص استفاده می‌کند، ممکن است نتایج بهتری ارائه دهد.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚠️ قوانین ارزیابی: چگونه نتایج معتبر بگیریم؟</p>
<p style="font-size:16pt; text-align:right;">
در ارزیابی طبقه‌بندی، جداسازی دقیق بین مجموعه آموزشی، توسعه و آزمون بسیار مهم است. هرگز نباید در حین توسعه سیستم به داده‌های آزمون دسترسی داشت یا روی آن‌ها برنامه اجرا کرد. در عوض، یک مجموعه توسعه برای آزمون در نظر گرفته شده و سیستم روی آن تنظیم می‌شود.</p>
<p style="font-size:16pt; text-align:right;">
این مجموعه توسعه همچنین <strong>داده‌های نگه‌داشته</strong> (held-out data) نامیده می‌شود. این رویکرد تضمین می‌کند که نتایج گزارش‌شده نماینده عملکرد واقعی سیستم در دنیای واقعی باشند، نه نتیجه تنظیم بیش از حد روی داده‌های آزمون.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در مسابقات کاگل (Kaggle) که شرکت‌ها در آن الگوریتم‌های خود را برای مسائل داده‌کاوی آزمایش می‌کنند، این قوانین به شدت رعایت می‌شود. شرکت‌کنندگان فقط به مجموعه آموزشی دسترسی دارند و نتایج الگوریتم‌هایشان فقط روی مجموعه آزمونی ارزیابی می‌شود که هرگز در حین توسعه دیده نشده است. این تضمین می‌کند که برنده مسابقه واقعاً بهترین الگوریتم را دارد، نه الگوریتمی که بیش از حد روی داده‌های آزمون تنظیم شده است.</p>

```python
# پیاده‌سازی معیارهای ارزیابی طبقه‌بندی متن

import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict

def calculate_precision(y_true, y_pred, class_label):
    """
    محاسبه دقت (Precision) برای یک کلاس خاص
    
    دقت نسبت اسنادی که به درستی به کلاس اختصاص داده شده‌اند به کل اسنادی که به آن کلاس اختصاص داده شده‌اند.
    
    پارامترها:
    y_true: لیست برچسب‌های واقعی
    y_pred: لیست برچسب‌های پیش‌بینی شده
    class_label: کلاسی که می‌خواهیم دقت را برای آن محاسبه کنیم
    
    خروجی: مقدار دقت برای کلاس مورد نظر
    """
    true_positives = sum(1 for true, pred in zip(y_true, y_pred) 
                          if true == class_label and pred == class_label)
    predicted_positives = sum(1 for pred in y_pred if pred == class_label)
    
    if predicted_positives == 0:
        return 0.0
    
    return true_positives / predicted_positives

def calculate_recall(y_true, y_pred, class_label):
    """
    محاسبه بازیابی (Recall) برای یک کلاس خاص
    
    بازیابی نسبت اسنادی که به درستی به کلاس اختصاص داده شده‌اند به کل اسنادی که باید به آن کلاس اختصاص داده می‌شدند.
    
    پارامترها:
    y_true: لیست برچسب‌های واقعی
    y_pred: لیست برچسب‌های پیش‌بینی شده
    class_label: کلاسی که می‌خواهیم بازیابی را برای آن محاسبه کنیم
    
    خروجی: مقدار بازیابی برای کلاس مورد نظر
    """
    true_positives = sum(1 for true, pred in zip(y_true, y_pred) 
                          if true == class_label and pred == class_label)
    actual_positives = sum(1 for true in y_true if true == class_label)
    
    if actual_positives == 0:
        return 0.0
    
    return true_positives / actual_positives

def calculate_f1(precision, recall):
    """
    محاسبه معیار F1
    
    F1 میانگین هارمونیک دقت و بازیابی است و معیاری متعادل برای ارزیابی عملکرد طبقه‌بند.
    
    پارامترها:
    precision: مقدار دقت
    recall: مقدار بازیابی
    
    خروجی: مقدار F1
    """
    if precision + recall == 0:
        return 0.0
    return 2 * (precision * recall) / (precision + recall)

def calculate_macro_avg(y_true, y_pred, classes):
    """
    محاسبه میانگین کلان (Macro-averaging) برای معیارها
    
    میانگین کلان به هر کلاس وزن برابر می‌دهد و برای ارزیابی عملکرد روی کلاس‌های کوچک مناسب است.
    
    پارامترها:
    y_true: لیست برچسب‌های واقعی
    y_pred: لیست برچسب‌های پیش‌بینی شده
    classes: لیست کلاس‌ها
    
    خروجی: دیکشنری شامل معیارهای میانگین کلان
    """
    precision_sum = 0.0
    recall_sum = 0.0
    f1_sum = 0.0
    
    for c in classes:
        precision = calculate_precision(y_true, y_pred, c)
        recall = calculate_recall(y_true, y_pred, c)
        f1 = calculate_f1(precision, recall)
        
        precision_sum += precision
        recall_sum += recall
        f1_sum += f1
    
    n_classes = len(classes)
    return {
        'precision': precision_sum / n_classes,
        'recall': recall_sum / n_classes,
        'f1': f1_sum / n_classes
    }

def calculate_micro_avg(y_true, y_pred, classes):
    """
    محاسبه میانگین ریز (Micro-averaging) برای معیارها
    
    میانگین ریز تصمیمات طبقه‌بندی را در تمام کلاس‌ها ادغام کرده و سپس معیار را محاسبه می‌کند.
    این معیار بیشتر تحت تأثیر کلاس‌های بزرگ قرار می‌گیرد.
    
    پارامترها:
    y_true: لیست برچسب‌های واقعی
    y_pred: لیست برچسب‌های پیش‌بینی شده
    classes: لیست کلاس‌ها
    
    خروجی: دیکشنری شامل معیارهای میانگین ریز
    """
    # ایجاد جدول ترکیبی برای تمام کلاس‌ها
    true_positives = 0
    predicted_positives = 0
    actual_positives = 0
    
    for c in classes:
        true_positives += sum(1 for true, pred in zip(y_true, y_pred) 
                               if true == c and pred == c)
        predicted_positives += sum(1 for pred in y_pred if pred == c)
        actual_positives += sum(1 for true in y_true if true == c)
    
    # محاسبه معیارهای ریز
    precision = true_positives / predicted_positives if predicted_positives > 0 else 0.0
    recall = true_positives / actual_positives if actual_positives > 0 else 0.0
    f1 = calculate_f1(precision, recall)
    
    return {
        'precision': precision,
        'recall': recall,
        'f1': f1
    }

def visualize_confusion_matrix(y_true, y_pred, classes):
    """
    مصورسازی ماتریس درهم‌ریختگی برای نمایش عملکرد طبقه‌بند
    
    این تابع ماتریس درهم‌ریختگی را نمایش می‌دهد که نشان می‌دهد چگونه طبقه‌بند
    اسناد را به کلاس‌های مختلف اختصاص داده است.
    
    پارامترها:
    y_true: لیست برچسب‌های واقعی
    y_pred: لیست برچسب‌های پیش‌بینی شده
    classes: لیست کلاس‌ها
    
    خروجی: نمایش ماتریس درهم‌ریختگی
    """
    n_classes = len(classes)
    # ایجاد ماتریس درهم‌ریختگی
    confusion_matrix = np.zeros((n_classes, n_classes), dtype=int)
    
    # پر کردن ماتریس
    class_to_index = {c: i for i, c in enumerate(classes)}
    for true, pred in zip(y_true, y_pred):
        true_idx = class_to_index[true]
        pred_idx = class_to_index[pred]
        confusion_matrix[true_idx, pred_idx] += 1
    
    # نمایش ماتریس
    # خطا: ax.figure.colorbar(im)
    # صحیح: fig.colorbar(im)
    fig, ax = plt.subplots(figsize=(8, 6))
    im = ax.imshow(confusion_matrix, interpolation='nearest', cmap=plt.cm.Blues)
    fig.colorbar(im)
    
    # تنظیم محورها
    ax.set_xticks(np.arange(n_classes))
    ax.set_yticks(np.arange(n_classes))
    ax.set_xticklabels(classes, rotation=45)
    ax.set_yticklabels(classes)
    ax.set_xlabel('Predicted Label')
    ax.set_ylabel('True Label')
    ax.set_title('Confusion Matrix')
    
    # اضافه کردن اعداد به هر سلول
    for i in range(n_classes):
        for j in range(n_classes):
            ax.text(j, i, str(confusion_matrix[i, j]),
                    ha="center", va="center", color="red")
    
    plt.tight_layout()
    plt.show()

def compare_macro_micro(y_true, y_pred, classes):
    """
    مقایسه میانگین‌گیری کلان و ریز با مثال‌های واقعی
    
    این تابع تفاوت بین دو روش میانگین‌گیری را با مثال‌های واقعی نشان می‌دهد.
    """
    # محاسبه معیارها
    macro_avg = calculate_macro_avg(y_true, y_pred, classes)
    micro_avg = calculate_micro_avg(y_true, y_pred, classes)
    
    # محاسبه دقت کلی
    accuracy = sum(1 for true, pred in zip(y_true, y_pred) if true == pred) / len(y_true)
    
    print("مقایسه روش‌های ارزیابی:")
    print("="*50)
    print(f"دقت کلی (Accuracy): {accuracy:.4f}")
    print(f"میانگین کلان دقت: {macro_avg['precision']:.4f}")
    print(f"میانگین ریز دقت: {micro_avg['precision']:.4f}")
    print()
    print(f"میانگین کلان بازیابی: {macro_avg['recall']:.4f}")
    print(f"میانگین ریز بازیابی: {micro_avg['recall']:.4f}")
    print()
    print(f"میانگین کلان F1: {macro_avg['f1']:.4f}")
    print(f"میانگین ریز F1: {micro_avg['f1']:.4f}")
    
    # نمایش ماتریس درهم‌ریختگی
    visualize_confusion_matrix(y_true, y_pred, classes)

def demonstrate_accuracy_problem():
    """
    نمایش مشکل دقت (Accuracy) برای مجموعه‌های نامتعادل
    
    این تابع با یک مثال نشان می‌دهد که چرا دقت ممکن است معیار گمراه‌کننده باشد.
    """
    print("نمایش مشکل دقت (Accuracy) برای مجموعه‌های نامتعادل:")
    print("="*60)
    
    # مجموعه نامتعادل: 99% اسناد در کلاس A و 1% در کلاس B
    y_true = ['A'] * 99 + ['B']  # 100 سند با برچسب واقعی
    y_pred_perfect = ['A'] * 99 + ['B']  # طبقه‌بند کامل
    y_pred_naive = ['A'] * 100  # طبقه‌بند ساده که همیشه A را پیش‌بینی می‌کند
    
    # محاسبه دقت برای هر طبقه‌بند
    accuracy_perfect = sum(1 for true, pred in zip(y_true, y_pred_perfect) if true == pred) / len(y_true)
    accuracy_naive = sum(1 for true, pred in zip(y_true, y_pred_naive) if true == pred) / len(y_true)
    
    print(f"دقت طبقه‌بند کامل: {accuracy_perfect:.2f}")
    print(f"دقت طبقه‌بند ساده (همیشه A را پیش‌بینی می‌کند): {accuracy_naive:.2f}")
    print()
    print("با وجود اینکه طبقه‌بند ساده هیچ سند B را به درستی تشخیص نمی‌دهد، دقت آن 99% است!")
    print("این مثال نشان می‌دهد که چرا دقت به تنهایی برای مجموعه‌های نامتعادل معیار مناسبی نیست.")
    print()
    
    # محاسبه F1 برای هر طبقه‌بند
    macro_perfect = calculate_macro_avg(y_true, y_pred_perfect, ['A', 'B'])
    macro_naive = calculate_macro_avg(y_true, y_pred_naive, ['A', 'B'])
    
    print("اما معیار F1 تفاوت را به خوبی نشان می‌دهد:")
    print(f"F1 طبقه‌بند کامل: {macro_perfect['f1']:.4f}")
    print(f"F1 طبقه‌بند ساده: {macro_naive['f1']:.4f}")
    print()
    print("F1 طبقه‌بند ساده برای کلاس B صفر است، زیرا هیچ سند B را به درستی تشخیص نمی‌دهد.")

# مثال 1: مقایسه طبقه‌بندهای مختلف بر اساس معیارهای ارزیابی
print("مثال 1: مقایسه طبقه‌بندهای مختلف بر اساس معیارهای ارزیابی")
print("="*60)

# داده‌های شبیه‌سازی شده از جدول 13.9 کتاب
# فرض می‌کنیم 1000 سند با توزیع کلاس‌های زیر داریم:
# کلاس earn: 300 سند
# کلاس acq: 200 سند
# کلاس money-fx: 100 سند
# کلاس grain: 150 سند
# کلاس crude: 120 سند
# کلاس corn: 80 سند
# کلاس trade: 50 سند

# ایجاد داده‌های واقعی و پیش‌بینی شده برای دو طبقه‌بند مختلف
np.random.seed(42)
classes = ['earn', 'acq', 'money-fx', 'grain', 'crude', 'corn', 'trade']

# ایجاد داده‌های واقعی
y_true = []
for c, count in [('earn', 300), ('acq', 200), ('money-fx', 100), 
                  ('grain', 150), ('crude', 120), ('corn', 80), ('trade', 50)]:
    y_true.extend([c] * count)

# ایجاد پیش‌بینی‌های طبقه‌بند 1 (Naive Bayes) - عملکرد بهتر روی کلاس‌های بزرگ
y_pred_nb = []
for c, count in [('earn', 300), ('acq', 200), ('money-fx', 100), 
                  ('grain', 150), ('crude', 120), ('corn', 80), ('trade', 50)]:
    if c in ['earn', 'acq']:  # عملکرد خوب روی کلاس‌های بزرگ
        y_pred_nb.extend([c] * int(count * 0.9))  # 90% دقت
    else:  # عملکرد ضعیف‌تر روی کلاس‌های کوچک
        y_pred_nb.extend([c] * int(count * 0.6))  # 60% دقت

# ایجاد پیش‌بینی‌های طبقه‌بند 2 (SVM) - عملکرد بهتر روی کلاس‌های کوچک
y_pred_svm = []
for c, count in [('earn', 300), ('acq', 200), ('money-fx', 100), 
                  ('grain', 150), ('crude', 120), ('corn', 80), ('trade', 50)]:
    if c in ['earn', 'acq']:  # عملکرد خوب روی کلاس‌های بزرگ
        y_pred_svm.extend([c] * int(count * 0.85))  # 85% دقت
    else:  # عملکرد بهتر روی کلاس‌های کوچک
        y_pred_svm.extend([c] * int(count * 0.75))  # 75% دقت

# محاسبه و مقایسه معیارها
print("ارزیابی Naive Bayes:")
macro_nb = calculate_macro_avg(y_true, y_pred_nb, classes)
micro_nb = calculate_micro_avg(y_true, y_pred_nb, classes)
print(f"میانگین کلان F1: {macro_nb['f1']:.4f}")
print(f"میانگین ریز F1: {micro_nb['f1']:.4f}")
print()

print("ارزیابی SVM:")
macro_svm = calculate_macro_avg(y_true, y_pred_svm, classes)
micro_svm = calculate_micro_avg(y_true, y_pred_svm, classes)
print(f"میانگین کلان F1: {macro_svm['f1']:.4f}")
print(f"میانگین ریز F1: {micro_svm['f1']:.4f}")
print()

# مقایسه دو طبقه‌بند
print("مقایسه دو طبقه‌بند:")
print(f"تفاوت در میانگین کلان F1: {macro_svm['f1'] - macro_nb['f1']:.4f}")
print(f"تفاوت در میانگین ریز F1: {micro_svm['f1'] - micro_nb['f1']:.4f}")

# نمایش مشکل دقت برای مجموعه‌های نامتعادل
demonstrate_accuracy_problem()

# مثال 2: مقایسه میانگین‌گیری کلان و ریز
print("\nمثال 2: مقایسه میانگین‌گیری کلان و ریز")
print("="*60)

# ایجاد یک مثال ساده با دو کلاس برای نمایش تفاوت میانگین‌گیری کلان و ریز
# کلاس 1: 100 سند (کلاس بزرگ)
# کلاس 2: 20 سند (کلاس کوچک)
y_true = ['class1'] * 100 + ['class2'] * 20

# داده‌های واقعی
y_true_example = ['class1'] * 100 + ['class2'] * 20

# پیش‌بینی‌ها
# 90 درست برای class1
# 5 درست برای class2
# 10 اشتباه (class2 واقعی، class1 پیش‌بینی)
# 15 اشتباه (class1 واقعی، class2 پیش‌بینی) - این بخش برای سادگی حذف شده و به class1 اضافه می‌شود
y_pred_example = ['class1'] * 90 + ['class2'] * 5 + ['class1'] * 10 + ['class1'] * 15 # طول 120


compare_macro_micro(y_true_example, y_pred_example, ['class1', 'class2'])

print("\nنتیجه‌گیری:")
print("="*60)
print("1. میانگین‌گیری کلان به هر کلاس وزن برابر می‌دهد و برای ارزیابی عملکرد روی کلاس‌های کوچک مناسب است.")
print("2. میانگین‌گیری ریز به هر تصمیم طبقه‌بندی وزن برابر می‌دهد و بیشتر تحت تأثیر کلاس‌های بزرگ قرار می‌گیرد.")
print("3. در مجموعه‌های نامتعادل، میانگین‌گیری کلان معیار قابل اعتمادتری برای ارزیابی عملکرد کلی است.")
```

<li>مدل یادگیری نظارت‌شده را تعریف کردیم: تابع طبقه‌بندی <code>γ : X → C</code> با استفاده از مجموعهٔ آموزش برچسب‌خورده.</li>
<li>مدل Naive Bayes را به‌عنوان یک روش آماری ساده و مؤثر معرفی کردیم و فرمول MAP را برای انتخاب کلاس بهینه استخراج کردیم.</li>
<li>دو نسخهٔ مدل NB را بررسی کردیم: مدل چندجمله‌ای (multinomial) و مدل برنولی (Bernoulli) — با تفاوت در نحوهٔ استفاده از فراوانی واژه و غیاب واژه‌ها.</li>
<li>مفهوم smoothing را برای مقابله با داده‌های پراکنده معرفی کردیم و از نرم‌سازی لاپلاس (add-one) استفاده کردیم.</li>
<li>ارتباط مدل NB با مدل زبانی چندجمله‌ای را نشان دادیم و پیچیدگی زمانی آموزش و تست را تحلیل کردیم.</li>
</ul>
