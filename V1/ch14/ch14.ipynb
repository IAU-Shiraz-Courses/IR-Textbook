<p style="font-size:18pt; text-align:center;">
<span id="section14.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">14 طبقه‌بندی فضای برداری: چگونه با بردارها، متون را درک کنیم؟</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 گوگل ترنسلیت و چالش طبقه‌بندی متون: چگونه زبان‌ها را از هم تشخیص دهیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم گوگل ترنسلیت کار می‌کنید و باید سیستمی طراحی کنید که متون‌ها را به زبان‌های مختلفی (انگلیسی، فرانسوی، آلمانی، اسپانیایی و...) طبقه‌بندی کند. چگونه می‌توانید با استفاده از بردارهای عددی، زبان متن را بدون نیاز به درک معنای آن تشخیص دهید؟</p>
<p style="font-size:16pt; text-align:right;">
این دقیقاً همان چالشی است که مدل فضای برداری با آن روبرو است: چگونه می‌توانیم با استفاده از ویژگی‌های عددی (مانند وزن tf-idf) به جای وجود یا عدم وجود واژه‌ها، متون‌ها را از هم تشخیص دهیم؟</p>
</div>

<p style="font-size:16pt; text-align:right;">
در فصل 13، با مدل بیز ساده کار کردیم که در آن هر سند به صورت دنباله‌ای از واژه‌ها یا بردار دوتایی نمایش داده می‌شد. در این فصل، از نمایش متفاوتی استفاده می‌کنیم: <strong>مدل فضای برداری</strong> که در فصل 6 توسعه داده شد. در این مدل، هر سند به صورت یک بردار با یک مولفه حقیقی برای هر واژه (معمولاً وزن tf-idf) نمایش داده می‌شود. بنابراین، فضای سند X، دامنه تابع طبقه‌بندی γ، به صورت ℝ<sup>|V|</sup> است.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧠 فرضیه تداوم(Contiguity hypothesis): مرزهای نامرئی بین زبان‌ها</p>
<p style="font-size:16pt; text-align:right;">
فرضیه اصلی در استفاده از مدل فضای برداری، <strong>فرضیه تداوم</strong> است: اسناد یک کلاس ناحیه‌ای پیوسته تشکیل می‌دهند و مناطق کلاس‌های مختلف با هم همپوشانی ندارند.</p>
<p style="font-size:16pt; text-align:right;">
در بسیاری از وظایف طبقه‌بندی متن، به خصوص نوعی که در فصل 13 با آن مواجه شدیم، کلاس‌ها را می‌توان با الگوهای واژگانی متمایز کرد. برای مثال، اسناد کلاس China معمولاً مقادیر بالایی در ابعاد مربوط به Chinese، Beijing و Mao دارند، در حالی که اسناد کلاس UK مقادیر بالایی برای London، British و Queen دارند. بنابراین، اسناد دو کلاس مناطق مجزایی را تشکیل می‌دهند که در شکل 14.1 نشان داده شده است و می‌توانیم مرزهایی را برای جداسازی آن‌ها ترسیم کرد.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی مقالات علمی آرکایو (ArXiv)، مقالات مربوط به هوش مصنوعی معمولاً واژه‌هایی مانند "neural network"، "deep learning" و "algorithm" دارند، در حالی که مقالات مربوط به فیزیک نظری واژه‌هایی مانند "quantum"، "particle" و "field" دارند. این دو مجموعه در فضای برداری مناطق مجزایی را تشکیل می‌دهند که با مرزهای خطی قابل تفکیک هستند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📐 اهمیت نمایش سند: چرا کلمات توقف مهم هستند؟</p>
<p style="font-size:16pt; text-align:right;">
اینکه آیا یک مجموعه از اسناد به مناطق مجزایی نگاشته می‌شود یا نه، به انتخاب‌هایی که برای نمایش سند انجام می‌دهیم بستگی دارد. برای مثال، وقوع مکرر ضمیر اول شخص "I" شواهدی برای کلاس "نوشته توسط یک نفر" است. اما این اطلاعات ممکن است از نمایش سند حذف شود اگر از لیست کلمات توقف استفاده کنیم.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی نظرات مشتریان آمازون، کلماتی مانند "the"، "is"، "and" و "a" تقریباً در همه نظرات وجود دارند اما اطلاعات کمی در مورد دسته محصول ارائه می‌دهند. حذف این کلمات به سیستم کمک می‌کند تا روی کلمات کلیدی مانند "excellent"، "poor" یا "packaging" تمرکز شود که اطلاعات بیشتری در مورد نظر مشتری ارائه می‌دهند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🎯 روش‌های طبقه‌بندی فضای برداری</p>
<p style="font-size:16pt; text-align:right;">
این فصل دو روش طبقه‌بندی فضای برداری را معرفی می‌کند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>طبقه‌بندی Rocchio</strong> (بخش 14.2): فضای برداری را به مناطقی تقسیم می‌کند که در مرکز مراکز یا نمونه‌های اولیه قرار دارند، یک برای هر کلاس که به صورت مرکز جرم تمام اسناد آن کلاس محاسبه می‌شود. طبقه‌بندی Rocchio ساده و کارآمد است، اما اگر کلاس‌ها تقریباً کروی با شعاع‌های مشابه نباشند، دقیق نیست.</li>
<li><strong>طبقه‌بندی kNN</strong> (بخش 14.3): به سند آزمون، کلاس اکثریت از k نزدیک‌ترین همسایه‌ها را اختصاص می‌دهد. kNN نیازی به آموزش صریح ندارد و می‌تواند مستقیماً از مجموعه آموزشی استفاده کند. اگر مجموعه آموزشی بزرگ باشد، kNN می‌تواند کلاس‌های غیرکروی و پیچیده‌تر را بهتر از Rocchio مدیریت کند.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی ایمیل‌های گوگل، Rocchio برای فیلتر اسپم‌های واضح (مانند ایمیل‌های تبلیغاتی که الگوی مشخصی دارند) استفاده می‌شود، در حالی که kNN برای فیلتر اسپم‌های پیچیده‌تر که الگوهای پنهان دارند یا از تکنیک‌های جدیدی استفاده می‌کنند، بهتر عمل می‌کند.</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">📈 طبقه‌بندهای خطی: مرزهای مستقیم در فضای برداری</p>
<p style="font-size:16pt; text-align:right;">
تعداد زیادی از طبقه‌بندهای متن را می‌توان به عنوان <strong>طبقه‌بندهای خطی</strong> در نظر گرفت - طبقه‌بندهایی که بر اساس ترکیب خطی ساده‌ای از ویژگی‌ها طبقه‌بندی می‌کنند. چنین طبقه‌بندهایی فضای ویژگی‌ها را به مناطقی جدا شده توسط هایپرپلن‌های تصمیم خطی تقسیم می‌کنند.</p>
<p style="font-size:16pt; text-align:right;">
وقتی از طبقه‌بندهای دو کلاسی برای مسائل با بیش از دو کلاس استفاده می‌کنیم، دو نوع وظیفه وجود دارد:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>وظایف یک-از-چند</strong> (one-of): سند باید دقیقاً به یکی از چندین کلاس انحصاری متقابل اختصاص داده شود.</li>
<li><strong>وظایف هر-از-چند</strong> (any-of): سند می‌تواند به هر تعدادی از کلاس‌ها اختصاص داده شود (در بخش 14.5 توضیح داده خواهد شد).</li>
</ul>

<p style="font-size:16pt; text-align:right;">
طبقه‌بندهای دو کلاسی وظایف هر-از-چند را حل می‌کنند و می‌توانند برای حل وظایف یک-از-چند ترکیب شوند.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section14.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">14.1 نمایش اسناد و معیارهای وابستگی در فضاهای برداری</span> 📊
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧠 بردارها به صورت نقطاطی: از ابعاد به مرزها</p>
<p style="font-size:16pt; text-align:right;">
برای نمایش ویژگی‌های بردار اسناد در طبقه‌بندی فضای برداری، این بردارها را به صورت نقاطی در یک صفحه نمایش می‌دهیم مانند مثال در شکل 14.1. در واقعیت، بردارهای سند بردارهای یکانی نرمال‌شده هستند که به سطح یک ابرکره (hypersphere) اشاره می‌کنند.</p>
<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig14.1.png" alt="fig14.1" style="width:800px; border:1px solid #ccc; border-radius:12px;">
</div>
<p style="font-size:16pt; text-align:right;">
شکل 14.2 تصویر ابعاد کوچک از سطح یک ابرکره را به یک صفحه نشان می‌دهد. سمت چپ: تصویر دایره‌ای دو بعدی به یک بعدی. برای نقاط x<sub>1</sub>, x<sub>2</sub>, x<sub>3</sub>, x<sub>4</sub>, x<sub>5</sub> در مختصات x، فاصله |x<sub>2</sub>x<sub>3</sub>| تقریباً 0.201 است و تنها 0.5% با |x| تفاوت دارد. سمت راست: تصویر مربوط به نیمکره سه بعدی به دو بعدی.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📏 فاصله‌ها و شباهت‌ها: چگونه "نزدیکی" را اندازه‌گیری کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصمیمات بسیاری از طبقه‌بندهای فضای برداری بر اساس مفهوم <strong>فاصله</strong> است، مثلاً هنگام محاسبه نزدیک‌ترین همسایه‌ها در طبقه‌بندی kNN. در این فصل، از فاصله اقلیدسی (Euclidean distance) به عنوان معیار فاصله اصلی استفاده می‌کنیم.</p>
<p style="font-size:16pt; text-align:right;">
همانطور که قبلاً مشاهده شد (تمرین 6.18، صفحه 131 کتاب)، ارتباط مستقیمی بین شباهت کسینوسی (cosine similarity) و فاصله اقلیدسی برای بردارهای نرمال‌شده وجود دارد. در طبقه‌بندی فضای برداری، به ندرست اهمیت ندارد که وابستگی دو سند به صورت شباهت یا فاصله بیان شود.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🎯 مراکزها و میانگین‌ها: نقاط مرجع در فضای برداری</p>
<p style="font-size:16pt; text-align:right;">
علاوه بر اسناد، <strong>مراکزها</strong> یا میانگین بردارها نیز نقش مهمی در طبقه‌بندی فضای برداری ایفا می‌کنند. مراکزها نرمال‌شده نیستند. برای بردارهای نرمال‌شده، حاصل ضرب داخلی، شباهت کسینوسی و فاصله اقلیدسی رفتار متفاوتی در حالت کلی دارند.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی مقالات علمی آرکایو (ArXiv)، مراکز هر حوزه (مانند یادگیری ماشین، فیزیک نظری، زیست‌شناسی) به صورت میانگین بردارهای تمام مقالات آن حوزه محاسبه می‌شود. وقتی مقاله جدیدی به سیستم داده می‌شود، با محاسبه فاصله آن از مراکزهای مختلف، می‌توان تشخیص داد که به کدام حوزه نزدیک‌تر است.</p>

<p style="font-size:16pt; text-align:right;">
این تصویر نشان می‌دهد که چگونه تصویر کردن به یک صفحه می‌تواند باعث تغییر شکل شود. برای مثال، تصویر دایره‌ای دو بعدی به یک بعدی به دلیل انحنای آن، کشیده شده و به صورت بیضی در می‌آید. این همان پدیده‌ای است که در سیستم‌های طبقه‌بندی متون رخ می‌دهد: وقتی زبان‌های مختلف را به فضای برداری کاهش می‌دهیم، ممکن است اطلاعات مهمی از دست بدهیم.</p>

این تمرین نشان می‌دهد که حتی برای مناطق کوچک، تصویر کردن به یک صفحه کمتر باعث تغییر شکل قابل توجهی می‌شود. برای مثال، تصویر نیمکره سه بعدی به دو بعدی می‌تواند باعث تغییر شکل 18% شود (مثال سمت راست).</p>
<p style="font-size:16pt; text-align:right;">
<b>تفسیر عملی:</b> در سیستم طبقه‌بندی متون گوگل، وقتی زبان‌های مختلف را به فضای برداری با ابعاد کمتر کاهش می‌دهیم (برای افزایش سرعت)، ممکن است برخی ویژگی‌های ظریف که زبان‌ها را از هم متمایز می‌کنند از دست بدهیم. این می‌تواند باعث کاهش دقت طبقه‌بندی شود، به خصوص برای زبان‌هایی با ویژگی‌های ظریف که ساختار پیچیده‌تری دارند.</p>

```python
# نمایش اسناد به صورت بردار و محاسبه معیارهای وابستگی در فضاهای برداری

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize

def euclidean_distance(v1, v2):
    """
    محاسبه فاصله اقلیدسی بین دو بردار
    
    این تابع فاصله اقلیدسی استاندارد بین دو بردار را محاسبه می‌کند.
    
    پارامترها:
    v1, v2: دو بردار که می‌خواهیم فاصله بین آن‌ها را محاسبه کنیم
    
    خروجی: مقدار فاصله اقلیدسی
    """
    return np.sqrt(np.sum((v1 - v2) ** 2))

def cosine_similarity(v1, v2):
    """
    محاسبه شباهت کسینوسی بین دو بردار
    
    این تابع شباهت کسینوسی را بین دو بردار محاسبه می‌کند.
    
    پارامترها:
    v1, v2: دو بردار که می‌خواهیم شباهت بین آن‌ها را محاسبه کنیم
    
    خروجی: مقدار شباهت کسینوسی (بین 0 و 1)
    """
    dot_product = np.dot(v1, v2)
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)
    
    if norm_v1 == 0 or norm_v2 == 0:
        return 0
    
    return dot_product / (norm_v1 * norm_v2)

def visualize_vectors_2d(vectors, labels, centroids=None):
    """
    نمایش بردارهای دو بعدی در یک صفحه
    
    این تابع بردارهای دو بعدی را به صورت نقاطی در یک صفحه نمایش می‌دهد.
    همچنین می‌تواند مراکزها را در صورت وجود نمایش دهد.
    
    پارامترها:
    vectors: لیست بردارهای دو بعدی
    labels: لیست برچسب‌های مربوط به هر بردار
    centroids: لیست مراکزها (اختیاری) برای نمایش
    
    خروجی: نمایش بردارها و مراکزها در یک صفحه
    """
    plt.figure(figsize=(10, 8))
    
    # رسم بردارها
    colors = ['red', 'blue', 'green', 'purple', 'orange']
    for i, (vec, label) in enumerate(zip(vectors, labels)):
        plt.quiver(0, 0, vec[0], vec[1], angles='xy', scale_units='xy', scale=1, 
                   color=colors[i % len(colors)], alpha=0.7, width=0.005)
        plt.scatter(vec[0], vec[1], s=100, c=colors[i % len(colors)], label=label)
    
    # رسم مراکزها در صورت وجود
    if centroids is not None:
        for i, (centroid, label) in enumerate(zip(centroids, ['Centroid 1', 'Centroid 2'])):
            plt.scatter(centroid[0], centroid[1], s=200, c='black', marker='x', label=label)
    
    plt.axhline(0, color='gray', linestyle='--', alpha=0.3)
    plt.axvline(0, color='gray', linestyle='--', alpha=0.3)
    plt.grid(True, alpha=0.3)
    plt.xlabel('first dimension')
    plt.ylabel('second dimension')
    plt.title('2D vector demonstration')
    plt.legend()
    plt.axis('equal')
    plt.show()

def angle_distortion(angle_deg):
    """
    محاسبه نسبت تغییر شکل α/sin(α) برای یک زاویه
    
    این تابع نسبت تغییر شکل α/sin(α) را برای یک زاویه محاسبه می‌کند.
    این معیار نشان می‌دهد که چقدر تصویر یک زاویه کوچک در یک صفحه تغییر شکل می‌دهد.
    
    پارامترها:
    angle_deg: زاویه بر حسب درجه
    
    خروجی: مقدار α/sin(α)
    """
    angle_rad = np.radians(angle_deg)
    return angle_rad / np.sin(angle_rad)

# تمرین 14.1: محاسبه تغییر شکل برای زوایای کوچک
print("تمرین 14.1: محاسبه تغییر شکل برای زوایای کوچک")
print("="*50)
angles = [1, 5, 10]
for angle in angles:
    distortion = angle_distortion(angle)
    print(f"زاویه {angle} درجه: نسبت تغییر شکل α/sin(α) = {distortion:.4f}")

# ایجاد چند بردار نمونه برای نمایش
vectors = np.array([
    [2, 3],    # کلاس 1
    [4, 1],    # کلاس 2
    [1, 4],    # کلاس 3
    [3, 2]     # کلاس 4
])
labels = ["class 1", "class 2", "class 3", "class 4"]

# محاسبه و نمایش بردارها
print("\nنمایش بردارهای دو بعدی:")
visualize_vectors_2d(vectors, labels)

# محاسبه شباهت کسینوسی و فاصله اقلیدسی بین بردارها
print("\nمعیارهای وابستگی بین بردارها:")
for i in range(len(vectors)):
    for j in range(i+1, len(vectors)):
        cos_sim = cosine_similarity(vectors[i], vectors[j])
        euc_dist = euclidean_distance(vectors[i], vectors[j])
        print(f"شباهت کسینوسی بین بردار {i+1} و {j+1}: {cos_sim:.4f}")
        print(f"فاصله اقلیدسی بین بردار {i+1} و {j+1}: {euc_dist:.4f}")

# نرمال‌سازی بردارها و مقایسه مجدد
normalized_vectors = normalize(vectors, norm='l2')
print("\nپس از نرمال‌سازی:")
for i in range(len(normalized_vectors)):
    for j in range(i+1, len(normalized_vectors)):
        cos_sim = cosine_similarity(normalized_vectors[i], normalized_vectors[j])
        euc_dist = euclidean_distance(normalized_vectors[i], normalized_vectors[j])
        print(f"شباهت کسینوسی بین بردار نرمال {i+1} و {j+1}: {cos_sim:.4f}")
        print(f"فاصله اقلیدسی بین بردار نرمال {i+1} و {j+1}: {euc_dist:.4f}")

# نمایش تفاوت بین بردارهای نرمال‌شده و نرمال‌نشده
print("\nمقایسه بردارهای نرمال‌شده و نرمال‌نشده:")
plt.figure(figsize=(12, 5))
orig_vectors = np.array([[2, 3], [4, 1]])  # بردارهای اصلی
norm_vectors = normalize(orig_vectors, norm='l2')  # بردارهای نرمال‌شده

# نمایش بردارهای اصلی
plt.subplot(1, 2, 1)
plt.quiver(0, 0, orig_vectors[0, 0], orig_vectors[0, 1], angles='xy', scale_units='xy', scale=1, 
           color='red', alpha=0.7, width=0.005)
plt.quiver(0, 0, orig_vectors[1, 0], orig_vectors[1, 1], angles='xy', scale_units='xy', scale=1, 
           color='blue', alpha=0.7, width=0.005)
plt.scatter(orig_vectors[:, 0], orig_vectors[:, 1], s=100, c=['red', 'blue'])
plt.title('Original vectors')
plt.axis('equal')
plt.grid(True, alpha=0.3)

# نمایش بردارهای نرمال‌شده
plt.subplot(1, 2, 2)
plt.quiver(0, 0, norm_vectors[0, 0], norm_vectors[0, 1], angles='xy', scale_units='xy', scale=1, 
           color='red', alpha=0.7, width=0.005)
plt.quiver(0, 0, norm_vectors[1, 0], norm_vectors[1, 1], angles='xy', scale_units='xy', scale=1, 
           color='blue', alpha=0.7, width=0.005)
plt.scatter(norm_vectors[:, 0], norm_vectors[:, 1], s=100, c=['red', 'blue'])
plt.title('Normalized vectors')
plt.axis('equal')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# نمایش تفاوت در معیارهای وابستگی
print("\nتفسیر عملی:")
print("="*50)
print("1. برای زوایای کوچک، تصویر کردن به یک صفحه کمتر باعث تغییر شکل قابل توجهی می‌شود.")
print("2. در سیستم طبقه‌بندی متون، وقتی زبان‌ها را به فضای برداری کاهش می‌دهیم، ممکن است برخی ویژگی‌های ظریف که زبان‌ها را از هم متمایز می‌کنند از دست بدهیم.")
print("3. این می‌تواند باعث کاهش دقت طبقه‌بندی شود، به خصوص برای زبان‌هایی با ویژگی‌های ظریف که ساختار پیچیده‌تری دارند.")
print("4. با این حال، سیستم‌هایی مانند گوگل ترنسلیت ممکن است بین سرعت و دقت تعادل برقرار کنند.")
```

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌍 گوگل نیوز و چالش طبقه‌بندی اخبار: چگونه اخبار مرتبط با یک کشور را از اخبار دیگر جدا کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم گوگل نیوز کار می‌کنید و باید سیستمی طراحی کند که اخبار مرتبط با کشورهای مختلف (چین، بریتانیا، آمریکا، ژاپن و...) را از هم جدا کند. این چالش پیچیده‌ای است زیرا اخبار مربوط به یک کشور ممکن است واژه‌های مشابهی با اخبار کشور دیگر داشته باشد.</p>
<p style="font-size:16pt; text-align:right;">
طبقه‌بندی Rocchio یکی از بهترین روش‌ها برای محاسبه مرزهای تصمیم خوب در فضای برداری است. این روش از <strong>مراکزها</strong> یا <strong>نمونه‌های اولیه</strong> برای تعریف کلاس‌ها استفاده می‌کند و سپس بر اساس فاصله، سند را به کلاسی اختصاص می‌دهد که مراکز آن به سند نزدیک‌تر است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی Rocchio، مراکز کلاس c به صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

 $$ \begin{aligned}
&\tilde{\mu}(c) = \frac{1}{|D_c|} \sum_{d \in D_c} \tilde{\mathbf{v}}(d)
\end{aligned}
$$ 
</div>

<p style="font-size:16pt; text-align:right;">
که در آن D<sub>c</sub> مجموعه اسنادی از کلاس c و ṽ(d) بردار نرمال‌شده سند d است.
</p>

<p style="font-size:16pt; text-align:right;">
مرز بین دو کلاس در طبقه‌بندی Rocchio، مجموعه‌ای از نقاط است که فاصله آن‌ها از دو مراکز برابر است. در فضای دو بعدی، این مرز یک خط است. در فضای M بعدی، این مرز یک هایپرپلن است که به صورت زیر تعریف می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

$$ \begin{aligned}
&\tilde{\mathbf{w}} \cdot \tilde{\mathbf{x}} = b
\end{aligned}
$$ 
</div>

<p style="font-size:16pt; text-align:right;">
که در آن ṽ(w) بردار نرمال‌شده هایپرپلن و b یک ثابت است. این تعریف شامل خطوط (هر خط در فضای 2D) و صفحات (هر صفحه در فضای 3D) و هایپرپلن‌ها در فضاهای با ابعاد بالاتر می‌شود.
</p>

</div>

<p style="font-size:16pt; text-align:right;">
الگوریتم به این صورت عمل می‌کند که برای هر کلاس c<sub>j</sub>:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>تمام اسناد مربوط به آن کلاس (D<sub>j</sub>) را جمع می‌کند.</li>
<li>مراکز ṽ(µ<sub>j</sub>) را به صورت میانگین بردارهای آن اسناد محاسبه می‌کند.</li>
<li>برای طبقه‌بندی یک سند جدید d، بردار آن ṽ(d) را محاسبه کرده و به هر مراکز ṽ(µ<sub>j</sub>) فاصله را محاسبه می‌کند.</li>
<li>سند را به کلاسی اختصاص می‌دهد که مراکز آن به سند نزدیک‌تر است.</li>

<p style="font-size:16pt; text-align:right;">
با این الگوریتم، سیستم می‌تواند تمام مراکزها و سپس قوانین طبقه‌بندی را برای طبقه‌بندی اسناد جدید محاسبه و ذخیره کند.
</p>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🎯 پیچیدگی‌های Rocchio: چه زمانی کار نمی‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
با وجود سادگی و کارایی بالا، طبقه‌بندی Rocchio محدودیت‌هایی دارد:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>کلاس‌های چندوجهی (multimodal)</strong>: Rocchio فرض می‌کند که کلاس‌ها تقریباً کروی با شعاع‌های مشابه هستند. این فرض برای بسیاری از وظایف طبقه‌بندی متن درست نیست. مثال کلاسی "a" در شکل 14.5 شامل دو خوشه مجزا با شعاع‌های مشابه است.</li>
<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig14.5.png" alt="fig14.5" style="width:600px; border:1px solid #ccc; border-radius:12px;">
</div>
<li><strong>توزیع نابرابر</strong>: Rocchio فقط از فاصله از مرکز استفاده می‌کند و جزئیات توزیع نقاط در هر کلاس را نادیده می‌گیرد. در شکل 14.3، Rocchio کلاس Kenya را به China اختصاص می‌دهد زیرا توزیع نقاط آن پراکندگی بیشتری دارد، در حالی که توزیع نقاط China فشرده‌تر است.</li>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی مقالات علمی آرکایو (ArXiv)، برخی حوزه‌ها مانند "یادگیری ماشین" ممکن است شامل زیرشاخه‌های متفاوتی (مانند یادگیری عمیق در مقابل یادگیری تقویتی) باشند که ساختار پیچیده‌ای دارند. طبقه‌بندی Rocchio ممکن است در تشخیص این زیرشاخه‌ها دچار شود و همه آن‌ها را به یک کلاس اصلی اختصاص دهد.</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚙️ پیچیدگی زمانی: چگونه کارایی را محاسبه کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
جدول 14.2 زمان آموزش و آزمون Rocchio را با روش‌های دیگر مقایسه می‌کند. پیچیدگی زمانی Rocchio به صورت زیر محاسبه می‌شود:
</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;

Θ(|D|L<sub>ave</sub> + |C||V|)
</div>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">

<p style="font-size:16pt; text-align:right;">
که در آن |D| تعداد اسناد، L<sub>ave</sub> میانگین تعداد توکن در هر سند، |C| تعداد کلاس‌ها و |V| تعداد ویژگی‌ها است. این نشان می‌دهد که پیچیدگی زمانی Rocchio خطی در اندازه مجموعه آموزشی است.
</p>

<p style="font-size:16pt; text-align:right;">
در مقایسه، طبقه‌بندی Rocchio و بیز ساده هر دو پیچیدگی زمانی خطی یکسانی دارند، زیرا هر دو روش تنها نیاز به محاسبه مراکزها و سپس مقایسه فاصله‌ها دارد. این در حالی که هر دو روش به اندازه مجموعه آموزشی خطی است.
</p>
</div>

</div>

```python
# پیاده‌سازی طبقه‌بندی Rocchio

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize

def train_rocchio(documents, labels):
    """
    آموزش طبقه‌بند Rocchio
    
    این تابع مراکزها را برای هر کلاس محاسبه کرده و برمی‌گرداند.
    """
    # تبدیل اسناد به بردارهای tf-idf
    vocab = set()
    for doc in documents:
        vocab.update(doc)
    
    vocab_list = list(vocab)
    # محاسبه tf برای هر سند و واژه
    tf_matrix = np.zeros((len(documents), len(vocab_list)))
    for i, doc in enumerate(documents):
        for j, term in enumerate(vocab_list):
            tf_matrix[i, j] = doc.count(term)
    
    # محاسبه df برای هر واژه
    df = np.zeros(len(vocab_list))
    for j, term in enumerate(vocab_list):
        df[j] = sum(1 for doc in documents if term in doc)
    
    # محاسبه idf
    n_docs = len(documents)
    idf = np.log(n_docs / (df + 1))
    
    # محاسبه tf-idf
    tfidf_matrix = tf_matrix * idf
    
    # نرمال‌سازی بردارها
    normalized_vectors = normalize(tfidf_matrix, norm='l2')
    
    # محاسبه مراکز برای هر کلاس
    classes = list(set(labels))
    centroids = {}
    
    for c in classes:
        # شاخص‌های مربوط به کلاس c
        class_indices = [i for i, label in enumerate(labels) if label == c]
        class_vectors = normalized_vectors[class_indices]
        
        # محاسبه مراکز به صورت میانگین بردارهای آن کلاس
        centroids[c] = np.mean(class_vectors, axis=0)
    
    # بازگرداندن مراکزها، واژگان، و مقادیر idf برای استفاده در مرحله آزمون
    return centroids, vocab_list, idf, normalized_vectors

def apply_rocchio(centroids, vocab_list, idf, test_doc):
    """
    اعمال طبقه‌بند Rocchio روی یک سند آزمون
    
    این تابع سند آزمون را به کلاسی اختصاص می‌دهد که مراکز آن به سند نزدیک‌تر است.
    """
    # تبدیل سند آزمون به بردار tf-idf
    test_vector = np.zeros(len(vocab_list))
    for j, term in enumerate(vocab_list):
        test_vector[j] = test_doc.count(term)
    
    # محاسبه tf-idf با استفاده از idf از مرحله آموزش
    tfidf_vector = test_vector * idf
    
    # نرمال‌سازی بردار آزمون
    normalized_test_vector = normalize(tfidf_vector.reshape(1, -1), norm='l2')
    
    # محاسبه فاصله از هر مراکز
    min_distance = float('inf')
    predicted_class = None
    
    for c, centroid in centroids.items():
        distance = np.linalg.norm(normalized_test_vector - centroid)
        if distance < min_distance:
            min_distance = distance
            predicted_class = c
    
    return predicted_class, normalized_test_vector

# مثال کاربردی: طبقه‌بندی اخبار مربوط به کشورها
documents = [
    ["Chinese", "Beijing", "Mao"],
    ["Chinese", "Shanghai", "Mao"],
    ["Chinese", "Taiwan", "Mao"],
    ["Chinese", "Tokyo", "Mao"],
    ["British", "London", "Queen"],
    ["British", "London", "Queen"],
    ["Kenyan", "Nairobi", "tourism"],
    ["Kenyan", "Nairobi", "safari"]
]
labels = ["China", "China", "China", "China", "UK", "UK", "Kenya", "Kenya"]

# آموزش طبقه‌بند Rocchio
centroids, vocab, idf_values, normalized_vectors = train_rocchio(documents, labels)

# نمایش مراکزها
print("مراکزهای محاسبه شده برای هر کلاس:")
for c, centroid in centroids.items():
    print(f"کلاس {c}: {centroid}")

# اعمال روی سند آزمون
test_doc = ["Chinese", "Beijing", "economy"]
predicted_class, normalized_test_vector = apply_rocchio(centroids, vocab, idf_values, test_doc)
print(f"\nسند آزمون: {test_doc}")
print(f"کلاس پیش‌بینی شده: {predicted_class}")

# نمایش فضای برداری و مرزهای تصمیم
plt.figure(figsize=(12, 8))
colors = {'China': 'red', 'UK': 'blue', 'Kenya': 'green'}

# نمایش اسناد آموزشی
for i, label in enumerate(labels):
    plt.scatter(normalized_vectors[i, 0], normalized_vectors[i, 1], 
               c=colors[label], alpha=0.7, label=label)

# نمایش مراکزها
for c, centroid in centroids.items():
    plt.scatter(centroid[0], centroid[1], s=200, c='black', marker='x', 
               label=f"Centroid {c}")

# نمایش سند آزمون
plt.scatter(normalized_test_vector[0, 0], normalized_test_vector[0, 1], 
           c='purple', marker='*', s=200, label="Test document")

# رسم مرزهای تصمیم (فقط بین دو کلاس اول)
centroid_china_2d = np.array([centroids["China"][0], centroids["China"][1]])
centroid_uk_2d = np.array([centroids["UK"][0], centroids["UK"][1]])

# محاسبه نقطه میانی خط
midpoint = (centroid_china_2d + centroid_uk_2d) / 2
# محاسبه بردار جهت
direction = centroid_uk_2d - centroid_china_2d
# محاسبه بردار عمود بر خط
normal = np.array([-direction[1], direction[0]])
normal = normal / np.linalg.norm(normal)

# رسم بخشی از خط
line_length = 0.8
line_start = midpoint - normal * line_length / 2
line_end = midpoint + normal * line_length / 2

plt.plot([line_start[0], line_end[0]], [line_start[1], line_end[1]], 
        '--', alpha=0.5, color='gray')

plt.xlabel('first dimension')
plt.ylabel('second dimension')
plt.title('Rocchio classification for news articles')
plt.legend()
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.show()

# نمایش پیچیدگی Rocchio: کلاس‌های چندوجهی
print("\nپیچیدگی‌های Rocchio:")
print("1. کلاس‌های چندوجهی (multimodal): Rocchio فرض می‌کند که کلاس‌ها تقریباً کروی با شعاع‌های مشابه هستند.")
print("2. توزیع نابر: Rocchio فقط از فاصله از مرکز استفاده می‌کند و جزئیات توزیع نقاط در هر کلاس را نادیده می‌گیرد.")
print("3. مثال واقعی: در سیستم طبقه‌بندی مقالات علمی، برخی حوزه‌ها ممکن است شامل زیرشاخه‌های متفاوتی (مانند یادگیری ماشین) باشند که ساختار پیچیده‌ای دارند.")

# نمایش پیچیدگی زمانی
print("\nپیچیدگی زمانی:")
print("پیچیدگی زمانی Rocchio به صورت زیر محاسبه می‌شود:")
print("Θ(|D|Lave + |C||V|)")
print("که در آن |D| تعداد اسناد، Lave میانگین تعداد توکن در هر سند، |C| تعداد کلاس‌ها و |V| تعداد ویژگی‌ها است.")
print("این نشان می‌دهد که پیچیدگی زمانی Rocchio خطی در اندازه مجموعه آموزشی است.")
print("در مقایسه، طبقه‌بندی Rocchio و بیز ساده هر دو پیچیدگی زمانی خطی یکسانی دارند.")
```

<p style="font-size:18pt; text-align:center;">
<span id="section14.3" style="font-size:24pt; font-weight:bold; color:#1B263B;">14.3 طبقه‌بندی k نزدیک‌ترین همسایه (kNN): مرزهای پیچیده با جستجوی محلی</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 اسپاتیفای و چالش جستجوی تصاویر: چگونه تصاویر مشابه را پیدا کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم اسپاتیفای (Spotify) کار می‌کنید و باید سیستمی طراحی کنید که موسیقی مشابه را بر اساس ویژگی‌های تصویری (رنگ، ریتم، ساز، ...) پیدا کند. این دقیقاً همان چالشی است که طبقه‌بندی k نزدیک‌ترین همسایه (kNN) با آن روبرو است: چگونه می‌توانیم با استفاده از شباهت بین بردارها، اسناد مشابه را پیدا کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
برخلاف Rocchio که از مراکزها استفاده می‌کند، kNN به طور مستقیم از مجموعه آموزشی استفاده می‌کند و نیازی به مرحله آموزش صریح ندارد. این روش به خصوص برای مسائلی که داده‌های جدید به طور مداوم در حال تغییر هستند، بسیار مفید است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی k نزدیک‌ترین همسایه (kNN)، هر سند آزمون به کلاسی اختصاص داده می‌شود که اکثریت از k نزدیک‌ترین همسایه‌های آن در مجموعه آموزشی به آن تعلق دارند. به طور معادل، ما k نزدیک‌ترین همسایه‌ها را پیدا می‌کنیم و سپس سند را به کلاس اکثریت اختصاص می‌دهیم.</p>
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📐 الگوریتم kNN بر اساس فرضیه تداوم</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig14.7.bmp" alt="" style="width:800px; border:1px solid #ccc; border-radius:12px;">
</div>

<p style="font-size:16pt; text-align:right;">
الگوریتم kNN بر اساس فرضیه تداوم است که اسناد نزدیک به هم در فضای برداری به هم تعلق دارند. این روش به طور مستقیم از مجموعه آموزشی استفاده می‌کند و نیازی به مرحله آموزش صریح ندارد.</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم kNN به این صورت عمل می‌کند که برای هر سند آزمون d، k نزدیک‌ترین همسایه‌های آن در مجموعه آموزشی را پیدا می‌کند:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>فاصله هر سند آزمون d از هر کلاس c محاسبه می‌شود.</li>
<li>سند به کلاسی اختصاص داده می‌شود که دارای کمترین فاصله.</li>
</ol>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم اسپاتیفای، kNN برای پیدا کردن موسیقی مشابه بر اساس ویژگی‌های صوتی بسیار مفید است. این سیستم می‌تواند دو آهنگ با ریتم‌های متفاوت را به عنوان "مشابه" شناسایی کند، حتی اگر متون آن‌ها کاملاً متفاوت باشند. این کار به ویژگی‌های صوتی که مستقل از زبان و فرهنگ موسیقی هستند تکیه می‌دهد.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📈 شکل 14.6: تقسیم فضای برداری به موزاییک‌های ورونویی</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig14.6.bmp" alt="" style="width:800px; border:1px solid #ccc; border-radius:12px;">
</div>

<p style="font-size:16pt; text-align:right;">
شکل 14.6 تقسیم فضای برداری را به موزاییک‌های ورونویی (Voronoi tessellation) تقسیم می‌کند. در این تقسیم، هر سند به یک سلول اختصاص داده می‌شود و هر سلول شامل اسنادی است که به آن کلاس تعلق دارند. مرزهای بین سلول‌ها هایپرپلن‌ها هستند که فضای بین سلول‌ها را به مناطقی تقسیم می‌کنند.</p>
<p style="font-size:16pt; text-align:right;">
در M بعدی، هایپرپلن‌ها (M-1) فضای بین سلول‌ها را تقسیم می‌کنند که به صورت چندوجهی (M-1) بعدی تقسیم می‌شوند. این تقسیم‌ها می‌توانند ساختارهای پیچیده‌ای مانند کره‌ها، دونات و اشکال را نمایش دهند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🎯 پیچیدگی‌های kNN: چه زمانی کار نمی‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
با وجود سادگی و کارایی بالا، طبقه‌بندی kNN محدودیت‌هایی دارد:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>حساسیت بالا در مجموعه‌های بزرگ</strong>: برای مجموعه‌های بزرگ، محاسبه فاصله برای هر سند آزمون ممکن است زمان‌بر باشد.</li>
<li><strong>حساسیت ذخیره</strong>: kNN به حافظه نیاز دارد تا تمام اسناد آموزشی را ذخیره کند.</li>
<li><strong>عدم تعادل وزن‌دهی ویژگی‌ها</strong>: ویژگی‌هایی با مقادیر بزرگ ممکن است بر ویژگی‌های با مقادیر کوچک تأثیر بیشتری داشته باشند.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی تصاویر پینترست (Pinterest)، kNN برای پیدا کردن تصاویر مشابه بر اساس ویژگی‌های بصری ممکن است بسیار کارآمد باشد. با این حال، سیستم می‌تواند تصاویر مشابه را حتی اگر برچسب‌ها یا توضیحات متفاوتی داشته باشند پیدا کند.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">⚙️ پیچیدگی زمانی: چگونه کارایی را بهبود آوریم؟</p>
<p style="font-size:16pt; text-align:right;">
پیچیدگی زمانی kNN به صورت زیر محاسبه می‌شود:
</p>

  <div style="

Θ(|D|L<sub>ave</sub> + |C||V|)

<p style="font-size:16pt; text-align:right;">
که در آن |D| تعداد اسناد، L<sub>ave</sub> میانگین تعداد توکن در هر سند، |C| تعداد کلاس‌ها و |V| تعداد ویژگی‌ها است. این نشان می‌دهد که پیچیدگی زمانی kNN خطی در اندازه مجموعه آموزشی است.
</p>

<p style="font-size:16pt; text-align:right;">
در مقایسه، پیچیدگی زمانی kNN معمولاً از Rocchio بدتر است زیرا نیاز به محاسبه و ذخیره تمام اسناد آموزشی دارد. با این حال، پیچیدگی زمانی Rocchio خطی در اندازه مجموعه آموزشی است، زیرا هر دو روش تنها نیاز به محاسبه مراکزها و سپس مقایسه فاصله‌ها دارد.</p>

```python
# پیاده‌سازی طبقه‌بندی k نزدیک‌ترین همسایه (kNN)

import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.decomposition import PCA

def euclidean_distance(v1, v2):
    """
    محاسبه فاصله اقلیدسی بین دو بردار
    
    این تابع فاصله اقلیدسی استاندارد بین دو بردار را محاسبه می‌کند.
    
    پارامترها:
    v1, v2: دو بردار که می‌خواهیم فاصله بین آن‌ها را محاسبه کنیم.
    
    خروجی: مقدار فاصله اقلیدسی
    """
    return np.sqrt(np.sum((v1 - v2) ** 2))

def find_k_nearest_neighbors(train_docs, train_labels, test_doc, k, vocab_list):
    """
    پیدا کردن k نزدیک‌ترین همسایه‌ها برای یک سند آزمون
    """
    # محاسبه tf برای سند آزمون
    test_vector = np.zeros(len(vocab_list))
    for j, term in enumerate(vocab_list):
        test_vector[j] = test_doc.count(term)
    
    # محاسبه فاصله از هر سند آزمون به تمام اسناد آموزشی
    distances = []
    for i, (doc, label) in enumerate(zip(train_docs, train_labels)):
        # تبدیل سند آموزشی به بردار tf
        train_vector = np.zeros(len(vocab_list))
        for j, term in enumerate(vocab_list):
            train_vector[j] = doc.count(term)
        
        # محاسبه فاصله اقلیدسی
        distance = euclidean_distance(test_vector, train_vector)
        distances.append((distance, label))
    
    # مرتب‌سازی بر اساس فاصله و انتخاب k نزدیک‌ترین
    distances.sort(key=lambda x: x[0])
    k_nearest = distances[:k]
    
    # شمارش آرا برای هر کلاس در k نزدیک‌ترین همسایه‌ها
    class_votes = Counter([label for _, label in k_nearest])
    
    # انتخاب کلاس با بیشترین آرا
    predicted_class = class_votes.most_common(1)[0][0]
    
    return predicted_class, k_nearest

def classify_knn(train_docs, train_labels, test_docs, k):
    """
    طبقه‌بندی اسناد با استفاده از kNN
    """
    # ایجاد واژگان مشترک
    vocab = set()
    for doc in train_docs + test_docs:
        vocab.update(doc)
    vocab_list = list(vocab)
    
    predictions = []
    for test_doc in test_docs:
        predicted_class, _ = find_k_nearest_neighbors(train_docs, train_labels, test_doc, k, vocab_list)
        predictions.append(predicted_class)
    
    return predictions

def evaluate_knn(train_docs, train_labels, test_docs, test_labels, k):
    """
    ارزیابی عملکرد kNN
    """
    predictions = classify_knn(train_docs, train_labels, test_docs, k)
    correct = sum(1 for predicted, true_label in zip(predictions, test_labels) if predicted == true_label)
    
    accuracy = correct / len(test_docs)
    return accuracy

def plot_knn_decision_boundary(train_docs, train_labels, k=3):
    """
    نمایش مرزهای تصمیم kNN (نسخه ساده از Voronoi tessellation)
    """
    plt.figure(figsize=(10, 8))
    
    # ایجاد واژگان مشترک
    vocab = set()
    for doc in train_docs:
        vocab.update(doc)
    vocab_list = list(vocab)
    
    # تبدیل اسناد به بردارهای tf
    doc_vectors = []
    for doc in train_docs:
        doc_vector = np.zeros(len(vocab_list))
        for j, term in enumerate(vocab_list):
            doc_vector[j] = doc.count(term)
        doc_vectors.append(doc_vector)
    
    # کاهش ابعاد با PCA برای نمایش در فضای دو بعدی
    pca = PCA(n_components=2)
    doc_vectors_2d = pca.fit_transform(doc_vectors)
    
    # نمایش اسناد آموزشی
    colors = {'China': 'red', 'UK': 'blue', 'Kenya': 'green'}
    
    for i, (vector_2d, label) in enumerate(zip(doc_vectors_2d, train_labels)):
        plt.scatter(vector_2d[0], vector_2d[1], 
                   c=colors[label], alpha=0.7, label=label)
    
    unique_labels = list(set(train_labels))
    label_to_numeric = {label: i for i, label in enumerate(unique_labels)}
    
    # نمایش مرزهای تصمیم (فقط بین دو کلاس اول)
    # ایجاد یک شبکه برای نمایش مرزهای تصمیم
    x_min, x_max = doc_vectors_2d[:, 0].min() - 1, doc_vectors_2d[:, 0].max() + 1
    y_min, y_max = doc_vectors_2d[:, 1].min() - 1, doc_vectors_2d[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                         np.arange(y_min, y_max, 0.1))
    
    # طبقه‌بندی هر نقطه در شبکه
    Z = []
    for point in np.c_[xx.ravel(), yy.ravel()]:
        # بازگشت به فضای اصلی با استفاده از معکوس PCA
        point_original = pca.inverse_transform(point.reshape(1, -1))
        
        # پیدا کردن نزدیک‌ترین همسایه
        min_distance = float('inf')
        nearest_label = None
        for i, (vector, label) in enumerate(zip(doc_vectors, train_labels)):
            distance = euclidean_distance(point_original, vector)
            if distance < min_distance:
                min_distance = distance
                nearest_label = label
        
        Z.append(label_to_numeric[nearest_label])
    
    Z = np.array(Z).reshape(xx.shape)
    
    cmap = plt.cm.colors.ListedColormap([colors[label] for label in unique_labels])
    
    # نمایش مرزهای تصمیم
    plt.contourf(xx, yy, Z, alpha=0.2, cmap=cmap)
    
    plt.xlabel('first principal component')
    plt.ylabel('second principal component')
    plt.title('Decision boundary of kNN (k=3)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.axis('equal')
    plt.show()

# مثال کاربردی: طبقه‌بندی اخبار مربوط به کشورها
documents = [
    ["Chinese", "Beijing", "Mao"],
    ["Chinese", "Shanghai", "Mao"],
    ["Chinese", "Taiwan", "Mao"],
    ["Chinese", "Tokyo", "Mao"],
    ["British", "London", "Queen"],
    ["British", "London", "Queen"],
    ["Kenyan", "Nairobi", "tourism"],
    ["Kenyan", "Nairobi", "safari"]
]
labels = ["China", "China", "China", "China", "UK", "UK", "Kenya", "Kenya"]

# طبقه‌بندی با k=1
predicted_1nn = classify_knn(documents, labels, documents, 1)
print(f"پیش‌بینی kNN (k=1): {predicted_1nn}")

# طبقه‌بندی با k=3
predicted_3nn = classify_knn(documents, labels, documents, 3)
print(f"پیش‌بینی kNN (k=3): {predicted_3nn}")

# ارزیابی عملکرد
accuracy_1nn = evaluate_knn(documents, labels, documents, labels, 1)
accuracy_3nn = evaluate_knn(documents, labels, documents, labels, 3)
print(f"\nدقت kNN (k=1): {accuracy_1nn:.2f}")
print(f"دقت kNN (k=3): {accuracy_3nn:.2f}")

# نمایش مرزهای تصمیم
plot_knn_decision_boundary(documents, labels, k=3)

# نمایش پیچیدگی‌های kNN
print("\nپیچیدگی‌های kNN:")
print("1. حساسیت بالا در مجموعه‌های بزرگ: برای مجموعه‌های بزرگ، محاسبه فاصله برای هر سند آزمون ممکن است زمان‌بر باشد.")
print("2. حساسیت ذخیره: kNN به حافظه نیاز دارد تا تمام اسناد آموزشی را ذخیره کند.")
print("3. عدم تعادل وزن‌دهی ویژگی‌ها: ویژگی‌هایی با مقادیر بزرگ ممکن است بر ویژگی‌های با مقادیر کوچک تأثیر بیشتری داشته باشند.")

# نمایش پیچیدگی زمانی
print("\nپیچیدگی زمانی kNN:")
print("پیچیدگی زمانی kNN به صورت زیر محاسبه می‌شود:")
print("Θ(|D|L<sub>ave</sub> + |C||V|)")
print("که در آن |D| تعداد اسناد، L<sub>ave</sub> میانگین تعداد توکن در هر سند، |C| تعداد کلاس‌ها و |V| تعداد ویژگی‌ها است.")
print("این نشان می‌دهد که پیچیدگی زمانی kNN خطی در اندازه مجموعه آموزشی است.")
print("در مقایسه، پیچیدگی زمانی kNN معمولاً از Rocchio بدتر است زیرا نیاز به محاسبه و ذخیره تمام اسناد آموزشی دارد.")

# مثال واقعی: سیستم اسپاتیفای
print("\nمثال واقعی: سیستم اسپاتیفای")
print("در سیستم اسپاتیفای، kNN برای پیدا کردن موسیقی مشابه بر اساس ویژگی‌های صوتی بسیار مفید است.")
print("این سیستم می‌تواند دو آهنگ با ریتم‌های متفاوت را به عنوان \"مشابه\" شناسایی کند، حتی اگر متون آن‌ها کاملاً متفاوت باشند.")
print("این کار به ویژگی‌های صوتی که مستقل از زبان و فرهنگ موسیقی هستند تکیه می‌دهد.")

# نمایش پیچیدگی زمانی
print("\nپیچیدگی زمانی: چگونه کارایی را بهبود آوریم؟")
print("1. بهینه‌سازی ساختار داده: استفاده از ساختارهای داده کارآمد مانند KD-tree برای جستجوی سریع‌تر.")
print("2. استفاده از تقریبندی تقریبی: محدود کردن k به تعداد معقولی برای کاهش محاسبات.")
print("3. استفاده از فاصله‌های جایگزین مانند فاصله کسینوسی که به وزن‌دهی ویژگی‌ها توجه بیشتری به موارد مهم می‌دهد.")
print("4. استفاده از روش‌های احتمالی مانند Local Sensitive Hashing (LSH) که به ویژگی‌های با وزن کمتر وزن بیشتری می‌دهد.")
```

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">💾 یادگیری مبتنی بر حافظه: نقاط قوت و ضعف</p>
<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی kNN، برخلاف طبقه‌بندی Rocchio (مراکز) یا بیز ساده (اولویت‌ها و احتمالات شرطی)، هیچ تخمین پارامتری انجام نمی‌دهیم. kNN صرفاً تمام مثال‌های مجموعه آموزشی را به خاطر می‌سپارد و سپس سند تست را با آن‌ها مقایسه می‌کند. به همین دلیل، kNN همچنین یادگیری مبتنی بر حافظه یا یادگیری مبتنی بر نمونه نامیده می‌شود.</p>

<b>مثال واقعی:</b> در سیستم جستجوی گوگل، برای پیدا کردن صفحات وب مشابه با یک کوئری، از نمایه معکوس استفاده می‌شود تا جستجو را به اسنادی محدود کند که حداقل یک کلمه مشترک با کوئری دارند. این رویکرد می‌تواند زمان جستجو را به طور قابل توجهی کاهش دهد، به خصوص برای کوئری‌های کوتاه که با استفاده از لیست متوقف کلمات بزرگ، همپوشانی کمی با اسناد دارند.</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">📊 بهینگی kNN: چقدر دقیق است؟</p>
<p style="font-size:16pt; text-align:right;">
همانطور که در فصل بعد خواهیم دید، اثربخشی kNN نزدیک به دقیق‌ترین روش‌های یادگیری در طبقه‌بندی متن است (جدول 15.2، صفحه 334). یک معیار برای کیفیت یک روش یادگیری، نرخ خطای بیز آن است، یعنی میانگین نرخ خطای طبقه‌بندهای یادگرفته شده توسط آن برای یک مسئله خاص.</p>

<p style="font-size:16pt; text-align:right;">
kNN برای مسائلی با نرخ خطای بیز غیرصفر بهینه نیست - یعنی مسائلی که حتی بهترین طبقه‌بند ممکن نرخ خطای طبقه‌بندی غیرصفر دارد. خطای 1NN به صورت مجانبی (با افزایش مجموعه آموزشی) توسط دو برابر نرخ خطای بیز محدود می‌شود. یعنی اگر طبقه‌بند بهینه نرخ خطای x داشته باشد، 1NN نرخ خطای مجانبی کمتر از 2x خواهد داشت.</p>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم تشخیص اسپم جیمیل، حتی بهترین طبقه‌بند ممکن است به دلیل تکنیک‌های پیچیده اسپمرها، نرخ خطای 1% داشته باشد. طبقه‌بند 1NN در این حالت ممکن است نرخ خطای تا 2% داشته باشد، که هنوز برای کاربران قابل قبول است. اما برای مسائل با نرخ خطای بیز صفر (مانند تشخیص اعداد دستنویس در شرایط ایده‌آل)، خطای 1NN با افزایش اندازه مجموعه آموزشی به صفر نزدیک می‌شود.</p>
</div>

</div>

<p style="font-size:18pt; text-align:center;">
<span id="section14.4" style="font-size:24pt; font-weight:bold; color:#1B263B;">14.4 طبقه‌بندهای خطی در مقابل غیرخطی: مرزهای ساده در برابر پیچیدگی‌های دنیای واقعی</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎬 چالش طبقه‌بندی فیلم در نتفلیکس: یک خط ساده کافی است؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم طبقه‌بندی محتوای نتفلیکس (Netflix) کار می‌کنید. وظیفه شما این است که فیلم‌ها را به دو دسته "اکشن" و "کمدی" تقسیم کنید. شاید بتوانید با ویژگی‌هایی مانند "تعداد انفجارها در دقیقه" یا "تعداد جوک‌ها در دقیقه" یک خط مرزی ساده بین این دو ژانر رسم کنید. این یک <strong>طبقه‌بند خطی</strong> است.</p>
<p style="font-size:16pt; text-align:right;">
حالا تصور کنید وظیفه شما پیشنهاد فیلم به کاربران است. مرز بین "فیلم‌هایی که کاربر دوست دارد" و "فیلم‌هایی که دوست ندارد" بسیار پیچیده و شخصی است. آیا یک خط ساده می‌تواند این تمایز را ایجاد کند؟ به احتمال زیاد خیر. این یک <strong>مسئله غیرخطی</strong> است که به الگوریتم‌های پیچیده‌تری مانند kNN نیاز دارد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این بخش، با تفاوت طبقه‌بندهای خطی و غیرخطی آشنا می‌شویم. <strong>طبقه‌بند خطی</strong> طبقه‌بندی را با مقایسه یک ترکیب خطی از ویژگی‌ها با یک آستانه (threshold) انجام می‌دهد. در فضای دو بعدی، این طبقه‌بند یک خط است و در فضاهای با ابعاد بالاتر، یک هایپرپلن.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📐 شکل 14.8: بی‌نهایت خط برای جدا کردن دو کلاس</p>
<p style="font-size:16pt; text-align:right;">
اگر دو کلاس کاملاً از هم جدا باشند (linearly separable)، بی‌نهایت خط (یا هایپرپلن) می‌تواند آن‌ها را از هم جدا کند. چالش اصلی، انتخاب بهترین خط است که روی داده‌های جدید بهترین عملکرد را داشته باشد.</p>

</div>

</div>

<p style="font-size:16pt; text-align:right;">
الگوریتم یک طبقه‌بند خطی بسیار ساده است: یک امتیاز (score) بر اساس مجموع وزن‌دار ویژگی‌ها محاسبه شده و با آستانه مقایسه می‌شود.</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚙️ شکل 14.9: الگوریتم طبقه‌بندی خطی</p>
<p style="font-size:16pt; text-align:right;">
این الگوریتم نشان می‌دهد که چگونه یک بردار ویژگی (<strong>~x</strong>) با استفاده از بردار وزن‌ها (<strong>~w</strong>) و آستانه (<strong>b</strong>) طبقه‌بندی می‌شود.</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig14.9.bmp" alt="Linear classification algorithm pseudocode" style="width:600px; border:1px solid #ccc; border-radius:12px;">
</div>
    
<p style="font-size:16pt; text-align:right;">
اگر مجموع وزن‌دار ویژگی‌ها (<strong>score</strong>) از آستانه <strong>b</strong> بیشتر باشد، سند به کلاس 1 و در غیر این صورت به کلاس 0 تعلق می‌گیرد. چالش واقعی در آموزش این طبقه‌بند است، یعنی پیدا کردن مقادیر بهینه برای <strong>~w</strong> و <strong>b</strong>.</p>
</div>

<p style="font-size:16pt; text-align:right;">
شاید تعجب‌آور باشد، اما الگوریتم‌هایی که تاکنون بررسی کردیم، یعنی <strong>Rocchio</strong> و <strong>Naive Bayes</strong>، هر دو نمونه‌هایی از طبقه‌بندهای خطی هستند.</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>Rocchio:</strong> مرز تصمیم در Rocchio، مجموعه‌ای از نقاط است که فاصله آن‌ها از دو مرکز (centroid) کلاس‌ها برابر است. می‌توان ثابت کرد که این مرز یک هایپرپلن است.</li>
<li><strong>Naive Bayes:</strong> قانون تصمیم Naive Bayes بر اساس بیشترین مقدار P(c|d) است. اگر از لگاریتم نسبت احتمالات (log odds) استفاده کنیم، معادله به صورت یک ترکیب خطی درمی‌آید:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
log P(c|d) / P(c¯|d) = log P(c) / P(c¯) + Σ log P(tk|c) / P(tk|c¯)
</div>
این معادله دقیقاً همان ساختار یک طبقه‌بند خطی (<strong>~wT~x > b</strong>) را دارد.</li>
</ul>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📈 جدول 14.4: یک طبقه‌بند خطی برای دسته‌بندی اخبار اقتصادی</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🎯 چالش‌های طبقه‌بندهای خطی: نویز و جداپذیری خطی</p>
<p style="font-size:16pt; text-align:right;">
طبقه‌بندهای خطی با چالش‌های خاصی روبرو هستند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>اسناد نویزی (Noise Documents):</strong> اسنادی وجود دارند که در منطقه‌ای از فضا قرار دارند که اکثر اسناد متعلق به کلاس دیگری هستند. این اسناد می‌توانند الگوریتم آموزش را گمراه کنند. در شکل 14.10، سه سند با علامت پیکان، اسناد نویزی هستند.</li>
<li><strong>جداپذیری خطی (Linear Separability):</strong> اگر یک هایپرپلن بتواند دو کلاس را به طور کامل از هم جدا کند، می‌گوییم کلاس‌ها "خطی جداپذیر" هستند. در این صورت، بی‌نهایت هایپرپلن جداکننده وجود دارد (مانند شکل 14.8). چالش این است که بهترین هایپرپلن را انتخاب کنیم که روی داده‌های جدید بهترین عملکرد را داشته باشد.</li>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🌀 طبقه‌بندهای غیرخطی: زمانی که یک خط کافی نیست</p>
<p style="font-size:16pt; text-align:right;">
گاهی اوقات، مرز بین دو کلاس به سادگی یک خط یا هایپرپلن نیست. شکل 14.11 یک <strong>مسئله غیرخطی</strong> را نشان می‌دهد. یک طبقه‌بند خطی نمی‌تواند "clave" دایره‌ای شکل در گوشه بالا سمت چپ را به درستی طبقه‌بندی کند.</p>

<p style="font-size:16pt; text-align:right;">
در این موارد، <strong>طبقه‌بندهای غیرخطی</strong> مانند kNN عملکرد بهتری دارند. مرز تصمیم kNN (خطوط دوتایی در شکل 14.6) از قطعات خطی محلی تشکیل شده است، اما به طور کلی شکلی پیچیده دارد که معادل یک خط یا هایپرپلن واحد نیست.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم تشخیص چهره فیس‌بوک، مرز بین "چهره شما" و "چهره افراد دیگر" بسیار پیچیده است. یک طبقه‌بند خطی ساده نمی‌تواند این تمایز را با دقت بالا ایجاد کند. در مقابل، kNN می‌تواند با مقایسه یک چهره جدید با نزدیک‌ترین چهره‌های موجود در پایگاه داده، این کار را با دقت بالایی انجام دهد.</p>

<p style="font-size:16pt; text-align:right;">
در نهایت، انتخاب بین طبقه‌بند خطی و غیرخطی به ماهیت مسئله بستگی دارد. اگر مسئله خطی است، یک طبقه‌بند خطی ساده و کارآمد بهترین انتخاب است. اما اگر مرزهای کلاس‌ها پیچیده و غیرخطی هستند، طبقه‌بندهای غیرخطی مانند kNN دقت بالاتری خواهند داشت.</p>

```python
# پیاده‌سازی و مقایسه طبقه‌بندهای خطی و غیرخطی

import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# تولید داده‌های خطی جداشد با نویز
def generate_linear_data(n_samples=200, noise_level=0.1):
    """
    تولید داده‌های خطی جداشد با نویز برای شبیه‌سازی چالش‌های دنیای واقعی
    """
    np.random.seed(42)
    
    # ایجاد دو کلاس خطی جداشد
    X_class1 = np.random.randn(n_samples, 2) + np.array([2, 2])
    X_class2 = np.random.randn(n_samples, 2) + np.array([-2, -2])
    
    # ایجاد برچسب‌ها
    y_class1 = np.ones(n_samples)
    y_class2 = np.zeros(n_samples)
    
    # ترکیب داده‌ها
    X = np.vstack([X_class1, X_class2])
    y = np.hstack([y_class1, y_class2])
    
    # افزودن نویز به درصدی از داده‌ها (اسناد نویزی)
    n_noise = int(noise_level * n_samples)
    noise_indices = np.random.choice(len(X), n_noise, replace=False)
    
    # معکوس کردن برچسب برخی از نقاط نویز
    y[noise_indices] = 1 - y[noise_indices]
    
    return X, y

# تولید داده‌های غیرخطی
def generate_nonlinear_data(n_samples=200):
    """
    تولید داده‌های واقعاً غیرخطی برای نمایش محدودیت‌های طبقه‌بندهای خطی
    """
    np.random.seed(42)
    # ایجاد داده‌های شکل ماه (غیرخطی)
    n = n_samples // 2
    X1 = np.zeros((n, 2))
    X2 = np.zeros((n, 2))
    
    # کلاس اول: نیمه بالایی ماه
    for i in range(n):
        angle = np.pi * i / n
        X1[i, 0] = np.cos(angle)
        X1[i, 1] = np.sin(angle)
    
    # کلاس دوم: نیمه پایینی ماه (بدون شیب رو به پایین)
    for i in range(n):
        angle = np.pi + np.pi * i / n
        X2[i, 0] = np.cos(angle)
        X2[i, 1] = np.sin(angle)  # حذف شیب رو به پایین
    
    # ترکیب داده‌ها
    X = np.vstack([X1, X2])
    y = np.hstack([np.zeros(n), np.ones(n)])
    
    return X, y

# پیاده‌سازی طبقه‌بند خطی Rocchio
class RocchioClassifier:
    """
    پیاده‌سازی ساده طبقه‌بند Rocchio که یک طبقه‌بند خطی است
    """
    def __init__(self):
        self.centroids = {}
    
    def fit(self, X, y):
        """
        محاسبه مرکزoids برای هر کلاس
        """
        # دریافت کلاس‌های یکتا
        classes = np.unique(y)
        
        # محاسبه مرکز برای هر کلاس
        for c in classes:
            # انتخاب تمام نقاط داده برای کلاس c
            X_c = X[y == c]
            # محاسبه میانگین (مرکز)
            self.centroids[c] = np.mean(X_c, axis=0)
    
    def predict(self, X):
        """
        پیش‌بینی کلاس برای هر نقطه داده بر اساس نزدیک‌ترین مرکز
        """
        y_pred = []
        
        # برای هر نقطه داده
        for x in X:
            # مقداردهی اولیه با فاصله بزرگ
            min_dist = float('inf')
            best_class = None
            
            # برای هر مرکز کلاس
            for c, centroid in self.centroids.items():
                # محاسبه فاصله تا مرکز
                dist = np.linalg.norm(x - centroid)
                
                # اگر این مرکز نزدیک‌تر است، به‌روزرسانی کن
                if dist < min_dist:
                    min_dist = dist
                    best_class = c
            
            # افزودن بهترین کلاس به پیش‌بینی‌ها
            y_pred.append(best_class)
        
        return np.array(y_pred)

# پیاده‌سازی طبقه‌بند خطی Naive Bayes
class NaiveBayesClassifier:
    """
    پیاده‌سازی ساده طبقه‌بند Naive Bayes که یک طبقه‌بند خطی است
    """
    def __init__(self):
        self.class_priors = {}
        self.class_feature_means = {}
        self.class_feature_vars = {}
    
    def fit(self, X, y):
        """
        محاسبه احتمالات اولیه کلاس و توزیع ویژگی‌ها
        """
        # دریافت کلاس‌های یکتا
        classes = np.unique(y)
        
        # محاسبه احتمال اولیه برای هر کلاس
        for c in classes:
            # احتمال اولیه: P(c) = تعداد(c) / مجموع_تعداد
            self.class_priors[c] = np.mean(y == c)
            
            # محاسبه میانگین و واریانس برای هر ویژگی
            X_c = X[y == c]
            self.class_feature_means[c] = np.mean(X_c, axis=0)
            self.class_feature_vars[c] = np.var(X_c, axis=0) + 1e-9  # افزودن مقدار کوچک برای جلوگیری از تقسیم بر صفر
    
    def predict(self, X):
        """
        پیش‌بینی کلاس برای هر نقطه داده با استفاده از Naive Bayes
        """
        y_pred = []
        
        # برای هر نقطه داده
        for x in X:
            # مقداردهی اولیه با مقدار بزرگ منفی (فضای لگاریتمی)
            max_log_prob = -float('inf')
            best_class = None
            
            # برای هر کلاس
            for c in self.class_priors:
                # شروع با لگاریتم احتمال اولیه
                log_prob = np.log(self.class_priors[c])
                
                # افزودن لگاریتم احتمال برای هر ویژگی
                for i in range(len(x)):
                    # محاسبه احتمال مقدار ویژگی با توجه به کلاس
                    # با استفاده از توزیع گوسی
                    mean = self.class_feature_means[c][i]
                    var = self.class_feature_vars[c][i]
                    
                    # لگاریتم احتمال مقدار ویژگی
                    log_prob += -0.5 * np.log(2 * np.pi * var) - 0.5 * ((x[i] - mean) ** 2) / var
                
                # اگر این کلاس احتمال بالاتری دارد، به‌روزرسانی کن
                if log_prob > max_log_prob:
                    max_log_prob = log_prob
                    best_class = c
            
            # افزودن بهترین کلاس به پیش‌بینی‌ها
            y_pred.append(best_class)
        
        return np.array(y_pred)

# پیاده‌سازی طبقه‌بند غیرخطی kNN
class KNNClassifier:
    """
    پیاده‌سازی ساده طبقه‌بند k-نزدیک‌ترین همسایه که یک طبقه‌بند غیرخطی است
    """
    def __init__(self, k=5):
        self.k = k
        self.X_train = None
        self.y_train = None
    
    def fit(self, X, y):
        """
        ذخیره داده‌های آموزشی (kNN یک یادگیرنده تنبل است)
        """
        self.X_train = X
        self.y_train = y
    
    def predict(self, X):
        """
        پیش‌بینی کلاس برای هر نقطه داده با استفاده از kNN
        """
        y_pred = []
        
        # برای هر نقطه داده
        for x in X:
            # محاسبه فاصله تا تمام نقاط آموزشی
            distances = np.array([np.linalg.norm(x - x_train) for x_train in self.X_train])
            
            # دریافت شاخص‌های k نزدیک‌ترین همسایه‌ها
            k_nearest_indices = np.argsort(distances)[:self.k]
            
            # دریافت برچسب‌های k نزدیک‌ترین همسایه‌ها
            k_nearest_labels = self.y_train[k_nearest_indices]
            
            # پیش‌بینی با رأی اکثریت
            counts = np.bincount(k_nearest_labels.astype(int))
            y_pred.append(np.argmax(counts))
        
        return np.array(y_pred)

# تابع برای رسم مرز تصمیم
def plot_decision_boundary(classifier, X, y, title, ax=None):
    """
    رسم مرز تصمیم یک طبقه‌بند
    """
    if ax is None:
        ax = plt.gca()
    
    # ایجاد یک شبکه مش
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                         np.arange(y_min, y_max, 0.1))
    
    # پیش‌بینی برای هر نقطه در شبکه مش
    Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    
    # نمایش مرز تصمیم و مناطق
    ax.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.Paired)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Paired, edgecolors='k')
    
    ax.set_title(title)
    ax.set_xlabel('Feature 1')
    ax.set_ylabel('Feature 2')

# تابع برای مقایسه طبقه‌بندها روی داده‌های خطی
def compare_linear_classifiers():
    """
    مقایسه طبقه‌بندهای خطی و غیرخطی روی داده‌های خطی جداشد با نویز
    """
    # تولید داده‌های خطی با نویز
    X, y = generate_linear_data(200, noise_level=0.1)
    
    # تقسیم داده‌ها
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    # ایجاد طبقه‌بندها
    rocchio = RocchioClassifier()
    naive_bayes = NaiveBayesClassifier()
    knn = KNNClassifier(k=5)
    
    # آموزش طبقه‌بندها
    rocchio.fit(X_train, y_train)
    naive_bayes.fit(X_train, y_train)
    knn.fit(X_train, y_train)
    
    # ارزیابی طبقه‌بندها
    rocchio_acc = accuracy_score(y_test, rocchio.predict(X_test))
    naive_bayes_acc = accuracy_score(y_test, naive_bayes.predict(X_test))
    knn_acc = accuracy_score(y_test, knn.predict(X_test))
    
    print(f"Rocchio Accuracy: {rocchio_acc:.3f}")
    print(f"Naive Bayes Accuracy: {naive_bayes_acc:.3f}")
    print(f"kNN Accuracy: {knn_acc:.3f}")
    
    # نمایش مرزهای تصمیم
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    plot_decision_boundary(rocchio, X, y, 'Rocchio (Linear)', axes[0])
    plot_decision_boundary(naive_bayes, X, y, 'Naive Bayes (Linear)', axes[1])
    plot_decision_boundary(knn, X, y, 'kNN (Nonlinear)', axes[2])
    
    plt.tight_layout()
    plt.show()

# تابع برای مقایسه طبقه‌بندها روی داده‌های غیرخطی
def compare_nonlinear_classifiers():
    """
    مقایسه طبقه‌بندهای خطی و غیرخطی روی داده‌های غیرخطی جداشد
    """
    # تولید داده‌های غیرخطی
    X, y = generate_nonlinear_data(200)
    
    # تقسیم داده‌ها
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    # ایجاد طبقه‌بندها
    rocchio = RocchioClassifier()
    naive_bayes = NaiveBayesClassifier()
    knn = KNNClassifier(k=5)
    
    # آموزش طبقه‌بندها
    rocchio.fit(X_train, y_train)
    naive_bayes.fit(X_train, y_train)
    knn.fit(X_train, y_train)
    
    # ارزیابی طبقه‌بندها
    rocchio_acc = accuracy_score(y_test, rocchio.predict(X_test))
    naive_bayes_acc = accuracy_score(y_test, naive_bayes.predict(X_test))
    knn_acc = accuracy_score(y_test, knn.predict(X_test))
    
    print(f"Rocchio Accuracy: {rocchio_acc:.3f}")
    print(f"Naive Bayes Accuracy: {naive_bayes_acc:.3f}")
    print(f"kNN Accuracy: {knn_acc:.3f}")
    
    # نمایش مرزهای تصمیم
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    plot_decision_boundary(rocchio, X, y, 'Rocchio (Linear)', axes[0])
    plot_decision_boundary(naive_bayes, X, y, 'Naive Bayes (Linear)', axes[1])
    plot_decision_boundary(knn, X, y, 'kNN (Nonlinear)', axes[2])
    
    plt.tight_layout()
    plt.show()

# اجرای مقایسه روی داده‌های خطی
print("Comparison of Classifiers on Linear Data with Noise:")
print("="*50)
compare_linear_classifiers()

# اجرای مقایسه روی داده‌های غیرخطی
print("\nComparison of Classifiers on Non-linear Data:")
print("="*50)
compare_nonlinear_classifiers()
```

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌐 چالش طبقه‌بندی خبر در گوگل نیوز: یک خبر می‌تواند چند برچسب داشته باشد؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم طبقه‌بندی محتوای گوگل نیوز کار می‌کنید و باید سیستمی طراحی کنید که اخبار را به دسته‌های مختلفی مانند "سیاست"، "اقتصاد"، "فناوری" و "ورزشی" تقسیم کند. آیا یک خبر می‌تواند همزمان به چند دسته تعلق داشته باشد؟</p>
<p style="font-size:16pt; text-align:right;">
مثلاً، خبری درباره "قرارداد چند میلیارد دلاری شرکت اپل با دولت چین برای تولید آیفون" می‌تواند همزمان به دسته‌های "فناوری"، "اقتصاد" و "سیاست" تعلق داشته باشد. این یک <strong>مسئله طبقه‌بندی هر-از-چند</strong> است.</p>
<p style="font-size:16pt; text-align:right;">
حالا تصور کنید وظیفه شما تشخیص زبان یک سند است. یک سند می‌تواند همزمان به چند زبان نوشته شده باشد؟ به ندرت. این یک <strong>مسئله طبقه‌بندی یک-از-چند</strong> است.</p>

<p style="font-size:16pt; text-align:right;">
در مسائل طبقه‌بندی با بیش از دو کلاس، دو نوع اصلی وظیفه وجود دارد که بسته به ماهیت کلاس‌ها انتخاب می‌شود:</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🏷️ طبقه‌بندی هر-از-چند (Any-of Classification)</p>
<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی هر-از-چند (که به آن چندبرچسبی یا چندمقداری نیز گفته می‌شود)، کلاس‌ها انحصاری متقابل نیستند. یک سند می‌تواند همزمان به چندین کلاس تعلق داشته باشد، یا به هیچ‌کدام از کلاس‌ها تعلق نداشته باشد.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم توصیه فیلم آمازون پرایم، یک فیلم می‌تواند همزمان برچسب‌های "اکشن"، "علمی-تخیلی" و "ماجراجویی" را داشته باشد. تصمیم برای یک برچسب، تصمیم برای برچسب‌های دیگر را محدود نمی‌کند.</p>
<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی هر-از-چند، برای هر کلاس یک طبقه‌بند جداگانه ساخته می‌شود. برای هر کلاس، مجموعه آموزشی شامل اسناد متعلق به آن کلاس (برچسب‌های مثبت) و مکمل آن‌ها (برچسب‌های منفی) است. هنگام طبقه‌بندی یک سند جدید، هر طبقه‌بند به طور جداگانه اعمال می‌شود و تصمیم یک طبقه‌بند تأثیری بر تصمیمات طبقه‌بندهای دیگر ندارد.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 طبقه‌بندی یک-از-چند (One-of Classification)</p>
<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی یک-از-چند (که به آن چندوجهی، چندکلاسی یا تک‌برچسبی نیز گفته می‌شود)، کلاس‌ها انحصاری متقابل هستند. هر سند باید دقیقاً به یکی از کلاس‌ها تعلق داشته باشد.</p>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم تشخیص زبان گوگل ترنسلیت، یک سند باید دقیقاً به یک زبان تعلق داشته باشد: انگلیسی، فرانسوی، اسپانیایی، و غیره. در این حالت، فرض یک-از-چند یک تقریب خوب است، زیرا اکثر متون فقط به یک زبان نوشته می‌شوند.</p>
<p style="font-size:16pt; text-align:right;">
مسائل واقعی یک-از-چند در طبقه‌بندی متن کمتر از مسائل هر-از-چند رایج هستند. با کلاس‌هایی مانند "بریتانیا"، "چین"، "طیور" یا "قهوه"، یک سند می‌تواند همزمان به چند موضوع مرتبط باشد - مانند وقتی که نخست‌وزیر بریتانیا برای صحبت درباره تجارت قهوه و طیور به چین سفر می‌کند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📐 شکل 14.12: چالش تقسیم فضا با چند هایپرپلن</p>
<p style="font-size:16pt; text-align:right;">
چالش اصلی در طبقه‌بندی یک-از-چند با استفاده از طبقه‌بندهای خطی این است که J هایپرپلن فضای برداری را به J ناحیه مجزا تقسیم نمی‌کنند. شکل 14.12 این مشکل را نشان می‌دهد: هایپرپلن‌ها یکدیگر را قطع می‌کنند و یک سند ممکن است همزمان به چندین ناحیه تعلق داشته باشد.</p>

<p style="font-size:16pt; text-align:right;">
برای حل این مشکل، از روش ترکیبی استفاده می‌شود: ابتدا کلاس‌ها بر اساس امتیاز رتبه‌بندی می‌شوند و سپس کلاس برتر انتخاب می‌شود. از نظر هندسی، رتبه‌بندی می‌تواند بر اساس فاصله از J جداکننده خطی باشد. اسنادی که نزدیک به جداکننده یک کلاس هستند، احتمالاً اشتباه طبقه‌بندی می‌شوند، بنابراین هرچه فاصله از جداکننده بیشتر باشد، احتمال صحت تصمیم‌گیری مثبت بیشتر است.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">📈 جدول 14.5: ماتریس سردرگمی برای تحلیل عملکرد طبقه‌بند</p>
<p style="font-size:16pt; text-align:right;">
ماتریس سردرگمی ابزار مهمی برای تحلیل عملکرد طبقه‌بندهای چندکلاسی است. این ماتریس برای هر جفت کلاس (c<sub>1</sub>, c<sub>2</sub>) نشان می‌دهد که چند سند از کلاس c<sub>1</sub> به اشتباه به کلاس c<sub>2</sub> طبقه‌بندی شده‌اند.</p>
<p style="font-size:16pt; text-align:right;">
جدول 14.5 یک ماتریس سردرگمی را برای طبقه‌بندی شش کلاس در مجموعه داده Reuters-21578 نشان می‌دهد. طبقه‌بند موفق می‌شود سه کلاس مالی (money-fx, trade, interest) را از سه کلاس کشاورزی (wheat, corn, grain) تشخیص دهد، اما در داخل هر یک از این دو گروه اشتباهات زیادی می‌کند.</p>

<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی ایمیل گوگل، ماتریس سردرگمی می‌تواند نشان دهد که چرا برخی ایمیل‌های "تبلیغاتی" به اشتباه به دسته "اسپم" ارسال شده‌اند. با تحلیل این ماتریس، می‌توان ویژگی‌های جدیدی معرفی کرد که این دو دسته را بهتر از هم تمایز دهد.</p>
<p style="font-size:16pt; text-align:right;">
ماتریس سردرگمی می‌تواند به شناسایی فرصت‌هایی برای بهبود دقت سیستم کمک کند. برای مثال، برای رسیدگی به دومین خطای بزرگ در جدول 14.5 (14 در سطر grain)، می‌توان تلاش کرد ویژگی‌هایی را معرفی کند که اسناد wheat را از اسناد grain متمایز کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، انتخاب بین طبقه‌بندی هر-از-چند و یک-از-چند به ماهیت مسئله بستگی دارد. اگر کلاس‌ها انحصاری متقابل هستند، طبقه‌بندی یک-از-چند می‌تواند با اعمال یک محدودیت، دقت را افزایش دهد. اما اگر کلاس‌ها می‌توانند همپوشانی داشته باشند، طبقه‌بندی هر-از-چند انتخاب مناسب‌تری است.</p>

</div>

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# مرحله ۱: تولید داده‌های مصنوعی برای ۳ کلاس
# هر کلاس در یک منطقه خاص از فضا متمرکز شده است تا مسئله طبقه‌بندی قابل مشاهده باشد.
def generate_multi_class_data():
    """
    داده‌هایی برای سه کلاس تولید می‌کند.
    کلاس 0: بالا-چپ (مثلاً "ورزشی")
    کلاس 1: بالا-راست (مثلاً "سیاسی")
    کلاس 2: پایین (مثلاً "فناوری")
    """
    np.random.seed(42)
    # کلاس 0 (ورزشی)
    X0 = np.random.randn(50, 2) + np.array([-2, 2])
    y0 = np.zeros(50)
    
    # کلاس 1 (سیاسی)
    X1 = np.random.randn(50, 2) + np.array([2, 2])
    y1 = np.ones(50)
    
    # کلاس 2 (فناوری)
    X2 = np.random.randn(50, 2) + np.array([0, -2])
    y2 = np.full(50, 2)

    X = np.vstack([X0, X1, X2])
    y = np.hstack([y0, y1, y2])
    return X, y

# مرحله ۲: پیاده‌سازی یک طبقه‌بند خطی ساده (از صفر)
# این طبقه‌بند فقط دو کلاس را از هم جدا می‌کند (یکی در مقابل بقیه)
class SimpleLinearClassifier:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None

    def fit(self, X, y_binary):
        """
        مدل را با داده‌های آموزشی آموزش می‌دهد.
        y_binary باید برچسب‌های 0 و 1 باشد.
        """
        n_samples, n_features = X.shape
        # مقداردهی اولیه وزن‌ها و بایاس
        self.weights = np.zeros(n_features)
        self.bias = 0
        
        # الگوریتم بهینه‌سازی ساده (گرادیان کاهشی)
        for _ in range(self.n_iterations):
            # مدل خطی: z = w.x + b
            linear_model = np.dot(X, self.weights) + self.bias
            # تابع فعال‌سازی ساده (پله‌ای)
            y_predicted = self._step_function(linear_model)
            
            # به‌روزرسانی وزن‌ها و بایاس
            dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y_binary))
            db = (1 / n_samples) * np.sum(y_predicted - y_binary)
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

    def predict(self, X):
        """
        برچسب کلاس (0 یا 1) را برای داده‌های جدید پیش‌بینی می‌کند.
        """
        linear_model = np.dot(X, self.weights) + self.bias
        y_predicted = self._step_function(linear_model)
        return y_predicted
    
    def decision_function(self, X):
        """
        امتیاز (فاصله از مرز تصمیم) را برای هر نقطه محاسبه می‌کند.
        این امتیاز برای مقایسه بین طبقه‌بندها در استراتژی یک-در-برابر-بقیه حیاتی است.
        """
        return np.dot(X, self.weights) + self.bias

    def _step_function(self, x):
        """
        تابع پله‌ای ساده: اگر ورودی بزرگتر از 0 بود، 1 برمی‌گرداند، در غیر این صورت 0.
        """
        return np.where(x >= 0, 1, 0)

# مرحله ۳: ساخت طبقه‌بند چندکلاسی با استراتژی "یک-در-برابر-بقیه"
class OneVsRestClassifier:
    def __init__(self):
        self.classifiers = {}
        self.classes = None

    def fit(self, X, y):
        """
        برای هر کلاس، یک طبقه‌بند دوتایی آموزش می‌دهد.
        این پیاده‌سازی استراتژی "یک-از-چند" است که در مارک‌دون توضیح داده شد.
        """
        self.classes = np.unique(y)
        for cls in self.classes:
            print(f"Training classifier for class {int(cls)} vs rest...")
            # تبدیل برچسب‌ها به حالت دوتایی: 1 برای کلاس فعلی، 0 برای بقیه
            y_binary = np.where(y == cls, 1, 0)
            classifier = SimpleLinearClassifier()
            classifier.fit(X, y_binary)
            self.classifiers[cls] = classifier

    def predict(self, X):
        """
        برای هر نقطه داده، امتیاز آن را از تمام طبقه‌بندها دریافت کرده
        و کلاسی را انتخاب می‌کند که بالاترین امتیاز را داشته باشد.
        این روش رتبه‌بندی و انتخاب برتر، راه‌حل مشکل تقسیم فضا توسط چند هایپرپلن است.
        """
        predictions = []
        for x in X:
            scores = {}
            for cls, classifier in self.classifiers.items():
                scores[cls] = classifier.decision_function(x)
            
            # کلاس با بالاترین امتیاز انتخاب می‌شود
            predicted_class = max(scores, key=scores.get)
            predictions.append(predicted_class)
        return np.array(predictions)

# مرحله ۴: اجرای کل فرآیند و نمایش نتایج

# تولید داده‌ها
X, y = generate_multi_class_data()

# آموزش طبقه‌بند چندکلاسی
ovr_classifier = OneVsRestClassifier()
ovr_classifier.fit(X, y)

# پیش‌بینی روی کل مجموعه داده برای تحلیل
y_pred = ovr_classifier.predict(X)

# مرحله ۵: محاسبه و نمایش ماتریس سردرگمی (از صفر)
def calculate_confusion_matrix(y_true, y_pred, classes):
    n_classes = len(classes)
    matrix = np.zeros((n_classes, n_classes), dtype=int)
    # ایجاد نگاشت از کلاس به اندیس ماتریس
    class_to_index = {cls: i for i, cls in enumerate(classes)}
    
    for true_label, pred_label in zip(y_true, y_pred):
        i = class_to_index[true_label]
        j = class_to_index[pred_label]
        matrix[i, j] += 1
    return matrix

conf_matrix = calculate_confusion_matrix(y, y_pred, ovr_classifier.classes)

# چاپ نتایج
print("\n--- Confusion Matrix ---")
print("Rows: Actual Class, Columns: Predicted Class")
print(conf_matrix)

# نمایش ماتریس سردرگمی با نمودار حرارتی
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=[f"Pred {int(c)}" for c in ovr_classifier.classes],
            yticklabels=[f"Actual {int(c)}" for c in ovr_classifier.classes])
plt.title("Confusion Matrix (from scratch)")
plt.ylabel("Actual Class")
plt.xlabel("Predicted Class")
plt.show()

# نمایش مرزهای تصمیم
def plot_decision_boundaries_ovr(X, y, classifier):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                         np.arange(y_min, y_max, 0.1))

    # پیش‌بینی برای تمام نقاط در شبکه
    Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.4, cmap=plt.cm.Paired)
    scatter = plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Paired, edgecolors='k')
    plt.title("Decision Boundaries (One-vs-Rest from scratch)")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    # ایجاد لیجنند (راهنما) برای کلاس‌ها
    legend1 = plt.legend(*scatter.legend_elements(),
                        title="Classes")
    plt.gca().add_artist(legend1)
    plt.show()

print("\n--- Plotting Decision Boundaries ---")
plot_decision_boundaries_ovr(X, y, ovr_classifier)
```

<p style="font-size:18pt; text-align:center;">
<span id="section14.6" style="font-size:24pt; font-weight:bold; color:#1B263B;">14.6 مبادله بایاس-واریانس: چرا هیچ الگوریتمی کامل نیست؟</span> ⚖️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🏥 چالش تشخیص پزشکی: آیا یک مدل ساده بهتر است یا یک مدل پیچیده؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در حال ساخت یک سیستم هوشمند برای تشخیص یک بیماری نادر بر اساس علائم بیماران هستید. دو رویکرد مختلف وجود دارد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مدل ساده (بایاس بالا، واریانس پایین):</strong> یک مدل بسیار ساده که می‌گوید "هیچ‌کس بیمار نیست" چون بیماری نادر است. این مدل واریانس پایینی دارد (پیش‌بینی‌اش هرگز تغییر نمی‌کند) اما بایاس بسیار بالایی دارد (برای بیماران واقعی به طور مداوم اشتباه می‌کند). این مدل به دلیل سادگی بیش از حد، ناتوان است در الگوهای واقعی بیماری را تشخیص دهد.</li>
<li><strong>مدل پیچیده (بایاس پایین، واریانس بالا):</strong> یک مدل بسیار پیچیده که داده‌های آموزشی را به طور کامل به خاطر می‌سپارد، حتی شامل خطاهای اندازه‌گیری. این مدل بایاس پایینی روی داده‌های آموزشی دارد اما واریانس بالایی است. وقتی بیمار جدیدی با علائم کمی متفاوت وارد می‌شود، پیش‌بینی ممکن است به شدت اشتباه باشد. این مدل به دلیل پیچیدگی بیش از حد، به نویز در داده‌های آموزشی حساس است.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
هدف در یادگیری ماشین، یافتن تعادل بین این دو رویکرد است: مدلی که به اندازه کافی پیچیده است تا الگوهای واقعی را تشخیص دهد، اما آنقدر پیچیده نیست که توسط نویز در داده‌های آموزشی گمراه شود.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در یادگیری ماشین، مفهوم <strong>مبادله بایاس-واریانس</strong> نشان می‌دهد که هیچ روش یادگیری به طور جهانی بهینه نیست. این مبادله توضیح می‌دهد که افزایش پیچیدگی یک طبقه‌بند معمولاً باعث کاهش بایاس (افزایش دقت) و افزایش واریانس (کاهش پایداری) می‌شود و برعکس.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 اندازه‌گیری خطا: خطای میانگین مربع (MSE)</p>
<p style="font-size:16pt; text-align:right;">
در طبقه‌بندی، هدف ما بهینه‌سازی خطای طبقه‌بندی بر روی داده‌های جدید است، نه صرفاً به حداقل رساندن خطا روی داده‌های آموزشی. خطای طبقه‌بندی را با معیار <strong>خطای میانگین مربع (MSE)</strong> تعریف می‌کنیم:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">

 $$ \text{MSE}(\gamma) = \mathbb{E}_{\mathbf{d}}[\gamma(\mathbf{d}) - P(c|\mathbf{d})]^2 $$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن γ یک طبقه‌بند است، d یک سند، و P(c|d) احتمال تعلق سند d به کلاس c است. MSE میانگین خطای طبقه‌بند را بر روی تمام اسناد اندازه‌گیری می‌کند و هم خطای ناشی از بایاس و هم خطای ناشی از واریانس را در نظر می‌گیرد.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 بایاس و واریانس در طبقه‌بندهای خطی و غیرخطی</p>
<p style="font-size:16pt; text-align:right;">
مبادله بایاس-واریانس را می‌توان با مقایسه طبقه‌بندهای خطی و غیرخطی بهتر درک کرد:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>طبقه‌بندهای خطی (مانند Rocchio و Naive Bayes):</strong> این طبقه‌بندها <strong>بایاس بالایی</strong> دارند زیرا فقط می‌توانند یک نوع مرز کلاس (هایپرپلن) را مدل کنند. اگر مرز واقعی بین کلاس‌ها پیچیده باشد (مانند شکل 14.11)، این طبقه‌بندها به طور مداوم اشتباه طبقه‌بندی می‌کنند. با این حال، این طبقه‌بندها <strong>واریانس پایینی</strong> دارند زیرا مرز تصمیم آن‌ها پایدار است و با تغییرات جزئی در مجموعه آموزشی، به شدت تغییر نمی‌کند.</li>
<li><strong>طبقه‌بندهای غیرخطی (مانند kNN):</strong> این طبقه‌بندها <strong>بایاس پایینی</strong> دارند زیرا می‌توانند مرزهای تصمیم پیچیده‌ای ایجاد کنند که داده‌های آموزشی را به خوبی مدل می‌کنند. با این حال، این طبقه‌بندها <strong>واریانس بالایی</strong> دارند زیرا به اسناد نویزی حساس هستند. یک سند نویزی در مجموعه آموزشی می‌تواند مرز تصمیم را در ناحیه اطراف خود به شدت تغییر دهد.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم فیلتر اسپم جیمیل، یک طبقه‌بند خطی ممکن است به طور مداوم ایمیل‌های تبلیغاتی خاصی را اشتباه طبقه‌بندی کند (بایاس بالا)، اما پیش‌بینی‌های آن برای کاربران مختلف پایدار است (واریانس پایین). در مقابل، یک طبقه‌بند غیرخطی ممکن است بتواند اسپم‌های جدید و پیچیده‌تر را بهتر تشخیص دهد، اما ممکن است برای کاربران مختلف، رفتار متفاوتی از خود نشان دهد (واریانس بالا).</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🤔 چرا بسیاری از الگوریتم‌های برتر طبقه‌بندی متن، خطی هستند؟</p>
<p style="font-size:16pt; text-align:right;">
شاید تعجب‌آور باشد که بسیاری از بهترین الگوریتم‌های طبقه‌بندی متن، خطی هستند. مبادله بایاس-واریانس به موفقیت آن‌ها کمک می‌کند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>در طبقه‌بندی متن، معمولاً با ابعاد بسیار بالایی سروکار داریم (هزاران یا میلیون‌ها ویژگی، یکی برای هر کلمه).</li>
<li>در فضاهی با ابعاد بالا، احتمال جداپذیری خطی به سرعت افزایش می‌یابد. این یعنی یک هایپرپلن ساده اغلب می‌تواند کلاس‌ها را به خوبی از هم جدا کند.</li>
<li>بنابراین، "بایاس بالا" مدل‌های خطی در عمل در مسائل طبقه‌بندی متن مشکلی جدی ایجاد نمی‌کند.</li>
<li>مزیت بزرگ مدل‌های خطی، "واریانس پایین" آن‌هاست که باعث می‌شود در برابر نویز مقاوم‌تر باشند و کمتر دچار بیش‌برازش (overfitting) شوند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
<b>مثال واقعی:</b> در سیستم طبقه‌بندی اسناد حقوقی شرکت‌هایی مانند Google یا Microsoft، با وجود پیچیدگی ظاهری زبان حقوقی، یک طبقه‌بند خطی ساده اغلب بهترین عملکرد را دارد. زیرا با هزاران ویژگی (کلمات)، یک مرز خطی می‌تواند اسناد مربوط به قراردادهای مختلف را به خوبی از هم جدا کند، در حالی که به نویز و تغییرات جزئی در زبان حساس نیست.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، انتخاب بین یک روش یادگیری ساده (بایاس بالا، واریانس پایین) و یک روش پیچیده (بایاس پایین، واریانس بالا) به ماهیت مسئله، حجم داده‌ها و سطح نویز بستگی دارد. مبادله بایاس-واریانس به ما کمک می‌کند تا این تصمیم مهم را آگاهانه بگیریم و بهترین روش را برای مسئله خاص خود انتخاب کنیم.</p>

</div>

```python
# نمایش مبادله بایاس-واریانس در طبقه‌بندی

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

def generate_synthetic_data(n_samples=1000, noise_level=0.2):
    """
    تولید داده‌های مصنوعی برای نمایش مبادله بایاس-واریانس
    """
    np.random.seed(42)
    
    # ایجاد دو کلاس با توزیع‌های متفاوت
    X_class1 = np.random.randn(n_samples//2, 2) + np.array([2, 2])
    X_class2 = np.random.randn(n_samples//2, 2) + np.array([-2, -2])
    
    y = np.hstack([np.ones(n_samples//2), np.zeros(n_samples//2)])
    
    # ترکیب داده‌ها
    X = np.vstack([X_class1, X_class2])
    
    # افزودن نویز به درصدی از داده‌ها
    n_noise = int(noise_level * n_samples)
    noise_indices = np.random.choice(len(X), n_noise, replace=False)
    
    # معکوس کردن برچسب برخی از نقاط نویز
    y[noise_indices] = 1 - y[noise_indices]
    
    return X, y

def fit_classifiers(X_train, y_train):
    """
    آموزش چندین طبقه‌بند با پیچیدگی‌های متفاوت
    """
    # طبقه‌بند خطی (زیاد بایاس، کم واریانس)
    linear = LogisticRegression(max_iter=1000)
    linear.fit(X_train, y_train)
    
    # طبقه‌بند kNN (کم بایاس، زیاد واریانس)
    knn = KNeighborsClassifier(n_neighbors=5)
    knn.fit(X_train, y_train)
    
    # طبقه‌بند SVM (کم بایاس، زیاد واریانس)
    svm = SVC(kernel='rbf', C=1.0, gamma='scale')
    svm.fit(X_train, y_train)
    
    return {'linear': linear, 'knn': knn, 'svm': svm}

def evaluate_classifiers(classifiers, X_test, y_test):
    """
    ارزیابی طبقه‌بندها با محاسبه دقت
    """
    results = {}
    
    for name, clf in classifiers.items():
        y_pred = clf.predict(X_test)
        accuracy = accuracy_score(y_test, y_pred)
        results[name] = accuracy
    
    return results

def plot_decision_boundary(classifier, X, y, title, ax=None):
    """
    نمایش مرز تصمیم برای یک طبقه‌بند
    """
    if ax is None:
        ax = plt.gca()
    
    # ایجاد شبکه برای نمایش مرز
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                         np.arange(y_min, y_max, 0.1))
    
    # پیش‌بینی تمام نقاط شبکه
    Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    
    # نمایش مرز تصمیم و مناطق
    ax.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.Paired)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Paired, edgecolors='k')
    
    ax.set_title(title)
    ax.set_xlabel('Feature 1')
    ax.set_ylabel('Feature 2')
    
    return ax

def calculate_bias_variance(classifiers, X_train, y_train):
    """
    محاسبه بایاس و واریانس برای طبقه‌بندها
    """
    results = {}
    
    for name, clf in classifiers.items():
        #  استفاده از اعتبارسنجی متقابل برای تخمین بایاس و واریانس
        from sklearn.model_selection import cross_val_score
        
        # محاسبه بایاس به عنوان 1 - میانگین دقت در اعتبارسنجی متقابل
        scores = cross_val_score(clf, X_train, y_train, cv=5)
        bias = 1 - np.mean(scores)
        
        #  محاسبه واریانس به عنوان واریانس دقت در اعتبارسنجی متقابل
        variance = np.var(scores)
        
        results[name] = {'bias': bias, 'variance': variance}
    
    return results

# مثال کاربردی
print("Generating synthetic data to show bias-variance tradeoff:")
print("="*60)

# تولید داده‌های آموزشی و آزمونی
X, y = generate_synthetic_data(n_samples=1000, noise_level=0.2)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# آموزش طبقه‌بندها
classifiers = fit_classifiers(X_train, y_train)

# محاسبه بایاس و واریانس
bias_variance = calculate_bias_variance(classifiers, X_train, y_train)
print("\nBias and Variance of classifiers:")
for name, metrics in bias_variance.items():
    print(f"{name}: Bias={metrics['bias']:.4f}, Variance={metrics['variance']:.4f}")

# ارزیابی روی داده‌های آزمون
results = evaluate_classifiers(classifiers, X_test, y_test)
print("\nAccuracy of classifiers on test data:")
for name, accuracy in results.items():
    print(f"{name}: {accuracy:.4f}")

# نمایش مرزهای تصمیم
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

plot_decision_boundary(classifiers['linear'], X_test, y_test, 'Linear Classifier (High Bias, Low Variance)', axes[0])
plot_decision_boundary(classifiers['knn'], X_test, y_test, 'kNN Classifier (Low Bias, High Variance)', axes[1])
plot_decision_boundary(classifiers['svm'], X_test, y_test, 'SVM Classifier (Low Bias, High Variance)', axes[2])

plt.tight_layout()
plt.show()

# Fixed: تحلیل اصلاح‌شده بر اساس نتایج واقعی
print("\nتحلیل:")
print("- طبقه‌بند خطی: بایاس بالا اما واریانس کم، دقت قابل قبول")
print("- طبقه‌بند kNN: بایاس و واریانس متوسط، دقت کمی پایین‌تر")
print("- طبقه‌بند SVM: بایاس و واریانس متوسط، بالاترین دقت")
print("\nاین مثال نشان می‌دهد که طبقه‌بند SVM با وجود پیچیدگی بیشتر، بهترین عملکرد را در این مجموعه داده دارد.")
```

<span id="section14.7" style="font-size:24pt; font-weight:bold; color:#8549EB;">جمع‌بندی فصل ۱۴: طبقه‌بندی فضای برداری</span> 📊

<p style="font-size:16pt; text-align:right;">
چگونه می‌توانیم اسناد را بر اساس شباهتشان در یک فضای چندبعدی دسته‌بندی کنیم؟ فصل ۱۴ ما را به دنیای **طبقه‌بندی فضای برداری** می‌برد؛ رویکردی که اسناد را به صورت بردارهایی در یک فضای چندبعدی نمایش می‌دهد و بر اساس فاصله یا شباهتشان تصمیم‌گیری می‌کند. این فصل نشان می‌دهد که چگونه می‌توان با استفاده از مفاهیم هندسی و جبری، مرزهای تصمیم‌گیری را بین کلاس‌های مختلف ترسیم کرد و اسناد جدید را به درستی طبقه‌بندی نمود.
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فرض **تداوم (Contiguity Hypothesis)** را به عنوان اساس طبقه‌بندی فضای برداری معرفی کردیم. این فرض می‌گوید اسناد مربوط به یک کلاس، ناحیه‌های پیوسته‌ای را در فضای برداری تشکیل می‌دهند که با ناحیه‌های کلاس‌های دیگر همپوشانی ندارند.</li>
<li>الگوریتم **روچیو (Rocchio)** را به عنوان روشی برای طبقه‌بندی بر اساس **مراکز (Centroids)** بررسی کردیم. در این روش، هر کلاس با مرکز جرمی اسنادش نمایش داده می‌شود و مرزهای تصمیم‌گیری به صورت ابرصفحه‌هایی بین این مراکز ترسیم می‌گردند.</li>
<li>الگوریتم **k نزدیک‌ترین همسایه (kNN)** را به عنوان روشی برای طبقه‌بندی محلی بررسی کردیم. در این روش، هر سند جدید بر اساس اکثریت کلاس k همسایه نزدیک‌ترش طبقه‌بندی می‌شود و مرزهای تصمیم‌گیری به صورت قطعه‌هایی از **تسلیم‌بندی ورونویی (Voronoi Tessellation)** شکل می‌گیرند.</li>
<li>تفاوت بین **طبقه‌بندهای خطی و غیرخطی** را بررسی کردیم. طبقه‌بندهای خطی مانند روچیو و بیز ساده، مرزهای تصمیم‌گیری خطی (ابرصفحه‌ها) ایجاد می‌کنند، در حالی که روش‌هایی مانند kNN می‌توانند مرزهای تصمیم‌گیری پیچیده‌تر و غیرخطی ایجاد کنند.</li>
<li>روش‌های طبقه‌بندی برای مسائل با **بیش از دو کلاس** را بررسی کردیم: مسائل **یکی-از (One-of)** که در آن هر سند باید دقیقاً به یک کلاس تعلق گیرد، و مسائل **هر-از (Any-of)** که در آن یک سند می‌تواند به چندین کلاس تعلق گیرد.</li>
<li>مفهوم **مبادله-واریانس (Bias-Variance Tradeoff)** را به عنوان یکی از بنیادی‌ترین مفاهیم در یادگیری ماشین معرفی کردیم. این مفهوم توضیح می‌دهد که چرا هیچ روش طبقه‌بندی به طور جهانی بر دیگری برتری ندارد و چگونه باید بین پیچیدگی مدل و حساسیت به داده‌های آموزشی تعادل برقرار کرد.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
<b>🔹 مبادله-واریانس در سیستم‌های تشخیص پزشکی:</b> در سیستم‌های تشخیص پزشکی، باید بین پیچیدگی مدل و حساسیت به داده‌های آموزشی تعادل برقرار کرد. یک مدل با واریانس بالا (مانند kNN با k کوچک) ممکن است داده‌های آموزشی را به خوبی به خاطر بسپارد، اما در تشخیص بیماران جدید عملکرد ضعیفی داشته باشد. از سوی دیگر، یک مدل با مبادله بالا (مانند یک طبقه‌بند خطی ساده) ممکن است نتواند الگوهای پیچیده بیماری را تشخیص دهد. در عمل، پزشکان و محققان باید با توجه به مقدار داده‌های در دسترس و پیچیدگی مسئله، مدل مناسبی را انتخاب کنند که مبادله و واریانس متعادلی داشته باشد.
