<p style="font-size:18pt; text-align:center;">
<span id="section20.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">20، 20.1 و ویژگی‌های کرالر وب: از تئوری تا عمل</span> 🕷️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌐 گوگل: چگونه میلیاردها صفحه وب را فهرست می‌کنند؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم مهندسی گوگل کار می‌کنید و مسئول بهبود موتور جستجوی این شرکت هستید. هر روز میلیاردها صفحه وب توسط ربات‌های گوگل (Crawlers) بازدید و ایندکس می‌شوند. این یک فرآیند بسیار پیچیده است، زیرا وب با سرعت فوق‌العاده در حال تغییر است.</p>
<p style="font-size:16pt; text-align:right;">
چالش‌های اصلی که تیم شما با آن‌ها روبرو است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>دامنه‌های کرالر:</strong> وب‌سایت‌هایی با محدودیت‌های مختلف (مانند فروشگاه‌های آنلاین با میلیون‌ها محصول).</li>
<li><strong>تله‌های کرالر:</strong> صفحاتی که به‌طور نامتناهی تولید می‌شوند و ربات‌ها را در حلقه‌های بی‌نهایت قرار می‌دهند.</li>
<li><strong>سرورهای پرسرعت:</strong> وب‌سایت‌هایی که بار ترافیکی بالایی دارند و به‌روزرسانی مداوم نیاز دارند.</li>
</ul>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🕷️ ربات‌های گوگل: از تئوری تا عمل</p>
<p style="font-size:16pt; text-align:right;">
ربات‌های گوگل (Googlebots) باید ویژگی‌های کلیدی که در کتاب ذکر شد را داشته باشند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>استحکام:</strong> مقاومت در برابر <strong>دام‌های کرالر</strong> که به‌طور نامتناهی تولید می‌شوند.</li>
<li><strong>ادب:</strong> رعایت فایل <code>robots.txt</code> و محدود کردن نرخ درخواست به هر دامنه.</li>
<li><strong>توزیع‌پذیری:</strong> امکان اجرا روی چندین ماشین برای پوشش سریع‌تر وب.</li>
<li><strong>مقیاس‌پذیری:</strong> افزایش نرخ کرالینگ با اضافه کردن منابع.</li>
<li><strong>کارایی:</strong> استفاده بهینه از منابع سیستم (CPU، ذخیره، پهنای باند).</li>
<li><strong>کیفیت:</strong> اولویت‌دهی به صفحات <strong>مفیدتر</strong> (مثلاً با بر اساس اعتبار دامنه).</li>
<li><strong>تازگی:</strong> به‌روزرسانی مداوم صفحات بر اساس <strong>نرخ تغییر</strong> آن‌ها.</li>
<li><strong>گسترش‌پذیری:</strong> معماری ماژولار برای پشتیبانی از پروتکل‌ها یا فرمت‌های داده جدید.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این ویژگی‌ها به ربات‌های گوگل اجازه می‌دهند تا با چالش‌های بزرگ وب مقابله کنند و در عین حال از بار سرورها و شبکه به صورت بهینه استفاده کنند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚫 تله‌های کرالر: چالش‌های پیشرفته</p>
<p style="font-size:16pt; text-align:right;">
یکی از بزرگترین چالش‌ها برای کرالرهای وب، <strong>تله‌های کرالر</strong> (Spider Traps) است. این صفحات به‌طور عمدی یا ناشیانه تولید می‌شوند تا ربات‌ها را به در حلقه‌های بی‌نهایت بیندازند.</p>
<p style="font-size:16pt; text-align:right;">
انواع رایج تله‌های کرالر:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>تله‌های تقویمی:</strong> صفحاتی با لینک‌های بی‌پایان که کاربر را به صفحات بی‌نهایت هدایت می‌کنند.</li>
<li><strong>تله‌های زمانی:</strong> صفحاتی با رویدادهای زمانی بی‌پایان که کاربر را در وب نگه می‌دارند.</li>
<li><strong>تله‌های نرم‌افزاری:</strong> صفحاتی که نیاز به نصب نرم‌افزار برای نمایش محتوای کامل دارند.</li>
<li><strong>تله‌های منطقی:</strong> ساختارهای پیچیده با مسیرهای تکراری که ربات‌ها را در مسیرهای بی‌پایان به دام می‌اندازند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
ربات‌های گوگل با الگوریتم‌های پیشرفته خود را در برابر این تله‌ها مقاومه می‌کنند. برای مثال، اگر یک صفحه بیش از 1000 لینک خروجی داشته باشد، ربات ممکن است آن را به عنوان یک تله تقویمی شناسایی و از کرالینگ کامل آن خودداری کند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🏗 معماری توزیع‌شده: برای کرالینگ در مقیاس بزرگ</p>
<p style="font-size:16pt; text-align:right;">
برای کرالینگ وب در مقیاس بزرگ (مانند کل وب)، گوگل از معماری <strong>توزیع‌شده</strong> استفاده می‌کند. در این معماری:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>URL Server:</strong> مسئولیت تخصیص URLها به ربات‌های مختلف برای کرالینگ موازی.</li>
<li><strong>Crawler:</strong> کامپیوترهایی که صفحات وب را بارگیری و تحلیل می‌کنند.</li>
<li><strong>Links Database:</strong> پایگاه داده‌ای که لینک‌های بین صفحات را ذخیره می‌کند.</li>
<li><strong>Document Store:</strong> ذخیره محتوای صفحات پس از تحلیل.</li>
<li><strong>Indexer:</strong> سیستمی‌هایی که ایندکس‌های جستجو را ایجاد و به‌روزرسانی می‌کنند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این معماری به گوگل اجازه می‌دهد تا میلیاردها صفحه را به صورت موازی کرالینگ کند و با سرعت بسیار بالا ایندکس‌ها را به‌روزرسانی کند. هر بخش می‌تواند روی ماشین‌های مختلف اجرا شود و به صورت مستقل به‌روزرسانی شود.</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🌐 کاربردهای عملی: از موتورهای جستجو تا توصیه‌گر محتوا</p>
<p style="font-size:16pt; text-align:right;">
کرالینگ وب تنها مرحله اول از زنجیره بازیابی اطلاعات است. مرحله بعد، <strong>ایندکس‌گذاری و جستجو</strong> است که در آن کاربران می‌توانند به اطلاعات دسترسی پیدا کنند. در این مرحله نیز، رویکردهای یادگیری ماشین نقش کلیدی دارند.</p>
<p style="font-size:16pt; text-align:right;">
کاربردهای عملی که از این رویکردها استفاده می‌کنند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>موتورهای جستجو:</strong> برای رتبه‌بندی نتایج جستجو بر اساس ارتباط با پرس‌وجو کاربر.</li>
<li><strong>سیستم‌های توصیه:</strong> برای ارائه پیشنهادهای شخصی‌سازی بر اساس تاریخچه و رفتار کاربر.</li>
<li><strong>پلتفرم‌های تجارت الکترونیک:</strong> برای ترکیب داده‌های مختلف (جستجو، خرید، بازدید محصول) برای ارائه پیشنهادهای جامع.</li>
<li><strong>شبکه‌های تحلیل داده:</strong> برای درک الگوهای پیچیده در رفتار کاربران و بهبود الگوریتم‌ها.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این کاربردها نشان می‌دهند که چگونه کرالینگ وب و ایندکس‌گذاری می‌توانند پایه‌ای برای سیستم‌های هوشمندتر باشند که تجربه کاربری را به شکل معنادارتری به کاربران ارائه دهند.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section20.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">20.2 کرالینگ: کاوشگران دنیای وب</span> 🕷️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌐 سفر به دنیای وب: چگونه ربات‌های گوگل میلیاردها صفحه را کشف می‌کنند؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در حال طراحی یک سیستم برای گوگل هستید که باید تمام صفحات وب را کشف و ایندکس کند. این فرآیند که به آن "کرالینگ" می‌گویند، مانند سفر یک کاوشگر به دنیای ناشناخته وب است.</p>
<p style="font-size:16pt; text-align:right;">
ربات‌های گوگل (Googlebots) این سفر را به این صورت انجام می‌دهند:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>نقطه شروع:</strong> از چند URL معروف مانند <code>google.com</code>، <code>wikipedia.org</code> و <code>twitter.com</code> شروع می‌کنند. این‌ها "مجموعه دانه" (seed set) نامیده می‌شوند.</li>
<li><strong>بازدید از صفحه:</strong> ربات به این آدرس‌ها سر زده و محتوای صفحه را دریافت می‌کند.</li>
<li><strong>تحلیل محتوا:</strong> متن صفحه برای ایندکس‌گذاری استخراج می‌شود و تمام لینک‌های داخل صفحه شناسایی می‌شوند.</li>
<li><strong>ادامه سفر:</strong> لینک‌های جدید به صفی به نام "مرز URL" (URL Frontier) اضافه می‌شوند تا در آینده بازدید شوند.</li>
</ol>
<p style="font-size:16pt; text-align:right;">
این فرآیند به صورت بازگشتی ادامه می‌یابد و ربات‌های گوگل روزانه میلیاردها صفحه جدید را کشف و ایندکس می‌کنند.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🚀 چالش‌های عملی: از تئوری تا اجرا</p>
<p style="font-size:16pt; text-align:right;">
در نگاه اول، کرالینگ ممکن است ساده به نظر برسد، اما در عمل با چالش‌های بزرگی روبرو است. برای درک این چالش‌ها، به مثال زیر توجه کنید:</p>
<p style="font-size:16pt; text-align:right;">
شرکت "دیجی‌کالا" می‌خواهد تمام محصولات خود را در گوگل ایندکس کند. با میلیون‌ها محصول و هزاران صفحه جدید هر روز، این کار به سادگی امکان‌پذیر نیست. ربات‌های گوگل باید:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مقیاس‌پذیر باشند:</strong> بتوانند میلیاردها صفحه را در زمان معقولی پردازش کنند.</li>
<li><strong>کارا باشند:</strong> از منابع سیستم بهینه استفاده کنند.</li>
<li><strong>مودب باشند:</strong> به سرورها فشار نیاورند.</li>
<li><strong>مقاوم باشند:</strong> در برابر خطاها و تله‌های وب مقاومت کنند.</li>
<li><strong>قابل گسترش باشند:</strong> بتوانند با تغییرات وب سازگار شوند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
برای مثال، کرالر Mercator که پایه بسیاری از کرالرهای تجاری است، می‌تواند صدها صفحه در ثانیه را پردازش کند. این یعنی برای ایندکس کردن یک میلیارد صفحه، فقط چند هفته زمان نیاز است!</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🤝 ادب در دنیای دیجیتال: چرا ربات‌ها باید مودب باشند؟</p>
<p style="font-size:16pt; text-align:right;">
شاید فکر کنید ربات‌ها می‌توانند هر طور که می‌خواهند به سرورها درخواست ارسال کنند، اما در واقعیت، این کار می‌تواند به سرورها آسیب بزند. برای درک این موضوع، به این مثال توجه کنید:</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید یک فروشگاه کوچک آنلاین دارید که روزانه ۱۰۰ بازدیدکننده دارد. ناگهان یک ربات به صورت همزمان ۱۰۰۰ درخواست به سرور شما ارسال می‌کند. این کار باعث می‌شود سرور شما از کار بیفتد و مشتریان واقعی نتوانند به فروشگاه شما دسترسی پیدا کنند.</p>
<p style="font-size:16pt; text-align:right;">
برای جلوگیری از این مشکلات، کرالرهای حرفه‌ای باید این قوانین را رعایت کنند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فقط <strong>یک اتصال همزمان</strong> به هر سرور.</li>
<li><strong>تأخیر چند ثانیه‌ای</strong> بین درخواست‌های متوالی به یک سرور.</li>
<li>رعایت سیاست‌های فایل <code>robots.txt</code> که مشخص می‌کند کدام بخش‌های سایت نباید کرال شوند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این قوانین به عنوان "ادب کرالر" (Crawler Politeness) شناخته می‌شوند و برای سلامت اکوسیستم وب ضروری هستند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔄 کرالینگ مداوم: چگونه گوگل اطلاعات را به‌روز نگه می‌دارد؟</p>
<p style="font-size:16pt; text-align:right;">
وب به سرعت در حال تغییر است و صفحات جدید دائماً ایجاد می‌شوند. برای اینکه موتورهای جستجو اطلاعات به‌روزی ارائه دهند، باید به صورت مداوم وب را کرال کنند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، سایت‌های خبری مانند <code>irna.ir</code> یا <code>isna.ir</code> هر چند دقیقه یک بار به‌روزرسانی می‌شوند. ربات‌های گوگل این سایت‌ها را بیشتر از صفحاتی که به ندرت تغییر می‌کنند (مانند یک مقاله علمی قدیمی) بازدید می‌کنند.</p>
<p style="font-size:16pt; text-align:right;">
در <strong>کرالینگ مداوم</strong> (Continuous Crawling)، URLهای بازدیدشده دوباره به مرز URL اضافه می‌شوند تا در آینده دوباره بازدید شوند. این فرآیند به گوگل اجازه می‌دهد تا همیشه جدیدترین اطلاعات را در اختیار کاربران قرار دهد.</p>
</div>

</div>

```python
import time
import random
from collections import deque, defaultdict
from urllib.parse import urlsplit

class WebCrawlerSimulator:
    """
    شبیه‌سازی یک کرالر وب ساده که:
    - از یک مجموعه دانه شروع می‌کند
    - لینک‌ها را استخراج و به مرز URL اضافه می‌کند
    - برای هر میزبان، فقط یک درخواست همزمان انجام می‌دهد
    - بین درخواست‌ها به یک میزبان، تأخیر رعایت می‌شود
    """
    
    def __init__(self, seed_urls, politeness_delay=2):
        """
        مقداردهی اولیه کرالر.
        - seed_urls: لیست URLهای اولیه
        - politeness_delay: تأخیر حداقل بین درخواست‌ها به یک میزبان (ثانیه)
        """
        self.seed_urls = seed_urls
        self.politeness_delay = politeness_delay
        self.url_frontier = deque(seed_urls)  # صف URLهای بازدید‌نشده
        self.fetched_urls = set()             # مجموعه URLهای بازدیدشده
        self.last_fetch_time = defaultdict(float)  # آخرین زمان درخواست به هر میزبان
        self.host_queues = defaultdict(deque) # صف URLها برای هر میزبان
    
    def extract_host(self, url):
        """استخراج نام میزبان از URL"""
        return urlsplit(url).netloc.lower()
    
    def simulate_fetch_page(self, url):
        """
        شبیه‌سازی دریافت صفحه و استخراج لینک‌ها.
        در واقعیت، این تابع باید HTML را دریافت و پارس کند.
        """
        print(f"  دریافت صفحه: {url}")
        # شبیه‌سازی زمان پردازش
        time.sleep(0.1)
        
        # تولید لینک‌های تصادفی برای شبیه‌سازی
        host = self.extract_host(url)
        new_links = []
        for i in range(random.randint(0, 3)):
            new_links.append(f"http://{host}/page_{random.randint(1, 100)}")
        return new_links
    
    def is_polite_to_fetch(self, host):
        """بررسی اینکه آیا می‌توان به این میزبان درخواست فرستاد یا خیر"""
        current_time = time.time()
        last_time = self.last_fetch_time[host]
        return (current_time - last_time) >= self.politeness_delay
    
    def crawl(self, max_pages=10):
        """
        اجرای فرآیند کرالینگ تا حداکثر تعداد صفحات مشخص‌شده.
        """
        print(f"شروع کرالینگ از {len(self.seed_urls)} URL اولیه...")
        pages_fetched = 0
        
        while self.url_frontier and pages_fetched < max_pages:
            # توزیع URLها در صف‌های میزبان
            while self.url_frontier:
                url = self.url_frontier.popleft()
                if url in self.fetched_urls:
                    continue
                host = self.extract_host(url)
                self.host_queues[host].append(url)
                self.fetched_urls.add(url)  # جلوگیری از بازدید مجدد
            
            # پردازش هر میزبان که شرایط ادب را دارد
            hosts_ready = [host for host in self.host_queues if self.is_polite_to_fetch(host)]
            
            if not hosts_ready:
                # اگر هیچ میزبانی آماده نیست، کمی صبر کن
                time.sleep(0.5)
                continue
            
            # انتخاب یک میزبان آماده
            host = random.choice(hosts_ready)
            if not self.host_queues[host]:
                continue
            
            url = self.host_queues[host].popleft()
            self.last_fetch_time[host] = time.time()
            
            # شبیه‌سازی دریافت صفحه و استخراج لینک‌ها
            new_links = self.simulate_fetch_page(url)
            pages_fetched += 1
            
            # اضافه کردن لینک‌های جدید به مرز URL
            for link in new_links:
                if link not in self.fetched_urls:
                    self.url_frontier.append(link)
            
            print(f"  لینک‌های جدید: {len(new_links)} | صفحات بازدیدشده: {pages_fetched}/{max_pages}")
            print("-" * 50)
        
        print(f"\nکرالینگ به پایان رسید. تعداد کل صفحات بازدیدشده: {pages_fetched}")

# مثال اجرا
if __name__ == "__main__":
    seed_urls = [
        "http://example.com/home",
        "http://sample.org/index",
        "http://test.net/main"
    ]
    
    crawler = WebCrawlerSimulator(seed_urls, politeness_delay=1)
    crawler.crawl(max_pages=8)
    
    print("\nنکته:")
    print("این شبیه‌سازی، مکانیزم اصلی کرالینگ را نشان می‌دهد:")
    print("- مدیریت مرز URL (URL Frontier)")
    print("- رعایت ادب میزبان (یک اتصال، تأخیر بین درخواست‌ها)")
    print("- جلوگیری از بازدید مجدد از URLهای یکسان")
    print("در پیاده‌سازی واقعی، نیاز به پردازش HTTP، HTML و پایگاه داده است.")
```

<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم مهندسی گوگل کار می‌کنید و مسئول طراحی کرالری هستید که باید هر روز میلیاردها صفحه وب را ایندکس کند. این کرالر مانند یک ارتش کوچک از ربات‌های هوشمند عمل می‌کند که هر کدام وظیفه خاصی دارند.</p>
<p style="font-size:16pt; text-align:right;">
معماری این کرالر از چندین ماژول اصلی تشکیل شده است که با هم هماهنگ کار می‌کنند:</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📋 مرز URL: صف انتظار دیجیتال</p>
<p style="font-size:16pt; text-align:right;">
مرز URL (URL Frontier) مانند یک صف انتظار بی‌پایان است که تمام URLهایی که باید بازدید شوند را نگهداری می‌کند. در کرالینگ مداوم، حتی URLهایی که قبلاً بازدید شده‌اند نیز به این صف بازمی‌گردند تا در آینده دوباره بررسی شوند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وقتی گوگل سایت دیجی‌کالا را کرال می‌کند، تمام صفحات محصولات در این صف قرار می‌گیرند. با توجه به اینکه قیمت‌ها و موجودی محصولات به طور مداوم تغییر می‌کنند، این URLها به طور منظم به صف بازگردانده می‌شوند تا اطلاعات به‌روز باقی بمانند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🌐 ماژول DNS: دفترچه تلفن اینترنت</p>
<p style="font-size:16pt; text-align:right;">
قبل از اینکه کرالر بتواند صفحه‌ای را دریافت کند، باید آدرس IP سرور را پیدا کند. این کار توسط ماژول DNS انجام می‌شود که مانند دفترچه تلفن اینترنت عمل می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وقتی کرالر می‌خواهد صفحه <code>https://www.bing.com/search?q=crawling</code> را دریافت کند، ابتدا باید آدرس IP سرور bing.com را پیدا کند. این فرآیند ممکن است چند میلی‌ثانیه طول بکشد، اما برای کرالرهایی که میلیاردها درخواست ارسال می‌کنند، این زمان اضافی می‌تواند تأثیر زیادی بر کارایی داشته باشد.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📥 ماژول Fetch: پیک اینترنت</p>
<p style="font-size:16pt; text-align:right;">
ماژول Fetch مانند یک پیک دیجیتال است که به سرورها مراجعه کرده و محتوای صفحات وب را تحویل می‌گیرد. این ماژول از پروتکل HTTP برای برقراری ارتباط با سرورها استفاده می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وقتی کرالر گوگل می‌خواهد صفحه اصلی سایت ورزش سه را دریافت کند، یک درخواست HTTP به سرور ورزش سه ارسال می‌کند و سرور محتوای HTML صفحه را به کرالر برمی‌گرداند.</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">📝 ماژول Parse: خواننده هوشمند</p>
<p style="font-size:16pt; text-align:right;">
ماژول Parse مانند یک خواننده هوشمند عمل می‌کند که محتوای HTML را تحلیل کرده و اطلاعات مهم را استخراج می‌کند. این ماژول متن اصلی صفحه، لینک‌ها و سایر اطلاعات مفید را جدا می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وقتی کرالر صفحه اصلی سایت تپسل (tapsell.ir)را تحلیل می‌کند، این ماژول تمام لینک‌های موجود در صفحه را استخراج کرده و برای بررسی‌های بعدی به  مرز URL اضافه می‌کند. همچنین، متن اصلی صفحه برای ایندکس‌گذاری به سیستم جستجو ارسال می‌شود.</p>
</div>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🚫 حذف محتوای تکراری: جلوگیری از اتلاف منابع</p>
<p style="font-size:16pt; text-align:right;">
یکی از چالش‌های بزرگ کرالرها، جلوگیری از پردازش محتوای تکراری است. این کار با استفاده از اثر انگشت (Fingerprint) یا Shingling انجام می‌شود.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، ممکن است یک مقاله خبری در چندین سایت خبری مختلف منتشر شود. کرالر باید بتواند تشخیص دهد که این محتوا تکراری است و از پردازش مجدد آن خودداری کند. این کار با محاسبه اثر انگشت محتوا و مقایسه آن با اثر انگشت‌های موجود انجام می‌شود.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🤖 احترام به حریم خصوصی: پروتکل Robots.txt</p>
<p style="font-size:16pt; text-align:right;">
وب‌مسترها می‌توانند با استفاده از فایل robots.txt مشخص کنند که کدام بخش‌های سایتشان نباید توسط کرالرها بازدید شود. این فایل مانند یک تابلو "ورود ممنوع" برای ربات‌ها عمل می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، فایل robots.txt سایت آمازون ممکن است مشخص کند که کرالرها نباید صفحات پرداخت و اطلاعات شخصی کاربران را بازدید کنند. این فایل باید مستقیماً قبل از دریافت صفحه بررسی شود تا از نقض قوانین سایت جلوگیری شود.</p>
<p style="font-size:16pt; text-align:right;">
در اینجا یک نمونه از فایل robots.txt سایت توییتر آمده است:</p>
<pre style="background:#f8f9fa; padding:10px; border-radius:5px; overflow-x:auto; font-family:monospace; font-size:14pt; text-align:left;">
User-agent: *
Disallow: /search
Disallow: /sms
Disallow: /sessions
Disallow: /settings

User-agent: Googlebot
Allow: /hashtags
Allow: /i/
Allow: /intent/
</pre>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🌍 کرالینگ توزیع‌شده: کار تیمی در مقیاس جهانی</p>
<p style="font-size:16pt; text-align:right;">
برای کرالینگ وب در مقیاس بزرگ، شرکت‌هایی مانند گوگل و مایکروسافت از معماری توزیع‌شده استفاده می‌کنند. در این معماری، کرالرها در سراسر جهان توزیع شده‌اند و هر کرالر مسئولیت کرالینگ مجموعه‌ای از سایت‌ها را بر عهده دارد.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، گوگل ممکن است کرالرهایی در اروپا داشته باشد که کرالینگ سایت‌های اروپایی، و کرالرهایی در آسیا که سایت‌های آسیایی را کرال می‌کنند. این توزیع جغرافیایی به کاهش تأخیر و افزایش کارایی کمک می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
در این معماری، یک "تقسیم‌کننده میزبان" (Host Splitter) مشخص می‌کند که هر URL به کدام کرالر ارسال شود. این تقسیم‌بندی معمولاً بر اساس هش نام دامنه انجام می‌شود تا توزیع بار به صورت یکنواخت انجام شود.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚡ چالش‌های عملی: از تئوری تا پیاده‌سازی</p>
<p style="font-size:16pt; text-align:right;">
پیاده‌سازی یک کرالر در مقیاس بزرگ با چالش‌های زیادی همراه است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مدیریت حافظه:</strong> مرز URL ممکن است حاوی میلیاردها آدرس باشد که نیاز به مدیریت حافظه هوشمند دارد.</li>
<li><strong>تحمل خطا:</strong> اگر یک کرالر از کار بیفتد، باید مکانیزمی برای ادامه کار از آخرین نقطه وجود داشته باشد.</li>
<li><strong>توزیع بار:</strong> باید اطمینان حاصل شود که هیچ سروری تحت فشار بیش از حد قرار نگیرد.</li>
<li><strong>به‌روزرسانی مداوم:</strong> وب دائماً در حال تغییر است و کرالر باید بتواند این تغییرات را پیگیری کند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
برای مقابله با این چالش‌ها، شرکت‌های بزرگ از سیستم‌های پیچیده‌ای استفاده می‌کنند که شامل مکانیزم‌های ذخیره‌سازی توزیع‌شده، الگوریتم‌های هوشمند برای اولویت‌بندی URLها و سیستم‌های نظارتی پیشرفته است.</p>

<p style="font-size:18pt; text-align:center;">
<span id="section20.2.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">20.2.2 تحلیل DNS: دفترچه تلفن اینترنت</span> 🌐
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📱 از آمازون تا گوگل: چرا DNS برای کرالرها حیاتی است؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید می‌خواهید به دوستتان پیام دهید، اما فقط شماره تلفن او را دارید، نه نامش. حالا برعکس: فقط نام دوستتان را می‌دانید، اما شماره تلفنش را نه. این دقیقاً همان چیزی است که در اینترنت اتفاق می‌افتد.</p>
<p style="font-size:16pt; text-align:right;">
وقتی در مرورگر خود <code>www.amazon.com</code> را تایپ می‌کنید، کامپیوتر شما نمی‌داند که این سایت در کجای اینترنت قرار دارد. باید نام دامنه به آدرس IP عددی (مثل <code>174.129.25.17</code>) تبدیل شود. این فرآیند را <strong>تحلیل DNS</strong> (DNS Resolution) می‌نامیم.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">⏱️ گلوگاه سرعت: چرا DNS کرالرها را کند می‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
برای یک کاربر عادی، تحلیل DNS فقط چند میلی‌ثانیه طول می‌کشد و اصلاً قابل توجه نیست. اما برای یک کرالر که باید در ثانیه صدها صفحه را دریافت کند، این فرآیند می‌تواند یک فاجعه باشد!</p>
<p style="font-size:16pt; text-align:right;">
چرا؟</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>تحلیل DNS ممکن است <strong>چند ثانیه</strong> طول بکشد</li>
<li>برای دامنه‌های ناشناخته یا مشکوک، این زمان حتی بیشتر هم می‌شود</li>
<li>اگر کرالر بخواهد ۱۰۰۰ صفحه در ثانیه دریافت کند، هر تأخیر چند ثانیه‌ای یک مانع بزرگ است</li>
</ul>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">💾 راه‌حل اولیه: کش کردن نتایج DNS</p>
<p style="font-size:16pt; text-align:right;">
اولین راه‌حلی که به ذهن می‌رسد، <strong>کش کردن</strong> نتایج DNS است. مانند اینکه شماره تلفن دوستتان را در دفترچه تلفن خود یادداشت کنید تا دیگر نیازی به جستجو نداشته باشید.</p>
<p style="font-size:16pt; text-align:right;">
اما این راه‌حل محدودیت‌هایی دارد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>به دلیل <strong>محدودیت‌های ادب کرالینگ</strong> (فاصله بین درخواست‌ها به یک سرور)، نرخ استفاده مجدد از کش کاهش می‌یابد</li>
<li>آدرس‌های IP ممکن است تغییر کنند، بنابراین کش‌ها باید به‌طور منظم به‌روز شوند</li>
<li>برای کرالرهای بزرگ، حتی کش هم کافی نیست</li>
</ul>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🚫 مشکل عمیق‌تر: رفتار همگام کتابخانه‌های DNS</p>
<p style="font-size:16pt; text-align:right;">
مشکل اصلی اینجاست: کتابخانه‌های استاندارد DNS <strong>همگام</strong> (Synchronous) عمل می‌کنند. یعنی وقتی یک کرالر درخواست DNS ارسال می‌کند، تمام فعالیت‌های دیگر متوقف می‌شوند تا پاسخ دریافت شود.</p>
<p style="font-size:16pt; text-align:right;">
این مانند این است که در یک رستوران، گارسون فقط بتواند یک سفارش را بگیرد و تا زمانی که غذا برای آن میز آماده نشده، از میزهای دیگر سفارش نگیرد!</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🏗️ راه‌حل گوگل: DNS غیرهمگام سفارشی</p>
<p style="font-size:16pt; text-align:right;">
شرکت‌هایی مانند گوگل و آمازون که کرالرهای بسیار بزرگی دارند، این مشکل را با پیاده‌سازی <strong>resolver DNS سفارشی و غیرهمگام</strong> حل کرده‌اند. این سیستم چگونه کار می‌کند؟</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>هر رشته کرالر، یک پیام DNS ارسال کرده و به کار خود ادامه می‌دهد (منتظر نمی‌ماند)</li>
<li>یک <strong>رشته اختصاصی DNS</strong> به‌طور مداوم پاسخ‌ها را دریافت کرده و به رشته‌های مربوطه تحویل می‌دهد</li>
<li>اگر پاسخی در زمان مشخصی (مثلاً ۱ ثانیه) دریافت نشد، درخواست مجدداً ارسال می‌شود</li>
<li>این تلاش‌ها معمولاً تا ۵ بار تکرار می‌شوند و زمان انتظار به‌صورت نمایی افزایش می‌یابد (از ۱ ثانیه تا ۹۰ ثانیه)</li>
</ol>
<p style="font-size:16pt; text-align:right;">
این طراحی به کرالر اجازه می‌دهد تا به جای انتظار برای پاسخ‌های DNS، به کارهای دیگر ادامه دهد و در نتیجه سرعت کرالینگ به‌شدت افزایش می‌یابد.</p>
</div>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎯 نتیجه‌گیری: چرا این موضوع مهم است؟</p>
<p style="font-size:16pt; text-align:right;">
بدون راه‌حل‌های هوشمند برای تحلیل DNS، کرالرهای مدرن نمی‌توانستند با سرعت مورد نیاز عمل کنند. این به‌ویژه برای شرکت‌هایی مانند گوگل که باید میلیاردها صفحه را در روز ایندکس کنند، حیاتی است.</p>
<p style="font-size:16pt; text-align:right;">
در واقع، بهینه‌سازی DNS یکی از رازهای موفقیت کرالرهای بزرگ است که به آنها اجازه می‌دهد وب را با سرعت فوق‌العاده‌ای پوشش دهند.</p>
</div>

<p style="font-size:18pt; text-align:center;">
<span id="section20.2.3" style="font-size:24pt; font-weight:bold; color:#1B263B;">20.2.3 مرز URL: رهبر ترافیک اینترنت</span> 🚦
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌐 چالش گوگل: چگونه میلیاردها صفحه را منظم کرال کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید رهبر تیم کرالینگ گوگل هستید. هر ثانیه، هزاران صفحه جدید به وب اضافه می‌شود و شما باید تصمیم بگیرید کدام صفحه را اول بازدید کنید. این مثل مدیریت ترافیک یک کلان شهر است که در آن همه جاده‌ها یک‌باره شلوغ شده‌اند.</p>
<p style="font-size:16pt; text-align:right;">
شما با دو چالش بزرگ روبرو هستید:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>اولویت‌دهی هوشمند:</strong> صفحاتی که مهم هستند و به سرعت تغییر می‌کنند (مانند خبرهای فوری در <code>bbc.com</code> یا قیمت‌های روز در <code>amazon.com</code>) باید سریع‌تر از صفحات کم‌اهمیت‌تر کرال شوند.</li>
<li><strong>ادب در ترافیک:</strong> شما نباید با ارسال همزمان صدها درخواست به یک سرور (مثلاً <code>apple.com</code> در زمان عرضه یک محصول جدید)، آن را از کار بیندازید.</li>
</ol>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🏗️ راه‌حل: معماری دو لایه مرز URL</p>
<p style="font-size:16pt; text-align:right;">
برای حل این چالش، کرالرهای مدرن از یک سیستم صف هوشمند به نام <strong>مرز URL</strong> (URL Frontier) استفاده می‌کنند. این سیستم مانند یک مرکز کنترل ترافیک پیشرفته عمل می‌کند که از دو بخش اصلی تشکیل شده است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>صف‌های جلو (Front Queues):</strong> این صف‌ها مانند خطوط ویژه وسایل نقلیه اضطراری عمل می‌کنند. هر کدام از این صف‌ها برای سطح اولویت خاصی طراحی شده‌اند. برای مثال، URLهای مربوط به خبرهای فوری در بالاترین صف قرار می‌گیرند، در حالی که URLهای وبلاگ‌های شخصی در صف‌های با اولویت پایین‌تر قرار می‌گیرند.</li>
<li><strong>صف‌های عقب (Back Queues):</strong> این صف‌ها مانند چراغ‌های راهنمایی برای هر خیابان (میزبان) خاص عمل می‌کنند. هر صف عقب فقط شامل URLهای یک میزبان خاص است. این تضمین می‌کند که در هر لحظه فقط یک درخواست به یک سرور خاص ارسال شود.</li>
</ul>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚙️ هماهنگ‌کننده اصلی: هیپ زمان‌بندی</p>
<p style="font-size:16pt; text-align:right;">
بین این دو لایه، یک <strong>هیپ</strong> (Heap) قرار دارد که مانند یک ساعت هوشمند عمل می‌کند. این هیپ برای هر میزبان، زودترین زمانی که می‌توان دوباره به آن درخواست ارسال کرد را ذخیره می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، اگر آخرین بار در ساعت ۱۰:۰۰:۰۰ به سرور <code>nytimes.com</code> درخواست ارسال شده و این درخواست ۲ ثانیه طول کشیده باشد، هیپ ممکن است زمان مجاز بعدی را ساعت ۱۰:۰۰:۲۰ (یعنی ۱۰ برابر زمان درخواست قبلی) تنظیم کند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔄 سفر یک URL از ورود تا خروج</p>
<p style="font-size:16pt; text-align:right;">
بیایید سفر یک URL خبری از <code>reuters.com</code> را دنبال کنیم:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>ورود و اولویت‌دهی:</strong> URL خبری به سیستم وارد شده و به دلیل اهمیت بالای خبر، به بالاترین صف جلو اضافه می‌شود.</li>
<li><strong>انتقال به صف عقب:</strong> وقتی یک صف عقب خالی می‌شود، یک "کنترل‌کننده هوشمند" URL را از صف‌های جلو (با تمایل به صف‌های با اولویت بالاتر) برداشته و آن را به صف عقب مربوط به <code>reuters.com</code> اضافه می‌کند.</li>
<li><strong>انتظار برای زمان مجاز:</strong> کرالر به هیپ نگاه می‌کند و می‌بیند که آیا زمان مجاز برای ارسال درخواست به <code>reuters.com</code> فرا رسیده است یا خیر.</li>
<li><strong>کرالینگ:</strong> پس از رسیدن به زمان مجاز، کرالر URL را از صف عقب برداشته و صفحه را دریافت می‌کند.</li>
<li><strong>به‌روزرسانی زمان:</strong> پس از اتمام کرالینگ، زمان مجاز بعدی برای <code>reuters.com</code> در هیپ به‌روزرسانی می‌شود.</li>
</ol>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">💾 مدیریت حجم عظیم: وقتی حافظه RAM کافی نیست</p>
<p style="font-size:16pt; text-align:right;">
در کرالینگ در مقیاس وب، مرز URL ممکن است به میلیاردها آدرس برسد که بسیار بزرگ‌تر از حافظه RAM یک سرور است. راه‌حل این است که بخش اعظم مرز URL روی <strong>دیسک</strong> ذخیره شود.</p>
<p style="font-size:16pt; text-align:right;">
این مانند یک پارکینگ خیلی بزرگ است که اکثر خودروها در آن منتظر می‌مانند و فقط تعداد محدودی در مرکز کنترل ترافیک (حافظه RAM) برای پردازش فوری قرار دارند. این طراحی به کرالرها اجازه می‌دهد تا کل وب را مدیریت کنند، حتی با منابع محاسباتی محدود.</p>
</div>

</div>

```python
import heapq
import time
import random
from collections import deque, defaultdict

class URLFrontierSimulator:
    """
    شبیه‌سازی مرز URL با دو لایه:
    - صف‌های جلو (Front Queues): برای اولویت‌دهی بر اساس کیفیت و نرخ تغییر
    - صف‌های عقب (Back Queues): برای رعایت ادب (یک میزبان در هر صف)
    - هیپ زمان‌بندی: برای مدیریت زمان مجاز درخواست به هر میزبان
    """
    
    def __init__(self, num_front_queues=3, num_back_queues=10):
        # لایه اولویت‌دهی
        self.front_queues = [deque() for _ in range(num_front_queues)]  # F صف جلو
        self.num_front = num_front_queues
        
        # لایه ادب
        self.back_queues = [deque() for _ in range(num_back_queues)]   # B صف عقب
        self.host_to_back_queue = {}  # نگاشت میزبان -> شماره صف عقب
        self.next_back_queue = 0      # برای تخصیص چرخشی صف عقب جدید
        
        # هیپ زمان‌بندی: (زمان_مجاز, شماره_صف_عقب)
        self.time_heap = []
        
        # سایر تنظیمات
        self.host_last_fetch = {}     # آخرین زمان درخواست به هر میزبان
        self.host_fetch_duration = defaultdict(lambda: 1.0)  # مدت‌زمان تخمینی Fetch
        
    def extract_host(self, url):
        """استخراج نام میزبان از URL (ساده‌شده)"""
        if "://" in url:
            url = url.split("://", 1)[1]
        return url.split("/")[0].lower()
    
    def assign_priority(self, url):
        """
        تخصیص اولویت به URL (۰ = پایین‌ترین، self.num_front-1 = بالاترین)
        در واقعیت، این تابع بر اساس تاریخچه تغییر و کیفیت صفحه تصمیم می‌گیرد.
        """
        host = self.extract_host(url)
        # شبیه‌سازی: دامنه‌های خبری اولویت بالاتری دارند
        if any(keyword in host for keyword in ["news", "bbc", "cnn"]):
            return self.num_front - 1
        # صفحات معمولی اولویت تصادفی دریافت می‌کنند
        return random.randint(0, self.num_front - 2)
    
    def add_url(self, url):
        """افزودن URL جدید به مرز"""
        priority = self.assign_priority(url)
        self.front_queues[priority].append(url)
        print(f"  ➕ افزودن URL به صف جلو (اولویت {priority}): {url}")
    
    def _get_or_create_back_queue(self, host):
        """دریافت یا ایجاد صف عقب برای یک میزبان"""
        if host in self.host_to_back_queue:
            return self.host_to_back_queue[host]
        
        # تخصیص یک صف عقب جدید
        back_queue_id = self.next_back_queue % len(self.back_queues)
        self.next_back_queue += 1
        self.host_to_back_queue[host] = back_queue_id
        
        # اولین زمان مجاز برای این میزبان
        current_time = time.time()
        heapq.heappush(self.time_heap, (current_time, back_queue_id))
        
        return back_queue_id
    
    def _refill_back_queue(self, back_queue_id):
        """پر کردن یک صف عقب خالی از صف‌های جلو (با اولویت بالاتر)"""
        # جستجوی صف‌های جلو از اولویت بالا به پایین
        for priority in range(self.num_front - 1, -1, -1):
            if self.front_queues[priority]:
                url = self.front_queues[priority].popleft()
                host = self.extract_host(url)
                target_back_id = self._get_or_create_back_queue(host)
                
                # URL را به صف عقب صحیح خودش اضافه کن
                self.back_queues[target_back_id].append(url)
                
                # اگر صفی که پر کردیم همان صف درخواستی بود، کار تمام است
                if target_back_id == back_queue_id:
                    return True
                # در غیر این صورت، به جستجو برای پر کردن صف درخواستی ادامه بده
        return False
    
    def get_next_url(self):
        """دریافت URL بعدی برای کرالینگ (با رعایت ادب و اولویت)"""
        # اگر هیچ زمان‌بندی وجود ندارد، سعی کن یک صف را برای شروع پر کنی
        if not self.time_heap:
            # تمام صف‌های عقب را با بالاترین اولویت ممکن پر کن
            for i in range(len(self.back_queues)):
                self._refill_back_queue(i)
            
            # اگر هنوز هیپ خالی است، یعنی هیچ URLای وجود ندارد
            if not self.time_heap:
                return None

        # دریافت زمان مجاز بعدی
        earliest_time, back_queue_id = heapq.heappop(self.time_heap)
        current_time = time.time()
        
        # اگر هنوز زمان مجاز نرسیده، منتظر بمان
        if current_time < earliest_time:
            time.sleep(earliest_time - current_time)
            current_time = earliest_time
        
        # اگر صف عقب خالی است، سعی کن آن را پر کنی
        if not self.back_queues[back_queue_id]:
            if not self._refill_back_queue(back_queue_id):
                # اگر نتوانستیم آن را پر کنیم، یعنی هیچ URLی باقی نمانده
                return None

        # اگر بعد از پر کردن، هنوز هم صف خالی بود، به سراغ زمان بعدی برو
        if not self.back_queues[back_queue_id]:
            return self.get_next_url()

        # دریافت URL از صف عقب
        url = self.back_queues[back_queue_id].popleft()
        host = self.extract_host(url)
        
        # به‌روزرسانی زمان‌ها
        self.host_last_fetch[host] = current_time
        fetch_duration = self.host_fetch_duration[host]
        next_allowed_time = current_time + 10 * fetch_duration  # سیاست Mercator
        
        # زمان مجاز بعدی را به هیپ اضافه کن
        heapq.heappush(self.time_heap, (next_allowed_time, back_queue_id))
        
        return url

# مثال استفاده
if __name__ == "__main__":
    # ایجاد مرز URL با 3 صف اولویت و 5 صف ادب
    frontier = URLFrontierSimulator(num_front_queues=3, num_back_queues=5)
    
    # افزودن URLهای نمونه
    sample_urls = [
        "http://news-example.com/latest",
        "http://blog-site.org/post1",
        "http://cnn-breaking.news/update",
        "http://regular-site.com/page",
        "http://bbc-world.news/today",
        "http://another-blog.net/article"
    ]
    
    print("افزودن URLها به مرز:")
    for url in sample_urls:
        frontier.add_url(url)
    
    print("\nشبیه‌سازی کرالینگ:")
    
    # حالا حلقه اصلی کرالینگ را اجرا می‌کنیم
    for i in range(6):
        url = frontier.get_next_url()
        if url:
            print(f"  📥 کرالینگ: {url}")
            # شبیه‌سازی زمان پردازش
            time.sleep(0.1)
        else:
            print("  ⏹️ هیچ URLی برای کرالینگ باقی نمانده")
            break
    
    print("\n💡 نکته:")
    print("این شبیه‌سازی مکانیزم اصلی مرز URL را نشان می‌دهد:")
    print("- اولویت‌دهی بر اساس میزبان (دامنه‌های خبری اولویت بالاتر)")
    print("- رعایت ادب با صف‌های میزبان‌محور")
    print("- مدیریت زمان‌بندی با هیپ")
    print("در پیاده‌سازی واقعی، اولویت‌ها بر اساس تاریخچه تغییر صفحه محاسبه می‌شوند.")
```

```python
import time
import random
import re
from collections import deque, defaultdict
from urllib.parse import urljoin, urlparse

# --- توابع کمکی برای شبیه‌سازی ---

def simulate_dns_lookup(host):
    """شبیه‌سازی حل DNS با تأخیر تصادفی"""
    print(f"  🔍 [DNS] در حال پیدا کردن آدرس IP برای {host}...")
    time.sleep(random.uniform(0.1, 0.5)) # شبیه‌سازی تأخیر شبکه
    return f"192.168.1.{random.randint(1, 254)}"

def simulate_http_fetch(url):
    """شبیه‌سازی دریافت صفحه با تأخیر تصادفی"""
    print(f"  🌐 [HTTP] در حال دریافت صفحه از {url}...")
    time.sleep(random.uniform(0.5, 1.5)) # شبیه‌سازی تأخیر دریافت
    # یک محتوای HTML ساختگی برای شبیه‌سازی برمی‌گردانیم
    return f"""
    <html>
    <head><title>Page at {url}</title></head>
    <body>
        <h1>Welcome to {urlparse(url).netloc}</h1>
        <p>This is a sample page.</p>
        <a href="{urljoin(url, '/page1.html')}">Link 1</a><br>
        <a href="{urljoin(url, '/page2.html')}">Link 2</a><br>
        <a href="http://external-site.com/about">External Link</a>
    </body>
    </html>
    """

def extract_links(html_content, base_url):
    """استخراج لینک‌ها از محتوای HTML (ساده‌شده)"""
    # پیدا کردن تمام لینک‌ها در تگ <a href="...">
    links = re.findall(r'<a\s+(?:[^>]*?)href="([^"]*)"', html_content, re.IGNORECASE)
    # تبدیل لینک‌های نسبی به مطلق
    absolute_links = [urljoin(base_url, link) for link in links]
    return list(set(absolute_links)) # حذف لینک‌های تکراری

def check_robots_txt(url, robots_cache):
    """بررسی فایل robots.txt برای یک URL"""
    host = urlparse(url).netloc
    if host in robots_cache:
        print(f"  🤖 [Robots] استفاده از کش robots.txt برای {host}")
        return robots_cache[host]
    
    print(f"  🤖 [Robots] در حال دریافت robots.txt برای {host}...")
    time.sleep(0.2) # شبیه‌سازی دریافت فایل
    # شبیه‌سازی یک فایل robots.txt ساده که همه چیزها مجاز هستند
    robots_cache[host] = {'allowed': True}
    return robots_cache[host]

# --- کلاس اصلی کرالر ---

class WebCrawler:
    """
    یک کرالر وب ساده که مفاهیم اصلی مانند مرز URL، ادب،
    حذف تکراری و احترام به robots.txt را پیاده‌سازی می‌کند.
    """
    def __init__(self, seed_urls, max_pages=20, politeness_delay=3):
        """
        مقداردهی اولیه کرالر
        seed_urls: لیست URLهای شروع
        max_pages: حداکثر تعداد صفحات برای دریافت
        politeness_delay: تأخیر به ثانیه بین درخواست‌ها به یک میزبان
        """
        self.max_pages = max_pages
        self.politeness_delay = politeness_delay
        
        # مرز URL: صفی از URLها که باید بازدید شوند
        self.url_frontier = deque(seed_urls)
        
        # مجموعه‌ای از URLهایی که قبلاً دیده‌ایم (برای جلوگیری از تکرار)
        self.visited_urls = set()
        
        # کش برای فایل‌های robots.txt هر میزبان
        self.robots_cache = {}
        
        # صف برای هر میزبان برای رعایت ادب (یک میزبان، یک صف)
        self.host_queues = defaultdict(deque)
        
        # زمان آخرین درخواست به هر میزبان
        self.last_fetch_time = defaultdict(float)
        
        # آمار کرالر
        self.crawled_count = 0
        self.total_links_found = 0

    def _is_polite(self, host):
        """آیا می‌توان به این میزبان درخواست فرستاد؟"""
        last_time = self.last_fetch_time[host]
        return (time.time() - last_time) >= self.politeness_delay

    def _add_to_frontier(self, new_urls, source_url):
        """اضافه کردن لینک‌های جدید به مرز URL"""
        for url in new_urls:
            # 1. حذف تکراری: اگر قبلاً دیدیم یا در مرز است، اضافه نمی‌کنیم
            if url in self.visited_urls or url in self.url_frontier:
                continue
            
            # 2. احترام به robots.txt
            if not check_robots_txt(url, self.robots_cache)['allowed']:
                print(f"  ❌ لینک {url} به دلیل محدودیت robots.txt رد شد.")
                continue
            
            # 3. اضافه به مرز
            self.url_frontier.append(url)
            print(f"  ✅ لینک جدید {url} (از {source_url}) به مرز اضافه شد.")

    def crawl_page(self, url):
        """کرال کردن یک صفحه: دریافت، تحلیل و اضافه کردن لینک‌ها"""
        print(f"\n--- [CRAWL] شروع کرال صفحه: {url} ---")
        
        host = urlparse(url).netloc
        # مرحله 1: حل DNS
        ip_address = simulate_dns_lookup(host)
        
        # مرحله 2: دریافت صفحه
        html_content = simulate_http_fetch(url)
        
        # مرحله 3: تحلیل صفحه و استخراج لینک‌ها
        new_links = extract_links(html_content, url)
        
        # مرحله 4: اضافه کردن لینک‌های جدید به مرز
        self._add_to_frontier(new_links, source_url=url)
        
        # به‌روزرسانی آمار و وضعیت
        self.visited_urls.add(url)
        self.crawled_count += 1
        self.total_links_found += len(new_links)
        self.last_fetch_time[host] = time.time()
        
        print(f"--- [CRAWL] کرال صفحه {url} تمام شد. {len(new_links)} لینک جدید پیدا شد. ---")

    def start_crawling(self):
        """شروع فرآیند اصلی کرالینگ"""
        print("🚀 کرالر وب شروع به کار کرد...")
        
        while self.url_frontier and self.crawled_count < self.max_pages:
            # مرحله 1: انتخاب یک URL از مرز
            current_url = self.url_frontier.popleft()
            
            host = urlparse(current_url).netloc
            
            # مرحله 2: رعایت ادب
            if not self._is_polite(host):
                print(f"  ⏳ منتظر ماندن برای ادب به میزبان {host}...")
                time.sleep(self.politeness_delay)
            
            # مرحله 3: کرال کردن صفحه
            self.crawl_page(current_url)
        
        print("\n✅ کرالینگ به پایان رسید.")
        print(f"📊 آمار نهایی:")
        print(f"   - تعداد کل صفحات کرال شده: {self.crawled_count}")
        print(f"   - تعداد کل لینک‌های کشف شده: {self.total_links_found}")
        print(f"   - تعداد URLهای باقی‌مانده در مرز: {len(self.url_frontier)}")


# --- مثال اجرا ---
if __name__ == "__main__":
    # لیست URLهای شروع (seed set)
    SEED_URLS = [
        "http://example.com/",
        "http://university.edu/departments",
        "http://blog-site.org/index"
    ]
    
    # ساخت و اجرای کرالر
    crawler = WebCrawler(seed_urls=SEED_URLS, max_pages=10, politeness_delay=2)
    crawler.start_crawling()
```

```python
from requests import Session
from requests.adapters import HTTPAdapter
try:
    # urllib3 v2+
    from urllib3.util.retry import Retry
except Exception:
    # fallback for some environments
    from requests.packages.urllib3.util.retry import Retry

from bs4 import BeautifulSoup
from collections import deque
from concurrent.futures import ThreadPoolExecutor, wait
from threading import Lock
from typing import List, Optional, Dict, Any
import json
import time
import random


class IMDbCrawler:
    """
    IMDb crawler (Top 250–seeded) that only extracts 4 fields:
      - title : str
      - genres : List[str]
      - stars : List[str]
      - summaries : List[str]

    Networking:
    - Session with retries/backoff for 429/5xx
    - Global rate limiter shared by threads
    """

    BASE_URL = "https://www.imdb.com"
    TOP_250_URL = f"{BASE_URL}/chart/top/"

    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/122.0.0.0 Safari/537.36"
        )
    }

    def __init__(
        self,
        crawling_threshold: int = 1000,
        *,
        max_workers: int = 3,
        min_interval: float = 1.0,
        timeout: int = 30,
        retries: int = 6
    ) -> None:
        self.crawling_threshold = crawling_threshold
        self.not_crawled: deque[str] = deque()
        self.crawled: List[Dict[str, Any]] = []
        self.added_ids: set[str] = set()

        self.add_list_lock = Lock()
        self.add_queue_lock = Lock()

        self.max_workers = max_workers
        self.min_interval = float(min_interval)
        self.timeout = timeout
        self.session = self._build_session(retries)

        self._last_request_ts = 0.0
        self._rate_lock = Lock()

    # --------------------------- Networking -----------------------------

    def _build_session(self, retries: int) -> Session:
        s = Session()
        try:
            retry = Retry(
                total=retries,
                connect=retries,
                read=retries,
                status=retries,
                backoff_factor=1.0,
                status_forcelist=[429, 500, 502, 503, 504],
                allowed_methods=frozenset({"GET"}),
                raise_on_status=False,
            )
        except TypeError:
            retry = Retry(
                total=retries,
                connect=retries,
                read=retries,
                status=retries,
                backoff_factor=1.0,
                status_forcelist=[429, 500, 502, 503, 504],
                method_whitelist=frozenset({"GET"}),
                raise_on_status=False,
            )
        adapter = HTTPAdapter(max_retries=retry, pool_connections=100, pool_maxsize=100)
        s.mount("https://", adapter)
        s.mount("http://", adapter)
        s.headers.update({
            "User-Agent": self.headers.get("User-Agent", ""),
            "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
            "Accept-Language": "en-US,en;q=0.9",
            "Accept-Encoding": "gzip, deflate, br",
            "Connection": "keep-alive",
            "Referer": "https://www.imdb.com/",
        })
        return s

    def crawl(self, url: str):
        """GET with retries/timeout + GLOBAL rate limit."""
        with self._rate_lock:
            now = time.time()
            wait_time = self.min_interval - (now - self._last_request_ts)
            if wait_time > 0:
                time.sleep(wait_time)
            time.sleep(random.uniform(0.05, 0.25))  # jitter
            self._last_request_ts = time.time()

        try:
            resp = self.session.get(url, timeout=self.timeout)
            if resp.status_code == 200:
                return resp
            print(f"Failed to crawl {url}: Status code {resp.status_code}")
        except Exception as e:
            print(f"An error occurred while crawling {url}: {e}")
        return None

    # ----------------------------- Seeds --------------------------------

    def extract_top_250(self) -> None:
        """Seed the queue from IMDb Top 250 page."""
        try:
            response = self.crawl(self.TOP_250_URL)
            if response is None:
                print("Failed to retrieve Top 250 movies list.")
                return

            soup = BeautifulSoup(response.content, "html.parser")
            movies = soup.find_all("li", class_="ipc-metadata-list-summary-item")

            self.not_crawled.clear()
            self.added_ids.clear()

            for movie in movies:
                link = movie.find("a", href=True)
                if not link:
                    continue
                href = link.get("href", "")
                parts = href.split("/")
                if len(parts) > 2:
                    movie_id = parts[2]
                    if movie_id and movie_id not in self.added_ids:
                        self.not_crawled.append(f"{self.BASE_URL}/title/{movie_id}/")
                        self.added_ids.add(movie_id)
        except Exception as e:
            print(f"An error occurred while extracting Top 250 movies: {e}")

    # ----------------------------- Public --------------------------------

    def start_crawling(self) -> None:
        """Crawl titles from Top 250 (and related) until threshold/queue empty."""
        self.extract_top_250()

        futures = []
        crawled_counter = 0

        with ThreadPoolExecutor(max_workers=self.max_workers) as executor:
            while crawled_counter < self.crawling_threshold and self.not_crawled:
                url = self.not_crawled.popleft()
                futures.append(executor.submit(self.crawl_page_info, url, crawled_counter))
                crawled_counter += 1

                if not self.not_crawled:
                    wait(futures)
                    futures.clear()

            if futures:
                wait(futures)

    def crawl_page_info(self, url: str, crawler_counter: int) -> Optional[Dict[str, Any]]:
        """
        Crawl a single title page and extract ONLY:
          - title
          - genres
          - stars
          - summaries
        Related links may be appended to the queue.
        """
        response = self.crawl(url)
        if response is None:
            print(f"Failed to crawl page: {url}")
            return None

        soup = BeautifulSoup(response.content, "html.parser")

        movie = self.get_imdb_instance()
        movie["id"] = self.get_id_from_URL(url)
        movie["title"] = self.get_title(soup)
        movie["genres"] = self.get_genres(soup)
        movie["stars"] = self.get_stars(soup)

        # Prefer full plot summaries page; fallback to first page short plot if empty
        summaries = self.get_summaries(url)
        if not summaries:
            short = self.get_first_page_summary(soup)
            if short and short != "Summary not found.":
                summaries = [short]
        movie["summaries"] = summaries

        with self.add_list_lock:
            self.crawled.append(movie)

        # Optional queue expansion via "More like this"
        related_links = self.get_related_links(soup)
        if related_links:
            with self.add_queue_lock:
                for link in related_links:
                    movie_id = self.get_id_from_URL(link)
                    if movie_id and movie_id not in self.added_ids:
                        self.not_crawled.append(link)
                        self.added_ids.add(movie_id)

        print(crawler_counter, f"Successfully crawled and processed: {url}")
        return movie

    # --------------------------- Serialization ---------------------------

    def save_to_json(self) -> None:
        """Write crawled results to a JSON file."""
        try:
            with open("IMDB_crawled.json", "w", encoding="utf-8") as f:
                json.dump(self.crawled, f, ensure_ascii=False, indent=4)
            print(f"✅ اطلاعات {len(self.crawled)} فیلم با موفقیت در فایل IMDB_crawled.json ذخیره شد.")
        except Exception as e:
            print(f"An error occurred while writing to JSON file: {e}")

    # ------------------------- Extraction helpers ------------------------

    @staticmethod
    def get_imdb_instance() -> Dict[str, Any]:
        return {
            "id": None,
            "title": None,
            "genres": None,
            "stars": None,
            "summaries": None,
        }

    @staticmethod
    def get_id_from_URL(url: str) -> Optional[str]:
        """https://www.imdb.com/title/tt0111161/ -> tt0111161"""
        try:
            parts = url.split("/")
            title_index = parts.index("title")
            if title_index + 1 < len(parts):
                return parts[title_index + 1]
            print("Error: 'title' segment found, but no ID follows in the URL.")
        except ValueError:
            print("Error: The URL does not contain a 'title' segment.")
        except Exception as e:
            print(f"An unexpected error occurred: {e}")
        return None

    def get_summary_link(self, url: str) -> Optional[str]:
        """ /title/<id>/ -> /title/<id>/plotsummary """
        try:
            movie_id = self.get_id_from_URL(url)
            if movie_id:
                return f"{self.BASE_URL}/title/{movie_id}/plotsummary"
            print("Failed to extract movie ID from URL.")
        except Exception as e:
            print(f"Failed to get summary link: {e}")
        return None

    def get_related_links(self, soup: BeautifulSoup) -> List[str]:
        """Collect 'More like this' title links to expand crawling (optional)."""
        try:
            related_links: List[str] = []
            section = soup.find("section", {"data-testid": "MoreLikeThis"})
            if not section:
                return related_links
            links = section.find_all("a", class_="ipc-lockup-overlay ipc-focusable")
            for link in links:
                href = link.get("href")
                if href:
                    related_links.append(f"https://www.imdb.com{href}")
            return related_links
        except Exception as e:
            print(f"Failed to get related links: {e}")
        return []

    @staticmethod
    def get_title(soup: BeautifulSoup) -> Optional[str]:
        try:
            tag = soup.find("h1")
            return tag.text.strip() if tag else None
        except Exception as e:
            print(f"Failed to get title: {e}")
            return None

    @staticmethod
    def get_first_page_summary(soup: BeautifulSoup) -> str:
        try:
            tag = soup.find("span", {"data-testid": "plot-xl"})
            return tag.text.strip() if tag else "Summary not found."
        except Exception as e:
            print(f"Failed to get summary: {e}")
            return "Summary not found."

    @staticmethod
    def _get_principal_names(soup: BeautifulSoup, label_text: str) -> List[str]:
        """Extract names (e.g., Stars) from principal credit blocks."""
        try:
            names: List[str] = []
            blocks = soup.find_all("li", {"data-testid": "title-pc-principal-credit"})
            for block in blocks:
                for tag_name, cls in [
                    ("span", "ipc-metadata-list-item__label"),
                    ("a", "ipc-metadata-list-item__label"),
                ]:
                    label_tag = block.find(tag_name, class_=cls)
                    if label_tag and label_tag.text.strip() == label_text:
                        for a in block.find_all("a", class_="ipc-metadata-list-item__list-content-item"):
                            txt = a.text.strip()
                            if txt:
                                names.append(txt)
            return list(set(names)) if names else []
        except Exception as e:
            print(f"Failed to get {label_text.lower()}: {e}")
            return []

    def get_stars(self, soup: BeautifulSoup) -> List[str]:
        return self._get_principal_names(soup, "Stars")

    @staticmethod
    def get_genres(soup: BeautifulSoup) -> List[str]:
        genres: List[str] = []
        try:
            scroller = soup.find("div", class_="ipc-chip-list__scroller")
            if not scroller:
                return genres
            for span in scroller.find_all("span", class_="ipc-chip__text"):
                txt = span.text.strip()
                if txt:
                    genres.append(txt)
        except Exception as e:
            print(f"Failed to get genres: {e}")
        return genres

    def get_summaries(self, url: str) -> List[str]:
        summaries: List[str] = []
        try:
            summaries_url = self.get_summary_link(url)
            if not summaries_url:
                return summaries
            resp = self.crawl(summaries_url)
            if not resp:
                return summaries
            soup = BeautifulSoup(resp.content, "html.parser")
            section = soup.find("div", {"data-testid": "sub-section-summaries"})
            if not section:
                return summaries
            for li in section.find_all("li"):
                div = li.find("div", class_="ipc-html-content-inner-div")
                if div and div.text:
                    summaries.append(div.text.strip())
        except Exception as e:
            print(f"Failed to get summaries: {e}")
        return summaries


def main() -> None:
    imdb_crawler = IMDbCrawler(
        crawling_threshold=10,  # فقط ۱۰ فیلم صرفا جهت تمرین 
        max_workers=3,
        min_interval=1.0,
        timeout=30,
        retries=6,
    )
    imdb_crawler.start_crawling()
    
    # ذخیره اطلاعات در فایل پس از اتمام کرالینگ
    imdb_crawler.save_to_json()


if __name__ == "__main__":
    main()
```

<li>هنگام جستجو، ابتدا فقط ایندکس‌های اولیه جستجو می‌شوند. اگر نتایج کافی نبود، سیستم به جستجو در ایندکس‌های ثانویه می‌پردازد.</li>

<span id="section20.4" style="font-size:24pt; font-weight:bold; color:#1B263B;">20.4 سرورهای همبستگی: نقشه راهنمای وب</span> 🔗

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🗺️ چرا گوگل به نقشه وب نیاز دارد؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم مهندسی گوگل کار می‌کنید و می‌خواهید الگوریتم PageRank را بهبود دهید. برای این کار، باید بدانید کدام وب‌سایت‌های معتبر به صفحه اصلی دانشگاه استنفورد لینک داده‌اند. یا شاید بخواهید بدانید کدام صفحات از سایت <code>bbc.com</code> به یک خبر خاص اشاره دارند.</p>
<p style="font-size:16pt; text-align:right;">
این پرس‌وجوها که به آنها <strong>پرس‌وجوهای همبستگی</strong> (Connectivity Queries) می‌گویند، برای موتورهای جستجو حیاتی هستند. اما پاسخ به آنها چالش‌برانگیز است، زیرا نیاز به ذخیره‌سازی کل گراف وب (که میلیاردها صفحه و تریلیون‌ها لینک دارد) در حافظه دارد.</p>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">💾 چالش ذخیره‌سازی: وقتی وب بزرگتر از حافظه است</p>
<p style="font-size:16pt; text-align:right;">
بیایید ابعاد این چالش را درک کنیم. اگر وب شامل ۴ میلیارد صفحه باشد و هر صفحه به طور متوسط ۱۰ لینک داشته باشد، برای ذخیره‌سازی هر لینک (با شناسه مبدأ و مقصد ۳۲ بیتی)، به <strong>۳۲۰ گیگابایت</strong> حافظه نیاز داریم! این مانند تلاش برای بارگذاری کل کتابخانه کنگره در RAM یک لپ‌تاپ است.</p>
<p style="font-size:16pt; text-align:right;">
راه‌حل، فشرده‌سازی هوشمندانه‌ای است که هم فضا را کاهش دهد و هم <strong>پرس‌وجوهای سریع</strong> را پشتیبانی کند. این تفاوت اصلی با فشرده‌سازی معمولی است؛ ما نه تنها داده‌ها را فشرده می‌کنیم، بلکه باید بتوانیم به سرعت به آنها دسترسی پیدا کنیم.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 سه تکنیک کلیدی برای فشرده‌سازی گراف وب</p>
<p style="font-size:16pt; text-align:right;">
مهندسان شرکت‌هایی مانند گوگل و آمازون از ترکیبی از سه تکنیک هوشمندانه برای حل این مشکل استفاده می‌کنند:</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>فریب هوشمندانه: کدگذاری فاصله‌ها (Gap Encoding)</strong><br>
به جای ذخیره شماره مطلق هر صفحه مقصد، <strong>تفاضل آن با صفحه قبلی</strong> را ذخیره می‌کنیم. چون لیست‌ها مرتب هستند، این فاصله‌ها (گَپ‌ها) معمولاً اعداد کوچکی هستند و فضای بسیار کمتری اشغال می‌کنند. این تکنیکی است که در فشرده‌سازی ایندکس‌های معکوس نیز استفاده می‌شود.</li>

<li><strong>پیدا کردن الگوها: شباهت بین ردیف‌ها</strong><br>
بسیاری از وب‌سایت‌ها از الگوی یکسانی برای ساختار خود استفاده می‌کنند. برای مثال، تقریباً تمام صفحات دانشگاه استنفورد لینک‌هایی به صفحه اصلی، نقشه سایت، و صفحه کپی‌رایت دارند. به جای ذخیره این لیست برای هر صفحه، می‌توانیم یک <strong>ردیف نمونه</strong> (prototype) ایجاد کرده و تفاوت‌های هر صفحه را با آن نمونه ذخیره کنیم.</li>

<li><strong>هم‌جواری: لینک‌های محلی</strong><br>
بسیاری از لینک‌ها به صفحات <strong>هم‌میزبان</strong> اشاره می‌کنند. در لیست مرتب‌شده URLها، این یعنی شماره‌های مقصد نزدیک به هم هستند و گَپ‌های کوچکتری تولید می‌شود، که فشرده‌سازی را مؤثرتر می‌کند.</li>
</ol>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🏗️ معماری عملی: از تئوری تا پیاده‌سازی</p>
<p style="font-size:16pt; text-align:right;">
این تکنیک‌ها چگونه در عمل ترکیب می‌شوند؟</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مرتب‌سازی و شماره‌گذاری:</strong> ابتدا تمام URLها را به‌صورت لکسیکوگرافیک مرتب کرده و به هر کدام یک شماره منحصربه‌فرد (ID عددی) می‌دهیم. این کار پایه و اساس سیستم ماست.</li>
<li><strong>ایجاد جدول مجاورت:</strong> جدولی ایجاد می‌کنیم که هر ردیف آن شامل لیست مرتب‌شده‌ای از ID صفحاتی است که به آن صفحه لینک داده‌اند (in-links) یا از آن لینک گرفته‌اند (out-links).</li>
<li><strong>کدگذاری هوشمند:</strong> از بالا به پایین جدول را پیمایش می‌کنیم و هر ردیف را بر اساس ۷ ردیف قبلی فشرده می‌کنیم. چرا ۷؟ چون صفحات یک وب‌سایت در این مرتب‌سازی کنار هم قرار می‌گیرند و ۷ ردیف قبلی به احتمال زیاد به همان سایت تعلق دارند. این محدودیت باعث می‌شود جستجو برای ردیف نمونه سریع باشد (فقط ۳ بیت برای ذخیره Offset نیاز است).</li>
<li><strong>بازسازی در زمان پرس‌وجو:</strong> هنگام پاسخ به یک پرس‌وجو، ابتدا ID صفحه مورد نظر را پیدا کرده، سپس ردیف فشرده‌شده آن را بازسازی می‌کنیم. اگر ردیف به یک ردیف نمونه اشاره کند، آن ردیف نمونه نیز بازسازی می‌شود.</li>
</ol>
<p style="font-size:16pt; text-align:right;">
این تکنیک‌ها به‌طور ترکیبی، مصرف حافظه را از <strong>۶۴ بیت به ازای هر لینک</strong> در روش ساده، به تنها <strong>۳ بیت به ازای هر لینک</strong> کاهش می‌دهند - یک پیشرفت چشمگیر!</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">⚖️ تعادل بین فشرده‌سازی و سرعت</p>
<p style="font-size:16pt; text-align:right;">
یک چالش مهم در این سیستم، پیدا کردن تعادل مناسب است. اگر آستانه شباهت برای انتخاب ردیف نمونه خیلی بالا باشد، از این تکنیک به ندرت استفاده می‌شود و فشرده‌سازی ضعیف می‌شود. اگر آستانه خیلی پایین باشد، ردیف‌های زیادی به ردیف‌های نمونه اشاره می‌کنند و بازسازی یک ردیف ساده ممکن است به چندین سطح از ارجاعات نیاز داشته باشد که سرعت را کاهش می‌دهد.</p>
<p style="font-size:16pt; text-align:right;">
این تعادل دقیق یکی از رازهای موفقیت موتورهای جستجوی مدرن است که به آنها اجازه می‌دهد کل وب را در حافظه مدیریت کرده و به پرس‌وجوهای پیچیده در کسری از ثانیه پاسخ دهند.</p>
</div>

</div>

```python
# شبیه‌سازی فشرده‌سازی گراف وب

import random
from typing import List, Tuple, Optional, Dict

class WebGraphCompressor:
    """
    شبیه‌سازی فشرده‌سازی گراف وب با سه تکنیک:
    1. کدگذاری گَپ (Gap Encoding)
    2. کدگذاری نسبت به ردیف نمونه (Prototype-based Encoding)
    """
    
    def __init__(self, similarity_threshold=0.6):
        self.similarity_threshold = similarity_threshold
        self.compressed_rows = []  # لیست ردیف‌های فشرده‌شده
        self.url_to_id = {}        # نگاشت URL به ID عددی
        self.id_to_url = []        # نگاشت ID به URL
        self.next_id = 0
    
    def register_url(self, url: str) -> int:
        """ثبت یک URL و بازگرداندن ID عددی منحصربه‌فرد آن"""
        if url not in self.url_to_id:
            self.url_to_id[url] = self.next_id
            self.id_to_url.append(url)
            self.next_id += 1
        return self.url_to_id[url]
    
    def encode_gap(self, sorted_ids: List[int]) -> List[int]:
        """کدگذاری گَپ: تبدیل لیست شماره‌های مرتب به تفاضل‌های متوالی"""
        if not sorted_ids:
            return []
        gaps = [sorted_ids[0]]
        for i in range(1, len(sorted_ids)):
            gaps.append(sorted_ids[i] - sorted_ids[i-1])
        return gaps
    
    def decode_gap(self, gaps: List[int]) -> List[int]:
        """بازگردانی لیست اصلی از روی گَپ‌ها"""
        if not gaps:
            return []
        ids = [gaps[0]]
        for i in range(1, len(gaps)):
            ids.append(ids[-1] + gaps[i])
        return ids
    
    def calculate_similarity(self, row1: List[int], row2: List[int]) -> float:
        """محاسبه شباهت بین دو ردیف بر اساس اشتراک مجموعه‌ای"""
        if not row1 or not row2:
            return 1.0
        set1, set2 = set(row1), set(row2)
        intersection = len(set1 & set2)
        union = len(set1 | set2)
        return intersection / union if union > 0 else 0.0
    
    def find_best_prototype(self, current_row: List[int], window_size: int = 7) -> Optional[Tuple[int, Dict[str, List[int]]]]:
        """
        یافتن بهترین ردیف نمونه از بین 'window_size' ردیف آخر.
        بازگشت: (offset, differences) یا None اگر شباهت کافی وجود نداشته باشد.
        """
        if not self.compressed_rows:
            return None
        
        num_rows = len(self.compressed_rows)
        start_idx = max(0, num_rows - window_size)

        best_similarity = 0.0
        best_offset = 0
        best_diff = {"remove": [], "add": []}
        
        for i in range(num_rows - 1, start_idx - 1, -1):
            # بازسازی ردیف نمونه (همیشه ذخیره‌شده به صورت گَپ)
            prototype_gaps = self.compressed_rows[i]["data"]
            prototype_row = self.decode_gap(prototype_gaps)
            
            similarity = self.calculate_similarity(current_row, prototype_row)
            
            if similarity > best_similarity:
                best_similarity = similarity
                best_offset = num_rows - i  # فاصله از ردیف فعلی
                
                # محاسبه تفاوت‌ها: عناصری که باید حذف/افزوده شوند
                remove = list(set(prototype_row) - set(current_row))
                add = list(set(current_row) - set(prototype_row))
                best_diff = {"remove": sorted(remove), "add": sorted(add)}
        
        if best_similarity >= self.similarity_threshold:
            return (best_offset, best_diff)
        return None
    
    def add_adjacency_row(self, out_links: List[str]) -> None:
        """
        افزودن یک ردیف مجاورت (لیست لینک‌های خروجی یک صفحه) به گراف فشرده‌شده.
        """
        # تبدیل URLها به IDهای عددی
        link_ids = [self.register_url(url) for url in out_links]
        link_ids.sort()  # لیست باید مرتب باشد
        
        # تلاش برای یافتن یک ردیف نمونه مناسب
        prototype_info = self.find_best_prototype(link_ids)
        
        # اگر ردیف نمونه مناسبی پیدا شد، آن را فشرده می‌کنیم
        if prototype_info is not None:
            offset, diff = prototype_info
            self.compressed_rows.append({
                "type": "prototype",
                "offset": offset,
                "diff": diff
            })
        else:
            # اگر ردیف نمونه مناسبی نبود، از کدگذاری گَپ استفاده می‌کنیم
            gaps = self.encode_gap(link_ids)
            self.compressed_rows.append({
                "type": "gap",
                "data": gaps
            })
    
    def reconstruct_row(self, row_index: int) -> List[int]:
        """
        بازسازی لیست لینک‌های یک ردیف از فرم فشرده‌شده.
        """
        if row_index >= len(self.compressed_rows) or row_index < 0:
            return []
        
        row = self.compressed_rows[row_index]
        if row["type"] == "gap":
            return self.decode_gap(row["data"])
        elif row["type"] == "prototype":
            # بازسازی ردیف نمونه اول
            prototype_index = row_index - row["offset"]
            prototype_row = self.reconstruct_row(prototype_index)
            
            # اعمال تغییرات
            result_set = set(prototype_row)
            for item in row["diff"]["remove"]:
                result_set.discard(item)
            for item in row["diff"]["add"]:
                result_set.add(item)
            return sorted(list(result_set))
        return []
    
    def get_original_links(self, row_index: int) -> List[str]:
        """بازگرداندن لینک‌های اصلی به صورت URL برای یک ردیف"""
        ids = self.reconstruct_row(row_index)
        return [self.id_to_url[i] for i in ids if i < len(self.id_to_url)]

# --- شبیه‌سازی سرور همبستگی ---
class ConnectivityServer:
    """
    شبیه‌سازی یک سرور همبستگی که پرس‌وجوهایی را مدیریت می‌کند.
    """
    def __init__(self, compressor: WebGraphCompressor):
        self.compressor = compressor
        self.inverted_index = {}  # نگاشت لینک -> لیست اسناد
    
    def add_document(self, doc_id: int, links: List[str]):
        """افزودن یک سند جدید به ایندکس معکوس"""
        for link in links:
            # --- اصلاح: استفاده از متدهای دیکشنری ---
            if link not in self.inverted_index:
                self.inverted_index[link] = set()
            self.inverted_index[link].add(doc_id)
    
    def handle_query(self, query_links: List[str]) -> List[int]:
        """
        پاسخ به یک پرس‌وجو: لیست اسنادهایی که شامل تمام لینک‌های پرس‌وجو هستند.
        """
        result_sets = [set(self.inverted_index.get(link, set())) for link in query_links]
        
        if not result_sets:
            return []
        
        # پیدا کردن اشتراک بین تمام مجموعه‌ها
        common_docs = result_sets[0]
        for i in range(1, len(result_sets)):
            common_docs.intersection_update(result_sets[i])
        
        # تبدیل مجموعه اسناد به لیست مرتب
        return sorted(list(common_docs))

# --- سناریوی اصلی ---
def main():
    """
    سناریوی اصلی: ایجاد گراف، فشرده‌سازی، و پاسخ به پرس‌وجوها.
    """
    # ۱. تعریف صفحات و لینک‌ها (مطابق شکل ۲۰.۵)
    pages = {
        1: {"id": 1, "links": [2, 3, 4, 8, 16, 32, 64]},
        2: {"id": 2, "links": [1, 3, 5, 6, 7, 8, 9]},
        3: {"id": 3, "links": [1, 4, 5, 6, 7, 8]},
        4: {"id": 4, "links": [1, 2, 3, 4, 5, 6]},
        5: {"id": 5, "links": [1, 2, 3, 4, 5]},
        6: {"id": 6, "links": [1, 2, 3, 4]}
    }
    
    # ۲. ایجاد فشرده‌ساز و سرور
    compressor = WebGraphCompressor()
    server = ConnectivityServer(compressor)
    
    # ۳. افزودن اسناد به سرور
    for doc_id, page_data in pages.items():
        server.add_document(doc_id, page_data["links"])
    
    # ۴. سناریوی اصلی: پاسخ به پرس‌وجوها
    queries = [
        [2, 4],      # پرس‌وجو: صفحاتی که به صفحه ۲ لینک دارند
        [1, 5, 6],   # پرس‌وجو: صفحاتی که به صفحات ۱، ۳ و ۵ لینک دارند
        [3, 4, 5],   # پرس‌وجو: صفحاتی که به صفحات ۲، ۳ و ۴ لینک دارند
    ]
    
    print("🔗 شبیه‌سازی سرور همبستگی:")
    print("-" * 40)
    
    for i, query in enumerate(queries):
        print(f"\nپرس‌وجو {i+1}: {query} -> {server.handle_query(query)}")
        result = server.handle_query(query)
        print(f"پاسخ: {result}")

if __name__ == "__main__":
    main()

print("نکته مهم: تکنیک فشرده‌سازی گراف، هرچقدر قدرتمند، برای کاهش حجم داده‌ها عالی است، اما در عملکرد جستجو (Query Processing) بسیار ناکارآمد است و در مقیاس‌های بزرگ کارایی ندارد.")
```

<p style="font-size:16pt; text-align:right;">
در این فصل:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فرآیند <strong> خزنده‌برداری وب یا به اختصار خزش وب </strong> (Web crawling) را به‌عنوان گامی حیاتی برای جمع‌آوری صفحات وب و ساخت ایندکس معرفی کردیم.</li>
<li>ویژگی‌های ضروری خزنده‌ها را بررسی کردیم:  
  – <strong>ربات‌گیر</strong> (Robustness): مقاومت در برابر دام‌های خزنده (spider traps)،  
  – <strong>ادب</strong> (Politeness): رعایت محدودیت‌های سرورها در سرعت درخواست.</li>
<li>معماری یک خزندهٔ واقعی را تحلیل کردیم:  
  – <strong>مرز URL</strong> (URL frontier) برای مدیریت صف درخواست‌ها،  
  – ماژول‌های <strong>DNS</strong>، <strong>دریافت</strong> (fetch)، <strong>تجزیه</strong> (parse) و <strong>حذف تکرار</strong> (duplicate elimination).</li>
<li>چالش‌های خزنده‌های توزیع‌شده را پوشش دادیم:  
  – پارتیشن‌بندی میزبان‌ها بین گره‌ها،  
  – مدیریت سراسری مجموعهٔ اثرانگشت‌ها برای شناسایی صفحات تکراری.</li>
<li>راهکارهای هوشمند برای <strong>مرز URL</strong> را معرفی کردیم:  
  – صف‌های جلو برای <strong>اولویت‌بندی</strong> (کیفیت و تازگی)،  
  – صف‌های عقب برای رعایت <strong>ادب</strong> (فاصلهٔ زمانی بین درخواست‌ها به یک میزبان).</li>
<li>دو روش اصلی <strong>توزیع ایندکس</strong> را مقایسه کردیم:  
  – پارتیشن‌بندی بر اساس <strong>اصطلاحات</strong> (term partitioning) — چالش‌برانگیز برای پرسمان‌های چندکلمه‌ای،  
  – پارتیشن‌بندی بر اساس <strong>اسناد</strong> (document partitioning) — رایج‌تر و ساده‌تر برای پیاده‌سازی.</li>
<li>نیاز به <strong>سرورهای اتصال</strong> (Connectivity servers) را تبیین کردیم: ذخیرهٔ ساختار گراف وب برای پشتیبانی از تحلیل پیوندها (فصل ۲۱) با استفاده از فشرده‌سازی هوشمند لیست‌های مجاورت.</li>
</ul>
