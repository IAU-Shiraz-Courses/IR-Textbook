<p style="font-size:18pt; text-align:center;">
<span id="section21.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.1 وب به‌عنوان یک گراف: قدرت نادیده‌گرفته لینک‌ها</span> 🔗
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 از ارجاعات علمی تا لینک‌های وب</p>
<p style="font-size:16pt; text-align:right;">
در دنیای آکادمیک، اهمیت یک مقاله علمی با تعداد دفعاتی که توسط مقالات دیگر به آن ارجاع داده می‌شود، سنجیده می‌شود. این ایده که «یک ارجاع، نوعی انتقال اعتبار است»، پایه و اساس یک حوزه به نام <strong>Bibliometrics</strong> است.</p>
<p style="font-size:16pt; text-align:right;">
موتورهای جستجوی وب، به‌ویژه گوگل، این ایده را به دنیای وب آوردند. آنها فرض کردند که یک <strong>لینک</strong> از صفحه A به صفحه B، مانند یک ارجاع علمی، نوعی <strong>تأیید اعتبار</strong> برای صفحه B توسط سازنده صفحه A است. این ایده ساده اما قدرتمند، اساس الگوریتم‌های رتبه‌بندی مدرن را شکل داد.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📝 قدرت متن لنگر: وقتی وب‌سایت شما خودش را نمی‌شناسد</p>
<p style="font-size:16pt; text-align:right;">
اولین شهودی که از تحلیل لینک به دست می‌آید، قدرت <strong>متن لنگر</strong> است - متنی که در یک لینک قابل کلیک قرار می‌گیرد. اغلب، این متن توصیف دقیق‌تری از صفحه مقصد ارائه می‌دهد تا خود صفحه.</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مثال واقعی را بررسی کنیم: صفحه اصلی شرکت Airbnb در گذشته، کلمه «اجاره آپارتمان» را در متن خود نداشت! این صفحه بیشتر بر روی «تجربه منحصربه‌فرد» و «میزبان‌های محلی» تمرکز داشت. اما وبلاگ‌های مسافرتی و رسانه‌ها به این صفحه با متن‌های لنگری مانند «اجاره خانه برای سفر» یا «پلتفرم اجاره اقامتگاه» لینک داده‌اند.</p>
<p style="font-size:16pt; text-align:right;">
نتیجه؟ وقتی کاربران عبارت «اجاره آپارتمان» را جستجو می‌کردند، صفحه Airbnb در نتایج بالا ظاهر می‌شد - نه به خاطر محتوای صفحه، بلکه به خاطر توصیفی که جامعه وب از آن ارائه کرده بود. این نشان می‌دهد که چگونه <strong>جامعه نویسندگان وب</strong> به‌طور ناخواسته نقش یک سیستم ایندکس‌دهی جمعی را ایفا می‌کنند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🏷️ لینک به‌عنوان رأی اعتماد: چگونه گوگل «Big Blue» را کشف کرد</p>
<p style="font-size:16pt; text-align:right;">
وقتی در گوگل عبارت «Big Blue» را جستجو می‌کنید، اولین نتیجه صفحه اصلی IBM است. چرا؟ چون هزاران وب‌سایت و وبلاگ با این متن لنگر به صفحه IBM لینک داده‌اند. گوگل از طریق تحلیل میلیون‌ها لینک، فهمید که «Big Blue» یک نام مستعار برای IBM است.</p>
<p style="font-size:16pt; text-align:right;">
این مثال نشان می‌دهد که چگونه لینک‌ها می‌توانند ارتباطات معنایی را که در متن صفحات وجود ندارد، آشکار کنند. موتورهای جستجو این متن‌های لنگر را با وزن‌دهی هوشمند (مشابه مفهوم IDF در ایندکس‌گذاری) به واژگان ایندکس صفحه مقصد اضافه می‌کنند و معمولاً به آنها اهمیت بالاتری نسبت به متن خود صفحه می‌دهند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚠️ محدودیت‌ها: چه زمانی یک لینک، رأی اعتماد نیست؟</p>
<p style="font-size:16pt; text-align:right;">
همه لینک‌ها یکسان نیستند. موتورهای جستجو به سرعت فهمیدند که برخی لینک‌ها واقعاً رأی اعتماد نیستند. برای مثال:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>تقریباً تمام صفحات یک وب‌سایت دانشگاهی لینکی به صفحه «قوانین کپی‌رایت» یا «اطلاعات تماس با ما» دارند.</li>
<li>آیا این یعنی صفحه «قوانین کپی‌رایت» مهم‌ترین صفحه آن وب‌سایت است؟ قطعاً خیر.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این نوع لینک‌ها که از یک <strong>قالب مشترک</strong> در تمام صفحات یک سایت ناشی می‌شوند، معمولاً توسط الگوریتم‌های تحلیل لینک نادیده گرفته یا وزن بسیار کمی به آنها اختصاص داده می‌شود، زیرا آنها نمایانگر یک تأیید واقعی و آگاهانه از محتوای صفحه مقصد نیستند.</p>
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🎯 اسپم متن لنگر: وقتی توصیف‌ها به دروغ تبدیل می‌شوند</p>
<p style="font-size:16pt; text-align:right;">
قدرت متن لنگر همچنین زمینه‌ی سوءاستفاده را فراهم می‌کند. در گذشته، رقبا به صورت برنامه‌ریزی‌شده به صفحات یک شرکت با متن‌های لنگر منفی مانند «شرکت کلاهبرداری» لینک می‌دادند تا رتبه آن را در نتایج جستجو کاهش دهند.</p>
<p style="font-size:16pt; text-align:right;">
این پدیده که به آن <strong>اسپم متن لنگر</strong> (Anchor Text Spam) گفته می‌شود، یکی از چالش‌های مهم موتورهای جستجو است. برای مقابله با آن، الگوریتم‌های پیچیده‌ای توسعه داده شده‌اند که الگوهای غیرطبیعی در توزیع متن‌های لنگر را شناسایی و وزن آنها را کاهش می‌دهند یا به طور کامل نادیده می‌گیرند.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section21.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.2 PageRank: چگونه گوگل اهمیت را در وب اندازه‌گیری می‌کند</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧭 یک جهانگرد در دنیای وب: ایده‌ی پشت PageRank</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید یک جهانگرد دیجیتال در حال گشت‌وگذار در اینترنت است. او از صفحه‌ای شروع کرده و در هر گام، یکی از لینک‌های موجود در صفحه را به‌صورت تصادفی انتخاب می‌کند و به آن صفحه می‌رود.</p>
<p style="font-size:16pt; text-align:right;">
با این منطق، صفحاتی که جهانگرد بیشتر به آنها سر می‌زند، باید مهم‌تر باشند. چرا؟ چون احتمالاً صفحات مهم‌تر، لینک‌های بیشتری از صفحات دیگر دریافت کرده‌اند و جهانگرد را به سمت خود جذب می‌کنند. این ایده‌ی ساده اما ناب، اساس الگوریتم PageRank است.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🚫 دو چالش بزرگ: بن‌بست‌ها و جزیره‌های جداافتاده</p>
<p style="font-size:16pt; text-align:right;">
اما جهانگرد ما با دو مشکل اساسی روبرو است:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>بن‌بست‌ها (Sink Nodes):</strong> صفحه‌ای بدون هیچ لینک خروجی. جهانگرد به این صفحه می‌رسد و در آن گیر می‌کند.</li>
<li><strong>جزیره‌های جداافتاده:</strong> ممکن است جهانگرد در یک بخش کوچک از وب (مثلاً مجموعه‌ای از سایت‌های یک دانشگاه خاص) گیر کند و هرگز به بخش‌های دیگر وب نرسد.</li>
</ol>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">✨ راه‌حل جادویی: عملیات تله‌پورت (Teleport)</p>
<p style="font-size:16pt; text-align:right;">
برای حل این مشکلات، به جهانگرد ما یک توانایی جادویی می‌دهیم: <strong>تله‌پورت</strong>. در هر گام، جهانگرد با احتمال کوچکی (معمولاً ۰.۱ یا ۱۰٪) از صفحه فعلی «ناپدید» شده و به‌صورت کاملاً تصادفی به هر صفحه‌ای در وب می‌رود.</p>
<p style="font-size:16pt; text-align:right;">
این کار مانند این است که جهانگرد خسته از صفحه فعلی شده و یک آدرس جدید و تصادفی در نوار آدرس مرورگر خود تایپ کند. این مکانیزم ساده دو مشکل را همزمان حل می‌کند: هم از بن‌بست‌ها خارج می‌شود و هم به تمام بخش‌های وب دسترسی پیدا می‌کند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🌐 PageRank در دنیای واقعی: داستان یک وبلاگ تکنولوژی</p>
<p style="font-size:16pt; text-align:right;">
بیایید یک مثال واقعی را بررسی کنیم. دو وبلاگ تکنولوژی جدید، «تک‌بین» و «کدپارسی»، همزمان فعالیت خود را آغاز می‌کنند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>وبلاگ «تک‌بین»</strong> موفق می‌شود از یک رسانه بزرگ معتبر مانند <strong>وی‌جی‌بی‌آی</strong> یک لینک دریافت کند.</li>
<li><strong>وبلاگ «کدپارسی»</strong> از ۵۰ وبلاگ کوچک‌تر اما تخصصی برنامه‌نویسی لینک می‌گیرد.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
کدام وبلاگ احتمالاً رتبه بالاتری در گوگل خواهد داشت؟</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم PageRank به ما می‌گوید که لینک از <strong>وی‌جی‌بی‌آی</strong> (که خود PageRank بالایی دارد) بسیار ارزشمند است و به «تک‌بین» امتیاز قابل توجهی می‌دهد. اما در عین حال، مجموع امتیاز ۵۰ لینک از وبلاگ‌های کوچک‌تر نیز می‌تواند به همان اندازه یا حتی بیشتر باشد.</p>
<p style="font-size:16pt; text-align:right;">
این مثال نشان می‌دهد که PageRank چگونه <strong>اعتبار دیجیتال</strong> را از طریق ساختار لینک‌های وب اندازه‌گیری می‌کند و به موتورهای جستجو کمک می‌کند تا صفحات معتبرتر را شناسایی کرده و در نتایج جستجو نمایش دهند.</p>
</div>

</div>

```python
# پیاده‌سازی الگوریتم PageRank

import random

class PageRankSimulator:
    """
    شبیه‌سازی الگوریتم PageRank با دو روش:
    1. روش تکراری (Iterative) — پایه‌ی اصلی محاسبه PageRank
    2. شبیه‌سازی Random Surfer — برای درک شهودی الگوریتم
    """
    
    def __init__(self, graph, alpha=0.1, tolerance=1e-6, max_iterations=100):
        """
        مقداردهی اولیه شبیه‌ساز PageRank.
        
        ورودی‌ها:
        - graph: دیکشنری نمایش‌دهنده گراف وب
                 مثال: {"A": ["B", "C"], "B": ["C"], "C": ["A"]}
        - alpha: احتمال تله‌پورت (معمولاً 0.1)
        - tolerance: آستانه همگرایی برای روش تکراری
        - max_iterations: حداکثر تعداد تکرار برای جلوگیری از حلقه نامتناهی
        """
        self.graph = graph
        self.alpha = alpha
        self.tolerance = tolerance
        self.max_iterations = max_iterations
        
        # لیست تمام گره‌ها (صفحات)
        self.nodes = set(graph.keys())
        for neighbors in graph.values():
            self.nodes.update(neighbors)
        self.nodes = list(self.nodes)
        self.num_nodes = len(self.nodes)
        
        # تبدیل نام گره‌ها به اندیس برای سرعت بیشتر
        self.node_to_index = {node: i for i, node in enumerate(self.nodes)}
    
    def iterative_pagerank(self):
        """
        محاسبه PageRank با روش تکراری (پایه‌ی موتورهای جستجوی واقعی).
        فرمول اصلی:
        PR(A) = (1 - α)/N + α * Σ(PR(Pi)/L(Pi))
        که در آن:
        - N = تعداد کل صفحات
        - Pi = صفحاتی که به A لینک داده‌اند
        - L(Pi) = تعداد لینک‌های خروجی صفحه Pi
        """
        N = self.num_nodes
        # مقداردهی اولیه: همه صفحات امتیاز یکسان دارند
        pagerank = [1.0 / N] * N
        new_pagerank = [0.0] * N
        
        for iteration in range(self.max_iterations):
            # مرحله 1: محاسبه مقدار تله‌پورت (1-α)/N برای همه صفحات
            teleport_value = (1.0 - self.alpha) / N
            new_pagerank = [teleport_value] * N
            
            # مرحله 2: توزیع امتیازات از طریق لینک‌ها
            for node in self.nodes:
                i = self.node_to_index[node]
                out_links = self.graph.get(node, [])
                if not out_links:
                    continue  # صفحه بدون لینک خروجی، امتیازی منتقل نمی‌کند
                
                # هر صفحه، امتیاز خود را به‌طور یکنواخت بین لینک‌های خروجی تقسیم می‌کند
                share = (self.alpha * pagerank[i]) / len(out_links)
                for neighbor in out_links:
                    j = self.node_to_index[neighbor]
                    new_pagerank[j] += share
            
            # بررسی همگرایی
            diff = sum(abs(new_pagerank[i] - pagerank[i]) for i in range(N))
            if diff < self.tolerance:
                print(f"✅ همگرایی پس از {iteration + 1} تکرار")
                break
            
            pagerank = new_pagerank[:]
        
        # تبدیل نتیجه به دیکشنری خوانا
        result = {}
        for i, node in enumerate(self.nodes):
            result[node] = pagerank[i]
        return result
    
    def simulate_random_surfer(self, steps=100000):
        """
        شبیه‌سازی مستقیم Random Surfer برای درک شهودی PageRank.
        این روش برای آموزش مناسب است، اما در عمل کارآمد نیست.
        """
        # شروع از یک صفحه تصادفی
        current = random.choice(self.nodes)
        visit_count = {node: 0 for node in self.nodes}
        
        for _ in range(steps):
            visit_count[current] += 1
            
            # تصمیم‌گیری: تله‌پورت یا پیمایش لینک؟
            if random.random() < self.alpha:
                # تله‌پورت: رفتن به صفحه‌ای تصادفی
                current = random.choice(self.nodes)
            else:
                # پیمایش لینک: انتخاب یک لینک خروجی به‌صورت تصادفی
                out_links = self.graph.get(current, [])
                if out_links:
                    current = random.choice(out_links)
                else:
                    # اگر لینک خروجی نداشت، تله‌پورت اجباری
                    current = random.choice(self.nodes)
        
        # تبدیل شمارش بازدیدها به احتمال (PageRank تقریبی)
        total_visits = sum(visit_count.values())
        approx_pagerank = {node: count / total_visits for node, count in visit_count.items()}
        return approx_pagerank

# مثال کاربردی: گراف نمونه از شکل 21.1
if __name__ == "__main__":
    # گراف نمونه: A → {B, C, D}, B → {C}, C → {A}, D → {}
    sample_graph = {
        "A": ["B", "C", "D"],
        "B": ["C"],
        "C": ["A"],
        "D": []  # صفحه بدون لینک خروجی (Sink Node)
    }
    
    print("مثال آموزشی PageRank برای گراف نمونه:")
    print("-" * 50)
    print("ساختار گراف:")
    for node, links in sample_graph.items():
        print(f"  {node} → {links if links else '[] (بدون لینک خروجی)'}")
    
    # محاسبه PageRank با روش تکراری
    pr_simulator = PageRankSimulator(sample_graph, alpha=0.1)
    iterative_result = pr_simulator.iterative_pagerank()
    
    print("\n📊 نتایج PageRank (روش تکراری):")
    sorted_pages = sorted(iterative_result.items(), key=lambda x: x[1], reverse=True)
    for rank, (page, score) in enumerate(sorted_pages, 1):
        print(f"  {rank}. صفحه {page}: {score:.6f}")
    
    # تأیید با شبیه‌سازی Random Surfer (برای همگرایی، مقدار steps را بالا ببرید)
    print("\n🔄 شبیه‌سازی Random Surfer (100,000 گام):")
    surfer_result = pr_simulator.simulate_random_surfer(steps=100000)
    surfer_sorted = sorted(surfer_result.items(), key=lambda x: x[1], reverse=True)
    for rank, (page, score) in enumerate(surfer_sorted, 1):
        print(f"  {rank}. صفحه {page}: {score:.6f}")
    
    print("\n💡 نکته:")
    print("- صفحه 'C' بالاترین امتیاز را دارد چون هم از A و هم از B لینک دریافت می‌کند.")
    print("- صفحه 'D' امتیاز پایینی دارد چون لینک خروجی ندارد و فقط از A لینک دریافت می‌کند.")
    print("- پارامتر α (تله‌پورت) از بن‌بست صفحه 'D' جلوگیری می‌کند.")
```

<p style="font-size:18pt; text-align:center;">
<span id="section21.2.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.2.1 زنجیره‌های مارکوف: سفر تصادفی در وب</span> 🌐
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎲 زنجیره مارکوف: هر قدم فقط به جای فعلی بستگی دارد</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در یک شهر جدید هستید و از یک توریست بپرسید: "اگر من الان در میدان اصلی هستم، کجا احتمالاً بعدی است؟" او به شما لیستی از مکان‌ها و احتمال رفتن به هر کدام را می‌دهد. سپس از میدان اصلی به یکی از آن مکان‌ها می‌روید و دوباره از یک توریست در آنجا می‌پرسید: "اگر من الان اینجا هستم، کجا بعدی می‌روم؟"</p>
<p style="font-size:16pt; text-align:right;">
این یک <strong>زنجیره مارکوف</strong> است! در دنیای وب، زنجیره مارکوف یک مدل ریاضی برای توصیف حرکت یک کاربر بین صفحات وب است. ویژگی کلیدی این است که <strong>آینده فقط به حالت فعلی بستگی دارد، نه به گذشته</strong>. این یعنی اگر شما در صفحه A هستید، احتمال رفتن شما به صفحه B فقط به صفحه فعلی (A) بستگی دارد، نه به صفحاتی که قبلاً بازدید کرده‌اید.</p>
<p style="font-size:16pt; text-align:right;">
این ویژگی که به آن <strong>خاصیت مارکوف</strong> می‌گویند، به ما اجازه می‌دهد تا رفتار کاربران در وب را با یک مدل ساده اما قدرتمند توصیف کنیم.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔍 ماتریس انتقال احتمال: نقشه احتمالات وب</p>
<p style="font-size:16pt; text-align:right;">
در زنجیره مارکوف، از یک <strong>ماتریس انتقال احتمال</strong> برای نمایش احتمال حرکت از یک صفحه به صفحه دیگر استفاده می‌شود. این ماتریس به شکل N×N است که N تعداد صفحات وب است.</p>
<p style="font-size:16pt; text-align:right;">
هر درایه P<sub>ij</sub> در این ماتریس نشان‌دهنده احتمال حرکت از صفحه i به صفحه j است. ویژگی‌های این ماتریس عبارتند از:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>هر درایه P<sub>ij</sub> در بازه [0, 1] قرار دارد.</li>
<li>مجموع درایه‌های هر سطر برابر با 1 است (∑<sub>j</sub> P<sub>ij</sub> = 1).</li>
<li>چنین ماتریسی یک <strong>ماتریس تصادفی</strong> (Stochastic Matrix) نامیده می‌شود.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
برای درک بهتر، به مثال زیر توجه کنید:</p>
<div style="background:#f8f9fa; border-right:3px solid #6c757d; padding:10px; margin:10px 0; border-radius:0 8px 8px 0; font-family:monospace; font-size:14pt;">
فرض کنید سه سایت خبری A، B و C داریم:
- سایت A به سایت‌های B و C لینک دارد.
- سایت B فقط به سایت A لینک دارد.
- سایت C فقط به سایت A لینک دارد.
</div>
<p style="font-size:16pt; text-align:right;">
ماتریس انتقال احتمال برای این شبکه به شکل زیر خواهد بود:</p>
<div style="background:#f8f9fa; border-right:3px solid #6c757d; padding:10px; margin:10px 0; border-radius:0 8px 8px 0; font-family:monospace; font-size:14pt;">
    A   B   C
A [ 0  0.5 0.5 ]
B [ 1   0   0  ]
C [ 1   0   0  ]
</div>
<p style="font-size:16pt; text-align:right;">
این ماتریس نشان می‌دهد که اگر در سایت A باشیم، 50% احتمال رفتن به B و 50% احتمال رفتن به C داریم. اما اگر در B یا C باشیم، با احتمال 100% به A بازمی‌گردیم.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🌐 ساخت ماتریس انتقال از گراف وب</p>
<p style="font-size:16pt; text-align:right;">
برای ساخت ماتریس انتقال P از گراف وب، مراحل زیر را طی می‌کنیم:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>ابتدا <strong>ماتریس مجاورت</strong> A را می‌سازیم: اگر از صفحه i به صفحه j لینک باشد، A<sub>ij</sub> = 1، در غیر این صورت A<sub>ij</sub> = 0.</li>
<li>اگر صفحه‌ای لینک خروجی نداشته باشد (صفحه تله)، تمام درایه‌های سطر آن برابر 1/N می‌شود (تله‌پورت اجباری).</li>
<li>در سایر صفحات، هر لینک خروجی احتمال 1/(تعداد لینک‌ها) دریافت می‌کند.</li>
<li>کل ماتریس در (1−α) ضرب می‌شود.</li>
<li>مقدار α/N به هر درایه اضافه می‌شود تا تله‌پورت جهانی را مدل کند.</li>
</ol>
<p style="font-size:16pt; text-align:right;">
پارامتر α (معمولاً بین 0.1 تا 0.2) احتمال <strong>تله‌پورت</strong> یا پرش تصادفی کاربر به هر صفحه وب را نشان می‌دهد، حتی اگر لینکی به آن صفحه وجود نداشته باشد. این ویژگی مدل را واقعی‌تر می‌کند، زیرا کاربران واقعی همیشه از طریق لینک‌ها حرکت نمی‌کنند و ممکن است مستقیماً آدرس صفحه‌ای را در نوار آدرس تایپ کنند.</p>
<p style="font-size:16pt; text-align:right;">
به عبارت دیگر، تله‌پورت به کاربر اجازه می‌دهد که در هر لحظه با احتمال α به هر صفحه‌ای در وب (حتی صفحاتی که به آن لینک ندارد) پرش کند. این ویژگی باعث می‌شود که زنجیره مارکوف <strong>ارگودیک</strong> شود و تضمین کند که حالت پایدار وجود دارد.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 توزیع احتمال و بردار احتمال: کجا هستیم؟</p>
<p style="font-size:16pt; text-align:right;">
در هر لحظه، موقعیت کاربر در وب را می‌توان با یک <strong>بردار احتمال</strong> نمایش داد. این بردار به ما می‌گوید که کاربر با چه احتمالی در هر صفحه قرار دارد.</p>
<p style="font-size:16pt; text-align:right;">
اگر در زمان t=0 کاربر در صفحه i باشد، بردار احتمال اولیه به این صورت است: درایه i برابر 1 و بقیه درایه‌ها برابر 0. پس از یک گام (t=1)، توزیع احتمال جدید با ضرب بردار احتمال فعلی در ماتریس انتقال به دست می‌آید: ~x<sub>1</sub> = ~x<sub>0</sub>P.</p>
<p style="font-size:16pt; text-align:right;">
به طور کلی، پس از t گام، توزیع احتمال برابر است با ~x<sub>t</sub> = ~x<sub>0</sub>P<sup>t</sup>.</p>
<p style="font-size:16pt; text-align:right;">
این فرآیند را می‌توان به این صورت تصور کرد: در هر گام، بردار احتمال فعلی (که نشان می‌دهد کاربر با چه احتمالی در هر صفحه است) با ماتریس انتقال ضرب می‌شود تا بردار احتمال جدید به دست آید.</p>
</div>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎯 حالت پایدار و PageRank: مقصد نهایی سفر تصادفی</p>
<p style="font-size:16pt; text-align:right;">
اگر زنجیره مارکوف <strong>ارگودیک</strong> باشد (یعنی هم <strong>غیرقابل‌تجزیه</strong> و هم <strong>نا‌دوره‌ای</strong> باشد)، پس از تعداد زیادی گام، توزیع احتمال به یک <strong>حالت پایدار</strong> همگرا می‌شود. در این حالت، احتمال حضور در هر صفحه دیگر تغییر نمی‌کند.</p>
<p style="font-size:16pt; text-align:right;">
این حالت پایدار را می‌توان با بردار ~π نمایش داد که معادله زیر را برآورده می‌کند:</p>
<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
~π = ~πP
  </div>
</div>
<p style="font-size:16pt; text-align:right;">
این بردار پایدار، <strong>بردار ویسته چپ اصلی</strong> ماتریس P است که به بزرگترین مقدار ویژه (که برابر 1 است) تعلق دارد.</p>
<p style="font-size:16pt; text-align:right;">
حالا نکته کلیدی: <strong>PageRank هر صفحه برابر با احتمال حضور در آن صفحه در حالت پایدار است!</strong> به عبارت دیگر، PageRank صفحه i برابر با π(i) است.</p>
<p style="font-size:16pt; text-align:right;">
این تعریف زیبا به ما می‌گوید که PageRank یک صفحه، احتمال حضور یک کاربر تصادفی در آن صفحه پس از گشت‌وگذار طولانی در وب است. صفحاتی که PageRank بالاتری دارند، صفحاتی هستند که کاربران تصادفی بیشتر به آن‌ها سر می‌زنند.</p>
<p style="font-size:16pt; text-align:right;">
قضیه 21.1 این را به صورت رسمی بیان می‌کند: برای هر زنجیره مارکوف ارگودیک، یک بردار احتمال پایدار منحصربه‌فرد ~π وجود دارد که بردار ویسته چپ اصلی P است. اگر η(i, t) تعداد بازدیدها از حالت i در t گام باشد، آنگاه:</p>
<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
lim<sub>t→∞</sub> η(i, t) / t = π(i)
  </div>
</div>
<p style="font-size:16pt; text-align:right;">
که در آن π(i) احتمال پایدار برای حالت i است و π(i) > 0 است.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🏢 مثال واقعی: PageRank در وب‌سایت‌های خبری</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید یک شبکه کوچک از وب‌سایت‌های خبری داریم:</p>

<p style="font-size:18pt; text-align:center;">
<span id="section21.2.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.2.2 محاسبه PageRank: از تئوری تا عمل</span> 🔢
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎯 چگونه PageRank محاسبه می‌شود؟</p>
<p style="font-size:16pt; text-align:right;">
PageRank هر صفحه برابر است با <strong>مقدار ورودی متناظر در بردار ویژه چپ اصلی</strong> ماتریس انتقال احتمال <em>P</em> است. این بردار <span style="direction: ltr;">~π</span> در معادله زیر صدق می‌کند:
</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
~π P = λ ~π
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
که در آن <em>λ</em> یک مقدار ویژه است. برای توزیع ایستا، <em>λ = 1</em> است، بنابراین:
</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
~π P = ~π
  </div>
</div>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔄 روش تکرار توان: شبیه‌سازی حرکت کاربر در وب</p>
<p style="font-size:16pt; text-align:right;">
برای محاسبه این بردار، از روشی به نام <strong>تکرار توان</strong> (Power Iteration) استفاده می‌شود. این روش را می‌توان به این صورت تصور کرد:</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید یک کاربر اینترنت هستید که به صورت تصادفی روی لینک‌ها کلیک می‌کند. از صفحه‌ای شروع می‌کنید و سپس لینک‌های تصادفی را دنبال می‌کنید. گاهی هم (با احتمال α) به صورت تصادفی به هر صفحه‌ای در اینترنت می‌پرید (تله‌پورت). اگر این کار را برای مدت طولانی ادامه دهید، در نهایت متوجه می‌شوید که درصد زمانی که در هر صفحه می‌گذرانید به یک مقدار ثابت همگرا می‌شود. این درصد زمانی همان PageRank آن صفحه است!</p>
<p style="font-size:16pt; text-align:right;">
از نظر ریاضی، اگر <span style="direction: ltr;">~x</span> توزیع اولیه احتمال روی حالت‌ها باشد، آنگاه توزیع در زمان <em>t</em> برابر است با <span style="direction: ltr;">~x P<sup>t</sup></span>. با افزایش <em>t</em>، این توزیع به یک <strong>توزیع ایستا</strong> همگرا می‌شود که مستقل از توزیع اولیه است (طبق قضیه 21.1).</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 مثال عملی: شبکه کوچک فروشگاه‌های آنلاین</p>
<p style="font-size:16pt; text-align:right;">
فرض کنید سه فروشگاه آنلاین داریم: فروشگاه اصلی (A) و دو فروشگاه وابسته (B و C). فروشگاه اصلی به هر دو فروشگاه وابسته لینک دارد، و هر دو فروشگاه وابسته فقط به فروشگاه اصلی لینک دارند. با نرخ تله‌پورت <em>α = 0.5</em>، ماتریس انتقال احتمال به‌صورت زیر است:</p>

<p style="font-size:16pt; text-align:right;">
در این مثال ساده، می‌توان توزیع ایستا را مستقیماً با استفاده از تقارن گراف محاسبه کرد: حالت‌های 1 و 3 متقارن هستند (چون سطرهای اول و سوم ماتریس <em>P</em> یکسان‌اند). بنابراین، اگر احتمال ایستای هر یک از این دو حالت را <em>p</em> فرض کنیم، آنگاه توزیع ایستا به‌صورت <span style="direction: ltr;">~π = (p, 1 − 2p, p)</span> خواهد بود. با جایگذاری در معادله <span style="direction: ltr;">~π = ~π P</span>، معادله ساده‌ای به‌دست می‌آید که راه‌حل آن <em>p = 5/18</em> است.</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 PageRank در موتورهای جستجوی واقعی</p>
<p style="font-size:16pt; text-align:right;">
مقادیر PageRank صفحات (و رتبه‌بندی ضمنی آن‌ها) <strong>مستقل از هر پرس‌وجویی</strong> است که کاربر ممکن است مطرح کند؛ بنابراین، PageRank یک معیار ایستا و مستقل از پرس‌وجو برای کیفیت ایستای هر صفحه وب است.</p>
<p style="font-size:16pt; text-align:right;">
اما در موتورهای جستجوی واقعی مانند گوگل، PageRank فقط یکی از صدها عامل در رتبه‌بندی نتایج جستجو است. برای مثال، وقتی شما عبارت "خرید لپ‌تاپ" را جستجو می‌کنید، گوگل نه تنها PageRank صفحات را در نظر می‌گیرد، بلکه به عواملی مانند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>مرتبط بودن محتوای صفحه با عبارت جستجو</li>
<li>کیفیت و اعتبار صفحه</li>
<li>تجربه کاربری صفحه (سرعت بارگذاری، طراحی ریسپانسیو و ...)</li>
<li>موقعیت جغرافیایی کاربر</li>
<li>تاریخچه جستجوی کاربر</li>
</ul>
<p style="font-size:16pt; text-align:right;">
نیز توجه می‌کند. در واقع، سهم نسبی PageRank در امتیاز کلی ممکن است توسط الگوریتم‌های یادگیری ماشین تعیین شود که با تحلیل میلیاردها جستجو، بهترین ترکیب از عوامل را برای هر نوع پرس‌وجو مشخص می‌کنند.</p>
<p style="font-size:16pt; text-align:right;">
این رویکرد ترکیبی به گوگل اجازه می‌دهد تا نه تنها صفحات با اعتبار عمومی بالا را نمایش دهد، بلکه نتایجی را ارائه کند که دقیقاً به نیاز کاربر پاسخ می‌دهند.</p>
</div>

```python
# پیاده‌سازی روش Power Iteration برای محاسبه PageRank

def power_iteration_pagerank(transition_matrix, initial_vector=None, tolerance=1e-6, max_iterations=100):
    """
    محاسبه بردار ویژه چپ اصلی (توزیع ایستا) با روش Power Iteration.
    
    ورودی‌ها:
    - transition_matrix: ماتریس انتقال احتمال P (لیستی از لیست‌ها)
    - initial_vector: بردار اولیه توزیع احتمال (پیش‌فرض: توزیع یکنواخت)
    - tolerance: آستانه همگرایی
    - max_iterations: حداکثر تعداد تکرار
    
    خروجی:
    - بردار ایستای ~π که PageRank صفحات را نشان می‌دهد
    """
    N = len(transition_matrix)
    
    # مقداردهی اولیه: توزیع یکنواخت اگر بردار اولیه داده نشده باشد
    if initial_vector is None:
        x = [1.0 / N] * N
    else:
        x = initial_vector[:]
    
    print("شروع روش Power Iteration برای محاسبه PageRank...")
    print(f"{'تکرار':<6} | {'بردار توزیع احتمال'}")
    print("-" * 55)
    print(f"{0:<6} | {[f'{v:.6f}' for v in x]}")
    
    for iteration in range(1, max_iterations + 1):
        # محاسبه x_new = x * P
        x_new = [0.0] * N
        for j in range(N):  # برای هر ستون j
            for i in range(N):  # برای هر سطر i
                x_new[j] += x[i] * transition_matrix[i][j]
        
        # بررسی همگرایی: تفاوت بین x و x_new
        diff = sum(abs(x_new[i] - x[i]) for i in range(N))
        print(f"{iteration:<6} | {[f'{v:.6f}' for v in x_new]}")
        
        if diff < tolerance:
            print(f"\n✅ همگرایی پس از {iteration} تکرار (تفاوت = {diff:.2e})")
            return x_new
        
        x = x_new[:]
    
    print(f"\n⚠️ هشدار: همگرایی پس از {max_iterations} تکرار حاصل نشد.")
    return x

# مثال 1: ماتریس انتقال از شکل 21.3 (مثال کتاب)
if __name__ == "__main__":
    print("مثال 1: محاسبه PageRank برای ماتریس انتقال شکل 21.3")
    print("=" * 60)
    
    # ماتریس انتقال P از فرمول (21.3)
    P_example = [
        [1/6, 2/3, 1/6],   # سطر اول
        [5/12, 1/6, 5/12], # سطر دوم
        [1/6, 2/3, 1/6]    # سطر سوم
    ]
    
    # بردار اولیه: شروع از حالت اول (x0 = [1, 0, 0])
    x0 = [1.0, 0.0, 0.0]
    
    result = power_iteration_pagerank(P_example, initial_vector=x0, tolerance=1e-8)
    
    print(f"\nتوزیع ایستای نهایی: {result}")
    print(f"بررسی: {result[0]:.6f} ≈ 5/18 = {5/18:.6f}, {result[1]:.6f} ≈ 4/9 = {4/9:.6f}")
    
    print("\n" + "="*60)
    print("مثال 2: محاسبه PageRank برای گراف شکل 21.4")
    print("="*60)
    
    # ماتریس انتقال برای گراف 7 گره‌ای شکل 21.4 (با α=0.14)
    P_fig21_4 = [
        [0.02, 0.02, 0.88, 0.02, 0.02, 0.02, 0.02],
        [0.02, 0.45, 0.45, 0.02, 0.02, 0.02, 0.02],
        [0.31, 0.02, 0.31, 0.31, 0.02, 0.02, 0.02],
        [0.02, 0.02, 0.02, 0.45, 0.45, 0.02, 0.02],
        [0.02, 0.02, 0.02, 0.02, 0.02, 0.02, 0.88],
        [0.02, 0.02, 0.02, 0.02, 0.02, 0.45, 0.45],
        [0.02, 0.02, 0.02, 0.31, 0.31, 0.02, 0.31]
    ]
    
    result2 = power_iteration_pagerank(P_fig21_4, tolerance=1e-8)
    print(f"\nبردار PageRank نهایی: {[f'{v:.6f}' for v in result2]}")
    print("نکته: صفحه d6 (ایندکس 6) بالاترین امتیاز را دارد، همان‌طور که در متن کتاب توضیح داده شده.")
    
    print("\n💡 نکته:")
    print("- این پیاده‌سازی نشان می‌دهد که چگونه می‌توان PageRank را بدون جبر خطی پیچیده محاسبه کرد.")
    print("- روش Power Iteration، پایه‌ی الگوریتم‌های واقعی محاسبه PageRank در موتورهای جستجو است.")
    print("- توزیع ایستا صرفاً به ساختار گراف و مقدار α بستگی دارد، نه به نقطه شروع.")
```

<p style="font-size:18pt; text-align:center;">
<span id="section21.2.3" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.2.3 PageRank شخصی‌سازی‌شده: وقتی گوگل شما را می‌شناسد</span> 🧠
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎯 فراتر از PageRank عمومی: دنیای PageRank موضوعی</p>
<p style="font-size:16pt; text-align:right;">
تا اینجا، PageRank را به‌عنوان معیاری برای اهمیت عمومی یک صفحه وب شناختیم. اما یک سؤال مهم: آیا اهمیت همه صفحات برای همه کاربران یکسان است؟ یک ورزشکار حرفه‌ای به دنبال نتایج زنده فوتبال است و یک محقق سیاسی به دنبال تحلیل‌های روز. آیا باید هر دو نتایج یکسانی دریافت کنند؟</p>
<p style="font-size:16pt; text-align:right;">
اینجاست که ایده <strong>PageRank موضوعی</strong> (Topic-specific PageRank) وارد می‌شود. در این مدل، ما فرض می‌کنیم کاربران به موضوعات خاصی علاقه دارند. برای مثال، یک کاربر علاقه‌مند به ورزش، ترجیح می‌دهد صفحات مرتبط با ورزش در نتایج جستجویش رتبه بالاتری داشته باشند.</p>
<p style="font-size:16pt; text-align:right;">
چگونه این کار را انجام می‌دهیم؟ با تغییر قوانین بازی تله‌پورت! به‌جای اینکه Random Surfer ما به یک صفحه وب کاملاً تصادفی پرش کند، او را به یک صفحه وب <strong>تصادفی اما مرتبط با موضوع مورد علاقه</strong> پرش می‌دهیم.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، اگر موضوع مورد نظر "ورزش" باشد، ما مجموعه‌ای از صفحات ورزشی معتبر (مثلاً از طریق دایرکتوری‌هایی مانند DMOZ یا با استفاده از دانش خود موتور جستجو) را شناسایی می‌کنیم. سپس، هر بار که سورفینگ ما تصمیم به تله‌پورت می‌گیرد، او را به یکی از این صفحات ورزشی می‌فرستیم.</p>
<p style="font-size:16pt; text-align:right;">
از آنجایی که وب‌سایت‌های ورزشی تمایل دارند به یکدیگر لینک دهند (مثلاً ESPN به سایت رسمی یک تیم)، این سورفینگ ورزشی علاقه‌مند به احتمال زیاد در اکوسیستم ورزش باقی می‌ماند. در نتیجه، <strong>توزیع ایستای جدید</strong> که به دست می‌آید (آن را <span style="direction: ltr;">~π<sub>ورزش</sub></span> می‌نامیم)، امتیاز بالاتری به صفحات ورزشی اختصاص می‌دهد.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">👤 PageRank شخصی‌سازی‌شده: ترکیب علایق</p>
<p style="font-size:16pt; text-align:right;">
اما افراد معمولاً فقط به یک موضوع علاقه ندارند. یک کاربر ممکن است هم به ورزش و هم به سیاست علاقه داشته باشد. اینجاست که ایده <strong>PageRank شخصی‌سازی‌شده</strong> (Personalized PageRank) به کمک ما می‌آید.</p>
<p style="font-size:16pt; text-align:right;">
فرض کنید کاربری به نام سارا وجود دارد که علایق او ترکیبی از ورزش (۶۰٪) و سیاست (۴۰٪) است. چگونه می‌توانیم PageRank را برای او شخصی‌سازی کنیم؟</p>
<p style="font-size:16pt; text-align:right;">
روند سورفینگ سارا به این صورت خواهد بود:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>هر بار که سارا تصمیم به تله‌پورت می‌گیرد، ابتدا سیستم تصمیم می‌گیرد که او را به دنیای ورزش بفرستد یا به دنیای سیاست. این تصمیم با احتمال ۶۰٪ به نفع ورزش و با احتمال ۴۰٪ به نفع سیاست گرفته می‌شود.</li>
<li>سپس، اگر قرار بود به دنیای ورزش برود، او را به یک صفحه تصادفی از میان مجموعه صفحات ورزشی می‌فرستد. اگر هم قرار بود به دنیای سیاست برود، او را به یک صفحه تصادفی از میان مجموعه صفحات سیاسی می‌فرستد.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این مدل منجر به یک زنجیره مارکوف جدید با توزیع ایستای منحصربه‌فردی می‌شود که دقیقاً متناسب با علایق سارا است.</p>
</div>


<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig21.5.bmp" alt="Topic-specific PageRank illustration" style="width:600px; border:1px solid #ccc; border-radius:12px;">
  <p style="font-size:14pt; text-align:center; color:#666;">شکل 21.5: نمایش PageRank موضوعی. در این مثال، کاربری با علایق ترکیبی (۶۰٪ ورزش و ۴۰٪ سیاست) در نظر گرفته شده است. اگر احتمال تله‌پورت ۱۰٪ باشد، این کاربر به این صورت مدل می‌شود که ۶٪ مواقع به صفحات ورزشی و ۴٪ مواقع به صفحات سیاسی تله‌پورت می‌کند.</p>
</div>


<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚡ معجزه خطی: محاسبه کارآمد برای میلیون‌ها کاربر</p>
<p style="font-size:16pt; text-align:right;">
در نگاه اول، ممکن است این ایده بسیار پرهزینه به نظر برسد. آیا باید برای هر کاربر با هر ترکیبی از علایق، یک محاسبه جداگانه انجام دهیم؟ این کار غیرممکن به نظر می‌رسد!</p>
<p style="font-size:16pt; text-align:right;">
اما اینجاست که جادوی جبر خطی به کمک ما می‌آید. نیازی نیست برای هر کاربر یک PageRank جداگانه محاسبه کنیم. به‌جای آن، فقط کافی است <strong>بردارهای PageRank موضوعی پایه</strong> را یک‌بار برای همیشه محاسبه کنیم (یعنی یک PageRank برای ورزش، یک PageRank برای سیاست، و غیره).</p>
<p style="font-size:16pt; text-align:right;">
سپس، PageRank شخصی‌سازی‌شده برای هر کاربر به‌سادگی یک <strong>ترکیب خطی</strong> از این بردارهای پایه است. برای سارا (۶۰٪ ورزش و ۴۰٪ سیاست)، PageRank شخصی‌سازی‌شده او به این صورت محاسبه می‌شود:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
~π<sub>سارا</sub> = 0.6 ~π<sub>ورزش</sub> + 0.4 ~π<sub>سیاست</sub>
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
این ویژگی، استفاده از PageRank شخصی‌سازی‌شده را حتی برای میلیاردها کاربر با پروفایل‌های مختلف ممکن می‌سازد. این روش به موتورهای جستجو اجازه می‌دهد تا نتایجی را ارائه دهند که دقیقاً متناسب با علایق فردی هر کاربر است.</p>
</div>

</div>

```python
# پیاده‌سازی PageRank موضوعی (Topic-specific)

def topic_specific_pagerank(graph, topic_nodes, alpha=0.1, tolerance=1e-6, max_iterations=100):
    """
    محاسبه PageRank موضوعی برای یک زیرمجموعه از صفحات (topic_nodes).
    
    ورودی‌ها:
    - graph: دیکشنری گراف وب (صفحه → لیست لینک‌های خروجی)
    - topic_nodes: مجموعه‌ای از صفحات مرتبط با یک موضوع خاص
    - alpha: احتمال تله‌پورت (معمولاً 0.1)
    - tolerance, max_iterations: پارامترهای همگرایی
    
    خروجی:
    - دیکشنری PageRank که فقط برای صفحات در مجموعه Y (صفحات قابل‌دسترسی از topic_nodes) غیرصفر است
    """
    # پیدا کردن تمام گره‌های مرتبط با موضوع (Y = صفحاتی که می‌توان به آن‌ها از topic_nodes رسید)
    all_nodes = set(graph.keys())
    for neighbors in graph.values():
        all_nodes.update(neighbors)
    
    # نگاشت گره‌ها به اندیس
    nodes = list(all_nodes)
    node_to_index = {node: i for i, node in enumerate(nodes)}
    N = len(nodes)
    
    # مقداردهی اولیه PageRank
    pagerank = [1.0 / N] * N
    
    # توزیع تله‌پورت: فقط به گره‌های موضوعی
    teleport_dist = [0.0] * N
    topic_count = 0
    for i, node in enumerate(nodes):
        if node in topic_nodes:
            teleport_dist[i] = 1.0
            topic_count += 1
    
    # اگر هیچ گره موضوعی وجود نداشت، تله‌پورت یکنواخت باشد
    if topic_count == 0:
        teleport_dist = [1.0 / N] * N
    else:
        teleport_dist = [p / topic_count for p in teleport_dist]  # توزیع یکنواخت روی موضوع
    
    # روش تکراری
    for iteration in range(max_iterations):
        new_pagerank = [0.0] * N
        
        # بخش تله‌پورت: (1 - α) * توزیع موضوعی
        for i in range(N):
            new_pagerank[i] += (1.0 - alpha) * teleport_dist[i]
        
        # بخش انتشار از طریق لینک‌ها: α * Σ(PR(Pi)/L(Pi))
        for node in nodes:
            i = node_to_index[node]
            if pagerank[i] == 0:
                continue
            out_links = graph.get(node, [])
            if not out_links:
                continue
            share = (alpha * pagerank[i]) / len(out_links)
            for neighbor in out_links:
                if neighbor in node_to_index:
                    j = node_to_index[neighbor]
                    new_pagerank[j] += share
        
        # بررسی همگرایی
        diff = sum(abs(new_pagerank[i] - pagerank[i]) for i in range(N))
        if diff < tolerance:
            break
        pagerank = new_pagerank[:]
    
    # ساخت دیکشنری نتیجه
    result = {}
    for i, node in enumerate(nodes):
        result[node] = pagerank[i]
    return result

def personalized_pagerank(sports_pagerank, politics_pagerank, sports_weight=0.6):
    """
    ساخت PageRank شخصی‌سازی‌شده با ترکیب خطی دو PageRank موضوعی.
    """
    all_nodes = set(sports_pagerank.keys()) | set(politics_pagerank.keys())
    personalized = {}
    for node in all_nodes:
        sp = sports_pagerank.get(node, 0.0)
        pp = politics_pagerank.get(node, 0.0)
        personalized[node] = sports_weight * sp + (1 - sports_weight) * pp
    return personalized

# مثال کاربردی: گراف نمونه با دو حوزه موضوعی
if __name__ == "__main__":
    # گراف نمونه: گره‌های s0, s1, s2 (ورزشی) و p0, p1, p2 (سیاسی)
    sample_graph = {
        # صفحات ورزشی (lینک‌های درون‌موضوعی زیاد)
        "s0": ["s1", "s2", "p0"],
        "s1": ["s0", "s2"],
        "s2": ["s0", "s1"],
        # صفحات سیاسی (lینک‌های درون‌موضوعی زیاد)
        "p0": ["p1", "p2", "s0"],
        "p1": ["p0", "p2"],
        "p2": ["p0", "p1"],
        # صفحه عمومی
        "general": ["s0", "p0"]
    }
    
    # تعریف مجموعه‌های موضوعی
    sports_nodes = {"s0", "s1", "s2"}
    politics_nodes = {"p0", "p1", "p2"}
    
    print("مثال آموزشی PageRank موضوعی و شخصی‌سازی‌شده")
    print("=" * 55)
    
    # محاسبه PageRank موضوعی برای ورزش
    sports_pr = topic_specific_pagerank(
        sample_graph, 
        sports_nodes, 
        alpha=0.1
    )
    
    # محاسبه PageRank موضوعی برای سیاست
    politics_pr = topic_specific_pagerank(
        sample_graph, 
        politics_nodes, 
        alpha=0.1
    )
    
    # ساخت PageRank شخصی‌سازی‌شده (60% ورزش، 40% سیاست)
    personalized_pr = personalized_pagerank(sports_pr, politics_pr, sports_weight=0.6)
    
    # نمایش نتایج
    print("\n📊 PageRank موضوعی (ورزش) — ۱۰ گره برتر:")
    sorted_sports = sorted(sports_pr.items(), key=lambda x: x[1], reverse=True)[:5]
    for node, score in sorted_sports:
        if score > 0:
            print(f"  {node}: {score:.6f}")
    
    print("\n📊 PageRank موضوعی (سیاست) — ۱۰ گره برتر:")
    sorted_politics = sorted(politics_pr.items(), key=lambda x: x[1], reverse=True)[:5]
    for node, score in sorted_politics:
        if score > 0:
            print(f"  {node}: {score:.6f}")
    
    print("\n🎯 PageRank شخصی‌سازی‌شده (60% ورزش + 40% سیاست) — ۱۰ گره برتر:")
    sorted_personal = sorted(personalized_pr.items(), key=lambda x: x[1], reverse=True)[:5]
    for node, score in sorted_personal:
        print(f"  {node}: {score:.6f}")
    
    print("\n💡 نکته:")
    print("- PageRank موضوعی، امتیاز بیشتری به صفحات درون یک حوزه اختصاص می‌دهد.")
    print("- PageRank شخصی‌سازی‌شده با ترکیب خطی ساخته می‌شود — نیازی به محاسبه مجدد نیست!")
    print("- این روش، امکان سفارشی‌سازی رتبه‌بندی برای میلیون‌ها کاربر را فراهم می‌کند.")
```

<p style="font-size:18pt; text-align:center;">
<span id="section21.3" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.3 هاب‌ها و مقام‌ها: کشف گنجینه‌های اطلاعاتی در وب</span> 🗺️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 سناریوی واقعی: جستجوی اطلاعات پزشکی</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید به تازگی به بیماری "میگرن" مبتلا شده‌اید و می‌خواهید اطلاعات معتبر و قابل اعتماد درباره آن پیدا کنید. اگر عبارت "میگرن" را در یک موتور جستجو وارد کنید، با هزاران نتیجه مواجه می‌شوید. کدام صفحات معتبر هستند و کدام صرفاً ادعا می‌کنند که متخصص هستند؟</p>
<p style="font-size:16pt; text-align:right;">
در اینجا، دو نوع صفحه وب وجود دارد که برای شما مفید خواهند بود:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>صفحات مقام (Authority)</strong>: این صفحات منابع اصلی و معتبر اطلاعات هستند. در مورد میگرن، این صفحات شامل وب‌سایت‌های انجمن‌های پزشکی معتبر، مقالات علمی در مجلات پزشکی، یا صفحات وب‌سایت‌های سازمان‌های بهداشتی معتبر است.</li>
<li><strong>صفحات هاب (Hub)</strong>: این صفحات خودشان حاوی اطلاعات اصلی نیستند، بلکه مجموعه‌ای از لینک‌ها به منابع معتبر را گردآوری کرده‌اند. مثلاً یک وبلاگ که توسط یک بیمار میگرنی اداره می‌شود و بهترین منابع اطلاعاتی درباره میگرن را معرفی کرده است.</li>
</ol>
<p style="font-size:16pt; text-align:right;">
الگوریتم <strong>HITS</strong> (Hyperlink-Induced Topic Search) دقیقاً برای همین منظور طراحی شده است: برای هر پرس‌وجو، دو امتیاز جداگانه به هر صفحه اختصاص می‌دهد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>امتیاز هاب</strong> (Hub Score): نشان‌دهنده توانایی صفحه در اشاره به صفحات معتبر است.</li>
<li><strong>امتیاز مقام</strong> (Authority Score): نشان‌دهنده اعتبار صفحه به‌عنوان منبع اطلاعات است.</li>
</ul>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔄 رابطه متقابل هاب‌ها و مقام‌ها</p>
<p style="font-size:16pt; text-align:right;">
ایده کلیدی پشت الگوریتم HITS این است که هاب‌ها و مقام‌ها به‌صورت متقابل یکدیگر را تقویت می‌کنند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>یک <strong>هاب خوب</strong> صفحه‌ای است که به مقام‌های خوب لینک می‌دهد.</li>
<li>یک <strong>مقام خوب</strong> صفحه‌ای است که توسط هاب‌های خوب به آن لینک داده شده است.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این رابطه متقابل را می‌توان با معادلات زیر نشان داد:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
h(v) ← Σ<sub>v→y</sub> a(y)<br>
a(v) ← Σ<sub>y→v</sub> h(y)
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
- <em>h(v)</em> امتیاز هاب صفحه <em>v</em> است،  
- <em>a(v)</em> امتیاز مقام صفحه <em>v</em> است.  
- مجموع اول برابر است با مجموع امتیازات مقام صفحاتی که <em>v</em> به آن‌ها لینک داده است.
- مجموع دوم برابر است با مجموع امتیازات هاب صفحاتی که به <em>v</em> لینک داده‌اند.</p>

<p style="font-size:16pt; text-align:right;">
این معادلات به‌صورت ماتریسی بازنویسی می‌شوند:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
~h ← A ~a<br>
~a ← A<sup>T</sup> ~h
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
که در آن <em>A</em> ماتریس مجاورت گراف وب است. با جایگزینی مکرر، به معادلات ویژه‌مقداری می‌رسیم:</p>

<div style="text-align: center; margin: 15px 0;">
  <div style="
    display: inline-block;
    direction: ltr;
    font-family: monospace;
    font-size: 14pt;
    background: #f0f0f0;
    padding: 10px;
    border-radius: 20px;
    text-align: left;
  ">
~h = (1/λ<sub>h</sub>) A A<sup>T</sup> ~h<br>
~a = (1/λ<sub>a</sub>) A<sup>T</sup> A ~a
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
بنابراین، بردارهای نهایی هاب و مقام، <strong>بردارهای ویژه اصلی</strong> ماتریس‌های <span style="direction: ltr;">A A<sup>T</sup></span> و <span style="direction: ltr;">A<sup>T</sup> A</span> هستند. در عمل، این بردارها با روش‌های تکراری (مثل Power Iteration) محاسبه می‌شوند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 مثال کاربردی: جستجوی اطلاعات در مورد هوش مصنوعی</p>
<p style="font-size:16pt; text-align:right;">
فرض کنید می‌خواهید اطلاعات معتبر در مورد "یادگیری عمیق" پیدا کنید. در این حالت:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مقام‌ها</strong> ممکن است شامل صفحات وب‌سایت‌های دانشگاهی معتبر (مثل استنفورد، ام‌آی‌تی)، مقالات بنیادی در این زمینه، یا مستندات رسمی فریمورک‌های معتبر یادگیری عمیق (مثل TensorFlow یا PyTorch) باشند.</li>
<li><strong>هاب‌ها</strong> ممکن است شامل وبلاگ‌های تخصصی که بهترین منابع یادگیری عمیق را معرفی کرده‌اند، یا صفحه‌هایی که مجموعه‌ای از مقالات مهم و آموزش‌های باکیفیت را گردآوری کرده‌اند، باشند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
الگوریتم HITS با تحلیل لینک‌ها بین این صفحات، می‌تواند به شما کمک کند تا هم منابع اصلی و معتبر (مقام‌ها) و هم راهنماهای مفید برای پیدا کردن این منابع (هاب‌ها) را شناسایی کنید.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 مراحل اجرای الگوریتم HITS</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم HITS به‌صورت زیر عمل می‌کند:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>برای پرس‌وجوی داده‌شده، یک زیرمجموعه از صفحات وب مرتبط را جمع‌آوری می‌کنیم (این فرآیند در بخش بعدی توضیح داده می‌شود).</li>
<li>برای هر صفحه در این زیرمجموعه، امتیاز هاب و مقام اولیه را 1 در نظر می‌گیریم.</li>
<li>به‌صورت تکراری، امتیازات هاب و مقام را با استفاده از معادلات به‌روزرسانی می‌کنیم تا زمانی که مقادیر همگرا شوند.</li>
<li>در نهایت، صفحات را بر اساس امتیازات هاب و مقام رتبه‌بندی کرده و دو لیست جداگانه ارائه می‌دهیم.</li>
</ol>
<p style="font-size:16pt; text-align:right;">
نکته کلیدی این است که HITS فقط ساختار لینک‌ها را تحلیل می‌کند و به محتوای متن صفحات توجهی ندارد (مگر در مرحله جمع‌آوری زیرمجموعه اولیه).</p>
</div>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔬 مثال عددی: تحلیل یک شبکه کوچک</p>
<p style="font-size:16pt; text-align:right;">
در مثال 21.2 برای پرس‌وجوی "jaguar" (که هم می‌تواند به حیوان و هم به خودرو اشاره کند)، الگوریتم HITS صفحات زیر را به‌عنوان مقام‌ها و هاب‌های اصلی شناسایی می‌کند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>صفحه q3 به‌عنوان <strong>مقام اصلی</strong> شناخته می‌شود، زیرا دو هاب (q2 و q6) از طریق لینک‌هایی با متن "jaguar" به آن اشاره کرده‌اند.</li>
<li>صفحات q2 و q6 به‌عنوان <strong>هاب‌های اصلی</strong> شناخته می‌شوند، زیرا به مقام‌های معتبر لینک داده‌اند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این مثال نشان می‌دهد که چگونه HITS می‌تواند با تحلیل ساختار لینک‌ها، صفحات معتبر را از صفحات کم‌اهمیت‌تر تمیز دهد، حتی وقتی کلمات کلیدی به تنهایی کافی نیستند.</p>
</div>

</div>

```python
# پیاده‌سازی الگوریتم HITS (Hubs and Authorities)

def hits_algorithm(graph, max_iterations=100, tolerance=1e-6):
    """
    محاسبه امتیازهای هاب و مقام برای گراف داده‌شده.
    
    ورودی‌ها:
    - graph: دیکشنری نمایش‌دهنده گراف جهت‌دار
             مثال: {"A": ["B", "C"], "B": ["C"], "C": ["A"]}
    - max_iterations: حداکثر تعداد تکرار
    - tolerance: آستانه همگرایی
    
    خروجی:
    - دیکشنری شامل دو کلید:
        'hubs': دیکشنری امتیاز هاب‌ها
        'authorities': دیکشنری امتیاز مقام‌ها
    """
    # جمع‌آوری تمام گره‌ها
    all_nodes = set(graph.keys())
    for neighbors in graph.values():
        all_nodes.update(neighbors)
    nodes = sorted(all_nodes)  # مرتب‌سازی برای نتایج قابل‌پیش‌بینی
    
    # مقداردهی اولیه: همه امتیازها برابر 1
    hubs = {node: 1.0 for node in nodes}
    authorities = {node: 1.0 for node in nodes}
    
    for iteration in range(max_iterations):
        # مرحله 1: به‌روزرسانی امتیاز مقام‌ها
        new_authorities = {}
        for node in nodes:
            auth_score = 0.0
            # جمع امتیاز هاب‌هایی که به این گره لینک داده‌اند
            for source, targets in graph.items():
                if node in targets:
                    auth_score += hubs.get(source, 0.0)
            new_authorities[node] = auth_score
        
        # مرحله 2: به‌روزرسانی امتیاز هاب‌ها
        new_hubs = {}
        for node in nodes:
            hub_score = 0.0
            # جمع امتیاز مقام‌هایی که این گره به آن‌ها لینک داده
            for target in graph.get(node, []):
                hub_score += new_authorities.get(target, 0.0)
            new_hubs[node] = hub_score
        
        # نرمال‌سازی برای جلوگیری از انفجار عددی
        # نرمال‌سازی L2 (رایج‌ترین روش در HITS)
        def normalize(score_dict):
            total_sq = sum(v**2 for v in score_dict.values())
            if total_sq == 0:
                return {k: 0.0 for k in score_dict}
            norm = total_sq ** 0.5
            return {k: v / norm for k, v in score_dict.items()}
        
        new_authorities = normalize(new_authorities)
        new_hubs = normalize(new_hubs)
        
        # بررسی همگرایی
        auth_diff = sum(abs(new_authorities[n] - authorities[n]) for n in nodes)
        hub_diff = sum(abs(new_hubs[n] - hubs[n]) for n in nodes)
        
        if auth_diff < tolerance and hub_diff < tolerance:
            print(f"✅ همگرایی پس از {iteration + 1} تکرار")
            break
        
        authorities = new_authorities
        hubs = new_hubs
    
    return {"hubs": hubs, "authorities": authorities}

# مثال کاربردی: گراف شکل 21.4 (ساده‌شده برای آموزش)
if __name__ == "__main__":
    # گراف نمونه مبتنی بر شکل 21.4 با گره‌های d0 تا d6
    sample_graph = {
        "d0": ["d2", "d1"],
        "d1": ["d2"],
        "d2": ["d0", "d3"],
        "d3": ["d4", "d6"],
        "d4": ["d6"],
        "d5": ["d6"],
        "d6": ["d3"]
    }
    
    print("مثال آموزشی الگوریتم HITS برای گراف شکل 21.4")
    print("=" * 50)
    
    # اجرای الگوریتم HITS
    results = hits_algorithm(sample_graph)
    hubs = results["hubs"]
    authorities = results["authorities"]
    
    # نمایش ۳ گره برتر در هر دسته
    print("\n🏆 ۳ گره برتر از نظر امتیاز مقام (Authorities):")
    top_authorities = sorted(authorities.items(), key=lambda x: x[1], reverse=True)[:3]
    for rank, (node, score) in enumerate(top_authorities, 1):
        print(f"  {rank}. {node}: {score:.6f}")
    
    print("\n🌐 ۳ گره برتر از نظر امتیاز هاب (Hubs):")
    top_hubs = sorted(hubs.items(), key=lambda x: x[1], reverse=True)[:3]
    for rank, (node, score) in enumerate(top_hubs, 1):
        print(f"  {rank}. {node}: {score:.6f}")
    
    print("\n💡 نکته:")
    print("- الگوریتم HITS دو نوع صفحه را شناسایی می‌کند:")
    print("  • مقام‌ها: صفحاتی که توسط هاب‌های متعددی لینک شده‌اند")
    print("  • هاب‌ها: صفحاتی که به مقام‌های متعددی لینک داده‌اند")
    print("- برخلاف PageRank، HITS وابسته به پرس‌وجو است و روی یک زیرگراف موضوعی اجرا می‌شود.")
    print("- نرمال‌سازی L2 جلوی رشد نامحدود امتیازها را می‌گیرد و همگرایی را تضمین می‌کند.")
```

<p style="font-size:18pt; text-align:center;">
<span id="section21.3.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">21.3.1 انتخاب زیرمجموعه وب: هنر پیدا کردن گنجینه‌های پنهان</span> 🗺️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 چالش اصلی: صفحات معتبر بدون کلمات کلیدی</p>
<p style="font-size:16pt; text-align:right;">
یک چالش بزرگ در الگوریتم HITS این است که بسیاری از صفحات معتبر ممکن است کلمه کلیدی پرس‌وجو را در محتوای خود نداشته باشند. این پدیده به خصوص برای شرکت‌های بزرگ و برندهای معتبر رایج است.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وب‌سایت شرکت اپل منبع معتبری برای اطلاعات درباره آیفون است، اما ممکن است صفحه اصلی آن کلمه "آیفون" را نداشته باشد و فقط از برند "Apple" استفاده کند. یا وب‌سایت سازمان بهداشت جهانی (WHO) منبع معتبری برای اطلاعات درباره کرونا است، اما ممکن است در صفحه اصلی خود از کلمه "کرونا" استفاده نکند و عبارت "COVID-19" را به کار ببرد.</p>
<p style="font-size:16pt; text-align:right;">
این چالش باعث می‌شود که اگر فقط صفحاتی را که شامل کلمه کلیدی پرس‌وجو هستند در نظر بگیریم، ممکن است منابع معتبر و اصلی را از دست بدهیم.</p>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🎯 راه‌حل: از مجموعه ریشه تا مجموعه پایه</p>
<p style="font-size:16pt; text-align:right;">
برای حل این مشکل، از یک فرآیند دو مرحله‌ای برای ساخت مجموعه پایه استفاده می‌کنیم:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مجموعه ریشه (Root Set)</strong>: تمام صفحاتی که کلمه کلیدی پرس‌وجو را در متن خود دارند. برای مثال، اگر پرس‌وجوی ما "هوش مصنوعی در پزشکی" باشد، مجموعه ریشه شامل تمام صفحاتی است که این عبارت را در متن خود دارند.</li>
<li><strong>مجموعه پایه (Base Set)</strong>: شامل مجموعه ریشه به‌علاوه تمام صفحاتی که یا به صفحات ریشه لینک داده‌اند یا توسط صفحات ریشه لینک شده‌اند.</li>
</ol>
<p style="font-size:16pt; text-align:right;">
برای درک بهتر، این مثال را در نظر بگیرید: فرض کنید در حال جستجوی اطلاعات درباره "سرطان سینه" هستیم. مجموعه ریشه شامل صفحاتی است که مستقیماً از این عبارت استفاده می‌کنند. اما وب‌سایت انجمن سرطان آمریکا، که یک منبع بسیار معتبر است، ممکن است در صفحه اصلی خود از عبارت "سرطان سینه" استفاده نکند و فقط از عبارت "Breast Cancer" استفاده کند.</p>
<p style="font-size:16pt; text-align:right;">
با این حال، اگر یک وبلاگ پزشکی (که در مجموعه ریشه قرار دارد) به وب‌سایت انجمن سرطان آمریکا لینک داده باشد، این وب‌سایت معتبر وارد مجموعه پایه می‌شود و در نتیجه در تحلیل HITS شرکت داده خواهد شد.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">💡 مزایای کلیدی این رویکرد</p>
<p style="font-size:16pt; text-align:right;">
این روش گسترش مجموعه ریشه به مجموعه پایه سه مزیت کلیدی دارد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>شناسایی مقام‌های پنهان</strong>: صفحات معتبری که کلمه کلیدی را ندارند، از طریق لینک‌های ورودی از هاب‌های مرتبط شناسایی می‌شوند.</li>
<li><strong>گسترش شبکه مقام‌ها</strong>: اگر یک هاب معتبر در مجموعه ریشه باشد، تمام مقام‌هایی که به آن‌ها لینک داده، در مجموعه پایه گنجانده می‌شوند.</li>
<li><strong>شناسایی هاب‌های مرتبط</strong>: اگر یک مقام معتبر در مجموعه ریشه باشد، تمام هاب‌هایی که به آن لینک داده‌اند، وارد مجموعه پایه می‌شوند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
به طور خلاصه، این گسترش به ما کمک می‌کند تا یک شبکه غنی از هاب‌ها و مقام‌های مرتبط با موضوع مورد نظر ایجاد کنیم، حتی اگر برخی از این صفحات مستقیماً کلمه کلیدی پرس‌وجو را در متن خود نداشته باشند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🌐 پدیده شگفت‌انگیز: بازیابی بین‌زبانی</p>
<p style="font-size:16pt; text-align:right;">
یکی از جالب‌ترین نتایج الگوریتم HITS این است که می‌تواند صفحات را به زبان‌های مختلف بازیابی کند، حتی اگر پرس‌وجو به یک زبان خاص باشد. این پدیده که <strong>بازیابی بین‌زبانی</strong> (Cross-Language Retrieval) نامیده می‌شود، کاملاً از طریق تحلیل لینک (نه ترجمه زبانی) حاصل می‌شود.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، اگر شما عبارت "japan elementary schools" را به انگلیسی جستجو کنید، ممکن است صفحات ژاپنی (به زبان ژاپنی) در نتایج ظاهر شوند. چطور این اتفاق می‌افتد؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید یک وبلاگ انگلیسی‌زبان درباره سیستم آموزشی ژاپن وجود دارد که به وب‌سایت‌های مدارس ژاپنی (به زبان ژاپنی) لینک داده است. این وبلاگ انگلیسی در مجموعه ریشه قرار می‌گیرد (چون حاوی کلمات کلیدی انگلیسی است)، اما وب‌سایت‌های ژاپنی که به آن‌ها لینک داده شده، در مجموعه پایه قرار می‌گیرند. سپس الگوریتم HITS این وب‌سایت‌های ژاپنی را به‌عنوان مقام‌های معتبر شناسایی می‌کند، حتی اگر به زبان انگلیسی نباشند.</p>
</div>

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">⚙️ نکات اجرایی و بهینه‌سازی</p>
<p style="font-size:16pt; text-align:right;">
در پیاده‌سازی عملی الگوریتم HITS، چند نکته مهم برای بهینه‌سازی عملکرد وجود دارد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>اندازه مجموعه ریشه</strong>: نیازی به شامل کردن همه صفحات مرتبط با پرس‌وجو نیست. معمولاً حدود 200 صفحه برای مجموعه ریشه کافی است تا نتایج باکیفیتی تولید شود.</li>
<li><strong>تعداد تکرارها</strong>: برای محاسبه امتیازهای هاب و مقام، نیازی به دقت بالا نیست. معمولاً 5 تا 10 تکرار از معادلات HITS برای یافتن رتبه‌بندی نسبی کافی است.</li>
<li><strong>بهینه‌سازی محاسبات</strong>: به‌دلیل پراکنده‌بودن گراف وب (هر صفحه به طور متوسط به حدود 10 صفحه دیگر لینک می‌دهد)، به‌روزرسانی‌ها به‌صورت جمع‌های افزایشی (نه ضرب ماتریسی) انجام می‌شود تا محاسبات کارآمدتر شوند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این بهینه‌سازی‌ها باعث می‌شوند که الگوریتم HITS بتواند به سرعت و با کارایی بالا نتایج باکیفیتی تولید کند، حتی برای مجموعه‌های بزرگ از صفحات وب.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#f0f8ff; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section21.4" style="font-size:24pt; font-weight:bold; color:#8549EB;">جمع‌بندی فصل ۲۱: تحلیل پیوندها</span> 🔗
</p>

<p style="font-size:16pt; text-align:right;">
چگونه می‌توان اهمیت و اعتبار یک صفحه وب را بر اساس ساختار پیوندهای آن در وب سنجید؟ فصل ۲۱ به ما نشان می‌دهد که چگونه تحلیل گراف وب (شبکه‌ای از صفحات و پیوندها) منجر به توسعه الگوریتم‌های رتبه‌بندی انقلابی مانند PageRank و HITS شده است. این الگوریتم‌ها با تفسیر هر پیوند به عنوان یک «رأی اعتماد»، توانستند کیفیت صفحات را به شیوه‌ای فراتر از تطبیق متن ساده ارزیابی کنند و به موتورهای جستجو کمک کنند تا نتایج مرتبط‌تر و معتبرتری را به کاربران ارائه دهند.
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>با **مفهوم متن لنگر (Anchor Text)** آشنا شدیم. این متن، همان متن قابل کلیکی است که در یک پیوند hyperlink قرار دارد و اغلب توصیف دقیقی از صفحه مقصد ارائه می‌دهد. این ویژگی به ویژه برای ایندکس‌گذاری صفحاتی که خودشان به درستی توصیف نشده‌اند (مانند صفحه اصلی IBM که ممکن است کلمه "computer" را نداشته باشد) بسیار مفید است.</li>
<li>الگوریتم **PageRank** را بررسی کردیم. این الگوریتم، که توسط گوگل توسعه یافته، هر صفحه را بر اساس تعداد و کیفیت پیوندهایی که به آن داده شده است، امتیازی بین ۰ و ۱ اختصاص می‌دهد. این مدل بر اساس مفهوم «کاربر تصادفی» (Random Surfer) عمل می‌کند که در وب به صورت تصادفی پیوندها را دنبال می‌کند یا به صورت تصادفی به صفحه‌ای دیگر «تلپورت» می‌کند. PageRank، احتمال حضور کاربر در یک صفحه در حالت پایدار را نشان می‌دهد و یک معیار کیفیت ایستا و مستقل از پرس‌وجو است.</li>
<li>با **PageRank موضوع-محور (Topic-Specific PageRank)** آشنا شدیم. در این رویکرد، به جای تلپورت کردن کاربر به یک صفحه کاملاً تصادفی، او به صفحات مرتبط با یک موضوع خاص (مانند ورزش یا سیاست) تلپورت می‌کند. این کار باعث می‌شود صفحات مرتبط با آن موضوع، امتیاز PageRank بالاتری کسب کنند. این ایده اساس **PageRank شخصی‌سازی‌شده (Personalized PageRank)** را نیز فراهم می‌کند که می‌تواند بر اساس علایق ترکیبی یک کاربر، امتیازهای PageRank مختلف را با هم ترکیب کند.</li>
<li>الگوریتم **HITS (Hyperlink-Induced Topic Search)** را مطالعه کردیم. این الگوریتم برای هر صفحه دو امتیار محاسبه می‌کند: **امتیاز هاب (Hub Score)** و **امتیاز اتوریتی (Authority Score)**. صفحات **اتوریتی** منابع معتبر و اصلی در یک موضوع هستند (مانند صفحه یک مؤسسه تحقیقاتی معتبر). صفحات **هاب** صفحاتی هستند که به منابع معتبر زیادی لینک می‌دهند (مانند یک فهرست دست‌نویس شده از منابع مفید). این الگوریتم به صورت تکراری این دو امتیاز را محاسبه می‌کند تا بهترین هاب‌ها و اتوریتی‌ها را برای یک پرس‌وجوی مشخص پیدا کند.</li>
</ul>

<div style="background:#e8f4fd; border-radius:8px; padding:15px; margin:15px 0;">
<p style="font-size:16pt; text-align:right;">
<b>بیایید این مفاهیم را با چند مثال واقعی و کاربردی زنده کنیم:</b>
</p>
<hr style="border-top: 1px solid #ccc; margin: 10px 0;">

<p style="font-size:16pt; text-align:right;">
<b>🔗 قدرت متن لنگر:</b> ممکن است صفحه اصلی شرکت IBM (ibm.com) کلمه "computer" را در متن خود نداشته باشد. اما هزاران وب‌سایت دیگر در هنگام لینک دادن به IBM، از کلمه "computer" در متن لنگر خود استفاده می‌کنند. موتورهای جستجو از این سیگنال استفاده می‌کنند و به صفحه IBM امتیاز بالایی برای جستجوی "computer" می‌دهند. به همین دلیل، جستجوی "big blue" (لقب معروف IBM) نیز معمولاً صفحه اصلی این شرکت را به عنوان نتیجه برتر نشان می‌دهد.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🏆 PageRank به مثابۀ رأی اعتماد:</b> تصور کنید هر پیوند به یک صفحه، یک رأی اعتماد به آن صفحه است. اما همه رأی‌ها برابر نیستند. یک پیوند از یک وب‌سایت بسیار معتبر و محبوب (مانند یک خبرگزاری بزرگ) بسیار ارزشمندتر از یک پیوند از یک وبلاگ ناشناس است. PageRank دقیقاً همین منطق را پیاده‌سازی می‌کند؛ صفحاتی که از منابع معتبرتری لینک دریافت کرده‌اند، رتبه بالاتری خواهند داشت، زیرا کاربر تصادفی احتمال بیشتری دارد که از طریق آن پیوندهای معتبر به این صفحات برسد.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🎯 HITS: پیدا کردن مراجع و راهنماها:</b> اگر شما عبارت "لوکیمی" را جستجو کنید، الگوریتم HITS تلاش می‌کند دو نوع صفحه را پیدا کند: ۱) صفحات **اتوریتی** مانند وب‌سایت موسسه ملی سرطان که منبع اصلی و معتبر اطلاعات در این زمینه است. ۲) صفحات **هاب** مانند وبلاگ یک پزشک یا صفحه‌ای که یک بیمار برای جمع‌آوری منابع مفید درباره لوکیمی ساخته است و به منابع معتبر متعددی (از جمله موسسه ملی سرطان) لینک داده است. HITS با تحلیل این پیوندها، بهترین منابع (اتوریتی‌ها) و بهترین راهنماها (هاب‌ها) را شناسایی می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🧭 PageRank شخصی‌سازی‌شده:</b> یک کاربر ممکن است ۶۰٪ به ورزش و ۴۰٪ به سیاست علاقه داشته باشد. موتور جستجو می‌تواند با ترکیب PageRank موضوع-محور ورزش (که در آن تلپورت به صفحات ورزشی است) و PageRank موضوع-محور سیاست، یک رتبه‌بندی شخصی‌سازی‌شده برای این کاربر ایجاد کند. در این رتبه‌بندی، صفحات مرتبط با ورزش و سیاست بر اساس ترکیب علایق او، امتیاز بالاتری کسب می‌کنند و نتایج جستجوی شخصی‌تری به او نمایش داده می‌شود.
</p>

</div>

<p style="font-size:16pt; text-align:right;">
این فصل نشان داد که تحلیل پیوندها با تبدیل وب به یک گراف بزرگ و استفاده از هوش جمعی سازندگان صفحات وب (از طریق پیوندهایشان)، انقلابی در فناوری جستجو ایجاد کرد. الگوریتم‌هایی مانند PageRank و HITS با فراتر رفتن از تطبیق ساده متن، توانستند اعتبار و کیفیت صفحات را بسنجند و این یکی از دلایل اصلی موفقیت موتورهای جستجوی مدرن در ارائه نتایج باکیفیت و مرتبط است.
</p>
