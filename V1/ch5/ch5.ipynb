<p style="font-size:18pt; text-align:center;">
<span id="section5.1" style="font-size:24pt; font-weight:bold; color:#FF6700;">فصل ۵: فشرده‌سازی ایندکس - هنر کوچک کردن داده‌ها بدون از دست دادن اطلاعات</span> 🗜️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌍 تصور کنید مدیر داده گوگل هستید</p>
<p style="font-size:16pt; text-align:right;">
هر روز میلیاردها جستجو در گوگل انجام می‌شود. برای پاسخ به این جستجوها در کسری از ثانیه، گوگل باید یک ایندکس عظیم از کل وب را مدیریت کند. این ایندکس آنقدر بزرگ است که اگر بخواهیم آن را بدون فشرده‌سازی ذخیره کنیم، به هزاران سرور نیاز داریم! اما گوگل چگونه این کار را انجام می‌دهد؟ پاسخ در فشرده‌سازی هوشمندانه ایندکس‌هاست.</p>
</div>

<p style="font-size:16pt; text-align:right;">
فشرده‌سازی ایندکس یکی از مهم‌ترین تکنیک‌ها در سیستم‌های بازیابی اطلاعات است. در این فصل، با روش‌های فشرده‌سازی دیکشنری و ایندکس معکوس آشنا می‌شویم که برای سیستم‌های بازیابی اطلاعات کارآمد ضروری هستند.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">💰 مزایای فشرده‌سازی: فراتر از صرفه‌جویی در فضا</p>
<p style="font-size:16pt; text-align:right;">
اولین مزیت فشرده‌سازی که به ذهن می‌رسد، کاهش فضای ذخیره‌سازی است. با نسبت فشرده‌سازی 1:4، می‌توانیم هزینه ذخیره‌سازی ایندکس را %75 کاهش دهیم. اما دو مزیت مهم دیگر هم وجود دارد:</p>
<p style="font-size:16pt; text-align:right;">
<strong>۱. افزایش استفاده از حافظه کش:</strong> تصور کنید در یک کتابخانه دیجیتال بزرگ، کاربران مدام به دنبال کلمه «کرونا» می‌گردند. اگر لیست ارسال این کلمه را در حافظه اصلی ذخیره کنیم، پاسخ به این جستجوها بسیار سریع خواهد بود. با فشرده‌سازی، می‌توانیم اطلاعات بیشتری را در حافظه اصلی نگه داریم و در نتیجه، پاسخ به جستجوها سریع‌تر انجام شود.</p>
<p style="font-size:16pt; text-align:right;">
<strong>۲. انتقال سریع‌تر داده از دیسک به حافظه:</strong> الگوریتم‌های بازفشرده‌سازی مدرن آنقدر سریع هستند که زمان انتقال داده فشرده از دیسک و سپس بازفشرده‌سازی آن، معمولاً کمتر از زمان انتقال همان حجم داده به صورت فشرده‌نشده است. برای مثال، در یک سیستم جستجوی علمی مانند PubMed، بارگذاری لیست ارسال فشرده برای یک اصطلاح پزشکی رایج و سپس بازفشرده‌سازی آن، سریع‌تر از بارگذاری لیست ارسال فشرده‌نشده است.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 ویژگی‌های آماری واژه‌ها: رازهای آمار در خدمت فشرده‌سازی</p>
<p style="font-size:16pt; text-align:right;">
برای فهم بهتر فشرده‌سازی، ابتدا باید با ویژگی‌های آماری واژه‌ها آشنا شویم. در مجموعه‌های متنی بزرگ، توزیع واژه‌ها از الگوهای خاصی پیروی می‌کند که می‌توانیم از آن‌ها برای فشرده‌سازی بهتر استفاده کنیم.</p>
<p style="font-size:16pt; text-align:right;">
جدول زیر تأثیر پیش‌پردازش بر تعداد واژه‌ها و لیست‌های ارسال را در مجموعه Reuters-RCV1 نشان می‌دهد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>حذف اعداد: کاهش %۲ در واژگان و %۸ در لیست‌های ارسال</li>
<li>تبدیل حروف به حالت یکسان (case folding): کاهش %۱۷ در واژگان</li>
<li>حذف ۳۰ واژه رایج: کاهش %۱۴ در لیست‌های ارسال</li>
<li>حذف ۱۵۰ واژه رایج: کاهش %۳۰ در لیست‌های ارسال</li>
<li>ریشه‌یابی (stemming): کاهش %۱۷ دیگر در واژگان</li>
</ul>
<p style="font-size:16pt; text-align:right;">
نکته جالب این است که حذف واژه‌های رایج باعث کاهش چشمگیر در تعداد postings می‌شود، اما تأثیر زیادی بر اندازه ایندکس فشرده ندارد، چون لیست‌های ارسال این واژه‌ها پس از فشرده‌سازی بسیار کوچک می‌شوند.</p>
</div>

<div style="background:#d4edda; border-right:4px solid #155724; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#155724;">🔄 فشرده‌سازی بدون اتلاف در مقابل فشرده‌سازی با اتلاف</p>
<p style="font-size:16pt; text-align:right;">
در این فصل، ما بر روی فشرده‌سازی بدون اتلاف (lossless compression) تمرکز می‌کنیم، جایی که تمام اطلاعات حفظ می‌شود. اما فشرده‌سازی با اتلاف (lossy compression) نیز کاربردهای مهمی دارد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>فشرده‌سازی بدون اتلاف:</strong> تمام اطلاعات حفظ می‌شود. تکنیک‌هایی که در این فصل بررسی می‌شوند از این نوع هستند.</li>
<li><strong>فشرده‌سازی با اتلاف:</strong> بخشی از اطلاعات حذف می‌شود. تبدیل حروف به حالت یکسان، ریشه‌یابی، و حذف واژه‌های توقف نمونه‌هایی از این نوع هستند. همچنین مدل فضای برداری (فصل ۶) و تکنیک‌های کاهش ابعاد مانند نمایه‌سازی معنایی پنهان (فصل ۱۸) نیز نمونه‌هایی از فشرده‌سازی با اتلاف هستند.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در جستجوی وب، چون کاربران معمولاً فقط نتایج اول را می‌بینند، می‌توانیم از فشرده‌سازی با اتلاف استفاده کنیم بدون اینکه تأثیر منفی بر تجربه کاربر داشته باشیم.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📈 قانون هپس: چرا واژگان مجموعه‌های متنی آنقدر بزرگ هستند؟</p>
<p style="font-size:16pt; text-align:right;">
شاید فکر کنید زبان‌ها واژگان محدودی دارند. برای مثال، دومین نسخه دیکشنری آکسفورد بیش از ۶۰۰۰۰۰ کلمه تعریف می‌کند. اما واژگان اکثر مجموعه‌های بزرگ بسیار بزرگ‌تر از دیکشنری آکسفورد است. چرا؟</p>
<p style="font-size:16pt; text-align:right;">
دیکشنری آکسفورد شامل اکثر نام افراد، مکان‌ها، محصولات یا موجودات علمی مانند ژن‌ها نمی‌شود. این نام‌ها باید در ایندکس معکوس گنجانده شوند تا کاربران بتوانند آن‌ها را جستجو کنند.</p>
<p style="font-size:16pt; text-align:right;">
قانون هپس (Heaps' law) رابطه بین اندازه مجموعه (تعداد توکن‌ها) و اندازه واژگان (تعداد واژه‌های منحصربه‌فرد) را توصیف می‌کند. طبق این قانون، اندازه واژگان با افزایش اندازه مجموعه به صورت زیر رشد می‌کند:</p>
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px; width:auto; ">
M = k * T^b
</div>
<p style="font-size:16pt; text-align:right;">
که در آن M تعداد واژه‌های منحصربه‌فرد، T تعداد کل توکن‌ها، k و b ثابت‌هایی هستند که معمولاً k بین 10 تا 100 و b بین 0.4 تا 0.6 هستند.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، در مجموعه Reuters-RCV1، بهترین برازش برای این رابطه k = 44 و b = 0.49 است. این یعنی با دو برابر کردن اندازه مجموعه، تعداد واژه‌های منحصربه‌فرد حدود %40 افزایش می‌یابد، نه دو برابر!</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig5.1.bmp" alt="Heaps' Law" style="width:800px; border:1px solid #ccc; border-radius:12px;">
    <p style="font-size:14pt; color:#084C61; font-weight:bold; text-align:center;">شکل 5.1 – قانون هپس .واژگان به اندازه M و تابعی از مجموعه ای به اندازه T ( تعداد توکن ها) برای Reuters-RCV1  . برای این داده ها ، خط نقطه چین بهترین کوچکترین توان مناسب است.</p>
</div>
</div>

<p style="font-size:16pt; text-align:right;">
در ادامه این فصل، با تکنیک‌های فشرده‌سازی دیکشنری (روش دیکشنری به صورت رشته و ذخیره‌سازی بلوکی) و فشرده‌سازی لیست‌های ارسال (کدگذاری بایت متغیر و کدگذاری گاما) آشنا خواهیم شد. این تکنیک‌ها به ما کمک می‌کنند تا سیستم‌های بازیابی اطلاعات کارآمدتر و سریع‌تری بسازیم.</p>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section5.1.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۱.۱ قانون هیپس: چرا واژگان اینقدر سریع رشد می‌کنند؟</span> 📈
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌐 یک مثال از دنیای واقعی: رشد واژگان در اینستاگرام</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید مدیر داده در اینستاگرام هستید و می‌خواهید پیش‌بینی کنید چقدر فضا برای ذخیره هشتگ‌های جدید نیاز دارید. در ابتدا، اینستاگرام فقط چند هزار هشتگ داشت، اما امروزه میلیون‌ها هشتگ منحصربه‌فرد وجود دارد. قانون هیپس به ما کمک می‌کند این رشد را پیش‌بینی کنیم!</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای تخمین تعداد واژه‌های متمایز \(M\) در یک مجموعهٔ متنی با \(T\) توکن، از <strong>قانون هیپس</strong> استفاده می‌شود. این قانون رابطه‌ای تجربی بین اندازهٔ مجموعه و اندازهٔ واژگان ارائه می‌دهد:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
M = k \cdot T^b
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>\(T\): تعداد توکن‌ها در مجموعه</li>
<li>\(M\): تعداد واژه‌های متمایز</li>
<li>\(k\): ضریب وابسته به نوع مجموعه (بین ۳۰ تا ۱۰۰)</li>
<li>\(b\): نمایی که معمولاً حدود ۰٫۵ است</li>
</ul>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔍 چرا این قانون مهم است؟</p>
<p style="font-size:16pt; text-align:right;">
قانون هیپس نشان می‌دهد که رشد واژگان با افزایش حجم متن متوقف نمی‌شود. این یعنی برای سیستم‌های بزرگی مانند گوگل یا اینستاگرام، همیشه باید برای واژگان جدید فضا در نظر گرفت. این دقیقاً همان چیزی است که فشرده‌سازی دیکشنری را برای سیستم‌های بازیابی اطلاعات حیاتی می‌کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در مجموعهٔ <strong>Reuters-RCV1</strong>، برازش خطی زیر به‌دست آمده:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
\log_{10} M = 0.49 \cdot \log_{10} T + 1.64
$$
</div>

<p style="font-size:16pt; text-align:right;">
که معادل است با:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
M = 44 \cdot T^{0.49}
$$
</div>

<p style="font-size:16pt; text-align:right;">
مثال عددی: برای \(T = 1{,}000{,}020\) توکن، قانون هیپس پیش‌بینی می‌کند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
M = 44 \cdot (1{,}000{,}020)^{0.49} \approx 38{,}323
$$
</div>

<p style="font-size:16pt; text-align:right;">
در حالی که مقدار واقعی برابر با 38,365 واژه بوده — که بسیار نزدیک به پیش‌بینی است.</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🧩 عوامل مؤثر بر پارامترها</p>
<p style="font-size:16pt; text-align:right;">
مقدار \(k\) وابسته به نوع مجموعه و نحوهٔ پیش‌پردازش آن است:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>استفاده از case folding و stemming باعث کاهش نرخ رشد واژگان می‌شود.</li>
<li>وجود اعداد، اشتباهات املایی و اسامی خاص باعث افزایش واژگان می‌شود.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
نتیجه‌گیری مهم:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>اندازهٔ واژگان با افزایش اسناد همچنان رشد می‌کند و به سقف نمی‌رسد.</li>
<li>برای مجموعه‌های بزرگ، اندازهٔ دیکشنری بسیار زیاد خواهد بود.</li>
<li>بنابراین، <strong>فشرده‌سازی دیکشنری</strong> برای کارایی سیستم بازیابی اطلاعات ضروری است.</li>
</ul>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۱.۲ قانون زیپف: چرا برخی کلمات بسیار بیشتر از بقیه تکرار می‌شوند؟</span> 📊
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🐦 یک مثال از دنیای واقعی: تحلیل توییتهای توییتر</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید محققان توییتر (X) می‌خواهند بفهمند کلمات در توییتهای فارسی چگونه توزیع شده‌اند. آنها متوجه می‌شوند که کلمه «از» در میلیون‌ها توییت ظاهر می‌شود، در حالی که کلمه «الگوریتم» بسیار نادر است. این توزیع ناهمگون چیزی است که قانون زیپف توصیف می‌کند!</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای طراحی الگوریتم‌های فشرده‌سازی لیست‌های ارسال، باید بدانیم واژه‌ها چگونه در مجموعه اسناد توزیع شده‌اند. یکی از مدل‌های رایج برای این هدف، <strong>قانون زیپف</strong> است.</p>

<p style="font-size:16pt; text-align:right;">
این قانون بیان می‌کند که اگر واژه \(t_1\) رایج‌ترین واژه باشد، \(t_2\) دومین واژه رایج، و همین‌طور تا \(t_i\)، آنگاه فراوانی مجموعه‌ای (collection frequency) واژه \(t_i\) به‌صورت زیر است:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
cf_i \propto \frac{1}{i}
$$
</div>

<p style="font-size:16pt; text-align:right;">
به زبان ساده:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>اگر رایج‌ترین واژه \(cf_1\) بار ظاهر شود</li>
<li>واژه دوم تقریباً \(cf_1 / 2\) بار</li>
<li>واژه سوم تقریباً \(cf_1 / 3\) بار</li>
<li>و همین‌طور الگو ادامه می‌یابد...</li>
</ul>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔍 اهمیت عملی قانون زیپف</p>
<p style="font-size:16pt; text-align:right;">
در تحلیل داده‌های توییتر، محققان متوجه شدند کلماتی مانند «از»، «به» و «در» تقریباً در هر توییت ظاهر می‌شوند، در حالی که کلمات تخصصی مانند «الگوریتم» یا «فشرده‌سازی» بسیار نادر هستند. این توزیع ناهمگون به ما کمک می‌کند تا الگوریتم‌های فشرده‌سازی بهینه طراحی کنیم.</p>
</div>

<p style="font-size:16pt; text-align:right;">
فرمول زیپف را می‌توان به‌صورت نمایی یا لگاریتمی نیز نوشت:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
cf_i = c \cdot i^k \quad \text{with } k = -1
$$
</div>

<p style="font-size:16pt; text-align:right;">
یا به‌صورت لگاریتمی:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
\log cf_i = \log c + k \cdot \log i
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن \(c\) یک ثابت است و \(k = -1\). این رابطه یک <strong>قانون توانی (power law)</strong> است که در بسیاری از پدیده‌های طبیعی و اجتماعی مشاهده می‌شود.</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/fig5.2.bmp" alt="Zipf's law for Reuters-RCV1" style="width:800px; border:1px solid #ccc; border-radius:12px;">
</div>

<p style="font-size:16pt; text-align:right;">
نمودار بالا برای مجموعه <strong>Reuters-RCV1</strong>، فراوانی واژه‌ها را بر حسب رتبه‌شان نشان می‌دهد. خط با شیب <span dir='ltr'>-1</span> تقریب خوبی از داده‌ها ارائه می‌دهد، هرچند کاملاً دقیق نیست.</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🧩 کاربرد در فشرده‌سازی</p>
<p style="font-size:16pt; text-align:right;">
قانون زیپف به ما می‌گوید که:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>بیشتر واژه‌ها بسیار کم‌تکرار هستند</li>
<li>فشرده‌سازی لیست‌های ارسال برای واژه‌های رایج بسیار مؤثر است</li>
<li>می‌توانیم الگوریتم‌های متفاوتی برای واژه‌های پرتکرار و کم‌تکرار طراحی کنیم</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
در بخش 5.3، خواهیم دید که چگونه این درک از توزیع واژه‌ها به طراحی الگوریتم‌های فشرده‌سازی بهینه کمک می‌کند.</p>

</div>



<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section5.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۲ فشرده‌سازی دیکشنری: چرا حتی بخش کوچک هم مهم است؟</span> 📘
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🛒 یک مثال از دنیای واقعی: جستجوی محصولات در آمازون</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید مدیر موتور جستجوی آمازون هستید. هر ثانیه، هزاران کاربر در حال جستجوی محصولات هستند. وقتی کاربری عبارت "لپ‌تاپ گیمینگ" را جستجو می‌کند، سیستم باید فوراً واژه‌های "لپ‌تاپ" و "گیمینگ" را در دیکشنری پیدا کند. اگر دیکشنری در حافظه اصلی نباشد، هر جستجو نیازمند چندین دسترسی به دیسک خواهد بود که باعث تأخیر قابل توجهی می‌شود. در دنیای رقابتی تجارت الکترونیک، حتی تأخیر ۱۰۰ میلی‌ثانیه‌ای می‌تواند منجر به از دست رفتن مشتری شود!</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نگاه اول، دیکشنری نسبت به فایل لیست‌های ارسال (postings file) بسیار کوچک‌تر است. پس چرا باید آن را فشرده کنیم؟ پاسخ در <strong>کارایی سیستم بازیابی اطلاعات</strong> نهفته است.</p>

<p style="font-size:16pt; text-align:right;">
یکی از عوامل اصلی در زمان پاسخ‌دهی سیستم IR، تعداد <strong>disk seek</strong>هایی است که برای پردازش یک پرسمان انجام می‌شود. اگر بخشی از دیکشنری روی دیسک باشد، هر جست‌وجو نیازمند چندین seek خواهد بود. بنابراین، هدف اصلی فشرده‌سازی دیکشنری این است که بتوان آن را به‌طور کامل (یا حداقل بخش بزرگی از آن) در حافظهٔ اصلی نگه داشت.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🚀 مزایای فشرده‌سازی دیکشنری در دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
در شرکت‌های بزرگی مانند گوگل یا مایکروسافت که باید میلیاردها سند را ایندکس کنند، فشرده‌سازی دیکشنری به آنها اجازه می‌دهد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>کاهش تعداد disk seek در زمان جست‌وجو (هر seek معادل چند میلی‌ثانیه تأخیر است)</li>
<li>افزایش throughput سیستم در پردازش پرسمان‌ها (مهم برای سرویس‌هایی با ترافیک بالا)</li>
<li>امکان اجرای سیستم روی سخت‌افزارهای محدود (مثل موبایل یا کامپیوترهای جاسازی‌شده در خودروها)</li>
<li>کاهش زمان راه‌اندازی سیستم (مهم برای سرویس‌های ابری که باید سریعاً راه‌اندازی شوند)</li>
<li>هم‌زیستی بهتر با نرم‌افزارهای دیگر در حافظه (در محیط‌های اشتراکی)</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
شکل زیر یک ساختار ساده برای ذخیره‌سازی دیکشنری را نشان می‌دهد:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
<table style="margin:auto; border-collapse:collapse; font-size:14pt;">
<thead>
<tr style="background:#e0e0e0;">
<th>term</th>
<th>document frequency</th>
<th>pointer to postings list</th>
</tr>
</thead>
<tbody>
<tr><td>a</td><td>656,265</td><td>→</td></tr>
<tr><td>aachen</td><td>65</td><td>→</td></tr>
<tr><td>...</td><td>...</td><td>...</td></tr>
<tr><td>zulu</td><td>221</td><td>→</td></tr>
</tbody>
</table>
</div>

<p style="font-size:16pt; text-align:right;">
در این ساختار، هر ورودی به‌صورت <strong>fixed-width entry</strong> ذخیره می‌شود. به‌طور مثال:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>term: 20 bytes</li>
<li>document frequency: 4 bytes</li>
<li>pointer to postings list: 4 bytes</li>
</ul>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📊 چالش در دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
برای یک شرکت بزرگ بین‌المللی که اسناد را به زبان‌های مختلف ایندکس می‌کند، دیکشنری می‌تواند به راحتی از حافظه یک کامپیوتر استاندارد فراتر رود. در چنین مواردی، فشرده‌سازی دیکشنری نه یک انتخاب، بلکه یک ضرورت فنی است. در ادامه این بخش، با روش‌های مختلف فشرده‌سازی دیکشنری آشنا می‌شویم که به ما کمک می‌کنند تا حتی در محیط‌های محدود نیز سیستم‌های جستجوی کارآمد داشته باشیم.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section5.2.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۲.۱ دیکشنری به‌صورت رشته‌ای: هوشمندانه‌تر استفاده از فضا</span> 🧵
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 یک مثال از دنیای واقعی: بهینه‌سازی جستجو در اسناد پزشکی</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در شرکت IBM Watson Health کار می‌کنید و در حال طراحی سیستم جستجوی پزشکی هستید که به پزشکان کمک می‌کند در میلیون‌ها مقاله پزشکی جستجو کنند. در این حوزه، واژه‌های تخصصی و طولانی مانند "hydrochlorofluorocarbons" یا "immunohistochemistry" بسیار رایج هستند. اگر از روش با طول ثابت استفاده کنید، بسیاری از این واژه‌ها کوتاه نمی‌آیند و فضای زیادی هدر می‌رود. اینجاست که روش رشته‌ای به کمک شما می‌آید!</p>
</div>

<p style="font-size:16pt; text-align:right;">
ساده‌ترین ساختار برای ذخیره‌سازی دیکشنری، استفاده از آرایه‌ای از ورودی‌های با طول ثابت (fixed-width) است. در این روش، واژگان به‌صورت مرتب‌شدهٔ الفبایی ذخیره می‌شوند و هر ورودی شامل موارد زیر است:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>term: 20 bytes</li>
<li>document frequency: 4 bytes</li>
<li>pointer to postings list: 4 bytes</li>
</ul>

<p style="font-size:16pt; text-align:right;">
برای مجموعهٔ <strong>Reuters-RCV1</strong> با 400,000 واژه، حافظهٔ موردنیاز برابر است با:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
400{,}000 \times (20 + 4 + 4) = 11.2 \text{ MB}
$$
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">⚠️ مشکلات روش با طول ثابت</p>
<p style="font-size:16pt; text-align:right;">
این روش بسیار ناکارآمد است، چون:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>طول متوسط واژه‌ها در انگلیسی حدود ۸ کاراکتر است ← ۱۲ بایت هدر در هر ورودی</li>
<li>واژه‌های طولانی‌تر از ۲۰ کاراکتر قابل ذخیره‌سازی نیستند</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در سیستم جستجوی پزشکی Watson، این مشکل جدی‌تر است، چون بسیاری از واژه‌های پزشکی بسیار طولانی هستند و نمی‌توانند در ۲۰ کاراکتر جای بگیرند!</p>
</div>

<p style="font-size:16pt; text-align:right;">
راه‌حل: ذخیره‌سازی واژه‌ها به‌صورت یک رشتهٔ طولانی از کاراکترها (dictionary-as-a-string). در این روش:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>همهٔ واژه‌ها پشت سر هم در یک رشته ذخیره می‌شوند</li>
<li>هر واژه با یک <strong>term pointer</strong> مشخص می‌شود (آدرس شروع واژه)</li>
<li>جست‌وجو با <strong>binary search</strong> روی جدول اشاره‌گرها انجام می‌شود</li>
</ul>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🧮 محاسبه فضای ذخیره‌سازی</p>
<p style="font-size:16pt; text-align:right;">
برای 400,000 واژه با طول متوسط ۸ کاراکتر، تعداد موقعیت‌های ممکن برابر است با:</p>
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
400{,}000 \times 8 = 3.2 \times 10^6
$$
</div>
<p style="font-size:16pt; text-align:right;">
برای آدرس‌دهی این تعداد موقعیت، به اشاره‌گرهایی با طول زیر نیاز داریم:</p>
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
\log_2(3.2 \times 10^6) \approx 22 \text{ bits} \Rightarrow 3 \text{ bytes}
$$
</div>
</div>

<p style="font-size:16pt; text-align:right;">
در این ساختار جدید، حافظهٔ موردنیاز برابر است با:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>term: 8 bytes (میانگین)</li>
<li>term pointer: 3 bytes</li>
<li>document frequency: 4 bytes</li>
<li>postings pointer: 4 bytes</li>
</ul>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
400{,}000 \times (8 + 3 + 4 + 4) = 7.6 \text{ MB}
$$
</div>

<p style="font-size:16pt; text-align:right;">
بنابراین، با استفاده از dictionary-as-a-string حدود <strong>%۳۳ صرفه‌جویی</strong> در حافظه نسبت به روش fixed-width خواهیم داشت.</p>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 نحوه کارکرد این روش</p>
<p style="font-size:16pt; text-align:right;">
در روش dictionary-as-a-string، واژه‌ها به صورت یک رشته پیوسته ذخیره می‌شوند و اشاره‌گرها ابتدای هر واژه را مشخص می‌کنند. برای مثال، اگر سه واژه "systile"، "syzygetic" و "syzygial" را داشته باشیم، در رشته به این شکل ذخیره می‌شوند:</p>
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
...sys t i l e s y z y g e t i c s y z y g i a l...
</div>
<p style="font-size:16pt; text-align:right;">
اشاره‌گر اول به ابتدای "systile"، اشاره‌گر دوم به ابتدای "syzygetic" و اشاره‌گر سوم به ابتدای "syzygial" اشاره می‌کند. این ساختار به ما اجازه می‌دهد تا با جستجوی دودویی در جدول اشاره‌گرها، واژه مورد نظر را به سرعت پیدا کنیم.</p>
</div>

</div>



<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section5.2.2" style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۲.۲ فشرده‌سازی بلوکی و کدگذاری پیشوندی</span> 📦
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📱 یک مثال از دنیای واقعی: بهینه‌سازی جستجو در اپلیکیشن موبایل</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم توسعه اپلیکیشن Spotify کار می‌کنید و مسئول بهینه‌سازی جستجوی آهنگ‌ها در نسخه موبایل هستید. اپلیکیشن شما باید بتواند نام صدها هزار هنرمند، آلبوم و آهنگ را در حافظه محدود موبایل ذخیره کند و همزمان جستجوی سریعی را ارائه دهد. فشرده‌سازی بلوکی و کدگذاری پیشوندی به شما اجازه می‌دهد تا دیکشنری بزرگ را در حافظه محدود موبایل جا دهید و همزمان سرعت جستجو را قابل قبول نگه دارید.</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای کاهش بیشتر اندازه دیکشنری، واژه‌ها را به بلوک‌هایی با اندازه \(k\) تقسیم می‌کنیم و فقط برای واژه اول هر بلوک یک اشاره‌گر ذخیره می‌کنیم. طول هر واژه نیز به‌صورت یک بایت در ابتدای آن ذخیره می‌شود.</p>

<p style="font-size:16pt; text-align:right;">
برای \(k = 4\):</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>صرفه‌جویی در اشاره‌گرها: \((k - 1) \times 3 = 9\) bytes</li>
<li>هزینه اضافه برای طول واژه‌ها: \(k = 4\) bytes</li>
<li>صرفه‌جویی خالص: \(5\) bytes per block</li>
<li>تعداد بلوک‌ها: \(400{,}000 / 4 = 100{,}000\)</li>
<li>صرفه‌جویی کل: \(100{,}000 \times 5 = 0.5 \text{ MB}\)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
حافظه نهایی: \(7.6 - 0.5 = 7.1 \text{ MB}\)</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">⚖️ معامله بین فضا و سرعت</p>
<p style="font-size:16pt; text-align:right;">
افزایش \(k\) باعث کاهش بیشتر حافظه می‌شود، اما زمان جست‌وجو افزایش می‌یابد. در اپلیکیشن Spotify، این معامله بسیار مهم است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>جست‌وجو در دیکشنری بدون فشرده‌سازی: میانگین 1.6 مرحله</li>
<li>جست‌وجو با \(k = 4\): میانگین 2 مرحله ← افزایش %25</li>
<li>جست‌وجو با \(k = 8\) یا \(k = 16\): کندتر خواهد شد و تجربه کاربری را تحت تأثیر قرار می‌دهد</li>
</ul>
<p style="font-size:16pt; text-align:right;">برای اپلیکیشن موبایل که باید پاسخگو باشد، تیم Spotify احتمالاً از \(k = 4\) استفاده می‌کند تا تعادل خوبی بین فضا و سرعت برقرار شود.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔤 کدگذاری پیشوندی (Front Coding)</p>
<p style="font-size:16pt; text-align:right;">تکنیک <strong>front coding</strong> از اشتراک پیشوند بین واژه‌های متوالی استفاده می‌کند. در اپلیکیشن Spotify، بسیاری از نام هنرمندان با پیشوندهای مشترک شروع می‌شوند:</p>
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
<pre>
9The Beatles
10The Rolling Stones
8The Doors
8The Who
⇓
3The∗
6⋄Beatles
12⋄Rolling Stones
5⋄Doors
3⋄Who
</pre>
</div>
<p style="font-size:16pt; text-align:right;">در این روش، پیشوند مشترک با ∗ علامت‌گذاری می‌شود و ادامه واژه‌ها با ⋄ و طول مشخص می‌شوند. این تکنیک در مجموعه Reuters باعث صرفه‌جویی <span dir='ltr'>1.2 MB</span> دیگر شد.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 مقایسه نهایی ساختارهای دیکشنری</p>
<p style="font-size:16pt; text-align:right;">جدول زیر مقایسه‌ای از فضا مورد نیاز برای روش‌های مختلف فشرده‌سازی دیکشنری را نشان می‌دهد:</p>
<table style="margin:auto; border-collapse:collapse; font-size:14pt; text-align:center;">
<thead>
<tr style="background:#e0e0e0;">
<th>ساختار</th>
<th>حافظه (MB)</th>
</tr>
</thead>
<tbody>
<tr><td>Fixed-width</td><td>11.2</td></tr>
<tr><td dir='ltr'>Term pointers into string</td><td>7.6</td></tr>
<tr><td dir='ltr'>Blocking (k = 4)</td><td>7.1</td></tr>
<tr><td dir='ltr'>Blocking + Front coding</td><td>5.9</td></tr>
</tbody>
</table>
<p style="font-size:16pt; text-align:right;">همانطور که می‌بینید، با ترکیب این تکنیک‌ها می‌توانیم تقریباً %50 در فضا صرفه‌جویی کنیم، که برای اپلیکیشن‌های موبایل با حافظه محدود بسیار حیاتی است.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section5.3" style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۳ فشرده‌سازی فایل لیست‌های ارسال: هنر ذخیره هوشمندانه</span> 🧮
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 یک مثال از دنیای واقعی: بهینه‌سازی جستجو در گوگل</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید مهندس گوگل هستید و مسئول بهینه‌سازی ایندکس جستجو برای میلیاردها صفحه وب هستید. هر روز، میلیون‌ها صفحه جدید به ایندکس اضافه می‌شود و کاربران میلیاردها جستجو انجام می‌دهند. اگر هر لیست ارسال را بدون فشرده‌سازی ذخیره کنید، به هزاران سرور با حافظه عظیم نیاز خواهید داشت! فشرده‌سازی لیست‌های ارسال به گوگل اجازه می‌دهد تا با منابع کمتر، پاسخ‌های سریع‌تری به کاربران ارائه دهد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این بخش، هدف ما کاهش اندازه فایل لیست‌های ارسال (postings file) است. برای مجموعه <strong>Reuters-RCV1</strong>:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>تعداد اسناد: 800,000</li>
<li>تعداد توکن‌ها در هر سند: 200</li>
<li>طول هر توکن: 6 کاراکتر</li>
<li>تعداد کل postings: 100,000,000</li>
</ul>

<p style="font-size:16pt; text-align:right;">
بنابراین:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>اندازه مجموعه: \(800{,}000 \times 200 \times 6 = 960 \text{ MB}\)</li>
<li>اندازه فایل لیست‌های ارسال بدون فشرده‌سازی: \(100{,}000{,}000 \times \frac{20}{8} = 250 \text{ MB}\)</li>
</ul>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">💡 ایده کلیدی: کدگذاری فاصله‌ها (Gap Encoding)</p>
<p style="font-size:16pt; text-align:right;">
برای کاهش این حجم، از ایده <strong>gap encoding</strong> استفاده می‌کنیم. به‌جای ذخیره مستقیم docIDها، فاصله بین آن‌ها را ذخیره می‌کنیم:</p>

<table style="margin:auto; border-collapse:collapse; font-size:14pt; text-align:center;">
<thead>
<tr style="background:#e0e0e0;">
<th>term</th>
<th>docIDs</th>
<th>gaps</th>
</tr>
</thead>
<tbody>
<tr><td>the</td><td>283042, 283043, 283044, 283045</td><td>1, 1, 1</td></tr>
<tr><td>computer</td><td>283047, 283154, 283159, 283202</td><td>107, 5, 43</td></tr>
<tr><td>arachnocentric</td><td>252000, 500100</td><td>252000, 248100</td></tr>
</tbody>
</table>

<p style="font-size:16pt; text-align:right;">
برای واژه‌های رایج مانند "the"، فاصله‌ها بسیار کوچک هستند (معمولاً ۱) و می‌توان آن‌ها را با بیت‌های بسیار کمتر ذخیره کرد. اما برای واژه‌های نادر مانند "arachnocentric"، فاصله‌ها بزرگ هستند و همچنان نیاز به بیت‌های کامل دارند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔧 روش‌های فشرده‌سازی</p>
<p style="font-size:16pt; text-align:right;">
برای فشرده‌سازی این فاصله‌ها، از دو نوع روش استفاده می‌کنیم:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>Bytewise Compression:</strong> فشرده‌سازی با حداقل تعداد بایت</li>
<li><strong>Bitwise Compression:</strong> فشرده‌سازی با حداقل تعداد بیت</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در گوگل، هر دو روش استفاده می‌شود. برای واژه‌های بسیار رایج که فاصله‌های کوچکی دارند، روش bitwise بسیار کارآمد است، زیرا می‌تواند فاصله ۱ را تنها با یک بیت ذخیره کند. برای واژه‌های کمتر رایج، روش bytewise ممکن است مناسب‌تر باشد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در بخش‌های بعدی، دو الگوریتم اصلی از هر دسته را بررسی و پیاده‌سازی می‌کنیم:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>Variable Byte Encoding (bytewise)</li>
<li>Gamma Encoding (bitwise)</li>
</ul>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 اهمیت عملی</p>
<p style="font-size:16pt; text-align:right;">
در سیستم‌های بزرگ مانند گوگل یا آمازون، فشرده‌سازی لیست‌های ارسال فقط یک بهینه‌سازی کوچک نیست، بلکه یک ضرورت فنی است. بدون این تکنیک‌ها، هزینه ذخیره‌سازی و پردازش ایندکس‌ها به قدری بالا می‌رود که ارائه خدمات جستجوی سریع و مقرون‌به‌صرفه غیرممکن می‌شود. با فشرده‌سازی هوشمندانه، این شرکت‌ها می‌توانند میلیاردها سند را با هزینه کمتر ایندکس کرده و پاسخ‌های سریع‌تری به کاربران ارائه دهند.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section5.3.1" style="font-size:24pt; font-weight:bold; color:#1B263B;">۵.۳.۱ کدگذاری بایتی متغیر: هنر ذخیره هوشمندانه اعداد</span> 🧩
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📱 یک مثال از دنیای واقعی: بهینه‌سازی جستجو در واتس‌اپ</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در تیم واتس‌اپ کار می‌کنید و مسئول بهینه‌سازی جستجوی پیام‌ها هستید. هر روز میلیاردها پیام ارسال می‌شود و کاربران باید بتوانند سریعاً پیام‌های قدیمی خود را پیدا کنند. اگر هر شناسه پیام را با ۴ بایت ذخیره کنید، حجم ایندکس به سرعت افزایش می‌یابد. با استفاده از کدگذاری بایتی متغیر، واتس‌اپ می‌تواند شناسه‌های پیام‌های نزدیک به هم را با فضای بسیار کمتری ذخیره کند و همزمان سرعت جستجو را بالا نگه دارد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در روش کدگذاری بایتی متغیر (Variable Byte Encoding)، هر عدد (مثلاً فاصله بین docIDها) با تعداد متغیری از بایت‌ها ذخیره می‌شود. این روش از یک ایده ساده اما هوشمندانه استفاده می‌کند:</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔍 ساختار یک بایت در کدگذاری VB</p>
<p style="font-size:16pt; text-align:right;">
هر بایت در این روش شامل دو بخش است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>۷ بیت داده (payload): بخشی از عدد را ذخیره می‌کند</li>
<li>۱ بیت ادامه (continuation bit): نشان می‌دهد آیا این بایت آخرین بایت عدد است یا خیر</li>
</ul>
<p style="font-size:16pt; text-align:right;">اگر continuation bit برابر ۱ باشد ← این بایت آخرین بایت عدد است</p>
<p style="font-size:16pt; text-align:right;">اگر continuation bit برابر ۰ باشد ← عدد در بایت‌های بعدی ادامه دارد</p>
</div>


<p style="font-size:16pt; text-align:right;">
برای مثال، عدد 214577 به صورت زیر کدگذاری می‌شود:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
214577 = 00001101 00001100 10110001  
→ 3 bytes: [13, 12, 177]  
→ continuation bits: 0, 0, 1
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔄 فرآیند کدگذاری و رمزگشایی</p>
<p style="font-size:16pt; text-align:right;">برای کدگذاری یک عدد:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>عدد را به مبنای 128 می‌بریم (چون 7 بیت برای داده داریم)</li>
<li>بایت‌ها را از پایین به بالا تولید می‌کنیم</li>
<li>به آخرین بایت، 128 اضافه می‌کنیم تا continuation bit آن 1 شود</li>
</ol>
<p style="font-size:16pt; text-align:right;">برای رمزگشایی:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>بایت‌ها را یکی‌یکی می‌خوانیم</li>
<li>تا زمانی که continuation bit صفر است، عدد را با ضرب در 128 ادامه می‌دهیم</li>
<li>وقتی continuation bit برابر ۱ شد، عدد کامل شده است</li>
</ol>
</div>

Extend یک لیست را بسط می دهد. به عنوان مثال : <span dir='ltr'>Extend(<1,2>,<3,4>) = <1,2,3,4></span> .</p>

<p style="font-size:16pt; text-align:right;">در آزمایش روی مجموعه Reuters-RCV1، اندازه فایل postings از 250MB به 116MB کاهش یافت — یعنی بیش از %۵۰ صرفه‌جویی در فضا.</p>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚖️ مزایا و معایب</p>
<p style="font-size:16pt; text-align:right;">کدگذاری VB تعادلی عالی بین سرعت و فشرده‌سازی ارائه می‌دهد:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>مزیت:</strong> پیاده‌سازی ساده، سرعت رمزگشایی بالا، و فشرده‌سازی قابل قبول</li>
<li><strong>معایب:</strong> برای اعداد بسیار بزرگ (مانند فاصله‌های بزرگ در واژه‌های نادر) کارایی کمتری دارد</li>
</ul>
<p style="font-size:16pt; text-align:right;">برای اکثر سیستم‌های بازیابی اطلاعات، این روش یک انتخاب عالی است. اما اگر فضا بسیار محدود است، می‌توان از روش‌های سطح بیت (مانند γ codes) که در بخش بعدی معرفی می‌شوند، استفاده کرد.</p>
</div>

</div>

```python
def vb_encode_number(n):
    """
    این تابع یک عدد را با استفاده از روش کدگذاری بایتی متغیر کدگذاری می‌کند.
    هر بایت شامل 7 بیت داده و 1 بیت ادامه (continuation bit) است.
    بیت ادامه برای تمام بایت‌ها به جز آخرین بایت تنظیم می‌شود (MSB=1).
    """
    # حالت خاص برای عدد صفر، یک بایت با مقدار صفر برمی‌گرداند
    if n == 0:
        return [0]

    bytes_ = []
    while True:
        # استخراج 7 بیت کم اهمیت
        byte = n & 0x7F
        n >>= 7
        
        # اگر هنوز بیت‌های باقی‌مانده وجود دارد، بیت ادامه را تنظیم کن
        if n > 0:
            byte |= 0x80
            
        bytes_.append(byte)
        if n == 0:
            break
            
    return bytes_

# تابع کدگذاری لیستی از اعداد (مثلاً gapها)
def vb_encode(numbers):
    """
    این تابع لیستی از اعداد را با استفاده از روش کدگذاری بایتی متغیر کدگذاری می‌کند.
    """
    bytestream = []
    for n in numbers:
        bytestream.extend(vb_encode_number(n))
    return bytestream

#  تابع رمزگشایی لیست بایت‌ها به لیست اعداد 
def vb_decode(bytestream):
    """
    این تابع لیست بایت‌های کدگذاری شده را به لیست اعداد اصلی بازمی‌گرداند.
    """
    numbers = []
    n = 0
    shift = 0  # شیفت برای قرار دادن بایت‌ها در موقعیت صحیح
    
    for byte in bytestream:
        # استخراج 7 بیت داده
        data = byte & 0x7F
        # قرار دادن بیت‌ها در موقعیت صحیح
        n |= (data << shift)
        
        # Fixed: اگر بیت ادامه تنظیم نشده باشد (byte < 128)، این آخرین بایت عدد است
        if byte < 128:
            numbers.append(n)
            n = 0
            shift = 0
        else:
            # در غیر این صورت، برای بایت بعدی شیفت را 7 واحد افزایش می‌دهیم
            shift += 7
            
    return numbers

# تابع محاسبه حجم قبل و بعد از فشرده‌سازی
def compare_space(gaps):
    """
    این تابع حجم لیست gapها را قبل و بعد از فشرده‌سازی مقایسه می‌کند.
    """
    # حجم قبل از فشرده‌سازی (هر عدد 4 بایت)
    original_size = len(gaps) * 4
    
    # حجم بعد از فشرده‌سازی
    encoded = vb_encode(gaps)
    compressed_size = len(encoded)
    
    return original_size, compressed_size

# مثال با داده‌های واقعی از کتاب
gaps = [5, 214577]
encoded = vb_encode(gaps)
decoded = vb_decode(encoded)

print("Gaps:", gaps)
print("Encoded bytes:", encoded)
print("Decoded:", decoded)
print()

# مقایسه حجم قبل و بعد از فشرده‌سازی
original_size, compressed_size = compare_space(gaps)
print(f"Original size: {original_size} bytes")
print(f"Compressed size: {compressed_size} bytes")
print(f"Compression ratio: {compressed_size/original_size:.2f}")

# مثال با لیست بزرگتر از gapها
large_gaps = [1, 1, 1, 1, 1, 107, 5, 43, 252000, 248100]
original_size, compressed_size = compare_space(large_gaps)
print()
print("Large gaps example:")
print(f"Original size: {original_size} bytes")
print(f"Compressed size: {compressed_size} bytes")
print(f"Compression ratio: {compressed_size/original_size:.2f}")
```

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 یک مثال از دنیای واقعی: بهینه‌سازی جستجو در گوگل</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در گوگل کار می‌کنید و مسئول بهینه‌سازی ایندکس وب هستید. گوگل باید میلیاردها صفحه وب را ایندکس کند و برای هر کلمه، لیستی از شناسه‌های صفحات (docID) را ذخیره کند. برای کلمه‌ای مانند "HTML" که در صفحات زیادی وجود دارد، لیست docIDها بسیار طولانی است. اگر هر docID را با ۳۲ بیت ذخیره کنید، حجم ایندکس به ترابایت‌ها می‌رسد! با استفاده از کدهای گاما، گوگل می‌تواند فاصله بین docIDها را با بیت‌های بسیار کمتر ذخیره کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در روش کدگذاری گاما (γ encoding)، هر عدد (مثلاً فاصله بین docIDها) به دو بخش تقسیم می‌شود:</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧩 ساختار کد گاما</p>
<p style="font-size:16pt; text-align:right;">
هر عدد \(G\) به دو بخش تقسیم می‌شود:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>Offset:</strong> نمایش دودویی \(G\) بدون بیت اول (مثلاً برای ۱۳ ← ۱۱۰۱ ← ۱۰۱)</li>
<li><strong>Length:</strong> طول offset به‌صورت کد unary (برای ۳ بیت ← ۱۱۱۰)</li>
</ul>
<p style="font-size:16pt; text-align:right;">
کد نهایی: اتصال length و offset ← برای ۱۳ می‌شود: <code>۱۱۱۰۱۰۱</code></p>
</div>

<p style="font-size:16pt; text-align:right;">
بیایید فرآیند کدگذاری و رمزگشایی را با یک مثال واقعی ببینیم. فرض کنید در ایندکس گوگل، docIDهای زیر برای کلمه "machine learning" وجود دارند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
docIDs: 824, 829, 215406<br>
gaps: 5, 214577
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔄 فرآیند کدگذاری عدد ۲۱۴۵۷۷:</p>
<p style="font-size:16pt; text-align:right;">
۱. عدد ۲۱۴۵۷۷ به دودویی تبدیل می‌شود: ۱۱۰۱۰۰۰۱۱۰۱۰۰۰۱<br>
۲. بیت اول حذف می‌شود: ۱۰۱۰۰۰۱۱۰۱۰۰۰۱ (offset)<br>
۳. طول offset محاسبه می‌شود: ۱۵ بیت<br>
۴. طول ۱۵ به صورت unary نوشته می‌شود: ۱۵ بار ۱ و سپس ۰ ← ۱۱۱۱۱۱۱۱۱۱۱۱۱۱۱۰<br>
۵. کد نهایی: ۱۱۱۱۱۱۱۱۱۱۱۱۱۱۱۰ + ۱۰۱۰۰۰۱۱۰۱۰۰۰۱ = ۱۱۱۱۱۱۱۱۱۱۱۱۱۱۱۰۱۰۱۰۰۰۱۱۰۱۰۰۰۱
</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای رمزگشایی:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>ابتدا unary code تا رسیدن به ۰ می‌خوانیم (۱۵ بار ۱ و سپس ۰) ← طول offset = ۱۵ بیت</li>
<li>سپس ۱۵ بیت بعدی را می‌خوانیم (offset)</li>
<li>یک ۱ به ابتدای offset اضافه می‌کنیم ← عدد اصلی</li>
</ol>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 ویژگی‌های برجسته کدهای گاما</p>
<p style="font-size:16pt; text-align:right;">
کدهای گاما سه ویژگی کلیدی دارند که آنها را برای فشرده‌سازی ایندکس ایده‌آل می‌کند:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>Universal:</strong> برای هر توزیع احتمالاتی، نزدیک به کد بهینه است. این یعنی بدون اینکه توزیع فاصله‌ها را بدانیم، می‌توانیم از این کد استفاده کنیم.</li>
<li><strong>Prefix-free:</strong> هیچ کد گاما پیشوند دیگری نیست. این ویژگی رمزگشایی را ساده می‌کند، چون نیازی به جداکننده بین کدها نیست.</li>
<li><strong>Parameter-free:</strong> نیازی به تنظیم پارامتر ندارد. این ویژگی برای ایندکس‌های پویا که توزیع فاصله‌ها ممکن است تغییر کند، بسیار مهم است.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
طول کد γ برای عدد \(G\) برابر است با:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
\text{length} = 2 \cdot \lfloor \log_2 G \rfloor + 1
$$
</div>

<p style="font-size:16pt; text-align:right;">
این کد همیشه طولی فرد دارد و در بدترین حالت، حداکثر دو برابر حد پایین اطلاعاتی است.</p>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🎯 کارایی در دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
در مجموعه Reuters-RCV1، کدگذاری گاما باعث کاهش اندازه فایل postings از ۲۵۰MB به ۱۰۱MB شد — حدود %۱۵ بهتر از Variable Byte Encoding. این صرفه‌جویی در مقیاس وب، که گوگل باید میلیاردها صفحه را ایندکس کند، به ترابایت‌ها فضا می‌رسد!</p>
<p style="font-size:16pt; text-align:right;">
برای تخمین اندازه ایندکس فشرده‌شده با γ، از مدل Zipf و فرمول زیر استفاده می‌شود:</p>
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:16pt; margin:15px 0;">
$$
\text{total bits} \approx \sum_{j=1}^{M/L_c} 2NL_c \cdot \frac{\log_2 j}{j}
$$
</div>
<p style="font-size:16pt; text-align:right;">
که برای \(M = 400{,}000\) و \(L_c = 15\) تقریباً برابر با ۲۲۴MB می‌شود — اما مقدار واقعی ۱۰۱MB است چون فرض‌های مدل ساده‌سازی شده‌اند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
کدهای گاما تعادل خوبی بین فشرده‌سازی و سرعت رمزگشایی ارائه می‌دهند. اگرچه رمزگشایی آنها از Variable Byte Encoding کندتر است، اما فشرده‌سازی بهتری ارائه می‌دهند. انتخاب بین این دو روش به نیازهای کاربرد بستگی دارد: اگر فضا محدود است، γ codes انتخاب بهتری هستند؛ اگر سرعت پردازش پرسمان مهم‌تر است، Variable Byte Encoding ممکن است مناسب‌تر باشد.</p>

</div>

```python
# تابع کدگذاری یک عدد با استفاده از روش گاما (Gamma Encoding)
def gamma_encode(n):
    """
    این تابع یک عدد صحیح را با استفاده از روش کدگذاری گاما کدگذاری می‌کند.
    فرآیند به این صورت است:
    1. عدد را به فرم دودویی تبدیل می‌کنیم
    2. بیت اول را حذف کرده (offset)
    3. طول offset را محاسبه کرده و به صورت unary کدگذاری می‌کنیم (length)
    4. length و offset را به هم متصل می‌کنیم
    """
    if n <= 0:
        raise ValueError("کدگذاری گاما فقط برای اعداد صحیح مثبت پشتیبانی می‌شود")
    
    # مرحله 1: تبدیل به دودویی و حذف پیشوند '0b'
    binary = bin(n)[2:]
    
    # مرحله 2: حذف بیت اول برای ساخت offset
    offset = binary[1:]
    
    # مرحله 3: محاسبه طول offset و ساخت کد unary برای length
    length = '1' * len(offset) + '0'
    
    # مرحله 4: اتصال length و offset
    return length + offset

# تابع رمزگشایی یک کد گاما
def gamma_decode(code):
    """
    این تابع یک کد گاما را به عدد اصلی بازمی‌گرداند.
    فرآیند به این صورت است:
    1. تعداد 1های اولیه را می‌شمارد تا به 0 برسیم (طول offset)
    2. به تعداد بیت offset را می‌خوانیم
    3. یک 1 به ابتدای offset اضافه کرده و به دودویی تبدیل می‌کنیم
    """
    # مرحله 1: شمارش تعداد 1ها برای پیدا کردن طول offset
    length = 0
    i = 0
    while i < len(code) and code[i] == '1':
        length += 1
        i += 1
    
    # مرحله 2: خواندن offset
    offset = code[i+1:i+1+length]
    
    # مرحله 3: بازسازی عدد اصلی
    return int('1' + offset, 2)

# تابع رمزگشایی یک جریان از کدهای گاما
def gamma_decode_stream(bit_stream):
    """
    این تابع یک جریان از بیت‌ها را که شامل چندین کد گاما است را رمزگشایی می‌کند.
    این تابع برای نمایش نحوه کار با لیست‌های ارسال فشرده‌شده مفید است.
    """
    numbers = []
    i = 0
    while i < len(bit_stream):
        # پیدا کردن طول کد بعدی
        length = 0
        while i < len(bit_stream) and bit_stream[i] == '1':
            length += 1
            i += 1
        
        if i >= len(bit_stream):
            break
        
        # خواندن offset
        offset = bit_stream[i+1:i+1+length]
        
        # بازسازی عدد و اضافه به لیست
        numbers.append(int('1' + offset, 2))
        
        # حرکت به ابتدای کد بعدی
        i += 1 + length
    
    return numbers

# تست توابع با اعداد نمونه از کتاب و مثال‌های واقعی
test_numbers = [1, 2, 3, 4, 9, 13, 24, 511, 1025, 5, 214577]

print("تست کدگذاری و رمزگشایی:")
print("-" * 60)
print(f"{'عدد':<10} | {'کد گاما':<25} | {'طول کد':<10} | {'رمزگشایی':<10}")
print("-" * 60)

for n in test_numbers:
    encoded = gamma_encode(n)
    decoded = gamma_decode(encoded)
    print(f"{n:<10} | {encoded:<25} | {len(encoded):<10} | {decoded:<10}")

print("\n\nتست رمزگشایی جریان بیت‌ها:")
print("-" * 60)
# ایجاد یک جریان از کدهای گاما برای تست
stream = ''.join([gamma_encode(n) for n in test_numbers])
decoded_stream = gamma_decode_stream(stream)
print(f"جریان بیت‌ها: {stream[:50]}...")
print(f"اعداد رمزگشایی شده: {decoded_stream}")

# محاسبه و نمایش طول کد برای چند عدد نمونه
print("\n\nتحلیل طول کدها:")
print("-" * 60)
print(f"{'عدد':<10} | {'طول کد گاما':<15} | {'2*floor(log2(n))+1':<15}")
print("-" * 60)

for n in [1, 2, 3, 4, 9, 13, 24, 511, 1025]:
    encoded = gamma_encode(n)
    theoretical_length = 2 * (len(bin(n)) - 3) + 1
    print(f"{n:<10} | {len(encoded):<15} | {theoretical_length:<15}")
```
