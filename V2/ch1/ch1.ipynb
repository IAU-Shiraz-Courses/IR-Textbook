<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section10" style="font-size:24pt; font-weight:bold; color:#FF6700;">۱. مقدمه‌ای بر بازیابی اطلاعات: سفر به قلب وب امروزی</span> 🧠
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 تصور کنید یک مغز دیجیتال فوق‌العاده</p>
<p style="font-size:16pt; text-align:right;">
شما هر روز چند بار از خود می‌پرسید: «اون ایمیل مهم کجا رفت؟»، «مقاله‌ای که دیروز خوندم رو کجا ذخیره کردم؟» یا «چطور می‌تونم یه فیلم خوب با بازیگرای مورد علاقه‌م پیدا کنم؟». در پشت هر یک از این سوالات ساده، یک غول نامرئی به نام <strong>بازیابی اطلاعات</strong> (Information Retrieval) کار می‌کند. این علم، هنر یافتن سوزنِ اطلاعات در انبار کاه دیجیتال ماست.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
دهه‌ها پیش، این یک مهارت تخصصی برای کتابداران و وکلای پژوهشگر بود. امروز، اما صدها میلیون نفر هر روز از آن استفاده می‌کنند—هر بار که در گوگل عبارتی را تایپ می‌کنند یا در صندوق ورودی ایمیل خود کلمه‌ای را جستجو می‌نمایند. بازیابی اطلاعات، دیگر یک فعالیت حاشیه‌ای نیست؛ بلکه به dominant form of information access تبدیل شده است.
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 چالش اصلی: داده‌های بی‌نظم و نامنظم</p>
<p style="font-size:16pt; text-align:right;">
قلب این علم، کار با داده‌های <strong>ناساختاریافته</strong> (Unstructured Data) است. داده‌های ساختاریافته، مانند یک جدول اکسل منظم یا یک پایگاه داده فروشگاه، ستون‌ها و ردیف‌های مشخصی دارند. اما بیشتر اطلاعات دنیای ما—مانند ایمیل‌ها، مقالات، اخبار و صفحات وب—مثل یک اتاق پر از وسایل هستند که هرچیز جایی ریخته و هیچ برچسبی ندارد. علم بازیابی اطلاعات به کامپیوتر یاد می‌دهد که چطور در این اتاق شلوغ، مورد مورد نظر ما را پیدا کند.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
البته، تقریباً هیچ داده‌ای کاملاً بی‌ساختار نیست. متون هم دارای ساختارهای پنهان زبانی هستند. سیستم‌های بازیابی مدرن از این ساختارها مانند عناوین، پاراگراف‌ها و حتی لینک‌ها (hyperlinks) بهره می‌برند تا جستجو را <strong>نیمه‌ساختاریافته</strong> (Semi-structured) و هوشمندانه‌تر کنند.
</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🗂️ فراتر از جستجو: سازماندهی و طبقه‌بندی</p>
<p style="font-size:16pt; text-align:right;">
بازیابی اطلاعات فقط پیدا کردن چند سند مرتبط نیست. این شامل وظایف هوشمندانه‌تری است:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>خوشه‌بندی (Clustering):</strong> تصور کنید صدها سند علمی در اختیار دارید. سیستم می‌تواند آن‌ها را به صورت خودکار به گروه‌های موضوعی تقسیم کند، درست مانند زمانی که کتاب‌های یک قفسه را بر اساس موضوع مرتب می‌کنید.</li>
<li><strong>طبقه‌بندی (Classification):</strong> سیستم ایمیل شما هر روز این کار را برای شما انجام می‌دهد. به صورت خودکار تشخیص می‌دهد که کدام ایمیل اسپم است و کدامیک مهم و آن را در پوشه مناسب قرار می‌دهد.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
با توجه به مقیاس عملکرد، سیستم‌های بازیابی اطلاعات به سه دسته اصلی تقسیم می‌شوند:</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>🌐 جستجوی وب (Web Scale):</strong> بزرگترین و پیچیده‌ترین حوزه. این سیستم‌ها باید بر روی میلیاردها سند در میلیون‌ها کامپیوتر جستجو کنند. چالش‌ها عظیم هستند: از جمع‌آوری اسناد با خزنده‌های وب (Crawlers) گرفته تا مقابله با ترفندهای سایتی برای دستکاری رتبه‌ها.</li>
<li><strong>💻 جستجوی شخصی (Personal Search):</strong> همان جستجویی که در کامپیوتر خود انجام می‌دهید (مانند Spotlight در مک یا Instant Search در ویندوز). اینجا تمرکز بر سرعت، سبکی و کارایی با منابع محدود است.</li>
<li><strong>🏢 جستجوی سازمانی/حوزه‌محور (Enterprise/Domain-Specific):</strong> جستجو در مجموعه‌های متمرکز مانند اسناد داخلی یک شرکت، پایگاه‌های ثبت اختراع یا مقالات تخصصی شیمی. این سیستم‌ها معمولاً روی چند سرور قدرتمند اجرا می‌شوند.</li>
</ol>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🖼️ یک تصویر به ارزش هزاران کلمه</p>
<p style="font-size:16pt; text-align:right;">
برای درک عمیق‌تر این مفهوم، تصور کنید کاربری که در اقیانوس بی‌پایان داده‌ها (سمت راست) یک نیاز اطلاعاتی دارد (سمت چپ). یک سیستم بازیابی اطلاعات هوشمند (مغز دیجیتال) باید بتواند از میان این حجم عظیم داده‌های نامنظم، دقیقاً همان اسناد مرتبط را استخراج و به کاربر ارائه دهد. این تصویر، قلب تلاش ما در این فصل است: چطور این مغز دیجیتال را بسازیم.
</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/1.png" alt="ocean of information and our role as a homo sapian" style="width:600px; border:1px solid #ccc; border-radius:12px;">
</div>
</div>

<p style="font-size:16pt; text-align:right;">
در این کتاب، ما به بررسی ایده‌ها و تکنیک‌هایی می‌پردازیم که این مغز دیجیتال را می‌سازند؛ از ساختارهای داده‌ای اولیه گرفته تا الگوریتم‌های پیچیده‌ای که دنیای وب را می‌چرخانند. 
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11" style="font-size:24pt; font-weight:bold; color:#FF6700;">۱.۱ یک مسئله نمونه از بازیابی اطلاعات: جستجو در اقیانوس دیجیتال</span> 🎭
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 تصور کنید مدیر یک سرویس استریم فیلمنامه</p>
<p style="font-size:16pt; text-align:right;">
شما مدیر آرشیو دیجیتال غول‌پیکری هستید که شامل تمام فیلمنامه‌های نوشته‌شده در تاریخ است. یک روز، یک کاربر پژوهشگر با یک درخواست بسیار خاص به شما مراجعه می‌کند: «می‌خواهم تمام فیلمنامه‌هایی را پیدا کنم که در آن‌ها <strong>بروتوس</strong> و <strong>سزار</strong> هر دو حضور دارند، اما <strong>کالپورنیا</strong> در هیچ‌کدام حضور ندارد.» چطور این جستجو را به صورت کارآمد انجام می‌دهید؟</p>
</div>

<p style="font-size:16pt; text-align:right;">
یک راه اولیه و شاید بلافاصله به ذهن می‌رسد، استخدام یک کتابدار بسیار پرتلاش و خواندن تک‌تک تمام فیلمنامه‌ها! این روش در دنیای کامپیوتر به <strong>پیمایش خطی</strong> یا <strong>گرِپ کردن (grepping)</strong> معروف است. برای مجموعه‌ای به کوچکی مانند آثار کامل شکسپیر، این روش حتی با سرعت کامپیوترهای امروزی هم قابل قبول است. اما برای میلیاردها سند یا برای پرسمان‌های پیچیده‌تر، این روش مانند تلاش برای پیدا کردن سوزن در انبار کاه است!
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧮 ایده اول: ماتریس وقوع بولی (یک برگه بازرسی غول‌پیکر)</p>
<p style="font-size:16pt; text-align:right;">
اولین راهکار هوشمندانه، ساخت یک <strong>ماتریس وقوع بولی</strong> است. تصور کنید یک برگه غول‌پیکر که در آن هر ردیف به یک واژه (مثل «بروتوس») و هر ستون به یک سند (مثل «هملت») اختصاص دارد. در خانه‌ای که واژه در سند حضور دارد، عدد ۱ و در غیر این صورت ۰ می‌نویسیم.
</p>
<p style="font-size:16pt; text-align:right;">
برای پاسخ به پرسمان <span style="font-family:monospace;">Brutus AND Caesar AND NOT Calpurnia</span>، کافی است بردارهای بولی مربوط به این سه واژه را برداشته و یک عملیات بیتی روی آن‌ها انجام دهیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
    
 $$ [1, 1, 0, 1, 0, 0] \land [1, 1, 0, 1, 1, 1] \land [1, 0, 1, 1, 1, 1] = [1, 0, 0, 1, 0, 0]
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
نتیجه این محاسبه به ما می‌گوید که سندهای شماره ۱ (هملت) و ۳ (مکبث) پاسخ‌های درست هستند. این روش، در تئوری بی‌نقص است، اما در عمل با یک مشکل اساسی روبرو است: <strong>اندازه!</strong>
</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚫 فاجعه مقیاس: ماتریس بیش از حد بزرگ!</p>
<p style="font-size:16pt; text-align:right;">
فرض کنید مجموعه ما فقط یک میلیون سند با ۵۰۰ هزار واژهٔ منحصربه‌فرد دارد. ماتریس ما ابعاد $500{,}000 \times 1{,}000{,}000$ خواهد داشت! این ماتریس حدود <strong>۵۰۰ میلیارد</strong> درایه دارد. حتی اگر هر درایه فقط یک بایت باشد، به بیش از ۴۶۰ گیگابایت فضا نیاز دارد. مشکل بزرگ‌تر این است که این ماتریس فوق‌العاده <strong>تنک (sparse)</strong> است. چون هر سند فقط حدود ۱۰۰۰ واژه دارد، از ۵۰۰ میلیارد درایه، کمتر از ۱ میلیارد آن‌ها ۱ هستند و بقیه (یعنی ۹۹.۸٪) همه صفر هستند! ما در حال ذخیره یک اقیانوس از صفرها هستیم!
</p>
</div>

<div style="background:#d4edda; border-right:4px solid #155724; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#155724;">💡 راه حل نهایی: ایندکس معکوس (نگاهی از زاویه‌ای دیگر)</p>
<p style="font-size:16pt; text-align:right;">
به‌جای پرسیدن «کدام اسناد شامل واژه X هستند؟»، مسئله را برعکس می‌کنیم و می‌پرسیم: «واژه X در کدام اسناد ظاهر شده است؟» این تغییر دیدگاه ساده، به <strong>ایندکس معکوس</strong> (Inverted Index) منجر می‌شود.
</p>
<p style="font-size:16pt; text-align:right;">
این ساختار هوشمندانه از دو بخش اصلی تشکیل شده است:
</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>📖 دیکشنری (Dictionary):</strong> یک فهرست مرتب‌شده از تمام واژه‌هایی که در مجموعه وجود دارند. این مانند فهرست مطالب کتاب است.</li>
<li><strong>📋 لیست ارسال (Postings List):</strong> برای هر واژه در دیکشنری، لیستی از تمام اسنادی که آن واژه در آن‌ها یافت شده است. این لیست‌ها معمولاً بر اساس شماره سند (docID) مرتب هستند.</li>
</ol>

<p style="font-size:16pt; text-align:right;">
برای مثال ما، ایندکس معکوس به این شکل خواهد بود:
</p>

<div style="text-align: center;">
  <div style="font-size:18pt; color:#333; background-color:rgba(255,255,255,0.8); padding:15px; border-radius:10px; box-shadow:0 2px 8px rgba(0,0,0,0.04); display:inline-block; width:auto; margin: 0 auto;">
    <span>Brutus → [1, 2, 4] | </span>
    <span>Caesar → [1, 2, 4, 5, 6] | </span>  
    <span>Calpurnia → [2]</span>
  </div>
</div>

<p style="font-size:16pt; text-align:right;">
    <br/>
حالا برای پاسخ به پرسمان، دیگر نیازی به محاسبات پیچیده و ماتریس عظیم نداریم. فقط کافی است لیست ارسال «بروتوس» و «سزار» را گرفته و اشتراکشان را محاسبه کنیم، و سپس از نتیجه، سند شماره ۲ (که در لیست ارسال «کالپورنیا» است) را حذف کنیم. این فرآیند فوق‌العاده سریع و بهینه است.
</p>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🖼️ تصویر نهایی: لحظهٔ «آها!»</p>
<p style="font-size:16pt; text-align:right;">
تصویر زیر (شکل ۱.۲ در کتاب) دقیقاً همان نتیجه‌ای است که از روش ایندکس معکوس به دست می‌آید. این تصویر، لحظه‌ای است که تلاش فکری ما به نتیجه ملموس تبدیل می‌شود. این تصویر نشان می‌دهد که اسناد ۱ و ۳ (هملت و مکبث) پاسخ صحیح پرسمان ما هستند، دقیقاً همان چیزی که با محاسبات بیتی به دست آوردیم. این تصویر، پلی است بین تئوری ماتریس و کاربرد عملی ایندکس معکوس.
</p>

<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/1.2.1.png" alt="Results from Shakespeare" style="width:800px; border:1px solid #ccc; border-radius:12px;">
</div>

<p style="font-size:16pt; text-align:right;">
ایدهٔ ایندکس معکوس، سنگ بنای بسیاری از سیستم‌های بازیابی اطلاعات مدرن، از گوگل گرفته تا جستجوی فایل‌های محلی شما، است. شما در حال یادگیری یکی از بنیادی‌ترین و قدرتمندترین مفاهیم در دنیای دیجیتال هستید. 🚀
</p>

</div>
</div>
</div>

```python
######## اگر یکسری از جاهای کد را متوجه نمیشید، نگران نباشید چون در ادامه‌ی جزوه در مورد تمام کارایی که اینجا داریم میکنیم بیشتر صحبت میکنیم. ########



# --- تعریف ساختار داده‌ها ---

# نمونه‌ای از مجموعه اسناد (در اینجا: نمایشنامه‌های شکسپیر)
documents = {
    "Antony": "Agrippa [Aside to Domitius Enobarbus]: Why, Enobarbus, When Antony found Julius Caesar dead, He cried almost to roaring; and he wept When at Philippi he found Brutus slain.",
    "Hamlet": "Lord Polonius: I did enact Julius Caesar: I was killed i' the Capitol; Brutus killed me.",
    "Othello": "Had I but one hour tonight-- Othello, I am not well.",
    "Macbeth": "If it were done when 'twere done, then 'twere well it were done quickly."
}
print(documents)
print()

# --- پیاده‌سازی الگوریتم‌ها ---

def build_term_document_matrix(docs):
    """
    این تابع ماتریس وقوع بولی (term-document incidence matrix) را می‌سازد.
    خروجی: یک دیکشنری که کلیدها واژه‌ها و مقادیر دیکشنری‌های دیگر هستند
    """
    print("مرحله ۱: ساخت ماتریس وقوع بولی...")
    all_terms = set()
    # استخراج تمام واژه‌های یکتا از همه اسناد
    for doc_name, content in docs.items():
        all_terms.update(content.lower().split())

    # ساخت ماتریس با مقادیر اولیه صفر
    term_doc_matrix = {term: {} for term in all_terms}

    # پر کردن ماتریس بر اساس وجود واژه در هر سند
    for term in all_terms:
        for doc_name, content in docs.items():
            term_doc_matrix[term][doc_name] = 1 if term in content.lower() else 0

    print("ماتریس وقوع بولی ساخته شد.")
    return term_doc_matrix

def boolean_query(matrix, query):
    """
    این تابع یک پرسمان بولی را بر اساس ماتریس وقوع اجرا می‌کند.
    ورودی: ماتریس وقوع و پرسمان (مثال: ['Brutus', 'AND', 'Caesar', 'AND', 'NOT', 'Calpurnia'])
    """
    print(f"\nمرحله ۲: اجرای پرسمان بولی: {' '.join(query)}")
    
    # نرمال‌سازی پرسمان: تبدیل واژه‌ها به حروف کوچک (اما نه عملگرها)
    normalized_query = []
    for token in query:
        if token in ['AND', 'OR', 'NOT']:
            normalized_query.append(token)
        else:
            normalized_query.append(token.lower())  # واژه‌ها را کوچک می‌کنیم
    
    # جداسازی پرسمان به بخش‌های مثبت و منفی
    positive_terms = []
    negative_terms = []
    i = 0
    while i < len(normalized_query):
        token = normalized_query[i]
        if token == 'NOT':
            if i + 1 < len(normalized_query) and normalized_query[i + 1] not in ['AND', 'OR', 'NOT']:
                negative_terms.append(normalized_query[i + 1])
                i += 2
                continue
        elif token not in ['AND', 'OR']:
            positive_terms.append(token)
        i += 1

    # اگر واژه‌ای وجود نداشت، هیچ سندی برگردانده نشود
    if not positive_terms:
        print("نتیجه پرسمان: []")
        return []

    # شروع با اسناد حاوی اولین واژه مثبت
    first_term = positive_terms[0]
    if first_term not in matrix:
        result_docs = set()
    else:
        result_docs = set(doc for doc, exists in matrix[first_term].items() if exists)

    # اعمال AND برای سایر واژه‌های مثبت
    for term in positive_terms[1:]:
        if term in matrix:
            term_docs = set(doc for doc, exists in matrix[term].items() if exists)
            result_docs = result_docs.intersection(term_docs)
        else:
            result_docs = set()  # واژه‌ای نیست → نتیجه خالی
            break

    # اعمال NOT
    for term in negative_terms:
        if term in matrix:
            term_docs = set(doc for doc, exists in matrix[term].items() if exists)
            result_docs = result_docs.difference(term_docs)

    print(f"نتیجه پرسمان: {sorted(result_docs)}")
    return sorted(result_docs)

def build_inverted_index(matrix):
    """
    این تابع یک ایندکس معکوس از روی ماتریس وقوع می‌سازد.
    این تابع، پل ارتباطی بین ماتریس و ایندکس معکوس را نشان می‌دهد.
    """
    print("\nمرحله ۳: ساخت ایندکس معکوس از ماتریس...")
    inverted_index = {}

    for term, doc_dict in matrix.items():
        # استخراج اسنادی که واژه در آن‌ها وجود دارد
        docs_with_term = [doc for doc, exists in doc_dict.items() if exists]
        if docs_with_term:
            inverted_index[term] = sorted(docs_with_term)

    print("ایندکس معکوس ساخته شد.")
    # print(inverted_index)
    return inverted_index

# --- اجرای مثال ---

if __name__ == "__main__":
    # مرحله ۱: ساخت ماتریس وقوع بولی
    term_doc_matrix = build_term_document_matrix(documents)

    # نمایش ماتریس برای چند واژه نمونه
    sample_terms = ['Brutus', 'Caesar', 'Calpurnia']
    print(f"\nنمونه‌ای از ماتریس وقوع برای واژه‌های {sample_terms}:")
    for term in sample_terms:
        if term in term_doc_matrix:
            print(f"{term}: {term_doc_matrix[term]}")

    # مرحله ۲: اجرای پرسمان بولی
    query = ['Brutus', 'AND', 'Caesar', 'AND', 'NOT', 'Calpurnia']
    result = boolean_query(term_doc_matrix, query)

    # مرحله ۳: ساخت ایندکس معکوس
    inverted_index = build_inverted_index(term_doc_matrix)

    # نمایش ساختار ایندکس معکوس
    sample_terms = ['Brutus', 'Caesar', 'Calpurnia']
    print(f"\nنمونه‌ای از ماتریس وقوع برای واژه‌های {sample_terms}:")
    for term in sample_terms:
        lower_term = term.lower()
        if lower_term in term_doc_matrix:
            print(f"{term}: {term_doc_matrix[lower_term]}")
        else:
            print(f"{term}: (وجود ندارد در ماتریس)")

    print("\nاین فرآیند، چگونگی تبدیل یک مسئله جستجوی خطی به یک راه‌حل بهینه (ایندکس معکوس) را نشان می‌دهد.")
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section12" style="font-size:24pt; font-weight:bold; color:#FF6700;">1.2 ساخت اولیه فهرست معکوس: پشت صحنه جادوی جستجو!</span> 🔍
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🚀 یک چالش دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید مدیر داده یک شبکه اجتماعی بزرگ مانند «Instagram» هستید. میلیون‌ها پست روزانه منتشر می‌شود. یک کاربر می‌خواهد تمام پست‌هایی را ببیند که همزمان دارای هشتگ‌های <code>#سفر</code> و <code>#غروب_آفتاب</code> هستند. چگونه می‌توانید این کار را در کسری از ثانیه انجام دهید؟ پیمایش تمام میلیون‌ها پست برای هر جستجو غیرممکن است! اینجاست که جادوی <strong>فهرست معکوس</strong> وارد می‌شود.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای بهره‌گیری از سرعت فوق‌العاده در زمان جستجو، باید این فهرست را <strong>از قبل</strong> بسازیم. این فرآیند مانند تهیه یک دستور آشپزی دقیق است و چهار مرحله اصلی دارد:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>۱. جمع‌آوری مواد اولیه (اسناد)</strong><br/>
همه پست‌های IG را جمع می‌کنیم. این «مجموعه اسناد» ماست.</li>
<li><strong>۲. خرد کردن (توکن‌سازی)</strong><br/>
هر پست را به واحدهای کوچک‌تری به نام «توکن» (Token) می شکنیم. در دنیای ما، این یعنی جدا کردن تمام کلمات و هشتگ‌ها. 🧩</li>
<li><strong>۳. تمیزکاری و آماده‌سازی (پیش‌پردازش زبانی)</strong><br/>
توکن‌ها را نرمال‌سازی می‌کنیم: همه را به حروف کوچک تبدیل می‌کنیم، علائم نگارشی مانند «!» و «#» را حذف می‌کنیم و ریشه کلمات را پیدا می‌کنیم. 🧠</li>
<li><strong>۴. ساخت دفترچه تلفن هوشمند (ایندکس‌سازی)</strong><br/>
حالا مهم‌ترین بخش کار: ساختن خود فهرست معکوس. 📚</li>
</ol>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎩 جادوی مرتب‌سازی: از آشفتگی تا نظم</p>
<p style="font-size:16pt; text-align:right;">
چگونه این دفترچه تلفن را می‌سازیم؟ روش اولیه و هوشمندانه، <strong>ایندکس‌سازی مبتنی بر مرتب‌سازی</strong> است:
</p>

<p style="font-size:16pt; text-align:right;">
<strong>مرحله الف: آشفت‌گاه اولیه</strong><br/>
ابتدا یک لیست بزرگ از تمام جفت‌های (توکن، شناسه_پست) برای کل مجموعه اسناد می‌سازیم. مثلاً:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
(sunset, 1542)<br/>
(travel, 1123)<br/>
(سفر, 1123)<br/>
(good, 8801)<br/>
(غروب_آفتاب, 1542)<br/>
...
</div>

<p style="font-size:16pt; text-align:right;">
<strong>مرحله ب: نظم جادویی</strong><br/>
حالا این لیست شلوغ را بر اساس توکن‌ها (الفبا) مرتب می‌کنیم. سپس موارد تکراری را گروه‌بندی می‌کنیم. نتیجه چیزی شبیه به این می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
good &rarr; 8801<br/>
sunset &rarr; 1542<br/>
travel &rarr; 1123<br/>
سفر &rarr; 1123<br/>
غروب_آفتاب &rarr; 1542<br/>
...
</div>

<p style="font-size:16pt; text-align:right;">
این ساختار مرتب، قلب تپنده فهرست معکوس ماست.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، این دفترچه تلفن هوشمند به دو بخش اصلی تقسیم می‌شود:
</p>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">📚 دو بخش اصلی فهرست معکوس</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>دیکشنری (Dictionary):</strong> این بخش مانند فهرست نام‌ها در دفترچه تلفن است. شامل تمام توکن‌های منحصربه‌فرد (مانند <code>سفر</code>، <code>غروب_آفتاب</code>) است. برای هر توکن، یک آمار کلیدی هم ذخیره می‌شود: <strong>فراوانی سند</strong> (Document Frequency یا $df_t$)؛ یعنی تعداد کل پست‌هایی که آن توکن در آن‌ها وجود دارد. این دیکشنری آنقدر کوچک است که می‌توانیم آن را کاملاً در حافظه اصلی (RAM) نگه داریم تا دسترسی به آن فوری باشد.</li>
<li><strong>لیست ارسال (Postings List):</strong> این بخش مانند شماره تلفن‌های مربوط به هر نام در دفترچه تلفن است. برای هر توکن در دیکشنری، لیستی از شناسه‌های پست‌هایی (docID) که آن توکن در آن‌ها وجود دارد، ذخیره می‌شود. این لیست‌ها بسیار حجیم هستند و روی دیسک ذخیره می‌شوند، اما <strong>مرتب‌شده بر اساس docID</strong> هستند. این مرتب‌سازی، کلید سرعت پردازش پرسمان‌های بولی است که در بخش قبل دیدیم.</li>
</ol>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🤓 عمیق‌تر: بهترین ساختار برای لیست ارسال چیست؟</p>
<p style="font-size:16pt; text-align:right;">
برای ذخیره لیست ارسال در حافظه، دو گزینه اصلی وجود دارد و انتخاب بین آن‌ها یک معامله بین انعطاف‌پذیری، سرعت و فضا است:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>لیست پیوندی تک‌طرفه (Singly Linked List):</strong> مثل قطاری از واگن‌ها که به هم متصل‌اند. <strong>مزیت:</strong> اضافه کردن یک پست جدید به لیست بسیار سریع و ارزان است. <strong>عیب:</strong> به حافظه پراکنده نیاز دارد و پیمایش آن ممکن است کندتر باشد. این ساختار، پایه‌ای برای تکنیک‌های پیشرفته‌تر مانند <em>Skip Lists</em> است.</li>
<li><strong>آرایه با طول متغیر (Variable Length Array):</strong> مثل یک قطار یکپارچه با تعداد واگن متغیر. <strong>مزیت:</strong> بسیار فشرده‌تر است و چون در حافظه پشت سر هم قرار گرفته، پیمایش آن در پردازنده‌های مدرن بسیار سریع‌تر است. <strong>عیب:</strong> اضافه کردن عضو جدید ممکن است نیازمند ساختن کل آرایه جدیدی باشد.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در دنیای واقعی، اگر به‌روزرسانی زیاد باشد، از لیست پیوندی استفاده می‌کنیم. اما اگر اسناد کمتر تغییر کنند، آرایه‌ها به دلیل فشردگی و سرعت بالا، برنده هستند! 💾
</p>
</div>

<p style="font-size:16pt; text-align:right;">
این ساختار به ظاهر ساده، <strong>بی‌رقیبترین و کارآمدترین روش</strong> برای پشتیبانی از جستجوی متنی موردی (ad-hoc) است. هر بار که در گوگل عبارتی را جستجو می‌کنید یا در یک پایگاه داده حقوقی به دنبال یک سند خاص می‌گردید، در حال استفاده از نسخه‌ای بسیار پیچیده‌تر از همین ایده ساده و زیبا هستید.
</p>

<div align="center">
  <img src="images/reverse indexing 2.png" width="70%">
  <p style="font-size:14pt; color:#084C61; font-weight:bold;"></p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section13" style="font-size:24pt; font-weight:bold; color:#FF6700;">1.3 پردازش پرسمان‌های بولی</span> 🔀
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎓 داستانی از کلاس درس</p>
<p style="font-size:16pt; text-align:right;">
"روز اول کلاس بازیابی اطلاعات بود. دکتر اسکندری وارد شد و پرسید: 'چطور می‌توانم در این کتابخانه‌های دیجیتال، کتاب‌هایی را پیدا کنم که هم درباره‌ی هوش مصنوعی و هم درباره‌ی VLMها باشند؟' سکوت حاکم شد. سپس دکتر لبخندی زد و گفت: 'این همان چیزی است که امروز می‌خواهیم یاد بگیریم - جادوی پرسمان‌های بولی!'"
</p>
</div>

<p style="font-size:16pt; text-align:right;">
در مدل بولی، هر پرسمان ترکیبی از واژه‌ها با عملگرهای <span style="font-family:monospace;">AND</span>، <span style="font-family:monospace;">OR</span>، و <span style="font-family:monospace;">NOT</span> است. این مدل ساده اما قدرتمند، پایه‌ی بسیاری از موتورهای جستجوی اولیه بود و هنوز هم در سیستم‌های اطلاعاتی مدرن کاربرد دارد.
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🧠 فکر کن!</p>
<p style="font-size:16pt; text-align:right;">
تا حالا به این فکر کرده‌اید که چطور گوگل نتایج جستجو را فیلتر می‌کند؟ وقتی تایپ می‌کنید "رستوران ایتالیایی نزدیک من" AND "باز تا ساعت ۱۰ شب"، در حال استفاده از منطق بولی هستید! 🍕🌙
</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای پاسخ به پرسمان ساده‌ای مانند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
    
 $$ \text{Brutus} \land \text{Calpurnia}
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
مراحل زیر را دنبال می‌کنیم:
</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>🔍 یافتن <span style="font-family:monospace;">Brutus</span> در دیکشنری</li>
<li>📋 بازیابی لیست ارسال آن</li>
<li>🔍 یافتن <span style="font-family:monospace;">Calpurnia</span> در دیکشنری</li>
<li>📋 بازیابی لیست ارسال آن</li>
<li>🔗 محاسبهٔ <strong>اشتراک</strong> دو لیست — یعنی یافتن اسنادی که هر دو واژه در آن‌ها ظاهر شده‌اند</li>
</ol>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎬 تصور کن...</p>
<p style="font-size:16pt; text-align:right;">
دو لیست شماره تلفن داریم: یکی از دوستانت به نام "علی" و دیگری از دوستانت که در "تهران" زندگی می‌کنند. می‌خواهی بدانی کدام علی‌ها در تهران زندگی می‌کنند. چه کار می‌کنی؟ دقیقاً! هر دو لیست را نگاه می‌کنی و شماره‌هایی که در هر دو لیست وجود دارند را پیدا می‌کنی. این همان الگوریتم <strong>اشتراک</strong> است!
</p>
</div>

<p style="font-size:16pt; text-align:right;">
چون لیست‌های ارسال <strong>مرتب‌شده بر اساس docID</strong> هستند، می‌توانیم از الگوریتم <strong>ادغام دو‌طرفه</strong> (two-pointer merge) برای یافتن اشتراک در زمان خطی استفاده کنیم. این الگوریتم در شکل ۱.۶ کتاب آمده است:
</p>

<div align="center">
  <img src="images/intersect.bmp" width="60%">
  <p style="font-size:14pt; color:#084C61; font-weight:bold;"></p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🚀 تحلیل پیچیدگی</p>
<p style="font-size:16pt; text-align:right;">
پیچیدگی زمانی این الگوریتم برابر $O(x + y)$ است، که $x$ و $y$ طول لیست‌های ارسال هستند. این بسیار سریع‌تر از اسکن خطی تمام مجموعه است، هرچند از نظر نماد $\Theta$ هم‌مرتبه باشد — اما ثابت عملی آن بسیار کوچک‌تر است.
</p>
<p style="font-size:16pt; text-align:right;">
<strong>تفکر سریع:</strong> اگر لیست اول ۱۰۰۰ آیتم و لیست دوم ۵۰۰ آیتم داشته باشد، در بدترین حالت چند مقایسه نیاز داریم؟ 🤔
</p>

<details>
<summary style="cursor:pointer; font-weight:bold;">پاسخ</summary>
در بدترین حالت، ۱۵۰۰ مقایسه (هر آیتم از هر دو لیست یک بار بررسی می‌شود).
</details>
</div>

<p style="font-size:16pt; text-align:right;">
برای پرسمان‌های چندجمله‌ای مانند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:20px 0;">

 $$ \text{Brutus} \land \text{Caesar} \land \text{Calpurnia}
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
بهینه‌ترین استراتژی، <strong>پردازش واژه‌ها به ترتیب فراوانی صعودی</strong> است: ابتدا واژه‌ای که در کمترین تعداد سند آمده را بررسی می‌کنیم، چون نتیجهٔ میانی از همان ابتدا کوچک خواهد بود و کار کمتری برای اشتراک بعدی لازم است. برای مثال، اگر:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><span style="font-family:monospace;">Calpurnia</span>:  2 سند</li>
<li><span style="font-family:monospace;">Brutus</span>:174 سند</li>
<li><span style="font-family:monospace;">Caesar</span>: 132 سند</li>
</ul>

<div style="background:#d4edda; border-right:4px solid #155724; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#155724;">🎯 استراتژی بهینه</p>
<p style="font-size:16pt; text-align:right;">
آنگاه بهتر است پرسمان را به‌صورت زیر ارزیابی کنیم:
</p>
</div>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
    
 $$ (\text{Calpurnia} \land \text{Brutus}) \land \text{Caesar}
 $$ 
</div>

<div style="background:#e2e3e5; border-right:4px solid #383d41; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#383d41;">🎪 مثال کاربردی</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در یک فروشگاه آنلاین کار می‌کنید و مشتری به دنبال محصولی است که همزمان این ویژگی‌ها را داشته باشد: "قیمت کمتر از ۱۰۰ هزار تومان" AND "رنگ آبی" AND "برند X". اگر بدانید که فقط ۵ محصول رنگ آبی دارند، ۵۰ محصول برند X هستند و ۲۰۰ محصول قیمت کمتر از ۱۰۰ هزار تومان دارند، از کدام فیلتر شروع می‌کنید؟ طبیعتاً از رنگ آبی چون نتایج اولیه را به ۵ محصول محدود می‌کند! 🛍️
</p>
</div>

<p style="font-size:16pt; text-align:right;">
این تکنیک، <strong>بهینه‌سازی پرسمان</strong> (Query Optimization) نام دارد و نیازمند ذخیرهٔ <strong>فراوانی سند</strong> ($df_t$) در دیکشنری است.
</p>

<p style="font-size:16pt; text-align:right;">
برای پرسمان‌های پیچیده‌تر مانند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
    
 $$ (\text{Brutus} \lor \text{Caesar}) \land \lnot (\text{Antony} \lor \text{Cleopatra})
 $$ 
</div>

<p style="font-size:16pt; text-align:right;">
ابتدا هر زیرعبارت را ارزیابی می‌کنیم:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><span style="font-family:monospace;">Brutus OR Caesar</span>: ادغام دو لیست (UNION)</li>
<li><span style="font-family:monospace;">Antony OR Cleopatra</span>: ادغام دو لیست دیگر</li>
<li><span style="font-family:monospace;">NOT (...)</span>: مکمل گرفتن نسبت به تمام اسناد مجموعه</li>
</ul>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">⚠️ چالش NOT</p>
<p style="font-size:16pt; text-align:right;">
اما <strong>پردازش NOT مشکل‌ساز است</strong>، چون نیاز به دانستن مجموعهٔ کامل اسناد دارد — و اگر مجموعه بزرگ باشد، ساخت مکمل بسیار هزینه‌بر است. راه‌حل‌های هوشمندانه‌تری (مثل استفاده از ایندکس معکوس معکوس!) نیاز است، که در اینجا به‌صورت ساده پیاده‌سازی می‌شود.
</p>
<p style="font-size:16pt; text-align:right;">
<strong>تقویت درک:</strong> تصور کنید در یک مهمانی هستید و می‌خواهید کسانی را پیدا کنید که "نوشیدنی نمی‌خورند". آیا راحت‌تر است همه‌ی مهمان‌ها را بررسی کنید یا لیست کسانی را که "نوشیدنی می‌خورند" تهیه کرده و از کل لیست مهمان‌ها کم کنید؟ 🍹
</p>
</div>

<p style="font-size:16pt; text-align:right;">
الگوریتم کلی برای پرسمان‌های ترکیبی در شکل ۱.۷ کتاب آمده است — که واژه‌ها را بر اساس فراوانی مرتب کرده و به‌صورت تدریجی اشتراک می‌گیرد. این، پایهٔ اصلی موتورهای جست‌وجوی بولی است.
</p>

<div align="center">
  <img src="images/fig 1.7.png" width="60%">
  <p style="font-size:14pt; color:#084C61; font-weight:bold;"></p>
</div>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🌐 کاربردهای امروزی</p>
<p style="font-size:16pt; text-align:right;">
اگرچه مدل بولی ساده به نظر می‌رسد، اما هنوز در بسیاری از سیستم‌های مدرن استفاده می‌شود. برای مثال:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>در پایگاه‌های داده حقوقی برای جستجوی دقیق قوانین و احکام</li>
<li>در جستجوی اسناد علمی برای یافتن مقالاتی که همزمان چند مفهوم را پوشش می‌دهند</li>
<li>در ویژگی‌های جستجوی پیشرفته گوگل (مانند جستجوی عبارت دقیق با استفاده از کوتیشن)</li>
<li>در سیستم‌های توصیه‌گر برای فیلتر کردن محتوا بر اساس علایق کاربران</li>
</ul>
</div>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">

<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">
🧪 آزمایش فکری
</p>

<p style="font-size:16pt; text-align:right;">
حالا که با مفاهیم آشنا شدی، بیا یک سناریوی واقعی را حل کنیم:
</p>

<p style="font-size:16pt; text-align:right;">
شما توسعه‌دهنده یک اپلیکیشن دوست‌یابی هستید. کاربر می‌خواهد فردی را پیدا کند که این ویژگی‌ها را داشته باشد:
"علاقه‌مند به سفر" AND "علاقه‌مند به کتاب" AND NOT "سیگار می‌کشد".
با توجه به آمار زیر، بهینه‌ترین استراتژی برای جستجو چیست؟
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>علاقه‌مند به سفر: ۱۰,۰۰۰ کاربر</li>
<li>علاقه‌مند به کتاب: ۵,۰۰۰ کاربر</li>
<li>سیگار می‌کشد: ۲,۰۰۰ کاربر</li>
</ul>

<details>
<summary style="cursor:pointer; font-weight:bold;">راهنمایی</summary>
فکر کن کدام شرط نتایج را به کمترین تعداد محدود می‌کند.
</details>

<details>
<summary style="cursor:pointer; font-weight:bold;">پاسخ</summary>
بهینه‌ترین استراتژی: ابتدا کاربرانی که "علاقه‌مند به کتاب" هستند را پیدا می‌کنیم (۵,۰۰۰ نفر)، سپس این لیست را با لیست "علاقه‌مند به سفر" اشتراک می‌گیریم، و در نهایت از لیست نهایی، کسانی که "سیگار می‌کشند" را حذف می‌کنیم.
</details>

</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">📚 برای مطالعه بیشتر</p>
<p style="font-size:16pt; text-align:right;">
اگر به این موضوع علاقه‌مند شدی، پیشنهاد می‌کنم مقاله "Beyond Boolean: The Evolution of Information Retrieval Models" را مطالعه کنی. این مقاله نشان می‌دهد که چگونه مدل‌های بولی به مدل‌های پیچیده‌تر مانند مدل فضای برداری و مدل‌های احتمالی تبدیل شده‌اند. 📖
</p>

</p>
</div>

</div>

```python
# --- تعریف ساختار داده‌ها ---

# نمونه‌ای از یک ایندکس معکوس (ساده‌شده)
# کلیدها واژه‌ها و مقادیر، لیست‌های ارسال (postings lists) مرتب‌شده هستند.
# این ساختار، قلب یک موتور جستجوی بولی است.
inverted_index = {
    "Brutus": [1, 2, 4, 11, 31, 45, 173, 174],
    "Calpurnia": [2, 31, 54, 101],
    "Caesar": [1, 2, 4, 5, 6, 16, 57, 132],
    "Antony": [5, 10, 31],
    "Cleopatra": [2, 3, 31, 64]
}

# فرض می‌کنیم مجموعه اسناد ما شامل شناسه‌های ۱ تا ۲۰۰ است.
# این لیست برای عملگر NOT ضروری است تا بتوانیم مکمل یک مجموعه را محاسبه کنیم.
ALL_DOCUMENTS = list(range(1, 201))


# --- پیاده‌سازی الگوریتم‌های اصلی ---

def intersect(p1, p2):
    """
    الگوریتم ادغام دوطرفه برای یافتن اشتراک دو لیست مرتب‌شده.
    این پیاده‌سازی دقیقاً مطابق با شبه‌کد کتاب و توضیحات مارک‌دان سل است.
    این الگوریتم به صورت خطی و با حداکثر کارایی O(len(p1) + len(p2)) عمل می‌کند.
    """
    print(f"در حال محاسبه اشتراک لیست‌ها: {p1} و {p2}")
    i, j = 0, 0  # دو نشانگر (pointer) برای حرکت در لیست‌ها
    result = []
    # تا زمانی که هر دو نشانگر به انتها نرسیده‌اند، به مقایسه ادامه می‌دهیم
    while i < len(p1) and j < len(p2):
        # اگر آیتم‌ها برابر بودند، در نتیجه ذکر شده و هر دو نشانگر جلو می‌روند
        if p1[i] == p2[j]:
            result.append(p1[i])
            i += 1
            j += 1
        # اگر آیتم لیست اول کوچکتر بود، فقط نشانگر اول را جلو می‌بریم
        elif p1[i] < p2[j]:
            i += 1
        # در غیر این صورت، نشانگر دوم را جلو می‌بریم
        else:
            j += 1
    print(f"نتیجه اشتراک: {result}")
    return result

def union(p1, p2):
    """
    الگوریتم ادغام برای یافتن اجتماع دو لیست مرتب‌شده (بدون تکرار).
    این تابع برای پردازش عملگر OR استفاده می‌شود.
    """
    print(f"در حال محاسبه اجتماع لیست‌ها: {p1} و {p2}")
    i, j = 0, 0
    result = []
    # تا زمانی که هر دو لیست آیتم دارند
    while i < len(p1) and j < len(p2):
        if p1[i] == p2[j]:
            result.append(p1[i])
            i += 1
            j += 1
        elif p1[i] < p2[j]:
            result.append(p1[i])
            i += 1
        else:
            result.append(p2[j])
            j += 1
    # اضافه کردن آیتم‌های باقی‌مانده در هر دو لیست
    result.extend(p1[i:])
    result.extend(p2[j:])
    print(f"نتیجه اجتماع: {result}")
    return result

def complement(postings_list):
    """
    تابع مکمل (NOT) برای پیاده‌سازی عملگر نفی.
    این عملیات هزینه‌بر است چون به کل مجموعه اسناد نیاز دارد.
    """
    print(f"در حال محاسبه مکمل لیست: {postings_list}")
    p1_iter = iter(postings_list)
    current_doc_to_exclude = next(p1_iter, None)
    result = []
    for doc_id in ALL_DOCUMENTS:
        if current_doc_to_exclude is not None and doc_id == current_doc_to_exclude:
            current_doc_to_exclude = next(p1_iter, None)
        else:
            result.append(doc_id)
    # برای نمایش بهتر، فقط ۱۰ سند اول نتیجه را چاپ می‌کنیم
    print(f"نتیجه مکمل (۱۰ سند اول): {result[:10]}...")
    return result

def get_postings(term):
    """تابع کمکی برای گرفتن لیست ارسال یک واژه."""
    return inverted_index.get(term, [])

def process_conjunctive_query(terms):
    """
    پردازش یک پرسمان ترکیبی (conjunctive query) با بهینه‌سازی.
    واژه‌ها بر اساس فراوانی (طول لیست ارسال) به ترتیب صعودی پردازش می‌شوند.
    این مهم‌ترین تکنیک بهینه‌سازی برای پرسمان‌های بولی است.
    """
    print(f"\n--- پردازش پرسمان AND: {' AND '.join(terms)} ---")
    
    # مرحله ۱: مرتب‌سازی واژه‌ها بر اساس طول لیست ارسال (بهینه‌سازی کلیدی)
    terms_with_freq = [(term, len(get_postings(term))) for term in terms]
    terms_with_freq.sort(key=lambda x: x[1])
    sorted_terms = [term for term, freq in terms_with_freq]
    print(f"ترتیب بهینه پردازش واژه‌ها (بر اساس فراوانی صعودی): {sorted_terms}")

    # مرحله ۲: شروع با کوچکترین لیست و اشتراک گرفتن تدریجی
    result = get_postings(sorted_terms[0])
    print(f"شروع با لیست کوچکترین واژه '{sorted_terms[0]}': {result}")
    
    for term in sorted_terms[1:]:
        if not result:  # اگر نتیجه میانی خالی شد، دیگر ادامه نده
            print("نتیجه میانی خالی شد، پردازش متوقف شد.")
            break
        result = intersect(result, get_postings(term))
        
    print(f"نتیجه نهایی پرسمان: {result}")
    return result

# --- اجرای مثال‌ها ---

if __name__ == "__main__":
    
    # مثال ۱: پرسمان ساده (Brutus AND Calpurnia)
    print("="*50)
    print("مثال ۱: پرسمان ساده")
    brutus_postings = get_postings("Brutus")
    calpurnia_postings = get_postings("Calpurnia")
    intersect(brutus_postings, calpurnia_postings)

    # مثال ۲: پرسمان ترکیبی با بهینه‌سازی (Brutus AND Caesar AND Calpurnia)
    print("\n" + "="*50)
    print("مثال ۲: پرسمان ترکیبی با بهینه‌سازی")
    process_conjunctive_query(["Brutus", "Caesar", "Calpurnia"])

    # مثال ۳: پرسمان پیچیده با OR و NOT
    print("\n" + "="*50)
    print("مثال ۳: پرسمان پیچیده با OR و NOT")
    # (Brutus OR Caesar) AND NOT (Antony OR Cleopatra)
    
    # مرحله ۱: محاسبه (Brutus OR Caesar)
    brutus_caesar_union = union(get_postings("Brutus"), get_postings("Caesar"))
    
    # مرحله ۲: محاسبه (Antony OR Cleopatra)
    antony_cleopatra_union = union(get_postings("Antony"), get_postings("Cleopatra"))
    
    # مرحله ۳: محاسبه NOT (Antony OR Cleopatra)
    # هشدار: این عملیات بسیار سنگین است!
    # antony_cleopatra_complement = complement(antony_cleopatra_union)
    
    # مرحله ۴: محاسبه نهایی (اشتراک)
    # نتیجه نهایی = intersect(brutus_caesar_union, antony_cleopatra_complement)
    # برای جلوگیری از اجرای عملیات سنگین NOT، آن را کامنت می‌کنیم.
    print("برای جلوگیری از عملیات سنگین NOT، محاسبه نهایی انجام نمی‌شود.")
    print("در عمل، موتورهای جستجو از تکنیک‌های هوشمندانه‌تری برای پردازش NOT استفاده می‌کنند.")
```

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #ccc; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section14" style="font-size:24pt; font-weight:bold; color:#FF6700;">1.4 مدل بولی گسترش‌یافته در مقابل بازیابی رتبه‌بندی‌شده</span> ⚖️
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 دو دستیار در ذهن شما تصور کنید</p>
<p style="font-size:16pt; text-align:right;">
دستیار اول بسیار دقیق و خشک است. اگر بگویید «قهوه و کیک»، دقیقاً اسنادی را پیدا می‌کند که هر دو کلمه را داشته باشند، نه یکی بیشتر. این دستیار، <strong>مدل بولی</strong> است.
<br/><br/>
دستیار دوم هوشمند و شهودی است. اگر بگویید «یک کافه دنج برای عصر جمعه»، او با توجه به کل جملات، مفهوم «دنج» و «عصر جمعه» را می‌فهمد و بهترین گزینه‌ها را به شما پیشنهاد می‌دهد، حتی اگر دقیقاً این کلمات در متن نباشند. این دستیار، <strong>مدل بازیابی رتبه‌بندی‌شده</strong> (Ranked Retrieval) است.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای دهه‌ها، دنیای جستجوی حرفه‌ای، به‌ویژه در حوزه‌های تخصصی مانند حقوق، تحت سلطه دستیار اول (مدل بولی) بود. اما نسخه‌ی ساده‌ی بولی که با `AND` و `OR` آشنا شدیم، برای کاربران حرفه‌ای بسیار محدود بود. آن‌ها به یک نسخه‌ی <strong>ابرسوپر</strong> نیاز داشتند!
</p>

<p style="font-size:16pt; text-align:right;">
مهم‌ترین ارتقا، اضافه شدن <strong>عملگرهای نزدیکی</strong> (Proximity Operators) بود. این عملگرها به دستیار ما کمک می‌کنند تا بفهمد دو کلمه فقط در یک سند وجود ندارند، بلکه «کنار هم» معنای خاصی ایجاد می‌کنند.
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🎬 مطالعه موردی: دنیای جستجوی حقوقی (Westlaw)</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید یک وکیل برجسته هستید و باید پرونده‌ای پیچیده درباره «مسئولیت میزبان در قبال مهمانان مست» را تحقیق کنید. در سیستم Westlaw، شما یک پرسمان غول‌پیکر و دقیق می‌نویسید، نه یک جستجوی ساده:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0; background:#f8f9fa; padding:10px; border-radius:5px;">
host! /p (responsib! liab!) /p (intoxicat! drunk!) /p guest
</div>

<p style="font-size:16pt; text-align:right;">
بیایید این جادو را با هم بشکافیم:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>host!</code>: هر کلمه‌ای که با <code>host</code> شروع شود (مثل host, host's).</li>
<li><code>/p</code>: معجون اصلی! یعنی «در همان <strong>پاراگراف</strong>».</li>
<li><code>(responsib! liab!)</code>: کلماتی مثل responsible یا liability.</li>
</ul>
<p style="font-size:16pt; text-align:right;">
این پرسمان می‌گوید: «اسنادی را پیدا کن که در یک پاراگراف، کلماتی از خانواده host، در پاراگراف دیگری کلماتی از خانواده مسئولیت، و در پاراگراف سومی کلماتی از خانواده مستی و مهمان وجود داشته باشد.» این سطح از کنترل، برای یک متخصص طلاست! 🏆
</p>
</div>

<p style="font-size:16pt; text-align:right;">
اما یک بحث بزرگ همیشه وجود داشته: کدام روش بهتر است؟
</p>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">⚖️ نبرد دقّت (Precision) در مقابل بازیابی (Recall)</p>
<p style="font-size:16pt; text-align:right;">
مدل بولی یک مشکل اساسی دارد:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>استفاده زیاد از <code>AND</code> → نتایج بسیار <strong>مربوط</strong> ولی <strong>کم</strong>. (دقت بالا، بازیابی پایین)</li>
<li>استفاده زیاد از <code>OR</code> → نتایج بسیار <strong>زیاد</strong> ولی بسیاری از آن‌ها <strong>غیرمربوط</strong>. (دقت پایین، بازیابی بالا)</li>
</ul>
<p style="font-size:16pt; text-align:right;">
یافتن نقطه‌ی طلایی بین این دو، مثل راه رفتن روی تیغ یک کارد است! به همین دلیل، مطالعات (مانند تحقیق Turtle در ۱۹۹۴) نشان دادند که حتی برای وکلا، پرسمان‌های متنی آزاد (دستیار دوم) اغلب نتایج بهتری تولید می‌کنند.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
پس راه حل چیست؟ چگونه می‌توانیم بهترین هر دو دنیا را داشته باشیم؟ اینجاست که ما به <strong>چهار ستون موتورهای جستجوی مدرن</strong> می‌رسیم:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>🔍 جستجوی بخشنده (Tolerant Search):</strong> سیستمی که اشتباهات املایی شما را می‌بخشد و می‌فهمد که «ماشین» و «دستگاه» می‌توانند مترادف باشند.</li>
<li><strong>🧩 جستجوی بافتار (Proximity & Phrase Search):</strong> سیستمی که تفاوت بین «سگ گاز می‌گیرد» و «انسان به سگ گاز می‌گیرد» را می‌فهمد. این یعنی درک نزدیکی و ترتیب کلمات.</li>
<li><strong>📊 اهمیت فراوانی (Term Frequency):</strong> سیستمی که می‌داند اگر کلمه «هوش مصنوعی» در یک سند ۱۰ بار تکرار شده، آن سند احتمالاً از سندی که فقط یک بار به آن اشاره کرده، مرتبط‌تر است.</li>
<li><strong>🥇 هنر رتبه‌بندی (Ranking):</strong> سیستمی که به جای گفتن «بله/خیر»، به هر سند یک امتیاز از ۱۰۰ می‌دهد و بهترین‌ها را در بالای لیست قرار می‌دهد.</li>
</ol>

<div style="background:#d4edda; border-right:4px solid #155724; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#155724;">🌐 سفر از کتابخانه حقوق تا گوگل</p>
<p style="font-size:16pt; text-align:right;">
این چهار ستون، فناوری است که امروزه دنیا را تسخیر کرده است. بار بعدی که در گوگل عبارتی را جستجو می‌کنید، به یاد داشته باشید که در پشت این کادر ساده، این ایده‌ها با قدرت در حال کار هستند. گوگل ممکن است ظاهراً یک جستجوی متنی آزاد ارائه دهد، اما در پشت صحنه از جستجوی عبارت (دستور دوم) و نسخه‌های ساده‌شده‌ای از منطق بولی استفاده می‌کند.
</p>
</div>

</div>

```python
# --- تعریف ساختار داده‌ها ---

# کلاس ساده برای نمایش یک سند با محتوا و پاراگراف‌های آن
class Document:
    def __init__(self, doc_id, content):
        self.doc_id = doc_id
        # تقسیم محتوا به پاراگراف‌ها بر اساس خطوط خالی
        self.paragraphs = [p.strip() for p in content.split('\n\n') if p.strip()]
        # ایجاد لیستی از کلمات در هر پاراگراف برای جستجوی سریع‌تر
        self.paragraph_words = [p.lower().split() for p in self.paragraphs]
    
    def __str__(self):
        return f"سند {self.doc_id}: {self.paragraphs[0][:50]}..."

# نمونه‌ای از مجموعه اسناد کوچک
documents = [
    Document(1, "مهمانان در مهمانی مست بودند. میزبان مسئولیت مواد الکلی را بر عهده داشت. مهمانان پس از ترک مهمانی تصادف کردند."),
    Document(2, "میزبان مهمانی دیگری برگزار کرد. در این مهمانی، مهمانان الکل مصرف نکردند. مسئولیت میزبان در این مورد کمتر بود."),
    Document(3, "قوانین مربوط به مسئولیت میزبان در بسیاری از کشورها وجود دارد. اگر مهمان مست باشد و حادثه‌ای رخ دهد، میزبان مسئول است."),
    Document(4, "مهمان و میزبان توافق‌نامه‌ای قبل از مهمانی امضا کردند. این توافق‌نامه مسئولیت‌ها را مشخص می‌کرد."),
    Document(5, "در این مهمانی، میزبان به مهمانان الکل نداد. با این حال، یکی از مهمانان مشروب خود را آورده بود و مست شد.")
]

# --- پیاده‌سازی الگوریتم‌های جستجو ---

def boolean_search(documents, query_terms):
    """
    جستجوی بولی ساده: اسنادی را برمی‌گرداند که تمام واژه‌های پرسمان را دارند.
    """
    results = []
    for doc in documents:
        # تبدیل کل سند به حروف کوچک برای جستجوی حساس‌به‌حروف بزرگ/کوچک نباشیم
        doc_text = ' '.join(doc.paragraphs).lower()
        if all(term.lower() in doc_text for term in query_terms):
            results.append(doc)
    return results

def proximity_search_paragraph(doc, term1, term2):
    """
    بررسی می‌کند که آیا دو واژه در یک پاراگراف از سند وجود دارند یا نه.
    این پیاده‌سازی برای عملگر /p (همان پاراگراف) است.
    """
    for words in doc.paragraph_words:
        has_term1 = any(term1.lower() in word for word in words)
        has_term2 = any(term2.lower() in word for word in words)
        if has_term1 and has_term2:
            return True
    return False

def extended_boolean_search(documents, term1, term2, operator):
    """
    جستجوی بولی گسترش‌یافته با عملگرهای نزدیکی.
    """
    results = []
    for doc in documents:
        if operator == "AND":
            # جستجوی بولی ساده
            doc_text = ' '.join(doc.paragraphs).lower()
            if term1.lower() in doc_text and term2.lower() in doc_text:
                results.append(doc)
        elif operator == "/p":
            # جستجوی نزدیکی در همان پاراگراف
            if proximity_search_paragraph(doc, term1, term2):
                results.append(doc)
    return results

def term_frequency(doc, term):
    """
    تعداد تکرار یک واژه در یک سند را می‌شمارد.
    """
    count = 0
    for paragraph in doc.paragraphs:
        words = paragraph.lower().split()
        count += sum(1 for word in words if term.lower() in word)
    return count

def ranked_retrieval(documents, query_terms):
    """
    بازیابی رتبه‌بندی‌شده: اسناد را بر اساس مجموع فراوانی واژه‌های پرسمان رتبه‌بندی می‌کند.
    """
    doc_scores = []
    for doc in documents:
        score = sum(term_frequency(doc, term) for term in query_terms)
        if score > 0:  # فقط اسنادی که حداقل یکی از واژه‌ها را دارند
            doc_scores.append((doc, score))
    
    # مرتب‌سازی اسناد بر اساس امتیاز (از بیشترین به کمترین)
    doc_scores.sort(key=lambda x: x[1], reverse=True)
    return doc_scores

# --- اجرای مثال‌ها ---

if __name__ == "__main__":
    
    # مثال ۱: مقایسه جستجوی بولی ساده و جستجوی نزدیکی
    print("="*60)
    print("مثال ۱: مقایسه جستجوی بولی ساده و جستجوی نزدیکی")
    term1 = "میزبان"
    term2 = "مهمان"
    
    # جستجوی بولی ساده (AND)
    boolean_results = boolean_search(documents, [term1, term2])
    print(f"\nنتایج جستجوی بولی ساده ({term1} AND {term2}):")
    for doc in boolean_results:
        print(f"  - {doc}")
    
    # جستجوی نزدیکی (/p)
    proximity_results = extended_boolean_search(documents, term1, term2, "/p")
    print(f"\nنتایج جستجوی نزدیکی ({term1} /p {term2}):")
    for doc in proximity_results:
        print(f"  - {doc}")
    
    # مثال ۲: مقایسه مدل بولی و مدل رتبه‌بندی‌شده
    print("\n" + "="*60)
    print("مثال ۲: مقایسه مدل بولی و مدل رتبه‌بندی‌شده")
    query_terms = ["میزبان", "مسئولیت"]
    
    # نتایج مدل بولی
    boolean_results = boolean_search(documents, query_terms)
    print(f"\nنتایج مدل بولی (بدون رتبه‌بندی):")
    for doc in boolean_results:
        print(f"  - {doc}")
    
    # نتایج مدل رتبه‌بندی‌شده
    ranked_results = ranked_retrieval(documents, query_terms)
    print(f"\nنتایج مدل رتبه‌بندی‌شده (با امتیاز):")
    for doc, score in ranked_results:
        print(f"  - امتیاز {score}: {doc}")
    
    # مثال ۳: نمایش فراوانی واژه در اسناد مختلف
    print("\n" + "="*60)
    print("مثال ۳: فراوانی واژه در اسناد مختلف")
    term = "میزبان"
    print(f"\nفراوانی واژه '{term}' در اسناد:")
    for doc in documents:
        freq = term_frequency(doc, term)
        print(f"  - {doc}: {freq} بار")
```

<p style="font-size:18pt; text-align:center;">
<span id="section15" style="font-size:24pt; font-weight:bold; color:#8549EB;">جمع‌بندی فصل ۱: بازیابی بولی</span> 📌
</p>

<p style="font-size:16pt; text-align:right;">
در این فصل با مفاهیم پایه‌ای بازیابی اطلاعات آشنا شدیم:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>مدل بولی را به‌عنوان سنگ بنای سیستم‌های جستجو معرفی کردیم. این مدل همان منطقی است که در کتابخانه‌ها با جستجوی «و» و «یا» استفاده می‌کنیم - مثلاً جستجوی «هوش مصنوعی و یادگیری ماشین» در پایگاه داده مقالات علمی.</li>
<li>ماتریس وقوع بولی و محدودیت‌های آن را بررسی کردیم. تصور کنید یک کتابخانه با هزاران کتاب دارید و می‌خواهید بدانید کدام کتاب‌ها کلمات کلیدی خاصی دارند - ماتریس وقوع این کار را انجام می‌دهد، اما برای کتابخانه‌های بزرگ بسیار ناکارآمد است!</li>
<li>فهرست معکوس را به‌عنوان راه‌حلی کارآمد برای ذخیره‌سازی و جست‌وجو معرفی کردیم. این همان فهرستی است که در انتهای کتاب‌ها می‌بینیم، اما برای جستجوی کامپیوتری بهینه شده است. گوگل در نسخه‌های اولیه خود از این روش استفاده می‌کرد.</li>
<li>الگوریتم‌های اشتراک (AND)، ادغام (OR) و بهینه‌سازی ترتیب پردازش را پیاده‌سازی کردیم. وقتی در اسنپ فود «پیتزا و نوشابه» را جستجو می‌کنید، در واقع از الگوریتم AND استفاده می‌کنید - فقط نتایجی نمایش داده می‌شوند که هر دو کلمه را داشته باشند.</li>
<li>تفاوت مدل بولی ساده، بولی گسترش‌یافته، و بازیابی رتبه‌بندی‌شده را بررسی کردیم. این همان تفاوتی است که بین جستجوی ساده در فایل‌های ویندوز و جستجوی پیشرفته در گوگل وجود دارد.</li>
</ul>

<div style="background:#e8f4fd; border-radius:8px; padding:15px; margin:15px 0;">
<p style="font-size:16pt; text-align:right;">
<b>بیایید این مفاهیم را با چند مثال واقعی مقایسه کنیم:</b>
</p>
<hr style="border-top: 1px solid #ccc; margin: 10px 0;">

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازیابی بولی ساده:</b> تصور کنید در دیجی‌کالا به دنبال «گوشی سامسونگ با حافظه ۲۵۶ گیگابایت و قیمت زیر ۱۰ میلیون تومان» می‌گردید. این جستجو دقیقاً از منطق بولی استفاده می‌کند: (گوشی AND سامسونگ) AND (حافظه ۲۵۶) AND (قیمت زیر ۱۰ میلیون). سیستم ابتدا تمام گوشی‌های سامسونگ را پیدا می‌کند، سپس از میان آن‌ها موارد با حافظه ۲۵۶ را جدا کرده و در نهایت آن‌هایی را نمایش می‌دهد که قیمت مناسب داشته باشند. نتیجه یا هست یا نیست؛ هیچ رتبه‌بندی وجود ندارد.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازیابی بولی گسترش‌یافته:</b> یک وکیل در پایگاه داده قوانین و آراء دادگاه به دنبال یک مورد خاص می‌گردد. او جستجوی زیر را انجام می‌دهد: <br/>
«<code>مسئولیت قراردادی NEAR/10 خسارت</code>» <br/>
در اینجا، `NEAR/10` یک عملگر گسترش‌یافته است که به سیستم می‌گوید فقط سندهایی را پیدا کن که عبارت «مسئولیت قراردادی» و «خسارت» در فاصله کمتر از ۱۰ کلمه از هم قرار دارند. این کار از پیدا کردن سندهایی که این دو کلمه را به صورت پراکنده در آن آورده‌اند، بسیار دقیق‌تر است.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازیابی رتبه‌بندی‌شده:</b> شما در گوگل عبارت «بهترین کافه‌های قصرالدشت» را جستجو می‌کنید. گوگل هزاران صفحه را پیدا می‌کند که این کلمات را دارند. اما به جای نمایش همه آن‌ها، آن‌ها را <b>رتبه‌بندی</b> می‌کند. نتیجه اول احتمالاً کافه‌ای است که امتیاز بالایی دارد، نظرات مثبت زیادی گرفته، به موقعیت مکانی شما نزدیک است و در وب‌سایت‌های معتبر به آن اشاره شده است. نتیجه پنجم ممکن است کافه خوبی باشد اما امتیاز کمتری دارد. این مدل، نتایج را بر اساس «مرتبطی» به شما نمایش می‌دهد، نه فقط بر اساس تطابق کلمات.
</p>

</div>

<p style="font-size:16pt; text-align:right;">
این فصل پایه‌های سخت‌افزاری و ساختاری سیستم‌های جست‌وجو را فراهم می‌کند. در فصل‌های بعدی، با مفاهیمی مانند:
</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>وزن‌دهی به واژه‌ها (tf-idf) - همان روشی که گوگل تشخیص می‌دهد کلمه «گوشی» در یک صفحه مهم‌تر از کلمه «و» است</li>
<li>مدل فضای برداری - تکنیکی که به سیستم‌های جستجو اجازه می‌دهد مفاهیم مشابه را پیدا کنند، حتی اگر کلمات یکسانی استفاده نشده باشد</li>
<li>ارزیابی سیستم‌ها (دقت، بازیابی، F1) - معیارهایی برای سنجش اینکه آیا موتور جستجوی ما واقعاً خوب کار می‌کند یا نه</li>
<li>خوشه‌بندی و طبقه‌بندی متن - روشی که گوگل اخبار را به دسته‌های ورزشی، سیاسی و اقتصادی تقسیم می‌کند</li>
</ul>
<p style="font-size:16pt; text-align:right;">
آشنا خواهیم شد — که همگی بر همین پایهٔ فهرست معکوس بنا شده‌اند.
</p>

    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Web</code> = <b>10000</b> و

  <div style="margin-bottom: 15px;">
    <span style="font-size: 18pt; font-weight: bold; color: #1B263B;">

    </span>
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Postings Lists</code>) 
    را بدون مرتب‌سازی 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Sorting</code>) 
    شناسه‌های سند 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">DocID</code>) 
    در فایل ذخیره کند، الگوریتم استاندارد دو اشاره‌گره 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Standard Merge Algorithm</code>) 
    در زمان اجرای پرسمانِ 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">A AND B</code> 
    چه رفتاری از خود نشان می‌دهد؟
  </div>

    </span>
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">
    دو کلمه <b>A</b> و <b>B</b> به ترتیب دارای لیست ارجاع 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Postings Lists</code>) 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">[1, 3, 5, 7, 9, 11]</code> 
    و 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">[2, 3, 4, 5, 11, 15]</code> 
    هستند. برای یافتن اشتراک این دو لیست با الگوریتم استاندارد ادغام 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Standard Merge Algorithm</code>)، 
    اشاره‌گرها دقیقاً چند مقایسه 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Comparison</code>) 
    انجام می‌دهند تا به انتهای لیست برسند؟
    <br>
    مسیر حرکت آن‌ها را ردیابی کنید.
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    را وارد کرده است. با استفاده از عملگرهای مجموعه، نحوه پردازش این پرسمان روی لیست‌های ارجاع 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Postings Lists</code>) 
    را شرح دهید.
    <br>
    چرا هزینه پردازش عملگر 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">NOT</code> 
    در موتورهای جستجو معمولاً بالاتر از 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">AND</code>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">DocID</code>) 
    را ذخیره می‌کند.
    <br>
    برای پشتیبانی از پرسمان‌های عبارتی دقیق 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Phrase Queries</code>) 
    مانند جستجوی عبارت 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">"Machine Learning"</code> 
    (دقیقاً پشت سر هم)، ساختار ذخیره‌سازی لیست‌های ارجاع 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Postings Lists</code>) 
    باید چه تغییری کند و چه اطلاعات اضافه‌ای باید نگهداری شود؟
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">OR</code> 
    هستند (مثل

    <br>
    صحت و کارایی این پیشنهاد را از نظر پیچیدگی زمانی

    </span>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    چگونه می‌توان با استفاده از عملگرهای مجاورت 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Proximity Operators</code>) 
    پرسمانی نوشت که کلمه «دانشگاه» و «تهران» را با فاصله حداکثر <b>2</b> کلمه از یکدیگر پیدا کند؟
    <br>
    پیاده‌سازی این عملگر در زمان اجرا چه سرباری دارد؟
  </div>

    </span>
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Exact Match</code>) 
    است؟
    <br>
    یک سناریوی ساده مثال بزنید که در آن کاربری به دنبال یک سند خاص است، اما به دلیل ماهیت <b>0</b> و <b>1</b> بودن این مدل، با خطای منطقی مواجه شده و هیچ نتیجه‌ای دریافت نمی‌کند.
  </div>

  <div style="margin-bottom: 15px;">
    <span style="font-size: 18pt; font-weight: bold; color: #1B263B;">
      مسئله 1.13: توزیع نمایه وارونه و مقایسه معماری‌های پارتیشن‌بندی ⭐
    </span>
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">
    فرض کنید حجم اسناد به قدری زیاد است که نمایه وارونه روی یک سرور جا نمی‌شود و باید آن را بین چندین سرور توزیع کنید 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Sharding</code>). 
    دو رویکرد 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Term-partitioning</code> 
    (تقسیم بر اساس کلمات) و 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Document-partitioning</code> 
    (تقسیم بر اساس اسناد) را با هم مقایسه کنید.
    <br>
    برای اجرای پرسمان 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">A AND B</code>، 
    شبکه در کدام روش بیشتر درگیر می‌شود و گلوگاه 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Bottleneck</code>) 
    سیستم کجاست؟

  <div style="margin-bottom: 15px;">

    </span>
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Skip Pointers</code> 
    همیشه و در هر شرایطی زمان اشتراک لیست‌ها را کاهش می‌دهد.
    <br>
    یک مثال نقض عددی طراحی کنید (شامل طول گام پرش و عناصر دو لیست) که در آن، استفاده از پرشگرها نه تنها سرعت را بالا نبرده، بلکه به دلیل سربار بررسی شرط‌های پرش، تعداد کل مقایسه‌ها را نسبت به حالت خطی معمولی افزایش می‌دهد.

    </span>
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">
    در یک موتور جستجوی واقعی، پرسمان‌های تکراری زیادی وجود دارد. اگر بخواهید نتایج پرسمان‌های رایج را کش کنید، آیا از نظر معماری بهتر است نتیجه نهایی پرسمان 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">DocID</code>های خروجی) 
    را کش کنید یا لیست‌های ارجاع 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Postings Lists</code>) 
    کلمات آن پرسمان را در 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">RAM</code> 
    نگه دارید؟
    <br>
    با تحلیل 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Trade-off</code> 
    بین فضای حافظه و سرعت به‌روزرسانی سیستم به این چالش پاسخ دهید.
  </div>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">

    هیچ نتیجه‌ای به کاربر نمایش داده نمی‌شود 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Famine</code>)، 
    در حالی که لپ‌تاپ‌های گیمینگ ایسوس <b>16</b> گیگابایتی موجود هستند که کاربر احتمالاً یکی را می‌خرید.
    <br>
    از منظر تئوری بازیابی، توضیح دهید چرا مدل 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Boolean</code> 
    در این سناریو به بن‌بست خورده است و چگونه گذر به مدل 
    <code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Ranked Retrieval</code> 
    و جستجوی متن آزاد 
    (<code style="background: #f1f1f1; padding: 2px 6px; border-radius: 6px; font-family: Consolas, 'Courier New', monospace; color: #6D2E46; direction: ltr; unicode-bidi: embed; display: inline-block;">Free text</code>) 
    این مشکل را حل می‌کند؟

مقدمه و طرح مسئله

<div style="text-align:justify; margin-bottom:15px;">
در سیستم‌های بازیابی اطلاعات <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Information Retrieval)</span></b>، رویکرد الگوریتمی به فرآیند جستجو، تعیین‌کننده اصلی کارایی سیستم است. برای یافتن یک عبارت خاص در مجموعه‌ای از اسناد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Corpus)</span></b>، دو پارادایم کلی وجود دارد: جستجوی ترتیبی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Sequential)</span></b> و استفاده از ساختارهای داده پیش‌پردازش‌شده. این مسئله به بررسی تفاوت بنیادین این دو روش و تحلیل مقیاس‌پذیری آن‌ها می‌پردازد.
</div>

</div>

<div style="text-align:justify; margin-bottom:15px;">
در رویکرد جستجوی خطی، سیستم در زمان ثبت درخواست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Query Time)</span></b>، محتوای اسناد را به صورت کاراکتر به کاراکتر یا کلمه به کلمه پیمایش می‌کند تا عبارت مورد نظر را بیابد. عملکرد ابزارهایی مانند <code style="background:#F3F3F3; padding:2px 6px; border-radius:6px; color:#6D2E46; font-family:monospace; direction:ltr; unicode-bidi:embed;">grep</code> در سیستم‌عامل‌های مبتنی بر یونیکس بر این اساس استوار است.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">پیچیدگی زمانی:</b> اگر حجم کل داده‌ها (تعداد کل کلمات یا کاراکترهای موجود در تمامی اسناد) را برابر با <b style="color:#6D2E46;">N</b> در نظر بگیریم، پیچیدگی زمانی این رویکرد از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(N)</span></b> خواهد بود. این بدان معناست که با افزایش حجم متون، زمان اجرای الگوریتم نیز به صورت خطی افزایش می‌یابد.
</div>

<div style="font-size:18pt; font-weight:bold; margin-top:25px; margin-bottom:10px;">
2. رویکرد نمایه وارونه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Inverted Index)</span></b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
نمایه وارونه ساختار داده‌ای است که پیش از زمان ثبت درخواست و طی فاز پیش‌پردازش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Pre-processing)</span></b> ایجاد می‌شود. در این رویکرد، یک دیکشنری از تمامی کلمات یکتای موجود در اسناد تشکیل شده و هر کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Term)</span></b> به لیستی از شناسه‌های اسناد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Document IDs)</span></b> که آن کلمه در آن‌ها رخ داده است <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings List)</span></b>، متصل می‌شود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">پیچیدگی زمانی:</b> هنگام جستجو، سیستم نیازی به پیمایش متن اسناد ندارد؛ بلکه مستقیماً به دیکشنری مراجعه می‌کند. اگر از ساختار جدول درهم‌سازی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Hash Table)</span></b> برای پیاده‌سازی دیکشنری استفاده شود، زمان جستجوی کلمه از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1)</span></b> و در صورت استفاده از درخت‌های جستجو، از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(log K)</span></b> خواهد بود (که <b style="color:#6D2E46;">K</b> تعداد کلمات یکتای دیکشنری است). در این حالت، هزینه زمانی اصلی تنها مربوط به واکشی و اشتراک‌گیری لیست‌های پیوندی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span></b> است و جستجو کاملاً مستقل از حجم کل داده‌ها (<b style="color:#6D2E46;">N</b>) انجام می‌پذیرد.
</div>

<div style="font-size:18pt; font-weight:bold; margin-top:25px; margin-bottom:10px;">
3. نقطه بحرانی در مقیاس داده‌ها <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Scalability Threshold)</span></b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
تفاوت این دو رویکرد از نظر پیچیدگی زمانی، وابستگی مستقیمی به مقیاس و حجم داده‌ها دارد:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مقیاس خرد و محلی (داده‌های در سطح مگابایت یا گیگابایت پایین):</b> در این مقیاس (مانند جستجو در فایل‌های متنی یک سیستم شخصی یا سورس‌کدهای یک پروژه)، استفاده از جستجوی خطی کاملاً بهینه و منطقی است. در این حالت، سربارِ <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Overhead)</span></b> محاسباتی و حافظه‌ای لازم برای ایجاد و نگهداری ساختار نمایه وارونه توجیه فنی ندارد.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مقیاس کلان و سازمانی (داده‌های وب، ترابایت و فراتر از آن):</b> با ورود به مقیاس کلان اطلاعاتی، تفاوت پیچیدگی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(N)</span></b> در برابر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1)</span></b> از نظر عملیاتی حیاتی می‌شود. در حجم داده‌های وب یا پایگاه‌داده‌های عظیم، جستجوی خطی منجر به زمان‌های پاسخ‌گویی غیرقابل قبول (از چند ساعت تا چند روز) می‌شود و عملاً سیستم را ناکارآمد می‌سازد. در این نقطه، پردازش خطی از نظر محاسباتی غیرعملی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Intractable)</span></b>تلقی شده و استفاده از نمایه وارونه از یک «انتخاب بهینه‌ساز» به یک «الزام معماری» برای پاسخ‌گویی درنگ‌درنگ <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">    (Real-time)</span></b> تبدیل می‌گردد.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری: تفاوت اصلی این دو روش در انتقال بار محاسباتی است؛ جستجوی خطی تمام هزینه را در زمان جستجو <span style="direction:ltr; unicode-bidi:embed;">(Query Time)</span> پرداخت می‌کند، در حالی که نمایه وارونه، با استفاده از حافظه بیشتر، هزینه محاسباتی را به فاز پیش‌پردازش <span style="direction:ltr; unicode-bidi:embed;">(Index Time)</span> منتقل کرده و زمان جستجو را به شدت کاهش می‌دهد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.2</b>: ماتریس تلاقی ترم-سند و تحلیل پدیده خلوت بودن <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Sparsity)</span></b>
</div>

<div style="font-size:18pt; font-weight:bold; margin-top:25px; margin-bottom:10px;">
1. تشکیل ماتریس تلاقی ترم-سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Term-Document Incidence Matrix)</span></b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای درک این ساختار، ابتدا فرض می‌کنیم سه سند کوتاه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">D1</span></b> تا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">D3</span></b>) با محتوای زیر در اختیار داریم و می‌خواهیم حضور یا عدم حضور <b style="color:#6D2E46;">5</b> کلمه کلیدی مشخص را در آن‌ها بررسی کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">سند <span style="direction:ltr; unicode-bidi:embed;">D1</span>:</b> «مبانی بازیابی اطلاعات»<br>
<b style="color:#6D2E46;">سند <span style="direction:ltr; unicode-bidi:embed;">D2</span>:</b> «طراحی موتور جستجو»<br>
<b style="color:#6D2E46;">سند <span style="direction:ltr; unicode-bidi:embed;">D3</span>:</b> «الگوریتم‌های بازیابی وب»
</div>

<div style="text-align:justify; margin-bottom:15px;">
کلمات کلیدی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Terms)</span></b> انتخاب شده: بازیابی، اطلاعات، موتور، جستجو، وب.
<br>
ماتریس تلاقی به صورت یک ماتریس دودویی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Binary)</span></b> تشکیل می‌شود که در آن سطرها نشان‌دهنده کلمات کلیدی، ستون‌ها نشان‌دهنده اسناد و درایه‌های ماتریس مقدار <b style="color:#6D2E46;">1</b> (حضور کلمه در سند) یا <b style="color:#6D2E46;">0</b> (عدم حضور) را می‌گیرند:
</div>

<div style="font-size:18pt; font-weight:bold; margin-top:25px; margin-bottom:10px;">
2. چالش ذخیره‌سازی در سیستم‌های واقعی و پدیده خلوت بودن <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Sparsity)</span></b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
در یک سیستم بازیابی اطلاعات واقعی (نظیر موتورهای جستجوی وب)، مقیاس این ماتریس به شدت بزرگ است. فرض کنید مجموعه‌ای متشکل از یک میلیون سند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">N = 1,000,000</span></b>) داریم و واژه‌نامه یکتای ما شامل <b style="color:#6D2E46;">500,000</b> کلمه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">M = 500,000</span></b>) است.
</div>

<div style="text-align:justify; margin-bottom:15px;">
اندازه ماتریس تلاقی برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">M &times; N</span></b> خواهد بود؛ یعنی به یک ماتریس با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">5 &times; 10<sup>11</sup></span></b> درایه (نصف یک تریلیون سلول) نیاز داریم. اگر هر درایه تنها <b style="color:#6D2E46;">1</b> بیت حافظه اشغال کند، ذخیره این ماتریس به ده‌ها گیگابایت حافظه اصلی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(RAM)</span></b> یکپارچه نیاز دارد.
</div>

<div style="text-align:justify; margin-bottom:15px;">
اما مشکل اصلی کجاست؟ <b style="color:#6D2E46;">پدیده خلوت بودن <span style="direction:ltr; unicode-bidi:embed;">(Sparsity)</span></b>.
<br>
هر سند متنی، معمولاً تنها شامل کسر بسیار کوچکی از کل واژگان موجود در زبان است (مثلاً یک سند <b style="color:#6D2E46;">1000</b> کلمه‌ای، نهایتاً از <b style="color:#6D2E46;">1000</b> کلمه یکتای واژه‌نامه <b style="color:#6D2E46;">500,000</b> کلمه‌ای استفاده کرده است). این یعنی بیش از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">99.9%</span></b> از درایه‌های ماتریس تلاقیِ ترم-سند، مقدار <b style="color:#6D2E46;">0</b> دارند. اختصاص دادن منابع عظیم حافظه برای ذخیره میلیاردها صفر، از نظر معماری سیستم و پیچیدگی فضایی، کاملاً غیرمنطقی و ناکارآمد است.
</div>

<div style="font-size:18pt; font-weight:bold; margin-top:25px; margin-bottom:10px;">
3. مدیریت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Sparsity</span></b> از طریق نمایه وارونه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Inverted Index)</span></b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای حل مشکل خلوت بودن ماتریس، نمایه وارونه رویکردی هوشمندانه اتخاذ می‌کند: <b style="color:#6D2E46;">تغییر از ذخیره‌سازی وضعیتی به ذخیره‌سازی رخدادی.</b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
به جای ذخیره کل ماتریس (شامل تمامی صفرها و <b style="color:#6D2E46;">1</b>ها)، سیستم نمایه وارونه تنها درایه‌هایی که مقدار <b style="color:#6D2E46;">1</b> دارند را ثبت می‌کند. این کار با استفاده از ساختار داده‌های لیست پیوندی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Linked List)</span></b> یا آرایه‌های پویا انجام می‌شود. در این ساختار:
<br>
هر کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Term)</span></b> به عنوان یک رکورد در دیکشنری ذخیره می‌شود و به جای یک سطر کامل از ماتریس، تنها یک <b style="color:#6D2E46;">لیست رخداد <span style="direction:ltr; unicode-bidi:embed;">(Postings List)</span></b> به آن متصل می‌گردد. این لیست صرفاً شامل شناسه اسنادی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Doc IDs)</span></b> است که کلمه در آن‌ها وجود دارد.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;"">نتیجه‌گیری ساختاری: </span>
استفاده از نمایه وارونه مشکل <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Sparsity</span></b> را به طور کامل برطرف می‌کند، زیرا هیچ حافظه‌ای برای ثبت عدم حضور (صفرها) تخصیص داده نمی‌شود. این بهینه‌سازی، پیچیدگی فضایی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Space Complexity)</span></b> سیستم را به طرز چشم‌گیری کاهش داده و امکان پیاده‌سازی سیستم‌های بازیابی در مقیاس وب را فراهم می‌آورد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.3</b>: معماری و اجزای اصلی نمایه وارونه
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه
</div>

<div style="text-align:justify; margin-bottom:24px;">
نمایه وارونه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Inverted Index)</span></b> به عنوان هسته مرکزی سیستم‌های بازیابی اطلاعات، از دو ساختار داده‌ی مجزا اما کاملاً وابسته به یکدیگر تشکیل شده است: واژه‌نامه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Dictionary)</span></b> و لیست‌های رخداد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span></b>. تفکیک این دو بخش، امکان بهینه‌سازی جستجو از نظر پیچیدگی زمانی و مدیریت حافظه را فراهم می‌آورد. در ادامه، ماهیت و وظیفه دقیق هر بخش تشریح می‌گردد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. واژه‌نامه <span style="direction:ltr; unicode-bidi:embed;">(Dictionary)</span> یا <span style="direction:ltr; unicode-bidi:embed;">(Lexicon)</span>
</div>

<div style="text-align:justify; margin-bottom:12px;">
واژه‌نامه یک ساختار داده جستجوپذیر (نظیر جدول درهم‌سازی، درخت جستجوی دودویی یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B-Tree</span></b>) است که شامل تمامی کلمات یکتای استخراج‌شده از کل مجموعه اسناد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Corpus)</span></b> است.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">محتویات:</b> این ساختار برای هر کلمه کلیدی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Term)</span></b>، خود آن کلمه، فرکانس تکرار آن در کل اسناد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Document Frequency)</span></b> و یک اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Pointer)</span></b> به لیست رخدادِ مربوط به آن کلمه را ذخیره می‌کند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">وظیفه در فرآیند جستجو:</b> واژه‌نامه به عنوان «دروازه ورود» به سیستم عمل می‌کند. وظیفه اصلی آن، پاسخ‌گویی درنگ‌درنگ <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Real-time)</span></b> به این سوال است که: <i>«آیا کلمه مورد جستجو اصلاً در اسناد ما وجود دارد یا خیر؟ و اگر وجود دارد، اطلاعات آن کجاست؟»</i> از آنجا که سرعت در این مرحله حیاتی است، واژه‌نامه معمولاً به صورت کامل در حافظه اصلی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(RAM)</span></b> بارگذاری می‌شود تا زمان جستجوی کلمه در آن از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1)</span></b> یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(log K)</span></b> باشد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. لیست‌های رخداد <span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span>
</div>

<div style="text-align:justify; margin-bottom:12px;">
لیست‌های رخداد، مجموعه‌ای از آرایه‌های پویا یا لیست‌های پیوندی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Linked Lists)</span></b> هستند. هر کلمه در واژه‌نامه، یک لیست رخداد اختصاصی برای خود دارد.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">محتویات:</b> هر لیست شامل شناسه‌های اسنادی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Document IDs)</span></b> است که آن کلمه در آن‌ها مشاهده شده است. در سیستم‌های پیشرفته‌تر، علاوه بر شناسه سند، اطلاعاتی نظیر موقعیت دقیق کلمه در متن <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Position)</span></b> و میزان تکرار کلمه در آن سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Term Frequency)</span></b> نیز در این لیست‌ها ذخیره می‌شود تا در فرآیند رتبه‌بندی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Ranking)</span></b> استفاده گردد. این شناسه‌ها همواره به صورت صعودی مرتب <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Sorted)</span></b> می‌شوند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">وظیفه در فرآیند جستجو:</b> لیست رخداد، حامل «پاسخ نهایی» است. وظیفه این بخش، ارائه دقیق آدرس تمامی اسنادی است که باید به عنوان نتیجه جستجو به کاربر نمایش داده شوند. به دلیل حجم بالای این لیست‌ها، معمولاً روی دیسک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Hard Drive/SSD)</span></b> ذخیره شده و تنها در زمان نیاز، واکشی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Fetch)</span></b> می‌شوند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. تشریح مرحله‌به‌مرحله فرآیند پیدا کردن یک کلمه
</div>

<div style="text-align:justify; margin-bottom:12px;">
برای درک بهتر تعامل این دو جزء، فرآیند پردازش یک تقاضای جستجو <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Query)</span></b> را به صورت الگوریتمی بررسی می‌کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام اول (جستجو در <span style="direction:ltr; unicode-bidi:embed;">Dictionary</span>):</b> کاربر کلمه‌ای (مثلاً «الگوریتم») را جستجو می‌کند. سیستم ابتدا به واژه‌نامه (که در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">RAM</span></b> قرار دارد) مراجعه کرده و کلمه «الگوریتم» را در میان کلیدها جستجو می‌کند.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام دوم (ارزیابی وضعیت):</b> اگر کلمه در واژه‌نامه یافت نشد، سیستم بلافاصله جستجو را خاتمه داده و اعلام می‌کند سندی یافت نشد. اگر کلمه پیدا شد، سیستم اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Pointer)</span></b> متصل به آن کلمه را می‌خواند.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام سوم (مراجعه به <span style="direction:ltr; unicode-bidi:embed;">Postings List</span>):</b> سیستم با استفاده از اشاره‌گر، به سراغ حافظه (دیسک) رفته و دقیقاً همان لیست رخداد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings List)</span></b> مختص به کلمه «الگوریتم» را واکشی می‌کند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">گام چهارم (بازگرداندن نتایج):</b> شناسه‌های اسناد موجود در آن لیست (مثلاً اسناد شماره 12، 45 و 89) خوانده شده و محتوای آن اسناد به عنوان نتیجه نهایی به کاربر ارجاع داده می‌شود.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری: طراحی دومرحله‌ای نمایه وارونه باعث می‌شود تا سیستم نیازی به پیمایش متن اسناد نداشته باشد. دیکشنری وظیفه «مکان‌یابی سریع اطلاعات» و لیست رخداد وظیفه «ذخیره‌سازی جامع داده‌های بازیابی» را بر عهده دارد. ادغام این دو، پردازش کارآمد روی کلان‌داده‌ها را ممکن می‌سازد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.4</b>: نقش پیش پردازش زبانی و توکنایزیشن در ساخت نمایه
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه
</div>

<div style="text-align:justify; margin-bottom:24px;">
در معماری سیستم‌های بازیابی اطلاعات، مرحله ساخت نمایه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Index Construction)</span></b> نمی‌تواند مستقیماً روی جریان خام کاراکترها <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Raw Text)</span></b> اعمال شود. پیش از آنکه سیستم بتواند شناسه‌های اسناد را به لیست‌های رخداد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span></b> اضافه کند، باید دقیقاً بداند که چه «موجودیت‌هایی» قرار است به عنوان کلید در واژه‌نامه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Dictionary)</span></b> ذخیره شوند. اینجاست که فاز پیش‌پردازش شامل نشانه‌سازی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Tokenization)</span></b> و ماژول‌های زبانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Linguistic Modules)</span></b> به عنوان یک لایه حیاتی وارد عمل می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. نشانه‌سازی یا تقطیع <span style="direction:ltr; unicode-bidi:embed;">(Tokenization)</span>
</div>

<div style="text-align:justify; margin-bottom:12px;">
فرآیند نشانه‌سازی، جریان پیوسته‌ای از کاراکترها را به واحدهای معنادار و مجزا به نام نشانه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Token)</span></b> یا کلمه تقسیم می‌کند. سیستم کامپیوتری به صورت پیش‌فرض درکی از کلمه ندارد و متن را صرفاً رشته‌ای از بایت‌ها می‌بیند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">ضرورت:</b> بدون نشانه‌سازی، سیستم نمی‌داند یک کلمه کجا شروع و کجا پایان می‌یابد. جدا کردن کلمات بر اساس فاصله‌ها <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Spaces)</span></b> و علائم نگارشی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Punctuation)</span></b> اولین قدم برای استخراج موجودیت‌های قابل جستجو است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. ماژول‌های پردازش زبانی <span style="direction:ltr; unicode-bidi:embed;">(Linguistic Modules)</span>
</div>

<div style="text-align:justify; margin-bottom:12px;">
پس از استخراج نشانه‌ها، این کلمات هنوز برای ورود به واژه‌نامه استاندارد نیستند. ماژول‌های زبانی شامل فرآیندهایی نظیر حذف کلمات توقف <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Stop-words Removal)</span></b>، یکسان‌سازی حروف <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Normalization)</span></b> و ریشه‌یابی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Stemming / Lemmatization)</span></b> هستند.
</div>

<div style="text-align:justify; margin-bottom:12px; padding-right:15px;">
<b style="color:#6D2E46;">ضرورت:</b> زبان‌های طبیعی پر از تنوع ظاهری برای یک مفهوم واحد هستند. هدف این ماژول‌ها، تبدیل کلمات مشتق‌شده و مختلف به یک «ریشه استاندارد» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Base Form)</span></b> است. این کار دو مزیت بزرگ دارد:
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:30px;">
<b style="color:#6D2E46;">1. کاهش حجم واژه‌نامه:</b> جلوگیری از ذخیره ده‌ها حالت مختلف برای یک کلمه.<br>
<b style="color:#6D2E46;">2. افزایش جامعیت <span style="direction:ltr; unicode-bidi:embed;">(Recall)</span>:</b> اطمینان از اینکه کاربر با جستجوی فعل مفرد، نتایج شامل افعال جمع یا مشتقات آن را نیز پیدا می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. بررسی یک مثال عملی: خروجی سیستم بدون فاز پیش‌پردازش
</div>

<div style="text-align:justify; margin-bottom:12px;">
برای درک فاجعه‌ای که در غیاب این مراحل رخ می‌دهد، فرض کنید سند زیر را در سیستم داریم و کاربر قصد جستجوی عبارت <b style="color:#6D2E46;">«کتاب»</b> را دارد:
</div>

<div style="text-align:justify; margin-bottom:20px; padding-right:15px;">
<b style="color:#6D2E46;">متن سند:</b> «او کتاب‌هایش را روی میز گذاشت.»
</div>

<div style="font-size:16pt; font-weight:bold; color:#1B263B; margin-bottom:10px;">
حالت الف: فقدان <span style="direction:ltr; unicode-bidi:embed;">Tokenization</span> و <span style="direction:ltr; unicode-bidi:embed;">Linguistic Modules</span>
</div>

<div style="text-align:justify; margin-bottom:12px;">
اگر هیچ پیش‌پردازشی انجام نشود، سیستم کل این جمله را به عنوان یک رشته یکپارچه می‌بیند.
</div>

<div style="text-align:justify; margin-bottom:20px; padding-right:15px;">
<b style="color:#6D2E46;">ورودی واژه‌نامه:</b> [او کتاب‌هایش را روی میز گذاشت.]<br>
<b style="color:#6D2E46;">نتیجه جستجو:</b> کاربر کلمه «کتاب» را جستجو می‌کند. چون سیستم کل جمله را به عنوان یک کلید واحد ثبت کرده است، هیچ تطابقی یافت نمی‌شود <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Miss)</span></b> و سند بازیابی نمی‌گردد.
</div>

<div style="font-size:16pt; font-weight:bold; color:#1B263B; margin-bottom:10px;">
حالت ب: انجام <span style="direction:ltr; unicode-bidi:embed;">Tokenization</span> اما فقدان <span style="direction:ltr; unicode-bidi:embed;">Linguistic Modules</span>
</div>

<div style="text-align:justify; margin-bottom:12px;">
حال فرض کنید سیستم کلمات را با فاصله‌ها جدا کرده است، اما ماژول‌های زبانی (مانند ریشه‌یابی و حذف علائم) اعمال نشده‌اند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">ورودی‌های واژه‌نامه:</b> <b style="color:#6D2E46;">[او]</b>، <b style="color:#6D2E46;">[کتاب‌هایش]</b>، <b style="color:#6D2E46;">[را]</b>، <b style="color:#6D2E46;">[روی]</b>، <b style="color:#6D2E46;">[میز]</b>، <b style="color:#6D2E46;">[گذاشت.]</b><br>
<b style="color:#6D2E46;">نتیجه جستجو:</b> مجدداً کاربر کلمه <b style="color:#6D2E46;">«کتاب»</b> را جستجو می‌کند. سیستم به واژه‌نامه مراجعه کرده و به دنبال تطابق دقیق <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Exact Match)</span></b> می‌گردد. چون در واژه‌نامه تنها کلمه <b style="color:#6D2E46;">[کتاب‌هایش]</b> وجود دارد، سیستم متوجه ارتباط معنایی این دو نمی‌شود و مجدداً سند بازیابی نمی‌گردد. (همچنین به نقطه متصل به کلمه <b style="color:#6D2E46;">[گذاشت.]</b> دقت کنید که به عنوان بخشی از کلمه ذخیره شده است).
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;";">نتیجه‌گیری ساختاری: </span>
عدم استفاده از نشانه‌سازی و ماژول‌های زبانی منجر به پدیده‌ای به نام <b style="color:#6D2E46;">«انفجار واژه‌نامه» <span style="direction:ltr; unicode-bidi:embed;">(Dictionary Blowup)</span></b> و <b style="color:#6D2E46;">«افت شدید جامعیت» <span style="direction:ltr; unicode-bidi:embed;">(Low Recall)</span></b> می‌شود. این مراحل پیش‌پردازشی، متون خام و ساختارنیافته را به مجموعه‌ای از مفاهیم استاندارد و قابل ردیابی تبدیل می‌کنند تا ساخت نمایه وارونه عملاً توجیه و کارایی داشته باشد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.5</b>: بهینه‌سازی پرسمان و ترتیب اجرای عملیات اشتراک <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Query Optimization and Intersection Order)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه و بیان منطق بهینه‌سازی
</div>

<div style="text-align:justify; margin-bottom:15px;">
در سیستم‌های بازیابی اطلاعات، عملگر منطقی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> (اشتراک) دارای خاصیت جابه‌جایی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Commutative)</span></b> و شرکت‌پذیری <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Associative)</span></b> است. این یعنی ترتیب قرارگیری پرانتزها در پرسمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Web AND Data) AND (Mining AND Search)</span></b> از نظر منطقی تفاوتی در نتیجه نهایی ایجاد نمی‌کند و سیستم می‌تواند پرانتزها را نادیده گرفته و پرسمان را به صورت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Web AND Data AND Mining AND Search</span></b> پردازش کند.
</div>

<div style="text-align:justify; margin-bottom:15px;">
هزینه محاسباتی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Cost)</span></b> اشتراک‌گیری بین دو لیست رخداد با طول‌های <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">x</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">y</span></b>، در بدترین حالت از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(x + y)</span></b> است. طول لیست رخدادِ حاصل از این اشتراک، هرگز نمی‌تواند از طول کوچک‌ترین لیست اولیه بزرگتر باشد؛ یعنی حداکثر برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">min(x, y)</span></b> خواهد بود.
</div>

<div style="text-align:justify; margin-bottom:24px;">
قاعده طلایی در بهینه‌سازی پرسمان این است: <b style="color:#6D2E46;">«برای کمینه‌سازی هزینه کل پردازش، همواره اشتراک‌گیری را از کوچک‌ترین لیست‌های رخداد (کمترین <span style="direction:ltr; unicode-bidi:embed;">Document Frequency</span>) آغاز کنید.»</b>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. مرتب‌سازی لیست‌ها بر اساس فرکانس سند <span style="direction:ltr; unicode-bidi:embed;">(DF)</span>
</div>

<div style="text-align:justify; margin-bottom:24px;">
در گام نخست، سیستم مقادیر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">DF</span></b> (طول لیست رخداد) هر کلمه را از واژه‌نامه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Dictionary)</span></b> واکشی کرده و آن‌ها را به صورت صعودی مرتب می‌کند:<br>
<b style="color:#6D2E46;">1. 2. 3. 4.</b> ترتیب بهینه برای پردازش به این شکل خواهد بود: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(((Search AND Mining) AND Data) AND Web)</span></b>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. اجرای مرحله‌به‌مرحله و تحلیل هزینه پردازش
</div>

<div style="text-align:justify; margin-bottom:12px;">
اکنون عملیات اشتراک را بر اساس ترتیب بهینه اجرا کرده و هزینه هر مرحله را محاسبه می‌کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مرحله اول: اشتراک دو لیست کوچک‌تر</b><br>
عملکرد: سیستم لیست‌های <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Search</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Mining</span></b> را واکشی و با هم اشتراک می‌گیرد.<br>
هزینه پردازش این مرحله: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(200 + 500) = O(700)</span></b><br>
خروجی: یک لیست میانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Temporary List)</span></b> ایجاد می‌شود. با توجه به اصل تقاطع، اندازه این لیست میانی حداکثر برابر با 200 خواهد بود (حالت بدبینانه). آن را <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b> می‌نامیم.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مرحله دوم: ادغام لیست میانی با کلمه بعدی</b><br>
عملکرد: سیستم لیست میانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b> را با لیست کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Data</span></b> اشتراک می‌گیرد.<br>
هزینه پردازش این مرحله: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(200 + 1000) = O(1200)</span></b><br>
خروجی: لیست میانی جدیدی به نام <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>2</sub></span></b> ایجاد می‌شود که اندازه آن مجدداً حداکثر 200 خواهد بود.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">مرحله سوم: ادغام نهایی با بزرگ‌ترین لیست</b><br>
عملکرد: در نهایت، سیستم لیست میانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>2</sub></span></b> را با لیست عظیم <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Web</span></b> اشتراک می‌گیرد.<br>
هزینه پردازش این مرحله: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(200 + 10000) = O(10200)</span></b><br>
خروجی: لیست نهایی اسناد (پاسخ پرسمان) به دست می‌آید.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. نتیجه‌گیری و مقایسه هزینه (چرا این روش بهینه است؟)
</div>

<div style="text-align:justify; margin-bottom:15px;">
مجموع هزینه پردازش در روش بهینه برابر است با:
<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">700 + 1200 + 10200 = 12100</span></b> عملیات.
</div>

<div style="text-align:justify; margin-bottom:12px;">
<b style="color:#6D2E46;">سناریوی فاجعه‌بار (بدون بهینه‌سازی):</b><br>
اگر سیستم بخواهد پرسمان را دقیقاً مطابق با پرانتزگذاری کاربر پردازش کند:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
1. ابتدا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Web</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Data</span></b> را اشتراک می‌گیرد: هزینه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(10000 + 1000) = O(11000)</span></b>. لیست میانی حاصل حداکثر طول 1000 خواهد داشت.<br>
2. سپس <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Mining</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Search</span></b> را اشتراک می‌گیرد: هزینه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(500 + 200) = O(700)</span></b>. لیست میانی حاصل حداکثر طول 200 خواهد داشت.<br>
3. در نهایت دو لیست میانی را اشتراک می‌گیرد: هزینه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1000 + 200) = O(1200)</span></b>.
</div>

<div style="text-align:justify; margin-bottom:24px;">
مجموع هزینه: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">11000 + 700 + 1200 = 12900</span></b> عملیات.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">نتیجه‌گیری ساختاری: </span>
در این مثال با وجود کوچک بودن اعداد، روش بهینه باعث صرفه‌جویی قابل توجهی شده است. در سیستم‌های واقعی که فرکانس کلماتی مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Web</span></b> ممکن است در مقیاس میلیارد باشد، شروع عملیات از کلمات عمومی و پرکاربرد منجر به ایجاد لیست‌های میانی غول‌پیکر و در نتیجه سرریز شدن حافظه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Memory Overflow)</span></b> و افت شدید سرعت سیستم می‌گردد. الگوریتم مرتب‌سازی بر اساس <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">DF</span></b> از بروز این مشکل جلوگیری می‌کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.6</b>: پیامد عدم مرتب‌سازی لیست‌های ارجاع در الگوریتم ادغام <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Consequences of Unsorted Postings in Merge Algorithm)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه و پیش‌فرض‌های الگوریتم استاندارد
</div>

<div style="text-align:justify; margin-bottom:24px;">
در قلب سیستم‌های بازیابی اطلاعات، کارایی خیره‌کننده عملگرهای منطقی (مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b>) وام‌دار «الگوریتم استاندارد ادغام دو اشاره‌گره» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Standard Two-Pointer Merge)</span></b> است. پیش‌نیاز قطعی و حیاتی برای اجرای این الگوریتم با پیچیدگی زمانی مطلوب <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(x + y)</span></b>، صعودی و <b style="color:#6D2E46;">مرتب بودن <span style="direction:ltr; unicode-bidi:embed;">(Sorted)</span></b> تمامی شناسه‌های سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(DocID)</span></b> در لیست‌های رخداد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span></b> است.
<br>
اگر ساختار مرتب‌سازی در زمان ذخیره‌سازی داده‌ها رعایت نشود، منطق خطی این الگوریتم به طور کامل دچار فروپاشی ساختاری <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Structural Breakdown)</span></b> می‌شود. در ادامه، رفتار دقیق سیستم در مواجهه با این خطای پیاده‌سازی بررسی می‌گردد.
</div>
<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. منطق تصمیم‌گیری در الگوریتم دو اشاره‌گره
</div>
<div style="text-align:justify; margin-bottom:24px;">
این الگوریتم بر اساس یک قانون ساده اما قدرتمند کار می‌کند: در هر مرحله، دو اشاره‌گر مقادیر فعلی خود را با هم مقایسه می‌کنند. اشاره‌گری که به <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">DocID</span></b> کوچکتر اشاره می‌کند، یک گام به جلو می‌رود (چون قطعاً مقدار کوچکتر در ادامه لیست دیگر که صعودی است یافت نخواهد شد). اگر مقادیر برابر بودند، یک اشتراک یافت شده است و هر دو اشاره‌گر پیش می‌روند.
</div>
<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. رفتار الگوریتم در مواجهه با لیست‌های نامرتب (بروز پدیده <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False Negative</span></b>)
</div>
<div style="text-align:justify; margin-bottom:24px;">
وقتی لیست‌ها نامرتب باشند، پیش‌فرض بنیادین الگوریتم (یعنی عدم بازگشت به مقادیر کوچکتر در ادامه لیست) نقض می‌شود. در این حالت، الگوریتم به جای یافتن تمامی اشتراک‌ها، به دلیل تصمیم‌گیری‌های غلط در حرکت دادن اشاره‌گرها، بسیاری از شناسه‌های مشترک را نادیده گرفته و اصطلاحاً آن‌ها را «رد می‌کند» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Skip)</span></b>. نتیجه مستقیم این رفتار، افت شدید و غیرقابل پیش‌بینیِ <b style="color:#6D2E46;">جامعیت <span style="direction:ltr; unicode-bidi:embed;">(Recall)</span></b> سیستم است؛ یعنی سیستم اسنادی را که کاملاً مرتبط هستند و کلمات مورد نظر را دارند، به کاربر نشان نمی‌دهد (منجر به تولید خروجی‌های <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False Negative</span></b> می‌شود).
</div>
<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. اثبات رفتار الگوریتم با بررسی گام‌به‌گام <span style="direction:ltr; unicode-bidi:embed;">(Trace)</span>
</div>
<div style="text-align:justify; margin-bottom:15px;">
برای درک دقیق این فروپاشی، دو لیست رخداد نامرتب را برای پرسمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A AND B</span></b> در نظر بگیرید:
</div>
<div style="text-align:justify; margin-bottom:15px; padding-right:15px; direction:ltr;">
List A: [45, 12, 89] <br>
List B: [12, 45, 70] <br>
<b style="color:#6D2E46;">True Intersection: 12, 45</b>
</div>

<div style="text-align:justify; margin-bottom:15px;">
اکنون رفتار الگوریتم استاندارد را روی این دو لیست ردیابی می‌کنیم:
</div>
<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام اول:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> روی <b style="color:#6D2E46;">45</b> و اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> روی <b style="color:#6D2E46;">12</b> است. چون <b style="color:#6D2E46;">12 < 45</b> است، اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> یک گام به جلو می‌رود.<br><br>
<b style="color:#6D2E46;">گام دوم:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> روی <b style="color:#6D2E46;">45</b> و اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> روی <b style="color:#6D2E46;">45</b> است. مقادیر برابرند. سند <b style="color:#6D2E46;">45</b> به عنوان خروجی ثبت می‌شود. هر دو اشاره‌گر پیش می‌روند.<br><br>
<b style="color:#6D2E46;">گام سوم:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> روی <b style="color:#6D2E46;">12</b> و اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> روی <b style="color:#6D2E46;">70</b> است. چون <b style="color:#6D2E46;">12 < 70</b> است، اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> یک گام به جلو می‌رود.<br><br>
<b style="color:#6D2E46;">گام چهارم:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> روی <b style="color:#6D2E46;">89</b> و اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> روی <b style="color:#6D2E46;">70</b> است. چون <b style="color:#6D2E46;">70 < 89</b> است، اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> باید به جلو برود، اما لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> به پایان رسیده است. الگوریتم متوقف می‌شود.
</div>
<div style="text-align:justify; margin-bottom:24px;">
<b style="color:#6D2E46;">نتیجه اجرای الگوریتم:</b> خروجی تولید شده تنها سند <b style="color:#6D2E46;">[45]</b> است! الگوریتم به دلیل نامرتب بودن لیست، سند <b style="color:#6D2E46;">12</b> را به طور کامل از دست داد.
</div>
<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
4. پیامدهای سیستمی و تغییر هزینه پردازشی
</div>
<div style="text-align:justify; margin-bottom:24px;">
وقتی لیست‌ها مرتب نباشند، امکان استفاده از الگوریتم بهینه دو اشاره‌گره از بین می‌رود و دانشجو مجبور به انتخاب یکی از دو رویکرد زیر است که هر دو فاجعه‌بار هستند:
<br>
<b style="color:#6D2E46;">استفاده از حلقه‌های تودرتو <span style="direction:ltr; unicode-bidi:embed;">(Nested Loops)</span>:</b> مقایسه هر عنصر از لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> با تمام عناصر لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b>. پیچیدگی زمانی از حالت خطی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(x+y)</span></b> به حالت مربعی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(x*y)</span></b> تنزل می‌یابد که در لیست‌های میلیونی، باعث توقف کامل <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Hang)</span></b> سیستم می‌گردد.
<br>
<b style="color:#6D2E46;">استفاده از جدول درهم‌سازی <span style="direction:ltr; unicode-bidi:embed;">(Hash Table)</span>:</b> بارگذاری عناصر لیست کوچکتر در جدول درهم‌سازی. اگرچه زمان اجرا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(x+y)</span></b> باقی می‌ماند، اما سیستم متحمل یک پیچیدگی فضایی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Space Complexity)</span></b> عظیم و غیرضروری از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(min(x, y))</span></b> می‌شود؛ در حالی که الگوریتم استاندارد برای اجرا نیازی به حافظه کمکی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(In-place)</span></b> ندارد.
</div>
<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری: این فروپاشی خاموش، یکی از رایج‌ترین خطاهای پیاده‌سازی است. ارائه خروجی ناقص مشابه مثال بالا و درخواست از دانشجو برای عارضه‌یابی <span style="direction:ltr; unicode-bidi:embed;">(Troubleshooting)</span>، روشی بسیار دقیق برای سنجش درک عمیق آن‌ها از معماری <span style="direction:ltr; unicode-bidi:embed;">Inverted Index</span> است.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.7</b>: محاسبه تعداد مقایسه‌ها در الگوریتم ادغام
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه و تنظیمات اولیه
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای یافتن اشتراک (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b>) میان دو لیست رخداد، الگوریتم استاندارد از مکانیزم دو اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Two-Pointer)</span></b> استفاده می‌کند. فرض می‌کنیم اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> برای پیمایش لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> و اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> برای پیمایش لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> تخصیص یافته‌اند.
</div>

<div style="text-align:justify; margin-bottom:15px;">
در نقطه شروع، هر دو اشاره‌گر روی اولین عنصر (اندیس صفر) لیست‌های خود قرار می‌گیرند. در هر گام، مقادیرِ تحت اشاره با یکدیگر مقایسه شده و سیستم بر اساس کوچکتر، بزرگتر یا مساوی بودن آن‌ها برای حرکت اشاره‌گرها تصمیم‌گیری می‌کند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">لیست <span style="direction:ltr; unicode-bidi:embed;">A</span>:</b> <span style="direction:ltr; unicode-bidi:embed; font-weight:bold; color:#6D2E46;">[1, 3, 5, 7, 9, 11]</span><br>
<b style="color:#6D2E46;">لیست <span style="direction:ltr; unicode-bidi:embed;">B</span>:</b> <span style="direction:ltr; unicode-bidi:embed; font-weight:bold; color:#6D2E46;">[2, 3, 4, 5, 11, 15]</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
ردیابی گام‌به‌گام مسیر حرکت اشاره‌گرها
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 1:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">1</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">2</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> 1 کوچکتر از 2 است.<br>
<b style="color:#6D2E46;">عملیات:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> یک خانه به جلو می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 2:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">3</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">2</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> 3 بزرگتر از 2 است.<br>
<b style="color:#6D2E46;">عملیات:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> یک خانه به جلو می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 3:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">3</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">3</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> مقادیر برابرند.<br>
<b style="color:#6D2E46;">عملیات:</b> عدد <b style="color:#6D2E46;">3</b> به عنوان اشتراک یافت‌شده در لیست خروجی ثبت می‌شود. هر دو اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> یک خانه پیش می‌روند.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 4:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">5</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">4</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> 5 بزرگتر از 4 است.<br>
<b style="color:#6D2E46;">عملیات:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> یک خانه به جلو می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 5:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">5</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">5</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> مقادیر برابرند.<br>
<b style="color:#6D2E46;">عملیات:</b> عدد <b style="color:#6D2E46;">5</b> در لیست خروجی ثبت می‌شود. هر دو اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> یک خانه پیش می‌روند.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 6:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">7</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">11</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> 7 کوچکتر از 11 است.<br>
<b style="color:#6D2E46;">عملیات:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> یک خانه به جلو می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام 7:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">9</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">11</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> 9 کوچکتر از 11 است.<br>
<b style="color:#6D2E46;">عملیات:</b> اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> یک خانه به جلو می‌رود.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">گام 8:</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> روی مقدار <b style="color:#6D2E46;">11</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> روی مقدار <b style="color:#6D2E46;">11</b> قرار دارد.<br>
<b style="color:#6D2E46;">مقایسه:</b> مقادیر برابرند.<br>
<b style="color:#6D2E46;">عملیات:</b> عدد <b style="color:#6D2E46;">11</b> در لیست خروجی ثبت می‌شود. هر دو اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> یک خانه پیش می‌روند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
شرط پایان و نتیجه‌گیری نهایی
</div>

<div style="text-align:justify; margin-bottom:24px;">
پس از اجرای گام هشتم، اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> از مرز انتهای لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> عبور می‌کند. از آنجا که شرط بقای حلقه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Loop Condition)</span></b> در الگوریتم ادغام، نرسیدنِ اشاره‌گرها به انتهای هر دو لیست است، الگوریتم بلافاصله در همین نقطه متوقف می‌شود. در نتیجه، مقدار <b style="color:#6D2E46;">15</b> در انتهای لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> اساساً نیازی به بررسی پیدا نمی‌کند.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin:24px 0 4px 0; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">نتیجه‌گیری:</span>
  <b style="color:#6D2E46;">تعداد دقیق مقایسه‌ها:</b> سیستم برای تکمیل این پردازش دقیقاً به <b style="color:#6D2E46;">8</b> عمل مقایسه <span style="direction:ltr; unicode-bidi:embed;">(Comparison)</span> نیاز داشت.<br>
<b style="color:#6D2E46;">خروجی نهایی سیستم:</b> لیست <span style="direction:ltr; unicode-bidi:embed; font-weight:bold;">[3, 5, 11]</span> به عنوان شناسه‌های مشترک بین دو سند بازگردانده می‌شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.8</b>: تحلیل پردازش پرسمان‌های منطقی و هزینه عملگر <span style="direction:ltr; unicode-bidi:embed;">NOT</span> <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Analysis of Logical Query Processing and the Cost of NOT Operator)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه و نگاشت به جبر مجموعه‌ها <span style="direction:ltr; unicode-bidi:embed;">(Set Algebra)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
در مدل بازیابی بولی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Boolean Retrieval Model)</span></b>، پرسمان‌های متنی به طور مستقیم به عملگرهای نظریه مجموعه‌ها ترجمه می‌شوند. برای درک نحوه پردازش سیستم، ابتدا باید معادل ریاضی هر عملگر را مشخص کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
 عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">OR</span></b> معادل <b style="color:#6D2E46;">اجتماع</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(&cup;)</span></b> است.<br>
 عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> معادل <b style="color:#6D2E46;">اشتراک</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(&cap;)</span></b> است.<br>
 عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NOT</span></b> معادل <b style="color:#6D2E46;">متمم‌گیری</b> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;"></span></b> نسبت به مجموعه مرجع (کل اسناد) است.
</div>

<div style="text-align:justify; margin-bottom:15px;">
بنابراین، پرسمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(A OR B) AND NOT C</span></b> از نظر منطق ریاضی به شکل زیر فرمول‌بندی می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \boldsymbol{(A \cup B) \cap C^c} $$
</div>

<div style="text-align:justify; margin-bottom:15px;">
که این رابطه در جبر مجموعه‌ها دقیقاً معادل عملگر <b style="color:#6D2E46;">تفاضل <span style="direction:ltr; unicode-bidi:embed;">(Set Difference)</span></b> است:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \boldsymbol{(A \cup B) - C} $$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. پردازش مرحله‌به‌مرحله روی لیست‌های ادعا <span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای اجرای این پرسمان بدون تحمیل بار اضافی به حافظه، سیستم آن را در دو فاز و با استفاده از لیست‌های میانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Temporary Lists)</span></b> پردازش می‌کند:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام اول (پردازش اجتماع - <span style="direction:ltr; unicode-bidi:embed;">OR</span>):</b> سیستم ابتدا اشاره‌گرهایی برای لیست ادعای کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> و لیست ادعای کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> اختصاص می‌دهد. در الگوریتم ادغامِ اجتماع، هر شناسه‌ای <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(DocID)</span></b> که در حداقل یکی از دو لیست وجود داشته باشد، خوانده شده و وارد یک لیست میانی (مثلاً <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b>) می‌شود. در صورت وجود شناسه‌های تکراری (اشتراک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b>)، تنها یک نسخه از آن در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b> ثبت می‌گردد.<br>
<i>پیچیدگی زمانی:</i> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(|A| + |B|)</span></b><br>
<i>طول لیست حاصل:</i> حداکثر برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">|A| + |B|</span></b>
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
 <b style="color:#6D2E46;">گام دوم (پردازش تفاضل - <span style="direction:ltr; unicode-bidi:embed;">AND NOT</span>):</b><br>
اکنون سیستم لیست میانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b> و لیست ادعای کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> را در اختیار دارد. دو اشاره‌گر در ابتدای این دو لیست قرار می‌گیرند. منطق حرکت در این مرحله، منطق «حذف» است:<br>
 اگر شناسه در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b> کوچکتر از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> باشد، آن شناسه به عنوان نتیجه قطعی ثبت شده و اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">R<sub>1</sub></span></b> پیش می‌رود.<br>
اگر شناسه‌ها برابر باشند، یعنی سند هم کلمات اولیه را دارد و هم شامل کلمه ممنوعه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> است. پس این شناسه <b style="color:#6D2E46;">نادیده گرفته شده</b> (حذف می‌شود) و هر دو اشاره‌گر پیش می‌روند.<br>
 اگر شناسه در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> کوچکتر باشد، اشاره‌گر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> صرفاً پیش می‌رود تا به شناسه‌های بزرگتر برسد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. تحلیل معماری: چرا هزینه پردازش <span style="direction:ltr; unicode-bidi:embed;">NOT</span> بسیار بالاتر از <span style="direction:ltr; unicode-bidi:embed;">AND</span> است؟
</div>

<div style="text-align:justify; margin-bottom:15px;">
در موتورهای جستجو و سیستم‌های بازیابی مقیاس‌بزرگ، عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NOT</span></b> یکی از پرهزینه‌ترین دستورات ارزیابی می‌شود که دلایل ساختاری آن به شرح زیر است:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">الف) از دست رفتن پدیده خلوت بودن <span style="direction:ltr; unicode-bidi:embed;">(Loss of Sparsity)</span>:</b><br>
قدرت اصلی نمایه وارونه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Inverted Index)</span></b> در استفاده از خاصیت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Sparsity</span></b> است؛ یعنی ما فقط اسنادی را پردازش می‌کنیم که یک کلمه در آن‌ها «وجود» دارد. عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> ماهیتی <b style="color:#6D2E46;">محدودکننده <span style="direction:ltr; unicode-bidi:embed;">(Restrictive)</span></b> دارد و به سرعت طول لیست خروجی را کاهش می‌دهد. اما عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NOT</span></b> (به تنهایی) ماهیتی <b style="color:#6D2E46;">بسط‌دهنده <span style="direction:ltr; unicode-bidi:embed;">(Expansive)</span></b> دارد. اگر سیستم بخواهد پرسمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NOT C</span></b> را پردازش کند، باید لیست ادعای <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> را از مجموعه کل اسناد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Universal Set</span></b> - که با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">N</span></b> نمایش داده می‌شود) کم کند. این یعنی سیستم مجبور است یک لیست خروجی با ابعاد تقریبی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(N)</span></b> بسازد که در مجموعه‌های چند میلیارد سندی، منجر به سرریز کامل حافظه اصلی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(RAM)</span></b> و فروپاشی کارایی می‌گردد.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">ب) تفاوت در شرایط توقف الگوریتم <span style="direction:ltr; unicode-bidi:embed;">(Stopping Criteria)</span>:</b><br>
در الگوریتم تقاطع استاندارد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A AND B</span></b>)، به محض اینکه یکی از لیست‌ها به انتها برسد، الگوریتم به دلیل عدم امکان یافتن اشتراک جدید، فوراً متوقف می‌شود. اما در پردازش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A AND NOT C</span></b> (که بهینه‌ترین حالت استفاده از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NOT</span></b> است)، حتی اگر لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">C</span></b> به پایان برسد، سیستم <b style="color:#6D2E46;">نمی‌تواند متوقف شود</b>؛ بلکه مجبور است تمام شناسه‌های باقی‌مانده در انتهای لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> را تا آخرین مورد به لیست خروجی اضافه کند. این پیمایش اجباری تا انتهای لیست پایه، هزینه پردازشی زمان اجرا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Query Time)</span></b> را در مقایسه با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> به شدت افزایش می‌دهد.
</div>

<div style="text-align:justify; margin:24px 0 4px 0; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46;">
<span style="font-size:17pt; font-weight:bold; color:#6D2E46; display:block; margin-bottom:10px;">نتیجه‌گیری:</span>
به دلیل همین هزینه‌های ساختاری، موتورهای جستجوی مدرن اجرای عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NOT</span></b> را به صورت مستقل <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Unary NOT)</span></b> ممنوع کرده یا نادیده می‌گیرند و تنها آن را در ترکیب با عملگرهای محدودکننده‌ای نظیر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> (به شکل فیلتر حذفی) مجاز می‌دانند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.9</b>: ارتقای نمایه برای پشتیبانی از پرسمان‌های عبارتی <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Upgrading Index to Support Phrasal Queries)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه: محدودیت مدل استاندارد <span style="direction:ltr; unicode-bidi:embed;">(Bag of Words)</span>
</div>

<div style="text-align:justify; margin-bottom:24px;">
در نمایه‌سازی استاندارد که تنها شناسه‌های سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(DocID)</span></b> ذخیره می‌شوند، سیستم از مدل «کیسه کلمات» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Bag of Words)</span></b> پیروی می‌کند. در این مدل، سیستم تنها از «حضور» یا «عدم حضور» یک کلمه در سند مطلع است و هیچ درکی از <b style="color:#6D2E46;">ترتیب</b>، <b style="color:#6D2E46;">مجاورت</b> یا <b style="color:#6D2E46;">ساختار جملات</b> ندارد. در نتیجه، اگر کاربری عبارت دقیق <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Machine Learning"</span></b> را جستجو کند، سیستم استاندارد آن را به عنوان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Machine AND Learning</span></b> پردازش کرده و سندی با متن <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Learning to operate this Machine"</span></b> را نیز به عنوان یک پاسخ کاملاً مرتبط برمی‌گرداند (یک خطای <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False Positive</span></b> بزرگ). برای حل این مشکل، معماری سیستم باید از سطح سند به سطح موقعیتِ کلمات ارتقا یابد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. تغییر ساختار ذخیره‌سازی: معرفی نمایه موقعیتی <span style="direction:ltr; unicode-bidi:embed;">(Positional Index)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای پشتیبانی از پرسمان‌های عبارتی، ساختار لیست‌های رخداد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings Lists)</span></b> باید تغییر بنیادی کند. به جای تخصیص یک لیست ساده از شناسه‌ها به هر کلمه، سیستم باید <b style="color:#6D2E46;">موقعیت <span style="direction:ltr; unicode-bidi:embed;">(Position)</span></b> دقیقِ هر بار تکرارِ آن کلمه در متن سند را نیز ذخیره کند.
</div>

<div style="text-align:justify; margin-bottom:15px;">
در این ساختار جدید (نمایه موقعیتی)، هر درایه از لیست رخداد، دیگر یک عدد منفرد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">DocID</span></b>) نیست؛ بلکه یک ساختار داده ترکیبی شامل شناسه سند و لیستی از موقعیت‌هاست:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مثال عملی:</b><br>
فرض کنید کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Machine"</span></b> در سند شماره 12، کلمات دهم و پنجاهم متن است. کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Learning"</span></b> نیز در همان سند، کلمات یازدهم و نودم متن است. لیست رخداد در نمایه موقعیتی به این شکل ذخیره می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{\text{Machine} \rightarrow [\dots, \text{DocID}_{12}: \langle 10, 50 \rangle, \dots]} $$

$$ \color{#6D2E46}{\text{Learning} \rightarrow [\dots, \text{DocID}_{12}: \langle 11, 90 \rangle, \dots]} $$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. الگوریتم پردازش پرسمان (اشتراک موقعیتی - <span style="direction:ltr; unicode-bidi:embed;">Positional Intersect</span>)
</div>

<div style="text-align:justify; margin-bottom:12px;">
زمان اجرای پرسمان، سیستم دیگر نمی‌تواند فقط به یافتن اسناد مشترک اکتفا کند، بلکه باید یک «الگوریتم اشتراک موقعیتی» را اجرا نماید:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام اول (یافتن اسناد مشترک):</b> ابتدا مانند یک پرسمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> معمولی، با استفاده از دو اشاره‌گر، اسنادی که هر دو کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Machine"</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Learning"</span></b> را دارند پیدا می‌کند (در اینجا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">DocID<sub>12</sub></span></b>).
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام دوم (بررسی مجاورت):</b> روی سند مشترک، سیستم موقعیت‌های کلمه اول را با کلمه دوم مقایسه می‌کند. شرط تطابق برای پرسمان عبارتی این است که موقعیت کلمه دوم، دقیقاً یک واحد بیشتر از کلمه اول باشد:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{\text{Pos}(\text{Term}_2) - \text{Pos}(\text{Term}_1) = 1} $$
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">گام سوم (ارزیابی مقادیر):</b> سیستم جفت <b style="color:#6D2E46;">(10, 11)</b> را بررسی می‌کند: <b style="color:#6D2E46;">11 - 10 = 1</b>. تطابق یافت شد!<br>
سپس جفت <b style="color:#6D2E46;">(50, 90)</b> را بررسی می‌کند که مجاور نیستند.<br>
به دلیل یافتن حداقل یک تطابق مجاور، سند 12 به عنوان پاسخ معتبر برای پرسمان عبارتی تایید و به کاربر بازگردانده می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. تحلیل پیچیدگی و هزینه‌های سیستمی این ارتقا
</div>

<div style="text-align:justify; margin-bottom:12px;">
تبدیل نمایه استاندارد به نمایه موقعیتی، هزینه‌های معماری قابل توجهی به همراه دارد که تحلیل آن برای طراحی سیستم‌های مقیاس‌بزرگ حیاتی است:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">پیچیدگی فضایی <span style="direction:ltr; unicode-bidi:embed;">(Space Complexity)</span>:</b> نمایه موقعیتی به شدت حجیم‌تر است. از آنجا که هر بار تکرار یک کلمه نیازمند ثبت موقعیت است، حجم نمایه وارونه بین 2 تا 4 برابر افزایش می‌یابد و بسته به زبان و ساختار داده‌ها، می‌تواند به 30 تا 50 درصد از حجم کل متن اصلی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Corpus)</span></b> برسد. استفاده از تکنیک‌های فشرده‌سازی (مانند گاما کدینگ برای ذخیره فاصله بین موقعیت‌ها به جای عدد مطلق) در اینجا الزامی است.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">پیچیدگی زمانی <span style="direction:ltr; unicode-bidi:embed;">(Time Complexity)</span>:</b> الگوریتم ادغام <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Merge)</span></b> دیگر محدود به طول لیست اسناد نیست. در بدترین حالت، زمان اجرای مقایسه‌ها وابسته به تعداد کل رخدادهای کلمات (فرکانس کلمه در کل متون یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Collection Frequency</span></b>) خواهد بود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
4. رویکرد جایگزین (معماری نمایه دوکلمه‌ای - <span style="direction:ltr; unicode-bidi:embed;">Bi-word Index</span>)
</div>

<div style="text-align:justify; margin-bottom:24px;">
قابل ذکر است که روش جایگزینی به نام <b style="color:#6D2E46;">نمایه دوکلمه‌ای</b> وجود دارد که در آن جفت کلمات متوالی (مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">"Machine Learning"</span></b>) مستقیماً به عنوان یک کلید مستقل در واژه‌نامه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Dictionary)</span></b> ثبت می‌شوند. این روش زمان اجرای پرسمان را به <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1)</span></b> کاهش می‌دهد، اما منجر به پدیده «انفجار واژه‌نامه» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Dictionary Blowup)</span></b> می‌شود؛ زیرا تعداد ترکیب‌های دوکلمه‌ای ممکن در یک زبان بی‌نهایت است.
</div>

<div style="text-align:justify; margin:24px 0 4px 0; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46;">
<span style="font-size:17pt; font-weight:bold; color:#6D2E46; display:block; margin-bottom:10px;">نتیجه‌گیری ساختاری:</span>
به همین دلیل، موتورهای جستجوی مدرن همواره از <b style="color:#6D2E46;">نمایه موقعیتی <span style="direction:ltr; unicode-bidi:embed;">(Positional Index)</span></b> به عنوان رویکرد استاندارد و مقیاس‌پذیر استفاده می‌کنند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.10</b>: تحلیل کارایی صف اولویت در بهینه‌سازی پرسمان‌های <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">OR</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه و طرح مسئله
</div>

<div style="text-align:justify; margin-bottom:24px;">
در سیستم‌های بازیابی اطلاعات، پردازش پرسمان‌هایی که شامل چندین عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">OR</span></b> متوالی هستند (مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A OR B OR C OR D</span></b>)، به معنای یافتن اجتماع <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Union)</span></b> چندین لیست رخداد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Postings List)</span></b> است. به این مسئله در علوم کامپیوتر، <b style="color:#6D2E46;">ادغام چندگانه <span style="direction:ltr; unicode-bidi:embed;">(K-way Merge)</span></b> گفته می‌شود که در آن <b style="color:#6D2E46;">K</b> تعداد لیست‌هاست. پیشنهاد تیم دانشجویی مبنی بر استفاده از صف اولویت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Priority Queue)</span></b> برای حل این مسئله، یک رویکرد کاملاً مهندسی‌شده و یکی از راهکارهای استاندارد در طراحی موتورهای جستجو است. در ادامه، صحت و برتری این الگوریتم را از منظر تحلیل پیچیدگی زمانی اثبات می‌کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. تحلیل رویکرد سنتی: ادغام دو به دو <span style="direction:ltr; unicode-bidi:embed;">(Pairwise Merge)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای درک ارزش پیشنهاد دانشجویان، ابتدا باید نقص روش سنتی را بررسی کنیم. فرض کنید <b style="color:#6D2E46;">K</b> لیست رخداد داریم که میانگین طول هر کدام <b style="color:#6D2E46;">N</b> است (مجموعاً <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">T = K &times; N</span></b> عنصر).
<br>
اگر سیستم بخواهد لیست‌ها را به صورت متوالی و دو به دو ادغام کند:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
ابتدا لیست اول و دوم را ادغام می‌کند: هزینه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(2N)</span></b><br>
سپس حاصل را با لیست سوم ادغام می‌کند: هزینه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(3N)</span></b><br>
این روند تا لیست <b style="color:#6D2E46;">K</b>-ام ادامه می‌یابد: هزینه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(KN)</span></b>
</div>

<div style="text-align:justify; margin-bottom:24px;">
مجموع پیچیدگی زمانی این رویکرد به صورت یک تصاعد حسابی رشد کرده و از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(K<sup>2</sup> N)</span></b> خواهد بود. در پرسمان‌هایی با تعداد متغیرهای بالا، این توان دوم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">K<sup>2</sup></span></b>) باعث افت شدید کارایی زمان اجرا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Query Time)</span></b> می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. تشریح الگوریتم پیشنهادی دانشجویان: استفاده از صف اولویت <span style="direction:ltr; unicode-bidi:embed;">(Min-Heap)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
پیشنهاد استفاده از صف اولویت (که معمولاً با ساختار داده <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Min-Heap</span></b> پیاده‌سازی می‌شود)، رویکرد حل مسئله را از حالت «دو به دو» به حالت «همزمان» تغییر می‌دهد. نحوه کارکرد این الگوریتم به شرح زیر است:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام راه‌اندازی <span style="direction:ltr; unicode-bidi:embed;">(Initialization)</span>:</b> سیستم یک اشاره‌گر به ابتدای هر یک از <b style="color:#6D2E46;">K</b> لیست اختصاص می‌دهد. سپس عنصر اول هر لیست (همراه با شناسه نشان‌دهنده این که عنصر متعلق به کدام لیست است) وارد یک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Min-Heap</span></b> می‌شود. از آنجا که درخت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Min-Heap</span></b> همواره کوچکترین مقدار را در ریشه خود نگه می‌دارد، کوچکترین شناسه سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(DocID)</span></b> در کل لیست‌ها، بلافاصله در ریشه درخت قرار می‌گیرد. اندازه این صف همواره دقیقاً برابر با <b style="color:#6D2E46;">K</b> است.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">گام استخراج و پیشروی <span style="direction:ltr; unicode-bidi:embed;">(Extract and Advance)</span>:</b><br>
<b style="color:#6D2E46;">1.</b> سیستم عنصر ریشه (کوچکترین <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">DocID</span></b>) را استخراج کرده و به لیست خروجی نهایی (پاسخ پرسمان) اضافه می‌کند.<br>
<b style="color:#6D2E46;">2.</b> اشاره‌گر لیستی که این عنصر از آن استخراج شده، یک خانه به جلو می‌رود.<br>
<b style="color:#6D2E46;">3.</b> سیستم عنصر جدیدِ آن لیست را وارد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Min-Heap</span></b> می‌کند. (اگر لیست به انتها رسیده باشد، عنصری اضافه نمی‌شود و اندازه صف کاهش می‌یابد).<br>
<b style="color:#6D2E46;">4.</b> درخت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Min-Heap</span></b> خود را مرتب می‌کند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Heapify)</span></b> تا کوچکترین عنصر جدید به ریشه بیاید.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">گام تکرار:</b> این فرآیند تا زمانی که صف اولویت کاملاً خالی شود (یعنی تمام لیست‌ها به انتها برسند) تکرار می‌گردد. برای مدیریت شناسه‌های تکراری (اشتراک‌ها در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">OR</span></b>)، سیستم به سادگی چک می‌کند که اگر عنصر استخراج‌شده از ریشه برابر با آخرین عنصر اضافه‌شده به لیست خروجی بود، آن را نادیده بگیرد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. تحلیل پیچیدگی زمانی رویکرد صف اولویت <span style="direction:ltr; unicode-bidi:embed;">(Min-Heap)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
اکنون کارایی این پیشنهاد را با نماد <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Big O</span></b> تحلیل می‌کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
 ساخت صف اولویت اولیه با <b style="color:#6D2E46;">K</b> عنصر، نیازمند زمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(K)</span></b> است.<br>
 در مجموع، تمام عناصر موجود در تمامی لیست‌ها (که تعداد آن‌ها <b style="color:#6D2E46;">T</b> است) دقیقاً یک بار وارد صف شده و یک بار از آن خارج می‌شوند.<br>
 در ساختار <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Min-Heap</span></b>، هزینه هر بار درج <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Insert)</span></b> یا استخراج و مرتب‌سازی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Extract-Min)</span></b>، وابسته به ارتفاع درخت است که با توجه به اندازه <b style="color:#6D2E46;">K</b>-تاییِ صف، برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(log K)</span></b> خواهد بود.<br>
 از آنجا که ما <b style="color:#6D2E46;">T</b> عنصر داریم و برای هر کدام عملیاتی از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(log K)</span></b> انجام می‌دهیم، زمان کل پردازشِ گام‌های تکرار شونده برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(T log K)</span></b> است.
</div>

<div style="text-align:justify; margin-bottom:15px;">
بنابراین پیچیدگی زمانی کل این رویکرد برابر است با:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \boldsymbol{\mathcal{O}(K) + \mathcal{O}(T \log K) = \mathcal{O}(T \log K)} $$
</div>

<div style="text-align:justify; margin:24px 0 4px 0; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46;">
<span style="font-size:17pt; font-weight:bold; color:#6D2E46; display:block; margin-bottom:10px;">4. نتیجه‌گیری و ارزیابی فنی:</span>
پیشنهاد این تیم دانشجویی <b style="color:#6D2E46;">کاملاً صحیح و بسیار بهینه</b> است.
<br>
با مقایسه پیچیدگی زمانی رویکرد سنتی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(K<sup>2</sup> N)</span></b> و رویکرد صف اولویت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(K N log K)</span></b> (با جایگذاری <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">T = KN</span></b>)، مشخص می‌شود که جایگزینی ضریب <b style="color:#6D2E46;">K</b> با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">log K</span></b>، یک بهبود دراماتیک در مقیاس‌پذیری ایجاد می‌کند. در سیستم‌های واقعی بازیابی وب که یک پرسمان ممکن است شامل ده‌ها مترادف و بسط کوئری <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Query Expansion)</span></b> با عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">OR</span></b> باشد، استفاده از صف اولویت نه تنها یک پیشنهاد خوب، بلکه استانداردِ صنعتی برای جلوگیری از گلوگاه‌های محاسباتی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Computational Bottlenecks)</span></b> است.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.11</b>: پیاده‌سازی و تحلیل سربار عملگرهای مجاورت <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Implementation and Overhead Analysis of Proximity Operators)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه
</div>

<div style="text-align:justify; margin-bottom:24px;">
مدل بولین استاندارد، به دلیل ماهیت صفر و یکی خود (حضور یا عدم حضور یک کلمه در سند)، توانایی درک فاصله و ارتباط معنایی کلمات مجاور را ندارد. برای رفع این محدودیت، «مدل بولین توسعه‌یافته» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Extended Boolean Model)</span></b> معرفی می‌شود که با استفاده از عملگرهای مجاورت (مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NEAR</span></b> یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">WITHIN</span></b>)، امکان جستجوی مبتنی بر فاصله کلمات را فراهم می‌کند. این قابلیت به ویژه برای یافتن اسامی خاص یا مفاهیم ترکیبی حیاتی است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. نگارش پرسمان <span style="direction:ltr; unicode-bidi:embed;">(Query Formulation)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
در سینتکس استاندارد سیستم‌های بازیابی اطلاعات، عملگر مجاورت معمولاً با کلمه کلیدی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NEAR</span></b> (مخفف <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">N</span></b>) یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">W</span></b> (مخفف <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Within</span></b>) به همراه یک عدد نشان داده می‌شود که حداکثر تعداد کلمات مجاز بین دو عبارت را مشخص می‌کند.
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; font-weight:bold; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; font-family:monospace;">
Daneshgah NEAR/2 Tehran
</div>

<div style="text-align:justify; margin-bottom:24px;">
این پرسمان به سیستم دستور می‌دهد: «اسنادی را بازیابی کن که در آن‌ها کلمات <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Daneshgah</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Tehran</span></b> حضور داشته باشند، به شرطی که حداکثر <b style="color:#6D2E46;">2</b> کلمه دیگر بتواند بین آن‌ها قرار گیرد.»
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. منطق محاسباتی و شرط تطابق موقعیت‌ها
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای اجرای این پرسمان، سیستم نیازمند استفاده از «نمایه موقعیتی» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Positional Index)</span></b> است. پس از یافتن اسنادی که هر دو کلمه را دارند، سیستم موقعیت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Position)</span></b> آن‌ها را در متن مقایسه می‌کند.
</div>

<div style="text-align:justify; margin-bottom:15px;">
اگر موقعیت کلمه اول <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>1</sub></span></b> و موقعیت کلمه دوم <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>2</sub></span></b> باشد و حداکثر کلمات مجاز بین آن‌ها <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">k</span></b> در نظر گرفته شود (در این مسئله <b style="color:#6D2E46;">k = 2</b>)، شرط تطابق به صورت ریاضی این‌گونه تعریف می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \boldsymbol{|P_1 - P_2| \le k + 1} $$
</div>

<div style="text-align:justify; margin-bottom:24px;">
بنابراین برای این مسئله، اختلاف موقعیت دو کلمه باید کوچکتر یا مساوی <b style="color:#6D2E46;">3</b> باشد. (به عنوان مثال، اگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Daneshgah</span></b> کلمه دهم و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Tehran</span></b> کلمه سیزدهم باشد، <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">|10 - 13| = 3 &le; 3</span></b>، شرط برقرار است و دقیقاً <b style="color:#6D2E46;">2</b> کلمه بین آن‌ها فاصله وجود دارد).
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. تحلیل سربار پردازشی در زمان اجرا <span style="direction:ltr; unicode-bidi:embed;">(Query-Time Overhead)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
پشتیبانی از عملگرهای مجاورت، بهای سنگینی برای معماری سیستم دارد و سربارهای <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Overheads)</span></b> زیر را در زمان اجرای پرسمان تحمیل می‌کند:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">الف) سربار سنگین ورودی/خروجی <span style="direction:ltr; unicode-bidi:embed;">(I/O Overhead)</span>:</b><br>
همان‌طور که می‌دانیم، لیست‌های رخداد موقعیتی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Positional Postings Lists)</span></b> بسیار حجیم‌تر از لیست‌های استاندارد هستند. وقتی کاربر از عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">NEAR</span></b> استفاده می‌کند، سیستم باید حجم عظیمی از داده‌های موقعیتی را از دیسک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Disk)</span></b> خوانده و به حافظه اصلی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(RAM)</span></b> منتقل کند. این انتقال داده، گلوگاه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Bottleneck)</span></b> اصلی در سرعت پاسخ‌گویی است.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">ب) پیچیدگی زمانی و سربار محاسباتی <span style="direction:ltr; unicode-bidi:embed;">(Computational Overhead)</span>:</b><br>
در مدل بولین استاندارد، یافتن اشتراک دو سند نیازمند یک مقایسه ساده روی شناسه‌های سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(DocID)</span></b> است. اما با عملگر مجاورت، سیستم باید یک «حلقه تودرتو» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Nested Loop)</span></b> روی موقعیت‌های کلمات اجرا کند.<br>
فرض کنید کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> در یک سند خاص <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">tf<sub>A</sub></span></b> بار تکرار شده و کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> در همان سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">tf<sub>B</sub></span></b> بار تکرار شده است. الگوریتم اشتراک موقعیتی برای بررسی شرط فاصله، باید این دو لیست از موقعیت‌ها را پیمایش کند که پیچیدگی زمانی آن برای هر تک‌سندِ مشترک، از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(tf<sub>A</sub> + tf<sub>B</sub>)</span></b> است.<br>
هنگامی که این محاسبه برای هزاران یا میلیون‌ها سندِ مشترک ضرب شود، بار محاسباتی عظیمی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(CPU Cycles)</span></b> به پردازنده‌های موتور جستجو تحمیل می‌گردد.
</div>

<div style="text-align:justify; margin:24px 0 4px 0; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46; color:#6D2E46;">
<span style="font-size:17pt; font-weight:bold; display:block; margin-bottom:10px;">نتیجه‌گیری ساختاری:</span>
به دلیل همین سربارهای نمایی در زمان اجرا، موتورهای جستجوی مقیاس‌بزرگ، جستجوی مجاورتی را به عنوان پیش‌فرض اعمال نمی‌کنند، بلکه آن را به عنوان یک فاز ثانویه (پس از فیلتر کردن اسناد کاندیدا با عملگر ساده <span style="direction:ltr; unicode-bidi:embed;">AND</span>) و تنها روی زیرمجموعه کوچکی از اسناد با بالاترین امتیاز <span style="direction:ltr; unicode-bidi:embed;">(Top-K Documents)</span> اجرا می‌کنند تا از افت کارایی جلوگیری شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.12</b>: محدودیت‌های تطابق دقیق در بازیابی بولین کلاسیک <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Exact Match Limitations in Classic Boolean Retrieval)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه
</div>

<div style="text-align:justify; margin-bottom:24px;">
مدل بازیابی بولین کلاسیک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Classic Boolean Model)</span></b>، نخستین و پایه‌ای‌ترین مدل در سیستم‌های بازیابی اطلاعات است. این مدل، با وجود سادگی و سرعت بالا در پردازش، از یک محدودیت ذاتی و بنیادین رنج می‌برد: ارزیابی میزان ارتباط اسناد به صورت قطعی و دودویی (باینری) انجام می‌شود و هیچ‌گونه درکی از «درجه ارتباط» <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">(Degree of Relevance)</span></b> یا وزن‌دهی به کلمات وجود ندارد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1.  چرا ارزیابی به صورت صفر و یک است؟
</div>

<div style="text-align:justify; margin-bottom:15px;">
علت این رویکرد به بنیان‌های ریاضی این مدل بازمی‌گردد. مدل بولین مستقیماً بر پایه <b style="color:#6D2E46;">جبر بول <span style="direction:ltr; unicode-bidi:embed;">(Boolean Algebra)</span></b> و <b style="color:#6D2E46;">نظریه مجموعه‌ها <span style="direction:ltr; unicode-bidi:embed;">(Set Theory)</span></b> بنا شده است.
</div>

<div style="text-align:justify; margin-bottom:24px;">
در این چارچوب ریاضی، جهان به صورت قطعی تعریف می‌شود: یک کلمه یا در یک سند وجود دارد (مقدار <b style="color:#6D2E46;">1</b> یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">True</span></b>) یا وجود ندارد (مقدار <b style="color:#6D2E46;">0</b> یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False</span></b>). بر همین اساس، تابع ارزیابی سیستم، تطابق پرسمان کاربر با یک سند را صرفاً به عنوان یک معادله منطقی حل می‌کند که خروجی آن تنها می‌تواند <b style="color:#6D2E46;">0</b> (سند نامرتبط - عدم بازیابی) یا <b style="color:#6D2E46;">1</b> (سند مرتبط - بازیابی قطعی) باشد. به این مکانیزم سخت‌گیرانه، <b style="color:#6D2E46;">تطابق دقیق <span style="direction:ltr; unicode-bidi:embed;">(Exact Match)</span></b> گفته می‌شود، زیرا سیستم تنها به دنبال کلماتی می‌گردد که دقیقاً با کاراکترهای پرسمان یکسان باشند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. بررسی یک سناریوی عملی: شکست منطقی مدل بولین
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای درک ضعف این رویکرد صفر و یکی، سناریوی پرکاربرد زیر را در نظر بگیرید:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
 <b style="color:#6D2E46;">نیاز اطلاعاتی کاربر:</b> کاربری به دنبال یافتن یک مقاله جامع درباره «بهینه‌سازی الگوریتم‌ها در زبان پایتون» است.<br>
 <b style="color:#6D2E46;">پرسمان ثبت‌شده:</b> کاربر برای رسیدن به دقیق‌ترین نتیجه، از عملگر محدودکننده <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b> استفاده کرده و پرسمان زیر را وارد می‌کند:
<div style="text-align:center; direction:ltr; font-weight:bold; color:#6D2E46; background-color:#F8F9FA; border-radius:6px; border:1px solid #EBEBEB; padding:10px; margin:10px 0; font-family:monospace; font-size:16pt;">
Python AND Algorithm AND Optimization
</div>
 <b style="color:#6D2E46;">وضعیت سند هدف <span style="direction:ltr; unicode-bidi:embed;">(Target Document)</span>:</b> یک مقاله مرجع، بسیار غنی و دقیقاً مرتبط با نیاز کاربر در پایگاه داده وجود دارد. در این مقاله کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Python</span></b> بیش از <b style="color:#6D2E46;">100</b> بار و کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Algorithm</span></b> بیش از <b style="color:#6D2E46;">50</b> بار تکرار شده است. اما نویسنده مقاله، بر اساس سلیقه نگارشی خود، به جای استفاده از کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Optimization</span></b>، در سراسر متن از کلمات مترادفی نظیر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Efficiency</span></b> یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Performance</span></b> استفاده کرده است.
</div>

<div style="font-size:16pt; font-weight:bold; color:#1B263B; margin-bottom:10px;">
تحلیل رفتار سیستم (رخداد خطای منطقی):
</div>

<div style="text-align:justify; margin-bottom:15px;">
هنگام اجرای الگوریتم، سیستم مدل بولین وضعیت این سند را به این شکل محاسبه می‌کند:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
 وجود <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Python</span></b> در سند = <b style="color:#6D2E46;">1</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">True</span></b>)<br>
 وجود <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Algorithm</span></b> در سند = <b style="color:#6D2E46;">1</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">True</span></b>)<br>
 وجود <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Optimization</span></b> در سند = <b style="color:#6D2E46;">0</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False</span></b>)
</div>

<div style="text-align:justify; margin-bottom:15px;">
نتیجه ارزیابی نهایی برای این سند:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; font-weight:bold; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; font-family:monospace;">
1 AND 1 AND 0 = 0
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. پیامدهای ساختاری این رویکرد
</div>

<div style="text-align:justify; margin-bottom:15px;">
همان‌طور که در سناریوی بالا مشاهده شد، خروجی نهایی برای این مقاله مرجع دقیقاً <b style="color:#6D2E46;">0</b> خواهد بود و سیستم آن را به طور کامل از لیست نتایج حذف می‌کند (بروز خطای منفیِ کاذب یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False Negative</span></b>).
</div>

<div style="text-align:justify; margin-bottom:24px;">
مشکل اصلی اینجاست که در چشم‌انداز مدل بولین کلاسیک، بین این مقاله مرجع (که <b style="color:#6D2E46;">99%</b> با نیاز کاربر هم‌پوشانی معنایی دارد اما یک کلمه را ندارد) و یک مقاله کاملاً بی‌ربط درباره «دستور پخت پیتزا» (که هیچ‌کدام از کلمات را ندارد)، <b style="color:#6D2E46;">هیچ تفاوتی وجود ندارد</b>؛ زیرا هر دو در نهایت امتیاز صفر می‌گیرند.
</div>

<div style="text-align:justify; margin:24px 0 4px 0; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46; color:#6D2E46;">
<span style="font-size:17pt; font-weight:bold; display:block; margin-bottom:10px;">پیامدهای نهایی: این رفتار صفر و یکی منجر به دو پدیده مخرب در سیستم‌های بازیابی می‌شود:</span>
<b style="color:#6D2E46;">1. عدم امکان رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed;">(No Ranking)</span>:</b> تمام اسنادی که خروجی <b style="color:#6D2E46;">1</b> می‌گیرند، در یک سطح قرار دارند. سیستم نمی‌تواند تشخیص دهد کدام سند مرتبط‌تر است (مثلاً سندی که کلمه را <b style="color:#6D2E46;">1</b> بار دارد با سندی که آن را <b style="color:#6D2E46;">100</b> بار دارد، یکسان ارزیابی می‌شود).<br>
<b style="color:#6D2E46;">2. قحطی یا ضیافت <span style="direction:ltr; unicode-bidi:embed;">(Feast or Famine)</span>:</b> پرسمان‌های کاربر معمولاً یا به قدری سخت‌گیرانه هستند (مانند استفاده زیاد از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">AND</span></b>) که هیچ سندی بازیابی نمی‌شود (قحطی)، و یا به قدری گشاده‌دستانه هستند (استفاده از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">OR</span></b>) که میلیون‌ها سند نامرتبط به کاربر نمایش داده می‌شود (ضیافت)، که در هر دو حالت کاربر به نتیجه مطلوب نمی‌رسد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:center;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.13</span></b>: توزیع نمایه وارونه و مقایسه معماری‌های پارتیشن‌بندی
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه: ضرورت توزیع‌سازی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Distributed Indexing)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
زمانی که حجم اسناد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Corpus</span>) در سیستم‌های بازیابی اطلاعات به مقیاس وب می‌رسد، ذخیره‌سازی نمایه وارونه روی یک سرور منفرد به دلیل محدودیت‌های حافظه اصلی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RAM</span>) و فضای ذخیره‌سازی غیرممکن می‌گردد. در این شرایط، پایگاه داده باید تکه‌تکه شده و روی خوشه‌ای از سرورها (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Cluster</span>) توزیع شود که به این عمل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Sharding</span> می‌گویند. برای این کار دو معماری بنیادین وجود دارد که رفتار سیستم را در زمان اجرای پرسمان به کلی تغییر می‌دهند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) رویکرد تقسیم بر اساس اسناد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Document-Partitioning / Local Index)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در این معماری، مجموعه کل اسناد به بخش‌های مساوی تقسیم شده و هر بخش به یک سرور مجزا (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Node</span>) اختصاص می‌یابد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
 <b style="color:#6D2E46;">ساختار:</b> هر سرور یک نمایه وارونه محلی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Local Index</span>) می‌سازد که صرفاً شامل کلمات موجود در اسنادِ اختصاص‌یافته به خودش است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
 <b style="color:#6D2E46;">نحوه اجرای پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A AND B</span>:</b> وقتی کاربر این پرسمان را ثبت می‌کند، سرور مرکزی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Broker</span> یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Aggregator</span>) این درخواست را به طور همزمان (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Broadcast</span>) برای <b style="color:#6D2E46;">تمامی</b> سرورهای خوشه ارسال می‌کند. هر سرور به صورت مستقل و محلی، اشتراک لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> را در نمایه خود محاسبه کرده و بهترین نتایج خود را (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Top-K</span>) به سرور مرکزی بازمی‌گرداند. در نهایت، سرور مرکزی این نتایج را ترکیب و مرتب کرده و به کاربر نشان می‌دهد.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) رویکرد تقسیم بر اساس کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term-Partitioning / Global Index)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در این معماری، واژه‌نامه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Dictionary</span>) قطعه‌قطعه می‌شود. به عنوان مثال، تمامی کلماتی که با حروف <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> تا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">M</span></b> شروع می‌شوند به سرور اول و حروف <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">N</span></b> تا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Z</span></b> به سرور دوم سپرده می‌شوند.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
 <b style="color:#6D2E46;">ساختار:</b> هر سرور دارای یک نمایه سراسری (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Global Index</span>) است؛ به این معنا که لیست رخداد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Postings List</span>) یک کلمه در یک سرور، حاوی شناسه تمامی اسنادی است که آن کلمه در آن‌ها وجود دارد (از کل اسناد سیستم).
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
 <b style="color:#6D2E46;">نحوه اجرای پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A AND B</span>:</b> سرور مرکزی بررسی می‌کند که کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> در کدام سرور و کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> در کدام سرور ذخیره شده است. اگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> در سرور شماره <b style="color:#6D2E46;">1</b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> در سرور شماره <b style="color:#6D2E46;">2</b> باشد، یکی از این دو سرور مجبور است <b style="color:#6D2E46;">کل لیست رخداد خود را از طریق شبکه</b> برای سرور دیگر (یا برای سرور مرکزی) ارسال کند تا عملگر اشتراک (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Intersection</span>) بتواند روی دو لیست انجام شود.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) مقایسه درگیری شبکه و تحلیل گلوگاه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Bottleneck Analysis)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
تفاوت اصلی این دو رویکرد در زمان اجرای پرسمان‌های چندکلمه‌ای (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">A AND B</span>) به شکل چشم‌گیری نمایان می‌شود:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
 <b style="color:#6D2E46;">گلوگاه در روش <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Term-Partitioning</span> (فاجعه شبکه):</b><br>
در این روش، <b style="color:#6D2E46;">شبکه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Network Bandwidth</span>)</b> به گلوگاه اصلی سیستم تبدیل می‌شود. لیست‌های رخداد برای کلمات پرکاربرد می‌توانند شامل صدها میلیون شناسه سند باشند (چندین گیگابایت داده). انتقال این حجم عظیم از داده‌ها بین سرورها در هر بار جستجو، پهنای باند شبکه داخلی را به سرعت اشغال کرده و زمان پاسخ‌گویی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Latency</span>) را به شدت افزایش می‌دهد. در اینجا سیستم به جای جابه‌جا کردن «کد پردازشی»، مجبور به جابه‌جا کردن «داده‌های عظیم» است که در معماری توزیع‌یافته یک خطای طراحی محسوب می‌شود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
 <b style="color:#6D2E46;">گلوگاه در روش <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Document-Partitioning</span> (بهینگی شبکه):</b>
در این روش، درگیری شبکه بسیار ناچیز است. سرور مرکزی تنها یک رشته متنی کوتاه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">A AND B</span>) را به سرورها می‌فرستد و هر سرور تنها حجم کوچکی از نتایج نهایی (مثلاً <b style="color:#6D2E46;">10</b> سند برتر) را برمی‌گرداند. لیست‌های عظیم رخداد هرگز در شبکه جابه‌جا نمی‌شوند و پردازش دقیقاً در همان جایی انجام می‌شود که داده‌ها قرار دارند.<br>
با این حال، گلوگاه این روش <b style="color:#6D2E46;">سرور مرکزی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Broker</span>)</b> است؛ زیرا با افزایش تعداد سرورها (مثلاً به ده‌ها هزار نود)، ارسال درخواست برای همه آن‌ها و ادغام همزمان نتایج می‌تواند باعث سربار پردازشی روی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Broker</span> شود.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری ساختاری: به دلیل هزینه غیرقابل جبرانِ انتقال لیست‌های رخداد در شبکه (در روش <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Term-Partitioning</span>)، تقریباً تمامی موتورهای جستجوی مدرن و مقیاس‌پذیر (نظیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Google</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">ElasticSearch</span>) از معماری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Document-Partitioning</span> استفاده می‌کنند. در این سیستم‌ها، مدیریت گلوگاهِ سرور مرکزی با استفاده از ساختار درختی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Tree-based Aggregation)</span> و مسیریاب‌های توزیع‌شده به راحتی قابل حل است.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.14</b>: مثال نقض و تحلیل سربار در اشاره‌گرهای پرشی <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Counterexample and Overhead Analysis in Skip Pointers)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. بیان منطق الگوریتم و ریشه سربار <span style="direction:ltr; unicode-bidi:embed;">(Overhead)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
در الگوریتم ادغام با پرشگر، زمانی که اشاره‌گر لیست اول <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> به مقداری کوچکتر از لیست دوم <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> اشاره می‌کند، سیستم پیش از اینکه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> را یک خانه به جلو ببرد، یک بررسی اضافه انجام می‌دهد:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
شرط پرش: «آیا در این خانه پرشگر وجود دارد؟ و آیا مقصدی که پرشگر به آن اشاره می‌کند، همچنان کوچکتر یا مساوی با مقدار فعلی در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> است؟»
</div>

<div style="text-align:justify; margin-bottom:24px;">
اگر پاسخ به این شرط منطقی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">False</span></b> باشد (یعنی مقدار مقصد بزرگتر از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> باشد)، پرش لغو می‌شود و اشاره‌گر به صورت عادی یک خانه جلو می‌رود. اما سیستم برای بررسی این شرط، <b style="color:#6D2E46;">یک مقایسه عددی اضافی</b> را به پردازنده تحمیل کرده است. اگر این اتفاق مدام تکرار شود، کارایی به شدت افت می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. طراحی مثال نقض عددی
</div>

<div style="text-align:justify; margin-bottom:15px;">
برای ایجاد این نقطه ضعف، باید دو لیست با شناسه‌های به شدت متراکم <span style="direction:ltr; unicode-bidi:embed; font-weight:bold; color:#6D2E46;">(Dense)</span> و درهم‌تنیده <span style="direction:ltr; unicode-bidi:embed; font-weight:bold; color:#6D2E46;">(Interleaved)</span> طراحی کنیم.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b>: <span style="direction:ltr; unicode-bidi:embed; font-weight:bold; color:#6D2E46;">[1, 3, 5, 7, 9, 11]</span> (طول لیست 6 است)
<br>
لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b>: <span style="direction:ltr; unicode-bidi:embed; font-weight:bold; color:#6D2E46;">[2, 4, 6, 8, 10, 12]</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
<b style="color:#6D2E46;">تنظیمات پرشگر برای لیست <span style="direction:ltr; unicode-bidi:embed;">A</span>:</b>
<br>
طبق فرمول استاندارد، طول گام پرش برابر با جذر طول لیست (یعنی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">&radic;6 &approx; 2</span></b>) در نظر گرفته می‌شود. بنابراین در لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b>، هر دو خانه یک پرشگر قرار می‌دهیم:
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
خانه با اندیس 0 (مقدار 1) دارای پرشگر به اندیس 2 (مقدار 5) است.
<br>
خانه با اندیس 2 (مقدار 5) دارای پرشگر به اندیس 4 (مقدار 9) است.
<br>
خانه با اندیس 4 (مقدار 9) دارای پرشگر به انتهای لیست است (بدون پرش در نظر می‌گیریم).
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. ردیابی الگوریتم ادغام خطی استاندارد (بدون پرشگر)
</div>

<div style="text-align:justify; margin-bottom:15px;">
در ادغام معمولی، ما فقط مقادیر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> را با هم مقایسه می‌کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
1 < 2 <span style="direction:ltr; unicode-bidi:embed;"></span> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> پیش می‌رود (1 مقایسه)
<br>
3 > 2 <span style="direction:ltr; unicode-bidi:embed;"></span> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> پیش می‌رود (1 مقایسه)
<br>
3 < 4 <span style="direction:ltr; unicode-bidi:embed;"></span> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> پیش می‌رود (1 مقایسه)
<br>
5 > 4 <span style="direction:ltr; unicode-bidi:embed;"></span> <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> پیش می‌رود (1 مقایسه)
<br>
این روند زیگزاگی دقیقاً تا انتهای لیست ادامه می‌یابد.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
تعداد کل مقایسه‌ها: به دلیل درهم‌تنیدگی کامل، برای رسیدن به انتهای لیست اول دقیقاً <b style="color:#6D2E46;">11 مقایسه</b> انجام می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
4. ردیابی الگوریتم همراه با پرشگر (بروز افت کارایی)
</div>

<div style="text-align:justify; margin-bottom:15px;">
اکنون همان دو لیست را با فعال بودن مکانیزم پرشگر ردیابی می‌کنیم:
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
گام اول (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub> = 1</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub> = 2</span></b>):
<br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">1 < 2</span></b> (نتیجه: 1 مقایسه)
<br>
چون <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A < B</span></b> است، سیستم شرط پرش را چک می‌کند: پرشگر به عدد 5 اشاره دارد. آیا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">5 &le; 2</span></b> است؟ خیر.
<br>
نتیجه شرط پرش: (1 مقایسه هدر رفته). <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> عادی پیش می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
گام دوم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub> = 3</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub> = 2</span></b>):
<br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">3 > 2</span></b> (نتیجه: 1 مقایسه). <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> پیش می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
گام سوم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub> = 3</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub> = 4</span></b>):
<br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">3 < 4</span></b> (نتیجه: 1 مقایسه). پرشگری وجود ندارد. <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> پیش می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
گام چهارم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub> = 5</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub> = 4</span></b>):
<br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">5 > 4</span></b> (نتیجه: 1 مقایسه). <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub></span></b> پیش می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
گام پنجم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub> = 5</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>B</sub> = 6</span></b>):
<br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">5 < 6</span></b> (نتیجه: 1 مقایسه).
<br>
سیستم شرط پرش را چک می‌کند: پرشگر به عدد 9 اشاره دارد. آیا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">9 &le; 6</span></b> است؟ خیر.
<br>
نتیجه شرط پرش: (1 مقایسه هدر رفته). <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">P<sub>A</sub></span></b> عادی پیش می‌رود.
</div>

<div style="text-align:justify; margin-bottom:15px;">
در هر گامی که لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> کوچکتر است و پرشگر دارد، سیستم مجبور به انجام یک ارزیابی بی‌فایده می‌شود.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
تعداد کل مقایسه‌ها با پرشگر: 11 مقایسه اصلی + 2 مقایسه باطل‌شده برای چک کردن پرشگرها = <b style="color:#6D2E46;">13 مقایسه</b>.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری تحلیلی: این مثال نقض نشان می‌دهد که در ادغام دو لیست با توزیع عددی بسیار نزدیک و طول‌های مشابه، الگوریتم پرشگر نه تنها زمان اجرا را از مرتبه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(N)</span> کاهش نمی‌دهد، بلکه با افزودن شاخه‌های شرطی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Conditional Branches)</span> بی‌حاصل، سربار پردازشی را افزایش داده و الگوریتم را از حالت بهینه خارج می‌کند. پرشگرها تنها زمانی مفیدند که یکی از لیست‌ها بسیار طولانی‌تر از دیگری بوده یا شکاف‌های عددی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Gaps)</span> بزرگی بین شناسه‌های اسناد وجود داشته باشد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:28px 32px; text-align:right; line-height:2.2; font-size:16pt; color:#2C3E50;">

<div style="font-size:20pt; font-weight:bold; margin-bottom:24px; text-align:right; color:#000000;">
پاسخنامه مسئله <b style="color:#6D2E46;">1.15</b>: معماری کش در موتورهای جستجو <span style="direction:ltr; unicode-bidi:embed; font-size:18pt; color:#6D2E46;">(Caching Architecture in Search Engines)</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
مقدمه: سطوح کش‌سازی در معماری موتورهای جستجو
</div>

<div style="text-align:justify; margin-bottom:24px;">
در موتورهای جستجوی مقیاس‌پذیر، مدیریت پرسمان‌های تکراری از طریق حافظه نهان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Cache</span></b>) یکی از ارکان کلیدی برای کاهش تاخیر (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Latency</span></b>) و صرفه‌جویی در منابع پردازشی است. معماران سیستم برای استقرار حافظه نهان با دو رویکرد بنیادین مواجه هستند: <b style="color:#6D2E46;">کش‌سازی نتایج <span style="direction:ltr; unicode-bidi:embed;">(Result Caching)</span></b> و <b style="color:#6D2E46;">کش‌سازی لیست‌ها <span style="direction:ltr; unicode-bidi:embed;">(List / Term Caching)</span></b>. انتخاب بین این دو نیازمند تحلیل دقیق بده‌بستان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Trade-off</span></b>) میان مصرف حافظه، سرعت پاسخ‌گویی و هزینه به‌روزرسانی است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
1. رویکرد اول: کش کردن نتایج پرسمان <span style="direction:ltr; unicode-bidi:embed;">(Result / Query Caching)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
در این معماری، سیستم مستقیماً خروجی نهایی یک پرسمان کامل (مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A AND B</span></b>) را کش می‌کند. یعنی کلید کش، رشته متنیِ پرسمان است و مقدار آن، لیست کوچکی از شناسه‌های اسناد برتر (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Top-K DocIDs</span></b>) یا حتی کدهای <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">HTML</span></b> نتایج است.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مزیت زمانی:</b> اگر کاربر دیگری دقیقاً همان پرسمان را وارد کند، پیچیدگی زمانی اجرای پرسمان از مرتبه پردازش اشتراک‌گیری، مستقیماً به <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1)</span></b> تقلیل می‌یابد؛ زیرا سیستم بدون هیچ محاسبه‌ای خروجی را از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">RAM</span></b> می‌خواند.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">مزیت فضایی در سطح پرسمان:</b> فضای اشغال شده برای هر پرسمان بسیار اندک است، زیرا سیستم فقط نتایج برتر (مثلاً 10 سند اول) را ذخیره می‌کند، نه کل اسناد مشترک را.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
2. رویکرد دوم: کش کردن لیست‌های رخداد <span style="direction:ltr; unicode-bidi:embed;">(List / Term Caching)</span>
</div>

<div style="text-align:justify; margin-bottom:15px;">
در این معماری، سیستم پرسمان‌های ترکیبی را کش نمی‌کند، بلکه لیست رخداد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Postings Lists</span></b>) کلماتِ پرکاربرد را از دیسک به حافظه اصلی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">RAM</span></b>) منتقل کرده و آنجا نگه می‌دارد.
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">مزیت نرخ اصابت <span style="direction:ltr; unicode-bidi:embed;">(Hit Rate)</span>:</b> بسیاری از پرسمان‌ها کاملاً جدید و منحصربه‌فرد هستند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Long-tail Queries</span></b>)، اما کلمات تشکیل‌دهنده آن‌ها تکراری است. در این روش، حتی اگر کل پرسمان در سیستم سابقه نداشته باشد، لیست کلمات آن در کش وجود دارد و از هزینه سنگین خواندن از دیسک (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Disk I/O</span></b>) جلوگیری می‌شود.
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">چالش پردازشی:</b> سرعت پاسخ‌گویی در این روش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(1)</span></b> نیست. سیستم همچنان موظف است الگوریتم ادغام (مانند دو اشاره‌گره) را با پیچیدگی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">O(x+y)</span></b> روی لیست‌های موجود در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">RAM</span></b> اجرا کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin-top:25px; margin-bottom:12px;">
3. تحلیل <span style="direction:ltr; unicode-bidi:embed;">Trade-off</span>: فضای حافظه در برابر به‌روزرسانی سیستم
</div>

<div style="text-align:justify; margin-bottom:15px;">
تفاوت حیاتی و پاشنه آشیل این دو روش، زمانی آشکار می‌شود که پایگاه داده اسناد پویا باشد (اسناد جدید ایندکس شوند یا اسناد قدیمی حذف گردند):
</div>

<div style="text-align:justify; margin-bottom:15px; padding-right:15px;">
<b style="color:#6D2E46;">تحلیل از منظر به‌روزرسانی و ابطال کش <span style="direction:ltr; unicode-bidi:embed;">(Cache Invalidation)</span>:</b>
<div style="padding-right:15px; margin-top:10px;">
<b style="color:#6D2E46;">در کش نتایج <span style="direction:ltr; unicode-bidi:embed;">(Result Cache)</span>:</b> مدیریت به‌روزرسانی یک فاجعه مهندسی است (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Cache Invalidation Nightmare</span></b>). اگر سندی به نام <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">D<sub>new</sub></span></b> به سیستم اضافه شود که حاوی کلمات <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> باشد، سیستم دقیقاً نمی‌داند این سندِ جدید روی کدام‌یک از میلیون‌ها پرسمانِ کش‌شده تاثیر می‌گذارد. برای حفظ دقت سیستم، با هر بار آپدیت شدن نمایه، معمولاً کل کشِ نتایج باید منقضی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Flush</span></b>) شده و از نو ساخته شود که منجر به افت شدید کارایی موقت می‌گردد.
<br>
<b style="color:#6D2E46;">در کش لیست‌ها <span style="direction:ltr; unicode-bidi:embed;">(List Cache)</span>:</b> به‌روزرسانی بسیار نرم و کارآمد است. با ورود سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">D<sub>new</sub></span></b>، سیستم صرفاً شناسه این سند را به انتهای لیست‌های <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">A</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">B</span></b> در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">RAM</span></b> اضافه می‌کند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Append</span></b>). هیچ کشی باطل نمی‌شود و یکپارچگی سیستم با کمترین هزینه حفظ می‌گردد.
</div>
</div>

<div style="text-align:justify; margin-bottom:24px; padding-right:15px;">
<b style="color:#6D2E46;">تحلیل از منظر فضای کل حافظه <span style="direction:ltr; unicode-bidi:embed;">(Memory Space)</span>:</b>
<div style="padding-right:15px; margin-top:10px;">
<b style="color:#6D2E46;">در کش نتایج:</b> با وجود اینکه هر پرسمان فضای کمی می‌گیرد، اما به دلیل تنوع بی‌نهایتِ پرسمان‌های کاربران، ذخیره تمامی ترکیبات ممکن باعث انفجار فضای کش می‌شود.
<br>
<b style="color:#6D2E46;">در کش لیست‌ها:</b> لیست‌های رخداد برای کلمات متداول نظیر «وب» یا «داده» بسیار غول‌پیکر هستند و نگهداری کامل آن‌ها در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">RAM</span></b> فضای عظیمی از حافظه را اشغال می‌کند.
</div>
</div>

<div style="text-align:justify; margin:24px 0 4px 0; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46; color:#6D2E46;">
<span style="font-size:17pt; font-weight:bold; display:block; margin-bottom:10px;">نتیجه‌گیری و راهکار:</span>
در سیستم‌های بازیابی واقعی (مانند موتورهای جستجوی تجاری)، تکیه بر تنها یک رویکرد غیرمنطقی است. معماری استاندارد بر اساس یک <b style="color:#6D2E46;">سیستم کش چندسطحی <span style="direction:ltr; unicode-bidi:embed;">(Multi-level Caching)</span></b> بنا می‌شود:
<div style="padding-right:15px; margin-top:10px;">
<b style="color:#6D2E46;">1.</b> سیستم از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Result Caching</span></b> تنها برای کسر بسیار کوچکی از پرسمان‌ها (شاید 1 تا 5 درصد) که به شدت تکراری و پرطرفدار هستند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">Trending / Head Queries</span></b>) استفاده می‌کند تا بار پردازشی را به شدت کاهش دهد. برای رفع مشکل آپدیت، معمولاً این کش را با زمان‌های انقضای کوتاه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">TTL - Time to Live</span></b>) تنظیم می‌کنند تا داده‌های کهنه به صورت خودکار حذف شوند.
<br>
<b style="color:#6D2E46;">2.</b> سیستم از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">List Caching</span></b> برای کلمات مستقل پرکاربرد به عنوان لایه دوم استفاده می‌کند تا سرعت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed;">I/O</span></b> را برای سایر پرسمان‌های ترکیبیِ غیرتکراری تضمین نماید. این ترکیب، بهینه‌ترین حالت از تعادل میان حافظه پردازشی و سرعت به‌روزرسانی را فراهم می‌آورد.
</div>
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:center;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.16</span></b>: تحلیل مشکل قحطی نتیجه و گذر به بازیابی رتبه‌بندی شده
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه و بیان مشکل
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
یکی از چالش‌های بنیادین در طراحی سیستم‌های بازیابی اطلاعات (به‌ویژه در پلتفرم‌های تجارت الکترونیک)، مدیریت پرسمان‌های سخت‌گیرانه کاربران است. در این سناریو، یک خطای کلاسیک به نام <b style="color:#6D2E46;">پدیده قحطی اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Information Famine)</span></b> رخ داده است؛ جایی که سیستم به جای ارائه بهترین جایگزین‌های موجود، با اعمال منطق صفر و یکی، هیچ نتیجه‌ای به کاربر نشان نمی‌دهد و منجر به شکست تجربه کاربری (و از دست رفتن فروش) می‌گردد. در ادامه ریشه‌های تئوریک این بن‌بست و راهکار معماری آن تحلیل می‌شود.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) ریشه بن‌بست در مدل بولین کلاسیک <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Exact Match)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
همان‌طور که در مسائل پیشین بررسی شد، مدل بولین بر پایه جبر گزاره‌ها و تطابق دقیق (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Exact Match</span>) عمل می‌کند. عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span></b> در این مدل، به معنای «اشتراک مطلق» است.
<br>
فرض کنید سیستم در حال ارزیابی سندی با عنوان «لپ‌تاپ گیمینگ ایسوس <b style="color:#6D2E46;">16</b> گیگابایت» برای پرسمان کاربر (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">laptop AND gaming AND asus AND 32GB</span></b>) است. ارزیابی بولی این سند به شکل زیر انجام می‌شود:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">laptop</span></b> در سند وجود دارد <b style="color:#6D2E46;">1</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">gaming</span></b> در سند وجود دارد <b style="color:#6D2E46;">1</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">asus</span></b> در سند وجود دارد <b style="color:#6D2E46;">1</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">32GB</span></b> در سند وجود ندارد <b style="color:#6D2E46;">0</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
ارزیابی نهایی سند برابر است با:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{1 \text{ AND } 1 \text{ AND } 1 \text{ AND } 0 = 0} $$
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
با وجود اینکه سند مورد نظر <b style="color:#6D2E46;">75</b> درصد با نیاز کاربر هم‌پوشانی دارد و یک جایگزین بسیار مناسب است، اما به دلیل ماهیت <b style="color:#6D2E46;">شکننده <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Brittle)</span></b> مدل بولین، یک تک‌کلمه غایب (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">32GB</span></b>) کل معادله را باطل کرده و سند کاملاً از لیست نتایج حذف می‌شود (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Famine</span>).
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) گذار معماری: جستجوی متن آزاد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Free-text Query)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
برای حل این مشکل، موتورهای جستجوی مدرن از پارادایم جبر بولی فاصله گرفته و به سمت «جستجوی متن آزاد» حرکت می‌کنند. در این حالت، کاربر نیازی به استفاده از عملگرهای محدودکننده مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span></b> یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">OR</span></b> ندارد. پرسمان صرفاً به مجموعه‌ای از کلمات (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Bag of Words</span>) تبدیل می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; font-weight:bold; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; font-family:monospace;">
[laptop, gaming, asus, 32GB]
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
این تغییر سینتکس، به سیستم اجازه می‌دهد تا به جای فیلتر کردن قطعی اسناد، به سمت «ارزیابی میزان ارتباط» حرکت کند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) مدل رتبه‌بندی‌شونده <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Ranked Retrieval)</span> و مکانیزم <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Partial Match</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در مدل‌های رتبه‌بندی‌شونده (نظیر مدل فضای برداری یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Vector Space Model</span>)، ارزیابی اسناد دیگر یک فرآیند باینری (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">True / False</span>) نیست، بلکه یک <b style="color:#6D2E46;">تابع پیوسته <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Continuous Function)</span></b> است که عددی بین <b style="color:#6D2E46;">0</b> و <b style="color:#6D2E46;">1</b> (یا یک امتیاز اعشاری) را به عنوان میزان شباهت بازمی‌گرداند.
<br>
این مدل‌ها از مفهوم <b style="color:#6D2E46;">تطابق نسبی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Partial Match)</span></b> پشتیبانی می‌کنند. زمانی که پرسمان متن آزاد وارد سیستم می‌شود:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">1. وزن‌دهی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Weighting)</span>:</b> سیستم با استفاده از الگوریتم‌هایی نظیر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">TF-IDF</span></b>، به هر کلمه در پرسمان وزن می‌دهد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">2. محاسبه شباهت <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Similarity Scoring)</span>:</b> سیستم میزان شباهت سند «لپ‌تاپ گیمینگ ایسوس <b style="color:#6D2E46;">16</b> گیگابایت» را با پرسمان کاربر محاسبه می‌کند. از آنجا که <b style="color:#6D2E46;">3</b> کلمه از <b style="color:#6D2E46;">4</b> کلمه پرسمان در سند وجود دارد، این سند به جای دریافت امتیاز <b style="color:#6D2E46;">0</b>، یک امتیاز شباهت بالا (مثلاً <b style="color:#6D2E46;">0.85</b>) دریافت می‌کند.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">3. رتبه‌بندی و نمایش <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Ranking)</span>:</b> از آنجا که لپ‌تاپ <b style="color:#6D2E46;">32</b> گیگابایتی در فروشگاه موجود نیست تا امتیاز <b style="color:#6D2E46;">1.0</b> بگیرد، همین لپ‌تاپ <b style="color:#6D2E46;">16</b> گیگابایتی (با امتیاز <b style="color:#6D2E46;">0.85</b>) به عنوان بالاترین رتبه در جایگاه اول نتایج (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Top-1</span>) به کاربر نمایش داده می‌شود.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری ساختاری: تفاوت این دو مدل، تفاوت میان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">"Data Retrieval"</span> (بازیابی قطعی داده‌ها نظیر دستورات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">SQL</span>) و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">"Information Retrieval"</span> (بازیابی معنایی اطلاعات) است. مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Ranked Retrieval</span> با درک این موضوع که «عدم تطابق یک کلمه نباید به معنای بی‌ارزش بودن کل سند باشد»، پدیده <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Famine</span> را از بین برده و با ارائه بهترین جایگزین‌های ممکن (مرتبط‌ترین اسناد)، نرخ موفقیت جستجو را به شدت افزایش می‌دهد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:center;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.17</span></b>: معماری دوگانه نمایه و مدیریت بروزرسانی بلادرنگ
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه و بیان مسئله
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
در سیستم‌های بازیابی اطلاعات پویای مقیاس‌بزرگ (مانند خبرگزاری‌ها یا شبکه‌های اجتماعی)، نمایه‌سازی ایستا (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Static Indexing</span></b>) پاسخگو نیست. نمایه وارونه اصلی برای کاهش حجم، با الگوریتم‌های پیچیده‌ای فشرده‌سازی شده و روی دیسک ذخیره می‌شود. ساختار این فشرده‌سازی‌ها به هم پیوسته است؛ یعنی درج یک سند جدید در میان یک لیست رخداد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Postings List</span></b>) فشرده‌شده، نیازمند بازنویسی کل آن فایل روی دیسک است که از نظر محاسباتی و زمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">I/O</span></b> برای آپدیت‌های لحظه‌ای کاملاً غیرممکن محسوب می‌شود. راهکار استانداردِ معماری برای این بن‌بست، استفاده از «سیستم نمایه‌سازی دوگانه» است.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) طراحی معماری نمایه دوگانه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Dual Index Architecture)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
برای پشتیبانی از درج بلادرنگ اسناد، سیستم نمایه‌سازی به دو بخش کاملاً مجزا با وظایف متفاوت تقسیم می‌شود:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">نمایه اصلی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Main Index)</span>:</b> این نمایه ساختاری عظیم، به شدت فشرده و <b>فقط‌خواندنی</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Read-only</span></b>) دارد که روی دیسک سخت ذخیره شده است. هیچ سند جدیدی مستقیماً به این نمایه اضافه نمی‌شود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">نمایه کمکی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Auxiliary Index)</span>:</b> یک نمایه وارونه کوچک، غیرفشرده (یا با فشرده‌سازی بسیار سبک) و کاملاً پویا است که منحصراً در <b>حافظه اصلی</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>) نگهداری می‌شود. تمامی اخبار و اسناد جدیدی که وارد سیستم می‌شوند، بلافاصله و با هزینه زمانی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(1)</span></b> به این نمایه کمکی اضافه می‌شوند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) منطق اجرای پرسمان در سیستم دوگانه
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
وقتی کاربر پرسمانی (مثلاً <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Q</span></b>) را جستجو می‌کند، سیستم دیگر نمی‌تواند فقط به یک نمایه رجوع کند. پردازشگر پرسمان، درخواست را به صورت موازی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Parallel</span></b>) برای هر دو نمایه ارسال می‌کند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">1.</b> جستجو در نمایه اصلی و استخراج نتایج (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">R<sub>main</sub></span></b>).
<br>
<b style="color:#6D2E46;">2.</b> جستجو در نمایه کمکی (موجود در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>) و استخراج نتایج اخبار جدید (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">R<sub>aux</sub></span></b>).
<br>
<b style="color:#6D2E46;">3.</b> سیستم این دو مجموعه نتیجه را با هم ترکیب و رتبه‌بندی کرده و خروجی نهایی یعنی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">R<sub>main</sub> &cup; R<sub>aux</sub></span></b> را به کاربر نمایش می‌دهد.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) چالش ادغام <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Index Merging)</span> و هزینه آن
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
نمایه کمکی به دلیل قرارگیری در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>، با محدودیت جدی فضا مواجه است. وقتی حافظه اختصاص‌یافته به این نمایه پر شود، سیستم باید عملیات سنگینی به نام ادغام (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Merge</span></b>) را آغاز کند.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
در این مرحله، سیستم نمی‌تواند نمایه اصلی را درجا آپدیت کند. بلکه باید نمایه اصلیِ قبلی را به طور کامل از دیسک بخواند، آن را با نمایه کمکی (از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>) به صورت ترتیبی ادغام کند، و یک <b>نمایه اصلی کاملاً جدید</b> روی دیسک بنویسد. پس از اتمام این عملیات <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">I/O</span></b> سنگین، نمایه کمکی در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b> پاک (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Flush</span></b>) شده و سیستم به نمایه اصلیِ جدید سوئیچ می‌کند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">4</span></b>) رویکرد مدیریت اسناد حذف‌شده <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Deletion Handling)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
چالش حیاتی بعدی، حذف یک خبر (مثلاً به دلیل دستور قضایی یا اشتباه محتوایی) است. از آنجا که نمایه اصلی فقط‌خواندنی و فشرده است، پاک کردن فیزیکی شناسه آن خبر (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocID</span></b>) از داخل لیست‌های رخداد دیسک امکان‌پذیر نیست.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
برای رفع این چالش، معماران سیستم از تکنیک <b>بردار ابطال</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Invalidation Bit Vector</span></b>) استفاده می‌کنند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">ساختار:</b> سیستم یک آرایه بیتی بسیار سریع در حافظه اصلی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>) ایجاد می‌کند. هر بیت در این آرایه نماینده یک شناسه سند در سیستم است.
<br>
<b style="color:#6D2E46;">منطق حذف:</b> اگر سندی (مثلاً <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocID<sub>x</sub></span></b>) حذف شود، سیستم هیچ تغییری در نمایه‌ها ایجاد نمی‌کند، بلکه صرفاً بیت مربوط به <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocID<sub>x</sub></span></b> را در آرایه ابطال به مقدار <b style="color:#6D2E46;">1</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">True</span></b>) تغییر می‌دهد.
<br>
<b style="color:#6D2E46;">منطق جستجو:</b> هنگام اجرای هر پرسمان، پس از اینکه نتایج از نمایه اصلی و کمکی استخراج شد، سیستم پیش از نمایش نتایج به کاربر، آن‌ها را از فیلتر آرایه ابطال عبور می‌دهد. هر سندی که بیت آن <b style="color:#6D2E46;">1</b> باشد، به عنوان سند مرده (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Tombstone</span></b>) شناخته شده و از نتایج نهایی دور ریخته می‌شود.
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{\text{Results} = (R_{\text{main}} \cup R_{\text{aux}}) - D} $$
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری ریاضی معماری: با در نظر گرفتن مجموعه اسناد حذف‌شده با متغیر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">D</span></b>، منطق ریاضی این معماری برای تولید خروجی نهایی در زمان جستجو همواره به شکل رابطه بالا فرمول‌بندی می‌شود که توازنی بی‌نظیر میان سرعت آپدیت و کارایی جستجو برقرار می‌سازد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:center;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.18</span></b>: استراتژی پردازش پرسمان ترکیبی و تحلیل هزینه
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
مقدمه و طرح مسئله
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
در پردازش پرسمان‌های بازیابی اطلاعات، مفسر پرسمان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Query Parser</span></b>) در حالت پیش‌فرض از قوانین تقدم عملگرهای ریاضی پیروی می‌کند؛ به این معنا که همواره عبارات داخل پرانتز را در اولویت محاسبه قرار می‌دهد. اما زمانی که با توزیع نامتقارن کلمات (یک کلمه به شدت نادر و دو کلمه به شدت پرتکرار) مواجه هستیم، این تقدم ریاضی می‌تواند منجر به فروپاشی کارایی سیستم شود. در این نقطه، ورود ماژول <b style="color:#6D2E46;">بهینه‌ساز پرسمان (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Query Optimizer</span>)</b> برای تغییر استراتژی پردازش (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Execution Plan</span></b>) الزامی است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) تحلیل رویکرد سنتی: آیا محاسبه زودهنگام <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(B OR C)</span> منطقی است؟
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
پاسخ قطعی به این سوال <b style="color:#6D2E46;">«خیر»</b> است. از منظر معماری سیستم‌های مقیاس‌بزرگ، اجرای این پرانتز در گام نخست یک خطای فاحش (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Anti-pattern</span></b>) محسوب می‌شود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">تحلیل هزینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Cost Analysis)</span>:</b> عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">OR</span></b> (اجتماع) در ذات خود یک عملگر «بسط‌دهنده» است. کلمات <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">C</span></b> هر کدام <b style="color:#6D2E46;">1</b> میلیون شناسه سند دارند. اگر سیستم بخواهد حاصل (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B OR C</span></b>) را محاسبه کند، باید دو لیست عظیم را با الگوریتم ادغام پیمایش کند که پیچیدگی زمانی آن از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(|B| + |C|)</span></b> خواهد بود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">سربار حافظه:</b> سیستم مجبور است یک «لیست میانی» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Intermediate List</span></b>) با حداکثر <b style="color:#6D2E46;">2</b> میلیون عنصر را در حافظه اصلی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>) ایجاد کند. سپس در گام دوم، باید این لیست <b style="color:#6D2E46;">2</b> میلیونی را با لیست کوچک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> (که تنها <b style="color:#6D2E46;">10</b> عنصر دارد) اشتراک بگیرد.<br>
این رویکرد، میلیون‌ها عملیات مقایسه‌ای و تخصیص حافظه بی‌فایده را به پردازنده تحمیل می‌کند، در حالی که خروجی نهایی به دلیل اشتراک با لیست نادرِ <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b>، هرگز نمی‌تواند بیشتر از <b style="color:#6D2E46;">10</b> سند باشد!
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تغییر پارادایم در بهینه‌سازی: رویکرد جایگزین و بهینه
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
بهینه‌ساز پرسمان با بررسی آمار فرکانس اسناد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Document Frequency</span></b>)، متوجه می‌شود که لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> با طول <b style="color:#6D2E46;">10</b>، محدودکننده‌ترین بخش این پرسمان است. بنابراین، قانون طلایی بهینه‌سازی را اعمال می‌کند: <b style="color:#6D2E46;">«لیست نادر همواره باید به عنوان لیست پیشران (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Driving List</span>) در نظر گرفته شود.»</b>
<br>
برای پردازش سریع‌تر این پرسمان، بهینه‌ساز از محاسبه اجتماع (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B OR C</span></b>) به طور کامل صرف‌نظر کرده و استراتژی <b style="color:#6D2E46;">«پیمایش روی لیست کوچک، جستجو در لیست‌های بزرگ»</b> را جایگزین می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) اجرای مرحله‌به‌مرحله استراتژی جایگزین <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Probing Strategy)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در این رویکرد، الگوریتم به جای استفاده از دو اشاره‌گر همگام (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Standard Merge</span></b>)، از مکانیزم جستجوی دودویی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Binary Search</span></b>) یا پرشگرها (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Skip Pointers</span></b>) بهره می‌گیرد:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">1. پیمایش لیست پیشران:</b> سیستم حلقه پردازش را صرفاً روی عناصر لیست <b style="color:#6D2E46;">10</b> تایی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> باز می‌کند.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">2. آزمون شرط <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Probing)</span>:</b> برای هر شناسه سند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocID</span></b>) موجود در لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> (مثلاً سند شماره <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">X</span></b>):
<div style="padding-right:15px; margin-top:10px;">
ابتدا با استفاده از جستجوی دودویی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Binary Search</span></b>) چک می‌کند که آیا سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">X</span></b> در لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> وجود دارد؟
<br>
اگر وجود داشت، ارزیابی موفق است و سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">X</span></b> به لیست خروجی اضافه می‌شود. سیستم مستقیماً سراغ عنصر بعدی در لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b> می‌رود (الگوریتم میان‌بر منطقی یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Short-circuit evaluation</span></b>).
<br>
اگر در لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> نبود، با جستجوی دودویی چک می‌کند که آیا سند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">X</span></b> در لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">C</span></b> وجود دارد؟ اگر بود، به خروجی اضافه می‌شود و در غیر این صورت کنار گذاشته می‌شود.
</div>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">4</span></b>) مقایسه پیچیدگی زمانی (تفاوت فاجعه تا بهینگی)
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">هزینه الگوریتم مبتنی بر پرانتز (سنتی):</b> حدود <b style="color:#6D2E46;">2,000,000</b> عملیات مقایسه.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">هزینه الگوریتم جایگزین (بهینه‌شده):</b>
<br>
برای هر یک از <b style="color:#6D2E46;">10</b> عنصر لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A</span></b>، در بدترین حالت دو بار جستجوی دودویی (یکی در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">B</span></b> و یکی در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">C</span></b>) انجام می‌شود. پیچیدگی زمانی جستجوی دودویی در لیستی به طول <b style="color:#6D2E46;">1</b> میلیون برابر است با:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{\mathbf{\text{Cost(Binary Search)}} \approx \log_2(1{,}000{,}000) \approx 20} $$
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
بنابراین کل عملیات مقایسه برای این پرسمان برابر است با:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{\mathbf{\text{Total Probing Cost}} = 10 \times (\approx 20 + \approx 20) \approx 400} $$
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری ساختاری: با تغییر زاویه دید از «حل معادلات جبری» به «انتخاب ساختار داده هوشمند»، زمان اجرای پرسمان از <b style="color:#6D2E46;">2</b> میلیون واحد محاسباتی به تنها <b style="color:#6D2E46;">400</b> مقایسه کاهش می‌یابد. در موتورهای جستجوی واقعی، بهینه‌سازها با بازنویسی فیزیکی نحوه اجرای پرسمان‌ها (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Physical Execution Plan</span>)، از تشکیل لیست‌های میانی غول‌پیکر برای عملگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">OR</span> جلوگیری می‌کنند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.19</span></b>: تحلیل پیچیدگی زمانی اشتراک چندگانه و ارزیابی جدول درهم‌سازی
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
در پردازش پرسمان‌هایی با چندین عبارت محدودکننده (مانند <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">T<sub>1</sub> AND T<sub>2</sub> AND ... AND T<sub>K</sub></span></b>)، انتخاب الگوریتم مناسب برای یافتن اشتراک <b style="color:#6D2E46;">K</b> لیست رخداد، مرز بین یک سیستم بازیابی بلادرنگ (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Real-time</span></b>) و یک سیستم ناکارآمد را مشخص می‌کند. در این مسئله، هزینه الگوریتم استاندارد متوالی بررسی شده و سپس ساختار جدول درهم‌سازی به عنوان یک راهکار جایگزینِ احتمالی نقد می‌شود.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) تحلیل پیچیدگی زمانی الگوریتم ادغام متوالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Sequential Merge)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
الگوریتم استاندارد برای ادغام <b style="color:#6D2E46;">K</b> لیست (با میانگین طول <b style="color:#6D2E46;">N</b>) شامل دو فاز است: مرتب‌سازی اولیه لیست‌ها و سپس اشتراک‌گیری دو‌به‌دو.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">فاز اول (مرتب‌سازی بر اساس <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DF</span>):</b> سیستم ابتدا فرکانس سند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DF</span></b>) یا همان طولِ هر <b style="color:#6D2E46;">K</b> لیست را خوانده و آن‌ها را به صورت صعودی مرتب می‌کند. پیچیدگی زمانی این فاز برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(K log K)</span></b> است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">فاز دوم (اشتراک‌گیری متوالی):</b> ادغام از دو لیست کوچک‌تر آغاز می‌شود. هزینه اشتراک دو لیست اول با الگوریتم دو اشاره‌گره، در بدترین حالت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(N + N) = O(N)</span></b> است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">بدترین حالت ممکن <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Worst-case Scenario)</span>:</b> بدترین حالت زمانی رخ می‌دهد که اشتراک اسناد بین لیست‌ها کامل باشد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Complete Overlap</span></b>)؛ یعنی طول «لیست میانی» تولید شده پس از هر بار اشتراک، کاهش نیابد و همچنان <b style="color:#6D2E46;">N</b> باقی بماند.
<br>
در این شرایط، سیستم باید (<b style="color:#6D2E46;">K - 1</b>) بار عملیات ادغام را انجام دهد که در هر بار، یک لیست به طول <b style="color:#6D2E46;">N</b> با لیست بعدی (که آن هم در بدترین حالت طول <b style="color:#6D2E46;">N</b> دارد) مقایسه می‌شود. بنابراین هزینه کل فاز دوم برابر خواهد بود با:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{\boldsymbol{\mathcal{O}((K - 1) \times N) \approx \mathcal{O}(K \times N)}} $$
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">مرتبه زمانی کل:</b> با ترکیب دو فاز، پیچیدگی زمانی کلی برابر با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(K log K + K N)</span></b> خواهد بود. از آنجا که در سیستم‌های واقعی تعداد کلمات پرسمان (<b style="color:#6D2E46;">K</b>) بسیار کوچکتر از طول لیست‌ها (<b style="color:#6D2E46;">N</b>) است (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">K &ll; N</span></b>)، بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">K log K</span></b> نادیده گرفته شده و مرتبه زمانی الگوریتم در بدترین حالت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(K N)</span></b> تقریب زده می‌شود.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) آیا جدول درهم‌سازی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Hash Table)</span> جایگزین بهتری است؟
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
پاسخ کوتاه به این ادعا <b style="color:#6D2E46;">«خیر»</b> است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
ایده استفاده از جدول درهم‌سازی (ساخت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Hash</span></b> از کوچکترین لیست و جستجوی عناصر لیست‌های دیگر در آن با زمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(1)</span></b> برای هر عنصر)، روی کاغذ جذاب به نظر می‌رسد و به پیچیدگی زمانی ظاهری <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(K N)</span></b> می‌رسد. اما در معماری موتورهای جستجو و مهندسی سیستم، این رویکرد به دلایل ساختاری زیر کاملاً مطرود (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Anti-pattern</span></b>) است:
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) معایب استفاده از <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Hash Table</span> در ادغام لیست‌های ادعا
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">الف) افت شدید کارایی حافظه نهان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Cache Locality Failure)</span>:</b>
<br>
بزرگترین نقطه ضعف جدول درهم‌سازی، نحوه دسترسی آن به حافظه است. الگوریتم دو اشاره‌گره (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Standard Merge</span></b>) روی آرایه‌ها حرکت می‌کند. پردازنده‌ها برای خواندن آرایه‌ها به شدت بهینه‌سازی شده‌اند و به دلیل ویژگی «مکان‌گزینی فضایی» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Spatial Locality</span></b>)، بلوک‌های بعدی لیست را از پیش در حافظه نهان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">CPU Cache</span></b>) بارگذاری می‌کنند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Prefetching</span></b>). اما جستجو در جدول درهم‌سازی نیازمند دسترسی تصادفی به حافظه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Random Memory Access</span></b>) است که منجر به خطاهای پیاپیِ کش (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Cache Miss</span></b>) شده و سرعت پردازنده را برای واکشی داده از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b> به شدت کاهش می‌دهد. در عمل، زمان پنهانِ یک <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Cache Miss</span></b> بسیار بیشتر از چند مقایسه ساده عددی است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">ب) پیچیدگی فضایی و سربار حافظه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Memory Overhead)</span>:</b>
<br>
الگوریتم استاندارد دو اشاره‌گره به صورت درجا (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">In-place</span></b>) عمل کرده و به فضای کمکی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Auxiliary Space</span></b>) نیازی ندارد؛ پیچیدگی فضایی آن <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(1)</span></b> است. اما ساخت یک جدول درهم‌سازی برای لیست‌های میلیونی، نیازمند تخصیص حافظه جانبی عظیم از مرتبه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(N)</span></b> است. علاوه بر این، برای جلوگیری از تصادم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Collisions</span></b>)، ضریب بار (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Load Factor</span></b>) جدول باید پایین نگه داشته شود که به معنای تخصیص حافظه‌ای به مراتب بیشتر از تعداد خود شناسه‌هاست.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">ج) از دست رفتن ترتیب اسناد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Loss of Sorting)</span>:</b>
<br>
در سیستم‌های بازیابی، خروجی پرسمان باید همواره بر اساس شناسه‌های سند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocID</span></b>) مرتب باشد تا در صورت نیاز به ترکیب با عملگرهای دیگر (نظیر فیلترهای زمانی، یا پرسمان‌های عبارتی) قابل استفاده باشد. جدول درهم‌سازی ترتیب اعداد را از بین می‌برد. اگر سیستم از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Hash</span></b> استفاده کند، در انتهای کار مجبور است لیست خروجی را مجدداً با زمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(R log R)</span></b> (که <b style="color:#6D2E46;">R</b> تعداد نتایج است) مرتب کند که این امر سربار زمانی اضافه‌ای را تحمیل می‌سازد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">د) ضرایب ثابت پنهان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(High Constant Factors)</span>:</b>
<br>
گرچه زمان جستجو در جدول درهم‌سازی با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(1)</span></b> نمایش داده می‌شود، اما این <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(1)</span></b> شامل محاسبه تابع درهم‌سازی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Hash Function</span></b>) و مدیریت تصادم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Collision Resolution</span></b>) برای هر تک‌تک شناسه‌هاست. محاسبه این تابع برای پردازنده، بسیار پرهزینه‌تر از یک عملگر مقایسه ساده (<b>&lt;</b> یا <b>&gt;</b>) در الگوریتم دو اشاره‌گره است.
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نکته کاربردی برای ارزیابی دانشجویان: در جلسات دفاع یا تصحیح برگه‌ها، اگر دانشجویی پیشنهاد استفاده از جدول درهم‌سازی را داد، از او بخواهید تفاوت زمان تئوری (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O</span>) و زمان واقعی سیستم (با در نظر گرفتن معماری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">CPU</span> و کش) را توضیح دهد. این سوال معیار بسیار دقیقی برای سنجش میزان تسلط دانشجو بر مفاهیم <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Systems Programming</span> در کنار الگوریتم‌هاست.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.20</span></b>: تحلیل خطای پیاده‌سازی عملگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">AND</span> و تداخل مدل‌های بازیابی
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه و طرح مسئله
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
در مهندسی سیستم‌های بازیابی اطلاعات، تفکیک دقیق میان پارادایم‌های مختلف جستجو از اهمیت بالایی برخوردار است. خطای فاحش تیم دانشجویی در این پروژه، ناشی از عدم درک مرزهای تئوریک میان «مدل بازیابی بولی کلاسیک» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Classic Boolean Retrieval</span></b>) و «مدل‌های بازیابی رتبه‌بندی‌شونده» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Ranked Retrieval Models</span></b>) است. آن‌ها به جای پیاده‌سازی منطق مجموعه‌ها، یک تابع امتیازدهی خام را در بستری اشتباه پیاده‌سازی کرده‌اند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) چرا این پیاده‌سازی در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Boolean</span> از اساس غلط است؟
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
مدل بازیابی بولین بر دو پایه اساسی استوار است: <b style="color:#6D2E46;">نظریه مجموعه‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Set Theory)</span></b> و مفهوم <b style="color:#6D2E46;">تطابق دقیق <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Exact Match)</span></b>.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">ارزیابی دودویی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Binary Evaluation)</span> در برابر فرکانس:</b> در مدل بولین کلاسیک، تنها شاخص معنادار، <b style="color:#6D2E46;">«حضور یا عدم حضور» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Presence)</span></b> یک کلمه در سند است. سیستم برای هر کلمه در یک سند تنها مقادیر <b style="color:#6D2E46;">1</b> (وجود دارد) یا <b style="color:#6D2E46;">0</b> (وجود ندارد) را می‌پذیرد. فرکانس کلمه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Term Frequency</span></b>)، یعنی اینکه کلمه <b style="color:#6D2E46;">1</b> بار در متن آمده یا <b style="color:#6D2E46;">1000</b> بار، در این مدل فاقد ارزش محاسباتی است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">نقض منطق عملگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span> (اشتراک مجموعه‌ها):</b> عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span></b> به معنای تقاطع قطعی لیست‌های رخداد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">A &cap; B</span></b>) است. وقتی دانشجویان به جای اشتراک، فرکانس کلمات را جمع می‌زنند، در واقع ماهیت محدودکننده <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span></b> را از بین برده و آن را تبدیل به یک عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">OR</span></b> با وزن‌دهی غلط کرده‌اند. در این پیاده‌سازی اشتباه، فقدان کامل یک کلمه می‌تواند توسط تکرار بیش از حد کلمه‌ای دیگر جبران شود، که این امر نقض صریح منطق جبر بول است.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) بررسی یک مثال نقض برای اثبات فروپاشی الگوریتم دانشجویان
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
فرض کنید کاربر پرسمان عبارتی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Security AND Algorithm</span></b> را وارد کرده است. سیستم باید اسنادی را برگرداند که <b>هر دو</b> کلمه در آن‌ها وجود داشته باشد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">سند شماره 1:</b> یک مقاله کاملاً نامرتبط درباره «انواع قفل‌های امنیتی درب». کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Security</span></b> در این متن <b style="color:#6D2E46;">100</b> بار تکرار شده است، اما کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Algorithm</span></b> اصلاً وجود ندارد (تکرار = <b style="color:#6D2E46;">0</b>).
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">سند شماره 2:</b> یک مقاله کاملاً مرتبط درباره «الگوریتم‌های رمزنگاری». کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Security</span></b> <b style="color:#6D2E46;">2</b> بار و کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Algorithm</span></b> <b style="color:#6D2E46;">3</b> بار در متن آمده است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
<b style="color:#6D2E46;">رفتار الگوریتم دانشجویی:</b>
<br>
جمع فرکانس در سند اول: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">100 + 0 = 100</span></b>
<br>
جمع فرکانس در سند دوم: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2 + 3 = 5</span></b>
<br><br>
الگوریتم دانشجویان سند کاملاً نامرتبطِ اول را با امتیاز <b style="color:#6D2E46;">100</b> به عنوان بهترین نتیجه برمی‌گرداند! زیرا سند اول با وجود نداشتن نیمی از پرسمان کاربر، به دلیل جمع بالای فرکانس کلمه دیگر، سند دوم را شکست داده است. این در حالی است که در مدل بولین صحیح، خروجی سند اول دقیقاً برابر با <b style="color:#6D2E46;">0</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1 &land; 0 = 0</span></b>) و خروجی سند دوم برابر با <b style="color:#6D2E46;">1</b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1 &land; 1 = 1</span></b>) است.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) مفهوم واردشده به اشتباه: تداخل با مدل فضای برداری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Vector Space Model)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
این گروه دانشجویی، به صورت ناخودآگاه و ناقص، دو مفهوم کلیدی از سیستم‌های رتبه‌بندی مدرن را وارد سیستم بولین کرده است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">مدل کیسه کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Bag of Words)</span>:</b> آن‌ها به جای نگاه باینری (ماتریس تلاقی)، به متن به عنوان ظرفی از کلمات نگاه کرده‌اند که تکرار در آن مجاز و معنادار است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">وزن‌دهی بر اساس فرکانس کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Frequency Weighting - TF)</span>:</b> جمع زدن تکرار کلمات، ابتدایی‌ترین شکل از امتیازدهی در مدل فضای برداری (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Vector Space Model</span></b>) است.
</div>

<div style="text-align:justify; font-size:16pt; color:#1B263B; margin-top:24px; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46;">
<div style="text-align:right; font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:10px;">نتیجه‌گیری ساختاری</div>
اما خطای مضاعف آن‌ها این است که این امتیازدهی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Scoring</span></b>) را به صورت ایزوله و خام اعمال کرده‌اند. در مدل‌های رتبه‌بندی استاندارد، استفاده از فرکانس کلمه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">TF</span></b>) به تنهایی فاجعه‌بار است و حتماً باید با فاکتورهای متعادل‌کننده‌ای نظیر <b style="color:#6D2E46;">فرکانس معکوس سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">IDF - Inverse Document Frequency</span>)</b> ترکیب شود (فرمول <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">TF-IDF</span></b>) و سپس توسط معیارهایی مانند <b style="color:#6D2E46;">شباهت کسینوسی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Cosine Similarity</span>)</b> نرمال‌سازی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Normalize</span></b>) گردد تا طول اسناد و اهمیت کلمات نایاب در محاسبه نهایی لحاظ شود. ترکیب کردن منطق بولین با فرکانس‌های خام، نه دقت مدل اول را دارد و نه هوشمندی مدل دوم را.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.21</span></b>: تحلیل شکست تجاری مدل بولین و گذر به بازیابی رتبه‌بندی شده
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
طراحی سیستم‌های جستجو در پلتفرم‌های تجارت الکترونیک (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">E-commerce</span></b>)، نیازمند درک عمیق از رفتار کاربر و انعطاف‌پذیری سیستم است. در این سناریو، فروشگاه با یکی از مخرب‌ترین خطاهای سیستم‌های بازیابی به نام <b style="color:#6D2E46;">«قحطی اطلاعات» (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Information Famine</span>)</b> مواجه شده است؛ وضعیتی که سیستم به جای ارائه بهترین جایگزین‌های ممکن، خروجی تهی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Null Result</span></b>) برمی‌گرداند و منجر به شکست قیف فروش می‌شود. در ادامه، ریشه‌های تئوریک این شکست و راهکار معماری آن تحلیل می‌گردد.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">الف</span></b>) تحلیل تئوریک بن‌بست در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Boolean</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Extended Boolean</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
ریشه این شکست تجاری، در ماهیت ریاضیاتی و سخت‌گیرانه مدل بولین نهفته است. مدل بولین بر پایه مفهوم <b style="color:#6D2E46;">«تطابق دقیق» (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Exact Match</span>)</b> استوار است. در این پارادایم، ارزیابی اسناد به صورت کاملاً دودویی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Binary</span></b>) انجام می‌شود و هیچ درکی از «درجه ارتباط» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Degree of Relevance</span></b>) یا «نزدیکی معنایی» وجود ندارد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
زمانی که سیستم در حال پردازش پرسمانِ مقیدشده با عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span></b> است، شرط لازم و کافی برای بازیابی یک سند، حضور <b style="color:#6D2E46;">تمامی</b> کلمات پرسمان در آن سند است.
<br><br>
رفتار سیستم را برای یک لپ‌تاپ «گیمینگ ایسوس 32 گیگابایت» (که یک جایگزین عالی است) بررسی می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">laptop</span></b> در سند موجود است <b style="color:#6D2E46;">1</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">gaming</span></b> در سند موجود است <b style="color:#6D2E46;">1</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">asus</span></b> در سند موجود است <b style="color:#6D2E46;">1</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
کلمه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">16GB</span></b> در سند <b>موجود نیست</b> <b style="color:#6D2E46;">0</b>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
ارزیابی نهایی خروجی سیستم برای این سند بر اساس جبر بول به شکل زیر خواهد بود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
$$ \color{#6D2E46}{1 \text{ AND } 1 \text{ AND } 1 \text{ AND } 0 = 0} $$
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
<b style="color:#6D2E46;">دلیل شکست تجاری:</b> با وجود اینکه این سند 75٪ با نیاز کاربر هم‌پوشانی دارد، اما مدل بولین به دلیل شکنندگی ذاتی خود (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Brittleness</span></b>)، با غیبت تنها یک کلمه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">16GB</span></b>)، کل سند را بی‌اعتبار دانسته و به آن امتیاز <b style="color:#6D2E46;">0</b> می‌دهد. مدل <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Extended Boolean</span></b> نیز (حتی با وجود عملگرهای مجاورت)، در نهایت بر پایه همین منطق حذفی کار می‌کند. در دنیای واقعی، کاربران معمولاً شرایط سخت‌گیرانه‌ای ندارند و آماده پذیرش جایگزین‌های نزدیک (مثل برند دیگر یا حافظه بالاتر) هستند؛ اما مدل بولین این شانس را از کاربر و کسب‌وکار می‌گیرد.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">ب</span></b>) گذار معماری به <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Ranked Retrieval</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Free-text Query</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
برای جلوگیری از این بن‌بست، موتورهای جستجوی مدرن از منطق حذفی فاصله‌گرفته و به سمت ارزیابی پیوسته حرکت می‌کنند. این تغییر معماری نیازمند دو گام اساسی است:
</div>

<div style="font-size:16pt; font-weight:bold; color:#6D2E46; margin-bottom:12px; margin-top:20px;">
1. حذف عملگرهای بولی و استفاده از <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Free-text Query</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در سیستم جدید، کاربر نیازی به وارد کردن عملگرهای منطقی ندارد. پرسمان از حالت یک «معادله شرطی» خارج شده و به یک «کیسه کلمات» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Bag of Words</span></b>) تبدیل می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; font-weight:bold; color:#6D2E46; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; font-family:monospace;">
[laptop, gaming, asus, 16GB]
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در این حالت، سیستم کلمات را به عنوان «ترجیحات کاربر» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Soft Constraints</span></b>) می‌بیند، نه «شروط الزام‌آور» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Hard Constraints</span></b>).
</div>

<div style="font-size:16pt; font-weight:bold; color:#6D2E46; margin-bottom:12px; margin-top:20px;">
2. استفاده از مکانیزم تطابق نسبی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Partial Match)</span> در <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Ranked Retrieval</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در مدل‌های رتبه‌بندی‌شونده (مانند مدل فضای برداری - <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Vector Space Model</span></b>)، ارزیابی اسناد دیگر خروجی <b style="color:#6D2E46;">0</b> و <b style="color:#6D2E46;">1</b> ندارد، بلکه تابع ارزیابی یک امتیاز پیوسته (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Score</span></b>) را برمی‌گرداند.
<br>
نحوه برخورد مفهومی سیستم جدید با این پرسمان به شرح زیر است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">وزن‌دهی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Weighting)</span>:</b> سیستم به هر کلمه در پرسمان و اسناد یک وزن عددی اختصاص می‌دهد (مثلاً با الگوریتم <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">TF-IDF</span></b>).
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">محاسبه شباهت <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Similarity Calculation)</span>:</b> سیستم میزان شباهت هر کالا را با پرسمان کاربر می‌سنجد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:30px;">
<b style="color:#6D2E46;">کالای الف (لپ‌تاپ گیمینگ ایسوس <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">32GB</span>):</b> 3 کلمه از 4 کلمه پرسمان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">laptop, gaming, asus</span></b>) را داراست. بنابراین امتیاز بالایی (مثلاً <b style="color:#6D2E46;">0.75</b>) دریافت می‌کند.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:30px;">
<b style="color:#6D2E46;">کالای ب (لپ‌تاپ گیمینگ لنوو <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">16GB</span>):</b> این کالا نیز 3 کلمه از 4 کلمه پرسمان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">laptop, gaming, 16GB</span></b>) را داراست و امتیاز مشابهی (مثلاً <b style="color:#6D2E46;">0.70</b>) می‌گیرد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">رتبه‌بندی و نمایش <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Ranking)</span>:</b> سیستم اسناد را بر اساس امتیاز کسب‌شده به صورت نزولی مرتب می‌کند.
</div>

<div style="text-align:right; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری ساختاری: در سیستم <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Ranked Retrieval</span>، با وجود اینکه کالای کاملاً منطبق (100٪) در انبار وجود ندارد، خروجی سیستم تهی نخواهد بود. سیستم بهترین جایگزین‌های موجود (لپ‌تاپ‌های ایسوس 32 گیگابایتی و لنوو 16 گیگابایتی) را در صدر نتایج (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Top-K</span>) به کاربر نمایش می‌دهد. این امر با ارضای نسبی نیاز کاربر، پدیده قحطی اطلاعات را حل کرده و مستقیماً باعث نجات نرخ تبدیل (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Conversion Rate</span>) فروشگاه می‌شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.22</span></b>: طراحی سناریوی بدترین حالت در اشاره‌گرهای پرشی
</span>
<br>
<span style="font-size:16pt; font-weight:bold; color:#6D2E46; direction:ltr; unicode-bidi:embed; display:inline-block; margin-top:8px;">
(Worst-case Scenario Design in Skip Pointers)
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه و طرح مسئله
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در پیاده‌سازی سیستم‌های بازیابی اطلاعات، استفاده از پرشگرها (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Skip Pointers</span></b>) با هدف کاهش پیچیدگی زمانی در ادغام لیست‌های طولانی (عملگر <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">AND</span></b>) پیشنهاد می‌شود. ایده اصلی این است که با پرش از روی بخش‌هایی از لیست که قطعاً در اشتراک نقشی ندارند، تعداد مقایسه‌ها کاهش یابد. اما پرشگرها یک هزینه پنهان (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Hidden Cost</span></b>) دارند: <b style="color:#6D2E46;">سربار ارزیابی شرط پرش</b>.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
الگوریتم برای انجام یک پرش، باید علاوه بر مقایسه اصلی دو عنصر، بررسی کند که آیا «مقصدِ پرشگر همچنان کوچکتر یا مساوی با مقدار فعلی در لیست دوم است یا خیر». اگر سیستم مدام این شرط را بررسی کند اما پرشی صورت نگیرد، الگوریتم از حالت خطی استاندارد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(x+y)</span></b>) نیز کندتر خواهد شد.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) طراحی سناریو و مثال نقض عددی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Worst-case Scenario)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
بدترین حالت برای پرشگرها زمانی رخ می‌دهد که شناسه‌های اسناد (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocIDs</span></b>) در دو لیست، به شدت متراکم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Dense</span></b>) و <b style="color:#6D2E46;">درهم‌تنیده (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Interleaved</span>)</b> باشند. در این حالت، مقادیر دو لیست یکی‌درمیان از یکدیگر بزرگتر می‌شوند و عملاً هیچ شکاف (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Gap</span></b>) بزرگی برای پرش وجود ندارد.
<br><br>
برای اثبات این موضوع، دو لیست زیر را در نظر می‌گیریم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">لیست <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">L<sub>1</sub></span>:</b> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">[1, 3, 5, 7, 9, 11]</span> (طول لیست: 6)<br>
<b style="color:#6D2E46;">لیست <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">L<sub>2</sub></span>:</b> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">[2, 4, 6, 8, 10, 12]</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
<b style="color:#6D2E46;">تنظیمات پرشگر برای لیست <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">L<sub>1</sub></span>:</b><br>
بر اساس فرمول استاندارد، طول گام پرش (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Skip Step</span></b>) معمولاً برابر با جذر طول لیست (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">&radic;N</span></b>) در نظر گرفته می‌شود. برای لیست <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">L<sub>1</sub></span></b>، طول گام <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">&radic;6 &approx; 2</span></b> است. بنابراین در هر دو خانه، یک پرشگر قرار می‌دهیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
از اندیس 0 (مقدار <b style="color:#6D2E46;">1</b>) پرشگر به اندیس 2 (مقدار <b style="color:#6D2E46;">5</b>) اشاره می‌کند.<br>
از اندیس 2 (مقدار <b style="color:#6D2E46;">5</b>) پرشگر به اندیس 4 (مقدار <b style="color:#6D2E46;">9</b>) اشاره می‌کند.<br>
از اندیس 4 (مقدار <b style="color:#6D2E46;">9</b>) پرشگر به انتهای لیست متصل است.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) ردیابی گام‌به‌گام و اثبات افت کارایی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Trace)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
اکنون رفتار الگوریتم دارای پرشگر را روی این دو لیست ردیابی می‌کنیم. فرض کنید <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub></span></b> اشاره‌گر لیست اول و <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub></span></b> اشاره‌گر لیست دوم است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">گام اول (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub> = 1</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub> = 2</span>):</b><br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1 < 2</span></b> (نتیجه: 1 مقایسه).<br>
بررسی شرط پرش: چون <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">L<sub>1</sub></span></b> کوچکتر است و پرشگر دارد، سیستم چک می‌کند که آیا مقدار مقصدِ پرشگر (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">value = 5</span></b>) کوچکتر یا مساوی با <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub></span></b> (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">value = 2</span></b>) است؟ <b style="color:#6D2E46;">خیر (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">5 &not;&le; 2</span>).</b><br>
نتیجه: پرش لغو می‌شود. <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub></span></b> یک خانه عادی به جلو می‌رود. <i>(1 مقایسه بی‌فایده به پردازنده تحمیل شد)</i>.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">گام دوم (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub> = 3</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub> = 2</span>):</b><br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3 > 2</span></b> (نتیجه: 1 مقایسه).<br>
چون <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">L<sub>2</sub></span></b> کوچکتر است، <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub></span></b> جلو می‌رود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">گام سوم (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub> = 3</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub> = 4</span>):</b><br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3 < 4</span></b> (نتیجه: 1 مقایسه). در اینجا پرشگری وجود ندارد. <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub></span></b> عادی جلو می‌رود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">گام چهارم (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub> = 5</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub> = 4</span>):</b><br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">5 > 4</span></b> (نتیجه: 1 مقایسه). <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub></span></b> جلو می‌رود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">گام پنجم (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub> = 5</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>2</sub> = 6</span>):</b><br>
مقایسه اصلی: <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">5 < 6</span></b> (نتیجه: 1 مقایسه).<br>
بررسی شرط پرش: پرشگر به عدد 9 اشاره دارد. آیا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">9 &le; 6</span></b> است؟ <b style="color:#6D2E46;">خیر.</b><br>
نتیجه: پرش لغو می‌شود. <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P<sub>1</sub></span></b> عادی به جلو می‌رود. <i>(1 مقایسه بی‌فایده دیگر)</i>.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
<b style="color:#6D2E46;">نتیجه‌گیری بخش ردیابی:</b> اگر این دو لیست را با الگوریتم استاندارد (بدون پرش) ادغام می‌کردیم، برای رسیدن به انتهای لیست اول دقیقاً به <b style="color:#6D2E46;">11 مقایسه</b> نیاز داشتیم. اما با فعال بودن پرشگرها، علاوه بر آن 11 مقایسه اصلی، سیستم مجبور شد شرط‌های پرشِ لغوشده را نیز بررسی کند که تعداد کل مقایسه‌ها را افزایش داد (به عنوان مثال <b style="color:#6D2E46;">13 مقایسه</b> یا بیشتر). در مقیاس‌های میلیونی، این شاخه‌های شرطیِ بی‌فایده (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Branch Prediction Misses</span></b>) باعث افت شدید سرعت <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">CPU</span></b> می‌شوند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) تحلیل تاثیر طول گام پرش <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Impact of Skip Step Size)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در این حالت بدبینانه (لیست‌های درهم‌تنیده)، تغییر طول گام پرش چه تاثیری دارد؟
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">حالت الف: گام‌های کوتاه (تعداد پرشگرهای زیاد - مثلاً پرش هر 1 یا 2 خانه):</b><br>
این <b style="color:#6D2E46;">فاجعه‌بارترین</b> سناریو است. هرچه گام‌ها کوتاه‌تر باشند، تعداد گره‌های دارای پرشگر بیشتر می‌شود. در نتیجه، الگوریتم در هر قدم تلاش می‌کند پرش کند، شرط پرش با شکست مواجه می‌شود، و سیستم بالاترین حد از مقایسه‌های زائد و سربار محاسباتی را متحمل می‌گردد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">حالت ب: گام‌های بلند (تعداد پرشگرهای کم - مثلاً پرش هر 1000 خانه):</b><br>
در این حالت، تعداد گره‌های دارای پرشگر در لیست کاهش می‌یابد. بنابراین الگوریتم کمتر با شرط‌های پرش برخورد می‌کند و تعداد خطاهای بررسی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Overhead</span></b>) کمتر می‌شود. در این سناریو، رفتار سیستم تا حد زیادی به همان ادغام خطی استاندارد نزدیک می‌شود (هرچند باز هم اندکی کندتر از آن است).
</div>

<div style="text-align:justify; font-size:16pt; color:#1B263B; margin-top:24px; background-color:#FFFFFF; padding:18px; border-radius:10px; border:1px dashed #6D2E46;">
<div style="text-align:right; font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:10px;">خلاصه معماری</div>
این مسئله اثبات می‌کند که پرشگرها یک راهکار جادویی برای افزایش سرعت نیستند. این ساختارها تنها زمانی مفیدند که با «لیست‌هایی با طول به شدت نامساوی» مواجه باشیم یا توزیع اعداد دارای «شکاف‌های بزرگ» (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Large Gaps</span></b>) باشد تا پرش‌های موفق بتوانند هزینه مقایسه‌های سربار را جبران کنند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:24px 28px; text-align:right; line-height:2.1;">

<div style="margin-bottom:20px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1.23</span></b>: معماری نمایه پویا
</span>
<br>
<span style="font-size:16pt; font-weight:bold; color:#6D2E46; direction:ltr; unicode-bidi:embed; display:inline-block; margin-top:8px;">
(Dynamic Indexing)
</span>
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
مقدمه و طرح چالش معماری
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify;">
در موتورهای جستجو و سیستم‌های بازیابی اطلاعات، نمایه وارونه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Inverted Index</span></b>) یک ساختار داده به شدت فشرده و به هم پیوسته است. هرگونه تلاش برای درج مستقیم (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">In-place Insertion</span></b>) یک سند جدید در نمایه‌ای که روی دیسک (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Disk</span></b>) قرار دارد، نیازمند بازنویسی بخش‌های عظیمی از فایل‌های روی دیسک است. این عملیات، سربار ورودی/خروجی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">I/O Overhead</span></b>) بسیار سنگینی دارد و معماری سیستم را در مواجهه با ده‌ها هزار آپدیت روزانه (نظیر خبرگزاری‌ها) فلج می‌کند.
<br><br>
برای حل این تضاد میان «حجم عظیم داده‌های ایستا» و «نیاز به آپدیت‌های بلادرنگ»، مهندسی نرم‌افزار رویکردی استاندارد تحت عنوان <b style="color:#6D2E46;">معماری نمایه دوگانه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Dual Index Architecture</span>)</b> را ارائه می‌دهد.
</div>



<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) ساختار معماری نمایه دوگانه
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
در این رویکرد، سیستم نمایه‌سازی از یکپارچگی خارج شده و به دو جزء با عملکردهای متفاوت و سلسله‌مراتب حافظه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Memory Hierarchy</span></b>) مجزا تقسیم می‌گردد:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">الف) نمایه اصلی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Main Index</span>):</b>
<br>
این نمایه حاوی آرشیو کلان خبرگزاری است. ساختار آن کاملاً فشرده، ایستا و <b style="color:#6D2E46;">فقط‌خواندنی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Read-only</span>)</b> است که منحصراً روی دیسک سخت (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">HDD/SSD</span></b>) نگهداری می‌شود. هیچ سند جدیدی به صورت مستقیم و در لحظه به این نمایه اضافه نمی‌گردد.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">ب) نمایه کمکی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Auxiliary Index</span>):</b>
<br>
یک نمایه وارونه کوچک، پویا و غیرفشرده است که به طور کامل در <b style="color:#6D2E46;">حافظه اصلی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span>)</b> سیستم مستقر می‌شود. از آنجا که عملیات در <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b> با سرعت پردازنده انجام می‌گیرد، هر خبر جدیدی که وارد خبرگزاری می‌شود، در زمان <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">O(1)</span></b> به این نمایه کمکی اضافه شده و فوراً قابل جستجو می‌گردد.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) منطق اجرای پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query Execution)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
زمانی که کاربر پرسمانی را در این خبرگزاری جستجو می‌کند، موتور جستجو به جای یک سیستم، به دو سیستم موازی متوسل می‌شود:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">1.</b> پردازشگر پرسمان، درخواست را همزمان به نمایه اصلی (دیسک) و نمایه کمکی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b>) ارسال می‌کند.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">2.</b> نتایج آرشیو قدیمی از نمایه اصلی استخراج می‌شود (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">R<sub>main</sub></span></b>).
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">3.</b> نتایج اخبار ثانیه‌های اخیر از نمایه کمکی استخراج می‌شود (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">R<sub>aux</sub></span></b>).
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
<b style="color:#6D2E46;">4.</b> سیستم در نهایت این دو مجموعه را ادغام (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Merge</span></b>) و بر اساس الگوریتم‌های رتبه‌بندی مرتب کرده و خروجی یکپارچه یعنی <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">R<sub>main</sub> &cup; R<sub>aux</sub></span></b> را به کاربر بازمی‌گرداند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) مدیریت سرریز حافظه و ادغام دوره‌ای <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Periodic Index Merging)</span>
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
ظرفیت حافظه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b> محدود است و نمایه کمکی نمی‌تواند تا ابد رشد کند. وقتی حجم نمایه کمکی به آستانه مشخصی (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Threshold</span></b>) رسید، سیستم باید فاز سنگین «ادغام» را آغاز کند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:24px; text-align:justify; padding-right:15px;">
در این مرحله، سیستم در پس‌زمینه (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Background Thread</span></b>)، نمایه اصلی قدیمی را از دیسک می‌خواند، آن را با نمایه کمکی ادغام می‌کند، و یک <b style="color:#6D2E46;">نمایه اصلی کاملاً جدید</b> روی دیسک می‌نویسد. پس از تکمیل این فرآیند، نمایه کمکی از <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b> پاک (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Flush</span></b>) شده و سیستم به نمایه اصلی جدید سوئیچ می‌کند.
</div>

<div style="font-size:17pt; font-weight:bold; color:#6D2E46; margin-bottom:12px;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">4</span></b>) چالش حذف اخبار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Deletion)</span> و بردار ابطال
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify;">
اگر خبری به دلایل حقوقی یا اشتباه محتوایی حذف شود، امکان پاک کردن آن از نمایه اصلیِ فقط‌خواندنی وجود ندارد. راهکار معماری برای این چالش، استفاده از <b style="color:#6D2E46;">بردار ابطال (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Invalidation Bit Vector</span>)</b> است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
سیستم یک آرایه بیتی بسیار سریع در حافظه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RAM</span></b> نگه می‌دارد که هر بیت آن نماینده شناسه یک سند (<b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">DocID</span></b>) است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:10px; text-align:justify; padding-right:15px;">
با حذف یک خبر، سیستم تنها بیت مربوط به شناسه آن خبر را در این آرایه برابر با <b style="color:#6D2E46;">1</b> قرار می‌دهد (نشانه سند مرده یا <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Tombstone</span></b>).
</div>

<div style="font-size:16pt; color:#1B263B; margin-bottom:15px; text-align:justify; padding-right:15px;">
در زمان اجرای پرسمان، فرمول خروجی نهایی به شکل منطقی زیر اصلاح می‌گردد تا اسناد حذف‌شده (مجموعه <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">D</span></b>) پیش از نمایش به کاربر، از نتایج فیلتر شوند:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto;">
  $$ \color{#6D2E46}{\boldsymbol{\text{Results}} = (R_{\text{main}} \cup R_{\text{aux}}) - D} $$
</div>

<div style="text-align:center; font-size:17pt; font-weight:bold; color:#6D2E46; margin-top:24px; background-color:#FFFFFF; padding:15px; border-radius:10px; border:1px dashed #6D2E46;">
نتیجه‌گیری ساختاری: این الگو که در دیتابیس‌های پیشرفته نظیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">ElasticSearch</span> و موتورهای جستجوی وب استفاده می‌شود، با انتقال هوشمندانه هزینه از «زمان درج» (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Write-Time</span>) به «زمان ادغام دوره‌ای»، تضمین می‌کند که پیچیدگی زمانی ذخیره و بازیابی اخبار جدید در پایین‌ترین سطح ممکن باقی بماند.
</div>
