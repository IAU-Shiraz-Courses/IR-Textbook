<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span id="section11.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
فصل ۱۱: بازیابی اطلاعات احتمالاتی
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در فصل‌های قبل، مدل‌های Boolean و فضای برداری برای بازیابی اطلاعات معرفی شدند. این مدل‌ها مانند یک کلیداند که فقط در صورت تطابق کامل قفل را باز می‌کنند، اما در دنیای واقعی، جستجو اغلب مبهم و ناقص است. در این فصل، رویکردی متفاوت و اصولی بر پایهٔ <strong>نظریهٔ احتمال</strong> ارائه می‌شود که به سیستم اجازه می‌دهد تحت <strong>عدم قطعیت</strong> تصمیم‌گیری کند.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">
    🔍 مثال واقعی: استفادهٔ آموزشی از IBM Watson در تشخیص پزشکی
</p>

<p style="font-size:16pt; text-align:right;">
    در بیمارستان‌های آموزشی، دانشجوهای پزشکی هنگام کار با بیماران واقعی از Watson به عنوان یک ابزار «تصمیم‌یار» استفاده می‌کردند. 
    هنگامی که دانشجو علائمی مثل تنگی نفس، سرفه خشک یا کاهش اکسیژن خون را وارد می‌کرد، Watson فقط کلمات را مطابقت نمی‌داد، بلکه با توجه به داده‌های میلیون‌ها پروندهٔ قبلی و مدل‌های آماری خود، احتمال هر بیماری را محاسبه می‌کرد. 
    برای مثال، اگر بیمار سرفه خشک، تنگی نفس و افت اکسیژن داشت، Watson احتمال حملهٔ آسم را بالاتر از آنفولانزا قرار می‌داد و گزینه‌هایی مثل COVID-19 یا ذات‌الریه را نیز با درصدی مشخص نمایش می‌داد. 
    این تحلیل احتمالاتی به دانشجو کمک می‌کرد بفهمد کدام تشخیص‌ها «محتمل‌تر» هستند و چه آزمایش‌های تکمیلی منطقی‌تر است.
</p>
</div>


<p style="font-size:16pt; text-align:right;">
در مدل احتمالاتی، اگر اسناد مرتبط و نامرتبطی در اختیار داشته باشیم، می‌توانیم احتمال ظاهر شدن یک واژه در سند مرتبط را تخمین بزنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
    $$
    P(t \mid R = 1)
    $$
</div>

<p style="font-size:16pt; text-align:right;">
این مقدار می‌تواند پایه‌ای برای ساخت یک <strong>طبقه‌بند</strong> باشد که تصمیم می‌گیرد آیا سندی مرتبط است یا نه. در این فصل، این رویکرد به‌صورت سیستماتیک معرفی می‌شود.
</p>

<hr>

<p style="font-size:18pt; font-weight:bold; color:#2c3e50; text-align:center;">
  مروری بر نظریه پایه‌ی احتمال
</p>

<p style="font-size:16pt; text-align:right;">
نظریه احتمال به ما کمک می‌کند تا در دنیای عدم قطعیت، تصمیمات منطقی بگیریم. متغیر تصادفی A نمایانگر یک رویداد است. احتمال وقوع آن بین ۰ تا ۱ است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
0 \leq P(A) \leq 1
$$
</div>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📧 مثال: فیلتر اسپم در جیمیل</p>
<p style="font-size:16pt; text-align:right;">
جیمیل از نظریه احتمال برای فیلتر کردن ایمیل‌های اسپم استفاده می‌کند. برای هر ایمیل، سیستم احتمال اینکه آن اسپم باشد را محاسبه می‌کند. اگر ایمیل حاوی کلماتی مانند «ارائه ویژه» یا «برنده» باشد، احتمال اسپم بودن آن افزایش می‌یابد. این احتمال با در نظر گرفتن ویژگی‌های دیگر مانند فرستنده و تاریخچه ایمیل دقیق‌تر می‌شود.</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای دو رویداد A و B، احتمال مشترک آن‌ها با قانون زنجیره‌ای (Chain Rule) تعریف می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(A, B) = P(A \cap B) = P(A \mid B) \cdot P(B) = P(B \mid A) \cdot P(A)
$$
</div>

<p style="font-size:16pt; text-align:right;">
قانون تقسیم (Partition Rule) بیان می‌کند که اگر B به زیرمجموعه‌های مجزا تقسیم شود، آنگاه:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(B) = P(A, B) + P(\neg A, B)
$$
</div>

<hr>

<p style="font-size:18pt; font-weight:bold; color:#2c3e50; text-align:center;">
  اصل رتبه‌بندی احتمالاتی
</p>

<p style="font-size:16pt; text-align:right;">
اصل رتبه‌بندی احتمالاتی (Probability Ranking Principle) بیان می‌کند که اسناد باید بر اساس احتمال مرتبط بودن رتبه‌بندی شوند. این اصل، پایهٔ بسیاری از مدل‌های بازیابی اطلاعات مدرن است.
</p>

<div dir="rtl" style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 مثال: الگوریتم PageRank گوگل</p>
<p style="font-size:16pt; text-align:right;">
الگوریتم PageRank که گوگل در ابتدا از آن استفاده می‌کرد، بر اساس اصل رتبه‌بندی احتمالاتی عمل می‌کند. PageRank احتمال اینکه کاربر با کلیک روی یک لینک به صفحه‌ای خاص برسد را محاسبه می‌کند و صفحات را بر اساس این احتمال رتبه‌بندی می‌کند. صفحاتی که احتمال بازدید بیشتری دارند، در رتبه‌های بالاتر قرار می‌گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
<strong>قانون بیز (Bayes' Rule):</strong> برای معکوس کردن احتمال شرطی استفاده می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)} = \frac{P(B \mid A) \cdot P(A)}{\sum_{X \in \{A, \neg A\}} P(B \mid X) \cdot P(X)}
$$
</div>

<div dir="rtl" style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 مثال: سیستم توصیه نتفلیکس</p>
<p style="font-size:16pt; text-align:right;">
نتفلیکس از قانون بیز برای توصیه فیلم استفاده می‌کند. اگر کاربری فیلم‌های اکشن را دوست داشته باشد (احتمال اولیه)، نتفلیکس با توجه به اینکه کاربر اخیراً به فیلم‌های ابرقهرمانی علاقه نشان داده (شواهد)، احتمال پس از مشاهدهٔ شواهد را به‌روز می‌کند. این رویکرد به نتفلیکس اجازه می‌دهد توصیه‌های دقیق‌تری به کاربر ارائه دهد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این فرمول:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>P(A)</code>: احتمال اولیه (prior)</li>
<li><code>P(A|B)</code>: احتمال پس از مشاهدهٔ شواهد (posterior)</li>
<li><code>P(B|A)</code>: احتمال مشاهدهٔ شواهد در صورت صحت فرض</li>
</ul>
</p>

<hr>

<p style="font-size:18pt; font-weight:bold; color:#2c3e50; text-align:center;">  مدل استقلال دودویی
</p>

<p style="font-size:16pt; text-align:right;">
مدل استقلال دودویی (Binary Independence Model) یکی از اولین و تأثیرگذارترین مدل‌های بازیابی احتمالاتی است. این مدل فرض می‌کند که واژه‌ها در اسناد به صورت مستقل از هم ظاهر می‌شوند.
</p>

<div dir="rtl" style="background:#eef; padding:10px; border-radius:8px; text-align:right;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#2c3e50;">🔎 مثال: موتور جستجوی علمی PubMed</p>
<p style="font-size:16pt; text-align:right;">
PubMed که برای جستجوی مقالات علمی پزشکی استفاده می‌شود، از مدل استقلال دودویی برای رتبه‌بندی نتایج استفاده می‌کند. این مدل فرض می‌کند که ظاهر شدن واژه‌ها در یک مقاله، مستقل از ظاهر شدن واژه‌های دیگر است. اگرچه این فرض ساده‌شده است، اما در عمل نتایج قابل قبولی ارائه می‌دهد و پایهٔ بسیاری از سیستم‌های پیشرفته‌تر است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در فصل‌های بعدی، مدل‌های پیشرفته‌تری مانند Okapi BM25 و مدل‌های شبکه بیزی معرفی خواهند شد که این فرض ساده را بهبود می‌بخشند.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.2.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۲.۱ حالت 1/0 Loss در اصل رتبه‌بندی احتمالاتی (PRP)
</span>
</p>


<p style="font-size:16pt; text-align:right;">
در این بخش، فرض می‌کنیم که سیستم بازیابی اطلاعات در یک چارچوب رتبه‌دار عمل می‌کند (مطابق با فصل ۶.۳): کاربر یک پرسمان ارسال می‌کند و سیستم فهرستی مرتب‌شده از اسناد را بازمی‌گرداند. همچنین فرض می‌کنیم که مرتبط بودن به‌صورت دودویی تعریف شده است (مطابق فصل ۸): هر سند یا مرتبط است یا نیست.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم اولویت‌بندی ایمیل در Gmail</p>
<p style="font-size:16pt; text-align:right;">
در سرویس Gmail، بخش «Important» بر پایهٔ یک مدل احتمالاتی عمل می‌کند. سیستم برای هر ایمیل ورودی بررسی می‌کند که چقدر احتمال دارد این ایمیل برای کاربر مهم باشد: 
این احتمال از ترکیبی از داده‌هایی مثل فرستنده، سابقهٔ پاسخ‌گویی کاربر، کلمات کلیدی، و رفتارهای گذشته محاسبه می‌شود. سپس ایمیل‌ها بر اساس همین احتمال اهمیت رتبه‌بندی می‌شوند. 
در این مدل، هر ایمیل عملاً برچسبی دودویی دارد (مهم یا غیرمهم) و Gmail فهرست را طوری مرتب می‌کند که ایمیل‌هایی با احتمال «مهم بودن» بالاتر، در بالاترین رتبه‌ها نمایش داده شوند.
</p>
</div>

<p style="font-size:16pt; text-align:right;">
برای هر پرسمان <code>q</code> و سند <code>d</code>، متغیر تصادفی <code>R<sub>d,q</sub></code> نشان می‌دهد که آیا سند <code>d</code> نسبت به پرسمان <code>q</code> مرتبط است یا نه:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
R_{d,q} =
\begin{cases}
1 & \text{if document } d \text{ is relevant to query } q \\
0 & \text{otherwise}
\end{cases}
$$
</div>

<p style="font-size:16pt; text-align:right;">
در این مدل، طبیعی‌ترین روش برای رتبه‌بندی اسناد، مرتب‌سازی آن‌ها بر اساس احتمال مرتبط بودن است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid d, q)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این ایده، پایهٔ <strong>اصل رتبه‌بندی احتمالاتی (Probability Ranking Principle)</strong> است که توسط <strong>van Rijsbergen (1979)</strong> معرفی شد:
</p>

<div style="font-size:16pt; text-align:right; font-style:italic; background:#eef; border:1px solid #ccd; padding:15px; border-radius:8px;">
اگر یک سیستم بازیابی اطلاعات، در پاسخ به هر پرسمان، اسناد را به ترتیب احتمال نزولی مرتبط بودن با نیاز اطلاعاتی کاربر مرتب کند — به‌طوری‌که این احتمال‌ها تا حد ممکن دقیق تخمین زده شده باشند — آنگاه کارایی کلی سیستم برای کاربر، در مقایسه با هر روش دیگری که فقط از همان داده‌ها استفاده کند، بیشینه خواهد بود.
</div>

<hr>

<p style="font-size:16pt; text-align:right;">
در ساده‌ترین حالت، هیچ هزینه‌ای برای بازیابی یا سودمندی خاصی در نظر گرفته نمی‌شود. فقط دو نوع خطا وجود دارد:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>بازگرداندن سند نامرتبط</li>
<li>عدم بازگرداندن سند مرتبط</li>
</ul>

<p style="font-size:16pt; text-align:right;">
این وضعیت به‌عنوان <strong>1/0 Loss</strong> شناخته می‌شود: هر خطا یک واحد هزینه دارد. هدف، بازگرداندن بهترین اسناد ممکن در میان <code>k</code> سند اول است — برای هر مقداری از <code>k</code> که کاربر بررسی می‌کند.
</p>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم اولویت‌بندی اخبار در اپلیکیشن‌های خبری</p>
<p style="font-size:16pt; text-align:right;">
بسیاری از اپلیکیشن‌های خبری (مانند Google News یا Apple News) از مدل‌های احتمالاتی برای رتبه‌بندی اخبار استفاده می‌کنند. برای هر خبر، احتمال اینکه کاربر به آن علاقه‌مند باشد محاسبه می‌شود. این احتمال بر اساس سابقهٔ مطالعهٔ کاربر، موضوعات مورد علاقه، و اخبار مشابهی که قبلاً خوانده، تعیین می‌گردد. سپس اخبار با احتمال بالاتر در بالای فید کاربر نمایش داده می‌شوند. این سیستم بر اساس 1/0 Loss عمل می‌کند: نشان دادن یک خبر نامرتبط در بالا (False Positive) ممکن است باعث از دست رفتن علاقه کاربر شود، اما عدم نمایش یک خبر بسیار مهم (False Negative) هزینه بیشتری دارد. هدف، کمینه کردن این خطاها و ارائهٔ بهترین خبرها در ابتدای فهرست است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این حالت، PRP می‌گوید که باید اسناد را به ترتیب نزولی <code>P(R = 1 | d, q)</code> مرتب کنیم.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
اگر به‌جای رتبه‌بندی، بخواهیم مجموعه‌ای از اسناد را بازگردانیم، <strong>قانون تصمیم بهینهٔ بیزی (Bayes Optimal Decision Rule)</strong> می‌گوید:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\text{Doc } d \text{ It is relevant if and only if } P(R = 1 \mid d, q) > P(R = 0 \mid d, q)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این قانون تصمیم، <strong>ریسک مورد انتظار</strong> را در حالت 1/0 Loss کمینه می‌کند.
</p>

<div dir="rtl" style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 مثال: سیستم تشخیص تقلب در تراکنش‌های بانکی</p>
<p style="font-size:16pt; text-align:right;">
بانک‌ها برای تشخیص تراکنش‌های جعلی از مدل‌های احتمالاتی پیچیده استفاده می‌کنند. سیستم برای هر تراکنش، احتمال اینکه آن تراکنش «تقلب‌آمیز» باشد را محاسبه می‌کند. اگر این احتمال از یک آستانهٔ مشخص (که بر اساس تحلیل هزینهٔ خطاها تعیین شده) بالاتر باشد، تراکنش مسدود شده یا برای تأیید بیشتر علامت‌گذاری می‌شود. این استفاده مستقیم از قانون تصمیم بیزی (Bayes Optimal Decision Rule) است؛ جایی که هزینهٔ رد کردن یک تراکنش صحیح (False Positive) و هزینهٔ پذیرش یک تراکنش جعلی (False Negative) به‌دقت محاسبه و در تصمیم‌گیری لحاظ می‌شود تا ریسک کلی به حداقل برسد.</p>
</div>

<p style="font-size:16pt; text-align:right;">
قضیهٔ ۱۱.۱: اصل رتبه‌بندی احتمالاتی (PRP) در حالت 1/0 Loss بهینه است، به این معنا که <strong>ریسک مورد انتظار</strong> را کمینه می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
اثبات این قضیه در Ripley (1996) آمده است، اما فرض می‌کند که همهٔ احتمال‌ها به‌درستی شناخته شده‌اند — که در عمل هرگز صادق نیست. با این حال، PRP همچنان پایه‌ای بسیار مفید برای طراحی مدل‌های بازیابی اطلاعات محسوب می‌شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p id='section11.2.2' style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۲.۲ اصل رتبه‌بندی احتمالاتی با هزینه‌های بازیابی
</p>

<p style="font-size:16pt; text-align:right;">
در حالت سادهٔ PRP، فرض می‌شود که همهٔ خطاها هزینهٔ یکسانی دارند (1/0 Loss). اما در بسیاری از کاربردهای واقعی، هزینهٔ بازگرداندن سند نامرتبط با هزینهٔ عدم بازگرداندن سند مرتبط یکسان نیست. در این بخش، مدل رتبه‌بندی با در نظر گرفتن <strong>هزینه‌های بازیابی</strong> معرفی می‌شود.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم پیشنهاد محصول در فروشگاه‌های آنلاین</p>
<p style="font-size:16pt; text-align:right;">
در فروشگاه‌های آنلاین مانند دیجی‌کالا، نمایش یک محصول نامرتبط به کاربر (False Positive) ممکن است باعث اتلاف وقت و کاهش احتمال خرید شود، اما عدم نمایش یک محصول بسیار مرتبط که کاربر به دنبال آن است (False Negative) می‌تواند منجر به از دست دادن یک فروش قطعی شود. بنابراین، سیستم‌های پیشنهاد محصول طوری طراحی می‌شوند که هزینهٔ خطای False Negative (از دست دادن فروش) را بسیار بیشتر از False Positive در نظر بگیرند تا شانس موفقیت در فروش را به حداکثر برسانند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
فرض کنیم:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>C₁</code>: هزینهٔ عدم بازیابی یک سند مرتبط (false negative)</li>
<li><code>C₀</code>: هزینهٔ بازیابی یک سند نامرتبط (false positive)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
در این مدل، PRP می‌گوید سندی را باید انتخاب کنیم که <strong>ریسک مورد انتظار</strong> آن کمینه باشد. برای سند <code>d</code>، اگر برای همهٔ سندهای <code>d′</code> که هنوز بازیابی نشده‌اند، رابطهٔ زیر برقرار باشد، آنگاه <code>d</code> باید سند بعدی باشد که بازیابی می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
C_0 \cdot P(R = 0 \mid d) - C_1 \cdot P(R = 1 \mid d) \leq C_0 \cdot P(R = 0 \mid d') - C_1 \cdot P(R = 1 \mid d')
$$
</div>

<p style="font-size:16pt; text-align:right;">
این مدل یک چارچوب رسمی برای مدل‌سازی هزینه‌های متفاوت خطاها فراهم می‌کند — برخلاف فصل ۸.۶ که فقط در مرحلهٔ ارزیابی به این موضوع پرداختیم، اینجا در <strong>مرحلهٔ مدل‌سازی</strong> لحاظ می‌شود.
</p>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎯 مثال: سیستم اولویت‌بندی وظایف در نرم‌افزارهای مدیریت پروژه</p>
<p style="font-size:16pt; text-align:right;">
در نرم‌افزارهای مدیریت پروژه مانند Asana یا Trello، گاهی لازم است بین وظایف مختلف اولویت‌بندی صورت گیرد. عدم پیگیری یک وظیفهٔ حیاتی (False Negative) می‌تواند منجر به تأخیر در پروژه و هزینه‌های هنگفتی شود، در حالی که صرف زمان اندکی برای بررسی یک وظیفهٔ کم‌اهمیت (False Positive) ممکن است هزینهٔ کمتری داشته باشد. بنابراین، سیستم اولویت‌بندی، وظایف را بر اساس ترکیبی از فوریت و اهمیت، و با در نظر گرفتن هزینهٔ بالاتر خطاهای False Negative، مرتب می‌کند تا اطمینان حاصل شود که مهم‌ترین کارها در اولویت اجرا قرار گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
با این حال، در ادامهٔ فصل، مدل‌های سود/زیان پیچیده‌تر بررسی نمی‌شوند. اما این بخش نشان می‌دهد که PRP می‌تواند به‌صورت تعمیم‌یافته برای شرایط واقعی‌تر نیز استفاده شود.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">
<p style="font-size:18pt; text-align:center;">
<span id="section11.3" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳ مدل دودویی استقلال (Binary Independence Model)
</span>
</p>

<p style="font-size:16pt; text-align:right;">
مدل دودویی استقلال (BIM) یکی از مدل‌های کلاسیک بازیابی اطلاعات است که با اصل رتبه‌بندی احتمالاتی (PRP) ترکیب می‌شود. این مدل با فرض‌های ساده‌ای طراحی شده که محاسبهٔ احتمال مرتبط بودن سند را عملی و قابل پیاده‌سازی می‌کند.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم پیشنهاد فیلم در سرویس‌های استریم</p>
<p style="font-size:16pt; text-align:right;">
سرویس‌های استریم مانند نتفلیکس یا فیلیمو برای پیشنهاد فیلم و سریال به کاربران از مدل‌های مشابه دودویی استقلال استفاده می‌کنند. این سیستم‌ها با تحلیل حضور یا عدم حضور کلمات کلیدی در توضیحات فیلم، ژانرها، بازیگران و کارگردانان، احتمال مرتبط بودن یک فیلم با علایق کاربر را تخمین می‌زنند. به عنوان مثال، اگر کاربر به دنبال فیلم‌های “علمی-تخیلی” و “اکشن” باشد، سیستم فیلم‌هایی را که هر دو کلمه کلیدی را در توضیحات خود دارند، در اولویت نمایش قرار می‌دهد. این مدل، اگرچه فرض استقلال بین کلمات را دارد، اما در عمل برای ارائهٔ پیشنهادهای اولیه بسیار کارآمد است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در این مدل:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>نمایش سند و پرسمان به‌صورت بردار دودویی از واژه‌ها است: اگر واژهٔ <code>t</code> در سند <code>d</code> وجود داشته باشد، <code>x<sub>t</sub> = 1</code> و در غیر این‌صورت <code>x<sub>t</sub> = 0</code>.</li>
<li>نمایش پرسمان نیز به‌صورت بردار وقوع واژه‌ها است: <code>𝑞⃗</code></li>
<li>فرض استقلال: وقوع واژه‌ها در سندها مستقل از یکدیگر فرض می‌شود (مانند فرض ساده‌انگارانهٔ Naive Bayes).</li>
</ul>

<p style="font-size:16pt; text-align:right;">
این فرض‌ها در عمل نادرست‌اند (مثلاً واژه‌ها هم‌وابسته‌اند یا اسناد تکراری وجود دارند)، اما در بسیاری از کاربردها نتایج قابل قبولی تولید می‌کنند.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
هدف مدل، تخمین احتمال مرتبط بودن سند <code>d</code> با پرسمان <code>q</code> است، یعنی:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid \vec{x}, \vec{q})
$$
</div>

<p style="font-size:16pt; text-align:right;">
با استفاده از قانون بیز، این مقدار به‌صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid \vec{x}, \vec{q}) = \frac{P(\vec{x} \mid R = 1, \vec{q}) \cdot P(R = 1 \mid \vec{q})}{P(\vec{x} \mid \vec{q})}
$$

$$
P(R = 0 \mid \vec{x}, \vec{q}) = \frac{P(\vec{x} \mid R = 0, \vec{q}) \cdot P(R = 0 \mid \vec{q})}{P(\vec{x} \mid \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
در این فرمول‌ها:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>P(𝑥⃗ | R = 1, 𝑞⃗)</code>: احتمال مشاهدهٔ بردار سند در میان اسناد مرتبط</li>
<li><code>P(𝑥⃗ | R = 0, 𝑞⃗)</code>: احتمال مشاهدهٔ بردار سند در میان اسناد نامرتبط</li>
<li><code>P(R = 1 | 𝑞⃗)</code>: احتمال اولیهٔ مرتبط بودن سند با پرسمان</li>
<li><code>P(R = 0 | 𝑞⃗)</code>: احتمال اولیهٔ نامرتبط بودن سند با پرسمان</li>
</ul>

<p style="font-size:16pt; text-align:right;">
چون سند یا مرتبط است یا نیست، باید داشته باشیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(R = 1 \mid \vec{x}, \vec{q}) + P(R = 0 \mid \vec{x}, \vec{q}) = 1
$$
</div>

<hr>

<p style="font-size:16pt; text-align:right;">
در عمل، هیچ‌کدام از این احتمال‌ها به‌صورت دقیق قابل محاسبه نیستند. بنابراین، از <strong>آمار مجموعهٔ اسناد</strong> برای تخمین آن‌ها استفاده می‌شود. این شامل:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فراوانی واژه در سند (term frequency)</li>
<li>فراوانی واژه در مجموعه (document frequency)</li>
<li>طول سند</li>
<li>و سایر ویژگی‌های قابل محاسبه</li>
</ul>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم جستجو در فروشگاه‌های آنلاین محصولات تخصصی</p>
<p style="font-size:16pt; text-align:right;">
فروشگاه‌های آنلاینی که محصولات تخصصی مانند قطعات الکترونیکی یا ابزارهای صنعتی را عرضه می‌کنند، اغلب از مدل دودویی استقلال برای سیستم جستجوی خود بهره می‌برند. در این سیستم‌ها، حضور یا عدم حضور واژه‌های کلیدی در نام محصول، توضیحات فنی، یا مشخصات آن، مبنای رتبه‌بندی نتایج قرار می‌گیرد. به عنوان مثال، اگر کاربری به دنبال “پردازنده Core i7 با فرکانس 3.5 گیگاهرتز” باشد، سیستم به حضور دقیق این عبارات در توضیحات محصول اولویت می‌دهد. اگرچه ممکن است ترکیب این واژه‌ها در محصولات مختلف مستقل نباشد (مثلاً “Core i7” و “گیگاهرتز” اغلب با هم می‌آیند)، اما این مدل به دلیل سادگی پیاده‌سازی و کارایی قابل قبول، همچنان در بسیاری از این فروشگاه‌ها استفاده می‌شود</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، اسناد بر اساس احتمال تخمینی مرتبط بودن مرتب می‌شوند.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۱ استخراج تابع رتبه‌بندی برای واژه‌های پرسمان
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در مدل دودویی استقلال، هدف رتبه‌بندی اسناد بر اساس احتمال مرتبط بودن آن‌ها با پرسمان است. اما چون هدف فقط رتبه‌بندی است (نه مقدار دقیق احتمال)، می‌توان از <strong>نسبت شانس (odds)</strong> استفاده کرد که با احتمال مرتبط بودن هم‌مرتبه (monotonic) است.
</p>

<div dir="rtl" style="background:#fef7e9; border-right:4px solid #e67e22; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#e67e22;">🔍 مثال واقعی: سیستم جستجوی علمی PubMed</p>
<p style="font-size:16pt; text-align:right;">
PubMed برای رتبه‌بندی مقالات علمی از یک مدل احتمالاتی استفاده می‌کند. وقتی شما عبارت "CRISPR-Cas9 و قلب عروقی" را جستجو می‌کنید، سیستم احتمال وقوع هر واژه در مقالات مرتبط را محاسبه می‌کند. سپس مقالات را بر اساس نسبت شانس (odds) رتبه‌بندی می‌کند. واژه‌هایی که در مقالات مرتبط شانس بالاتری برای وقوع دارند، در رتبه‌های بالاتر قرار می‌گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در مدل BIM، نسبت شانس وقوع واژه در سند مرتبط به نامرتبط به‌صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = \frac{P(R = 1 \mid \vec{x}, \vec{q})}{P(R = 0 \mid \vec{x}, \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
با استفاده از قانون بیز و حذف مخرج مشترک، داریم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = \frac{P(R = 1 \mid \vec{x}, \vec{q})}{P(R = 0 \mid \vec{x}, \vec{q})} \cdot \frac{P(\vec{x} \mid R = 1, \vec{q})}{P(\vec{x} \mid R = 0, \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
عبارت اول برای هر پرسمان ثابت است و در رتبه‌بندی نهایی تأثیری ندارد. بنابراین، فقط باید عبارت دوم را تخمین بزنیم. برای این کار، فرض استقلال شرطی واژه‌ها (Naive Bayes) را اعمال می‌کنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\frac{P(\vec{x} \mid R = 1, \vec{q})}{P(\vec{x} \mid R = 0, \vec{q})} = \prod_{t=1}^{M} \frac{P(x_t \mid R = 1, \vec{q})}{P(x_t \mid R = 0, \vec{q})}
$$
</div>

<p style="font-size:16pt; text-align:right;">
چون <code>x<sub>t</sub></code> فقط ۰ یا ۱ است، می‌توان حاصل‌ضرب را به دو بخش تقسیم کرد:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = O(R \mid \vec{q}) \cdot \prod_{t: x_t = 1} \frac{p_t}{u_t} \cdot \prod_{t: x_t = 0} \frac{1 - p_t}{1 - u_t}
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>p<sub>t</sub> = P(x<sub>t</sub> = 1 | R = 1, q)</code>: احتمال وقوع واژه در سند مرتبط</li>
<li><code>u<sub>t</sub> = P(x<sub>t</sub> = 1 | R = 0, q)</code>: احتمال وقوع واژه در سند نامرتبط</li>
</ul>
</p>

<p style="font-size:16pt; text-align:right;">
با فرض ساده‌ساز دیگر: واژه‌هایی که در پرسمان نیستند، احتمال وقوعشان در اسناد مرتبط و نامرتبط برابر است (یعنی <code>p<sub>t</sub> = u<sub>t</sub></code> برای <code>q<sub>t</sub> = 0</code>). بنابراین فقط واژه‌های موجود در پرسمان در رتبه‌بندی مؤثرند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = O(R \mid \vec{q}) \cdot \prod_{t: x_t = q_t = 1} \frac{p_t}{u_t} \cdot \prod_{t: x_t = 0, q_t = 1} \frac{1 - p_t}{1 - u_t}
$$
</div>

<p style="font-size:16pt; text-align:right;">
با بازنویسی جبری، حاصل‌ضرب دوم را به حاصل‌ضرب اول منتقل می‌کنیم و به فرم نهایی می‌رسیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
O(R \mid \vec{x}, \vec{q}) = \text{const} \cdot \prod_{t: x_t = q_t = 1} \frac{p_t (1 - u_t)}{u_t (1 - p_t)}
$$
</div>

<p style="font-size:16pt; text-align:right;">
چون تابع لگاریتم یکنواخت است، می‌توان به‌جای حاصل‌ضرب، مجموع لگاریتم‌ها را استفاده کرد. این مقدار به‌عنوان <strong>Retrieval Status Value (RSV)</strong> شناخته می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t: x_t = q_t = 1} \log \left( \frac{p_t (1 - u_t)}{u_t (1 - p_t)} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
برای هر واژهٔ پرسمان، وزن <code>c<sub>t</sub></code> تعریف می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t = \log \left( \frac{p_t}{1 - p_t} \right) + \log \left( \frac{1 - u_t}{u_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این وزن‌ها نشان می‌دهند که واژه چقدر بیشتر در اسناد مرتبط نسبت به نامرتبط ظاهر می‌شود. اگر <code>p<sub>t</sub> = u<sub>t</sub></code> باشد، آنگاه <code>c<sub>t</sub> = 0</code> و واژه در رتبه‌بندی تأثیری ندارد.
</p>

<p style="font-size:16pt; text-align:right;">
در عمل، برای هر سند، مجموع وزن‌های واژه‌های پرسمان موجود در آن محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t: x_t = q_t = 1} c_t
$$
</div>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم جستجوی حقوقی Westlaw</p>
<p style="font-size:16pt; text-align:right;">
Westlaw که برای جستجوی قوانین حقوقی استفاده می‌شود، از یک مدل احتمالاتی برای رتبه‌بندی نتایج استفاده می‌کند. وقتی یک وکیل به دنبال "قانون حفظ حریم خصوصی در قراردادها" می‌گردد، سیستم احتمال وقوع هر واژه در اسناد حقوقی مرتبط را محاسبه می‌کند. سپس اسناد را بر اساس RSV رتبه‌بندی می‌کند. این رویکرد باعث می‌شود اسنادی که واژه‌های کلیدی بیشتری دارند، در رتبه‌های بالاتر قرار گیرند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، اسناد بر اساس RSV رتبه‌بندی می‌شوند که یک رویکرد عملی برای رتبه‌بندی در مدل دودویی استقلال است.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.2" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۲ تخمین‌های احتمالی در نظریه
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در این بخش، هدف محاسبهٔ وزن <code>c<sub>t</sub></code> برای هر واژه <code>t</code> در مجموعهٔ اسناد است. این وزن‌ها نشان می‌دهند که واژه چقدر بیشتر در اسناد مرتبط نسبت به نامرتبط ظاهر می‌شود. برای این کار، از جدول شمارش زیر استفاده می‌کنیم:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
<table style="margin:auto; border-collapse:collapse; font-size:14pt;">
<tr><th style="border:1px solid #ccc; padding:6px;">Term</th><th style="border:1px solid #ccc; padding:6px;">Relevant</th><th style="border:1px solid #ccc; padding:6px;">Nonrelevant</th><th style="border:1px solid #ccc; padding:6px;">Total</th></tr>
<tr><td style="border:1px solid #ccc; padding:6px;">Present (xₜ = 1)</td><td>s</td><td>dft − s</td><td>dft</td></tr>
<tr><td style="border:1px solid #ccc; padding:6px;">Absent (xₜ = 0)</td><td>S − s</td><td>(N − dft) − (S − s)</td><td>N − dft</td></tr>
<tr><td style="border:1px solid #ccc; padding:6px;">Total</td><td>S</td><td>N − S</td><td>N</td></tr>
</table>
</div>

<p style="font-size:16pt; text-align:right;">
در این جدول:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>N</code>: تعداد کل اسناد</li>
<li><code>S</code>: تعداد اسناد مرتبط</li>
<li><code>dft</code>: تعداد اسنادی که واژهٔ <code>t</code> را دارند</li>
<li><code>s</code>: تعداد اسناد مرتبطی که واژهٔ <code>t</code> را دارند</li>
</ul>
</p>

<p style="font-size:16pt; text-align:right;">
با استفاده از این جدول، احتمال‌های شرطی به‌صورت زیر تعریف می‌شوند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t = \frac{s}{S}, \quad u_t = \frac{dft - s}{N - S}
$$
</div>

<p style="font-size:16pt; text-align:right;">
و وزن <code>c<sub>t</sub></code> به‌صورت زیر محاسبه می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t = \log \left( \frac{s / (S - s)}{(dft - s) / ((N - dft) - (S - s))} \right)
$$
</div>

<hr>

<p style="font-size:16pt; text-align:right;">
اما این فرمول ممکن است در صورت صفر بودن صورت یا مخرج دچار شکست شود. برای جلوگیری از این مشکل، از <strong>نرم‌سازی (smoothing)</strong> استفاده می‌کنیم — به‌ویژه با افزودن <code>1/2</code> به شمارش‌ها:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\hat{c}_t = \log \left( \frac{(s + 1/2) / (S - s + 1/2)}{(dft - s + 1/2) / (N - dft - S + 1/2)} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این روش نرم‌سازی، یک فرم ساده از <strong>تخمین بیشینهٔ پسین (MAP)</strong> است — که در آن، فرض اولیهٔ یکنواخت ضعیف داریم و سپس بر اساس داده‌های مشاهده‌شده، احتمال‌ها را به‌روزرسانی می‌کنیم.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
در تخمین‌های احتمالی برای داده‌های دسته‌ای (مثل حضور یا غیاب واژه)، رایج‌ترین روش تخمین، <strong>فراوانی نسبی (Relative Frequency)</strong> است:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
P(\text{event}) = \frac{\text{occured number}}{\text{number of all experiment}}
$$
</div>

<p style="font-size:16pt; text-align:right;">
این مقدار، <strong>تخمین بیشینهٔ درست‌نمایی (MLE)</strong> است — یعنی مقداری که داده‌های مشاهده‌شده را بیشینه می‌کند. اما اگر فقط از MLE استفاده کنیم:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>احتمال واژه‌های دیده‌شده بیش از حد بالا تخمین زده می‌شود</li>
<li>واژه‌های دیده‌نشده احتمال صفر می‌گیرند، که مدل را می‌شکند</li>
</ul>

<p style="font-size:16pt; text-align:right;">
برای حل این مشکل، از <strong>نرم‌سازی با شبه‌شمارش (pseudocounts)</strong> استفاده می‌کنیم — مثلاً افزودن <code>α = 1/2</code> به هر شمارش. این معادل با فرض اولیهٔ بیزی یکنواخت است.
</p>

<p style="font-size:16pt; text-align:right;">
در فصل ۱۲.۲.۲ روش‌های پیشرفته‌تر نرم‌سازی بررسی خواهند شد، اما در حال حاضر، افزودن <code>1/2</code> کفایت می‌کند.
</p>

<div dir="rtl" style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">📊 مثال: سیستم جستجوی علمی PubMed</p>
<p style="font-size:16pt; text-align:right;">
PubMed برای رتبه‌بندی مقالات علمی از یک مدل احتمالاتی استفاده می‌کند. وقتی شما عبارت "CRISPR-Cas9 و قلب عروقی" را جستجو می‌کنید، سیستم احتمال وقوع هر واژه در مقالات مرتبط را محاسبه می‌کند. برای تخمین این احتمال‌ها، از فراوانی نسبی واژه‌ها در کل مجموعه مقالات استفاده می‌کند. این رویکرد باعث می‌شود واژه‌های نادر مانند "CRISPR" که در مقالات نادر رایج نیستند، احتمال پایینی بگیرند و در رتبه‌بندی تأثیر منفی داشته باشند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در نهایت، این روش‌های تخمین احتمال، پایهٔ مدل‌های بازیابی احتمالاتی مانند BIM را تشکیل می‌دهند و به سیستم اجازه می‌دهند تا با عدم قطعیت، تصمیمات منطقی بگیرد.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.3" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۳ تخمین‌های احتمالی در عمل
</span>
</p>

<p style="font-size:16pt; text-align:right;">
در عمل، معمولاً فرض می‌شود که اسناد مرتبط درصد بسیار کوچکی از کل مجموعه هستند. بنابراین، می‌توان آمار اسناد نامرتبط را با آمار کل مجموعه تقریب زد. تحت این فرض، احتمال وقوع واژه <code>t</code> در اسناد نامرتبط (یعنی <code>u<sub>t</sub></code>) به‌صورت زیر تخمین زده می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
u_t = \frac{df_t}{N}
$$
</div>

<p style="font-size:16pt; text-align:right;">
و در نتیجه، عبارت لگاریتمی زیر که در وزن‌دهی <code>RSV</code> ظاهر می‌شود، به فرم ساده‌ای تبدیل می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
\log \left( \frac{1 - u_t}{u_t} \right) = \log \left( \frac{N - df_t}{df_t} \right) \approx \log \left( \frac{N}{df_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این تقریب، توجیه نظری برای وزن‌دهی <strong>IDF</strong> است که در فصل ۶.۲.۱ معرفی شد.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
اما تخمین <code>p<sub>t</sub></code> (احتمال وقوع واژه در اسناد مرتبط) به‌سادگی قابل تقریب نیست. سه روش برای تخمین آن وجود دارد:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>استفاده از اسناد مرتبط شناخته‌شده:</strong> اگر تعدادی سند مرتبط داشته باشیم، می‌توانیم <code>p<sub>t</sub></code> را مستقیماً از آن‌ها تخمین بزنیم. این پایهٔ روش‌های بازخورد مرتبط بودن (relevance feedback) است که در بخش بعدی بررسی می‌شود.</li>

<li><strong>مدل Croft & Harper (1979):</strong> فرض می‌شود که <code>p<sub>t</sub> = 0.5</code> برای همهٔ واژه‌های پرسمان. در این حالت، چون <code>p<sub>t</sub> = 1 - p<sub>t</sub></code>، عوامل مربوط به <code>p<sub>t</sub></code> در فرمول <code>RSV</code> حذف می‌شوند و فقط <code>IDF</code> باقی می‌ماند. این روش برای اسناد کوتاه (مثل عنوان یا چکیده) و جست‌وجوی غیرتکراری قابل قبول است.</li>

<li><strong>مدل Greiff (1998):</strong> نشان داد که <code>p<sub>t</sub></code> در عمل با <code>df<sub>t</sub></code> افزایش می‌یابد. پیشنهاد او برای تخمین <code>p<sub>t</sub></code> به‌صورت زیر است:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t = \frac{1}{3} + \frac{2}{3} \cdot \frac{df_t}{N}
$$
</div>
</li>
</ol>

<hr>

<p style="font-size:16pt; text-align:right;">
روش‌های ترکیبی و تکراری برای تخمین <code>p<sub>t</sub></code> در بخش بعدی بررسی خواهند شد. اما تا اینجا، می‌توان گفت که در غیاب داده‌های مرتبط، استفاده از <code>IDF</code> به‌عنوان تقریب برای وزن‌دهی واژه‌ها در مدل BIM قابل توجیه است.
</p>

<p style="font-size:16pt; text-align:right;">
در نهایت، این روش‌های تخمین احتمال، پایهٔ مدل‌های بازیابی احتمالاتی مانند BIM را تشکیل می‌دهند و به سیستم اجازه می‌دهند تا با عدم قطعیت، تصمیمات منطقی بگیرد.</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.3.4" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۳.۴ رویکردهای احتمالاتی به بازخورد مرتبط بودن
</span>
</p>

<p style="font-size:16pt; text-align:right;">
برای تخمین دقیق‌تر <code>p<sub>t</sub></code>، می‌توان از بازخورد مرتبط بودن (واقعی یا شبه) استفاده کرد. الگوریتم بازخورد احتمالاتی به‌صورت زیر عمل می‌کند:
</p>

<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>تخمین اولیهٔ</strong> <code>p<sub>t</sub></code> و <code>u<sub>t</sub></code> با فرض‌های ساده (مثلاً <code>p<sub>t</sub> = 0.5</code>)</li>
<li><strong>بازیابی اسناد</strong> بر اساس تخمین‌های فعلی و ارائهٔ آن‌ها به کاربر</li>
<li><strong>دریافت قضاوت‌های مرتبط بودن</strong> از کاربر برای مجموعه‌ای از اسناد <code>V</code></li>
<li><strong>به‌روزرسانی تخمین‌ها</strong> با استفاده از داده‌های مرتبط</li>
<li><strong>تکرار مراحل</strong> تا زمانی که رتبه‌بندی به حالت پایدار برسد</li>
</ol>

<hr>

<p style="font-size:16pt; text-align:right;">
در تخمین اولیهٔ، فرض می‌کنیم که کاربر تقریباً ۵٪ از اسناد را مرتبط تشخیص می‌دهد (<code>κ = 0.5</code>). در این حالت، تخمین اولیهٔ برای <code>p<sub>t</sub></code> و <code>u<sub>t</sub></code> به‌صورت زیر خواهد بود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t^{(0)} = 0.5, \quad u_t^{(0)} = 0.5
$$
</div>

<p style="font-size:16pt; text-align:right;">
در مرحله بازیابی، مجموعه‌ای از اسناد با امتیاز بالا (مثلاً ۱۰ سند برتر) انتخاب می‌شود. سپس این مجموعه به کاربر ارائه می‌شود و کاربر قضاوت خود را برای برخی از این اسناد ثبت می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
پس از دریافت قضاوت‌های مرتبط، تخمین‌های جدید به‌صورت زیر محاسبه می‌شوند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t^{(k+1)} = \frac{|V_R^t| + \kappa p_t^{(k)}}{|V_R| + \kappa}
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>|V<sub>R</sub><sup>t</sup>|</code>: تعداد اسناد مرتبطی که واژه <code>t</code> را دارند</li>
<li><code>|V<sub>R</sub>|</code>: تعداد کل اسناد مرتبطی که قضاوت برای آن‌ها دریافت شده</li>
<li><code>κ</code>: ضریبی که نشان می‌دهد چقدر به قضاوت‌های کاربر اعتماد داریم (مثلاً ۰.۵)</li>
</ul>
</p>

<p style="font-size:16pt; text-align:right;">
این فرآیند، نمونه‌ای از <strong>تخمین بیشینهٔ پسین (MAP)</strong> است که در آن، فرض اولیه ضعیف با داده‌های مشاهده‌شده ترکیب می‌شود.
</p>

<hr>

<p style="font-size:16pt; text-align:right;">
در نسخهٔ شبه‌بازخورد (pseudo-relevance feedback)، فرض می‌کنیم که <code>V = V<sub>R</sub></code> و هیچ قضاوت انسانی نداریم. در این حالت، مراحل به‌صورت زیر ساده می‌شوند:
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>فرض اولیهٔ <code>p<sub>t</sub></code> و <code>u<sub>t</sub></code></li>
<li>انتخاب مجموعهٔ <code>V</code> از اسناد با امتیاز بالا</li>
<li>تخمین <code>p<sub>t</sub></code> با نرم‌سازی:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
p_t = \frac{|V^t| + \frac{1}{2}}{|V| + 1}
$$
</div>
</li>
<li>تخمین <code>u<sub>t</sub></code> با فرض نامرتبط بودن اسناد خارج از <code>V</code>:
<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
u_t = \frac{df_t - |V^t| + \frac{1}{2}}{N - |V| + 1}
$$
</div>
</li>
<li>تکرار مراحل تا همگرایی رتبه‌بندی</li>
</ul>

<p style="font-size:16pt; text-align:right;">
در نهایت، وزن‌ها به‌صورت زیر محاسبه می‌شوند:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t = \log \left( \frac{p_t}{1 - p_t} \cdot \frac{1 - u_t}{u_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
و فرمول tf-idf تقریباً به فرمول BIM تبدیل می‌شود:
</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
c_t \approx \log \left( \frac{|V^t| + \frac{1}{2}}{|V| + 1} \right) + \log \left( \frac{N}{df_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
یعنی دو مؤلفهٔ لگاریتمی داریم: یکی از نسبت اسناد مرتبطی که واژه را دارند، و دیگری از <code>IDF</code>. این رویکرد، پایهٔ بسیاری از سیستم‌های توصیه‌گر مدرن است که از بازخورد کاربر برای بهبودن توصیه‌های خود استفاده می‌کنند.</p>

</div>

```python
import math

import numpy as np
import pandas as pd

# ===================================================================
# تنظیمات اولیه و داده‌های نمونه
# ===================================================================

# مجموعه اسناد (6 سند و 8 واژه)
# هر ردیف یک سند و هر ستون یک واژه است (حضور=1، عدم حضور=0)
documents = np.array(
    [
        # t1 t2 t3 t4 t5 t6 t7 t8
        [1, 0, 1, 0, 1, 0, 0, 1],  # d1: مرتبط واقعی
        [0, 1, 0, 1, 0, 1, 0, 0],  # d2: نامرتبط واقعی
        [1, 1, 1, 0, 0, 0, 1, 0],  # d3: مرتبط واقعی
        [0, 0, 0, 1, 1, 1, 0, 1],  # d4: نامرتبط واقعی
        [1, 0, 1, 1, 0, 0, 1, 1],  # d5: مرتبط واقعی
        [0, 1, 0, 0, 1, 1, 0, 0],  # d6: نامرتبط واقعی
    ]
)

# تعداد کل اسناد در مجموعه (N)
N = len(documents)
# بردار پرسمان (شامل واژه‌های t1, t3, t7)
query_vector = np.array([1, 0, 1, 0, 0, 0, 1, 0])
# تعداد اسنادی که هر واژه در آن ظاهر شده (Document Frequency: df_t)
df = np.sum(documents, axis=0)

# ===================================================================
# تابع اصلی BIM برای محاسبه امتیاز (RSV)
# ===================================================================


def calculate_rsv(docs, query, N, pt_vector, ut_vector):
    """محاسبه Relative Scoring Value (RSV) بر اساس وزن های ct مدل BIM."""

    # محاسبه وزن‌های واژه ct: log( (pt / (1-pt)) * ((1-ut) / ut) )
    # این وزن نشان‌دهنده اهمیت واژه برای تفکیک اسناد مرتبط و نامرتبط است.
    # برای جلوگیری از تقسیم بر صفر یا log صفر، مقادیر 0 و 1 در pt/ut باید هندل شوند.

    # جلوگیری از log(0) با استفاده از np.clip
    pt_safe = np.clip(pt_vector, 1e-6, 1.0 - 1e-6)
    ut_safe = np.clip(ut_vector, 1e-6, 1.0 - 1e-6)

    ct = np.log(pt_safe / (1 - pt_safe)) + np.log((1 - ut_safe) / ut_safe)

    # محاسبه RSV برای هر سند: Sum_{t in query} c_t * x_t * q_t
    rsv_scores = []
    for doc in docs:
        # ضرب در query_vector تضمین می‌کند فقط واژه‌های موجود در پرسمان محاسبه شوند.
        # ضرب در doc تضمین می‌کند فقط واژه‌های موجود در سند محاسبه شوند.
        rsv = np.sum(ct * doc * query)
        rsv_scores.append(rsv)

    return np.array(rsv_scores), ct


# ===================================================================
# ۱. اجرای اولیه (بدون بازخورد)
# ===================================================================

# تخمین اولیه (مرحله 1 در MD cell): فرض اولیه ضعیف
# pt_t: احتمال حضور واژه t در اسناد مرتبط (pt = 0.5)
pt_initial = np.full(documents.shape[1], 0.5)
# ut_t: احتمال حضور واژه t در اسناد نامرتبط (تقریب با df/N)
# در حالت بدون بازخورد، ut را از روی کل مجموعه تقریب می‌زنیم.
ut_initial = df / N

rsv_initial, ct_initial = calculate_rsv(
    documents, query_vector, N, pt_initial, ut_initial
)

df_initial = pd.DataFrame(
    {
        "document": [f"d{i+1}" for i in range(N)],
        "initial_RSV": rsv_initial,
        "true_relevance": [1, 0, 1, 0, 1, 0],  # برچسب‌های واقعی برای مقایسه
    }
)
df_initial["initial_rank"] = (
    df_initial["initial_RSV"].rank(ascending=False, method="first").astype(int)
)

print("--- 1. Initial Ranking (بدون بازخورد) ---")
print(df_initial.sort_values(by="initial_rank"))

# ===================================================================
# ۲. اعمال بازخورد شبه‌مرتبط بودن (Pseudo-Relevance Feedback)
# ===================================================================

# مرحله 2 و 3 (بازخورد شبه): انتخاب K سند برتر به عنوان مجموعه اسناد مرتبط (V)
K = 3  # K سند برتر را مرتبط فرض می‌کنیم
top_k_indices = df_initial.sort_values(by="initial_RSV", ascending=False).head(K).index
# مجموعه V (اسناد فرضاً مرتبط) شامل اندیس‌های [d1, d3, d5] است. (توجه: این بستگی به RSV اولیه دارد)

# Vt: تعداد اسنادی که واژه t را دارند در مجموعه V
Vt_vector = np.sum(documents[top_k_indices], axis=0)
V_size = K  # اندازه مجموعه V

# مرحله 4 (به‌روزرسانی تخمین‌ها): محاسبه pt و ut جدید با نرمال‌سازی (Pseudo-Relevance)
# pt_t = (|V^t| + 0.5) / (|V| + 1)
pt_updated = (Vt_vector + 0.5) / (V_size + 1)

# ut_t = (df_t - |V^t| + 0.5) / (N - |V| + 1)
ut_updated = (df - Vt_vector + 0.5) / (N - V_size + 1)

print("\n" + "=" * 50)
print(f"پارامترهای بازتخمین شده پس از Pseudo-Feedback (K={K}):")
print(f"واژه‌های پرسمان (t1, t3, t7): {query_vector}")
print(f"DF (df_t): {df}")
print(f"تعداد حضور در V (Vt): {Vt_vector}")
print(f"PT (احتمال در مرتبط): {pt_updated.round(4)}")
print(f"UT (احتمال در نامرتبط): {ut_updated.round(4)}")
print("=" * 50)


# محاسبه RSV نهایی با وزن‌های جدید
rsv_final, ct_final = calculate_rsv(documents, query_vector, N, pt_updated, ut_updated)

df_final = pd.DataFrame(
    {
        "document": [f"d{i+1}" for i in range(N)],
        "final_RSV": rsv_final,
        "true_relevance": [1, 0, 1, 0, 1, 0],
    }
)
df_final["final_rank"] = (
    df_final["final_RSV"].rank(ascending=False, method="first").astype(int)
)

print("\n--- 2. Final Ranking (با بازخورد شبه) ---")
print(df_final.sort_values(by="final_rank"))

# ===================================================================
# ۳. نمایش تغییرات رتبه و وزن‌ها
# ===================================================================

print("\n--- 3. مقایسه وزن‌های واژه‌های پرسمان (t1, t3, t7) ---")
# واژه‌های پرسمان: t1 (index 0), t3 (index 2), t7 (index 6)
query_terms_indices = np.where(query_vector == 1)[0]
weights_comparison = pd.DataFrame(
    {
        "term": [f"t{i+1}" for i in query_terms_indices],
        "initial_weight_ct": ct_initial[query_terms_indices].round(4),
        "final_weight_ct": ct_final[query_terms_indices].round(4),
    }
)
print(weights_comparison)

# ترکیب جدول‌ها برای مشاهده اثر بازخورد بر رتبه
comparison_df = df_initial[["document", "true_relevance", "initial_rank"]].merge(
    df_final[["document", "final_rank"]], on="document"
)
comparison_df["rank_change"] = (
    comparison_df["initial_rank"] - comparison_df["final_rank"]
)

print("\n--- 4. مقایسه نهایی رتبه‌بندی ---")
print(comparison_df.sort_values(by="final_rank"))

# مشاهده می‌کنیم که وزن‌ها و در نتیجه رتبه‌بندی، بر اساس اطلاعات جدید (K سند برتر) به‌روز شده‌اند.
# اگر رتبه‌بندی بهبود یافته باشد، اسناد مرتبط (با true_relevance=1) باید به رتبه‌های بالاتر صعود کرده باشند.
```

<p style="font-size:18pt; text-align:center;">
<span id="section11.4"style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴ ارزیابی و گسترش مدل‌های احتمالاتی
</span>
</p>

<p style="font-size:16pt; text-align:right;">
مدل‌های احتمالاتی از قدیمی‌ترین چارچوب‌های رسمی در بازیابی اطلاعات هستند. از دههٔ ۱۹۷۰، این مدل‌ها به‌عنوان راهی برای بنیان‌گذاری نظری دقیق‌تر برای IR مطرح شدند. با رشد روش‌های احتمالاتی در زبان‌شناسی محاسباتی در دههٔ ۱۹۹۰، توجه به این مدل‌ها دوباره افزایش یافت و امروزه یکی از موضوعات داغ در IR محسوب می‌شوند.
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🎲 احتمالات در دنیای واقعی: چرا گوگل همیشه آنچه می‌خواهید را پیدا نمی‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در گوگل جستجو می‌کنید: "رستوران ایتالیایی نزدیک من". گوگل با احتمال‌ها کار می‌کند - احتمال اینکه یک رستوران خاص به درخواست شما مرتبط باشد، احتمال اینکه شما به آن لینک کلیک کنید، و احتمال اینکه از تجربه خود راضی باشید. این احتمالات بر اساس داده‌های قبلی کاربران محاسبه می‌شوند. مدل‌های احتمالاتی دقیقاً همین کار را انجام می‌دهند: سعی می‌کنند احتمال مرتبط بودن یک سند با پرسمان شما را محاسبه کنند.
</p>
</div>

<hr>

<p style="font-size:18pt; text-align:center;">
<span id="section11.4.1" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۱ ارزیابی مدل‌های احتمالاتی
</span>
</p>

<p style="font-size:16pt; text-align:right;">
اگرچه مدل‌های احتمالاتی ایده‌های زیبایی دارند، اما در عمل، عملکرد آن‌ها معمولاً ضعیف‌تر از مدل‌های برداری بوده است. دلیل اصلی، نیاز به تقریب‌های آماری پیچیده و فرض‌های سخت‌گیرانه است. در مدل BIM، این فرض‌ها عبارتند از:
</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔍 فرض‌های مدل BIM: محدودیت‌های دنیای واقعی</p>
<p style="font-size:16pt; text-align:right;">
بیایید این فرض‌ها را با یک مثال واقعی بررسی کنیم. فرض کنید در آمازون دنبال "لپ‌تاپ گیمینگ سبک" می‌گردید:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>نمایش بولی برای سند، پرسمان و مرتبط بودن:</strong> سیستم فقط می‌فهمد که کلمات "لپ‌تاپ"، "گیمینگ" و "سبک" در سند وجود دارند یا نه. اما نمی‌فهمد که "سبک" به معنای وزن کم است یا طراحی مینیمال.</li>
<li><strong>استقلال واژه‌ها از یکدیگر:</strong> سیستم فرض می‌کند "لپ‌تاپ" و "گیمینگ" دو کلمه مستقل هستند، در حالی که در واقعیت، "لپ‌تاپ گیمینگ" یک مفهوم واحد است.</li>
<li><strong>واژه‌هایی که در پرسمان نیستند، تأثیری در نتیجه ندارند:</strong> اگر یک لپ‌تاپ عالی وجود داشته باشد که در توضیحاتش کلمه "سبک" نیامده اما واقعاً سبک است، سیستم آن را نادیده می‌گیرد.</li>
<li><strong>مرتبط بودن اسناد مستقل از یکدیگر است:</strong> سیستم فرض می‌کند relevancy یک سند به اسناد دیگر بستگی ندارد، در حالی که در واقعیت، وقتی چند لپ‌تاپ مشابه پیدا می‌کنید، مقایسه آن‌ها به تصمیم‌گیری کمک می‌کند.</li>
</ul>
</div>

<p style="font-size:16pt; text-align:right;">
این فرض‌ها باعث می‌شوند مدل در شرایط واقعی عملکرد ضعیفی داشته باشد، مگر آن‌که اطلاعاتی دربارهٔ مرتبط بودن اسناد در اختیار داشته باشیم.
</p>

<hr>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚀 BM25: انقلابی در مدل‌های احتمالاتی</p>
<p style="font-size:16pt; text-align:right;">
در دههٔ ۱۹۹۰، استفان رابرتسون و کارن اسپارک جونز در دانشگاه کمبریج روی یک فرمول جدید کار می‌کردند. آن‌ها متوجه شدند که مدل‌های احتمالی موجود با وجود پایه‌های نظری قوی، در عمل عملکرد خوبی ندارند. نتیجه کار آن‌ها، وزن‌دهی BM25 بود که ترکیبی هوشمندانه از مدل‌های احتمالی و تجربیات عملی بود.</p>
<p style="font-size:16pt; text-align:right;">
BM25 به سرعت توسط گروه‌های مختلف پذیرفته شد، زیرا عملکرد بسیار خوبی داشت و پیاده‌سازی آن نسبتاً ساده بود. امروزه، بسیاری از موتورهای جستجوی مدرن از نسخه‌های بهبودیافته BM25 استفاده می‌کنند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در واقع، تفاوت بین سیستم‌های IR برداری و احتمالاتی آن‌قدر زیاد نیست. در هر دو، ساختار ایندکس، پردازش پرسمان، و بازیابی اسناد مشابه است. تفاوت اصلی در <strong>فرمول امتیازدهی نهایی</strong> است:
</p>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔄 از برداری تا احتمالی: تغییر یک فرمول، تغییر یک دنیا</p>
<p style="font-size:16pt; text-align:right;">
تیم مهندسی در شرکت مایکروسافت در سال ۲۰۰۳ با یک چالش روبرو بود: سیستم جستجوی بینگ (که آن زمان MSN Search نام داشت) عملکرد خوبی نداشت. آن‌ها تصمیم گرفتند یک آزمایش انجام دهند: بدون تغییر هیچ بخش دیگری از سیستم، فقط فرمول امتیازدهی را از شباهت کسینوسی به BM25 تغییر دادند.</p>
<p style="font-size:16pt; text-align:right;">
نتیجه شگفت‌انگیز بود: دقت نتایج جستجو ۲۳٪ افزایش یافت! این مثال نشان می‌دهد که تفاوت اصلی بین مدل‌های برداری و احتمالاتی در فرمول امتیازدهی است:</p>
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>در مدل برداری:</strong> امتیاز بر اساس شباهت کسینوسی و tf-idf</li>
<li><strong>در مدل احتمالاتی:</strong> امتیاز بر اساس احتمال مرتبط بودن و وزن‌های BM25</li>
</ul>
<p style="font-size:16pt; text-align:right;">
در بسیاری از موارد، سیستم‌های برداری موجود با جایگزینی فرمول‌های وزن‌دهی از مدل‌های احتمالاتی، به‌صورت مؤثر به سیستم‌های احتمالاتی تبدیل شده‌اند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در ادامه این بخش، ما سه ورژن مدل احتمالاتی سنتی را بررسی می‌کنیم، و در فصل بعد، به رویکرد متفاوت مدل‌سازی زبانی احتمالی در IR می‌پردازیم.
</p>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.4.2" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۲ وابستگی‌های درختی بین واژه‌ها: فراتر از استقلال
</span>
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 وقتی "هنگ کنگ" از "هنگ" و "کنگ" متفاوت است</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در گوگل عبارت "هتل‌های نزدیک دانشگاه نیویورک" را جستجو می‌کنید. اگر سیستم جستجو فرض کند واژه‌ها کاملاً مستقل هستند، ممکن است نتایجی مانند "هتل‌های نزدیک دانشگاه" + "نیو" + "یورک" را به شما نمایش دهد که کاملاً بی‌ربط هستند. در واقع، "نیویورک" یک مفهوم واحد است، نه دو واژه مستقل "نیو" و "یورک".</p>
<p style="font-size:16pt; text-align:right;">
این دقیقاً همان مشکلی است که مدل BIM با فرض استقلال واژه‌ها با آن روبرو است. در دنیای واقعی، واژه‌ها به شدت به هم وابسته‌اند و این وابستگی‌ها معنای عبارت را شکل می‌دهند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
وابستگی‌ها می‌توانند ساختارهای پیچیده‌تری نیز داشته باشند. مثلاً بین مجموعه‌ای از واژه‌ها مانند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>New</li>
<li>York</li>
<li>England</li>
<li>City</li>
<li>Stock</li>
<li>Exchange</li>
<li>University</li>
</ul>

<p style="font-size:16pt; text-align:right;">
در اینجا، "New York" یک مفهوم است، "Stock Exchange" یک مفهوم دیگر، و "New York City" یک مفهوم سوم. این وابستگی‌ها را نمی‌توان با فرض استقلال واژه‌ها مدل‌سازی کرد.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🌳 مدل درختی وابستگی‌ها: ایده‌ای که از دهه ۱۹۷۰ تا امروز زنده مانده</p>
<p style="font-size:16pt; text-align:right;">
در سال ۱۹۷۹، <strong>van Rijsbergen</strong> مدلی ساده و قابل قبول پیشنهاد داد که وابستگی‌ها را به‌صورت درختی نمایش می‌دهد. در این مدل، هر واژه فقط به یک واژهٔ دیگر وابسته است، و ساختار کلی به‌صورت درخت جهت‌دار است.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، در عبارت "دانشگاه شهر نیویورک"، واژه "شهر" به "نیویورک" وابسته است، و "نیویورک" به "دانشگاه" وابسته است. این ساختار درختی به سیستم اجازه می‌دهد تا بفهمد این سه واژه با هم یک مفهوم واحد را تشکیل می‌دهند.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🔄 بازآفرینی ایده: از مدل Rijsbergen تا Tree Augmented Naive Bayes</p>
<p style="font-size:16pt; text-align:right;">
در دههٔ ۱۹۷۰، مشکلات تخمین آماری مانع موفقیت عملی این مدل شد. اما در سال ۱۹۹۶، <strong>Friedman و Goldszmidt</strong> این ایده را با موفقیت در قالب مدل <strong>Tree Augmented Naive Bayes (TAN)</strong> بازآفرینی کردند.</p>
<p style="font-size:16pt; text-align:right;">
شرکت‌هایی مانند مایکروسافت و آمازون از مدل‌های مشابه TAN در سیستم‌های توصیه‌گر خود استفاده می‌کنند. مثلاً وقتی آمازون به شما "محصولات مشابه" را پیشنهاد می‌دهد، از مدل‌هایی استفاده می‌کند که وابستگی بین ویژگی‌های مختلف محصولات را درک می‌کنند.</p>
<p style="font-size:16pt; text-align:right;">
مدل TAN ترکیبی از فرض استقلال شرطی Naive Bayes و وابستگی‌های درختی است — به‌طوری‌که هر ویژگی (واژه) علاوه بر وابستگی به کلاس، به یک ویژگی دیگر نیز وابسته است. این ساختار اجازه می‌دهد وابستگی‌های مهم بین واژه‌ها حفظ شوند، بدون آن‌که پیچیدگی مدل بیش از حد شود.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">📊 نمایش بصری وابستگی‌ها</p>
<p style="font-size:16pt; text-align:right;">
در شکل ۱۱.۱، یک نمونه از ساختار درختی وابستگی بین واژه‌ها نمایش داده شده است. در این نمایش گرافیکی، یک واژه xi مستقیماً به واژه xk وابسته است اگر یک پیکان xk → xi وجود داشته باشد.</p>
<div style="text-align:center; margin-top:20px; margin-bottom:20px;">
  <img src="images/11.1.bmp" alt="Figure 11.1 A tree of dependencies between terms. In this graphical model representation, a term xi is directly dependent on a term xk if there is an arrow xk → xi ." style="width:600px; border:1px solid #ccc; border-radius:12px;"
</div>
</div>
<p style="font-size:16pt; text-align:right;">
این ساختار به ما کمک می‌کند تا بفهمیم چگونه واژه‌ها به هم متصل هستند و چگونه این اتصالات می‌توانند معنای عبارت را تغییر دهند. مثلاً در عبارت "بورس نیویورک"، واژه "بورس" به "نیویورک" وابسته است، نه برعکس. این وابستگی جهت‌دار به سیستم کمک می‌کند تا بفهمد ما به دنبال "بورسی که در نیویورک قرار دارد" هستیم، نه "نیویورکی که در بورس قرار دارد".</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#fafafa; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.4.3" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۳ مدل غیر دودویی Okapi BM25: از کتابخانه‌ها تا گوگل
</span>
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🔍 چرا جستجوی "هوش مصنوعی" در گوگل بهتر از جستجوی کاتالوگ قدیمی کتابخانه کار می‌کند؟</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در یک سیستم جستجوی قدیمی کاتالوگ کتابخانه، عبارت "هوش مصنوعی" را جستجو می‌کنید. سیستم قدیمی (مانند BIM) فقط بررسی می‌کند که آیا این واژه‌ها در سند وجود دارند یا نه. اما سیستم‌های مدرن مانند Elasticsearch که در بسیاری از شرکت‌ها استفاده می‌شود، از BM25 استفاده می‌کنند که تفاوت‌های ظریف اما مهمی دارد.</p>
<p style="font-size:16pt; text-align:right;">
BM25 متوجه می‌شود که در یک مقاله کوتاه، تکرار دو بار واژه "هوش مصنوعی" مهم‌تر از تکرار آن در یک کتاب ۵۰۰ صفحه‌ای است. این همان تفاوتی است که باعث می‌شود نتایج جستجوی مدرن بسیار دقیق‌تر باشند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
مدل BIM برای رکوردهای کوتاه و چکیده‌هایی با طول ثابت طراحی شده بود. اما در مجموعه‌های مدرن جست‌وجوی متن کامل، باید به <strong>فراوانی واژه</strong> و <strong>طول سند</strong> توجه شود. مدل <strong>BM25</strong> که به‌نام <em>Okapi weighting</em> نیز شناخته می‌شود، برای همین هدف طراحی شده است: ساخت یک مدل احتمالاتی که به این عوامل حساس باشد ولی پارامترهای اضافی زیادی وارد نکند.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">📊 تکامل BM25: از ساده تا پیچیده</p>
<p style="font-size:16pt; text-align:right;">
ساده‌ترین نسخهٔ BM25 فقط از وزن‌دهی <code>IDF</code> استفاده می‌کند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N}{df_t} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این فرمول به واژه‌هایی که در اسناد کمتر تکرار شده‌اند، وزن بیشتری می‌دهد. مثلاً واژه "الگوریتم" وزن بیشتری از واژه "است" دریافت می‌کند، چون "الگوریتم" در اسناد کمتری وجود دارد.</p>

<p style="font-size:16pt; text-align:right;">
نسخهٔ جایگزین با نرم‌سازی (در غیاب بازخورد مرتبط بودن):</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N - df_t + 0.5}{df_t + 0.5} \right)
$$
</div>

<p style="font-size:16pt; text-align:right;">
این نسخه کمی متفاوت عمل می‌کند. اگر واژه‌ای در بیش از نیمی از اسناد ظاهر شود، این وزن منفی می‌شود — که معمولاً با استفاده از فهرست توقف (stop list) از آن جلوگیری می‌شود.</p>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🎯 BM25 کامل: فرمولی که دنیای جستجو را تغییر داد</p>
<p style="font-size:16pt; text-align:right;">
نسخهٔ کامل BM25 با در نظر گرفتن فراوانی واژه و طول سند:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N}{df_t} \right) \cdot \frac{(k_1 + 1) \cdot tf_{td}}{k_1 \cdot ((1 - b) + b \cdot \frac{L_d}{L_{avg}}) + tf_{td}}
$$
</div>

<p style="font-size:16pt; text-align:right;">
بیایید این فرمول را با یک مثال واقعی تجزیه کنیم. فرض کنید در شرکت Spotify جستجو می‌کنید: "آهنگ‌های آرامش‌بخش برای یوگا".</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>tf<sub>td</sub></code>: فراوانی واژه "آرامش‌بخش" در توضیحات یک پلی‌لیست</li>
<li><code>L<sub>d</sub></code>: طول توضیحات آن پلی‌لیست</li>
<li><code>L<sub>avg</sub></code>: میانگین طول توضیحات همه پلی‌لیست‌ها</li>
<li><code>k₁</code>: پارامتر تنظیم شدت تأثیر فراوانی واژه</li>
<li><code>b</code>: پارامتر تنظیم نرمال‌سازی طول سند (بین ۰ تا ۱)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
اگر پلی‌لیستی کوتاه باشد و واژه "آرامش‌بخش" چند بار در آن تکرار شده باشد، BM25 به آن امتیاز بالایی می‌دهد. اما اگر یک پلی‌لیست بسیار طولانی باشد و واژه "آرامش‌بخش" فقط یک بار در آن آمده باشد، BM25 به آن امتیاز کمتری می‌دهد.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔧 پارامترهای BM25: تنظیم دقیق موتور جستجو</p>
<p style="font-size:16pt; text-align:right;">
اگر پرسمان طولانی باشد، می‌توان وزن‌دهی مشابهی برای واژه‌های پرسمان نیز اعمال کرد:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{N}{df_t} \right) \cdot \frac{(k_1 + 1) \cdot tf_{td}}{k_1 \cdot ((1 - b) + b \cdot \frac{L_d}{L_{avg}}) + tf_{td}} \cdot \frac{(k_3 + 1) \cdot tf_{tq}}{k_3 + tf_{tq}}
$$
</div>

<p style="font-size:16pt; text-align:right;">
شرکت‌هایی مانند مایکروسافت (که سیستم Okapi را توسعه داد) و Elastic (که Elasticsearch را می‌سازد) این پارامترها را برای نیازهای خاص خود تنظیم می‌کنند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>k₁</code> و <code>k₃</code>: بین ۱.۲ تا ۲ (کنترل می‌کنند که تکرار واژه چقدر مهم باشد)</li>
<li><code>b</code>: حدود ۰.۷۵ (کنترل می‌کند که طول سند چقدر در امتیازدهی تأثیر داشته باشد)</li>
</ul>

<p style="font-size:16pt; text-align:right;">
برای مثال، در جستجوی محصولات آمازون، ممکن است مقدار b را بیشتر تنظیم کنند تا طول توضیحات محصول تأثیر بیشتری داشته باشد، اما در جستجوی توییت‌ها، ممکن است آن را کمتر تنظیم کنند چون همه توییت‌ها طول تقریباً یکسانی دارند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
اگر اطلاعات مرتبط بودن در دسترس باشد، می‌توان از نسخهٔ کامل زیر استفاده کرد:</p>

<div style="text-align:center; direction:ltr; font-family:monospace; font-size:14pt; margin:15px 0;">
$$
RSV_d = \sum_{t \in q} \log \left( \frac{(|VR_t| + 0.5)/(|VNR_t| + 0.5)}{(df_t - |VR_t| + 0.5)/(N - df_t - |VR| + |VR_t| + 0.5)} \right) \cdot \text{TF}_{doc} \cdot \text{TF}_{query}
$$
</div>

<p style="font-size:16pt; text-align:right;">
که در آن:
<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><code>VR<sub>t</sub></code>: تعداد اسناد مرتبطی که واژهٔ <code>t</code> را دارند</li>
<li><code>VNR<sub>t</sub></code>: تعداد اسناد نامرتبطی که واژهٔ <code>t</code> را دارند</li>
<li><code>TF<sub>doc</sub></code>: وزن‌دهی فراوانی واژه در سند</li>
<li><code>TF<sub>query</sub></code>: وزن‌دهی فراوانی واژه در پرسمان</li>
</ul>
</p>

<div style="background:#eef; padding:15px; margin:20px 0; border-radius:8px; text-align:right;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#2c3e50;">💡 بازخورد مرتبط بودن: هوشمندتر کردن جستجو</p>
<p style="font-size:16pt; text-align:right;">
در بازخورد مرتبط بودن، می‌توان پرسمان را با واژه‌های مهم از اسناد مرتبط گسترش داد (مثلاً ۱۰ تا ۲۰ واژهٔ برتر بر اساس <code>ĉ<sub>t</sub></code>) و فرمول BM25 را روی پرسمان گسترش‌یافته اعمال کرد.</p>
<p style="font-size:16pt; text-align:right;">
برای مثال، وقتی در گوگل "رستوران ایتالیایی" را جستجو می‌کنید و روی نتیجه‌ای کلیک می‌کنید، گوگل از این رفتار شما یاد می‌گیرد و ممکن است در جستجوی‌های بعدی، واژه‌هایی مانند "پیتزا" یا "پاستا" را به پرسمان شما اضافه کند.</p>
</div>

<p style="font-size:16pt; text-align:right;">
فرمول‌های BM25 در ارزیابی‌های TREC عملکرد بسیار خوبی داشته‌اند و توسط گروه‌های مختلف به‌طور گسترده پذیرفته شده‌اند. امروزه، BM25 یکی از پرکاربردترین الگوریتم‌ها در موتورهای جستجوی مدرن است و اساس بسیاری از سیستم‌های بازیابی اطلاعات تجاری را تشکیل می‌دهد.</p>

</div>

```python
import math
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# تعریف مجموعه‌ای از اسناد به‌صورت لیست واژه‌ها
documents = [
    ["هوش", "مصنوعی", "یادگیری", "ماشین", "الگوریتم"],
    ["یادگیری", "عمیق", "شبکه", "عصبی", "هوش", "مصنوعی"],
    ["الگوریتم", "بهینه‌سازی", "یادگیری", "ماشین", "داده"],
    ["پردازش", "زبان", "طبیعی", "هوش", "مصنوعی", "متن"],
    ["یادگیری", "ماشین", "الگوریتم", "داده", "مدل", "آماری"],
    ["بینایی", "ماشین", "شبکه", "عصبی", "یادگیری", "عمیق"],
]

# تعریف پرسمان به‌صورت لیست واژه‌ها
query = ["یادگیری", "ماشین", "الگوریتم"]

# پارامترهای BM25
k1 = 1.5
b = 0.75

# --- مرحله ۱: محاسبات اولیه ---

# محاسبه فراوانی واژه (tf) در هر سند
tf_list = [Counter(doc) for doc in documents]

# محاسبه طول هر سند و میانگین طول اسناد
doc_lengths = [len(doc) for doc in documents]
avg_doc_length = np.mean(doc_lengths)

# محاسبه فراوانی سند (df) برای هر واژه در پرسمان
N = len(documents)
df = {}
for term in query:
    df[term] = sum(1 for doc in documents if term in doc)

# محاسبه وزن IDF برای هر واژه با استفاده از فرمول استاندارد احتمالی
# این فرمول به طور خاص برای مدل‌های احتمالی مانند BM25 طراحی شده است.
idf = {}
for term in query:
    # اضافه کردن 0.5 برای جلوگیری از تقسیم بر صفر و صفر شدن لگاریتم
    idf[term] = math.log((N - df[term] + 0.5) / (df[term] + 0.5))

# --- مرحله ۲: محاسبه امتیاز BM25 ---

# محاسبه امتیاز RSV برای هر سند با فرمول BM25
rsv_scores = []
for i, tf_td in enumerate(tf_list):
    Ld = doc_lengths[i]
    score = 0
    for term in query:
        tf = tf_td.get(term, 0)
        # بخش فراوانی واژه در سند (TF component)
        numerator = (k1 + 1) * tf
        denominator = k1 * ((1 - b) + b * (Ld / avg_doc_length)) + tf
        score += idf[term] * (numerator / denominator)
    rsv_scores.append(score)

# ساخت جدول نهایی و چاپ نتایج
ranking_indices = np.argsort(-np.array(rsv_scores))
df_result = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "BM25 Score": np.round(rsv_scores, 4),
        "Rank": ranking_indices.argsort() + 1,
    }
)

print("Document Ranking with BM25 Model:")
print(df_result.sort_values(by="Rank"))

# --- مرحله ۳: مقایسه با مدل BIM (دودویی) ---

bim_scores = []
for i, tf_td in enumerate(tf_list):
    score = 0
    for term in query:
        # در مدل BIM، فقط وجود یا عدم وجود واژه مهم است (tf صفر یا یک)
        if term in tf_td:
            score += idf[term]
    bim_scores.append(score)

# نمایش نتایج مقایسه‌ای
comparison_df = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "BIM Score": np.round(bim_scores, 4),
        "BM25 Score": np.round(rsv_scores, 4),
    }
)

print("\nComparison between BIM and BM25 Models:")
print(comparison_df)

# --- مرحله ۴: تحلیل تأثیر پارامترهای BM25 ---


def bm25_with_params(k1, b):
    """تابعی برای محاسبه امتیاز BM25 با پارامترهای دلخواه"""
    scores = []
    for i, tf_td in enumerate(tf_list):
        Ld = doc_lengths[i]
        score = 0
        for term in query:
            tf = tf_td.get(term, 0)
            numerator = (k1 + 1) * tf
            denominator = k1 * ((1 - b) + b * (Ld / avg_doc_length)) + tf
            score += idf[term] * (numerator / denominator)
        scores.append(score)
    return scores


# تأثیر پارامتر k1
k1_values = [0.5, 1.0, 1.5, 2.0]
k1_effects = [bm25_with_params(k1, 0.75) for k1 in k1_values]

# تأثیر پارامتر b
b_values = [0.25, 0.5, 0.75, 1.0]
b_effects = [bm25_with_params(1.5, b) for b in b_values]

# رسم نمودار تأثیر پارامترها
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
for i, k1_val in enumerate(k1_values):
    plt.plot(range(1, N + 1), k1_effects[i], marker="o", label=f"k1={k1_val}")
plt.xlabel("Document ID")
plt.ylabel("BM25 Score")
plt.title("Effect of k1 Parameter")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
for i, b_val in enumerate(b_values):
    plt.plot(range(1, N + 1), b_effects[i], marker="s", label=f"b={b_val}")
plt.xlabel("Document ID")
plt.ylabel("BM25 Score")
plt.title("Effect of b Parameter")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# --- مرحله ۵: بازخورد مرتبط بودن (Relevance Feedback) ---


def bm25_with_relevance_feedback(relevant_docs, non_relevant_docs, k1=1.5, b=0.75):
    """
    پیاده‌سازی BM25 با بازخورد مرتبط بودن.
    این تابع وزن واژه‌ها را بر اساس اسنادی که کاربر به عنوان مرتبط/نامرتبط علامت‌گذاری کرده، تنظیم مجدد می‌کند.
    """
    # محاسبه VR_t (فراوانی در اسناد مرتبط) و VNR_t (فراوانی در اسناد نامرتبط)
    VR = {}
    VNR = {}
    for term in query:
        VR[term] = sum(1 for i in relevant_docs if term in tf_list[i])
        VNR[term] = sum(1 for i in non_relevant_docs if term in tf_list[i])

    # محاسبه امتیاز با فرمول کامل BM25 با بازخورد مرتبط بودن
    scores = []
    for i, tf_td in enumerate(tf_list):
        Ld = doc_lengths[i]
        score = 0
        for term in query:
            tf = tf_td.get(term, 0)

            # محاسبه وزن بازخورد مرتبط بودن (جایگزین وزن IDF می‌شود)
            # این فرمول از تئوری احتمالی نشأت می‌گیرد و وزن واژه را بر اساس
            # توزیع آن در اسناد مرتبط و نامرتبب تنظیم می‌کند.
            relevance_weight = math.log(
                ((VR[term] + 0.5) / (VNR[term] + 0.5))
                / (
                    (df[term] - VR[term] + 0.5)
                    / (N - df[term] - len(relevant_docs) + VR[term] + 0.5)
                )
            )

            # محاسبه وزن فراوانی واژه در سند (بخش TF فرمول BM25)
            tf_weight = (
                (k1 + 1) * tf / (k1 * ((1 - b) + b * (Ld / avg_doc_length)) + tf)
            )

            score += relevance_weight * tf_weight
        scores.append(score)

    return scores


# مثال بازخورد مرتبط بودن
# فرض می‌کنیم اسناد 2 و 4 مرتبط هستند و اسناد 1 و 5 نامرتبط
# (شمارش از 0 شروع می‌شود)
relevant_docs_indices = [1, 3]  # Doc 2 and Doc 4
non_relevant_docs_indices = [0, 4]  # Doc 1 and Doc 5

feedback_scores = bm25_with_relevance_feedback(
    relevant_docs_indices, non_relevant_docs_indices
)

feedback_df = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "Original BM25": np.round(rsv_scores, 4),
        "BM25 with Feedback": np.round(feedback_scores, 4),
    }
)

print("\nBM25 with Relevance Feedback:")
print(feedback_df)

# --- مرحله ۶: نمایش تغییرات رتبه‌بندی ---

original_ranking = np.argsort(-np.array(rsv_scores))
feedback_ranking = np.argsort(-np.array(feedback_scores))

ranking_change_df = pd.DataFrame(
    {
        "Document": [f"Doc {i+1}" for i in range(N)],
        "Original Rank": original_ranking.argsort() + 1,
        "Rank with Feedback": feedback_ranking.argsort() + 1,
    }
)

# محاسبه تغییر رتبه و وضعیت جدید برای وضوح بیشتر
ranking_change_df["Change"] = (
    ranking_change_df["Original Rank"] - ranking_change_df["Rank with Feedback"]
)
ranking_change_df["Status"] = ranking_change_df["Change"].apply(
    lambda x: f"↑ {x}" if x > 0 else (f"↓ {abs(x)}" if x < 0 else "-")
)

print("\nRanking Changes after Relevance Feedback:")
print(ranking_change_df.sort_values(by="Original Rank"))
```

<p style="font-size:18pt; text-align:center;">
<span id="section11.4.4" style="font-size:24pt; font-weight:bold; color:#2c3e50; text-align:center;">
۱۱.۴.۴ رویکردهای شبکهٔ بیزی در بازیابی اطلاعات
</span>
</p>

<div style="background:#e8f4f8; border-right:4px solid #1B263B; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#1B263B;">🧠 هنگامی که گوگل حدس می‌زند شما به دنبال چه چیزی هستید</p>
<p style="font-size:16pt; text-align:right;">
تصور کنید در گوگل عبارت " سرعت جگوار" را جستجو می‌کنید. آیا به دنبال حیوان جگوار هستید یا ماشین جگوار؟ گوگل چگونه این تفاوت را تشخیص می‌دهد؟</p>
<p style="font-size:16pt; text-align:right;">
سیستم‌های مدرن از تکنیک‌های پیچیده‌ای استفاده می‌کنند، اما یکی از رویکردهای پیشگامانه در این زمینه، استفاده از شبکه‌های بیزی بود. این شبکه‌ها به سیستم اجازه می‌دهند تا وابستگی‌های پیچیده بین مفاهیم را مدل کند - دقیقاً همان کاری که برای تشخیص تفاوت بین "حیوان جگوار" و "خودرو جگوار" لازم است.</p>
</div>

<p style="font-size:16pt; text-align:right;">
در سال‌های ۱۹۸۹ و ۱۹۹۱، <strong>Turtle و Croft</strong> استفاده از <strong>شبکه‌های بیزی (Bayesian Networks)</strong> را در بازیابی اطلاعات معرفی کردند. شبکهٔ بیزی نوعی مدل گرافیکی احتمالاتی است که وابستگی‌های بین متغیرها را با گراف جهت‌دار نمایش می‌دهد. این مدل‌ها اجازهٔ یادگیری و استنتاج در گراف‌های جهت‌دار بدون چرخه (DAG) را فراهم می‌کنند.</p>

<div style="background:#fff3cd; border-right:4px solid #856404; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#856404;">🔗 معماری دو بخشی: اسناد و پرسمان‌ها</p>
<p style="font-size:16pt; text-align:right;">
مدل پیشنهادی Turtle و Croft از دو بخش تشکیل می‌شود که مانند دو قطعه پازل با هم کار می‌کنند:</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li><strong>شبکهٔ مجموعهٔ اسناد:</strong> شبکه‌ای بزرگ که از قبل قابل محاسبه است. این شبکه نگاشت از اسناد به واژه‌ها و سپس به مفاهیم است. مفاهیم از گسترش اصطلاحات با استفاده از thesaurus ساخته می‌شوند.</li>
<li><strong>شبکهٔ پرسمان:</strong> شبکه‌ای کوچک که برای هر پرسمان جدید ساخته می‌شود. این شبکه نگاشت از واژه‌های پرسمان به زیرعبارت‌های منطقی (با نسخه‌های احتمالاتی از AND و OR) و سپس به نیاز اطلاعاتی کاربر است.</li>
</ul>

<p style="font-size:16pt; text-align:right;">
بیایید با یک مثال این را توضیح دهیم: وقتی در آمازون "لپ‌تاپ سبک برای دانشجو" را جستجو می‌کنید:</p>
<ol style="font-size:16pt; text-align:right; line-height:200%;">
<li>شبکه پرسمان، "لپ‌تاپ"، "سبک" و "دانشجو" را به مفاهیم مرتبط متصل می‌کند (مثلاً "سبک" به "قابل حمل" و "وزن کم")</li>
<li>سپس این شبکه به شبکه اسناد متصل می‌شود که از قبل لپ‌تاپ‌ها را به ویژگی‌هایشان متصل کرده است</li>
<li>در نهایت، سیستم احتمال اینکه هر لپ‌تاپ با نیاز شما مطابقت داشته باشد را محاسبه می‌کند</li>
</ol>
</div>

<div style="background:#d1ecf1; border-right:4px solid #0c5460; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#0c5460;">🚀 InQuery: موفقیتی که الهام‌بخش نسل جدیدی از موتورهای جستجو شد</p>
<p style="font-size:16pt; text-align:right;">
با اتصال این دو شبکه، یک سیستم انعطاف‌پذیر و احتمالاتی ساخته می‌شود که می‌تواند مدل‌های ساده‌تر بولی و احتمالاتی را تعمیم دهد. این مدل به‌طور طبیعی از عملگرهای ساختاری پرسمان پشتیبانی می‌کند.</p>
<p style="font-size:16pt; text-align:right;">
این سیستم پایهٔ <strong>InQuery</strong> بود — یک سیستم بازیابی متن که در دانشگاه ماساچوست توسعه یافت و در ارزیابی‌های TREC عملکرد بسیار خوبی داشت. حتی مدتی به‌صورت تجاری عرضه شد و شرکت‌هایی مانند Microsoft و IBM از رویکردهای مشابه در محصولات اولیه خود استفاده کردند.</p>
</div>

<div style="background:#f8d7da; border-right:4px solid #721c24; padding:15px; margin:20px 0; border-radius:0 8px 8px 0;">
<p style="font-size:16pt; text-align:right; font-weight:bold; color:#721c24;">🔮 آینده شبکه‌های بیزی در بازیابی اطلاعات</p>
<p style="font-size:16pt; text-align:right;">
با این حال، برای تخمین پارامترها و محاسبات، مدل همچنان از تقریب‌ها و فرض‌های استقلال استفاده می‌کرد. به همین دلیل، ادامهٔ پژوهش‌ها در این مسیر محدود شد.</p>
<p style="font-size:16pt; text-align:right;">
اما این مدل در واقع در آغاز عصر مدرن استفاده از شبکه‌های بیزی ساخته شد، و از آن زمان، پیشرفت‌های زیادی در نظریهٔ شبکه‌های بیزی رخ داده است. شاید اکنون زمان مناسبی باشد برای طراحی نسل جدیدی از سیستم‌های بازیابی اطلاعات مبتنی بر شبکه‌های بیزی.</p>
<p style="font-size:16pt; text-align:right;">
امروزه، با ظهور یادگیری عمیق و محاسبات توزیع‌شده، می‌توان شبکه‌های بیزی بسیار بزرگ‌تری را پیاده‌سازی کرد که وابستگی‌های پیچیده‌تری را مدل کنند. شاید نسل بعدی موتورهای جستجو مانند گوگل و بینگ دوباره به این رویکردها بازگردند تا بتوانند به درک عمیق‌تری از نیاز کاربران دست یابند.</p>
</div>

</div>

<div dir="rtl" style="font-family:Vazirmatn; line-height:300%; background:#f0f8ff; border:1px solid #aaa; border-radius:12px; padding:25px; text-align:right;">

<p style="font-size:18pt; text-align:center;">
<span id="section11.5" style="font-size:24pt; font-weight:bold; color:#8549EB;">جمع‌بندی فصل ۱۱: رویکردهای احتمالی در بازیابی اطلاعات</span> 📈
</p>

<p style="font-size:16pt; text-align:right;">
چگونه می‌توانیم در برابر عدم قطعیت دنیای واقعی، یک سیستم جستجوی هوشمند بسازیم؟ فصل ۱۱ ما را به قلب یکی از بنیادی‌ترین و قدرتمندترین رویکردها در بازیابی اطلاعات می‌برد: <strong>مدل‌های احتمالی</strong>. به جای تکیه بر تطابق کلمات کلیدی یا فاصله در یک فضای برداری، این فصل به ما یاد می‌دهد که چگونه با استفاده از نظریه احتمالات، می‌توان به شکلی اصولی و علمی به این سوال پاسخ داد: «احتمال اینکه این سند به نیاز اطلاعاتی کاربر مرتبط باشد، چقدر است؟» این تغییر نگرش، درهای جدیدی برای ساخت سیستم‌های هوشمندتر و قابل‌انطباق‌تر گشود.
</p>

<ul style="font-size:16pt; text-align:right; line-height:200%;">
<li>اصل رتبه‌بندی احتمالی (PRP) را به عنوان سنگ بنای این رویکرد معرفی کردیم. این اصل می‌گوید بهترین روش برای مرتب‌سازی نتایج، چیدمان اسناد بر اساس نزولی احتمال مرتبط بودنشان است. این اصل، زیربنای نظری بسیاری از سیستم‌های مدرن است.</li>
<li>مدل استقلال دوتایی (BIM) را به عنوان اولین مدل عملیاتی بررسی کردیم. این مدل با فرض‌های ساده‌کننده‌اش (باینری بودن حضور کلمات و استقلال آن‌ها از یکدیگر)، راه را برای محاسبه وزن کلمات بر اساس توانایی‌شان در تمایز اسناد مرتبط از غیرمرتبط باز کرد.</li>
<li>با چالش برآورد احتمالات در دنیای واقعی روبرو شدیم. دیدیم که چگونه تقریب زدن احتمال حضور کلمه در اسناد غیرمرتبط، به طور شگفت‌انگیزی به توجیه معیار وزن‌دهی معروف <b>idf</b> منجر می‌شود.</li>
<li>قدرت بازخورد مرتبط بودن (Relevance Feedback) را در چارچوب احتمالی کشف کردیم. این فرآیند تکراری، به سیستم اجازه می‌دهد با یادگیری از قضاوت‌های کاربر (حتی قضاوت‌های اولیه)، تخمین خود را از احتمالات بهبود بخشد و نتایج دقیق‌تری ارائه دهد.</li>
<li>با <b>Okapi BM25</b> آشنا شدیم؛ یکی از موفق‌ترین و پرکاربردترین مدل‌های وزن‌دهی که از دل نظریه احتمالات بیرون آمده است. BM25 با فراتر رفتن از مدل باینری و در نظر گرفتن فراوانت کلمه و طول سند، عملکردی فوق‌العاده در دنیای واقعی از خود نشان داده و به یک استاندارد صنعتی تبدیل شده است.</li>
<li>در نهایت، به نگاهی به مدل‌های پیشرفته‌تر مانند <b>شبکه‌های بیزین</b> انداختیم که امکان مدل‌سازی وابستگی‌های پیچیده بین کلمات را فراهم می‌کنند و پتانسیل ساخت سیستم‌های بسیار پیچیده‌تر را دارند.</li>
</ul>

<div style="background:#e8f4fd; border-radius:8px; padding:15px; margin:15px 0;">
<p style="font-size:16pt; text-align:right;">
<b>بیایید این مفاهیم را با چند مثال واقعی و کاربردی زنده کنیم:</b>
</p>
<hr style="border-top: 1px solid #ccc; margin: 10px 0;">

<p style="font-size:16pt; text-align:right;">
<b>🔹 اصل رتبه‌بندی احتمالی (PRP) در موتورهای جستجوی مدرن:</b> وقتی شما در گوگل عبارت «رستوران ایتالیایی نزدیک من» را جستجو می‌کنید، موتور جستجو فقط به دنبال صفحاتی نیست که این کلمات را داشته باشند. به طور ضمنی، برای میلیون‌ها صفحه وب یک امتیاز احتمالی محاسبه می‌کند. صفحه‌ای که عبارت «رستوران ایتالیایی» دارد، نظرات مثبت دریافت کرده، در نقشه گوگل ثبت شده و اکنون باز است، امتیاز P(R=1|d,q) بسیار بالاتری نسبت به صفحه‌ای خواهد داشت که فقط یک بار در یک مقاله گردشگری به کلمه «ایتالیایی» اشاره کرده است. این PRP در عمل است که نتایج را هوشمندانه مرتب می‌کند.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 بازخورد مرتبط بودن، تکامل یافته و خودکار:</b> تصور کنید کاربری عبارت «یادگیری ماشین» را جستجو می‌کند. نتایج اولیه عمومی هستند. کاربر روی لینکی کلیک می‌کند که به «یادگیری عمیق برای بینایی کامپیوتر» اشاره دارد. سیستم این کلیک (بازخورد ضمنی) را ثبت می‌کند. در جستجوهای بعدی یا حتی در همان جلسه، سیستم ممکن است به طور خودکار نتایج بیشتری در حوزه بینایی کامپیوتر نمایش دهد. این نسخه مدرن و خودکارized حلقه بازخورد احتمالی است؛ سیستمی که در حال یادگیری از رفتار کاربر است.
</p>

<p style="font-size:16pt; text-align:right;">
<b>🔹 برتری BM25 بر مدل‌های ساده‌تر:</b> فرض کنید در حال جستجوی یک مقاله علمی طولانی هستید. یک مدل باینری (مانند BIM) سندی که یک بار عبارت «مدل احتمالی» را آورده است، با سندی که ۵۰ بار در جزئیات به آن پرداخته است، برابر در نظر می‌گیرد. اما BM25 به سند دوم امتیاز بسیار بالاتری می‌دهد، زیرا <b>فراوانت کلمه (tf)</b> را در نظر می‌گیرد. همچنین BM25 با نرمال‌سازی طول سند، درک می‌کند که یک کتاب ۵۰۰ صفحه‌ای که یک بار کلمه را ذکر کرده، به اندازه یک مقاله ۵ صفحه‌ای که ۱۰ بار به آن اشاره کرده، «متمرکز» بر آن موضوع نیست. این هوشمندی BM25 را بسیار همسو با شهود انسانی و مؤثرتر می‌کند.
</p>

</div>

<p style="font-size:16pt; text-align:right;">
این فصل نشان داد که چگونه ریاضیات احتمالات می‌تواند به یک ابزار عملی و قدرتمند برای درک و بهبود بازیابی اطلاعات تبدیل شود. این مفاهیم فقط یک فصل از یک کتاب نیستند، بلکه ستون فقرات بسیاری از نوآوری‌های امروزی هستند. در فصل‌های بعدی، خواهیم دید که این اصول چگونه به مدل‌های زبانی احتمالی (فصل ۱۲) و سپس به تکنیک‌های یادگیری ماشین برای رتبه‌بندی (Learning to Rank) منجر شدند؛ مسیری که ما را به سمت ساخت سیستم‌های جستجویی هوشمندتر می‌برد که نه تنها کلمات را می‌فهمند، بلکه به نیاز پنهان و نیت کاربر نیز نزدیک‌تر می‌شوند.
</p>

    ">Okapi BM25</code> 
    تولید کند.
  </div>

</div>

    font-family: Vazirmatn, Tahoma, Arial, sans-serif;

">
  <div style="margin-bottom: 15px;">
    <span style="font-size: 18pt; font-weight: bold; color: #1B263B;">
      مسئله 11.32: پیاده‌سازی کامل سیستم رتبه‌بندی BM25 (Coding: Full BM25 Ranking System)⭐
    </span>

  <div style="font-size: 16pt; line-height: 2.2; color: #1B263B;">
    تابعی کامل در پایتون طراحی کنید که پارامترهای زیر را به عنوان ورودی دریافت کند:
    <br>
    1. 
    <code style="

    2. 
    <code style="

    (لیستی از اسناد که هر سند یک لیست از کلمات است)
    <br>
    3. 
    <code style="
        background:#f1f1f1;
        padding:2px 6px;

    و 
    <code style="

    <code style="

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.1</span></b>: محاسبه احتمال شرطی در بازیابی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Conditional Probability in Retrieval)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی و کالبدشکافی مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌های احتمالاتی بازیابی اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Probabilistic Retrieval Models)</span>، برای ارزیابی و تخمین میزان ارتباط یک سند با نیاز اطلاعاتی کاربر، از «قضیه بیز» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bayes' Theorem)</span> استفاده می‌شود. این قضیه ابزاری قدرتمند در ریاضیات است که به ما اجازه می‌دهد با در دست داشتن اطلاعات اولیه و ترکیب آن با شواهد جدید (مانند مشاهده یک کلمه خاص در متن سند)، احتمال وقوع یک رویداد (مرتبط بودن سند) را به‌روزرسانی و محاسبه کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری متغیرها
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای حل دقیق مسئله، ابتدا متغیرها را با استفاده از نمادهای استاندارد ریاضی مدل‌سازی می‌کنیم. فرض کنید رویداد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R = 1</span> نشان‌دهنده «مرتبط بودن سند» و رویداد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T</span> نشان‌دهنده «وجود کلمه هدف (هوش) در سند» باشد. بر اساس صورت مسئله، مقادیر زیر استخراج می‌شوند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">الف) احتمال پیشین <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Prior Probability)</span>:</b> احتمال مرتبط بودن یک سند تصادفی، پیش از بررسی محتوای متنی آن: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1) = 0.01</span></li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">ب) احتمال درست‌نمایی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Likelihood)</span>:</b> احتمال وجود کلمه «هوش» به شرطی که بدانیم سند قطعا مرتبط است: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(T|R=1) = 0.20</span></li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">ج) احتمال حاشیه‌ای <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Marginal Probability)</span>:</b> احتمال کل رخداد کلمه «هوش» در تمام اسناد مجموعه (چه مرتبط و چه غیرمرتبط): <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(T) = 0.05</span></li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
<b style="color:#6D2E46;">خواسته مسئله:</b> یافتن احتمال شرطی مرتبط بودن سند، با علم به اینکه کلمه «هوش» در آن مشاهده شده است؛ به زبان ریاضی باید مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|T)</span> را محاسبه کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) طراحی منطق حل و محاسبات گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
<b style="color:#6D2E46;">گام اول: بیان رابطه ریاضی پایه (بسط قضیه بیز)</b><br>
بر اساس فرمول استاندارد قضیه بیز برای این دو رویداد، رابطه ریاضی حاکم بر مسئله به شکل زیر تعریف می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(R=1|T) = \frac{P(T|R=1) \times P(R=1)}{P(T)}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
<b style="color:#6D2E46;">گام دوم: جایگذاری مقادیر استخراج‌شده در فرمول</b><br>
مقادیری که در بخش تحلیل داده‌ها فهرست کردیم را مستقیماً در رابطه بیز قرار می‌دهیم:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(R=1|T) = \frac{0.20 \times 0.01}{0.05}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
<b style="color:#6D2E46;">گام سوم: انجام محاسبات جبری و ساده‌سازی کسرها</b><br>
برای جلوگیری از خطای محاسباتی، ابتدا عملیات ضرب در صورت کسر را اعمال می‌کنیم:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(R=1|T) = \frac{0.002}{0.05}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
جهت حذف اعشار و ساده‌سازی بهتر در محاسبات دستی، صورت و مخرج کسر را در عدد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">1000</span> ضرب می‌کنیم:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(R=1|T) = \frac{0.002 \times 1000}{0.05 \times 1000} = \frac{2}{50}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
با ساده کردن کسر نهایی (تقسیم هم‌زمان صورت و مخرج بر مقسوم‌علیه مشترک یعنی عدد 2) به پاسخ نهایی می‌رسیم:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(R=1|T) = \frac{1}{25} = 0.04}$$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
پاسخ نهایی و نتیجه‌گیری سیستم
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
احتمال اینکه سندی حاوی کلمه «هوش» با نیاز اطلاعاتی کاربر مرتبط باشد، دقیقاً برابر با <b>0.04</b> یا به عبارتی <b>4 درصد</b> است.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته:</b><br>
درک تحلیلی این مسئله در مهندسی بازیابی اطلاعات بسیار حائز اهمیت است. توجه داشته باشید که در ابتدا (پیش از مشاهده کلمه «هوش»)، شانس مرتبط بودن یک سند تصادفی در مجموعه اسناد تنها 1 درصد بود (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1) = 0.01</span>). اما پس از رویت این کلمه کلیدی در متن سند، احتمال ارتباط آن به 4 درصد افزایش یافت (رشد 4 برابری شانس موثر). این تغییر ملموس آماری اثبات می‌کند که واژه «هوش» دارای ارزش اطلاعاتی بسیار بالایی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Information Value)</span> در فرآیند تفکیک است و موتور جستجو باید در مکانیزم رتبه‌بندی، به اسنادی که حاوی این واژه هستند وزن محاسباتی بیشتری اختصاص دهد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.2</span></b>: متغیرهای تصادفی در رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Random Variables in Ranking)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در اصل رتبه‌بندی احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Probability Ranking Principle)</span> یا به‌اختصار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">PRP</span>، فرض بر این است که سیستم بازیابی اطلاعات باید اسناد را بر اساس احتمال مرتبط بودنشان با درخواست کاربر مرتب کند. در این چارچوب نظری، «ارتباط» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Relevance)</span> یک ویژگی قطعی و سیاه‌وسفید نیست، بلکه به عنوان یک متغیر تصادفی دودویی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Binary Random Variable)</span> با نماد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R</span> مدل‌سازی می‌شود. این متغیر تنها می‌تواند دو مقدار بپذیرد:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">R = 1</b>: نشان‌دهنده مرتبط بودن سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Relevant)</span></li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">R = 0</b>: نشان‌دهنده غیرمرتبط بودن سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Non-relevant)</span></li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل معنایی متغیرها
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
هنگامی که یک سیستم بازیابی با یک سند خاص (که آن را با نماد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span> نشان می‌دهیم) مواجه می‌شود، باید تصمیم بگیرد که این سند چقدر برای کاربر مفید است. در اینجا دو مفهوم آماری شکل می‌گیرد:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:14px;"><b style="color:#6D2E46;">1. متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(R=1|d)</span>:</b> این عبارت به معنای «احتمال شرطی مرتبط بودن سند، با فرض مشاهده محتوای سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span>» است. به بیان ساده‌تر، این عدد نشان می‌دهد که بر اساس شواهد موجود در سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span> (مانند کلمات کلیدی، ساختار و محتوا)، چقدر احتمال دارد که این سند نیاز اطلاعاتی کاربر را برطرف کند.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">2. متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(R=0|d)</span>:</b> این عبارت به معنای «احتمال شرطی غیرمرتبط بودن سند، با فرض مشاهده محتوای سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span>» است. این عدد نشان می‌دهد که با وجود محتوای سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span>، چقدر احتمال دارد که این سند برای کاربر کاملاً بی‌فایده باشد.</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
طبق اصول پایه نظریه احتمالات، چون یک سند در نهایت یا مرتبط است یا غیرمرتبط (حالت سومی وجود ندارد)، مجموع این دو احتمال همواره برابر با 1 (یا 100 درصد) خواهد بود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(R=1|d) + P(R=0|d) = 1}$$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام با ارائه یک مثال کاربردی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک عمیق‌تر تفاوت معنایی این دو متغیر، فرایند جستجوی زیر را در نظر بگیرید:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:16px;">
        <b style="color:#6D2E46;">گام اول: تعریف سناریوی جستجو</b><br>
        فرض کنید کاربری در حال مطالعه برای یک آزمون سیستم‌عامل است و عبارت جستجوی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Query)</span> «الگوریتم زمان‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Round Robin</span>» را در موتور جستجو وارد می‌کند.
    </li>
    <li style="margin-bottom:16px;">
        <b style="color:#6D2E46;">گام دوم: بررسی سند اول <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(d<sub>1</sub>)</span></b><br>
        موتور جستجو سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d<sub>1</sub></span> را پیدا می‌کند که یک مقاله دانشگاهی معتبر با عنوان «تحلیل عملکرد زمان‌بندی پردازنده به روش راند-رابین» است. کلمات کلیدی کاربر بارها در متن این مقاله تکرار شده‌اند.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>تحلیل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(R=1|d<sub>1</sub>)</span>:</b> از آنجا که محتوای سند تطابق بالایی با نیاز کاربر دارد، سیستم محاسبه می‌کند که احتمال مرتبط بودن این سند بسیار بالاست. به عنوان مثال: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d<sub>1</sub>) = 0.95</span> (یعنی 95 درصد احتمال دارد این سند دقیقاً همان چیزی باشد که کاربر می‌خواهد).</li>
            <li style="margin-bottom:6px;"><b>تحلیل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(R=0|d<sub>1</sub>)</span>:</b> به صورت منطقی، احتمال نامربوط بودن این مقاله بسیار پایین است: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=0|d<sub>1</sub>) = 0.05</span>.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: بررسی سند دوم <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(d<sub>2</sub>)</span></b><br>
        سپس موتور جستجو سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d<sub>2</sub></span> را بررسی می‌کند که یک صفحه وب با عنوان «برنامه‌ریزی و زمان‌بندی در پروژه‌های عمرانی» است. کلمه «زمان‌بندی» در آن وجود دارد، اما هیچ ارتباطی به مفاهیم کامپیوتر یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Round Robin</span> ندارد.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>تحلیل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(R=1|d<sub>2</sub>)</span>:</b> شواهد موجود در این سند ارتباط معنایی با درخواست کاربر ندارند. بنابراین احتمال مرتبط بودن آن بسیار ناچیز ارزیابی می‌شود: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d<sub>2</sub>) = 0.02</span>.</li>
            <li style="margin-bottom:6px;"><b>تحلیل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(R=0|d<sub>2</sub>)</span>:</b> در مقابل، سیستم با اطمینان بالایی متوجه می‌شود که این سند برای کاربر غیرمرتبط و بی‌فایده است: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=0|d<sub>2</sub>) = 0.98</span>.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
تفاوت معنایی این دو متغیر در این است که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d)</span> <b>شانس موفقیت</b> یک سند در پاسخگویی به کاربر را اندازه‌گیری می‌کند، در حالی که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=0|d)</span> <b>ریسک بی‌فایده بودن</b> آن سند را نشان می‌دهد. در موتورهای جستجوی واقعی که بر اساس اصل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">PRP</span> کار می‌کنند، سیستم تمام اسناد بازیابی شده را بر اساس مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d)</span> از بزرگ به کوچک مرتب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Rank)</span> کرده و به کاربر نمایش می‌دهد تا کاربر ابتدا اسنادی را ببیند که بالاترین شانس ارتباط را دارند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.3</span></b>: فرض استقلال اسناد در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">PRP</span> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Document Independence Assumption in PRP)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در حالت پایه «اصل رتبه‌بندی احتمالی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Probability Ranking Principle)</span>، که معمولاً با تابع زیان 1 و 0 <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(1/0 Loss Case)</span> مدل‌سازی می‌شود، یک فرض اساسی وجود دارد: <b>استقلال اسناد</b>. این فرض بیان می‌کند که احتمال مرتبط بودن یک سند با نیاز کاربر، کاملاً مستقل از سایر اسنادِ موجود در مجموعه یا اسنادی است که پیش از آن به کاربر نمایش داده شده‌اند. به زبان ساده‌تر، سیستم فرض می‌کند که ارزش و سودمندی سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d<sub>i</sub></span> برای کاربر، هیچ تأثیری بر ارزش سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d<sub>j</sub></span> نمی‌گذارد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در یک سیستم ایده‌آل و تئوریک، اگر کاربر به دنبال مفهومی بگردد، سیستم تک‌تک اسناد را به صورت مجزا ارزیابی کرده و نمره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d)</span> را برای هرکدام محاسبه می‌کند. سپس اسناد را صرفاً بر اساس همین نمره‌ی مستقل، از بزرگ به کوچک مرتب می‌کند.
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
با این حال، در دنیای واقعی و در رفتار واقعی کاربران هنگام جستجو، این فرضِ استقلال منطقی نیست و به دلایل متعددی نقض می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام (تحلیل علل نقض فرض استقلال در واقعیت)
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک اینکه چرا ارتباط یک سند در واقعیت به اسناد دیگر وابسته است، 3 دلیل اصلی را بررسی می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: افزونگی اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Information Redundancy)</span></b><br>
        فرض کنید کاربری عبارت «اخبار زلزله امروز» را جستجو می‌کند. سیستم بازیابی، 10 خبرگزاری مختلف را پیدا می‌کند که همگی یک خبر واحد را کلمه‌به‌کلمه کپی کرده‌اند.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>دیدگاه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">PRP</span> (مستقل):</b> سیستم به هر 10 سند بالاترین امتیاز را می‌دهد و همه را در صدر نتایج رتبه‌بندی می‌کند.</li>
            <li style="margin-bottom:6px;"><b>واقعیت (وابسته):</b> وقتی کاربر سند اول را می‌خواند و نیاز اطلاعاتی‌اش برطرف می‌شود، سند دوم (که دقیقاً همان اطلاعات را دارد) دیگر برای او ارزشی نخواهد داشت و «غیرمرتبط» تلقی می‌شود. در اینجا، مرتبط بودن سند دوم کاملاً وابسته به این است که آیا کاربر سند اول را خوانده است یا خیر.</li>
        </ul>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: ابهام در پرس‌وجو و نیاز به تنوع <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query Ambiguity and Diversity)</span></b><br>
        کاربران غالباً از کلمات کلیدی کوتاه و مبهم استفاده می‌کنند. به عنوان مثال، جستجوی کلمه <code style="background:#F3F3F3; padding:2px 6px; border-radius:6px; color:#6D2E46; font-family:monospace; direction:ltr; unicode-bidi:embed;">Java</code> می‌تواند به معنای «زبان برنامه‌نویسی جاوا»، «جزیره جاوا در اندونزی» یا «قهوه جاوا» باشد.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>دیدگاه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">PRP</span> (مستقل):</b> اگر سیستم تشخیص دهد که بیشتر اسناد مجموعه درباره «زبان برنامه‌نویسی» هستند، ممکن است 20 سند اول را تماماً به برنامه‌نویسی اختصاص دهد.</li>
            <li style="margin-bottom:6px;"><b>واقعیت (وابسته):</b> یک سیستم هوشمند می‌داند که اگر سند اول و دوم درباره زبان برنامه‌نویسی بودند، سند سوم باید درباره جزیره یا قهوه باشد تا تمام جنبه‌های احتمالی نیاز کاربر پوشش داده شود. در اینجا، انتخاب سند سوم وابسته به محتوای اسناد قبلی است تا تنوع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Diversity)</span> نتایج حفظ شود.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: مکمل بودن اسناد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Complementarity)</span></b><br>
        گاهی اوقات یک سند به تنهایی پاسخگوی نیاز کاربر نیست، اما در کنار یک سند دیگر، ارزش بسیار بالایی پیدا می‌کند.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;">به عنوان مثال، یک دانشجو به دنبال یادگیری «نحوه پیاده‌سازی فشرده‌سازی بایت متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Variable Byte Code)</span>» است. سند اول فقط تئوری ریاضی را توضیح داده و سند دوم فقط کدهای پایتون را بدون توضیح قرار داده است. هر کدام از این اسناد به تنهایی ممکن است برای کاربر کاملاً مفید نباشند (ارتباط پایین)، اما ترکیب خواندن سند اول و سپس سند دوم، نیاز کاربر را کاملاً رفع می‌کند. در این حالت، ارزش سند دوم کاملاً وابسته به اطلاعات سند اول است.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
فرض استقلال اسناد در مدل پایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">PRP</span>، یک ساده‌سازی ریاضی است که از پیچیدگی‌های مسئله می‌کاهد، اما با رفتار واقعی کاربران هم‌خوانی کامل ندارد. در واقعیت، ارزش اطلاعاتی یک سند تابعی از اسنادی است که کاربر پیش از آن مشاهده کرده است.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
برای حل مشکل نقض فرض استقلال در موتورهای جستجوی مدرن، مفهوم «ارتباط حاشیه‌ای» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Marginal Relevance)</span> معرفی شده است. سیستم‌های امروزی از الگوریتم‌هایی مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">MMR (Maximal Marginal Relevance)</span> استفاده می‌کنند تا علاوه بر مرتبط بودن اسناد، میزان «جدید بودن» و «تکراری نبودن» اطلاعات آن‌ها نسبت به اسناد قبلیِ نمایش داده شده را نیز در رتبه‌بندی لحاظ کنند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.4</span></b>: هزینه خطای جستجو <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Search Error Costs)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در تئوری تصمیم‌گیری بیز <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bayesian Decision Theory)</span> که پایه‌ی مدل‌های پیشرفته‌ی بازیابی اطلاعات است، هدف تنها مرتب‌سازی اسناد نیست؛ بلکه حداقل کردن «ریسک» یا «هزینه»ی ناشی از تصمیمات اشتباه است. هر سیستم بازیابی برای هر سند در نهایت یک تصمیم دودویی می‌گیرد: «سند را به کاربر نمایش دهد» یا «سند را فیلتر کند». از آنجا که سیستم‌ها بی‌نقص نیستند، در این تصمیم‌گیری‌ها مرتکب خطا می‌شوند. برای مدل‌سازی ریاضیِ اثرات منفیِ این خطاها بر تجربه کاربر، از پارامترهای هزینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cost)</span> استفاده می‌کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در نمادگذاری استاندارد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>ij</sub></span>، زیرنویس اول (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span>) نشان‌دهنده‌ی «تصمیم سیستم» و زیرنویس دوم (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">j</span>) نشان‌دهنده‌ی «وضعیت واقعی سند» است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">1</b>: به معنای «مثبت» (بازیابی شدن توسط سیستم / مرتبط بودن در واقعیت)</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">0</b>: به معنای «منفی» (عدم بازیابی توسط سیستم / غیرمرتبط بودن در واقعیت)</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
بر این اساس، 2 نوع خطای اساسی و هزینه‌های مترتب بر آن‌ها در سیستم رخ می‌دهد که در ادامه آن‌ها را تحلیل می‌کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: تحلیل متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">C<sub>10</sub></span> (خطای بازیابی سند غیرمرتبط)</b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>مفهوم نماد:</b> سیستم تصمیم گرفته است سند را بازیابی کند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i = 1</span>)، در حالی که وضعیت واقعی سند، غیرمرتبط با نیاز کاربر است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">j = 0</span>).</li>
            <li style="margin-bottom:6px;"><b>معادل آماری:</b> این حالت دقیقاً بیانگر خطای <b>مثبت کاذب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(False Positive)</span></b> یا خطای نوع اول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Type I Error)</span> است. سیستم به اشتباه فکر کرده که این سند برای کاربر مفید است.</li>
            <li style="margin-bottom:6px;"><b>تأثیر بر کاربر:</b> هزینه‌ی این خطا به شکل «اتلاف وقت کاربر» خود را نشان می‌دهد. کاربر روی لینک کلیک می‌کند، متن را می‌خواند و متوجه می‌شود که محتوا هیچ ربطی به نیاز او ندارد. این خطا باعث کاهش دقت <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Precision)</span> سیستم می‌شود.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام دوم: تحلیل متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">C<sub>01</sub></span> (خطای عدم بازیابی سند مرتبط)</b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>مفهوم نماد:</b> سیستم تصمیم گرفته است سند را بازیابی نکند و آن را کنار بگذارد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i = 0</span>)، در حالی که در واقعیت، آن سند حاوی اطلاعات مرتبط و مفیدی برای کاربر بوده است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">j = 1</span>).</li>
            <li style="margin-bottom:6px;"><b>معادل آماری:</b> این حالت بیانگر خطای <b>منفی کاذب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(False Negative)</span></b> یا خطای نوع دوم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Type II Error)</span> است. سیستم نتوانسته است یک سند ارزشمند را تشخیص دهد.</li>
            <li style="margin-bottom:6px;"><b>تأثیر بر کاربر:</b> هزینه‌ی این خطا به شکل «از دست دادن اطلاعات حیاتی» بروز می‌کند. کاربر هرگز متوجه نمی‌شود که چنین سند مفیدی وجود داشته است. این خطا مستقیماً باعث کاهش فراخوانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Recall)</span> سیستم می‌شود.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
به‌طور خلاصه، در مدل‌سازی هزینه‌های سیستم بازیابی:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;">متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>10</sub></span> معادل خطای <b>مثبت کاذب <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(False Positive)</span></b> است (ارائه اطلاعات بی‌ارزش به کاربر).</li>
    <li style="margin-bottom:10px;">متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>01</sub></span> معادل خطای <b>منفی کاذب <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(False Negative)</span></b> است (پنهان ماندن اطلاعات ارزشمند از دید کاربر).</li>
</ul>
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
در طراحی یک موتور جستجوی واقعی، وزن این 2 هزینه معمولاً برابر نیست. به عنوان مثال، در یک موتور جستجوی وب عمومی مانند گوگل، هزینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>10</sub></span> بسیار بالاست؛ زیرا اگر در صفحه اول نتایج غیرمرتبط (اسپم) نمایش داده شود، کاربر بلافاصله ناراضی می‌شود. اما در جستجوهای حقوقی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Legal Search)</span> یا پزشکی، هزینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>01</sub></span> به شدت بحرانی است؛ زیرا پیدا نکردن یک پرونده قضایی مرتبط یا یک مقاله دارویی خاص (منفی کاذب) می‌تواند خسارات جبران‌ناپذیری به همراه داشته باشد. در چنین سیستم‌هایی، آستانه تصمیم‌گیری به گونه‌ای تنظیم می‌شود که به قیمت پذیرش چند مثبت کاذب، هیچ منفی کاذبی رخ ندهد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.5</span></b>: اصول مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM Principles)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Binary Independence Model)</span> یا به‌اختصار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، یکی از پایه‌ای‌ترین و شناخته‌شده‌ترین مدل‌های احتمالاتی در بازیابی اطلاعات است. این مدل برای غلبه بر پیچیدگی‌های محاسباتی در تخمین احتمال ارتباط یک سند، فرایند مدل‌سازی را بر اساس 2 فرض ساده‌کننده و در عین حال محدودکننده بنا می‌کند. نام این مدل مستقیماً از همین 2 فرض گرفته شده است: <b>بازنمایی باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Binary)</span></b> و <b>استقلال کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Independence)</span></b>.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
هدف این مسئله، کالبدشکافی این 2 کلمه در نام مدل و درک دقیق محدودیت‌هایی است که هر کدام به سیستم بازیابی (در نحوه نگاه به اسناد و پرسمان‌ها) تحمیل می‌کنند. در ادامه، هر یک از این اصول را به تفکیک بررسی و محدودیت‌های ناشی از آن‌ها را تحلیل می‌کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تحلیل مفهوم «باینری» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Binary Representation)</span> و محدودیت‌های آن</b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:12px;"><b>تعریف:</b> در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، هر سند و هر پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Query)</span> تنها به شکل یک بردار باینری (0 و 1) از کلمات بازنمایی می‌شود. اگر لغت‌نامه سیستم دارای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">M</span> کلمه باشد، یک سند به صورت بردار <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\vec{x} = (x_1, x_2, \dots, x_M)\)</span> نشان داده می‌شود که در آن <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(x_i \in \{0, 1\}\)</span> است. اگر کلمه در سند وجود داشته باشد مقدار آن <b>1</b> و در غیر این صورت <b>0</b> خواهد بود.</li>
            <li style="margin-bottom:6px;"><b>محدودیت‌های ایجادشده:</b>
                <ul style="margin-right:20px; padding-right:15px; margin-top:6px; list-style-type: decimal;">
                    <li style="margin-bottom:8px;"><b style="color:#6D2E46;">نادیده گرفتن بسامد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Frequency - TF)</span>:</b> این مدل هیچ تفاوتی بین سندی که کلمه «الگوریتم» در آن <b>50</b> بار تکرار شده با سندی که این کلمه تنها <b>1</b> بار در آن آمده، قائل نمی‌شود. هر دو سند برای این کلمه مقدار <b>1</b> را دریافت می‌کنند. این در حالی است که تکرار یک کلمه سیگنال قدرتمندی از موضوع اصلی سند است.</li>
                    <li style="margin-bottom:8px;"><b style="color:#6D2E46;">نادیده گرفتن طول سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Document Length)</span>:</b> در یک بردار باینری، طول واقعی متن گم می‌شود. یک مقاله <b>100</b> صفحه‌ای و یک پاراگراف <b>3</b> خطی در صورتی که کلمات منحصر‌به‌فرد یکسانی داشته باشند، دقیقاً دارای بردارهای باینری مشابهی خواهند بود، که این امر در رتبه‌بندی عادلانه اختلال ایجاد می‌کند.</li>
                </ul>
            </li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام دوم: تحلیل مفهوم «استقلال» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Independence Assumption)</span> و محدودیت‌های آن</b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:12px;"><b>تعریف:</b> این فرض بیان می‌کند که حضور یا عدم حضور یک کلمه در سند، کاملاً مستقل از حضور یا عدم حضور سایر کلمات در آن سند است (استقلال شرطی با فرض مشخص بودن وضعیت ارتباط). از نظر ریاضی، احتمال مشاهده بردار سند <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\vec{x}\)</span> برابر است با حاصل‌ضرب احتمال مشاهده تک‌تک کلمات آن:
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{P(\vec{x}|R) = \prod_{i=1}^{M} P(x_i|R)}$$
                </div>
            </li>
            <li style="margin-bottom:6px;"><b>محدودیت‌های ایجادشده:</b>
                <ul style="margin-right:20px; padding-right:15px; margin-top:6px; list-style-type: decimal;">
                    <li style="margin-bottom:8px;"><b style="color:#6D2E46;">نادیده گرفتن هم‌رخدادی کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Co-occurrence)</span>:</b> در زبان طبیعی، کلمات به شدت به هم وابسته‌اند. اگر در متنی کلمه «سیستم» بیاید، احتمال ظاهر شدن کلمه «عامل» بسیار بالاست. مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> این ارتباطات آماری و معنایی را کاملاً کورکورانه نادیده می‌گیرد و فرض می‌کند آمدن کلمه «سیستم» هیچ اطلاعات اضافه‌ای درباره احتمال آمدن کلمه «عامل» به ما نمی‌دهد.</li>
                    <li style="margin-bottom:8px;"><b style="color:#6D2E46;">از بین رفتن ساختار و عبارات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Phrases)</span>:</b> چون کلمات مستقل فرض می‌شوند، توالی آن‌ها بی‌اهمیت می‌شود. برای این مدل، عبارت «پایگاه داده» صرفاً ترکیب 2 کلمه مستقل است و مفهوم یکپارچه آن قابل درک نیست.</li>
                </ul>
            </li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> با استفاده از کلمه <b>«باینری»</b>، وزن کلمات و تکرار آن‌ها را حذف می‌کند و با کلمه <b>«استقلال»</b>، روابط معنایی و هم‌رخدادی کلمات در زبان طبیعی را از بین می‌برد. این 2 فرض، اگرچه از نظر ریاضی محاسبات احتمالات را بسیار ساده و سریع می‌کنند، اما باعث از دست رفتن بخش بزرگی از اطلاعات حیاتی متن می‌شوند و سقف دقت سیستم بازیابی را محدود می‌کنند.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
درک این محدودیت‌ها نقطه شروع بسیار خوبی برای ورود به مباحث پیشرفته‌تر است. در واقع، محققان برای جبران همین نقصِ «باینری بودن» در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، مفهوم فراوانی کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(TF)</span> و طول سند را به مدل‌های احتمالاتی اضافه کردند که نتیجه‌ی آن، تولد تابع رتبه‌بندی مشهور <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> بود؛ تابعی که امروزه به عنوان استاندارد صنعتی در موتورهای جستجو شناخته می‌شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.6</span></b>: نسبت شانس در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Odds Ratio in BIM)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌های احتمالاتی بازیابی اطلاعات، هدف اصلی مرتب‌سازی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Ranking)</span> اسناد بر اساس احتمال مرتبط بودن آن‌ها با درخواست کاربر است. در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، به جای محاسبه مستقیم احتمال شرطی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d)</span>، از مفهوم آماری «نسبت شانس» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Odds Ratio)</span> استفاده می‌شود. نسبت شانس، نسبتی است میان احتمال رخداد یک رویداد (مرتبط بودن) به احتمال عدم رخداد آن (غیرمرتبط بودن). استفاده از این ترفند، گره‌های محاسباتی بزرگی را در ریاضیاتِ بازیابی اطلاعات باز می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
طبق قضیه بیز، محاسبه مستقیم احتمال مرتبط بودن و غیرمرتبط بودن یک سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span>) به شکل زیر بسط داده می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{\begin{aligned} P(R=1|d) &= \frac{P(d|R=1) \times P(R=1)}{P(d)} \\[8pt] P(R=0|d) &= \frac{P(d|R=0) \times P(R=0)}{P(d)} \end{aligned}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
در این روابط، <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(d)</span> بیانگر احتمال حاشیه‌ای مشاهده‌ی دقیقِ بردار سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span> در کل مجموعه اسناد است. محاسبه این مخرج کسر در عمل بسیار دشوار و پرهزینه است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام و تحلیل مزیت‌های ریاضی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
استفاده از نسبت شانس <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">O(R|d)</span> <b>3</b> مزیت ریاضیاتی بنیادین برای سیستم ایجاد می‌کند که آن‌ها را مرحله‌به‌مرحله بررسی می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: حذف مخرج مشترک و پیچیده (حذف <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(d)</span>)</b><br>
        با تقسیم <b>2</b> رابطه بیز بر یکدیگر برای محاسبه نسبت شانس، عبارت <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(d)</span> که در مخرج هر <b>2</b> وجود دارد، به طور کامل ساده شده و حذف می‌گردد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{O(R|d) = \frac{P(R=1|d)}{P(R=0|d)} = \frac{P(d|R=1) \times P(R=1)}{P(d|R=0) \times P(R=0)}}$$
        </div>
        <b>مزیت اول:</b> سیستم دیگر نیازی به محاسبه احتمال رخداد تک‌تک اسناد در کل مجموعه ندارد. از آنجا که هدف فقط رتبه‌بندی اسناد نسبت به یکدیگر است، این نسبت به تنهایی برای مرتب‌سازی کفایت می‌کند (زیرا تابعی صعودی از احتمال است).
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: اعمال فرض استقلال و تبدیل به ضرب زنجیره‌ای</b><br>
        با اعمال فرض استقلال کلمات در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، احتمال مشاهده کل سند به حاصل‌ضرب احتمال مشاهده تک‌تک کلمات آن (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>i</sub></span>) تجزیه می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{O(R|d) = \left( \prod_{i=1}^{M} \frac{P(t_i|R=1)}{P(t_i|R=0)} \right) \times \frac{P(R=1)}{P(R=0)}}$$
        </div>
        <b>مزیت دوم:</b> اکنون سیستم می‌تواند شانس مرتبط بودن را با ضرب کردن وزن کلمات مجزای موجود در سند محاسبه کند.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: لگاریتم‌گیری و خطی‌سازی معادله <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Linearization)</span></b><br>
        ضرب تعداد زیادی عدد اعشاری کوچک (احتمالات) در کامپیوتر منجر به خطای «سرریز به پایین» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Arithmetic Underflow)</span> می‌شود و مقدار نهایی <b>0</b> مطلق در نظر گرفته می‌شود. برای حل این مشکل، از طرفین لگاریتم می‌گیریم. از آنجا که لگاریتم یک تابع اکیداً صعودی است، ترتیب رتبه‌بندی را تغییر نمی‌دهد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\log O(R|d) = \sum_{i=1}^{M} \log \frac{P(t_i|R=1)}{P(t_i|R=0)} + \log \frac{P(R=1)}{P(R=0)}}$$
        </div>
        <b>مزیت سوم:</b> با این کار، عملگر ضرب به عملگر جمع تبدیل می‌شود. کسر اول (احتمال پیشین) برای تمام اسناد ثابت است و در رتبه‌بندی تاثیری ندارد، بنابراین می‌توان آن را نادیده گرفت. معادله نهایی به یک جمع خطی ساده از وزن کلمات تبدیل می‌شود.
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
استفاده از نسبت شانس در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، به ظاهر یک تغییر ساده است، اما در واقعیت <b>3</b> چالش بزرگ را حل می‌کند: <b>1.</b> حذف متغیر غیرقابل محاسبه‌ی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(d)</span>، <b>2.</b> جلوگیری از خطای محاسباتی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Underflow</span> با امکان‌پذیر ساختن لگاریتم‌گیری و <b>3.</b> تبدیل یک فرمول پیچیده‌ی ضربی به یک تابع امتیازدهیِ خطی و جمعی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Linear Scoring Function)</span> که پردازش آن برای موتورهای جستجو بسیار سریع است.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
خروجی این فرمول لگاریتمی در متون علمی تحت عنوان <b>مقدار وضعیت بازیابی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Retrieval Status Value یا RSV)</span></b> شناخته می‌شود. عبارت داخل سیگما نیز دقیقاً همان منطقی است که بعدها پایه و اساس فرمول مشهور «وزن‌دهی رابرتسون-اسپارک جونز» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Robertson-Sparck Jones weight)</span> در بازیابی اطلاعات را شکل داد. در عمل، سیستم به جای احتمالات، برای هر کلمه یک «وزن» محاسبه کرده و وزن کلمات موجود در سند را با هم جمع می‌کند تا نمره نهایی سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RSV)</span> به دست آید.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.7</span></b>: حذف کلمات ناموجود <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Eliminating Non-occurring Terms)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، فرمول پایه برای محاسبه «مقدار وضعیت بازیابی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Retrieval Status Value یا RSV)</span>، در ابتدا بر روی تمام کلمات موجود در لغت‌نامه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Vocabulary)</span> سیستم بسط داده می‌شود. با این حال، در یک سیستم واقعی، پردازش ده‌ها هزار کلمه‌ی لغت‌نامه برای ارزیابی هر سند، از نظر محاسباتی غیرممکن است. یکی از ظرافت‌های مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> این است که با اعمال یک فرض منطقی و آماری درباره کلماتی که کاربر در جستجوی خود وارد نکرده است، توان محاسباتی مورد نیاز را به شدت کاهش داده و این کلمات را به صورت ریاضی از معادله حذف می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای اثبات این موضوع، ابتدا نمادهای استاندارد آماری را در تابع رتبه‌بندی تعریف می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{q_i \in \{0, 1\}}\)</span>: وضعیت حضور کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Query)</span> کاربر (<b>1</b> یعنی حضور، <b>0</b> یعنی عدم حضور).</li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{x_i \in \{0, 1\}}\)</span>: وضعیت حضور کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Document)</span>.</li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{p_i = P(x_i=1|R=1)}\)</span>: احتمال رخداد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در یک سند مرتبط.</li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{u_i = P(x_i=1|R=0)}\)</span>: احتمال رخداد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در یک سند غیرمرتبط.</li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{c_i}\)</span>: وزن اختصاص‌یافته به کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در معادله رتبه‌بندی.</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
<b style="color:#6D2E46;">گام اول: بیان تابع رتبه‌بندی پایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(RSV)</span></b><br>
بر اساس مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، پس از لگاریتم‌گیری و حذف مقادیر ثابت، نمره نهایی یک سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RSV<sub>d</sub></span>) به صورت مجموع وزن کلماتی که در آن سند حضور دارند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>i</sub>=1</span>) محاسبه می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{RSV_d = \sum_{x_i=1} c_i}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
در این مرحله، معادله روی تمام کلمات موجود در سند (چه در پرسمان باشند و چه نباشند) اعمال می‌شود.
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:24px;">
<b style="color:#6D2E46;">گام دوم: اعمال فرض تساوی احتمالات برای کلمات خارج از پرسمان</b><br>
اگر کلمه‌ای در پرسمان کاربر وجود نداشته باشد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">q<sub>i</sub>=0</span>)، سیستم فرض می‌کند که این کلمه هیچ اطلاعات و شواهدی درباره مرتبط بودن یا نبودن سند به ما نمی‌دهد. به بیان ریاضی، احتمال رخداد چنین کلمه‌ای در اسناد مرتبط، دقیقاً برابر با احتمال رخداد آن در اسناد غیرمرتبط است:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{p_i = u_i \quad \text{for} \quad q_i = 0}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:24px;">
<b style="color:#6D2E46;">گام سوم: محاسبه وزن ریاضی برای کلمات ناموجود</b><br>
اکنون فرض <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(p_i = u_i\)</span> را در فرمول محاسبه وزن (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(c_i\)</span>) جایگذاری می‌کنیم:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{c_i = \log \frac{p_i(1 - u_i)}{u_i(1 - p_i)} = \log \frac{u_i(1 - u_i)}{u_i(1 - u_i)}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
همان‌طور که مشاهده می‌شود، صورت و مخرج کسر کاملاً با هم برابر می‌شوند. با ساده شدن کسر، مقدار داخل لگاریتم برابر با <b>1</b> خواهد شد:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{c_i = \log(1) = 0}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:24px;">
<b style="color:#6D2E46;">گام چهارم: به‌روزرسانی و ساده‌سازی معادله نهایی</b><br>
از آنجا که ثابت کردیم وزن کلمات غایب در پرسمان (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">q<sub>i</sub>=0</span>) مطلقاً برابر با <b>0</b> است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">c<sub>i</sub>=0</span>)، این کلمات هیچ اثری در جمع نهایی نمره سند ندارند. در نتیجه، عملگر جمع را می‌توان محدود کرد؛ به طوری که تنها روی کلماتی اعمال شود که هم در پرسمان (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">q<sub>i</sub>=1</span>) و هم در سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>i</sub>=1</span>) حضور دارند:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{RSV_d = \sum_{x_i=1 \land q_i=1} c_i}$$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
از طریق معادلات جبری اثبات می‌شود که در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، کلماتی که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">q<sub>i</sub>=0</span> هستند، نسبت شانس برابر با <b>1</b> داشته و در نتیجه پس از لگاریتم‌گیری، وزن آن‌ها <b>0</b> می‌شود و به‌طور کامل از محاسبات خط می‌خورند.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این اثبات ریاضی، اهمیت بنیادینی در معماری سیستم‌های بازیابی اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(IR Architecture)</span> دارد. دقیقاً همین ویژگی است که به ما اجازه می‌دهد در موتورهای جستجو از ساختار «نمایه وارونه» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Inverted Index)</span> استفاده کنیم. به جای بررسی تمام کلمات یک سند، سیستم جستجو تنها لیست‌های ارسال <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Posting Lists)</span> مربوط به کلمات موجود در پرسمان کاربر را فراخوانی می‌کند. این امر پیچیدگی زمانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Time Complexity)</span> پردازش را از مرتبه‌ی «اندازه لغت‌نامه» به مرتبه‌ی «طول پرسمان» کاهش می‌دهد و جستجوی بی‌درنگ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Real-time Search)</span> را در پایگاه‌های داده عظیم امکان‌پذیر می‌سازد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.8</span></b>: جدول توافقی کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Term Contingency Table)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌سازی احتمالاتی و تخمین پارامترها در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، برای درک دقیق توزیع یک کلمه خاص در میان اسناد، از ابزاری آماری به نام «جدول توافقی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Contingency Table)</span> استفاده می‌شود. این جدول ماتریسی است که فراوانی (تعداد) اسناد را بر اساس تقاطع <b>2</b> متغیر دودویی دسته‌بندی می‌کند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">1. وضعیت ارتباط سند:</b> آیا سند مرتبط است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R=1</span>) یا غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R=0</span>)؟</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">2. وضعیت حضور کلمه:</b> آیا کلمه در سند حضور دارد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>i</sub>=1</span>) یا غایب است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>i</sub>=0</span>)؟</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل متغیرها و رسم جدول
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای ساخت این جدول، ابتدا نمادهای استاندارد فراوانی را برای یک مجموعه سند با اندازه کل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span> تعریف می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">S</b>: تعداد کل اسناد مرتبط با پرسمان کاربر.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">N - S</b>: تعداد کل اسناد غیرمرتبط با پرسمان.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">n</b>: تعداد کل اسنادی که کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در آن‌ها وجود دارد.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">s</b>: تعداد اسناد مرتبطی که کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در آن‌ها وجود دارد.</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
بر اساس این متغیرها، تقاطع حالات مختلف در یک جدول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">2 &times; 2</span> به شکل زیر تدوین می‌شود:
</div>

            <th style="padding:14px; border:1px solid #D8D2E6;">اسناد غیرمرتبط <span style="direction:ltr; display:inline-block; font-weight:normal;">(R=0)</span></th>

            <td style="padding:14px; border:1px solid #D8D2E6; font-weight:bold; direction:ltr;">s</td>

            <td style="padding:14px; border:1px solid #D8D2E6;"><b>غایب <span style="direction:ltr; display:inline-block; color:#6D2E46;">(x<sub>i</sub>=0)</span></b></td>
            <td style="padding:14px; border:1px solid #D8D2E6; font-weight:bold; direction:ltr;">S - s</td>

            <td style="padding:14px; border:1px solid #D8D2E6; font-weight:bold; direction:ltr; color:#6D2E46; font-size:18pt;">N</td>
        </tr>
    </tbody>

</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) محاسبه پارامترهای <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">p<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>i</sub></span>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
هدف اصلی از رسم این جدول در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، استخراج و تخمین <b>2</b> احتمال بنیادین <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> است. با استفاده از داده‌های جدول بالا، این <b>2</b> پارامتر را از طریق روابط «احتمال شرطی» محاسبه می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: استخراج پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">p<sub>i</sub></span></b><br>
        این پارامتر بیانگر «احتمال حضور کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> به شرط مرتبط بودن سند» است. برای محاسبه آن، تنها به ستون اول جدول (اسناد مرتبط) نگاه می‌کنیم. تعداد اسناد مرتبطی که کلمه را دارند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span>) را بر کل اسناد مرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span>) تقسیم می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_i = \frac{s}{S}}$$
        </div>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام دوم: استخراج پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>i</sub></span></b><br>
        این پارامتر بیانگر «احتمال حضور کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> به شرط غیرمرتبط بودن سند» است. برای محاسبه آن، تنها به ستون دوم جدول (اسناد غیرمرتبط) نگاه می‌کنیم. تعداد اسناد غیرمرتبطی که کلمه را دارند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n - s</span>) را بر کل اسناد غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N - S</span>) تقسیم می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{u_i = \frac{n - s}{N - S}}$$
        </div>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
جدول توافقی، پلی میان تئوری احتمالات انتزاعی و داده‌های قابل شمارش در یک مجموعه سند است. با جایگذاری مقادیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> (به دست آمده از این جدول) در تابع رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، فرمول مشهور <b>وزن‌دهی رابرتسون-اسپارک جونز <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Robertson-Sparck Jones Weight)</span></b> شکل می‌گیرد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
در عمل، سیستم‌های بازیابی اطلاعات در لحظه جستجو، مقدار دقیق <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span> را نمی‌دانند (زیرا هنوز نمی‌دانند کدام اسناد برای کاربر مرتبط هستند). بنابراین، در ابتدا مقادیر پیش‌فرضی برای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> در نظر می‌گیرند (مثلاً <b>0.5</b>) و پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> را با تقریب <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\frac{n}{N}\)</span> محاسبه می‌کنند (با این فرض که اکثر اسناد مجموعه غیرمرتبط هستند). سپس با استفاده از تکنیک «بازخورد کاذب» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Pseudo-Relevance Feedback)</span>، این جدول را به صورت پویا تخمین زده و به‌روزرسانی می‌کنند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.9</span></b>: مشکل احتمالات صفر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(The Zero Probability Problem)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، محاسبه وزن هر کلمه بر اساس داده‌های موجود در مجموعه اسناد انجام می‌شود. یکی از مشکلات رایج در تخمین احتمالات با استفاده از داده‌های تجربی، مسئله «پراکندگی داده‌ها» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Data Sparsity)</span> است. اگر یک کلمه کلیدی، با وجود اهمیتش، به صورت تصادفی در هیچ‌یک از اسنادی که تاکنون مرتبط تشخیص داده شده‌اند ظاهر نشود، احتمال شرطی آن برابر با <b>0</b> (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0</span>) خواهد شد. این مسئله در ظاهر یک اتفاق آماری ساده است، اما در ریاضیاتِ تابع رتبه‌بندی، منجر به یک فروپاشی محاسباتی می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
یادآوری می‌کنیم که فرمول پایه برای محاسبه وزن یک کلمه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">c<sub>i</sub></span>) در محاسبه مقدار وضعیت بازیابی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RSV)</span> به شکل زیر است:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{c_i = \log \frac{p_i(1 - u_i)}{u_i(1 - p_i)}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
حال بررسی می‌کنیم که اگر مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> برابر با صفر شود، چه اختلالی در ساختار این فرمول رخ می‌دهد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: بروز خطای ریاضی (محاسبه لگاریتم صفر)</b><br>
        اگر فرض کنیم هیچ سند مرتبطی حاوی کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> نیست، مقدار پارامتر مربوطه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0</span> خواهد شد. با قرار دادن این مقدار در معادله وزن کلمه خواهیم داشت:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{0 \times (1 - u_i)}{u_i(1 - 0)} = \log(0)}$$
        </div>
        از نظر ریاضیاتی، تابع لگاریتم برای عدد <b>0</b> تعریف نشده است (حد آن به سمت منفی بی‌نهایت میل می‌کند). در نتیجه، سیستم بازیابی در محاسبه نمره نهایی دچار خطا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Crash)</span> شده و از کار می‌افتد.<br>
        <span style="font-size:14pt; color:#666666;">*(توجه داشته باشید که همین مشکل زمانی که یک کلمه در تمام اسناد مرتبط ظاهر شود نیز رخ می‌دهد؛ یعنی در حالتی که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 1</span> باشد، مخرج کسر صفر شده و خطای تقسیم بر صفر به وجود می‌آید).*</span>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: راهکار تئوری (تکنیک هموارسازی یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Smoothing</span>)</b><br>
        برای حل این مشکل بنیادین، در تئوری احتمالات و بازیابی اطلاعات از تکنیکی به نام <b>هموارسازی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Smoothing)</span></b> استفاده می‌شود. ایده اصلی این است که سیستم هیچ‌گاه نباید به یک رویداد، احتمال مطلقاً صفر (غیرممکن) یا مطلقاً یک (قطعی) اختصاص دهد، زیرا داده‌های نمونه‌برداری‌شده همواره محدود هستند و عدم مشاهده یک کلمه در نمونه فعلی، به معنای عدم وجود مطلق آن در کل جامعه آماری نیست.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: اعمال هموارسازی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">0.5</span> (تعدیل رابرتسون)</b><br>
        رایج‌ترین روش برای هموارسازی در این مدل، افزودن یک مقدار ثابت و کوچک (معمولاً <b>0.5</b>) به تمام شمارنده‌های جدول توافقی است. با فرض اینکه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span> تعداد اسناد مرتبطِ حاوی کلمه و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span> کل اسناد مرتبط باشد، فرمول تخمین احتمالات از حالت پایه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = s / S</span>) به شکل زیر اصلاح می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_i = \frac{s + 0.5}{S + 1}}$$
        </div>
        <span style="font-size:14pt; color:#666666;">نکته: علت افزودن عدد <b>1</b> به مخرج این است که مقدار <b>0.5</b> به حالت موفقیت (حضور کلمه) و مقدار <b>0.5</b> به حالت شکست (عدم حضور کلمه) اضافه شده است که مجموعاً فضای نمونه در مخرج را <b>1</b> واحد افزایش می‌دهد.</span>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مشکل احتمالات صفر باعث تولید مقادیر تعریف‌نشده ریاضی مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">log(0)</span> یا تقسیم بر <b>0</b> در فرمول پایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RSV</span> می‌شود. راه حل قطعی این مشکل، استفاده از <b>تکنیک هموارسازی (افزودن <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">0.5</span>)</b> است. با این کار، حتی اگر کلمه‌ای در اسناد مرتبط یافت نشود (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s = 0</span>)، مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> به جای صفر مطلق، برابر با عددی بسیار کوچک (مثلاً <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\frac{0.5}{S + 1}\)</span>) خواهد شد که اجازه می‌دهد محاسبات لگاریتمی به صورت پایدار و بدون توقف انجام شوند.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این فرمول هموارشده، مستقیماً پایه و اساس فرمول مشهور «وزن‌دهی رابرتسون-اسپارک جونز» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Robertson-Sparck Jones)</span> را تشکیل داد که از نظر جبری به شکل زیر بسط داده می‌شود:
<div style="text-align:center; direction:ltr; font-size:16pt; margin:10px 0; color:#6D2E46;">
$$\boldsymbol{\log \left( \frac{(s + 0.5)(N - S - n + s + 0.5)}{(S - s + 0.5)(n - s + 0.5)} \right)}$$
</div>
امروزه این فرمول تعدیل‌شده در الگوریتم‌های پیشرفته‌ای مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> به عنوان هسته اصلی محاسبه وزن کلمات (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> پیشرفته) مورد استفاده قرار می‌گیرد و مقاومت سیستم را در برابر مجموعه‌داده‌های تُنُک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Sparse Data)</span> تضمین می‌کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.10</span></b>: هموارسازی نیم‌واحدی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Add-0.5 Smoothing)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌های احتمالاتی مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، اتکای صرف به داده‌های نمونه‌برداری‌شده برای تخمین احتمالات می‌تواند به مقادیر <b>0</b> یا <b>1</b> منجر شود. برای رفع این ناپایداری ریاضی و جلوگیری از خطای محاسبه لگاریتم، از تکنیک «هموارسازی نیم‌واحدی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Add-0.5 Smoothing)</span> استفاده می‌شود. این روش که توسط رابرتسون و اسپارک جونز معرفی شد، توزیع احتمالات را متعادل کرده و مانع از حذف کامل یک کلمه از محاسبات به دلیل کمبود داده <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Data Sparsity)</span> می‌گردد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل متغیرها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای فرمول‌بندی دقیق این تکنیک، ابتدا باید <b>4</b> متغیر کلیدی که فضای جستجو و وضعیت اسناد را توصیف می‌کنند، تعریف کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">N</b>: اندازه کل مجموعه (تعداد کل اسناد موجود در پایگاه داده).</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">S</b>: کل اسناد مرتبط (تعداد اسنادی که برای پرسمان جاری، مرتبط تشخیص داده شده‌اند).</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">df<sub>i</sub></b>: فراوانی سند برای کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> (تعداد کل اسنادی در مجموعه که حاوی کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> هستند). این متغیر معادل متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n</span> در نمادگذاری‌های رایج دیگر است.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">s</b>: تعداد اسناد مرتبطی که حاوی کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> هستند.</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: یادآوری احتمالات پایه (بدون هموارسازی)</b><br>
        در حالت پایه، احتمال حضور کلمه در اسناد مرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span>) و غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>) از طریق تقسیم مستقیم مشاهدات بر فضای نمونه به دست می‌آید:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_i = \frac{s}{S} \quad , \quad u_i = \frac{df_i - s}{N - S}}$$
        </div>
        <span style="font-size:14pt; color:#666666;">*(در مخرج <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>، کل اسناد غیرمرتبط برابر با <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N - S</span> و در صورت آن، تعداد اسناد غیرمرتبطِ حاوی کلمه برابر با <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub> - s</span> است).*</span>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: اعمال منطق هموارسازی نیم‌واحدی</b><br>
        برای جلوگیری از صفر شدن صورت کسر (در صورت نبود کلمه) یا مخرج آن (در صورت نبود سند مرتبط)، یک مقدار ثابت <b>0.5</b> به تعداد مشاهدات موفق (صورت کسر) اضافه می‌شود. از آنجا که این مقدار ثابت باید برای حالت ناموفق (عدم حضور کلمه) نیز به همین میزان لحاظ شود، مجموعاً فضای نمونه (مخرج کسر) به اندازه <b>0.5 + 0.5 = 1</b> واحد افزایش می‌یابد.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: بازنویسی فرمول‌های نهایی</b><br>
        با اعمال این منطق بر فرمول‌های پایه، روابط اصلاح‌شده برای تخمین احتمالات به شکل زیر تدوین می‌شوند:
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b>فرمول هموارشده <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">p<sub>i</sub></span>:</b>
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{p_i = \frac{s + 0.5}{S + 1}}$$
                </div>
            </li>
            <li style="margin-bottom:6px;"><b>فرمول هموارشده <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>i</sub></span>:</b>
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{u_i = \frac{df_i - s + 0.5}{N - S + 1}}$$
                </div>
            </li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
استفاده از هموارسازی نیم‌واحدی، پارامترهای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> را از خطرات ناشی از احتمالات صفر مطلق مصون می‌دارد. متغیرهای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span> به‌طور مستقیم نمایانگر اطلاعات مربوط به اسناد مرتبط هستند، در حالی که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub></span> نمایانگر اطلاعات سراسری مجموعه می‌باشند. ترکیب این <b>4</b> متغیر با ثابت‌های هموارسازی، فرمول‌های پایداری برای ارزیابی وزن کلمات ایجاد می‌کند.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
دلیل انتخاب عدد <b>0.5</b> (به جای مقادیر دیگر مانند <b>1</b> در هموارسازی لاپلاس) در بازیابی اطلاعات، ریشه در توزیع‌های آماری و مطالعات تجربی اولیه دارد. اگر این مقادیر هموارشده را مستقیماً در تابع پایه لگاریتمی قرار دهیم، به رابطه نهایی و بسیار مهم <b>وزن‌دهی رابرتسون-اسپارک جونز <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(RSJ Weight)</span></b> می‌رسیم. این فرمول همچنان به عنوان هسته اصلی برای محاسبه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> در الگوریتم‌های رتبه‌بندی مدرن کاربرد دارد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.11</span></b>: مقادیر پیش‌فرض احتمالات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Default Probability Values)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، سیستم برای محاسبه دقیق وزن کلمات به بازخورد کاربر و اطلاعات اسناد مرتبط نیاز دارد (تا مقادیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span> مشخص شوند). با این حال، در لحظه شروع جستجو (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Iteration 0</span>)، سیستم با مشکل «شروع سرد» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cold Start)</span> مواجه است؛ کاربر به تازگی پرسمان خود را وارد کرده و هنوز هیچ سندی به عنوان مرتبط ارزیابی نشده است. در این شرایط، سیستم ناچار است برای محاسبه رتبه‌بندی اولیه و ارائه سری اول نتایج به کاربر، از فرضیات منطقی و مقادیر پیش‌فرض آماری برای تخمین <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> استفاده کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و فرضیات
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در لحظه شروع جستجو داده‌های ما بدین شکل است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;">تعداد کل اسناد مرتبط شناخته‌شده برابر با <b>0</b> است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S = 0</span>).</li>
    <li style="margin-bottom:10px;">تعداد اسناد مرتبطِ حاوی کلمه هدف برابر با <b>0</b> است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s = 0</span>).</li>
    <li style="margin-bottom:10px;">اندازه کل مجموعه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span>) و فراوانی کلمه در کل مجموعه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub></span>) به عنوان داده‌های سراسریِ نمایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Index)</span> از قبل در دسترس سیستم هستند.</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: تعیین مقدار پیش‌فرض برای متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">p<sub>i</sub></span></b><br>
        متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> بیانگر احتمال حضور کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در اسناد مرتبط است. از آنجا که سیستم هیچ دانشی درباره اسناد مرتبط ندارد، از اصل آماری «حداکثر آنتروپی» استفاده می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_i = 0.5}$$
        </div>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: تخمین مقدار پیش‌فرض برای متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>i</sub></span></b><br>
        با تقریب کل مجموعه اسناد معادل اسناد غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">\(N-S \approx N\)</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">\(df_i-s \approx df_i\)</span>):
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{u_i = \frac{df_i}{N}}$$
        </div>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: بررسی اثر این پیش‌فرض‌ها بر فرمول رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(RSV)</span></b><br>
        با جایگذاری این مقادیر در فرمول پایه، وزن نهایی به فرم زیر ساده می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{0.5 \times (1 - \frac{df_i}{N})}{\frac{df_i}{N} \times (1 - 0.5)} = \log \frac{1 - \frac{df_i}{N}}{\frac{df_i}{N}} \approx \log \frac{N}{df_i}}$$
        </div>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
هنگامی که جستجو آغاز می‌شود و بازخوردی وجود ندارد، مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> به صورت پیش‌فرض مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0.5</span> را در نظر می‌گیرد و مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> را برابر با <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\frac{df_i}{N}\)</span> تخمین می‌زند.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
فرمول نهایی به دست آمده، یعنی <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\log(\frac{N}{df_i})\)</span>، دقیقاً همان فرمول مشهور <b>فراوانی معکوس سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(IDF)</span></b> است.  این اثبات نشان می‌دهد که مفهوم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> که در ابتدا در مدل فضای برداری صرفاً بر اساس شمِ مهندسی ابداع شده بود، در واقع پشتوانه ریاضی قدرتمندی در مدل‌های احتمالاتی دارد. سیستم پس از محاسبه اولیه، با دریافت کلیک کاربر، مقادیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span> را آپدیت کرده و محاسبات را مجدداً تکرار می‌کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.12</span></b>: مقایسه احتمالات و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Probabilities vs. IDF)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مفهوم «فراوانی معکوس سند» یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF (Inverse Document Frequency)</span>، در ابتدا توسط کارن اسپارک جونز <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Karen Spärck Jones)</span> به صورت تجربی و مبتنی بر بینش مهندسی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Heuristic)</span> معرفی شد تا به کلماتی که در مجموعه اسناد نادرتر هستند، وزن بیشتری اختصاص دهد. با این حال، یکی از دستاوردهای بزرگ «مدل استقلال باینری» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> این است که نشان می‌دهد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> صرفاً یک فرمول ابتکاری نیست، بلکه تحت شرایط اولیه و فرضیات خاص (مانند زمانی که هیچ دانشی از اسناد مرتبط نداریم)، دارای پشتوانه و اثبات دقیق ریاضی در تئوری احتمالات است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای اثبات این ارتباط، نقطه شروع ما فرمول وزن‌دهی کلمات در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> است که در غیاب هموارسازی به شکل زیر تعریف می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{c_i = \log \frac{p_i(1 - u_i)}{u_i(1 - p_i)}}$$
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
در این مسئله، فرض بنیادین این است که سیستم در نقطه شروع جستجو قرار دارد؛ بنابراین به دلیل عدم قطعیت کامل، احتمال حضور کلمه در اسناد مرتبط برابر با شانس تصادفی در نظر گرفته می‌شود: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0.5</span>.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام 
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: جایگذاری فرض عدم قطعیت (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">p<sub>i</sub> = 0.5</span>) در فرمول اصلی</b><br>
        مقدار مفروض برای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> را در فرمول وزن‌دهی جایگذاری می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{0.5(1 - u_i)}{u_i(1 - 0.5)}}$$
        </div>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: ساده‌سازی جبری کسر</b><br>
        از آنجا که عبارت <b>(1 - 0.5)</b> در مخرج برابر با <b>0.5</b> است، ضریب <b>0.5</b> در صورت و مخرج با یکدیگر ساده شده و حذف می‌شوند:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{1 - u_i}{u_i}}$$
        </div>
        تا این مرحله، وزن کلمه تنها به احتمال رخداد آن در اسناد غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>) وابسته شده است.
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام سوم: جایگذاری تقریب آماری برای <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>i</sub></span></b><br>
        همان‌طور که در مسائل پیشین بررسی شد، در یک پایگاه داده بزرگ، تعداد اسناد مرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span>) نسبت به کل اسناد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span>) بسیار ناچیز است. بنابراین، احتمال حضور کلمه در اسناد غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>) با تقریب بسیار بالایی معادل احتمال حضور کلمه در کل مجموعه اسناد در نظر گرفته می‌شود. اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub></span> فراوانی سند برای کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> باشد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{u_i \approx \frac{df_i}{N}}$$
        </div>
        این تقریب را در معادله به دست آمده از گام دوم جایگذاری می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{1 - \frac{df_i}{N}}{\frac{df_i}{N}}}$$
        </div>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام چهارم: مخرج مشترک‌گیری و ساده‌سازی</b><br>
        در صورت کسر، مخرج مشترک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span> را در نظر می‌گیریم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{\frac{N - df_i}{N}}{\frac{df_i}{N}}}$$
        </div>
        با ساده کردن مخرج‌های <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span> در صورت و مخرج کسر اصلی، به رابطه زیر می‌رسیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{N - df_i}{df_i}}$$
        </div>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام پنجم: اعمال فرض نادر بودن کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Sparsity)</span></b><br>
        در زبان طبیعی، اکثر کلمات کلیدی تنها در بخش کوچکی از کل اسناد ظاهر می‌شوند. به عبارت دیگر، فراوانی سند برای یک کلمه معمولاً بسیار کوچک‌تر از اندازه کل مجموعه است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub> &ll; N</span>). بر این اساس، کم کردن مقدار کوچک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub></span> از عدد بسیار بزرگ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N</span> در صورت کسر، تغییر معناداری در مقدار آن ایجاد نمی‌کند و می‌توان از آن صرف‌نظر کرد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{N - df_i \approx N}$$
        </div>
        با اعمال این تقریب نهایی در معادله، خواهیم داشت:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_i = \log \frac{N}{df_i}}$$
        </div>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
از طریق معادلات جبری اثبات شد که در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، اگر فرض کنیم هیچ اطلاعات اولیه‌ای درباره اسناد مرتبط نداریم (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0.5</span>) و کلمات کلیدی نسبت به کل مجموعه اسناد پراکنده و نادر هستند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>i</sub> &ll; N</span>)، تابع وزن‌دهی احتمالی دقیقاً به فرمول کلاسیک <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">IDF = log(N / df<sub>i</sub>)</span></b> همگرا می‌شود. این اثبات، استحکام نظری و ریشه‌های آماریِ مفهوم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> را در علم بازیابی اطلاعات تایید می‌کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.13</span></b>: بازخورد شبه‌ارتباط <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Pseudo-relevance Feedback)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌های احتمالاتی بازیابی اطلاعات مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، دستیابی به بالاترین سطح دقت نیازمند تخمین درست متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> (احتمال حضور کلمه در اسناد مرتبط) است. محاسبه دقیق این متغیر به صورت سنتی نیازمند «بازخورد صریح» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Explicit Feedback)</span> کاربر است تا مشخص کند کدام اسناد واقعاً مرتبط هستند. اما از آنجا که کاربران در دنیای واقعی تمایل چندانی به صرف زمان برای ارائه بازخورد ندارند، سیستم‌های بازیابی از رویکردی هوشمندانه به نام «بازخورد شبه‌ارتباط» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Pseudo-relevance Feedback یا Blind Relevance Feedback)</span> استفاده می‌کنند. در این روش، سیستم فرآیند بازخورد را بدون نیاز به دخالت انسان و به‌صورت کاملاً خودکار شبیه‌سازی می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل رویکرد و الگوریتم
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
ایده‌ی بنیادین بازخورد شبه-ارتباط بر یک «فرض خوش‌بینانه» استوار است: سیستم فرض می‌کند که الگوریتم رتبه‌بندی اولیه‌اش به قدری خوب کار می‌کند که چند سندِ اولِ بازیابی‌شده، به احتمال بسیار زیاد با نیاز کاربر مرتبط هستند. بر اساس این فرض، الگوریتم طی <b>4</b> مرحله متوالی اجرا می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) مراحل اجرای بازخورد شبه ارتباط
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: جستجوی اولیه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Initial Retrieval)</span></b><br>
        هنگامی که کاربر پرسمان خود را وارد می‌کند، سیستم هیچ اطلاعاتی از اسناد مرتبط ندارد. بنابراین، همان‌طور که در مباحث پیشین ثابت شد، با در نظر گرفتن فرض عدم قطعیت (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0.5</span>) و استفاده از تقریب‌های <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span>، یک جستجوی اولیه روی کل پایگاه داده انجام می‌دهد و اسناد را رتبه‌بندی می‌کند.
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: اعمال فرض شبه‌ارتباط <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(The Pseudo-relevance Assumption)</span></b><br>
        سیستم به جای اینکه منتظر کلیک یا تایید کاربر بماند، تعداد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k</span> سندِ برتر از نتایج اولیه (به عنوان مثال <b>10</b> یا <b>20</b> سند اول) را جدا می‌کند. سپس به صورت خودکار (و کورکورانه) برچسب «مرتبط» (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R = 1</span>) را به تمام این <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k</span> سند اختصاص می‌دهد. ما این مجموعه اسنادِ فرضاً مرتبط را مجموعه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">V</span> می‌نامیم (بنابراین کل اسناد مرتبط <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S = k</span> خواهد بود).
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام سوم: استخراج و به‌روزرسانی پارامترها <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Parameter Updating)</span></b><br>
        اکنون سیستم یک مجموعه محلی از اسناد مرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">V</span>) در اختیار دارد و می‌تواند از حالت پیش‌فرضِ حدسی خارج شود. برای هر کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">i</span> در پرسمان، سیستم بررسی می‌کند که این کلمه در چند مورد از این <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k</span> سند برتر تکرار شده است (این مقدار را <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span> می‌نامیم). سپس با استفاده از فرمول هموارسازی نیم‌واحدی، مقدار واقعی‌تر و دقیق‌تری برای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> محاسبه می‌کند:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_i = \frac{s + 0.5}{k + 1}}$$
        </div>
        <span style="font-size:14pt; color:#666666;">سیستم می‌تواند با استفاده از کلمات جدیدی که در این <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k</span> سند پرکاربرد بوده‌اند، پرسمان را نیز بسط دهد که به آن <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Query Expansion</span> می‌گویند.</span>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام چهارم: رتبه‌بندی مجدد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Re-ranking)</span></b><br>
        با به دست آمدن مقادیر جدید و به‌روزرسانی‌شده‌ی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> (و در صورت نیاز به‌روزرسانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>)، وزن تمام کلماتِ پرسمان در فرمول اصلی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RSV</span> مجدداً محاسبه می‌شود. سیستم یک بار دیگر اسناد را با این اوزان جدید امتیازدهی کرده، رتبه‌بندی نهایی را شکل می‌دهد و این نتایج اصلاح‌شده را به کاربر نمایش می‌دهد. تمام این <b>4</b> گام در کسری از ثانیه و در پس‌زمینه رخ می‌دهد.
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
بازخورد شبه‌ارتباط به سیستم‌های بازیابی احتمالی اجازه می‌دهد تا با استفاده از نتایج اولیه خود به عنوان یک «مجموعه آموزشی موقت»، پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> را از یک مقدار ثابت و بی‌اثر (<b>0.5</b>) به یک مقدار پویا و وابسته به محتوای جستجو تبدیل کنند، بدون اینکه هیچ بارِ شناختی یا زحمتی به کاربر تحمیل شود.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
اگرچه این روش معمولاً باعث افزایش چشمگیر نرخ فراخوانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Recall)</span> و دقت سیستم می‌شود، اما دارای یک ریسک تئوریک بزرگ به نام <b>«انحراف پرسمان» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query Drift)</span></b> است. اگر پرسمان کاربر بسیار مبهم باشد و سیستم در گام اول (جستجوی اولیه) کاملاً اشتباه کند و اسناد نامرتبطی را در صدر نتایج بیاورد، سیستم در گام دوم آن اسناد نامرتبط را به عنوان پایه محاسبات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> قرار می‌دهد. این کار باعث می‌شود در رتبه‌بندی مجدد، سیستم با اطمینان بیشتری به سمت یک موضوع کاملاً اشتباه منحرف شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.14</span></b>: محدودیت‌های مدل‌های احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Limitations of Probabilistic Models)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> با وجود پایه‌گذاری یک چارچوب مستحکم ریاضی برای بازیابی اطلاعات، به دلیل فرضیات ساده‌کننده‌اش در بازنمایی متن، دارای نقاط کور قابل‌توجهی است. این ساده‌سازی‌ها اگرچه پیچیدگی محاسباتی را در زمان جستجو به شدت کاهش می‌دهند، اما باعث از دست رفتن بخش مهمی از سیگنال‌های معنایی و ساختاری متن می‌شوند که مستقیماً بر کیفیت و دقت خروجی تأثیر منفی می‌گذارند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک چرایی عدم کارایی مدل پایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> در سیستم‌های مدرن و محیط‌های واقعی، <b>2</b> مورد از بنیادی‌ترین ضعف‌های این مدل را که ریشه در ذات «باینری» بودن آن دارند، کالبدشکافی کرده و اثرات مخرب آن‌ها را بر تجربه جستجوی کاربر بررسی می‌کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تحلیل محدودیت اول - عدم استفاده از بسامد کلمه درون سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Frequency - TF)</span></b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ماهیت ضعف:</b> در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، بازنمایی اسناد صرفاً به صورت متغیرهای دودویی (حضور با مقدار <b>1</b> یا عدم حضور با مقدار <b>0</b>) انجام می‌شود. این مدل هیچ مکانیزمی برای شمارش دفعات تکرار یک کلمه در متن ندارد.</li>
            <li style="margin-bottom:8px;"><b>اثر بر کیفیت جستجو:</b> در دنیای واقعی، میزان تکرار یک کلمه، شاخصی قدرتمند از میزان تمرکز سند بر روی آن موضوع است. فرض کنید کاربری عبارت «درخت جستجوی دودویی» را جستجو می‌کند. سند اول یک فصل از کتاب ساختمان داده است که این عبارت <b>100</b> بار در آن تکرار شده است، در حالی که سند دوم یک مقاله عمومی است که تنها در چکیده خود <b>1</b> بار به این عبارت اشاره کرده است. مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> به دلیل ماهیت صفر و یکی خود، به هر <b>2</b> سند امتیاز برابری برای این کلمه اختصاص می‌دهد و نمی‌تواند سند تخصصی‌تر را تشخیص دهد. این ضعف باعث کاهش شدید <b>دقت <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Precision)</span></b> در نتایج برتر می‌شود، زیرا اسناد حاشیه‌ای هم‌رده با اسناد مرجع و اصلی قرار می‌گیرند.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام دوم: تحلیل محدودیت دوم - نادیده گرفتن طول سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Document Length Normalization)</span></b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ماهیت ضعف:</b> مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> به طول کلی سند (تعداد کل کلمات تشکیل‌دهنده آن) کاملاً بی‌تفاوت است. یک پاراگراف کوتاه <b>30</b> کلمه‌ای و یک گزارش مفصل <b>30000</b> کلمه‌ای، در صورتی که کلمات پرسمان را داشته باشند، در سیستم یکسان ارزیابی می‌شوند.</li>
            <li style="margin-bottom:8px;"><b>اثر بر کیفیت جستجو:</b> اسناد طولانی‌تر به طور طبیعی (و صرفاً به دلیل حجم بالای کلمات) احتمال بیشتری دارند که شامل دایره واژگان وسیع‌تری باشند. در نتیجه، در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> شانس اینکه یک سند طولانی به طور تصادفی با کلمات پرسمان مطابقت پیدا کند بسیار بالاست. این سوگیری ساختاری باعث می‌شود موتور جستجو نتایج را به نفع اسناد طولانی (که ممکن است شامل موضوعات پراکنده و نامرتبط باشند) مصادره کند و اسناد کوتاه، دقیق و بسیار مرتبط (مانند پرسش و پاسخ‌های متمرکز) را در رتبه‌های پایین‌تر قرار دهد. این امر موجب افت شدید کیفیت رتبه‌بندی می‌گردد.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
محدودیت‌های بنیادین مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> در نادیده گرفتن بسامد کلمات و بی‌توجهی به طول اسناد، باعث می‌شوند سیستم توانایی تمایز قائل شدن بین اسناد عمیقاً مرتبط و اسنادی که صرفاً اشاره‌ای گذرا یا تصادفی به موضوع داشته‌اند را از دست بدهد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
دقیقاً همین <b>2</b> نقطه ضعف بزرگ، انگیزه اصلی پژوهشگران برای توسعه مدل‌های پیشرفته‌تر گردید. با ادغام منطق بسامد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(TF)</span> برای برطرف کردن محدودیت اول، و معرفی ضریب «نرمال‌سازی طول سند» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Document Length Normalization)</span> برای جبران سوگیری در محدودیت دوم، سیستم‌های رتبه‌بندی جهش بزرگی را تجربه کردند. خروجی نهاییِ این اصلاحات ریاضی، تولد الگوریتم قدرتمند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> بود که امروزه به عنوان هسته استاندارد در موتورهای جستجوی متن‌باز نظیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Elasticsearch</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Lucene</span> استفاده می‌شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.15</span></b>: تفاوت بازخورد احتمالی و برداری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Probabilistic vs. Vector Space Feedback)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مفهوم «بازخورد ارتباط» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Relevance Feedback)</span> در سیستم‌های بازیابی اطلاعات، به معنای استفاده از بازخورد کاربر (یا بازخورد شبه‌ارتباط) برای اصلاح و بهبود پرسمان اولیه است. اگرچه هدف نهایی در هر <b>2</b> مدل «احتمالاتی» (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>) و «فضای برداری» (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span>) یکسان است (یعنی نزدیک کردن نتایج به نیاز واقعی کاربر)، اما این <b>2</b> مدل از نظر پارادایم ریاضی و نحوه اعمال این بازخورد تفاوت‌های بنیادینی با یکدیگر دارند. یکی بر پایه آمار و توزیع احتمالات بنا شده و دیگری بر پایه هندسه و فاصله‌یابی در فضای چندبعدی.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل مسئله و متغیرها
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک این تفاوت، باید رفتار ریاضی هر مدل را هنگام دریافت اسناد مرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>r</sub></span>) و اسناد غیرمرتبط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>nr</sub></span>) بررسی کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;">در مدل احتمالی، متغیر هدف ما <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> (احتمال رخداد کلمه در اسناد مرتبط) است.</li>
    <li style="margin-bottom:10px;">در مدل فضای برداری، ابزار اصلی ما الگوریتم <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Rocchio</span></b> است و متغیر هدف، بردار پرسمان اصلاح‌شده (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\vec{q}_m\)</span>) می‌باشد.</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: مکانیزم بازخورد در مدل احتمالی (تغییر باور آماری)</b><br>
        در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، پرسمان به عنوان یک موجودیت یکپارچه حرکت نمی‌کند. در عوض، سیستم به صورت کلمه‌به‌کلمه عمل می‌کند. هنگامی که اسناد مرتبط مشخص می‌شوند، سیستم صرفاً توزیع آماری کلمات را بازنگری می‌کند.
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_i = \frac{s + 0.5}{S + 1}}$$
        </div>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ماهیت عملکرد:</b> این روش یک «به‌روزرسانی بیزی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bayesian Update)</span> است. سیستم پیش از بازخورد، باور داشت که اهمیت کلمه تصادفی است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub> = 0.5</span>). پس از بازخورد، با شمارش کلمات در اسناد مرتبط، <b>وزن <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Weight)</span></b> آن کلمه را در فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RSV</span> تغییر می‌دهد. در حالت پایه، کلمات جدیدی به پرسمان اضافه نمی‌شوند، بلکه فقط اهمیت کلمات موجود بازتوزیع می‌شود.</li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: مکانیزم بازخورد در مدل فضای برداری (انتقال هندسی)</b><br>
        در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span>، اسناد و پرسمان به صورت بردارهایی در یک فضای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n</span>-بعدی (که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n</span> اندازه لغت‌نامه است) نشان داده می‌شوند. بازخورد ارتباط با استفاده از الگوریتم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Rocchio</span> اعمال می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\vec{q}_m = \alpha \vec{q}_0 + \beta \frac{1}{|D_r|} \sum_{\vec{d}_j \in D_r} \vec{d}_j - \gamma \frac{1}{|D_{nr}|} \sum_{\vec{d}_k \in D_{nr}} \vec{d}_k}$$
        </div>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ماهیت عملکرد:</b> این روش یک «انتقال فضایی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Spatial Translation)</span> است. بردار پرسمان اولیه (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\vec{q}_0\)</span>) به صورت فیزیکی در فضای مختصات حرکت می‌کند؛ به سمت مرکز ثقل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Centroid)</span> اسناد مرتبط کشیده می‌شود و از مرکز ثقل اسناد غیرمرتبط دور می‌گردد. </li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: تبیین تفاوت‌های بنیادی</b><br>
        از مقایسه این <b>2</b> مکانیزم، <b>3</b> تفاوت اساسی استخراج می‌شود:
        <ul style="margin-right:20px; padding-right:15px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:10px;"><b style="color:#6D2E46;">1. تفاوت در ماهیت ریاضی:</b> بازخورد احتمالی مبتنی بر <b>آمار و شمارش گسسته</b> (تغییر احتمالات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span>) است، در حالی که بازخورد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span> مبتنی بر <b>جبر خطی و هندسه پیوسته</b> (تغییر زاویه و طول بردار) است.</li>
            <li style="margin-bottom:10px;"><b style="color:#6D2E46;">2. تفاوت در بسط پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query Expansion)</span>:</b> در الگوریتم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Rocchio</span>، جمع کردن بردارهای اسناد مرتبط به صورت خودکار باعث می‌شود کلمات جدیدی که در پرسمان اولیه نبوده‌اند، وزن غیرصفر بگیرند (بسط خودکار پرسمان). اما در مدل احتمالیِ پایه، بازخورد تنها وزن کلمات فعلی را اصلاح می‌کند و برای اضافه کردن کلمات جدید به پرسمان، به یک مرحله مجزا و صریح برای انتخاب کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Term Selection)</span> نیاز است.</li>
            <li style="margin-bottom:10px;"><b style="color:#6D2E46;">3. تفاوت در اثر اسناد منفی:</b> در الگوریتم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Rocchio</span>، اسناد غیرمرتبط مستقیماً (با ضریب <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\gamma\)</span>) بردار پرسمان را از خود دور می‌کنند که می‌تواند باعث منفی شدن وزن برخی کلمات شود. اما در مدل احتمالی، اسناد غیرمرتبط صرفاً با افزایش مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>، وزن مثبتِ یک کلمه را کاهش می‌دهند و مفهوم «وزن منفی» در آن به شکل هندسیِ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span> وجود ندارد.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
تفاوت بنیادی این <b>2</b> رویکرد در این است که مدل احتمالی، بازخورد را به عنوان شواهدی برای <b>«یادگیری و اصلاح توزیع احتمالات کلمات»</b> می‌بیند تا فرمول امتیازدهی را دقیق‌تر کند، در حالی که الگوریتم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Rocchio</span> در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span>، بازخورد را به عنوان نیرویی برای <b>«هدایت فیزیکی بردار پرسمان»</b> به سمت مناطق غنی‌تر و مرتبط‌تر در فضای چندبعدی تفسیر می‌کند.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
از منظر مهندسی سیستم، الگوریتم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Rocchio</span> به دلیل جمع کردن بردارهای طولانی اسناد، پرسمان اصلاح‌شده را به یک «بردار چگال» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Dense Vector)</span> با کلمات بسیار زیاد تبدیل می‌کند که زمان پردازش جستجو را به شدت افزایش می‌دهد. به همین دلیل، در سیستم‌های واقعی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span>، پس از محاسبه بردار جدید <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\vec{q}_m\)</span>، معمولاً فقط <b>10</b> الی <b>20</b> کلمه‌ای که بالاترین وزن را دارند نگه داشته می‌شوند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Truncation)</span> تا جستجو همچنان سریع و کارآمد باقی بماند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.16</span></b>: وابستگی کلمات در پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Term Dependencies in Queries)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> بر پایه فرض «استقلال شرطی کلمات» بنا شده است. این فرض تئوریک بیان می‌کند که با مشخص بودن وضعیت مرتبط بودن یا نبودن یک سند، احتمال حضور یک کلمه در آن سند هیچ ارتباطی به حضور یا عدم حضور کلمات دیگر ندارد. در زبان طبیعی، این فرض به ندرت صادق است؛ زیرا زبان مملو از کلمات مرکب، عبارات اسمی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Named Entities)</span> و هم‌رخدادی‌های معنادار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Collocations)</span> است که در آن‌ها کلمات به شدت به یکدیگر وابسته‌اند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"New York"</span> را در نظر بگیرید. این پرسمان از <b>2</b> کلمه مجزا تشکیل شده است: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub> = "New"</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>2</sub> = "York"</span>. هدف کاربر یافتن اسنادی درباره شهر یا ایالت «نیویورک» است. اکنون بررسی می‌کنیم که چرا پردازش این پرسمانِ به ظاهر ساده، در ساختار مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> منجر به فروپاشی منطقِ رتبه‌بندی می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: رویکرد پردازشی مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span> (نگاه مستقل)</b><br>
        از نگاه ریاضیاتِ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، احتمال مشاهده همزمان این <b>2</b> کلمه در یک سند مرتبط، برابر با حاصل‌ضرب احتمال مشاهده مستقل آن‌هاست:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(t_1, t_2 | R=1) = P(t_1 | R=1) \times P(t_2 | R=1)}$$
        </div>
        بر همین اساس، در محاسبه نمره نهایی سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RSV)</span>، سیستم صرفاً وزن مستقل هر کلمه را محاسبه کرده و با هم جمع می‌کند:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{RSV_d = c_1 + c_2}$$
        </div>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: نقض واقعیت آماری (بروز خطای محاسباتی)</b><br>
        در دنیای واقعی و در متون انگلیسی، این <b>2</b> کلمه مستقل نیستند. اگر کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"York"</span> در متنی درباره جغرافیا یا اخبار آمریکا ظاهر شود، با احتمال بسیار بالایی کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"New"</span> نیز بلافاصله قبل از آن وجود دارد. به عبارت دیگر، احتمال شرطیِ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P("New" | "York")</span> بسیار بالا و نزدیک به <b>1</b> است.<br>
        مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> با نادیده گرفتن این وابستگی آماری شدید، دچار خطای <b>«شمارش مضاعف شواهد» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Double Counting of Evidence)</span></b> می‌شود. سیستم گمان می‌کند <b>2</b> شاهد کاملاً مستقل و مجزا برای مرتبط بودن سند پیدا کرده است، در حالی که در واقعیت، این <b>2</b> کلمه تنها یک مفهوم واحد (یک شهر) را نمایندگی می‌کنند.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: آسیب‌های وارد شده به کیفیت رتبه‌بندی</b><br>
        این خطای محاسباتی، <b>2</b> آسیب جدی به خروجی موتور جستجو وارد می‌کند:
        <ul style="margin-right:20px; padding-right:15px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> <b>ناتوانی در تشخیص ساختار عبارات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Loss of Phrase Structure)</span>:</b> چون مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> کلمات را مستقل فرض می‌کند، موقعیت و ترتیب کلمات در سند را از دست می‌دهد (تبدیل شدن به یک کیسه کلمات یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Bag of Words</span>). به عنوان مثال، سندی با محتوای: <i>"دانشگاه <b>York</b> یک روش <b>New</b> برای تدریس ارائه کرد"</i>، دقیقاً همان نمره‌ای را دریافت می‌کند که سندی با محتوای <i>"شهر <b>New York</b> دارای آسمان‌خراش است"</i> دریافت می‌کند. سیستم نمی‌تواند بین کلمات پراکنده و یک عبارت متصل تمایز قائل شود.</li>
            <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> <b>تولید مثبت کاذب <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(False Positives)</span> و افت دقت:</b> با جمع شدن وزن کلمات به صورت مستقل، اسنادی که به طور تصادفی شامل کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"New"</span> (که کلمه بسیار پرکاربردی است) و کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"York"</span> (در سیاقی دیگر) هستند، نمره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RSV)</span> بالایی دریافت کرده و به اشتباه در صدر نتایج (رتبه‌های بالا) قرار می‌گیرند. این امر باعث می‌شود کاربر با حجم زیادی از اسناد نامرتبط مواجه شود.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> در پردازش پرسمان‌هایی نظیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"New York"</span> دچار خطای محاسباتی می‌شود زیرا فرض استقلال، وابستگی آماری و معنایی شدید میان اجزای یک عبارت مرکب را نادیده می‌گیرد. این امر منجر به شمارش مضاعف شواهد، از دست رفتن ساختار عبارات و در نهایت نمایش اسناد نامرتبط در رتبه‌های برتر (افت <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Precision</span>) می‌گردد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
برای غلبه بر این محدودیتِ بزرگ، سیستم‌های بازیابی اطلاعات مدرن از معماری‌های پیشرفته‌تری استفاده می‌کنند. یکی از راهکارهای رایج، استفاده از «نمایه‌های موقعیتی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Positional Indexes)</span> و پشتیبانی از «پرسمان‌های عبارتی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Phrase Queries)</span> است (مثلاً قرار دادن پرسمان در داخل کوتیشن: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"New York"</span>). راهکار پیچیده‌تر، استفاده از مدل‌های زبانی مبتنی بر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N-gram</span> (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Bigram</span>ها) یا مدل‌های وابستگی کلمات (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Markov Random Fields</span> در بازیابی اطلاعات) است که در آن‌ها، احتمال رخداد یک کلمه به کلمه قبل از خود وابسته <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Conditioned)</span> می‌شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.17</span></b>: رتبه‌بندی با لحاظ هزینه در سیستم پزشکی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cost-Aware Ranking in Medical Systems)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در سیستم‌های اطلاعاتی حساس (مانند پزشکی یا حقوقی)، اصل رتبه‌بندی احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(PRP)</span> با در نظر گرفتن «تابع زیان» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Loss Function)</span> تعمیم می‌یابد. در این حالت، تصمیم‌گیری برای نمایش یا عدم نمایش یک سند، صرفاً بر اساس مرتب‌سازی احتمالات نیست؛ بلکه سیستم بر اساس تئوری تصمیم بیز <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bayesian Decision Theory)</span>، «ارزش مورد انتظار» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Expected Value)</span> خطاها را محاسبه می‌کند. سیستم تنها زمانی یک سند را به کاربر نشان می‌دهد که هزینه مورد انتظار ناشی از بازیابی آن، کمتر یا مساوی با هزینه مورد انتظار ناشی از عدم بازیابی آن باشد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و اصلاح فرضیات
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
پیش از حل مسئله، نیاز است تا یک تناقض در صورت سوال را از منظر ریاضی اصلاح کنیم. در متن اشاره شده است که «هزینه از دست دادن یک سند مرتبط (منفی کاذب) <b>10</b> برابر هزینه نشان دادن سند نامرتبط (مثبت کاذب) است».<br>
همان‌طور که در مسئله <b>11.4</b> تعریف کردیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;">منفی کاذب = <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>01</sub></span></li>
    <li style="margin-bottom:10px;">مثبت کاذب = <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>10</sub></span></li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
بنابراین، ترجمه دقیق ریاضیِ متن صورت مسئله به شکل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>01</sub> = 10 &times; C<sub>10</sub></span> است. رابطه داخل پرانتز در صورت سوال (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>10</sub> = 10 &times; C<sub>01</sub></span>) یک خطای تایپی رایج است که مفهوم را معکوس می‌کند. در این پاسخنامه، بر اساس منطق و حساسیت سیستم پزشکی (اولویت جان بیمار بر اتلاف وقت پزشک)، ما محاسبات را بر پایه فرمول صحیح یعنی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>01</sub> = 10 &times; C<sub>10</sub></span> پیش خواهیم برد.<br>
همچنین فرض می‌کنیم تصمیمات درست (بازیابی سند مرتبط و عدم بازیابی سند نامرتبط) هیچ هزینه‌ای ندارند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>11</sub> = 0</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>00</sub> = 0</span>).
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: فرمول‌بندی هزینه‌های مورد انتظار</b><br>
        برای هر سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span>)، سیستم <b>2</b> گزینه پیش رو دارد و باید هزینه مورد انتظار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Expected Cost)</span> هر کدام را محاسبه کند. برای سادگی، احتمال مرتبط بودن سند را با <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p = P(R=1|d)</span> و احتمال نامرتبط بودن آن را با <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">1 - p = P(R=0|d)</span> نشان می‌دهیم.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:12px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> <b>هزینه مورد انتظار در صورت بازیابی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Retrieval)</span>:</b> اگر سیستم سند را نمایش دهد، تنها در صورتی جریمه می‌شود که سند نامرتبط باشد (مثبت کاذب).
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{E[Cost | Retrieval] = (1 - p) \times C_{10}}$$
                </div>
            </li>
            <li style="margin-bottom:8px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> <b>هزینه مورد انتظار در صورت عدم بازیابی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Non-Retrieval)</span>:</b> اگر سیستم سند را فیلتر کند، تنها در صورتی جریمه می‌شود که سند مرتبط بوده باشد (منفی کاذب).
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{E[Cost | Non\text{-}Retrieval] = p \times C_{01}}$$
                </div>
            </li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: اعمال شرط تصمیم‌گیری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">PRP</span></b><br>
        طبق اصل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">PRP</span>، سیستم تنها زمانی سند را بازیابی می‌کند که هزینه نمایش آن، از هزینه پنهان کردنش کمتر (یا مساوی) باشد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{E[Cost | Retrieval] \le E[Cost | Non\text{-}Retrieval]}$$
        </div>
        با جایگذاری فرمول‌های گام اول در این نامعادله خواهیم داشت:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{(1 - p) \times C_{10} \le p \times C_{01}}$$
        </div>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: حل نامعادله و یافتن آستانه احتمال</b><br>
        اکنون رابطه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>01</sub> = 10 &times; C<sub>10</sub></span> را در نامعادله جایگذاری می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{(1 - p) \times C_{10} \le p \times (10 \times C_{10})}$$
        </div>
        چون <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>10</sub></span> یک عدد مثبت است، می‌توانیم طرفین را بر آن تقسیم کنیم تا ساده شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{1 - p \le 10p}$$
        </div>
        متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p</span> را به سمت راست نامعادله منتقل می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{1 \le 11p}$$
        </div>
        در نهایت، با تقسیم طرفین بر <b>11</b>، آستانه احتمال (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p</span>) به دست می‌آید:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p \ge \frac{1}{11}}$$
        </div>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای اینکه یک سند در این سیستم پزشکی به پزشک نمایش داده شود، احتمال مرتبط بودن آن (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1|d)</span>) باید <b>بزرگتر یا مساوی 1/11 (یعنی حدود 0.09 یا 9 درصد)</b> باشد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این نتیجه از نظر مهندسیِ سیستم‌های حساس بسیار منطقی و آموزنده است. در یک موتور جستجوی وب معمولی، سیستم معمولاً اسنادی را نشان می‌دهد که احتمال ارتباط آن‌ها بالای <b>50</b> درصد باشد. اما در یک سیستم پزشکی، چون هزینه از دست دادن اطلاعات حیاتی (منفی کاذب) به شدت بالاست، سیستم بسیار محتاطانه عمل می‌کند و آستانه تصمیم‌گیری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Threshold)</span> را تا <b>9</b> درصد پایین می‌آورد. این یعنی سیستم به پزشک می‌گوید: <i>"من تنها <b>9</b> درصد مطمئنم که این پرونده به بیمار شما مرتبط است، اما چون نخواندن آن ممکن است خطرناک باشد، آن را به شما نمایش می‌دهم تا خودتان بررسی کنید."</i> این استراتژی به شدت به نفع افزایش نرخ فراخوانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Recall)</span> عمل می‌کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.18</span></b>: سناریوی تکامل وزن‌دهی کلمات با بازخورد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Term Weight Evolution Scenario with Feedback)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
این مسئله، یک جمع‌بندی عملی از مباحث پیشین است و نشان می‌دهد که چگونه «بازخورد ارتباط» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Relevance Feedback)</span> به صورت ریاضی در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> اعمال می‌شود. در اینجا مشاهده می‌کنیم که وزن یک کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RSV)</span> یک مقدار ثابت نیست، بلکه با دریافت اطلاعات جدید از سمت کاربر، از طریق یک فرایند «به‌روزرسانی بیزی» تکامل یافته و اصلاح می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
ابتدا مقادیر داده شده در صورت مسئله را با نمادهای استاندارد مدل مدل‌سازی می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">اندازه کل مجموعه اسناد پایگاه داده:</b> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N = 1000</span></li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">فراوانی سند برای کلمه هدف:</b> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">df<sub>1</sub> = 100</span></li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">تعداد کل اسناد مرتبط شناخته‌شده پس از بازخورد:</b> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S = 3</span></li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">تعداد اسناد مرتبطِ حاوی کلمه هدف:</b> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s = 2</span></li>
</ul>
</div>

<div style="font-size:14pt; color:#666666; margin-top:10px; text-align:justify;">
<i>(نکته: تعداد <b>5</b> سندی که در دور اول برگردانده شده‌اند، در فرمول‌های پایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> تاثیری ندارند، زیرا در این مدل تنها اسناد مرتبط (تایید شده توسط کاربر) و داده‌های سراسری مجموعه برای تخمین احتمالات اهمیت دارند).</i>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:17pt; font-weight:bold; color:#1B263B; margin:20px 0 14px 0; background-color:#F0F0F5; padding:8px 15px; border-radius:6px; display:inline-block;">
الف) محاسبه وزن کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> قبل از بازخورد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Iteration 0)</span>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
پیش از دریافت بازخورد کاربر، سیستم هیچ دانشی درباره اسناد مرتبط ندارد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S = 0</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s = 0</span>). همان‌طور که در مسئله‌های پیشین اثبات شد، در این حالت از فرض عدم قطعیت (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>1</sub> = 0.5</span>) و تقریب سراسری (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>1</sub> = df<sub>1</sub> / N</span>) استفاده می‌شود.
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: جایگذاری مقادیر پیش‌فرض در فرمول وزن‌دهی</b><br>
        فرمول پایه وزن کلمه به این شکل ساده می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_1 = \log \frac{1 - u_1}{u_1}}$$
        </div>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: محاسبه مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>1</sub></span> و وزن نهایی</b><br>
        مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>1</sub></span> برابر است با <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">100 / 1000 = 0.1</span>. این مقدار را در فرمول قرار می‌دهیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_1 = \log \frac{1 - 0.1}{0.1} = \log \frac{0.9}{0.1} = \log(9)}$$
        </div>
        وزن کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> قبل از بازخورد برابر با <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">log(9)</span></b> است.
    </li>
</ul>
</div>

<div style="font-size:17pt; font-weight:bold; color:#1B263B; margin:30px 0 14px 0; background-color:#F0F0F5; padding:8px 15px; border-radius:6px; display:inline-block;">
ب) محاسبه وزن کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> پس از دریافت بازخورد
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
اکنون کاربر بازخورد داده است و مقادیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S = 3</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s = 2</span> در دسترس هستند. برای محاسبه دقیق و جلوگیری از خطای احتمالات <b>0</b>، از تکنیک «هموارسازی نیم‌واحدی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Robertson-Sparck Jones)</span> استفاده می‌کنیم.
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام اول: به‌روزرسانی احتمال <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">p<sub>1</sub></span> (حضور در اسناد مرتبط)</b>
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{p_1 = \frac{s + 0.5}{S + 1} = \frac{2 + 0.5}{3 + 1} = \frac{2.5}{4} = 0.625}$$
        </div>
        <span style="font-size:14pt; color:#666666;"><i>(احتمال عدم حضور در اسناد مرتبط: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">1 - p<sub>1</sub> = 0.375</span>)</i></span>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام دوم: به‌روزرسانی احتمال <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">u<sub>1</sub></span> (حضور در اسناد غیرمرتبط)</b>
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{u_1 = \frac{df_1 - s + 0.5}{N - S + 1} = \frac{100 - 2 + 0.5}{1000 - 3 + 1} = \frac{98.5}{998} \approx 0.0987}$$
        </div>
        <span style="font-size:14pt; color:#666666;"><i>(احتمال عدم حضور در اسناد غیرمرتبط: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">1 - u<sub>1</sub> &approx; 0.9013</span>)</i></span>
    </li>
    <li style="margin-bottom:18px;">
        <b style="color:#6D2E46;">گام سوم: محاسبه وزن جدید (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">RSV</span> جدید)</b><br>
        اکنون احتمالات به‌روزرسانی‌شده را در فرمول اصلی قرار می‌دهیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{c_1 = \log \frac{p_1(1 - u_1)}{u_1(1 - p_1)} = \log \frac{0.625 \times 0.9013}{0.0987 \times 0.375} = \log \frac{0.5633}{0.0370} \approx \log(15.22)}$$
        </div>
        وزن کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> پس از بازخورد برابر با <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">log(15.22)</span></b> است.
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی و تحلیل تغییرات
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">قبل از بازخورد:</b> نسبت شانس (آرگومان درون لگاریتم) برابر با <b>9</b> بود.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">پس از بازخورد:</b> نسبت شانس به <b>15.22</b> افزایش یافت.</li>
</ul>
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
افزایش وزن کلمه از <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">log(9)</span> به <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">log(15.22)</span> از نظر آماری کاملاً معنادار است. پیش از جستجو، سیستم تنها می‌دانست که این کلمه در <b>10</b> درصد از کل اسناد پایگاه داده (<b>100</b> از <b>1000</b>) وجود دارد. اما پس از بازخورد، سیستم متوجه شد که این کلمه در حدود <b>66</b> درصد از اسناد مرتبطِ کاربر (<b>2</b> از <b>3</b>) حضور دارد. چون تمرکز این کلمه در اسناد مرتبط بسیار بیشتر از توزیع عمومی آن در کل پایگاه داده است، سیستم نتیجه می‌گیرد که کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> یک «کلمه متمایزکننده» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Discriminative Term)</span> بسیار عالی برای نیاز این کاربر است؛ بنابراین وزن آن را در رتبه‌بندی افزایش می‌دهد تا در جستجوی دور دوم، اسنادی که حاوی این کلمه هستند شانس بیشتری برای نمایش در رتبه‌های بالا داشته باشند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.19</span></b>: مدل‌سازی وابستگی درختی عبارات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Modeling Tree-Structured Dependencies)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
همان‌طور که در مسائل پیشین بررسی شد، «فرض استقلال کلمات» در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> یک نقطه‌ضعف بنیادین است. برای غلبه بر این مشکل، پژوهشگران (از جمله ون رایزبرگن) «مدل وابستگی درختی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Tree-Structured Dependencies)</span> را پیشنهاد دادند. در این مدل، به جای استقلال کامل کلمات یا وابستگی کامل تمام کلمات به یکدیگر (که از نظر محاسباتی غیرممکن است)، یک حد وسط در نظر گرفته می‌شود: هر کلمه می‌تواند حداکثر به <b>1</b> کلمه‌ی دیگر وابسته باشد. این ساختار، <b>1</b> گراف بدون دور یا «درخت» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Tree)</span> تشکیل می‌دهد که در آن گره‌ها همان کلمات و یال‌ها نشان‌دهنده وابستگی آماری هستند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و نمادگذاری
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
پرسمان کاربر دارای <b>3</b> کلمه است. آن‌ها را با نمادهای متغیر تصادفی نام‌گذاری می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{t_1}\)</span>: نمایانگر کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Machine"</span></li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{t_2}\)</span>: نمایانگر کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Learning"</span></li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{t_3}\)</span>: نمایانگر کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Python"</span></li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
طبق صورت مسئله، روابط وابستگی به این شکل تعریف شده‌اند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">1.</b> وابستگی شدید میان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>2</sub></span> (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Machine Learning</span> یک عبارت متصل است).</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">2.</b> وابستگی میان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>2</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>3</sub></span> (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Python</span> معمولاً در کنار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Learning</span> و در بستر یادگیری ماشین معنا پیدا می‌کند).</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
این ساختار به شکل یک زنجیره یا درخت خطی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub> &rarr; t<sub>2</sub> &rarr; t<sub>3</sub></span>) مدل‌سازی می‌شود. هدف، محاسبه احتمال توأم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Joint Probability)</span> حضور این <b>3</b> کلمه در یک سند مرتبط، یعنی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(t<sub>1</sub>, t<sub>2</sub>, t<sub>3</sub> | R=1)</span> است.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: بیان رابطه توأم بر اساس قانون زنجیره‌ای (بدون فرض درختی)</b><br>
        در تئوری احتمالات، قانون زنجیره‌ای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Chain Rule)</span> برای هر <b>3</b> رویداد به صورت دقیق و بدون هیچ فرضی به شکل زیر بسط داده می‌شود (با فرض شرطی بودن روی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R=1</span>):
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(t_1, t_2, t_3 | R=1) = P(t_1 | R=1) \times P(t_2 | t_1, R=1) \times P(t_3 | t_1, t_2, R=1)}$$
        </div>
        این فرمول کامل است، اما محاسبه جمله آخر (که کلمه سوم به هر <b>2</b> کلمه قبلی وابسته است) در پایگاه‌داده‌های بزرگ بسیار پیچیده و پرهزینه است.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: اعمال فرض مدل‌سازی درختی (ویژگی مارکوف)</b><br>
        مدل وابستگی درختی بیان می‌کند که هر گره تنها به گره «پدر» (والد) خود در درخت متصل است و از سایر گره‌های پیشین مستقل می‌باشد. با در نظر گرفتن ساختار خطی مسئله (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub> &rarr; t<sub>2</sub> &rarr; t<sub>3</sub></span>)، کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>3</sub></span> تنها به کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>2</sub></span> وابسته است و حضور کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>1</sub></span> اطلاعات اضافه‌ای درباره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>3</sub></span> به ما نمی‌دهد. به بیان ریاضی:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(t_3 | t_1, t_2, R=1) \approx P(t_3 | t_2, R=1)}$$
        </div>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: جایگذاری در معادله و فرمول‌بندی نهایی</b><br>
        اکنون تقریب به‌دست‌آمده در گام دوم را در قانون زنجیره‌ای اصلی (گام اول) جایگذاری می‌کنیم تا فرمول نهایی مدل درختی به دست آید:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(t_1, t_2, t_3 | R=1) \approx P(t_1 | R=1) \times P(t_2 | t_1, R=1) \times P(t_3 | t_2, R=1)}$$
        </div>
        <span style="font-size:14pt; color:#666666;"><i>(نکته: در گراف‌های درختیِ بدون جهت، انتخاب ریشه اختیاری است. اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>2</sub></span> را به عنوان ریشه درخت انتخاب می‌کردیم، فرمول به شکل کاملاً معادلِ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(t<sub>2</sub>|R=1) &times; P(t<sub>1</sub>|t<sub>2</sub>, R=1) &times; P(t<sub>3</sub>|t<sub>2</sub>, R=1)</span> نوشته می‌شد).</i></span>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
با استفاده از تقریب درختی، احتمال مشترک حضور این <b>3</b> کلمه در یک سند مرتبط به جای حاصل‌ضرب <b>3</b> احتمال مستقل، به صورت <b>حاصل‌ضرب احتمالات شرطی جفتی</b> مدل‌سازی می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:18px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{P(t_1, t_2, t_3 | R=1) \approx P(t_1 | R=1) \times P(t_2 | t_1, R=1) \times P(t_3 | t_2, R=1)}$$
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این روش از نظر ریاضی بسیار ظریف است. در این مدل، سیستم متوجه می‌شود که اگر کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Machine"</span> در سند وجود داشته باشد، احتمال ظاهر شدن کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Learning"</span> بسیار بیشتر از حالت عادی است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(t<sub>2</sub>|t<sub>1</sub>)</span> بزرگ است). در پیاده‌سازی‌های واقعیِ بازیابی اطلاعات، برای استخراج خودکار این درخت‌های وابستگی از داخل پایگاه داده، از ابزارهایی مانند «الگوریتم چو-لیو» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Chow-Liu Algorithm)</span> استفاده می‌شود که با محاسبه «اطلاعات متقابل» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Mutual Information)</span> بین تمام جفت‌کلمات، گراف وابستگی را به شکل <b>1</b> درخت پوشای کمینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(MST)</span> رسم می‌کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.20</span></b>: الگوریتم ساخت درخت چو-لیو <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Chow-Liu Tree Algorithm)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌سازی وابستگی کلمات، سیستم نمی‌تواند به صورت تصادفی یا دستی مشخص کند که کدام کلمات به یکدیگر وابسته‌اند. برای کشف خودکارِ این روابط از درون داده‌ها، به یک ابزار ریاضی نیاز است که میزان هم‌بستگی آماریِ جفت‌کلمات را اندازه‌گیری کند. الگوریتم «چو-لیو» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Chow-Liu Algorithm)</span> یک روش بنیادین در نظریه اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Information Theory)</span> است که بهترین و دقیق‌ترین تقریب درختی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Tree Approximation)</span> را برای یک توزیع احتمال توأمِ پیچیده و چندمتغیره پیدا می‌کند. این الگوریتم تضمین می‌کند که درختِ حاصل، کمترین میزان خطا (پراکندگی کولبک-لیبلر) را نسبت به توزیع واقعی کلمات در اسناد داشته باشد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل مسئله و مدل‌سازی گراف
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای حل این مسئله، سیستم مجموعه کلمات پرسمان را به عنوان گره‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Vertices)</span> در یک گراف در نظر می‌گیرد. هدف نهایی این است که از میان تمام حالت‌های ممکن برای اتصال این گره‌ها، درختی استخراج شود که یال‌های آن <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Edges)</span>، قوی‌ترین وابستگی‌های آماری را نمایندگی کنند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تعریف معیار وزن‌دهی یال‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Mutual Information)</span></b><br>
        سیستم برای اینکه بفهمد <b>2</b> کلمه (مثلاً <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">t<sub>j</sub></span>) چقدر به هم وابسته‌اند، از معیار «اطلاعات متقابل» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Mutual Information یا MI)</span> استفاده می‌کند. این معیار نشان می‌دهد که با دانستن وضعیت حضور یک کلمه، چقدر از عدم قطعیت ما درباره کلمه دیگر کاسته می‌شود.<br>
        برای هر جفت کلمه، سیستم مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">MI</span> را بر اساس توزیع احتمالات آن‌ها در مجموعه اسناد محاسبه کرده و آن را به عنوان «وزن یال» (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">W<sub>ij</sub></span>) در نظر می‌گیرد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{W_{ij} = I(t_i, t_j) = \sum_{x_i \in \{0,1\}} \sum_{x_j \in \{0,1\}} P(x_i, x_j) \log \frac{P(x_i, x_j)}{P(x_i)P(x_j)}}$$
        </div>
        <span style="font-size:14pt; color:#666666;"><i>دقت کنید در این فرمول، <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>j</sub></span> نشان‌دهنده حضور (<b>1</b>) یا عدم حضور (<b>0</b>) کلمات در اسناد هستند. اگر <b>2</b> کلمه کاملاً مستقل باشند، صورت و مخرج کسرِ درون لگاریتم برابر شده و وزن یال <b>0</b> می‌شود.</i></span>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: تشکیل گراف کامل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Complete Graph)</span></b><br>
        سیستم وزنِ اطلاعات متقابل را برای تمام جفت‌کلماتِ ممکن در پرسمان محاسبه می‌کند. اگر پرسمان دارای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n</span> کلمه باشد، یک «گراف کامل بدون جهت» با <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n</span> گره و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n(n-1)/2</span> یال تشکیل می‌شود که روی هر یال، مقدار محاسبه‌شده در گام اول به عنوان وزن قرار گرفته است.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام سوم: استخراج درخت پوشای بیشینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Maximum Spanning Tree)</span></b><br>
        گراف کامل شامل حلقه‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cycles)</span> است و یک ساختار درختی نیست. برای تبدیل آن به یک مدل تقریب درختی (طبق فرمول‌بندی‌های مسئله پیشین)، سیستم باید زیرگرافی را انتخاب کند که:
        <ul style="margin-right:20px; padding-right:15px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> تمام گره‌ها را شامل شود <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Spanning)</span>.</li>
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> هیچ حلقه‌ای نداشته باشد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Tree)</span>.</li>
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.3</b> مجموع وزن یال‌های آن بیشترین مقدار ممکن باشد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Maximum)</span>.</li>
        </ul>
        برای این کار، سیستم از الگوریتم‌های استاندارد گراف مانند الگوریتم کراسکال <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Kruskal)</span> یا پریم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Prim)</span> استفاده می‌کند. الگوریتم یال‌ها را به ترتیب نزولیِ وزنِ (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">MI</span>) مرتب کرده و قوی‌ترین یال‌ها را یکی پس از دیگری به درخت اضافه می‌کند، به شرطی که حلقه‌ای ایجاد نشود. این کار تا زمانی ادامه می‌یابد که دقیقاً <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">n-1</span> یال انتخاب شود.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام چهارم: استخراج توزیع احتمالی نهایی</b><br>
        پس از تشکیل درخت پوشای بیشینه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">MST</span>)، سیستم یک گره را به دلخواه به عنوان ریشه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Root)</span> انتخاب کرده و جهت یال‌ها را از ریشه به سمت برگ‌ها هدایت می‌کند. اکنون سیستم می‌داند که دقیقاً کدام کلمات قوی‌ترین وابستگی‌ها را دارند و می‌تواند فرمول قانون زنجیره‌ای را دقیقاً بر روی یال‌های همین درخت (همانند مسئله <b>11.19</b>) اعمال کند.
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
الگوریتم چو-لیو به سیستم بازیابی اجازه می‌دهد تا با محاسبه <b>اطلاعات متقابل <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Mutual Information)</span></b> به عنوان وزن یال‌ها و سپس یافتن <b>درخت پوشای بیشینه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(MST)</span></b>، قوی‌ترین وابستگی‌های دوبه‌دو میان کلمات را استخراج کند. درخت حاصل، بهترین ساختار ممکن برای مدل‌سازی وابستگی عبارات است که هم از خطای «شمارش مضاعف» جلوگیری می‌کند و هم از نظر محاسباتی بسیار بهینه‌تر از محاسبه وابستگیِ کاملِ تمام کلمات است.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
دلیل استفاده از درخت پوشای «بیشینه» (و نه کمینه) این است که ما می‌خواهیم یال‌هایی در مدل نهایی باقی بمانند که دارای <b>بیشترین اطلاعات متقابل</b> هستند (یعنی کلماتی که به شدت با هم در اسناد هم‌رخدادی دارند، مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Machine</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Learning</span>). اثبات ریاضی چو-لیو نشان می‌دهد که مجموع وزن یال‌ها در این درخت، مستقیماً با به حداقل رساندن خطای تقریب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(KL Divergence)</span> بین مدل درختی و توزیع احتمال واقعی کلمات در زبان طبیعی، در ارتباط است.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.21</span></b>: کالیبراسیون <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> برای متون حقوقی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BM25 Calibration for Legal Texts)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
تابع رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> یکی از موفق‌ترین و پرکاربردترین مدل‌های بازیابی اطلاعات است که محدودیت‌های مدل‌های پایه احتمالی (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>) را با معرفی <b>2</b> پارامتر تنظیم‌پذیر برطرف می‌کند. این <b>2</b> پارامتر عبارتند از:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> <b>پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub></span>:</b> کنترل‌کننده سرعت اشباع بسامد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Term Frequency Saturation)</span>.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> <b>پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">b</span>:</b> کنترل‌کننده میزان نرمال‌سازی طول سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Document Length Normalization)</span>.</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
در سیستم‌های واقعی، مقادیر پیش‌فرض این پارامترها (معمولاً <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> = 1.2</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b = 0.75</span>) برای متون عمومی وب تنظیم شده‌اند. با این حال، برای دامنه‌های تخصصی مانند متون حقوقی که رفتار زبانی متفاوتی دارند، این پارامترها نیازمند کالیبراسیون و تحلیل مجدد هستند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
بر اساس صورت مسئله، با <b>2</b> ویژگی متمایز در مجموعه اسناد حقوقی مواجه هستیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">ویژگی اول:</b> تکرار زیاد یک کلمه کلیدی، ارزش اطلاعاتی آن را به همان نسبت بالا نمی‌برد و اشباع به سرعت رخ می‌دهد.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">ویژگی دوم:</b> اسناد طولانی به دلیل ماهیت جامع و چندموضوعی بودنِ پرونده‌های حقوقی قطور هستند، نه به دلیل استفاده از کلمات زائد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Verbosity)</span>؛ پس نباید جریمه شوند.</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تحلیل و تنظیم پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub></span> (اشباع بسامد کلمه)</b><br>
        پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> &ge; 0</span> تعیین می‌کند که افزایش تکرار یک کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(TF)</span>، تا چه حد و با چه سرعتی می‌تواند نمره سند را افزایش دهد.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>تحلیل رفتار ریاضی:</b> اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> عدد بزرگی باشد، نمره سند با تکرار کلمه به صورت خطی افزایش می‌یابد (اشباعِ دیررس). اما اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> به سمت <b>0</b> میل کند، رفتار تابع به مدل بولین و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> نزدیک می‌شود؛ یعنی صرفاً «حضور» کلمه مهم است و تکرار آن تاثیری در افزایش نمره نخواهد داشت (اشباعِ سریع).</li>
            <li style="margin-bottom:8px;"><b>تصمیم‌گیری برای متون حقوقی:</b> از آنجا که در متون حقوقی بیان شده است تکرار کلمات لزوماً به معنای مرتبط بودن بیشتر نیست (اشباع سریع فرکانس)، باید <b>مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub></span> را کاهش داده و به سمت 0 متمایل کنیم</b> (مثلاً در بازه <b>0.1</b> تا <b>0.5</b>). با این کار، سندی که کلمه را <b>1</b> بار آورده با سندی که آن را <b>20</b> بار تکرار کرده است، تفاوت امتیاز چشمگیری نخواهند داشت.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام دوم: تحلیل و تنظیم پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">b</span> (نرمال‌سازی طول)</b><br>
        پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b &isin; [0, 1]</span> تعیین می‌کند که طول یک سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">L<sub>d</sub></span>) نسبت به میانگین طول اسناد مجموعه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">L<sub>avg</sub></span>)، چقدر باید نمره نهایی آن سند را جریمه کند.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>تحلیل رفتار ریاضی:</b> اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b = 1</span> باشد، سیستم فرض می‌کند طولانی بودن سند صرفاً به دلیل پرگویی است و اسناد طولانی را به شدت جریمه می‌کند. در مقابل، اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b = 0</span> باشد، عامل طول سند به طور کامل از معادله حذف شده و هیچ سندی به خاطر طولانی بودن جریمه نمی‌شود.</li>
            <li style="margin-bottom:8px;"><b>تصمیم‌گیری برای متون حقوقی:</b> چون پرونده‌های حقوقی طبیعتاً طولانی هستند و این طولانی بودن نشان‌دهنده پوشش جامع‌تر اطلاعات قانونی است، نباید سیستم این اسناد ارزشمند را سرکوب کند. بنابراین باید <b>مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">b</span> را به شدت کاهش داده و به سمت 0 نزدیک کنیم</b> (مثلاً در بازه <b>0.1</b> تا <b>0.2</b>).</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای بهینه‌سازی موتور جستجوی حقوقی، مهندس جستجو باید هر <b>2</b> پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> را برخلاف مقادیر پیش‌فرضِ وب، <b>به سمت 0 میل دهد</b>. کاهش <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> اثر تکرار مکرر کلمات را خنثی می‌کند و کاهش <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> مانع از جریمه شدن غیرعادلانه پرونده‌های قطور و طولانی می‌شود.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
تاثیر مستقیم این تنظیمات را می‌توان در کسرِ مربوط به وزن <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> در فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> مشاهده کرد:
<div style="text-align:center; direction:ltr; font-size:18pt; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{\frac{TF \times (k_1 + 1)}{TF + k_1 \times (1 - b + b \times \frac{L_d}{L_{avg}})}}$$
</div>
اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b = 0</span> قرار دهیم، عبارت <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\frac{L_d}{L_{avg}}\)</span> کاملاً خنثی شده و طول سند بی‌اثر می‌شود. اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> = 0</span> قرار دهیم، کل این کسر مستقل از مقدار متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span>، همواره برابر با عدد <b>1</b> می‌شود که دقیقاً همان رفتارِ استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> را شبیه‌سازی می‌کند. در دامنه حقوقی، ما دقیقاً به دنبال همین رفتارِ شبه-باینری اما با مقداری انعطاف‌پذیریِ کنترل‌شده هستیم.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.22</span></b>: تحلیل ریاضی پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در تابع رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span>، پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> نقش «نرمال‌ساز طول سند» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Document Length Normalization)</span> را ایفا می‌کند. هدف این پارامتر، تعدیل نمره اسناد بر اساس طول آن‌ها نسبت به میانگین طول اسناد مجموعه است تا از سوگیری سیستم به سمت اسناد بسیار طولانی (که به صورت تصادفی کلمات بیشتری دارند) جلوگیری شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل فرمول
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
بخشی از فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> که مسئول وزن‌دهی به بسامد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(TF)</span> است، به صورت زیر تعریف می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{\frac{TF \times (k_1 + 1)}{TF + k_1 \times \left(1 - b + b \frac{dl}{avdl}\right)}}$$
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
در اینجا:
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{dl}\)</span>: طول سند مورد نظر</li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{avdl}\)</span>: میانگین طول اسناد در کل مجموعه</li>
    <li style="margin-bottom:10px;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{b \in [0, 1]}\)</span>: پارامتر کنترل‌کننده نرمال‌سازی</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام (تحلیل رفتار ریاضی)
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">حالت اول: تحلیل رفتار سیستم در صورت <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">b = 1</span></b><br>
        اگر مقدار پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> را برابر با <b>1</b> قرار دهیم، عبارت داخل پرانتز به صورت زیر ساده می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\left(1 - 1 + 1 \times \frac{dl}{avdl}\right) = \frac{dl}{avdl}}$$
        </div>
        با جایگذاری در فرمول وزن‌دهی:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\frac{TF \times (k_1 + 1)}{TF + k_1 \times \left(\frac{dl}{avdl}\right)}}$$
        </div>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>نتیجه تحلیل:</b> در این حالت، مخرج کسر به شدت وابسته به نسبت <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\frac{dl}{avdl}\)</span> است. اگر طول سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">dl</span>) بسیار بزرگتر از میانگین (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">avdl</span>) باشد، مخرج کسر بزرگ‌تر شده و در نتیجه، امتیاز نهایی سند (وزن کلمه) به شدت کاهش می‌یابد.</li>
            <li style="margin-bottom:8px;"><b>رفتار سیستم:</b> این تنظیم باعث می‌شود اسناد طولانی به طور سیستماتیک "جریمه" شوند. سیستم در این حالت فرض می‌کند که تکرار یک کلمه در یک سند طولانی، ارزش کمتری نسبت به تکرار همان کلمه در یک سند کوتاه دارد.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">حالت دوم: تحلیل رفتار سیستم در صورت <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">b = 0</span></b><br>
        اگر مقدار پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> را برابر با <b>0</b> قرار دهیم، عبارت داخل پرانتز به صورت زیر ساده می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\left(1 - 0 + 0 \times \frac{dl}{avdl}\right) = 1}$$
        </div>
        با جایگذاری در فرمول وزن‌دهی:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\frac{TF \times (k_1 + 1)}{TF + k_1}}$$
        </div>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>نتیجه تحلیل:</b> در این حالت، عبارت <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\frac{dl}{avdl}\)</span> به طور کامل از معادله حذف شده و عملاً طول سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">dl</span>) هیچ تأثیری بر مخرج کسر ندارد.</li>
            <li style="margin-bottom:8px;"><b>رفتار سیستم:</b> در این وضعیت، سیستم نسبت به طول سند کاملاً "کور" <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Blind)</span> است. تفاوتی نمی‌کند سند <b>100</b> کلمه باشد یا <b>10,000</b> کلمه؛ اگر میزان تکرار کلمه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span>) یکسان باشد، وزن نهایی آن کلمه در هر <b>2</b> سند دقیقاً برابر خواهد بود. در اینجا تأثیر طول سند به طور کامل خنثی شده است.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
از نظر ریاضی، پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> یک «اهرم تنظیم‌کننده» است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;">با <b style="color:#6D2E46; direction:ltr; display:inline-block;">b = 1</b>، سیستم به سمت <b>سخت‌گیری حداکثری</b> و جریمه کردن اسناد طولانی حرکت می‌کند.</li>
    <li style="margin-bottom:10px;">با <b style="color:#6D2E46; direction:ltr; display:inline-block;">b = 0</b>، سیستم <b>نرمال‌سازی را غیرفعال می‌کند</b> و طول سند را کاملاً نادیده می‌گیرد.</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
انتخاب مقدار بهینه برای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> (معمولاً بین <b>0.5</b> تا <b>0.8</b>)، تعادلی میان این <b>2</b> دیدگاه ایجاد می‌کند تا اسناد مفید و جامع، بدون جریمه شدن یا بدون سوگیریِ ناعادلانه به سمت طول زیاد، رتبه‌بندی شوند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.23</span></b>: شبکه بیزی 3 لایه برای بازیابی اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Three-Layer Bayesian Network for IR)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
استفاده از «شبکه‌های استنتاجی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Inference Networks)</span> که بر پایه شبکه‌های بیزی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bayesian Networks)</span> بنا شده‌اند، یکی از پیشرفته‌ترین چارچوب‌ها در تئوری بازیابی اطلاعات است (که برای اولین بار توسط <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Turtle</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Croft</span> معرفی شد). این مدل گرافیکی، فرآیند بازیابی را به عنوان یک مسئله «استنتاج احتمالی» مدل‌سازی می‌کند. در این ساختار، متغیرهای تصادفی به عنوان گره‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Nodes)</span> و وابستگی‌های علی یا منطقی میان آن‌ها به عنوان یال‌های جهت‌دار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Directed Edges)</span> در <b>3</b> لایه مجزا تعریف می‌شوند:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> <b>لایه اسناد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(D)</span>:</b> گره‌های ریشه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Roots)</span> که نشان‌دهنده تک‌تک اسناد در پایگاه داده هستند.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> <b>لایه کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(T)</span>:</b> گره‌های میانی که نشان‌دهنده مفاهیم یا کلمات نمایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Index Terms)</span> هستند.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.3</b> <b>لایه پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Q)</span>:</b> گره‌های برگ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Leaves)</span> که نشان‌دهنده نیاز اطلاعاتی یا پرسمان کاربر هستند.</li>
</ul>
</div>

<div style="text-align:center; margin:24px 0;">
    
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل ساختار و رسم مسیر جریان
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در این مدل، جهتِ فلش‌ها (یال‌ها) از سمت اسناد به کلمات (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D &rarr; T</span>) و از سمت کلمات به پرسمان (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T &rarr; Q</span>) است. این جهت‌گیری نشان‌دهنده یک «جریان تولیدی» است: یک سند مفاهیم را فعال می‌کند، و آن مفاهیم به نوبه خود پرسمان را ارضا می‌کنند. وابستگی‌های روی این یال‌ها توسط «جداول احتمالات شرطی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(CPT)</span> مدیریت می‌شوند.<br>
بر اساس صورت مسئله، ما روی مسیر اختصاصی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub> &rarr; T<sub>1</sub> &rarr; Q</span> متمرکز می‌شویم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام (تحلیل جریان استنتاج و محاسبه احتمال پسین)
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: مقداردهی اولیه شبکه (تعریف جداول شرطی و پیشین)</b><br>
        پیش از ورود پرسمان، شبکه در وضعیت خنثی قرار دارد:
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>احتمال پیشین اسناد:</b> هر سند در ابتدا دارای یک احتمال پیشین بسیار کوچک و مساوی است (مثلاً <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(D<sub>1</sub> = 1) = 1/N</span>).</li>
            <li style="margin-bottom:8px;"><b>وابستگی سند-کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(D &rarr; T)</span>:</b> صورت مسئله بیان می‌کند که «سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> دارای وزن بالایی برای کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>1</sub></span> است». در زبان شبکه بیزی، این بدان معناست که احتمال شرطی فعال شدن کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>1</sub></span> به شرط فعال بودن سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> بسیار بالاست. یعنی متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(T<sub>1</sub> = 1 | D<sub>1</sub> = 1)</span> نزدیک به <b>1</b> است.</li>
            <li style="margin-bottom:8px;"><b>وابستگی کلمه-پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(T &rarr; Q)</span>:</b> پرسمان کاربر مستقیماً به کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>1</sub></span> متصل است؛ بنابراین <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(Q = 1 | T<sub>1</sub> = 1)</span> نیز مقدار بالایی دارد.</li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: دریافت شواهد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Observing Evidence)</span></b><br>
        هنگامی که کاربر پرسمان خود را در سیستم ثبت می‌کند، این رویداد در شبکه به عنوان یک «شاهد قطعی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Evidence)</span> ثبت می‌شود. به بیان ریاضی، گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q</span> در شبکه مقدار «درست» یا <b>1</b> به خود می‌گیرد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q = 1</span>). هدف سیستم اکنون این است که متوجه شود کدام سند به احتمال بیشتری باعث به وجود آمدن این شاهد شده است.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام سوم: انتشار باور <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Belief Propagation)</span> و جریان استقرایی</b><br>
        با فعال شدن گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q</span>، اطلاعات برخلاف جهت فلش‌ها در شبکه به سمت بالا جریان می‌یابد (استنتاج تشخیصی یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Diagnostic Reasoning</span>). سیستم می‌خواهد <b>احتمال پسین <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Posterior Probability)</span></b> سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> را با توجه به شواهد جدید محاسبه کند، یعنی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(D<sub>1</sub> = 1 | Q = 1)</span>.<br>
        طبق قضیه بیز، این احتمال برابر است با:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(D_1 = 1 | Q = 1) = \frac{P(Q = 1 | D_1 = 1) \times P(D_1 = 1)}{P(Q = 1)}}$$
        </div>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام چهارم: محاسبه تأثیر لایه میانی (حاشیه‌سازی روی لایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">T</span>)</b><br>
        برای حل صورتِ کسر بالا، سیستم باید احتمال <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(Q = 1 | D<sub>1</sub> = 1)</span> را از طریق لایه میانی (کلمات) محاسبه کند. بر اساس قانون احتمالات کل و فرض استقلال شرطی در شبکه بیزی، این مقدار با جمع زدن روی تمام حالات ممکن گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>1</sub></span> به دست می‌آید:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(Q = 1 | D_1 = 1) = \sum_{t \in \{0, 1\}} P(Q = 1 | T_1 = t) \times P(T_1 = t | D_1 = 1)}$$
        </div>
        از آنجا که می‌دانیم مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(T<sub>1</sub> = 1 | D<sub>1</sub> = 1)</span> به دلیل «وزن بالای کلمه در سند» بسیار بزرگ است، و همچنین <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>1</sub></span> مستقیماً پرسمان را فعال می‌کند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(Q=1|T<sub>1</sub>=1)</span> بزرگ است)، حاصل‌ضرب این <b>2</b> مقدار در معادله بالا عدد بزرگی خواهد شد.
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
ورود پرسمان به عنوان شواهد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Evidence)</span>، گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q</span> را فعال می‌کند. این فعال‌سازی از طریق کلمه کلیدی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>1</sub></span> به عنوان یک پل ارتباطی، به سمت بالا منتقل شده و باورِ شبکه را نسبت به گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> به‌روزرسانی می‌کند. به دلیل وزن بالای ارتباط <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub> &rarr; T<sub>1</sub> &rarr; Q</span>، <b>احتمال پسین <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">P(D<sub>1</sub> = 1 | Q = 1)</span> جهش چشمگیری نسبت به احتمال پیشین خود پیدا می‌کند.</b> سیستم در نهایت تمام اسناد لایه ریشه را بر اساس همین احتمال پسین از بزرگ به کوچک مرتب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Rank)</span> کرده و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> را در صدر نتایج به کاربر نمایش می‌دهد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
قدرت خارق‌العاده شبکه‌های بیزی در بازیابی اطلاعات، در انعطاف‌پذیری «جداول احتمالات شرطی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(CPT)</span> نهفته است. اگر این جداول را با <b>0</b> و <b>1</b>های مطلق مقداردهی کنیم، شبکه بیزی دقیقاً رفتار «مدل منطق بولین» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Boolean Model)</span> را شبیه‌سازی می‌کند. اما اگر آن‌ها را با احتمالاتِ برگرفته از مقادیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF-IDF</span> یا هموارسازی‌های <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> مقداردهی کنیم، شبکه به یک سیستم رتبه‌بندی احتمالی بسیار قدرتمند تبدیل می‌شود. این معماری، مدلی جامع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Unified Model)</span> است که تئوری‌های مختلف بازیابی را زیر یک چتر ریاضی واحد گرد هم می‌آورد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.24</span></b>: تفاوت فلسفی مدل زبان و مدل احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Philosophical Difference of Language vs. Probabilistic Models)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در اواخر دهه <b>90</b> میلادی، معرفی «مدل‌های زبانی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Language Models)</span> یک تغییر پارادایم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Paradigm Shift)</span> بنیادین در تئوری بازیابی اطلاعات ایجاد کرد. تا پیش از آن، تمامی مدل‌های احتمالی کلاسیک (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>) حول محور پیش‌بینی یک متغیر تصادفیِ پنهان به نام «ارتباط» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Relevance)</span> می‌چرخیدند. اما مدل‌های زبانی با اتخاذ یک رویکرد کاملاً متفاوت و معکوس، مفهوم صریح «ارتباط» را از معادلات ریاضی خود حذف کردند و به جای آن، از مفهوم «تولید متن» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Text Generation)</span> بهره بردند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل و مدل‌سازی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک این تفاوت فلسفی، باید نگاه سیستم به کاربر و نحوه شکل‌گیری پرسمان را در <b>2</b> مدل مقایسه کنیم. این تفاوت در تابع هدف و جهتِ شرطی شدن متغیرها خود را نشان می‌دهد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: رویکرد احتمالی کلاسیک (نگاه تشخیصی در <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span>)</b><br>
        در مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span>، سیستم یک نگاه «تشخیصی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Diagnostic)</span> دارد. سیستم سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d</span>) و پرسمان (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">q</span>) را به عنوان داده‌های ورودی دریافت می‌کند و تلاش می‌کند احتمال مرتبط بودن آن‌ها را محاسبه کند.<br>
        تابع هدف در این حالت به صورت زیر تعریف می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(R=1 | d, q)}$$
        </div>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>فلسفه:</b> «با توجه به شواهدی که از سند و پرسمان می‌بینم، چقدر احتمال دارد که این سند برای کاربر مفید (مرتبط) باشد؟» در این نگاه، ارتباط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R</span>) یک ویژگی عینی است که سیستم باید آن را کشف کند و برای این کار، شدیداً به داده‌های آموزشی یا بازخورد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Relevance Feedback)</span> نیاز دارد.</li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: رویکرد مدل زبانی (نگاه تولیدی یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Generative</span>)</b><br>
        مدل زبانی، متغیر ارتباط (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R</span>) را به طور کامل کنار می‌گذارد. در این فلسفه، هر سند در پایگاه داده به عنوان یک «مدل زبانی مجزا» (که با نماد <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">M<sub>d</sub></span> نشان داده می‌شود) در نظر گرفته می‌شود؛ گویی هر سند، یک جهان کوچک با دایره واژگان و توزیع احتمالات خاص خود است.<br>
        تابع هدف در این مدل به صورت معکوس و به شکل زیر تعریف می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(q | M_d)}$$
        </div>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>فلسفه:</b> سیستم از خود می‌پرسد: «اگر کاربری در جهانِ کلماتِ این سندِ خاص زندگی می‌کرد و می‌خواست به صورت تصادفی کلماتی را از داخل این سند بیرون بکشد، چقدر احتمال داشت که دقیقاً کلمات پرسمانِ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">q</span> را تولید کند؟»</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: تحلیلِ معکوس بودنِ مدل‌سازی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(The Inverse Relationship)</span></b><br>
        تفاوت اصلی در <b>«جهت پیکان شرطی‌سازی»</b> است:
        <ul style="margin-right:20px; padding-right:15px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">در BIM</b> پیکان از سمت ویژگی‌های سند و پرسمان، به سمت متغیر ارتباط می‌رود (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d, q &rarr; R</span>). سیستم می‌پرسد: آیا این سند برای پرسمان، یک پاسخِ مرتبط است؟</li>
            <li style="margin-bottom:10px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">در مدل زبانی (LM)</b> پیکان از سمت سند به سمت پرسمان می‌رود (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d &rarr; q</span>). سیستم می‌پرسد: آیا این سند، <b>نویسنده (تولیدکننده)</b> این پرسمان است؟</li>
        </ul>
        در مدل زبانی، کاربر به عنوان شخصی در نظر گرفته می‌شود که یک سند ایده‌آل را در ذهن خود دارد و پرسمان را با انتخاب کلماتی از آن سند ذهنی تولید کرده است. کار موتور جستجو این است که بگردد و ببیند کدام سندِ واقعی در پایگاه داده، بیشترین شباهت را به آن مدل ذهنی دارد (بیشترین احتمال برای تولید آن پرسمان را دارد).
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
تفاوت فلسفی این <b>2</b> مدل در این است که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> یک مسئله‌ی <b>«دسته‌بندی دودویی»</b> (طبقه‌بندی اسناد به مرتبط و غیرمرتبط) است، در حالی که مدل زبانی یک مسئله‌ی <b>«تخمین توزیع احتمال»</b> است. در مدل زبانی، سندی مرتبط‌تر ارزیابی می‌شود که زبان و ادبیاتِ پرسمان، هم‌خوانی و هم‌پوشانی بیشتری با توزیع زبانیِ آن سند داشته باشد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
بزرگترین مزیت ریاضیِ این «مدل‌سازی معکوس» این است که مدل‌های زبانی به هیچ‌گونه بازخورد اولیه یا داده آموزشی (برای تخمین <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span>) نیاز ندارند. آن‌ها می‌توانند احتمالات را مستقیماً از روی بسامد کلمات درون خود سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(TF)</span> محاسبه کنند و از طریق تکنیک‌های هموارسازی (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Jelinek-Mercer</span> یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Dirichlet Prior</span>)، به صورت کاملاً طبیعی متغیرهایی مانند «طول سند» و «اهمیت کلمه در کل مجموعه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(IDF)</span>» را در دل فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(q|M<sub>d</sub>)</span> جای دهند، بدون اینکه نیازی به پارامترهای ابتکاری و دستی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Heuristics)</span> باشد.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.25</span></b>: معماری سیستم و فشرده‌سازی در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(System Architecture and Compression in BM25)</span>
</span>
</div>

بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در معماری موتورهای جستجو با مقیاس وب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Web-scale IR Architecture)</span>، محاسبه تابع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> برای رتبه‌بندی اسناد نیازمند دسترسی بی‌درنگ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Real-time)</span> به طول تک‌تک اسناد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">dl</span>) است. اگر پایگاه داده‌ای شامل میلیاردها سند باشد، ذخیره دقیق طول هر سند در قالب اعداد صحیح استاندارد (<b>32</b> بیتی)، به ده‌ها گیگابایت حافظه اصلی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RAM)</span> تنها برای نگهداری آرایه طول‌ها نیاز دارد. از آنجا که دسترسی به دیسک در زمان پردازش پرسمان باعث افت شدید سرعت می‌شود، مهندسان سیستم از ترکیب <b>2</b> تکنیک «فشرده‌سازی با اتلاف» (کوانتایزیشن/دسته‌بندی) و «فشرده‌سازی بدون اتلاف» (کدگذاری طول متغیر) استفاده می‌کنند تا این ساختار داده را در سریع‌ترین لایه‌های حافظه جا دهند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
فرض کنید موتور جستجویی دارای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">N = 10<sup>10</sup></span> (<b>10</b> میلیارد) سند است.
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">حالت سنتی:</b> ذخیره طول دقیق هر سند به عنوان یک عدد <b>32</b> بیتی (<b>4</b> بایت)، نیازمند <b>40</b> گیگابایت حافظه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RAM</span> است که از نظر معماری گلوگاه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bottleneck)</span> محسوب می‌شود.</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">رفتار ریاضی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BM25</span>:</b> فرمول نرمال‌سازی طول در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> در برابر تغییرات بسیار جزئیِ طول سند، مقاوم است.
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{1 - b + b \cdot \frac{dl}{avdl}}$$
        </div>
    به عنوان مثال، تفاوتی بین سندی با طول <b>1000</b> کلمه و سندی با طول <b>1005</b> کلمه در خروجی نهایی رتبه‌بندی احساس نمی‌شود. این ویژگی به ما اجازه می‌دهد به جای مقادیر دقیق، از مقادیر تقریبی استفاده کنیم.</li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) طراحی استراتژی معماری
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تقریب مقادیر از طریق دسته‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Length Quantization & Bucketing)</span></b><br>
        به جای ذخیره طول دقیق، فضای طول اسناد را به تعدادی «دسته» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bucket)</span> تقسیم می‌کنیم. این دسته‌بندی به صورت غیرخطی (معمولاً لگاریتمی) انجام می‌شود؛ زیرا برای اسناد کوتاه، تفاوت <b>10</b> کلمه بسیار مهم است، اما برای اسناد بسیار طولانی (مثلاً ده‌ها هزار کلمه)، تفاوت <b>500</b> کلمه نیز تأثیری در نرمال‌سازی ندارد.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;">سیستم تمام طول‌های ممکن را به تعداد محدودی دسته (مثلاً <b>256</b> دسته) نگاشت می‌کند.</li>
            <li style="margin-bottom:8px;">اکنون به جای ذخیره طولِ واقعی، تنها «شناسه دسته» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bucket ID)</span> را ذخیره می‌کنیم که عددی بین <b>0</b> تا <b>255</b> است.</li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: اعمال کدگذاری طول متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Variable-Length Encoding)</span></b><br>
        با کاهش مقادیر به شناسه‌های کوچک، اکنون از تکنیک‌های فشرده‌سازی بدون اتلاف مانند «کدگذاری بایت متغیر» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Variable Byte Code)</span> یا کدهای بیتی (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Elias Gamma</span>) برای ذخیره این شناسه‌ها در حافظه استفاده می‌کنیم.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;">از آنجا که در وب، توزیع طول اسناد از «قانون زیف» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Zipf's Law)</span> یا توزیع‌های توانی پیروی می‌کند، اکثریت قریب به اتفاق اسنادِ وب نسبتاً کوتاه هستند و در دسته‌های اولیه (با شناسه‌های بسیار کوچک مانند <b>1</b>، <b>2</b> یا <b>5</b>) قرار می‌گیرند.</li>
            <li style="margin-bottom:8px;">با استفاده از <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Variable Byte Code</span>، این شناسه‌های کوچک تنها در <b>1 بایت</b> (و گاهی در سطح بیت) ذخیره می‌شوند. این تکنیک، حجم آرایه را از <b>40</b> گیگابایت به حدود <b>10</b> گیگابایت یا حتی کمتر (بسته به الگوریتم بیتی) کاهش می‌دهد.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: ادغام در معماری زمانِ جستجو <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query-Time Architecture)</span></b><br>
        برای استفاده از این داده‌های فشرده‌شده در زمان اجرای پرسمان، سیستم ساختاری به نام «نمایه پیش‌رو» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Forward Index)</span> برای طول اسناد می‌سازد:
        <ul style="margin-right:20px; padding-right:15px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> هنگامی که یک کلمه جستجو می‌شود، سیستم «لیست ارسال» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Posting List)</span> آن کلمه را از نمایه وارونه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Inverted Index)</span> فراخوانی می‌کند.</li>
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> با خواندن شناسه سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(DocID)</span> از لیست ارسال، سیستم مستقیماً به آرایه فشرده‌شده‌ی طول‌ها در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">RAM</span> مراجعه می‌کند.</li>
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.3</b> مقدار فشرده‌شده (کد بایت متغیر) خوانده شده و در کسری از نانوثانیه توسط پردازنده (با دستورات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">SIMD</span>) از حالت فشرده خارج <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Decode)</span> می‌شود.</li>
            <li style="margin-bottom:6px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.4</b> شناسه دسته به دست می‌آید و مقدارِ تقریبی طول سند (از یک جدول مراجعه کوچک یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Lookup Table</span>) استخراج شده و به تابع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> تزریق می‌گردد.</li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
استراتژی ترکیبیِ «تقریب لگاریتمی طول‌ها» و متعاقب آن «فشرده‌سازی با کدهای طول متغیر»، یک راهکار معماری بهینه است که با قربانی کردن بخش بسیار ناچیزی از دقتِ غیرضروری در طول سند، حجم حافظه مورد نیاز را به صورت نمایی کاهش می‌دهد. این معماری تضمین می‌کند که آرایه طول اسناد همواره در سریع‌ترین لایه‌های حافظه پنهان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cache)</span> باقی بماند و سرعت محاسبه تابع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> در مقیاس‌های میلیارد سندی حفظ شود.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این دقیقاً همان استراتژی مهندسیِ درخشانی است که موتورهای جستجوی صنعتی و کتابخانه‌های متن‌باز نظیر <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">Apache Lucene</span></b> (هسته مرکزی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Elasticsearch</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Solr</span>) از آن استفاده می‌کنند. در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Lucene</span> کلاسیک، فاکتورهای نرمال‌سازی طول سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Norms)</span> با استفاده از یک سیستم دسته‌بندی دقیقاً در <b>1</b> بایت (<b>8</b> بیت) کدگذاری می‌شوند (ترکیب <b>3</b> بیت برای مانتیس و <b>5</b> بیت برای نما). درک این ترفندها به دانشجویان نشان می‌دهد که در سیستم‌های توزیع‌شده، تئوری خالصِ بازیابی اطلاعات همواره باید با محدودیت‌های سخت‌افزاری و معماری کامپیوتر هم‌سو و کالیبره شود.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.26</span></b>: مسئله مقداردهی اولیه در سیستم‌های فروشگاهی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Initialization Problem in E-commerce)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در موتورهای جستجوی تجارت الکترونیک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(E-commerce)</span>، یکی از چالش‌های بنیادین، پدیده «شروع سرد» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cold Start)</span> یا مسئله مقداردهی اولیه است. هنگامی که از مدل‌های بازخورد ارتباط احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Probabilistic Relevance Feedback)</span> استفاده می‌شود، سیستم برای ارتقای رتبه‌بندی نیازمند سیگنال‌های رفتاری کاربران (مانند کلیک، افزودن به سبد خرید و ثبت سفارش) است. محصولات جدیدی که به تازگی وارد کاتالوگ فروشگاه شده‌اند، فاقد این تاریخچه رفتاری هستند. در نتیجه، الگوریتم‌های احتمالی در ارزیابی ارزش واقعی این محصولات دچار نقص اطلاعاتی شده و آن‌ها را به صورت ناعادلانه‌ای جریمه می‌کنند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و صورت مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در مدل‌های احتمالی کلاسیک، وزن یک کلمه برای یک سند خاص، پس از دریافت بازخورد کاربر به‌روزرسانی می‌شود (متغیرهای <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">s</span>). برای محصولات قدیمی، کلیک‌های کاربران باعث می‌شود سیستم با اطمینان بالایی احتمال مرتبط بودن آن‌ها را تخمین بزند. اما برای یک محصول جدید (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">d<sub>new</sub></span>):
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;">تعداد کلیک‌ها برابر با <b>0</b> است.</li>
    <li style="margin-bottom:10px;">احتمال مرتبط بودن بر اساس رفتار، معادل شانس تصادفی یا <b>0</b> در نظر گرفته می‌شود.</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
در این رقابت نابرابر، محصول جدید تنها بر اساس تطابق پایه کلمات کلیدی (مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span>) امتیاز می‌گیرد و همواره زیر سایه محصولات قدیمی‌تر و پرکلیک (حتی اگر آن محصولات قدیمی از رده خارج شده باشند) دفن می‌شود.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام با طراحی سناریو
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تعریف سناریوی فروشگاهی</b><br>
        فرض کنید کاربری عبارت پرسمان «گوشی هوشمند سامسونگ با دوربین عالی» را در فروشگاه جستجو می‌کند.
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>سند الف (محصول قدیمی):</b> گوشی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Samsung Galaxy S23</span> که <b>100,000</b> بار کلیک و خرید داشته است. سیستم از طریق بازخورد کاربران آموخته است که این محصول دقیقاً با کلمات «دوربین عالی» مرتبط است (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> بسیار بالا).</li>
            <li style="margin-bottom:8px;"><b>سند ب (محصول جدید):</b> گوشی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Samsung Galaxy S24</span> که امروز صبح به سایت اضافه شده است. هیچ کلیکی ندارد.</li>
        </ul>
        با وجود اینکه سند «ب» مدل جدیدتر و مرتبط‌تری است، مدل پایه احتمالی به دلیل نبود داده‌های بازخورد، سند «الف» را در رتبه بسیار بالاتری قرار می‌دهد.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: راهکار حل مسئله (تغییر پارادایم از رفتار به محتوا)</b><br>
        برای حل این مشکل، باید متغیر پنهان «ارتباط» (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">R</span>) را به جای اینکه منحصراً به شناسه سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Document ID)</span> و کلیک‌های آن گره بزنیم، به «ویژگی‌های محتوایی و متادیتا» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Content Features)</span> متصل کنیم.<br>
        هر محصول جدید، اگرچه کلیک ندارد، اما دارای مجموعه‌ای از ویژگی‌های ساختاریافته <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">F(d)</span> است:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{F(d_{new}) = \{f_1, f_2, f_3, \dots, f_n\}}$$
        </div>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام سوم: فرمول‌بندی احتمال پیشین مبتنی بر ویژگی‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Feature-based Prior Probability)</span></b><br>
        در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> گسترش‌یافته، تابع امتیازدهی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(RSV)</span> شامل بخشی برای «احتمال پیشین» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Prior Probability)</span> است که معمولاً برای همه اسناد ثابت فرض شده و حذف می‌گردد. اکنون ما این مقدار ثابت را بازگردانده و آن را به عنوان یک تابع پویا از ویژگی‌های محصول جدید تعریف می‌کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{RSV_d = \log \frac{P(R=1 | F(d))}{P(R=0 | F(d))} + \sum_{x_i=1} \log \frac{p_i(1 - u_i)}{u_i(1 - p_i)}}$$
        </div>
        در این معادله، عبارت اول (کسرِ پیشین)، ارزش محتوایی محصول را قبل از هرگونه بررسی کلمات کلیدی تخمین می‌زند.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام چهارم: محاسبه شانس محتوایی از طریق تعمیم داده‌ها <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Generalization)</span></b><br>
        سیستم برای محاسبه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(R=1 | F(d<sub>new</sub>))</span> به پایگاه داده تاریخی خود مراجعه می‌کند. سیستم مشاهده می‌کند که برای جستجوی «گوشی هوشمند با دوربین عالی»، اسنادی که دارای ویژگی‌های <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">f<sub>1</sub> = Samsung</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">f<sub>2</sub> = Flagship</span> (پرچمدار) بوده‌اند، در گذشته نرخ کلیک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(CTR)</span> بسیار بالایی (مثلاً <b>40</b> درصد) داشته‌اند.<br>
        بنابراین، سیستم این نرخ موفقیت تاریخی را به عنوان احتمال پایه به محصول جدید (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">S24</span>) که دقیقاً همین ویژگی‌ها را دارد، تزریق می‌کند (به ارث می‌رساند).
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مسئله مقداردهی اولیه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Cold Start)</span> در سیستم‌های فروشگاهی با ادغام «ویژگی‌های محتوایی» در دل «مدل احتمالی» حل می‌شود. با این معماری، محصول جدیدی که هیچ کلیکی ندارد، اعتبار اولیه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Prior Belief)</span> خود را از موفقیتِ تاریخیِ محصولاتی که شبیه به آن بوده‌اند (هم‌خانواده، هم‌برند، یا هم‌رده) وام می‌گیرد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این روش در ادبیات بازیابی اطلاعات تحت عنوان <b>«هموارسازی مبتنی بر ویژگی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Feature-based Smoothing)</span></b> یا ترکیب مدل‌های محتوا-محور با فیلتر مشارکتی شناخته می‌شود. در موتورهای جستجوی مدرن تجارت الکترونیک (مانند آمازون)، بخش لگاریتمیِ اول فرمول ما، توسط مدل‌های یادگیری ماشین (مانند درخت‌های تصمیم تقویتی یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">GBDT</span>) تخمین زده می‌شود تا به محض ورود یک کالای جدید، یک امتیاز پایه یا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Boost</span> منطقی به آن اختصاص یابد و کالا شانس دیده شدن و جمع‌آوری کلیک‌های اولیه را پیدا کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.27</span></b>: مقایسه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Comparing Okapi BM25 and BIM)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> به عنوان نقطه شروع مدل‌های احتمالاتی، یک چارچوب ریاضی مستحکم برای بازیابی اطلاعات ارائه داد. اما صفت «باینری» در این مدل، پاشنه آشیل آن در مواجهه با متون واقعی بود؛ زیرا این مدل جهان متن را تنها به صورت «وجود» (<b>1</b>) یا «عدم وجود» (<b>0</b>) کلمات می‌دید و از سیگنال‌های حیاتی دیگر چشم‌پوشی می‌کرد.
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
مدل <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block; color:#6D2E46;">Okapi BM25 (Best Matching 25)</span></b> با حفظ هسته‌ی تئوریک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> (یعنی منطق رتبه‌بندی احتمالی و نسبت شانس)، از قید و بند باینری بودن رها شد. این مدل با معرفی <b>3</b> مولفه «غیر باینری» و پیوسته که ریشه در ویژگی‌های ساختاری متن دارند، ضعف‌های بنیادین <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> را به طور کامل پوشش داد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و مسئله
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک تحلیلی این گذار، باید بررسی کنیم که فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> چگونه بسامد کلمه، طول سند و ساختار پرسمان را از حالت <b>0</b> و <b>1</b> خارج کرده و به توابعی پویا <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Dynamic)</span> تبدیل کرده است. در ادامه، این <b>3</b> تفاوت ساختاری را به صورت گام‌به‌گام کالبدشکافی می‌کنیم.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: گذار از حضور باینری به اشباع بسامد کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Frequency Saturation)</span></b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ضعف <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span>:</b> در مدل باینری، ارزش سندی که کلمه هدف را <b>1</b> بار آورده با سندی که آن را <b>100</b> بار تکرار کرده برابر است (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{x_i \in \{0, 1\}}\)</span>). این یعنی از دست دادن سیگنالِ «تمرکز موضوعی» سند.</li>
            <li style="margin-bottom:8px;"><b>راهکار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BM25</span>:</b> این مدل متغیر دودویی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">x<sub>i</sub></span> را حذف کرده و مفهوم <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">TF</span> (بسامد کلمه)</b> را وارد معادله می‌کند. اما برای جلوگیری از غلبه‌ی اسنادی که یک کلمه را به صورت افراطی تکرار می‌کنند، از یک «تابع اشباع غیرخطی» با پارامتر کنترل‌کننده <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> استفاده می‌کند:
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{\frac{TF \times (k_1 + 1)}{TF + k_1}}$$
                </div>
                از نظر تحلیلی، این فرمول یک رفتار هوشمندانه دارد: در تکرارهای اولیه، نمره به سرعت افزایش می‌یابد (پاداش برای تمرکز روی موضوع)، اما با افزایش بیشتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span>، نمودار به مجانب <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> + 1</span> میل کرده و اشباع می‌شود تا اثر تکرارهای بی‌مورد مهار گردد.
            </li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: گذار از کوریِ ساختاری به نرمال‌سازی طول سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Document Length Normalization)</span></b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ضعف <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span>:</b> مدل باینری طول سند را نادیده می‌گیرد. یک پاراگراف و یک کتاب برای این مدل تفاوتی ندارند، که این امر موجب «سوگیری طول» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Length Bias)</span> به نفع اسناد طولانی می‌شود، زیرا اسناد طولانی به صورت تصادفی کلمات بیشتری را در خود جای می‌دهند.</li>
            <li style="margin-bottom:8px;"><b>راهکار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BM25</span>:</b> مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> مفهوم طول سند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">dl</span>) و میانگین طول اسناد مجموعه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">avdl</span>) را از طریق پارامتر تنظیمی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> مستقیماً به مخرجِ کسرِ بسامد کلمه تزریق می‌کند:
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{\frac{TF \times (k_1 + 1)}{TF + k_1 \times \left(1 - b + b \frac{dl}{avdl}\right)}}$$
                </div>
                این رویکرد کاملاً غیر باینری، جریمه‌ای متناسب و پیوسته اعمال می‌کند. سندی که طول آن دقیقاً برابر میانگین باشد (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{\frac{dl}{avdl} = 1}\)</span>)، هیچ تغییری در نمره احساس نمی‌کند؛ اما اسنادِ بیش از حد طولانی، با بزرگ شدن مخرج کسر، با کاهش هوشمندانه‌ی وزن کلماتشان مواجه می‌شوند.
            </li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: گذار از پرسمان باینری به وزن‌دهی بسامد پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query Term Frequency)</span></b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b>ضعف <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span>:</b> در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span>، پرسمانِ کاربر نیز به عنوان یک بردار باینری در نظر گرفته می‌شود (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{q_i \in \{0, 1\}}\)</span>). اگر کاربری در یک پرسمان طولانی (مانند جستجوی مفهومی یا توصیفی) یک کلمه کلیدی را چند بار تکرار کند تا اهمیت آن را نشان دهد، سیستم آن را نادیده می‌گیرد.</li>
            <li style="margin-bottom:8px;"><b>راهکار <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BM25</span>:</b> مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> به جای رفتار باینری با پرسمان، متغیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">qtf</span> (بسامد کلمه در پرسمان) را معرفی کرده و با استفاده از پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>3</sub></span>، وزن کلمات را در سمتِ پرسمان نیز تنظیم می‌کند:
                <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
                $$\boldsymbol{\frac{qtf \times (k_3 + 1)}{qtf + k_3}}$$
                </div>
                این ویژگی در سیستم‌های مدرن که پرسمان‌ها به شکل متون بلند (مثل پاراگرافی از یک مقاله برای پیدا کردن مقالات مشابه) وارد سیستم می‌شوند، به شدت حیاتی است و به مدل اجازه می‌دهد هسته معنایی پرسمان را از طریق تکرار کلمات در آن شناسایی کند.
            </li>
        </ul>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> یک شاهکار در مهندسی بازیابی اطلاعات است، زیرا توانست <b>تئوری احتمالات مستحکم <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span></b> (از طریق استفاده از منطق <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> پایه رابرتسون-اسپارک جونز) را با <b>ویژگی‌های وزن‌دهی پیوسته‌ی مدل فضای برداری</b> (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> و نرمال‌سازی طول) ترکیب کند. با کنار گذاشتن محدودیت‌های <b>0</b> و <b>1</b>، <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> به مدلی منعطف و قابل کالیبره (پارامتریک) تبدیل شد که تا به امروز به عنوان خط پایه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Baseline)</span> و استاندارد صنعتی در موتورهای جستجو شناخته می‌شود.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
اگر پارامترهای مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> را به صورت حدی دستکاری کنیم، می‌توانیم آن را به مدل‌های دیگر تقلیل دهیم. به عنوان مثال، اگر در فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> مقدار <b><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub> = 0</span></b> قرار داده شود، اثر بسامد کلمه و طول سند کاملاً از بین می‌رود و کل کسرِ مربوط به <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> برابر با عدد <b>1</b> می‌شود. در این حالت خاص، خروجی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> دقیقاً به فرمول پایه و باینریِ <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> (فقط جمع اوزان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span>) تبدیل خواهد شد. این امر نشان می‌دهد که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> صرفاً یک حالت خاص و محدودشده از مدل جامع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> است.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.28</span></b>: سناریوی شبکه‌های بیزی با کلمات مترادف <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Bayesian Networks Scenario with Synonyms)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
یکی از چالش‌های کلاسیک در مدل‌های بازیابی اطلاعات (به‌ویژه مدل‌های مبتنی بر تطابق دقیق کلمات مانند مدل فضای برداری پایه یا مدل بولین)، مسئله «اختلاف واژگانی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Vocabulary Mismatch)</span> است. کاربران غالباً برای بیان یک نیاز اطلاعاتی از کلماتی استفاده می‌کنند که نویسنده سند از آن‌ها استفاده نکرده است (مانند جستجوی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Car"</span> برای سندی که از کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Automobile"</span> استفاده کرده است).
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
در معماری «شبکه‌های استنتاجی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Inference Networks)</span> که مبتنی بر شبکه‌های بیزی هستند، این مشکل به ظریف‌ترین شکل ممکن و از طریق افزودن یک لایه جدید به نام <b>«لایه مفاهیم» <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Concept Layer)</span></b> حل می‌شود. این لایه به عنوان یک پل معنایی میان کلماتِ فیزیکیِ موجود در متن و کلماتِ پرسمان کاربر عمل می‌کند.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>)تحلیل ساختار
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای مدل‌سازی مترادف‌ها، معماری <b>3</b> لایه‌ی استاندارد (سند &rarr; کلمه &rarr; پرسمان) به یک معماری <b>4</b> لایه ارتقا می‌یابد. در این معماری، چندین کلمه فیزیکیِ مترادف به یک «گره مفهومیِ» واحد متصل می‌شوند.<br>
نمای شماتیک گراف برای سناریوی مورد بحث به شکل زیر است:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">لایه اسناد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Document Node)</span>:</b> گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> (سند شامل کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Automobile</span>)</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">لایه کلمات <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Term Nodes)</span>:</b> گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>a</sub></span> (کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Automobile</span>) و گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>c</sub></span> (کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Car</span>)</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">لایه مفاهیم <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Concept Node)</span>:</b> گره میانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>1</sub></span> (مفهوم انتزاعی «خودرو / وسیله نقلیه»)</li>
    <li style="margin-bottom:10px;"><b style="color:#6D2E46;">لایه پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Query Node)</span>:</b> گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q</span> (پرسمان ثبت‌شده توسط کاربر: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Car"</span>)</li>
</ul>
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify; margin-top:14px;">
<b style="color:#6D2E46;">مسیر اتصال در گراف <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Topology)</span>:</b>
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{D_1 \longrightarrow T_a \longrightarrow C_1 \longrightarrow Q \longleftarrow T_c}$$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>)تحلیل جریان انتشار احتمالات
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: مقداردهی جداول احتمالات شرطی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(CPT)</span> برای یال‌های شبکه</b><br>
        سیستم برای برقراری این اتصالات معنایی، جداول احتمالات را در لایه‌های مختلف مقداردهی می‌کند:
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> <b>یال <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">D<sub>1</sub> &rarr; T<sub>a</sub></span>:</b> چون کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Automobile"</span> صراحتاً در متن سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> وجود دارد، احتمال فعال شدن این کلمه توسط سند بسیار بالاست (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{P(T_a=1 | D_1=1) \approx 1}\)</span>).</li>
            <li style="margin-bottom:8px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;".>2</b> <b>یال‌های <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">T<sub>a</sub> &rarr; C<sub>1</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">T<sub>c</sub> &rarr; C<sub>1</sub></span>:</b> سیستم (از طریق یک هستان‌شناسی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Ontology)</span> یا گنج‌واژه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Thesaurus)</span> مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">WordNet</span>) می‌داند که هر <b>2</b> کلمه، نمایانگر مفهوم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>1</sub></span> هستند. بنابراین، فعال شدن هر یک از این کلمات، گره مفهومی را با احتمال بالایی فعال می‌کند (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(C<sub>1</sub>=1 | T<sub>a</sub>=1)</span> بزرگ است).</li>
            <li style="margin-bottom:8px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.3</b> <b>یال <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">C<sub>1</sub> &rarr; Q</span>:</b> پرسمان کاربر با استفاده از مفهومِ خودرو شکل گرفته است، بنابراین وابستگی شدیدی میان این <b>2</b> گره وجود دارد.</li>
        </ul>
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: مشاهده شواهد <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Observing the Evidence)</span></b><br>
        کاربر پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Car"</span> را وارد می‌کند. در شبکه بیزی، گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q</span> به عنوان یک شاهد قطعی مقدار «درست» به خود می‌گیرد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q = 1</span>). اکنون سیستم باید استنتاج تشخیصی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Diagnostic Inference)</span> را آغاز کرده و احتمالات را برخلاف جهت یال‌ها به سمت بالا (لایه اسناد) منتشر کند.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام سوم: انتشار باور از طریق گره میانی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Belief Propagation)</span></b>
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;">گره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Q</span> فعال شده است و به دنبال مفاهیمِ تولیدکننده‌اش می‌گردد. این امر باعث می‌شود گره مفهومی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>1</sub></span> در لایه بالاتر فعال شود.</li>
            <li style="margin-bottom:8px;">گره مفهومی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>1</sub></span> به <b>2</b> کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>a</sub></span> و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>c</sub></span> متصل است. سیستم بررسی می‌کند که کدام اسناد این کلمات را فعال کرده‌اند.</li>
            <li style="margin-bottom:8px;">سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>a</sub></span> (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Automobile"</span>) را فعال کرده است. این فعال‌سازی از طریق مسیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub> &rarr; T<sub>a</sub> &rarr; C<sub>1</sub></span> به گره مفهومی منتقل می‌شود.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام چهارم: محاسبه احتمال پسین <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Posterior Probability)</span></b><br>
        با وجود اینکه سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> هیچ اتصال مستقیمی به کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Car"</span> (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">T<sub>c</sub></span>) ندارد، اما به دلیل وجود مسیر جایگزین از طریق گره مفهومی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">C<sub>1</sub></span>، احتمال پسینِ مرتبط بودن سند با پرسمان، یعنی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">P(D<sub>1</sub>=1 | Q=1)</span>، مقدار قابل‌توجهی دریافت می‌کند. طبق قوانین استقلال شرطی در گراف، این مقدار از طریق جمع زدن روی حالات مختلف گره‌های میانی (حاشیه‌سازی) محاسبه می‌شود:
        <div style="text-align:center; direction:ltr; font-size:16pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{P(Q=1 | D_1=1) = \sum_{c \in \{0,1\}} \sum_{t \in \{0,1\}} P(Q=1 | C_1=c) P(C_1=c | T_a=t) P(T_a=t | D_1=1)}$$
        </div>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در شبکه‌های بیزی، مفهوم «کلمات مترادف» با اضافه کردن یک «لایه مفاهیم» در میانه شبکه و متصل کردن کلمات هم‌معنی به یک گره مفهومی واحد، مدل‌سازی می‌شود. این معماری به سیستم اجازه می‌دهد تا پرسمان <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Car"</span> را ثبت کرده و از طریق گره میانی (مفهوم خودرو)، جریانی از احتمالات را به سمت کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Automobile"</span> و در نهایت به سمت سند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">D<sub>1</sub></span> هدایت کند؛ در نتیجه، سند مرتبط با وجود عدم تطابق لغوی، با موفقیت بازیابی و رتبه‌بندی می‌شود.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
یکی از مزیت‌های قدرتمند این معماری نسبت به روش‌های سنتیِ «بسط پرسمان» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Query Expansion)</span> این است که سیستم درگیر اضافه‌بارِ کلمات زائد نمی‌شود. در روش‌های سنتی، کلمه <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">"Automobile"</span> به صورت دستی به پرسمان کاربر اضافه می‌شود که ممکن است تعادل وزنیِ فرمول رتبه‌بندی را به هم بزند. اما در شبکه بیزی، این ارتباطِ معنایی به صورت یک «احتمال شرطی کنترل‌شده» در پس‌زمینه گراف لحاظ می‌شود و سیستم می‌تواند وزنِ مترادف‌ها را به دقت تنظیم کند (مثلاً مترادف‌های دورتر، احتمال شرطی کمتری دریافت کنند).
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

<div style="margin-bottom:24px; text-align:right;">
<span style="font-size:20pt; font-weight:bold; color:#1B263B;">
پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.29</span></b>: تحلیل حساسیت پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> در عمل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Sensitivity Analysis of k<sub>1</sub> in Practice)</span>
</span>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در تابع رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span>، تأثیر تکرار یک کلمه بر نمره نهایی به صورت خطی نیست، بلکه از یک تابع «اشباع غیرخطی» پیروی می‌کند. پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> (که معمولاً مقدار پیش‌فرض آن بین <b>1.2</b> تا <b>2.0</b> تنظیم می‌شود)، به عنوان «کنترل‌کننده سرعت اشباع» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Saturation Rate)</span> عمل می‌کند. تحلیل حساسیت این پارامتر به ما نشان می‌دهد که چگونه می‌توان با تغییر آن، رفتار مدل احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> را بین <b>2</b> طیف کاملاً متفاوتِ «مدل بولی/باینری» و «مدل فضای برداری خام» جابه‌جا کرد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل داده‌ها و فرمول پایه
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای تحلیل اثر بسامد کلمه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span>) و پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span>، فرض می‌کنیم طول سند برابر با میانگین طول اسناد است تا اثر پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">b</span> خنثی شود. در این حالت، بخش مربوط به بسامد کلمه در فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> به شکل زیر ساده می‌شود:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{W(TF) = \frac{TF \times (k_1 + 1)}{TF + k_1}}$$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">3</span></b>) حل گام‌به‌گام (تحلیل نموداری و ریاضی)
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: تحلیل رفتار نمودار در مقادیر استاندارد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub> &approx; 1.2</span>)</b><br>
        اگر نمودار تابع <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">W(TF)</span> را رسم کنیم (محور افقی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> و محور عمودی نمره <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">W</span>)، برای مقادیر استانداردی مانند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> = 1.2</span>، نمودار در ابتدا با شیب تندی بالا می‌رود. این یعنی بار <b>1</b> و <b>2</b> که کلمه در متن تکرار می‌شود، ارزش بسیار زیادی دارد. اما به سرعت شیب نمودار کم شده و به یک خط افقی (مجانب) نزدیک می‌شود.<br>
        از نظر ریاضی، حد تابع در بی‌نهایت برابر است با:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\lim_{TF \to \infty} \frac{TF \times (k_1 + 1)}{TF + k_1} = k_1 + 1}$$
        </div>
        یعنی اگر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> = 1.2</span> باشد، نمره تکرار کلمه هرگز از عدد <b>2.2</b> فراتر نخواهد رفت؛ حتی اگر کلمه <b>1000</b> بار تکرار شده باشد.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: افزایش پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub></span> و تغییر شکل نمودار</b><br>
        هرچه مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> را افزایش دهیم (مثلاً <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> = 5</span>، سپس <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> = 20</span> و ...):
        <ul style="margin-right:15px; padding-right:10px; margin-top:8px; list-style-type: none;">
            <li style="margin-bottom:8px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.1</b> <b>تأخیر در اشباع:</b> مجانب نمودار (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> + 1</span>) بالاتر می‌رود.</li>
            <li style="margin-bottom:8px;"><b style="color:#6D2E46; direction:ltr; display:inline-block;">.2</b> <b>کاهش تحدب نمودار:</b> نمودار دیرتر خم می‌شود و بخش اولیه آن که شیب ثابتی دارد، طولانی‌تر می‌گردد. به بیان ساده‌تر، سیستم دیرتر از تکرار کلمات «خسته» می‌شود و برای تکرارهای <b>10</b> و <b>20</b> نیز همچنان پاداشِ قابل‌توجهی در نظر می‌گیرد.</li>
        </ul>
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: اثبات همگرایی به مدل فضای برداری خام (حالت <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">k<sub>1</sub> &rarr; &infin;</span>)</b><br>
        مدل فضای برداری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(VSM)</span> در حالت خام خود (بدون لگاریتم‌گیری از <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span>)، فرض می‌کند که ارزش تکرار کلمه به صورت کاملاً خطی افزایش می‌یابد (یعنی سندی که کلمه را <b>10</b> بار آورده، دقیقاً <b>10</b> برابر ارزشمندتر از سندی است که آن را <b>1</b> بار آورده است).<br>
        اکنون بررسی می‌کنیم که اگر در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span>، مقدار <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> را به سمت بی‌نهایت میل دهیم، چه اتفاقی می‌افتد:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\lim_{k_1 \to \infty} \frac{TF \times (k_1 + 1)}{TF + k_1}}$$
        </div>
        هنگامی که <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> بسیار بزرگتر از محدوده‌ی طبیعی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> باشد (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> &gg; TF</span>)، می‌توانیم از <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> در مخرج کسر و همچنین از عدد <b>1</b> در صورت کسر (در برابر بزرگی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span>) صرف‌نظر کنیم:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{\approx \frac{TF \times k_1}{k_1}}$$
        </div>
        بنابراین معادله به شکل زیر ساده می‌شود:
        <div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
        $$\boldsymbol{W(TF) \approx TF}$$
        </div>
    </li>
</ul>
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
نتیجه‌گیری نهایی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
با افزایش پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span>، سرعت اشباع کاهش یافته و نمودار از حالت یک منحنیِ تند، به یک <b>خط راستِ مورب</b> تبدیل می‌شود. از نظر ریاضی اثبات می‌گردد که وقتی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub> &rarr; &infin;</span>، مخرج کسر عملاً اثر خود را از دست می‌دهد و وزن اختصاص‌یافته به کلمه، مستقیماً و به صورت <b>خطی</b> برابر با خودِ فرکانس کلمه (<span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span>) می‌شود. در این حالتِ حدی، مدل احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> رفتار پیچیده خود را از دست داده و دقیقاً به مدل خطی و خام فضای برداری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Linear Vector Space Model)</span> تقلیل می‌یابد.
</div>

<div style="background-color:#F8F9FA; border-right:4px solid #6D2E46; padding:15px 20px; margin:20px 0; border-radius:4px; font-size:15pt; color:#1B263B; line-height:1.9; text-align:justify; direction:rtl;">
<b style="color:#6D2E46;">نکته آموزشی:</b><br>
این تحلیل، زیبایی و جامعیت معماری <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> را ثابت می‌کند. پارامتر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> در واقع یک «پل ارتباطی» میان مدل‌های مختلف بازیابی اطلاعات است:
<ul style="margin-right:20px; padding-right:15px; margin-top:8px; list-style-type: none;">
    <li style="margin-bottom:6px;">اگر <b style="color:#6D2E46; direction:ltr; display:inline-block;">k<sub>1</sub> = 0</b> تنظیم شود: سیستم دقیقاً مدل <b>استقلال باینری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(BIM)</span></b> را شبیه‌سازی می‌کند (نمره ثابت <b>1</b> برای هر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF &gt; 0</span>).</li>
    <li style="margin-bottom:6px;">اگر <b style="color:#6D2E46; direction:ltr; display:inline-block;">k<sub>1</sub> &approx; 1.2</b> تنظیم شود: سیستم مدل <b>اشباع غیرخطی (<span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BM25</span> استاندارد)</b> را اجرا می‌کند.</li>
    <li style="margin-bottom:6px;">اگر <b style="color:#6D2E46; direction:ltr; display:inline-block;">k<sub>1</sub> &rarr; &infin;</b> تنظیم شود: سیستم مدل <b>فضای برداری خام خطی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">(Linear TF)</span></b> را شبیه‌سازی می‌کند.</li>
</ul>
در نتیجه، سیستم با یک فرمول واحد، می‌تواند رفتار <b>3</b> خانواده‌ی بزرگ از الگوریتم‌های بازیابی را بازتولید کند.
</div>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Tahoma, Arial, sans-serif; background:#FFFFFF; border:1px solid #D8D2E6; border-radius:14px; padding:26px 30px; text-align:right; line-height:2.2; color:#1B263B;">

پاسخنامه مسئله <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">11.30</span></b>: خلاصه و ترکیب مفاهیم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Summary and Concept Integration)</span>
</span>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">
بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
در سیر تکامل علم بازیابی اطلاعات <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(IR)</span>، <b>3</b> مدل «فضای برداری» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(VSM)</span>، «استقلال باینری» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(BIM)</span> و «اوکاپی» <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Okapi BM25)</span> به عنوان ستون‌های اصلی تئوری و پیاده‌سازی سیستم‌های رتبه‌بندی شناخته می‌شوند. هر یک از این مدل‌ها با فلسفه ریاضی متفاوتی به مسئله نگاه می‌کنند: <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span> بر پایه هندسه و جبر خطی، <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> بر پایه تئوری احتمالات کلاسیک و باینری، و <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> بر پایه تلفیق احتمالات با رویکردهای تجربی و غیرخطی بنا شده است. خلاصه کردن ویژگی‌های ساختاری این <b>3</b> مدل، درک عمیقی از نحوه تکامل الگوریتم‌های جستجو به ما می‌دهد.
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:34px 0 16px 0;">
جدول مقایسه‌ای مدل‌های بنیادین بازیابی اطلاعات
</div>

<div style="overflow-x:auto;">
    <table style="width:100%; border-collapse: collapse; border: 1px solid #D8D2E6; margin: 10px 0; font-size: 14pt; line-height: 1.8;">
        <thead style="background-color: #F8F9FA; color: #6D2E46; font-weight: bold; text-align: center;">
            <tr>
                <th style="border: 1px solid #D8D2E6; padding: 14px; width: 16%;">معیار مقایسه</th>
                <th style="border: 1px solid #D8D2E6; padding: 14px; width: 28%;">مدل فضای برداری <span style="direction:ltr; display:inline-block;">(VSM)</span></th>
                <th style="border: 1px solid #D8D2E6; padding: 14px; width: 28%;">مدل استقلال باینری <span style="direction:ltr; display:inline-block;">(BIM)</span></th>
                <th style="border: 1px solid #D8D2E6; padding: 14px; width: 28%;">مدل اوکاپی <span style="direction:ltr; display:inline-block;">(BM25)</span></th>
            </tr>
        </thead>
        <tbody style="text-align: justify; color: #1B263B;">
            <tr>
                <td style="border: 1px solid #D8D2E6; padding: 14px; text-align: center;"><b>1. بسامد کلمه <span style="direction:ltr; display:inline-block; color:#6D2E46;">(TF)</span></b></td>
                <td style="border: 1px solid #D8D2E6; padding: 14px; background-color: #FAFAFC;"><b>پیوسته و غیرخطی:</b> از فرکانس کلمه برای نشان دادن وزن و اهمیت محتوایی استفاده می‌کند (مانند <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{1 + \log TF}\)</span>).</td>
                <td style="border: 1px solid #D8D2E6; padding: 14px;"><b>مطلقاً باینری:</b> بسامد کلمه را کاملاً نادیده می‌گیرد و صرفاً «حضور» (<b>1</b>) یا «عدم حضور» (<b>0</b>) کلمه در متن ملاک است.</td>
                <td style="border: 1px solid #D8D2E6; padding: 14px; background-color: #FAFAFC;"><b>اشباع غیرخطی:</b> از <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> به صورت پیوسته استفاده می‌کند، اما با پارامتر <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">k<sub>1</sub></span> سرعت رشد آن را کنترل و اشباع می‌کند.</td>
            </tr>
            <tr>
                <td style="border: 1px solid #D8D2E6; padding: 14px; text-align: center;"><b>2. نرمال‌سازی طول سند</b></td>
                <td style="border: 1px solid #D8D2E6; padding: 14px;"><b>نرمال‌سازی هندسی:</b> طول سند را از طریق محاسباتی مانند نرمال‌سازی کسینوسی در مخرج کسر خنثی می‌کند.</td>
                <td style="border: 1px solid #D8D2E6; padding: 14px; background-color: #FAFAFC;"><b>نادیده گرفتن کامل:</b> طول سند جایگاهی در فرمول‌بندی ندارد؛ اسناد طولانی شانس بازیابی بیشتری دارند.</td>
                <td style="border: 1px solid #D8D2E6; padding: 14px;"><b>نرمال‌سازی پویا:</b> با استفاده از نسبت (<span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{dl/avdl}\)</span>) و اعمال پارامتر <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">b</span>، جریمه طول را به صورت منعطف اعمال می‌کند.</td>
            </tr>
            <tr>
                <td style="border: 1px solid #D8D2E6; padding: 14px; text-align: center;"><b>3. بازخورد ارتباط</b></td>
                <td style="border: 1px solid #D8D2E6; padding: 14px; background-color: #FAFAFC;"><b>انتقال هندسی:</b> از طریق الگوریتم <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">Rocchio</span>، بردار پرسمان را در فضای چندبعدی به سمت اسناد مرتبط حرکت می‌دهد.</td>
                <td style="border: 1px solid #D8D2E6; padding: 14px;"><b>به‌روزرسانی بیزی:</b> با شمارش کلمات در اسناد مرتبط، پارامترهای احتمالی <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">p<sub>i</sub></span> و <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">u<sub>i</sub></span> را مجدداً محاسبه می‌کند.</td>
                <td style="border: 1px solid #D8D2E6; padding: 14px; background-color: #FAFAFC;"><b>به‌روزرسانی فرمول:</b> هسته <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> را حفظ کرده و با دریافت بازخورد، احتمالات هموارشده را جایگزین <span style="direction:ltr; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> اولیه می‌کند.</td>
            </tr>
        </tbody>
    </table>
</div>

بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">2</span></b>) تحلیل و ترکیب تئوریک مفاهیم

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
برای درک عمیق‌تر این جدول، سیر تکاملی این <b>3</b> مدل را در <b>3</b> گام کالبدشکافی می‌کنیم:
</div>

<div style="font-size:16pt; color:#1B263B; margin-top:12px;">
<ul style="margin-right:20px; padding-right:15px; list-style-type: none; direction:rtl;">
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام اول: ایده هندسی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">VSM</span></b><br>
        مدل فضای برداری با معرفی مفاهیم <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> و نرمال‌سازی طول سند، انقلابی در کیفیت رتبه‌بندی ایجاد کرد. این مدل به خوبی درک کرد که متون، ساختاری فراتر از <b>0</b> و <b>1</b> دارند. با این حال، بزرگترین ضعف <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span>، نبود یک پشتوانه تئوریکِ ساختاریافته در نظریه احتمالات بود؛ فرمول‌های وزن‌دهی آن (مانند <span style="direction:ltr;unicode-bidi:isolate;display:inline-block;color:#6D2E46;">\(\boldsymbol{TF \times IDF}\)</span>) بیشتر بر اساس شَم مهندسی و آزمایش و خطا شکل گرفته بودند.
    </li>
    <li style="margin-bottom:24px;">
        <b style="color:#6D2E46;">گام دوم: ساختار آماری <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span></b><br>
        مدل استقلال باینری برعکس <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">VSM</span>، از یک منطق ریاضی و احتمالیِ فوق‌العاده مستحکم (قضیه بیز و تئوری تصمیم‌گیری) بهره می‌برد. این مدل به ما آموخت که چگونه بازخورد کاربر را به صورت علمی پذیرا باشیم. اما این مدل برای دستیابی به این ریاضیات روان، مجبور شد چشمان خود را بر روی واقعیت‌های فیزیکی متن (مانند تکرار کلمات و طول اسناد) ببندد و همه‌چیز را باینری کند، که این امر کارایی آن را در عمل کاهش داد.
    </li>
    <li style="margin-bottom:10px;">
        <b style="color:#6D2E46;">گام سوم: تکامل نهایی در <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BM25</span></b><br>
        مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> به عنوان نقطه اوج این تکامل شناخته می‌شود. طراحان این مدل متوجه شدند که می‌توان <b>«روح احتمالی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">BIM</span>»</b> را با <b>«جسم محتوایی <span style="direction:ltr; unicode-bidi:embed; display:inline-block;">VSM</span>»</b> ترکیب کرد. <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> فرمول احتمالی نسبت شانس در <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BIM</span> را به عنوان هسته اصلی (بخش <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span>) حفظ کرد، اما محدودیت باینری بودن را شکست و توابع پیوسته و اشباع‌شونده‌ی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">TF</span> و نرمال‌سازی طول سند را به آن تزریق کرد.
    </li>
</ul>

بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
همان‌طور که در مباحث نظری بررسی شد، فرمول <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">IDF</span> در مدل <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Okapi BM25</span> با فرمول کلاسیک <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;"><span style="direction:ltr;unicode-bidi:isolate;display:inline-block;">\(\log(N/df_i)\)</span></span> تفاوت دارد. این فرمول بر پایه احتمالات هموارشده‌ی رابرتسون-اسپارک جونز <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Robertson-Sparck Jones)</span> بنا شده است. در پیاده‌سازی‌های عملی و استاندارد (مانند آنچه در موتورهای جستجوی واقعی نظیر <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">Lucene</span> استفاده می‌شود)، برای جلوگیری از تولید مقادیر منفی برای کلماتی که در بیش از نیمی از اسناد ظاهر شده‌اند <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">(Stop-words)</span>، عدد <b>1</b> به آرگومان لگاریتم اضافه می‌شود.<br>
فرمول استاندارد و ایمن برای پیاده‌سازی به شکل زیر است:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{IDF_i = \log \left( 1 + \frac{N - df_i + 0.5}{df_i + 0.5} \right)}$$
</div>

<div style="font-size:18pt; font-weight:bold; color:#6D2E46; margin:30px 0 14px 0;">

بخش <b style="color:#6D2E46;"><span style="direction:ltr; unicode-bidi:embed; display:inline-block;">1</span></b>) مفاهیم کلیدی
</div>

<div style="font-size:16pt; color:#1B263B; text-align:justify;">
پیاده‌سازی یک موتور رتبه‌بندی <span style="direction:ltr; unicode-bidi:embed; display:inline-block; font-weight:bold; color:#6D2E46;">BM25</span> نیازمند ترکیب <b>2</b> مرحله اصلی است: <b>پیش‌پردازش داده‌های سراسری</b> (مانند محاسبه میانگین طول اسناد و فراوانی کلمات) و <b>امتیازدهی محلی</b> (محاسبه نمره هر سند نسبت به پرسمان).
<br>
هسته محاسباتی این پیاده‌سازی بر اساس معادله استاندارد زیر بنا شده است:
</div>

<div style="text-align:center; direction:ltr; font-size:18pt; background-color:#F8F9FA; border-radius:10px; border:1px solid #EBEBEB; padding:16px; margin:14px 0; overflow-x:auto; color:#6D2E46;">
$$\boldsymbol{Score(Q, d) = \sum_{q_i \in Q} IDF(q_i) \times \frac{tf(q_i, d) \times (k_1 + 1)}{tf(q_i, d) + k_1 \times \left(1 - b + b \frac{dl}{avdl}\right)}}$$
</div>

    Ranks a list of documents against a given query using the Okapi BM25 algorithm.

        query (list of str): The search query tokenized into a list of words.
        documents (list of list of str): A list where each element is a document

    <span style="color:#C586C0;">if</span> N == <span style="color:#B5CEA8;">0</span>:

            denominator = tf + k1 * (<span style="color:#B5CEA8;">1.0</span> - b + b * (doc_len / avdl))
